## 1.4 Executive Summary

### Project Overview

The **Skin Disorder Prediction System** is an end-to-end machine learning project designed to classify structured dermatological cases into one of **six skin disorder classes**.

The project uses a structured clinical dataset containing **366 records and 34 predictive features**. The dataset contains a combination of numerical and categorical variables, with `Age` represented as the primary numerical feature and the remaining 33 predictors represented as categorical clinical attributes.

The project follows a complete machine learning engineering lifecycle, beginning with data acquisition and quality assessment and progressing through exploratory data analysis, preprocessing, baseline modeling, model optimization, evaluation, error analysis, model stability assessment, serialization, versioning, input validation, safe prediction, and API deployment.

A leakage-safe preprocessing pipeline was implemented using separate numerical and categorical transformations. Numerical data is handled using imputation and scaling, while categorical features are processed using imputation and one-hot encoding. The resulting preprocessing workflow produces **130 transformed features**.

Multiple model evaluation and validation procedures were incorporated to ensure that the final model was not selected solely on overall accuracy. The project evaluates balanced performance, macro-level metrics, probability quality, generalization behavior, model stability, prediction confidence, and observed classification errors.

The final production model is an **Optimized Random Forest Classifier** integrated into a serialized production pipeline.

The final model achieved:

* **97.30% Test Accuracy**
* **96.94% Balanced Accuracy**
* **96.94% Macro F1 Score**
* **97.30% Weighted F1 Score**
* **99.77% Macro ROC-AUC**
* **98.44% Macro Average Precision**
* **98.59% Cross-Validation Macro F1**

The trained model and supporting artifacts were subsequently packaged and exposed through a **FastAPI production service** with prediction, probability, confidence, and health-check endpoints.

## 1.5 Problem Statement

### 1.5.1 Problem Definition

The objective of this project is to develop a machine learning system that predicts the most likely skin disorder class from structured clinical observations.

The system receives a set of clinical and patient-related features and produces a prediction corresponding to one of six predefined skin disorder classes.

Formally, the problem can be represented as:

```text
Input:
Structured clinical features

        ↓

Machine Learning Prediction System

        ↓

Output:
Predicted Skin Disorder Class
+
Class Probabilities
+
Prediction Confidence
```

### 1.5.2 Machine Learning Problem

This project is formulated as a **supervised multiclass classification problem**.

The target variable is:

```text
class
```

The target contains six classes:

```text
1, 2, 3, 4, 5, 6
```

The model learns the relationship between the available clinical predictors and the corresponding target class using labeled training data.

### 1.5.3 Input Definition

The production system expects exactly **34 predictive features**.

The input contains:

* **1 numerical feature:** `Age`
* **33 categorical clinical features**

The target variable `class` is excluded from prediction input.

### 1.5.4 Output Definition

The production system provides:

1. Predicted skin disorder class
2. Probability for each supported class
3. Maximum prediction probability
4. Confidence score
5. Confidence status
6. Model and version metadata through the production architecture

### 1.5.5 Key Technical Challenges

The project addresses several practical machine learning engineering challenges:

#### Mixed Data Types

The dataset contains both numerical and categorical predictors requiring different preprocessing strategies.

#### Missing Values

The `Age` feature contains missing-value markers represented using `"?"`, requiring controlled conversion and imputation.

#### Class Imbalance

The six target classes are not perfectly balanced, requiring evaluation using balanced and macro-level metrics.

#### Small Dataset Size

The dataset contains only **366 observations**, increasing the importance of careful validation and generalization analysis.

#### High-Dimensional Transformed Space

Categorical encoding expands the original 34 predictors into **130 transformed features**.

#### Prediction Reliability

The project evaluates probability distributions and confidence levels rather than relying only on predicted class labels.

#### Production Input Safety

The deployed system validates feature count, feature names, data types, missing values, invalid tokens, numeric ranges, and finite numeric values before prediction.

#### Model Reproducibility

The production model and supporting artifacts are versioned and fingerprinted using SHA-256 hashes.

#### Deployment Reliability

The final API includes automated endpoint testing, health monitoring, OpenAPI documentation, and deployment-readiness validation.

## 2.2 Logging Configuration

This section establishes the centralized logging framework for the project.

A structured logging system is used instead of direct console printing to provide consistent execution tracking, debugging support, validation reporting, and deployment traceability.

The logging configuration provides:

- Console-based execution monitoring
- Persistent project log storage
- Standardized log formatting
- Informational, warning, and error messages
- Centralized project logger management

The configured logger will be reused throughout the notebook and deployment workflow.

````python

#  LOGGING CONFIGURATION



# Logging Imports


import logging
from pathlib import Path


# Log Directory Configuration

LOG_DIRECTORY = Path("artifacts") / "logs"
LOG_DIRECTORY.mkdir(parents=True, exist_ok=True)

LOG_FILE = LOG_DIRECTORY / "skin_disorder_project.log"


# Project Logger

LOGGER_NAME = "skin_disorder_project"

logger = logging.getLogger(LOGGER_NAME)
logger.setLevel(logging.INFO)

# Prevent duplicate propagation to the root logger
logger.propagate = False


# Clear Existing Handlers


if logger.handlers:
    logger.handlers.clear()


# Logging Formatter


LOG_FORMAT = (
    "%(asctime)s - %(levelname)s - %(message)s"
)

formatter = logging.Formatter(LOG_FORMAT)


# Console Handler


console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)


# File Handler


file_handler = logging.FileHandler(
    LOG_FILE,
    mode="a",
    encoding="utf-8"
)

file_handler.setLevel(logging.INFO)
file_handler.setFormatter(formatter)


# Register Handlers

logger.addHandler(console_handler)
logger.addHandler(file_handler)


# Logging Configuration Status


logger.info("=" * 70)
logger.info("Logging Configuration")
logger.info("=" * 70)
logger.info("Logger Name                    : %s", LOGGER_NAME)
logger.info("Log Level                     : INFO")
logger.info("Log Directory                 : %s", LOG_DIRECTORY)
logger.info("Log File                      : %s", LOG_FILE)
logger.info("Console Handler               : AVAILABLE")
logger.info("File Handler                  : AVAILABLE")
logger.info("Logging Propagation           : DISABLED")
logger.info("=" * 70)
logger.info("Logging configuration initialized successfully.")
logger.info("=" * 70)
````

## 3.4 Data Types

This section evaluates the raw data types of all dataset columns before any type conversion or preprocessing is applied.

The raw dermatology dataset may contain numerical clinical and histopathological attributes represented as text values. Therefore, the initial pandas data types are treated as raw ingestion types rather than final analytical feature types.

The analysis will:

- Identify the current pandas data type of each column
- Determine the number of columns by raw data type
- Inspect representative values from the dataset
- Identify columns containing numeric-like values
- Identify columns containing non-numeric or special tokens
- Preserve the original dataset without transformation

Final data type conversion will be performed later during the data preprocessing stage after the dataset structure and invalid-value patterns have been validated.

````python
 

logger.info("=" * 70)
logger.info("Data Type Analysis")
logger.info("=" * 70)

logger.info("-" * 70)
logger.info("RAW DATA TYPE SUMMARY")

raw_dtype_counts = df.dtypes.value_counts()

for dtype, count in raw_dtype_counts.items():
    logger.info(
        "%-30s : %d columns",
        str(dtype),
        count
    )

logger.info("-" * 70)
logger.info("COLUMN DATA TYPE INSPECTION")

for column in df.columns:
    logger.info(
        "%-12s | Raw Type : %s",
        str(column),
        str(df[column].dtype)
    )


# Numeric-like feature detection


numeric_like_columns = []
non_numeric_columns = []

for column in df.columns:

    # Convert invalid token to NaN temporarily for diagnostic purposes.
    converted = pd.to_numeric(
        df[column].replace(MISSING_VALUE_TOKEN, np.nan),
        errors="coerce"
    )

    valid_numeric_values = converted.notna().sum()
    total_values = len(df[column])

    if valid_numeric_values == total_values:
        numeric_like_columns.append(column)
    else:
        non_numeric_columns.append(column)

logger.info("-" * 70)

logger.info(
    "Numeric-Like Columns                 : %d",
    len(numeric_like_columns)
)

logger.info(
    "Columns With Non-Numeric Values      : %d",
    len(non_numeric_columns)
)


# Invalid token analysis

question_mark_columns = []

total_question_mark_tokens = int(
    (df.astype(str) == MISSING_VALUE_TOKEN).sum().sum()
)

for column in df.columns:

    token_count = int(
        (df[column].astype(str) == MISSING_VALUE_TOKEN).sum()
    )

    if token_count > 0:
        question_mark_columns.append(column)

logger.info("-" * 70)

logger.info(
    "Columns Containing '?' Token         : %d",
    len(question_mark_columns)
)

logger.info(
    "Total '?' Tokens Detected            : %d",
    total_question_mark_tokens
)


# Diagnostic validation


if len(df.columns) == 35:
    logger.info(
        "Data Type Column Count Validation   : PASSED"
    )
else:
    logger.error(
        "Data Type Column Count Validation   : FAILED"
    )
    raise ValueError(
        f"Expected 35 columns, found {len(df.columns)}."
    )

if total_question_mark_tokens >= 0:
    logger.info(
        "Invalid Token Detection             : PASSED"
    )
else:
    logger.error(
        "Invalid Token Detection             : FAILED"
    )
    raise ValueError(
        "Invalid token analysis failed."
    )

logger.info(
    "Raw Data Type Analysis               : PASSED"
)

logger.info("=" * 70)
````

````python

#  TARGET IDENTIFICATION


logger.info("=" * 70)
logger.info("Target Identification")
logger.info("=" * 70)

logger.info("-" * 70)
logger.info("TARGET VARIABLE IDENTIFICATION")
logger.info("-" * 70)


# Target column

# The dermatology dataset contains 34 predictor attributes followed by
# the diagnostic class label.
TARGET_COLUMN = df.columns[-1]

logger.info(
    "Target Column                     : %s",
    TARGET_COLUMN
)

logger.info(
    "Target Column Position             : %d",
    df.columns.get_loc(TARGET_COLUMN)
)

logger.info(
    "Target Raw Data Type               : %s",
    df[TARGET_COLUMN].dtype
)

logger.info("-" * 70)


# Target missing-value validation


target_missing_token_count = int(
    (df[TARGET_COLUMN].astype(str).str.strip() == MISSING_VALUE_TOKEN).sum()
)

target_null_count = int(
    df[TARGET_COLUMN].isna().sum()
)

logger.info(
    "Target '?' Token Count             : %d",
    target_missing_token_count
)

logger.info(
    "Target Null Count                  : %d",
    target_null_count
)


# Target class analysis


target_class_counts = (
    df[TARGET_COLUMN]
    .astype(str)
    .str.strip()
    .value_counts(dropna=False)
    .sort_index()
)

target_class_count = int(
    target_class_counts.shape[0]
)

logger.info(
    "Number of Target Classes           : %d",
    target_class_count
)

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)

for class_label, class_count in target_class_counts.items():

    class_percentage = (
        class_count / len(df)
    ) * 100

    logger.info(
        "Class %-5s | Samples : %-4d | Percentage : %.2f%%",
        class_label,
        class_count,
        class_percentage
    )

logger.info("-" * 70)


# Target validation


if target_missing_token_count == 0 and target_null_count == 0:

    logger.info(
        "Target Missing-Value Validation    : PASSED"
    )

else:

    logger.error(
        "Target Missing-Value Validation    : FAILED"
    )

    raise ValueError(
        "Target variable contains missing or invalid target values."
    )

if target_class_count > 1:

    logger.info(
        "Target Class Count Validation      : PASSED"
    )

else:

    logger.error(
        "Target Class Count Validation      : FAILED"
    )

    raise ValueError(
        "Target variable contains fewer than two classes."
    )


# Target configuration


TARGET_CLASSES = sorted(
    target_class_counts.index.astype(str).tolist()
)

logger.info(
    "Target Classes                     : %s",
    TARGET_CLASSES
)

logger.info(
    "Target Identification Validation   : PASSED"
)

logger.info("=" * 70)
````

## 3.7 Feature Group Identification

This section identifies and validates the logical feature groups within the Skin Disorder Prediction dataset.

The predictor variables are organized into three clinically meaningful groups:

- **Clinical Features** — 12 features representing observable clinical characteristics.
- **Histopathological Features** — 21 features representing microscopic tissue characteristics.
- **Age** — 1 demographic feature representing patient age.

The expected predictor structure is therefore:

- 12 Clinical Features
- 21 Histopathological Features
- 1 Age Feature
- 34 Total Predictor Features

The target variable is the `class` column representing the diagnosed skin disorder category.

This feature grouping provides a structured representation of the dataset for subsequent preprocessing, exploratory analysis, feature engineering, model development, and explainability.

````python


logger.info("=" * 70)
logger.info("Feature Group Identification")
logger.info("=" * 70)


# Resolve Loaded Dataset


dataset_dataframe = None

# First, check the expected project-level dataset variable.
if "df" in globals() and isinstance(df, pd.DataFrame):
    dataset_dataframe = df

# Fallback: identify an existing DataFrame in the notebook namespace.
if dataset_dataframe is None:
    dataframe_candidates = {
        name: value
        for name, value in globals().items()
        if isinstance(value, pd.DataFrame)
    }

    if len(dataframe_candidates) == 1:
        dataset_dataframe = next(iter(dataframe_candidates.values()))

    elif "data" in dataframe_candidates:
        dataset_dataframe = dataframe_candidates["data"]

    elif "dataset" in dataframe_candidates:
        dataset_dataframe = dataframe_candidates["dataset"]

# Validate dataset resolution.
if dataset_dataframe is None:
    logger.error(
        "Dataset Object Resolution           : FAILED"
    )
    raise NameError(
        "No loaded pandas DataFrame was found. "
        "Please execute Section 3.1 Dataset Loading before Section 3.7."
    )

logger.info(
    "Dataset Object Resolution           : PASSED"
)

# Use one consistent local reference from this point onward.
df_feature_groups = dataset_dataframe



# Expected Feature Group Structure

CLINICAL_FEATURES = [
    "erythema",
    "scaling",
    "definite_borders",
    "itching",
    "koebner_phenomenon",
    "polygonal_papules",
    "follicular_papules",
    "oral_mucosal_involvement",
    "knee_and_elbow_involvement",
    "scalp_involvement",
    "family_history",
    "melanin_incontinence"
]

HISTOPATHOLOGICAL_FEATURES = [
    "eosinophils_in_the_infiltrate",
    "PNL_infiltrate",
    "fibrosis_of_the_papillary_dermis",
    "exocytosis",
    "acanthosis",
    "hyperkeratosis",
    "parakeratosis",
    "clubbing_of_the_rete_ridges",
    "elongation_of_the_rete_ridges",
    "thinning_of_the_suprapapillary_epidermis",
    "spongiform_pustule",
    "munro_microabcess",
    "focal_hypergranulosis",
    "disappearance_of_the_granular_layer",
    "vacuolisation_and_damage_of_basal_layer",
    "spongiosis",
    "saw-tooth_appearance_of_retes",
    "follicular_horn_plug",
    "perifollicular_parakeratosis",
    "inflammatory_monoluclear_inflitrate",
    "band-like_infiltrate"
]

AGE_FEATURE = "Age"
TARGET_FEATURE = "class"

EXPECTED_CLINICAL_COUNT = 12
EXPECTED_HISTOPATHOLOGICAL_COUNT = 21
EXPECTED_AGE_COUNT = 1
EXPECTED_PREDICTOR_COUNT = 34


logger.info("-" * 70)
logger.info("EXPECTED FEATURE GROUP STRUCTURE")
logger.info("-" * 70)

logger.info(
    "Clinical Features                  : %d",
    EXPECTED_CLINICAL_COUNT
)

logger.info(
    "Histopathological Features         : %d",
    EXPECTED_HISTOPATHOLOGICAL_COUNT
)

logger.info(
    "Age Feature                        : %d",
    EXPECTED_AGE_COUNT
)

logger.info(
    "Expected Predictor Features        : %d",
    EXPECTED_PREDICTOR_COUNT
)

logger.info(
    "Target Feature                     : %s",
    TARGET_FEATURE
)



# Feature Presence Validation


missing_clinical_features = [
    feature
    for feature in CLINICAL_FEATURES
    if feature not in df_feature_groups.columns
]

missing_histopathological_features = [
    feature
    for feature in HISTOPATHOLOGICAL_FEATURES
    if feature not in df_feature_groups.columns
]

age_feature_present = AGE_FEATURE in df_feature_groups.columns
target_feature_present = TARGET_FEATURE in df_feature_groups.columns



# Clinical Feature Validation


if missing_clinical_features:

    logger.error(
        "Clinical Feature Group Validation : FAILED"
    )

    logger.error(
        "Missing Clinical Features         : %s",
        missing_clinical_features
    )

    raise ValueError(
        "Clinical feature validation failed. "
        f"Missing features: {missing_clinical_features}"
    )

else:

    logger.info(
        "Clinical Feature Group Validation : PASSED"
    )



# Histopathological Feature Validation


if missing_histopathological_features:

    logger.error(
        "Histopathological Feature Group Validation : FAILED"
    )

    logger.error(
        "Missing Histopathological Features         : %s",
        missing_histopathological_features
    )

    raise ValueError(
        "Histopathological feature validation failed. "
        f"Missing features: {missing_histopathological_features}"
    )

else:

    logger.info(
        "Histopathological Feature Group Validation : PASSED"
    )



# Age Feature Validation

if not age_feature_present:

    logger.error(
        "Age Feature Validation             : FAILED"
    )

    raise ValueError(
        f"Required age feature '{AGE_FEATURE}' was not found."
    )

else:

    logger.info(
        "Age Feature Validation             : PASSED"
    )



# Target Feature Validation


if not target_feature_present:

    logger.error(
        "Target Feature Validation          : FAILED"
    )

    raise ValueError(
        f"Required target feature '{TARGET_FEATURE}' was not found."
    )

else:

    logger.info(
        "Target Feature Validation          : PASSED"
    )



# Construct Feature Groups

clinical_features = [
    feature
    for feature in CLINICAL_FEATURES
    if feature in df_feature_groups.columns
]

histopathological_features = [
    feature
    for feature in HISTOPATHOLOGICAL_FEATURES
    if feature in df_feature_groups.columns
]

age_feature = AGE_FEATURE

predictor_features = (
    clinical_features
    + histopathological_features
    + [age_feature]
)



# Feature Group Count Validation


clinical_count = len(clinical_features)
histopathological_count = len(histopathological_features)
age_count = 1
predictor_count = len(predictor_features)


logger.info("-" * 70)
logger.info("FEATURE GROUP SUMMARY")
logger.info("-" * 70)

logger.info(
    "Clinical Features                  : %d",
    clinical_count
)

logger.info(
    "Histopathological Features         : %d",
    histopathological_count
)

logger.info(
    "Age Feature                        : %s",
    age_feature
)

logger.info(
    "Total Predictor Features           : %d",
    predictor_count
)

logger.info(
    "Target Feature                     : %s",
    TARGET_FEATURE
)


# ----------------------------------------------------------------
# Structural Validation
# ----------------------------------------------------------------

if clinical_count != EXPECTED_CLINICAL_COUNT:

    logger.error(
        "Clinical Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected {EXPECTED_CLINICAL_COUNT} clinical features, "
        f"but found {clinical_count}."
    )

else:

    logger.info(
        "Clinical Feature Count Validation : PASSED"
    )


if histopathological_count != EXPECTED_HISTOPATHOLOGICAL_COUNT:

    logger.error(
        "Histopathological Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected {EXPECTED_HISTOPATHOLOGICAL_COUNT} "
        f"histopathological features, "
        f"but found {histopathological_count}."
    )

else:

    logger.info(
        "Histopathological Feature Count Validation : PASSED"
    )


if predictor_count != EXPECTED_PREDICTOR_COUNT:

    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected {EXPECTED_PREDICTOR_COUNT} predictor features, "
        f"but found {predictor_count}."
    )

else:

    logger.info(
        "Predictor Feature Count Validation : PASSED"
    )


# ----------------------------------------------------------------
# Feature Group Disjointness Validation
# ----------------------------------------------------------------

clinical_set = set(clinical_features)
histopathological_set = set(histopathological_features)

overlapping_features = (
    clinical_set.intersection(histopathological_set)
)

if overlapping_features:

    logger.error(
        "Feature Group Overlap Validation    : FAILED"
    )

    logger.error(
        "Overlapping Features                : %s",
        sorted(overlapping_features)
    )

    raise ValueError(
        "Clinical and histopathological feature groups overlap."
    )

else:

    logger.info(
        "Feature Group Overlap Validation    : PASSED"
    )


# ----------------------------------------------------------------
# Predictor Schema Validation
# ----------------------------------------------------------------

expected_predictor_set = (
    clinical_set
    | histopathological_set
    | {AGE_FEATURE}
)

actual_predictor_set = set(
    df_feature_groups.columns
).difference({TARGET_FEATURE})

if actual_predictor_set != expected_predictor_set:

    missing_predictors = sorted(
        expected_predictor_set - actual_predictor_set
    )

    unexpected_predictors = sorted(
        actual_predictor_set - expected_predictor_set
    )

    logger.error(
        "Predictor Schema Validation         : FAILED"
    )

    logger.error(
        "Missing Predictors                  : %s",
        missing_predictors
    )

    logger.error(
        "Unexpected Predictors              : %s",
        unexpected_predictors
    )

    raise ValueError(
        "Predictor schema does not match the expected "
        "clinical, histopathological, and age feature structure."
    )

else:

    logger.info(
        "Predictor Schema Validation         : PASSED"
    )


# ----------------------------------------------------------------
# Final Feature Group Output
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE GROUP STRUCTURE")
logger.info("-" * 70)

logger.info(
    "Clinical Feature Count              : %d",
    len(clinical_features)
)

logger.info(
    "Histopathological Feature Count     : %d",
    len(histopathological_features)
)

logger.info(
    "Age Feature Count                   : %d",
    age_count
)

logger.info(
    "Total Predictor Feature Count       : %d",
    len(predictor_features)
)

logger.info(
    "Target Feature                      : %s",
    TARGET_FEATURE
)

logger.info("-" * 70)
logger.info(
    "Feature Group Identification        : PASSED"
)
logger.info("=" * 70)
````

````python


logger.info("=" * 70)
logger.info("Initial Dataset Validation")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# Expected Dataset Structure
# ------------------------------------------------------------------------------

EXPECTED_RECORD_COUNT = 366
EXPECTED_PREDICTOR_COUNT = 34
EXPECTED_TOTAL_COLUMNS = 35

EXPECTED_CLINICAL_COUNT = 12
EXPECTED_HISTOPATHOLOGICAL_COUNT = 21

EXPECTED_AGE_FEATURE = "Age"
EXPECTED_TARGET_FEATURE = "class"

logger.info("-" * 70)
logger.info("EXPECTED DATASET STRUCTURE")
logger.info("-" * 70)
logger.info(
    "Expected Records                    : %d",
    EXPECTED_RECORD_COUNT
)
logger.info(
    "Expected Total Columns              : %d",
    EXPECTED_TOTAL_COLUMNS
)
logger.info(
    "Expected Predictor Features         : %d",
    EXPECTED_PREDICTOR_COUNT
)
logger.info(
    "Expected Clinical Features          : %d",
    EXPECTED_CLINICAL_COUNT
)
logger.info(
    "Expected Histopathological Features : %d",
    EXPECTED_HISTOPATHOLOGICAL_COUNT
)
logger.info(
    "Expected Age Feature                : %s",
    EXPECTED_AGE_FEATURE
)
logger.info(
    "Expected Target Feature             : %s",
    EXPECTED_TARGET_FEATURE
)


# Dataset Object Validation
# ------------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation          : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading first."
    )

logger.info("-" * 70)
logger.info("DATASET OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Dataset Object                     : AVAILABLE"
)

if not isinstance(df, pd.DataFrame):
    logger.error(
        "Dataset DataFrame Type Validation : FAILED"
    )
    raise TypeError(
        "Dataset object 'df' is not a pandas DataFrame."
    )

logger.info(
    "Dataset DataFrame Type Validation  : PASSED"
)

# ------------------------------------------------------------------------------
# Dataset Dimension Validation
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET DIMENSION VALIDATION")
logger.info("-" * 70)

actual_records = df.shape[0]
actual_columns = df.shape[1]
actual_predictors = actual_columns - 1

logger.info(
    "Actual Records                     : %d",
    actual_records
)
logger.info(
    "Actual Total Columns               : %d",
    actual_columns
)
logger.info(
    "Actual Predictor Features          : %d",
    actual_predictors
)

if actual_records != EXPECTED_RECORD_COUNT:
    logger.error(
        "Record Count Validation            : FAILED"
    )
    raise ValueError(
        f"Expected {EXPECTED_RECORD_COUNT} records, "
        f"but found {actual_records}."
    )

logger.info(
    "Record Count Validation            : PASSED"
)

if actual_columns != EXPECTED_TOTAL_COLUMNS:
    logger.error(
        "Column Count Validation            : FAILED"
    )
    raise ValueError(
        f"Expected {EXPECTED_TOTAL_COLUMNS} columns, "
        f"but found {actual_columns}."
    )

logger.info(
    "Column Count Validation            : PASSED"
)

if actual_predictors != EXPECTED_PREDICTOR_COUNT:
    logger.error(
        "Predictor Count Validation         : FAILED"
    )
    raise ValueError(
        f"Expected {EXPECTED_PREDICTOR_COUNT} predictor features, "
        f"but found {actual_predictors}."
    )

logger.info(
    "Predictor Count Validation         : PASSED"
)


# Target Feature Validation


logger.info("-" * 70)
logger.info("TARGET FEATURE VALIDATION")
logger.info("-" * 70)

if EXPECTED_TARGET_FEATURE not in df.columns:
    logger.error(
        "Target Feature Availability        : FAILED"
    )
    raise ValueError(
        f"Target feature '{EXPECTED_TARGET_FEATURE}' "
        "is missing from the dataset."
    )

logger.info(
    "Target Feature                    : %s",
    EXPECTED_TARGET_FEATURE
)

logger.info(
    "Target Feature Availability       : PASSED"
)

target_null_count = int(
    df[EXPECTED_TARGET_FEATURE].isna().sum()
)

target_missing_token_count = int(
    df[EXPECTED_TARGET_FEATURE]
    .astype(str)
    .str.strip()
    .eq("?")
    .sum()
)

logger.info(
    "Target Null Count                 : %d",
    target_null_count
)

logger.info(
    "Target '?' Token Count            : %d",
    target_missing_token_count
)

if target_null_count > 0:
    logger.error(
        "Target Null Validation             : FAILED"
    )
    raise ValueError(
        "Target variable contains null values."
    )

if target_missing_token_count > 0:
    logger.error(
        "Target Missing Token Validation    : FAILED"
    )
    raise ValueError(
        "Target variable contains '?' tokens."
    )

logger.info(
    "Target Missing-Value Validation    : PASSED"
)


# Feature Group Validation


logger.info("-" * 70)
logger.info("FEATURE GROUP VALIDATION")
logger.info("-" * 70)

# These lists are created in Section 3.7.
required_feature_groups = [
    "CLINICAL_FEATURES",
    "HISTOPATHOLOGICAL_FEATURES",
    "AGE_FEATURE"
]

missing_group_objects = [
    group_name
    for group_name in required_feature_groups
    if group_name not in globals()
]

if missing_group_objects:
    logger.error(
        "Feature Group Object Validation    : FAILED"
    )
    raise NameError(
        "Required feature-group objects are missing: "
        + ", ".join(missing_group_objects)
        + ". Run Section 3.7 Feature Group Identification first."
    )

logger.info(
    "Feature Group Objects              : AVAILABLE"
)


# Clinical Features


clinical_features = list(CLINICAL_FEATURES)

missing_clinical_features = [
    feature
    for feature in clinical_features
    if feature not in df.columns
]

if missing_clinical_features:
    logger.error(
        "Clinical Feature Group Validation  : FAILED"
    )
    raise ValueError(
        "Missing clinical features: "
        + ", ".join(missing_clinical_features)
    )

logger.info(
    "Clinical Features                  : %d",
    len(clinical_features)
)

logger.info(
    "Clinical Feature Validation        : PASSED"
)


# Histopathological Features


histopathological_features = list(
    HISTOPATHOLOGICAL_FEATURES
)

missing_histopathological_features = [
    feature
    for feature in histopathological_features
    if feature not in df.columns
]

if missing_histopathological_features:
    logger.error(
        "Histopathological Group Validation : FAILED"
    )
    raise ValueError(
        "Missing histopathological features: "
        + ", ".join(missing_histopathological_features)
    )

logger.info(
    "Histopathological Features         : %d",
    len(histopathological_features)
)

logger.info(
    "Histopathological Feature Validation : PASSED"
)


# Age Feature


age_feature = AGE_FEATURE

if age_feature not in df.columns:
    logger.error(
        "Age Feature Validation             : FAILED"
    )
    raise ValueError(
        f"Age feature '{age_feature}' is missing from the dataset."
    )

logger.info(
    "Age Feature                        : %s",
    age_feature
)

logger.info(
    "Age Feature Validation             : PASSED"
)


# Feature Group Count Validation


if len(clinical_features) != EXPECTED_CLINICAL_COUNT:
    logger.error(
        "Clinical Feature Count Validation : FAILED"
    )
    raise ValueError(
        f"Expected {EXPECTED_CLINICAL_COUNT} clinical features, "
        f"but found {len(clinical_features)}."
    )

if len(histopathological_features) != EXPECTED_HISTOPATHOLOGICAL_COUNT:
    logger.error(
        "Histopathological Count Validation : FAILED"
    )
    raise ValueError(
        f"Expected {EXPECTED_HISTOPATHOLOGICAL_COUNT} "
        f"histopathological features, "
        f"but found {len(histopathological_features)}."
    )

logger.info(
    "Clinical Feature Count Validation : PASSED"
)

logger.info(
    "Histopathological Count Validation : PASSED"
)


# Feature Group Overlap Validation


clinical_set = set(clinical_features)
histopathological_set = set(histopathological_features)
age_set = {age_feature}

group_overlap = (
    clinical_set & histopathological_set
) | (
    clinical_set & age_set
) | (
    histopathological_set & age_set
)

if group_overlap:
    logger.error(
        "Feature Group Overlap Validation   : FAILED"
    )
    raise ValueError(
        "Feature overlap detected: "
        + ", ".join(sorted(group_overlap))
    )

logger.info(
    "Feature Group Overlap Validation   : PASSED"
)


# Predictor Schema Validation


expected_predictor_features = (
    clinical_features
    + histopathological_features
    + [age_feature]
)

actual_predictor_features = [
    column
    for column in df.columns
    if column != EXPECTED_TARGET_FEATURE
]

if set(expected_predictor_features) != set(actual_predictor_features):
    missing_predictors = sorted(
        set(expected_predictor_features)
        - set(actual_predictor_features)
    )

    unexpected_predictors = sorted(
        set(actual_predictor_features)
        - set(expected_predictor_features)
    )

    logger.error(
        "Predictor Schema Validation         : FAILED"
    )

    raise ValueError(
        "Predictor schema mismatch. "
        f"Missing: {missing_predictors}; "
        f"Unexpected: {unexpected_predictors}"
    )

logger.info(
    "Predictor Schema Validation         : PASSED"
)


# Missing-Value Token Inspection


logger.info("-" * 70)
logger.info("RAW DATA MISSING-VALUE TOKEN INSPECTION")
logger.info("-" * 70)

question_mark_counts = (
    df.astype(str)
    .apply(
        lambda column: column.str.strip().eq("?").sum()
    )
)

total_question_marks = int(
    question_mark_counts.sum()
)

columns_with_question_marks = int(
    (question_mark_counts > 0).sum()
)

logger.info(
    "Columns Containing '?' Token        : %d",
    columns_with_question_marks
)

logger.info(
    "Total '?' Tokens                    : %d",
    total_question_marks
)

logger.info(
    "Missing-Value Token Detection       : PASSED"
)


# Target Class Validation


target_class_count = int(
    df[EXPECTED_TARGET_FEATURE].nunique()
)

logger.info("-" * 70)
logger.info("TARGET CLASS VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Class Count                  : %d",
    target_class_count
)

if target_class_count < 2:
    logger.error(
        "Target Class Validation             : FAILED"
    )
    raise ValueError(
        "Target variable must contain at least two classes."
    )

logger.info(
    "Target Class Validation             : PASSED"
)


# Final Dataset Readiness


logger.info("-" * 70)
logger.info("FINAL DATASET READINESS")
logger.info("-" * 70)

validation_checks = {
    "Dataset Object": True,
    "Record Count": actual_records == EXPECTED_RECORD_COUNT,
    "Column Count": actual_columns == EXPECTED_TOTAL_COLUMNS,
    "Predictor Count": actual_predictors == EXPECTED_PREDICTOR_COUNT,
    "Target Feature": EXPECTED_TARGET_FEATURE in df.columns,
    "Clinical Features": len(clinical_features) == EXPECTED_CLINICAL_COUNT,
    "Histopathological Features": (
        len(histopathological_features)
        == EXPECTED_HISTOPATHOLOGICAL_COUNT
    ),
    "Age Feature": age_feature in df.columns,
    "Feature Group Overlap": not bool(group_overlap),
    "Predictor Schema": (
        set(expected_predictor_features)
        == set(actual_predictor_features)
    ),
    "Target Classes": target_class_count >= 2
}

total_checks = len(validation_checks)
passed_checks = sum(validation_checks.values())
failed_checks = total_checks - passed_checks

logger.info(
    "Total Validation Checks            : %d",
    total_checks
)

logger.info(
    "Passed Validation Checks           : %d",
    passed_checks
)

logger.info(
    "Failed Validation Checks           : %d",
    failed_checks
)

if failed_checks != 0:
    logger.error(
        "Initial Dataset Validation         : FAILED"
    )
    raise ValueError(
        "One or more initial dataset validation checks failed."
    )

logger.info("-" * 70)
logger.info(
    "Dataset Readiness                  : READY"
)

logger.info(
    "Initial Dataset Validation         : PASSED"
)

logger.info("=" * 70)
````

## 4.**DATA QUALITY ASSESSMENT** ##

## 4.1 Missing Value Analysis

This section evaluates the dataset for explicit missing values represented as null or `NaN` entries.

Missing-value analysis is performed before preprocessing to establish the raw data quality baseline and prevent unexpected missing values from propagating into downstream modeling stages.

The analysis includes:

- Total missing-value count across the dataset
- Number of columns containing missing values
- Number of complete columns
- Column-level missing-value inspection
- Target-variable missing-value validation
- Dataset-level missing-value accounting

Literal missing-value tokens such as `?` are treated separately and will be analyzed in Section 4.2 Invalid Token Detection.

The raw dataset is not modified during this assessment.

````python


logger.info("=" * 70)
logger.info("Missing Value Analysis")
logger.info("=" * 70)


# Dataset Object Validation


if "df" not in globals():
    logger.error(
        "Dataset Object Validation          : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.1."
    )

logger.info("-" * 70)
logger.info("DATASET MISSING-VALUE INSPECTION")
logger.info("-" * 70)


# Null Value Analysis


null_counts = df.isnull().sum()

total_null_values = int(null_counts.sum())
columns_with_null_values = int((null_counts > 0).sum())

complete_columns = int((null_counts == 0).sum())

logger.info(
    "Total Missing Values              : %d",
    total_null_values
)

logger.info(
    "Columns With Missing Values       : %d",
    columns_with_null_values
)

logger.info(
    "Columns With Complete Values      : %d",
    complete_columns
)


# Column-Level Missing Value Inspection

if columns_with_null_values > 0:

    logger.info("-" * 70)
    logger.info("COLUMNS CONTAINING NULL VALUES")
    logger.info("-" * 70)

    for column in df.columns:

        missing_count = int(null_counts[column])

        if missing_count > 0:

            missing_percentage = (
                missing_count / len(df)
            ) * 100

            logger.info(
                "%-35s | Missing : %-5d | Percentage : %.2f%%",
                str(column),
                missing_count,
                missing_percentage
            )

else:

    logger.info(
        "Null Value Status                  : NO NULL VALUES DETECTED"
    )


# Target Variable Null Validation

if TARGET_FEATURE not in df.columns:

    logger.error(
        "Target Feature Availability       : FAILED"
    )

    raise ValueError(
        f"Target feature '{TARGET_FEATURE}' is not present in dataset."
    )

target_null_count = int(
    df[TARGET_FEATURE].isnull().sum()
)

logger.info("-" * 70)
logger.info("TARGET NULL-VALUE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Feature                    : %s",
    TARGET_FEATURE
)

logger.info(
    "Target Null Count                 : %d",
    target_null_count
)

if target_null_count == 0:

    logger.info(
        "Target Missing-Value Validation  : PASSED"
    )

else:

    logger.warning(
        "Target Missing-Value Validation  : WARNING"
    )


# Dataset-Level Validation

logger.info("-" * 70)
logger.info("MISSING-VALUE VALIDATION")
logger.info("-" * 70)

expected_total_columns = len(df.columns)

if complete_columns + columns_with_null_values == expected_total_columns:

    logger.info(
        "Column Missing-Value Accounting   : PASSED"
    )

else:

    logger.error(
        "Column Missing-Value Accounting   : FAILED"
    )

    raise ValueError(
        "Missing-value column accounting validation failed."
    )

logger.info(
    "Missing Value Analysis             : PASSED"
)

logger.info("=" * 70)
````

## 4.2 Invalid Token Detection

This section identifies invalid or non-standard data entries that may not be represented as conventional null values.

The raw dataset is inspected for:

- Literal `?` tokens
- Blank or empty-string entries
- Whitespace-only entries
- Non-numeric values within numeric clinical and histopathological features
- Invalid entries within the Age feature
- Invalid entries within the target variable

The Dermatology dataset contains categorical and ordinal clinical observations stored as raw values, with `?` used to represent missing feature information.

These invalid tokens are identified and quantified at the raw-data stage. No imputation, encoding, or data transformation is performed in this section.

The results from this analysis will be used to design the preprocessing strategy in the subsequent data preparation stage.

````python


logger.info("=" * 70)
logger.info("Invalid Token Detection")
logger.info("=" * 70)


# Dataset Object Validation

if "df" not in globals():
    logger.error(
        "Dataset Object Validation          : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.2."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

logger.info("-" * 70)
logger.info("INVALID TOKEN INSPECTION")
logger.info("-" * 70)


# Define Invalid Tokens


MISSING_TOKEN = "?"
BLANK_TOKEN = ""
INVALID_TOKENS = {
    MISSING_TOKEN,
    BLANK_TOKEN
}

logger.info(
    "Primary Missing Token               : %s",
    MISSING_TOKEN
)


# '?' Token Analysis


question_mark_counts = (
    df.astype(str)
    .apply(lambda column: column.str.strip().eq(MISSING_TOKEN).sum())
)

question_mark_total = int(
    question_mark_counts.sum()
)

columns_with_question_mark = int(
    (question_mark_counts > 0).sum()
)

logger.info("-" * 70)
logger.info("QUESTION-MARK TOKEN ANALYSIS")
logger.info("-" * 70)

logger.info(
    "Columns Containing '?' Token        : %d",
    columns_with_question_mark
)

logger.info(
    "Total '?' Tokens Detected           : %d",
    question_mark_total
)

if columns_with_question_mark > 0:

    for column in df.columns:

        token_count = int(
            question_mark_counts[column]
        )

        if token_count > 0:

            logger.info(
                "%-35s | '?' Tokens : %d",
                str(column),
                token_count
            )

    logger.info(
        "Question-Mark Token Detection      : PASSED"
    )

else:

    logger.info(
        "Question-Mark Token Detection      : PASSED"
    )


# Blank / Empty String Analysis

blank_counts = (
    df.astype(str)
    .apply(lambda column: column.str.strip().eq(BLANK_TOKEN).sum())
)

blank_total = int(
    blank_counts.sum()
)

columns_with_blank_values = int(
    (blank_counts > 0).sum()
)

logger.info("-" * 70)
logger.info("BLANK VALUE ANALYSIS")
logger.info("-" * 70)

logger.info(
    "Columns Containing Blank Values     : %d",
    columns_with_blank_values
)

logger.info(
    "Total Blank Values Detected         : %d",
    blank_total
)

if columns_with_blank_values > 0:

    for column in df.columns:

        blank_count = int(
            blank_counts[column]
        )

        if blank_count > 0:

            logger.info(
                "%-35s | Blank Values : %d",
                str(column),
                blank_count
            )

    logger.warning(
        "Blank Value Detection              : VALUES DETECTED"
    )

else:

    logger.info(
        "Blank Value Detection              : PASSED"
    )


# Whitespace-Only Value Analysis

whitespace_counts = (
    df.astype(str)
    .apply(
        lambda column: (
            column.ne(column.str.strip())
            & column.str.strip().ne("")
        ).sum()
    )
)

whitespace_total = int(
    whitespace_counts.sum()
)

logger.info("-" * 70)
logger.info("WHITESPACE VALUE ANALYSIS")
logger.info("-" * 70)

logger.info(
    "Whitespace-Affected Values         : %d",
    whitespace_total
)

if whitespace_total == 0:

    logger.info(
        "Whitespace Validation              : PASSED"
    )

else:

    logger.warning(
        "Whitespace Validation              : VALUES DETECTED"
    )


# Target Variable Invalid Token Analysis


target_question_marks = int(
    df[TARGET_FEATURE]
    .astype(str)
    .str.strip()
    .eq(MISSING_TOKEN)
    .sum()
)

target_blank_values = int(
    df[TARGET_FEATURE]
    .astype(str)
    .str.strip()
    .eq(BLANK_TOKEN)
    .sum()
)

logger.info("-" * 70)
logger.info("TARGET INVALID TOKEN VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Feature                    : %s",
    TARGET_FEATURE
)

logger.info(
    "Target '?' Token Count             : %d",
    target_question_marks
)

logger.info(
    "Target Blank Value Count           : %d",
    target_blank_values
)

if target_question_marks == 0 and target_blank_values == 0:

    logger.info(
        "Target Invalid Token Validation   : PASSED"
    )

else:

    logger.error(
        "Target Invalid Token Validation   : FAILED"
    )

    raise ValueError(
        "Invalid tokens detected in target variable."
    )


# Feature Invalid Token Summary


feature_columns = [
    column
    for column in df.columns
    if column != TARGET_FEATURE
]

feature_question_marks = int(
    question_mark_counts[feature_columns].sum()
)

feature_blank_values = int(
    blank_counts[feature_columns].sum()
)

logger.info("-" * 70)
logger.info("FEATURE INVALID TOKEN SUMMARY")
logger.info("-" * 70)

logger.info(
    "Predictor Features                 : %d",
    len(feature_columns)
)

logger.info(
    "Feature '?' Tokens                 : %d",
    feature_question_marks
)

logger.info(
    "Feature Blank Values               : %d",
    feature_blank_values
)


# Final Validation


logger.info("-" * 70)
logger.info("INVALID TOKEN VALIDATION")
logger.info("-" * 70)

if question_mark_total == feature_question_marks + target_question_marks:

    logger.info(
        "Question-Mark Accounting          : PASSED"
    )

else:

    logger.error(
        "Question-Mark Accounting          : FAILED"
    )

    raise ValueError(
        "Question-mark token accounting validation failed."
    )

if blank_total == feature_blank_values + target_blank_values:

    logger.info(
        "Blank Value Accounting             : PASSED"
    )

else:

    logger.error(
        "Blank Value Accounting             : FAILED"
    )

    raise ValueError(
        "Blank-value accounting validation failed."
    )

logger.info(
    "Invalid Token Detection            : PASSED"
)

logger.info("=" * 70)
````

## 4.3 Duplicate Record Analysis

This section evaluates the raw dataset for completely duplicated records.

Duplicate records are identified by comparing all feature and target columns across the complete dataset. The analysis determines:

- Total number of records
- Number of duplicated records
- Number of unique records
- Duplicate-record percentage
- Whether duplicated records exist in the dataset

Duplicate records are assessed before preprocessing to establish the integrity of the original dataset and prevent repeated observations from unintentionally influencing downstream model training and evaluation.

No records are removed during this assessment. Any duplicate records detected will be handled explicitly during the data-cleaning stage.

````python


logger.info("=" * 70)
logger.info("Duplicate Record Analysis")
logger.info("=" * 70)


# Dataset Object Validation


if "df" not in globals():
    logger.error(
        "Dataset Object Validation          : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.3."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

logger.info("-" * 70)
logger.info("DUPLICATE RECORD INSPECTION")
logger.info("-" * 70)


# Basic Dataset Information


total_records = int(len(df))

logger.info(
    "Total Records                       : %d",
    total_records
)


# Duplicate Record Detection


duplicate_mask = df.duplicated(
    keep=False
)

duplicate_record_count = int(
    duplicate_mask.sum()
)

duplicate_group_count = int(
    df.duplicated(
        keep="first"
    ).sum()
)

unique_record_count = int(
    total_records - duplicate_group_count
)


# Duplicate Percentage

if total_records > 0:

    duplicate_percentage = (
        duplicate_record_count / total_records
    ) * 100

else:

    duplicate_percentage = 0.0

logger.info(
    "Duplicate Records                  : %d",
    duplicate_record_count
)

logger.info(
    "Duplicate Record Occurrences      : %d",
    duplicate_group_count
)

logger.info(
    "Unique Records                     : %d",
    unique_record_count
)

logger.info(
    "Duplicate Record Percentage        : %.2f%%",
    duplicate_percentage
)


# Duplicate Record Validation


logger.info("-" * 70)
logger.info("DUPLICATE RECORD VALIDATION")
logger.info("-" * 70)

if duplicate_record_count == 0:

    logger.info(
        "Duplicate Records Status           : NONE DETECTED"
    )

    logger.info(
        "Duplicate Record Validation        : PASSED"
    )

else:

    logger.warning(
        "Duplicate Records Status           : %d DETECTED",
        duplicate_record_count
    )

    logger.warning(
        "Duplicate Record Validation        : REVIEW REQUIRED"
    )


# Record Accounting Validation


if unique_record_count + duplicate_group_count == total_records:

    logger.info(
        "Record Accounting Validation       : PASSED"
    )

else:

    logger.error(
        "Record Accounting Validation       : FAILED"
    )

    raise ValueError(
        "Duplicate record accounting validation failed."
    )


# Final Validation


if duplicate_record_count >= duplicate_group_count:

    logger.info(
        "Duplicate Count Consistency        : PASSED"
    )

else:

    logger.error(
        "Duplicate Count Consistency        : FAILED"
    )

    raise ValueError(
        "Duplicate count consistency validation failed."
    )

logger.info("-" * 70)

logger.info(
    "Duplicate Record Analysis           : PASSED"
)

logger.info("=" * 70)
````

## 4.4 Duplicate Feature Vector Analysis

This section evaluates the dataset for duplicate predictor feature vectors independently of the target variable.

Unlike complete duplicate record analysis, which compares all dataset columns including the target variable, duplicate feature vector analysis compares only the predictor features.

The analysis determines:

- Total predictor feature vectors
- Number of unique feature vectors
- Number of duplicated feature vectors
- Number of duplicate feature-vector occurrences
- Duplicate feature-vector percentage
- Whether identical predictor patterns are associated with different target classes

This analysis helps identify repeated predictor patterns and potential target conflicts that may affect model learning and evaluation.

No records are removed or modified during this assessment.

````python


logger.info("=" * 70)
logger.info("Duplicate Feature Vector Analysis")
logger.info("=" * 70)


# Dataset Object Validation


if "df" not in globals():

    logger.error(
        "Dataset Object Validation           : FAILED"
    )

    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.4."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)


# Target Feature Resolution

TARGET_FEATURE = "class"

if TARGET_FEATURE not in df.columns:

    logger.error(
        "Target Feature Resolution           : FAILED"
    )

    raise ValueError(
        f"Target feature '{TARGET_FEATURE}' is not present in the dataset."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)


# Predictor Feature Construction

# Do NOT depend on PREDICTOR_FEATURES from Section 3.7.
# Predictor features are derived directly from the loaded DataFrame.

PREDICTOR_FEATURES_44 = [
    column
    for column in df.columns
    if column != TARGET_FEATURE
]


# Predictor Feature Validation


logger.info("-" * 70)
logger.info("PREDICTOR FEATURE VALIDATION")
logger.info("-" * 70)

total_predictor_features = len(
    PREDICTOR_FEATURES_44
)

logger.info(
    "Total Predictor Features            : %d",
    total_predictor_features
)

logger.info(
    "Expected Predictor Features         : 34"
)

if total_predictor_features != 34:

    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected 34 predictor features, "
        f"but found {total_predictor_features}."
    )

logger.info(
    "Predictor Feature Count Validation : PASSED"
)


# Target Exclusion Validation

if TARGET_FEATURE in PREDICTOR_FEATURES_44:

    logger.error(
        "Target Exclusion Validation        : FAILED"
    )

    raise ValueError(
        "Target feature was incorrectly included in predictor features."
    )

logger.info(
    "Target Exclusion Validation        : PASSED"
)


# Predictor Matrix Construction


X_raw_44 = df[
    PREDICTOR_FEATURES_44
].copy()

logger.info("-" * 70)
logger.info("FEATURE VECTOR INSPECTION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    len(X_raw_44)
)

logger.info(
    "Total Predictor Features            : %d",
    X_raw_44.shape[1]
)

logger.info(
    "Feature Vector Shape                : %s",
    X_raw_44.shape
)


# Feature Vector Column Validation


if X_raw_44.shape[1] != 34:

    logger.error(
        "Feature Vector Dimension Validation: FAILED"
    )

    raise ValueError(
        "Feature vector does not contain exactly 34 predictor features."
    )

logger.info(
    "Feature Vector Dimension Validation: PASSED"
)


# Duplicate Feature Vector Detection


duplicate_vector_mask = X_raw_44.duplicated(
    keep=False
)

duplicate_vector_occurrence_mask = X_raw_44.duplicated(
    keep="first"
)

duplicate_vector_record_count = int(
    duplicate_vector_mask.sum()
)

duplicate_vector_occurrence_count = int(
    duplicate_vector_occurrence_mask.sum()
)

unique_feature_vector_count = int(
    len(X_raw_44)
    - duplicate_vector_occurrence_count
)


# Duplicate Feature Vector Percentage

if len(X_raw_44) > 0:

    duplicate_vector_percentage = (
        duplicate_vector_record_count
        / len(X_raw_44)
    ) * 100

else:

    duplicate_vector_percentage = 0.0


# Results


logger.info("-" * 70)
logger.info("DUPLICATE FEATURE VECTOR RESULTS")
logger.info("-" * 70)

logger.info(
    "Duplicate Feature Vector Records  : %d",
    duplicate_vector_record_count
)

logger.info(
    "Duplicate Vector Occurrences       : %d",
    duplicate_vector_occurrence_count
)

logger.info(
    "Unique Feature Vectors             : %d",
    unique_feature_vector_count
)

logger.info(
    "Duplicate Feature Vector Percentage: %.2f%%",
    duplicate_vector_percentage
)


# Target Consistency Analysis


logger.info("-" * 70)
logger.info("TARGET CONSISTENCY ANALYSIS")
logger.info("-" * 70)

duplicate_vector_groups = 0
target_conflict_count = 0

if duplicate_vector_record_count > 0:

    duplicate_data = X_raw_44[
        duplicate_vector_mask
    ].copy()

    duplicate_data["_TARGET_CHECK_"] = df.loc[
        duplicate_vector_mask,
        TARGET_FEATURE
    ].values

    grouped_targets = (
        duplicate_data
        .groupby(
            PREDICTOR_FEATURES_44,
            dropna=False,
            sort=False
        )["_TARGET_CHECK_"]
        .nunique()
    )

    duplicate_vector_groups = int(
        len(grouped_targets)
    )

    target_conflict_count = int(
        (grouped_targets > 1).sum()
    )

logger.info(
    "Duplicate Feature Vector Groups   : %d",
    duplicate_vector_groups
)

logger.info(
    "Vectors With Conflicting Targets   : %d",
    target_conflict_count
)


# Duplicate Feature Vector Status


logger.info("-" * 70)
logger.info("DUPLICATE FEATURE VECTOR VALIDATION")
logger.info("-" * 70)

if duplicate_vector_record_count == 0:

    logger.info(
        "Duplicate Feature Vectors Status  : NONE DETECTED"
    )

    logger.info(
        "Target Conflict Status             : NOT APPLICABLE"
    )

else:

    logger.warning(
        "Duplicate Feature Vectors Status  : %d DETECTED",
        duplicate_vector_record_count
    )

    if target_conflict_count == 0:

        logger.info(
            "Target Conflict Status             : NONE DETECTED"
        )

    else:

        logger.warning(
            "Target Conflict Status             : %d GROUPS DETECTED",
            target_conflict_count
        )


# Feature Vector Accounting Validation


if (
    unique_feature_vector_count
    + duplicate_vector_occurrence_count
    == len(X_raw_44)
):

    logger.info(
        "Feature Vector Accounting          : PASSED"
    )

else:

    logger.error(
        "Feature Vector Accounting          : FAILED"
    )

    raise ValueError(
        "Feature vector accounting validation failed."
    )


# Predictor Schema Validation


if (
    len(PREDICTOR_FEATURES_44) == 34
    and TARGET_FEATURE not in PREDICTOR_FEATURES_44
):

    logger.info(
        "Predictor Schema Validation        : PASSED"
    )

else:

    logger.error(
        "Predictor Schema Validation        : FAILED"
    )

    raise ValueError(
        "Predictor schema validation failed."
    )


# Final Validation


logger.info("-" * 70)
logger.info(
    "Duplicate Feature Vector Analysis   : PASSED"
)
logger.info("=" * 70)
````

## 4.5 Constant Feature Detection

Constant feature detection identifies predictor variables that contain only one unique value across all observations.

A constant feature does not provide discriminatory information to a machine learning model because its value remains unchanged across the dataset. Such features are therefore identified as potential candidates for removal during later preprocessing.

This section evaluates all predictor features and determines:

* Total number of predictor features
* Number of constant features
* Number of non-constant features
* Unique-value count for each predictor feature
* Constant feature status
* Feature accounting consistency

A feature is classified as constant when it contains one or fewer unique values, including missing-value tokens where applicable.

No feature is removed during this section. The purpose is to assess dataset quality and document potentially non-informative features for later preprocessing.


````python


import logging
import pandas as pd
from IPython.display import display

logger = logging.getLogger(__name__)

# =============================================================================
# SECTION HEADER
# =============================================================================

logger.info("=" * 70)
logger.info("Constant Feature Detection")
logger.info("=" * 70)

# =============================================================================
# DATASET OBJECT VALIDATION
# =============================================================================

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.5."
    )

if not isinstance(df, pd.DataFrame):
    logger.error(
        "Dataset DataFrame Type Validation  : FAILED"
    )
    raise TypeError(
        "Object 'df' exists but is not a pandas DataFrame."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

logger.info(
    "Dataset DataFrame Type Validation   : PASSED"
)

# =============================================================================
# TARGET FEATURE RESOLUTION
# =============================================================================

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE

elif "class" in df.columns:
    target_feature = "class"

else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "TARGET_FEATURE is not available and target column "
        "'class' was not found in the dataset."
    )

if target_feature not in df.columns:
    logger.error(
        "Target Feature Availability         : FAILED"
    )
    raise ValueError(
        f"Target feature '{target_feature}' is not present in df."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# =============================================================================
# PREDICTOR FEATURE RESOLUTION
# =============================================================================

if "PREDICTOR_FEATURES" in globals():

    predictor_features = list(PREDICTOR_FEATURES)

else:

    predictor_features = [
        column
        for column in df.columns
        if column != target_feature
    ]

logger.info("-" * 70)
logger.info("PREDICTOR FEATURE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Expected Predictor Features         : 34"
)

# =============================================================================
# PREDICTOR COUNT VALIDATION
# =============================================================================

if len(predictor_features) != 34:

    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected 34 predictor features, "
        f"but found {len(predictor_features)}."
    )

logger.info(
    "Predictor Feature Count Validation : PASSED"
)

# =============================================================================
# TARGET EXCLUSION VALIDATION
# =============================================================================

if target_feature in predictor_features:

    logger.error(
        "Target Exclusion Validation        : FAILED"
    )

    raise ValueError(
        f"Target feature '{target_feature}' is incorrectly "
        "included in predictor features."
    )

logger.info(
    "Target Exclusion Validation        : PASSED"
)

# =============================================================================
# PREDICTOR COLUMN AVAILABILITY
# =============================================================================

missing_predictors = [
    feature
    for feature in predictor_features
    if feature not in df.columns
]

if missing_predictors:

    logger.error(
        "Predictor Column Availability       : FAILED"
    )

    raise ValueError(
        "Missing predictor columns: "
        + ", ".join(missing_predictors)
    )

logger.info(
    "Predictor Column Availability       : PASSED"
)

# =============================================================================
# CONSTANT FEATURE INSPECTION
# =============================================================================

logger.info("-" * 70)
logger.info("CONSTANT FEATURE INSPECTION")
logger.info("-" * 70)

constant_features = []
non_constant_features = []

feature_unique_counts = {}

for feature in predictor_features:

    unique_count = df[feature].nunique(
        dropna=False
    )

    feature_unique_counts[feature] = unique_count

    if unique_count <= 1:

        constant_features.append(feature)

    else:

        non_constant_features.append(feature)

# =============================================================================
# CONSTANT FEATURE RESULTS
# =============================================================================

logger.info("-" * 70)
logger.info("CONSTANT FEATURE RESULTS")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Constant Features                   : %d",
    len(constant_features)
)

logger.info(
    "Non-Constant Features               : %d",
    len(non_constant_features)
)

# =============================================================================
# CONSTANT FEATURE DETAILS
# =============================================================================

if constant_features:

    logger.info("-" * 70)
    logger.info("CONSTANT FEATURE DETAILS")
    logger.info("-" * 70)

    for feature in constant_features:

        logger.info(
            "Feature : %-35s | Unique Values : %d",
            feature,
            feature_unique_counts[feature]
        )

else:

    logger.info(
        "Constant Feature Status            : NONE DETECTED"
    )

# =============================================================================
# FEATURE ACCOUNTING VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE ACCOUNTING VALIDATION")
logger.info("-" * 70)

total_accounted_features = (
    len(constant_features)
    + len(non_constant_features)
)

logger.info(
    "Constant Features                   : %d",
    len(constant_features)
)

logger.info(
    "Non-Constant Features               : %d",
    len(non_constant_features)
)

logger.info(
    "Total Accounted Features            : %d",
    total_accounted_features
)

logger.info(
    "Expected Predictor Features         : %d",
    len(predictor_features)
)

if total_accounted_features != len(predictor_features):

    logger.error(
        "Feature Accounting Validation      : FAILED"
    )

    raise ValueError(
        "Constant/non-constant feature accounting "
        "does not match predictor feature count."
    )

logger.info(
    "Feature Accounting Validation      : PASSED"
)

# =============================================================================
# CREATE FEATURE ANALYSIS TABLE
# =============================================================================

constant_feature_summary = pd.DataFrame(
    {
        "Feature": predictor_features,
        "Unique Values": [
            feature_unique_counts[feature]
            for feature in predictor_features
        ],
        "Constant": [
            feature in constant_features
            for feature in predictor_features
        ]
    }
)

# =============================================================================
# CONSTANT FEATURE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("CONSTANT FEATURE VALIDATION")
logger.info("-" * 70)

if len(constant_features) == 0:

    constant_feature_status = "NONE DETECTED"

    logger.info(
        "Constant Feature Status            : NONE DETECTED"
    )

else:

    constant_feature_status = (
        f"{len(constant_features)} DETECTED"
    )

    logger.info(
        "Constant Feature Status            : %d DETECTED",
        len(constant_features)
    )

logger.info(
    "Constant Feature Detection          : PASSED"
)

# =============================================================================
# VISIBLE NOTEBOOK OUTPUT
# =============================================================================

print()
print("=" * 70)
print("4.5 CONSTANT FEATURE DETECTION — SUMMARY")
print("=" * 70)

print(f"Total Predictor Features : {len(predictor_features)}")
print(f"Constant Features        : {len(constant_features)}")
print(f"Non-Constant Features    : {len(non_constant_features)}")

print("-" * 70)

if constant_features:

    print("Constant Features:")
    for feature in constant_features:
        print(
            f"  - {feature} "
            f"(Unique Values: {feature_unique_counts[feature]})"
        )

else:

    print("Constant Features        : NONE DETECTED")

print("-" * 70)
print("Feature Accounting       : PASSED")
print("Constant Feature Status  :", constant_feature_status)
print("=" * 70)

# =============================================================================
# DISPLAY FEATURE SUMMARY TABLE
# =============================================================================

display(
    constant_feature_summary
)

# =============================================================================
# FINAL SECTION STATUS
# =============================================================================

logger.info("-" * 70)
logger.info(
    "Constant Feature Detection          : PASSED"
)
logger.info("=" * 70)
````

## 4.6 Unique Value Analysis

This section analyzes the number of distinct values present in each predictor feature.

The objective is to understand feature cardinality and identify:

- Binary features with exactly two unique values.
- Low-cardinality categorical features.
- Higher-cardinality features such as Age.
- Single-value features that may require removal.
- Fully unique predictor features that may represent identifiers.

The target variable `class` is excluded from predictor-level analysis.

Unique-value analysis is performed for data-quality assessment only. No feature is removed solely on the basis of cardinality at this stage.

````python


import logging
import pandas as pd
from IPython.display import display

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Unique Value Analysis")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.6."
    )

if not isinstance(df, pd.DataFrame):
    logger.error(
        "Dataset DataFrame Type Validation  : FAILED"
    )
    raise TypeError(
        "Dataset object 'df' is not a pandas DataFrame."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE
elif "class" in df.columns:
    target_feature = "class"
else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "Target feature could not be resolved."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Feature Resolution
# -----------------------------------------------------------------------------

if "PREDICTOR_FEATURES" in globals():
    predictor_features = list(PREDICTOR_FEATURES)
else:
    predictor_features = [
        column
        for column in df.columns
        if column != target_feature
    ]

logger.info("-" * 70)
logger.info("PREDICTOR FEATURE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Expected Predictor Features         : 34"
)

if len(predictor_features) != 34:
    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )
    raise ValueError(
        f"Expected 34 predictor features, "
        f"but found {len(predictor_features)}."
    )

logger.info(
    "Predictor Feature Count Validation : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Column Availability
# -----------------------------------------------------------------------------

missing_predictors = [
    feature
    for feature in predictor_features
    if feature not in df.columns
]

if missing_predictors:
    logger.error(
        "Predictor Column Availability       : FAILED"
    )
    raise ValueError(
        "Missing predictor columns: "
        + ", ".join(missing_predictors)
    )

logger.info(
    "Predictor Column Availability       : PASSED"
)

# -----------------------------------------------------------------------------
# Target Exclusion Validation
# -----------------------------------------------------------------------------

if target_feature in predictor_features:
    logger.error(
        "Target Exclusion Validation        : FAILED"
    )
    raise ValueError(
        "Target feature is incorrectly included "
        "in predictor features."
    )

logger.info(
    "Target Exclusion Validation        : PASSED"
)

# -----------------------------------------------------------------------------
# Unique Value Analysis
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("UNIQUE VALUE INSPECTION")
logger.info("-" * 70)

unique_value_records = []

for feature in predictor_features:

    unique_count = int(
        df[feature].nunique(dropna=False)
    )

    unique_percentage = (
        unique_count / len(df)
    ) * 100

    unique_value_records.append(
        {
            "Feature": feature,
            "Unique Values": unique_count,
            "Unique %": round(unique_percentage, 2)
        }
    )

# -----------------------------------------------------------------------------
# Create Analysis DataFrame
# -----------------------------------------------------------------------------

unique_value_df = pd.DataFrame(
    unique_value_records
)

logger.info(
    "Features Analyzed                   : %d",
    len(unique_value_df)
)

# -----------------------------------------------------------------------------
# Feature Cardinality Classification
# -----------------------------------------------------------------------------

single_value_features = []
binary_features = []
low_cardinality_features = []
high_cardinality_features = []
fully_unique_features = []

for record in unique_value_records:

    feature = record["Feature"]
    unique_count = record["Unique Values"]

    if unique_count == 1:

        single_value_features.append(feature)

    elif unique_count == 2:

        binary_features.append(feature)

    elif 3 <= unique_count <= 10:

        low_cardinality_features.append(feature)

    elif unique_count == len(df):

        fully_unique_features.append(feature)
        high_cardinality_features.append(feature)

    else:

        high_cardinality_features.append(feature)

# -----------------------------------------------------------------------------
# Cardinality Summary
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("UNIQUE VALUE SUMMARY")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Single-Value Features               : %d",
    len(single_value_features)
)

logger.info(
    "Binary Features                     : %d",
    len(binary_features)
)

logger.info(
    "Low-Cardinality Features (3-10)    : %d",
    len(low_cardinality_features)
)

logger.info(
    "High-Cardinality Features (>10)     : %d",
    len(high_cardinality_features)
)

logger.info(
    "Fully Unique Features               : %d",
    len(fully_unique_features)
)

# -----------------------------------------------------------------------------
# Display Feature-Level Unique Value Table
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE UNIQUE VALUE DETAILS")
logger.info("-" * 70)

display(
    unique_value_df
)

# -----------------------------------------------------------------------------
# Single-Value Feature Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SINGLE-VALUE FEATURE INSPECTION")
logger.info("-" * 70)

if single_value_features:

    logger.info(
        "Single-Value Features Detected     : %d",
        len(single_value_features)
    )

    for feature in single_value_features:

        logger.info(
            "Feature : %s | Unique Values : 1",
            feature
        )

else:

    logger.info(
        "Single-Value Feature Status         : NONE DETECTED"
    )

# -----------------------------------------------------------------------------
# Binary Feature Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("BINARY FEATURE INSPECTION")
logger.info("-" * 70)

if binary_features:

    logger.info(
        "Binary Features Detected            : %d",
        len(binary_features)
    )

    for feature in binary_features:

        logger.info(
            "Feature : %-40s | Unique Values : 2",
            feature
        )

else:

    logger.info(
        "Binary Feature Status               : NONE DETECTED"
    )

# -----------------------------------------------------------------------------
# Fully Unique Feature Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FULLY UNIQUE FEATURE INSPECTION")
logger.info("-" * 70)

if fully_unique_features:

    logger.info(
        "Fully Unique Features Detected      : %d",
        len(fully_unique_features)
    )

    for feature in fully_unique_features:

        logger.info(
            "Feature : %-40s | Unique Values : %d",
            feature,
            int(df[feature].nunique(dropna=False))
        )

else:

    logger.info(
        "Fully Unique Feature Status         : NONE DETECTED"
    )

# -----------------------------------------------------------------------------
# Feature Accounting Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE CARDINALITY ACCOUNTING")
logger.info("-" * 70)

accounted_features = (
    len(single_value_features)
    + len(binary_features)
    + len(low_cardinality_features)
    + len(high_cardinality_features)
)

logger.info(
    "Single-Value Features               : %d",
    len(single_value_features)
)

logger.info(
    "Binary Features                     : %d",
    len(binary_features)
)

logger.info(
    "Low-Cardinality Features            : %d",
    len(low_cardinality_features)
)

logger.info(
    "High-Cardinality Features           : %d",
    len(high_cardinality_features)
)

logger.info(
    "Accounted Predictor Features        : %d",
    accounted_features
)

logger.info(
    "Expected Predictor Features         : %d",
    len(predictor_features)
)

if accounted_features != len(predictor_features):

    logger.error(
        "Feature Cardinality Accounting     : FAILED"
    )

    raise ValueError(
        "Feature cardinality accounting does not "
        "match the predictor feature count."
    )

logger.info(
    "Feature Cardinality Accounting     : PASSED"
)

# -----------------------------------------------------------------------------
# Final Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("UNIQUE VALUE VALIDATION")
logger.info("-" * 70)

if len(single_value_features) == 0:

    logger.info(
        "Single-Value Feature Status         : NONE DETECTED"
    )

else:

    logger.info(
        "Single-Value Feature Status         : %d DETECTED",
        len(single_value_features)
    )

logger.info(
    "Unique Value Analysis               : PASSED"
)

logger.info("=" * 70)
````

## 4.7 Data Type Validation

This section validates the data types of the dataset before preprocessing.

The objectives are to:

- Validate the raw data type of all predictor features.
- Validate the target variable data type.
- Confirm that all expected predictor columns are present.
- Identify numeric, categorical, and object-type features.
- Verify that no unexpected data types are present.
- Confirm that the raw dataset schema is consistent with the expected dataset structure.

Because the dataset is initially loaded using string-preserving mode, categorical predictors and the `Age` feature are expected to be represented as `object` at this stage.

No data type conversion is performed in this section. Data transformation and numerical encoding will be handled during the preprocessing stage.

````python


import logging
import pandas as pd
from IPython.display import display

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Data Type Validation")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.7."
    )

if not isinstance(df, pd.DataFrame):
    logger.error(
        "Dataset DataFrame Type Validation  : FAILED"
    )
    raise TypeError(
        "Dataset object 'df' is not a pandas DataFrame."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE
elif "class" in df.columns:
    target_feature = "class"
else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "Target feature could not be resolved."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Feature Resolution
# -----------------------------------------------------------------------------

if "PREDICTOR_FEATURES" in globals():
    predictor_features = list(PREDICTOR_FEATURES)
else:
    predictor_features = [
        column
        for column in df.columns
        if column != target_feature
    ]

logger.info("-" * 70)
logger.info("EXPECTED DATA TYPE STRUCTURE")
logger.info("-" * 70)

logger.info(
    "Total Dataset Columns               : %d",
    len(df.columns)
)

logger.info(
    "Expected Predictor Features         : 34"
)

logger.info(
    "Expected Target Feature             : %s",
    target_feature
)

logger.info(
    "Expected Raw Predictor Type         : object"
)

logger.info(
    "Expected Raw Target Type            : object"
)

# -----------------------------------------------------------------------------
# Predictor Feature Count Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREDICTOR FEATURE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

if len(predictor_features) != 34:

    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected 34 predictor features, "
        f"but found {len(predictor_features)}."
    )

logger.info(
    "Predictor Feature Count Validation : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Column Availability
# -----------------------------------------------------------------------------

missing_predictors = [
    feature
    for feature in predictor_features
    if feature not in df.columns
]

if missing_predictors:

    logger.error(
        "Predictor Column Availability       : FAILED"
    )

    raise ValueError(
        "Missing predictor columns: "
        + ", ".join(missing_predictors)
    )

logger.info(
    "Predictor Column Availability       : PASSED"
)

# -----------------------------------------------------------------------------
# Target Exclusion Validation
# -----------------------------------------------------------------------------

if target_feature in predictor_features:

    logger.error(
        "Target Exclusion Validation        : FAILED"
    )

    raise ValueError(
        "Target feature is incorrectly "
        "included in predictor features."
    )

logger.info(
    "Target Exclusion Validation        : PASSED"
)

# -----------------------------------------------------------------------------
# Raw Data Type Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RAW DATA TYPE INSPECTION")
logger.info("-" * 70)

dtype_records = []

for feature in predictor_features:

    raw_dtype = str(df[feature].dtype)

    dtype_records.append(
        {
            "Feature": feature,
            "Raw Data Type": raw_dtype
        }
    )

# Target data type

target_dtype = str(df[target_feature].dtype)

dtype_records.append(
    {
        "Feature": target_feature,
        "Raw Data Type": target_dtype
    }
)

dtype_df = pd.DataFrame(dtype_records)

# -----------------------------------------------------------------------------
# Data Type Category Counts
# -----------------------------------------------------------------------------

object_features = [
    feature
    for feature in predictor_features
    if str(df[feature].dtype) == "object"
]

numeric_features = [
    feature
    for feature in predictor_features
    if pd.api.types.is_numeric_dtype(df[feature])
]

other_dtype_features = [
    feature
    for feature in predictor_features
    if (
        feature not in object_features
        and feature not in numeric_features
    )
]

logger.info(
    "Object-Type Predictor Features      : %d",
    len(object_features)
)

logger.info(
    "Numeric Predictor Features           : %d",
    len(numeric_features)
)

logger.info(
    "Other-Type Predictor Features        : %d",
    len(other_dtype_features)
)

# -----------------------------------------------------------------------------
# Feature-Level Data Type Table
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DATA TYPE DETAILS")
logger.info("-" * 70)

display(dtype_df)

# -----------------------------------------------------------------------------
# Age Data Type Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE FEATURE DATA TYPE VALIDATION")
logger.info("-" * 70)

age_feature = "Age"

if age_feature not in df.columns:

    logger.error(
        "Age Feature Availability            : FAILED"
    )

    raise ValueError(
        "Expected Age feature was not found "
        "in the dataset."
    )

logger.info(
    "Age Feature                         : %s",
    age_feature
)

logger.info(
    "Age Raw Data Type                   : %s",
    str(df[age_feature].dtype)
)

if str(df[age_feature].dtype) != "object":

    logger.error(
        "Age Raw Data Type Validation       : FAILED"
    )

    raise TypeError(
        "Expected Age to have raw dtype "
        "'object' because the dataset was "
        "loaded using dtype=str."
    )

logger.info(
    "Age Raw Data Type Validation       : PASSED"
)

# -----------------------------------------------------------------------------
# Target Data Type Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET DATA TYPE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Feature                     : %s",
    target_feature
)

logger.info(
    "Target Raw Data Type               : %s",
    target_dtype
)

if target_dtype != "object":

    logger.error(
        "Target Data Type Validation       : FAILED"
    )

    raise TypeError(
        "Expected target feature to have "
        "raw dtype 'object'."
    )

logger.info(
    "Target Data Type Validation       : PASSED"
)

# -----------------------------------------------------------------------------
# Raw Predictor Data Type Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREDICTOR DATA TYPE VALIDATION")
logger.info("-" * 70)

unexpected_dtype_features = [
    feature
    for feature in predictor_features
    if str(df[feature].dtype) != "object"
]

logger.info(
    "Expected Raw Predictor Type         : object"
)

logger.info(
    "Object-Type Predictor Features      : %d",
    len(object_features)
)

logger.info(
    "Unexpected Predictor Data Types     : %d",
    len(unexpected_dtype_features)
)

if unexpected_dtype_features:

    logger.error(
        "Predictor Data Type Validation     : FAILED"
    )

    raise TypeError(
        "Unexpected raw data types detected "
        "for predictor features: "
        + ", ".join(unexpected_dtype_features)
    )

logger.info(
    "Predictor Data Type Validation     : PASSED"
)

# -----------------------------------------------------------------------------
# Data Type Accounting
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA TYPE ACCOUNTING VALIDATION")
logger.info("-" * 70)

accounted_predictors = (
    len(object_features)
    + len(numeric_features)
    + len(other_dtype_features)
)

logger.info(
    "Object-Type Features                : %d",
    len(object_features)
)

logger.info(
    "Numeric Features                    : %d",
    len(numeric_features)
)

logger.info(
    "Other-Type Features                 : %d",
    len(other_dtype_features)
)

logger.info(
    "Accounted Predictor Features        : %d",
    accounted_predictors
)

logger.info(
    "Expected Predictor Features         : %d",
    len(predictor_features)
)

if accounted_predictors != len(predictor_features):

    logger.error(
        "Data Type Accounting Validation    : FAILED"
    )

    raise ValueError(
        "Data type accounting does not "
        "match predictor feature count."
    )

logger.info(
    "Data Type Accounting Validation    : PASSED"
)

# -----------------------------------------------------------------------------
# Final Validation Summary
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA TYPE VALIDATION SUMMARY")
logger.info("-" * 70)

logger.info(
    "Dataset Columns                    : %d",
    len(df.columns)
)

logger.info(
    "Predictor Features                 : %d",
    len(predictor_features)
)

logger.info(
    "Object-Type Predictors             : %d",
    len(object_features)
)

logger.info(
    "Numeric Predictors                 : %d",
    len(numeric_features)
)

logger.info(
    "Other-Type Predictors              : %d",
    len(other_dtype_features)
)

logger.info(
    "Target Data Type                   : %s",
    target_dtype
)

logger.info(
    "Age Data Type                      : %s",
    str(df[age_feature].dtype)
)

logger.info("-" * 70)

if (
    len(unexpected_dtype_features) == 0
    and target_dtype == "object"
    and str(df[age_feature].dtype) == "object"
):
    logger.info(
        "Data Type Status                   : VALID"
    )
else:
    logger.error(
        "Data Type Status                   : INVALID"
    )

    raise ValueError(
        "One or more data type validation "
        "conditions failed."
    )

logger.info(
    "Data Type Validation                : PASSED"
)

logger.info("=" * 70)
````

## 4.8 Feature Range Validation

This section validates whether predictor feature values fall within the expected ranges of the dermatology dataset.

The objectives are to:

- Validate the value range of all 34 predictor features.
- Detect unexpected or out-of-range feature values.
- Validate categorical clinical and histopathological feature values.
- Validate the `Age` feature separately.
- Handle the raw `?` token without modifying the original dataset.
- Confirm that no invalid predictor values are present.
- Preserve the original raw dataset for subsequent preprocessing.

The dermatology dataset contains categorical clinical and histopathological attributes represented by coded values. The `Age` feature is numeric in meaning but is currently stored as a string because the raw dataset was loaded using `dtype=str`.

No permanent data transformation is performed in this section.

````python


import logging
import pandas as pd
import numpy as np
from IPython.display import display

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Feature Range Validation")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():

    logger.error(
        "Dataset Object Validation           : FAILED"
    )

    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.8."
    )

if not isinstance(df, pd.DataFrame):

    logger.error(
        "Dataset DataFrame Type Validation  : FAILED"
    )

    raise TypeError(
        "Dataset object 'df' is not a pandas DataFrame."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():

    target_feature = TARGET_FEATURE

elif "class" in df.columns:

    target_feature = "class"

else:

    logger.error(
        "Target Feature Resolution           : FAILED"
    )

    raise NameError(
        "Target feature could not be resolved."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Feature Resolution
# -----------------------------------------------------------------------------

if "PREDICTOR_FEATURES" in globals():

    predictor_features = list(PREDICTOR_FEATURES)

else:

    predictor_features = [
        column
        for column in df.columns
        if column != target_feature
    ]

# -----------------------------------------------------------------------------
# Predictor Feature Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREDICTOR FEATURE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Expected Predictor Features         : 34"
)

if len(predictor_features) != 34:

    logger.error(
        "Predictor Feature Count Validation : FAILED"
    )

    raise ValueError(
        f"Expected 34 predictor features, "
        f"but found {len(predictor_features)}."
    )

logger.info(
    "Predictor Feature Count Validation : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Availability
# -----------------------------------------------------------------------------

missing_predictors = [
    feature
    for feature in predictor_features
    if feature not in df.columns
]

if missing_predictors:

    logger.error(
        "Predictor Column Availability       : FAILED"
    )

    raise ValueError(
        "Missing predictor columns: "
        + ", ".join(missing_predictors)
    )

logger.info(
    "Predictor Column Availability       : PASSED"
)

# -----------------------------------------------------------------------------
# Target Exclusion
# -----------------------------------------------------------------------------

if target_feature in predictor_features:

    logger.error(
        "Target Exclusion Validation        : FAILED"
    )

    raise ValueError(
        "Target feature is incorrectly "
        "included in predictor features."
    )

logger.info(
    "Target Exclusion Validation        : PASSED"
)

# -----------------------------------------------------------------------------
# Expected Dataset Feature Ranges
# -----------------------------------------------------------------------------
#
# The dermatology dataset uses:
#
#   Clinical / Histopathological features:
#       0, 1, 2, 3
#
#   Age:
#       0 - 100 years
#
# The raw dataset may contain '?' for missing feature values.
# These tokens are excluded from range calculations and are handled later.
#
# IMPORTANT:
# The original df is NOT modified.
# -----------------------------------------------------------------------------

EXPECTED_CATEGORICAL_RANGE = {0, 1, 2, 3}

AGE_MIN = 0
AGE_MAX = 100

# -----------------------------------------------------------------------------
# Feature Group Resolution
# -----------------------------------------------------------------------------

if "CLINICAL_FEATURES" in globals():

    clinical_features = list(CLINICAL_FEATURES)

else:

    clinical_features = [
        "erythema",
        "scaling",
        "definite_borders",
        "itching",
        "koebner_phenomenon",
        "polygonal_papules",
        "follicular_papules",
        "oral_mucosal_involvement",
        "knee_and_elbow_involvement",
        "scalp_involvement",
        "family_history",
        "melanin_incontinence"
    ]

if "HISTOPATHOLOGICAL_FEATURES" in globals():

    histopathological_features = list(
        HISTOPATHOLOGICAL_FEATURES
    )

else:

    histopathological_features = [
        "eosinophils_in_the_infiltrate",
        "PNL_infiltrate",
        "fibrosis_of_the_papillary_dermis",
        "exocytosis",
        "acanthosis",
        "hyperkeratosis",
        "parakeratosis",
        "clubbing_of_the_rete_ridges",
        "elongation_of_the_rete_ridges",
        "thinning_of_the_suprapapillary_epidermis",
        "spongiform_pustule",
        "munro_microabcess",
        "focal_hypergranulosis",
        "disappearance_of_the_granular_layer",
        "vacuolisation_and_damage_of_basal_layer",
        "spongiosis",
        "saw-tooth_appearance_of_retes",
        "follicular_horn_plug",
        "perifollicular_parakeratosis",
        "inflammatory_monoluclear_inflitrate",
        "band-like_infiltrate"
    ]

age_feature = "Age"

# -----------------------------------------------------------------------------
# Feature Group Availability Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE GROUP AVAILABILITY")
logger.info("-" * 70)

expected_group_features = (
    clinical_features
    + histopathological_features
    + [age_feature]
)

missing_group_features = [
    feature
    for feature in expected_group_features
    if feature not in df.columns
]

if missing_group_features:

    logger.error(
        "Feature Group Availability           : FAILED"
    )

    raise ValueError(
        "Missing feature-group columns: "
        + ", ".join(missing_group_features)
    )

logger.info(
    "Clinical Features                    : %d",
    len(clinical_features)
)

logger.info(
    "Histopathological Features           : %d",
    len(histopathological_features)
)

logger.info(
    "Age Feature                          : %s",
    age_feature
)

logger.info(
    "Feature Group Availability           : PASSED"
)

# -----------------------------------------------------------------------------
# Categorical Feature Range Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE RANGE INSPECTION")
logger.info("-" * 70)

categorical_features = (
    clinical_features
    + histopathological_features
)

range_results = []

invalid_categorical_features = []

total_question_tokens = 0

for feature in categorical_features:

    raw_series = df[feature].astype(str).str.strip()

    question_count = (
        raw_series.eq("?").sum()
    )

    total_question_tokens += question_count

    valid_series = raw_series[
        ~raw_series.eq("?")
        & ~raw_series.eq("")
    ]

    numeric_values = pd.to_numeric(
        valid_series,
        errors="coerce"
    )

    non_numeric_count = numeric_values.isna().sum()

    numeric_values = numeric_values.dropna()

    if len(numeric_values) > 0:

        minimum_value = numeric_values.min()
        maximum_value = numeric_values.max()

        unique_values = sorted(
            numeric_values.unique().tolist()
        )

    else:

        minimum_value = np.nan
        maximum_value = np.nan
        unique_values = []

    invalid_range_values = [
        value
        for value in unique_values
        if value not in EXPECTED_CATEGORICAL_RANGE
    ]

    invalid_count = len(invalid_range_values)

    if (
        non_numeric_count > 0
        or invalid_count > 0
    ):

        invalid_categorical_features.append(
            feature
        )

    range_results.append(
        {
            "Feature": feature,
            "Minimum": minimum_value,
            "Maximum": maximum_value,
            "Unique Values": len(unique_values),
            "Invalid Values": invalid_count,
            "'?' Tokens": question_count
        }
    )

# -----------------------------------------------------------------------------
# Categorical Range Summary
# -----------------------------------------------------------------------------

categorical_range_df = pd.DataFrame(
    range_results
)

logger.info(
    "Categorical Predictor Features      : %d",
    len(categorical_features)
)

logger.info(
    "Allowed Categorical Values          : {0, 1, 2, 3}"
)

logger.info(
    "Features With Invalid Values        : %d",
    len(invalid_categorical_features)
)

logger.info(
    "Question-Mark Tokens Excluded       : %d",
    total_question_tokens
)

display(categorical_range_df)

# -----------------------------------------------------------------------------
# Categorical Range Validation
# -----------------------------------------------------------------------------

if invalid_categorical_features:

    logger.error(
        "Categorical Feature Range Validation : FAILED"
    )

    raise ValueError(
        "Invalid categorical feature values detected in: "
        + ", ".join(invalid_categorical_features)
    )

logger.info(
    "Categorical Feature Range Validation : PASSED"
)

# -----------------------------------------------------------------------------
# Age Feature Range Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE FEATURE RANGE INSPECTION")
logger.info("-" * 70)

age_raw = df[age_feature].astype(str).str.strip()

age_question_count = age_raw.eq("?").sum()

age_blank_count = age_raw.eq("").sum()

age_valid = age_raw[
    ~age_raw.eq("?")
    & ~age_raw.eq("")
]

age_numeric = pd.to_numeric(
    age_valid,
    errors="coerce"
)

age_non_numeric_count = age_numeric.isna().sum()

age_numeric_valid = age_numeric.dropna()

if len(age_numeric_valid) > 0:

    age_minimum = age_numeric_valid.min()
    age_maximum = age_numeric_valid.max()

else:

    age_minimum = np.nan
    age_maximum = np.nan

age_out_of_range = age_numeric_valid[
    (age_numeric_valid < AGE_MIN)
    | (age_numeric_valid > AGE_MAX)
]

logger.info(
    "Age Feature                         : %s",
    age_feature
)

logger.info(
    "Age Minimum Expected                : %d",
    AGE_MIN
)

logger.info(
    "Age Maximum Expected                : %d",
    AGE_MAX
)

logger.info(
    "Age Minimum Observed                : %s",
    age_minimum
)

logger.info(
    "Age Maximum Observed                : %s",
    age_maximum
)

logger.info(
    "Age '?' Tokens                      : %d",
    age_question_count
)

logger.info(
    "Age Blank Values                    : %d",
    age_blank_count
)

logger.info(
    "Age Non-Numeric Values              : %d",
    age_non_numeric_count
)

logger.info(
    "Age Out-of-Range Values             : %d",
    len(age_out_of_range)
)

# -----------------------------------------------------------------------------
# Age Range Validation
# -----------------------------------------------------------------------------

if age_non_numeric_count > 0:

    logger.error(
        "Age Numeric Validation              : FAILED"
    )

    raise ValueError(
        "Non-numeric Age values detected "
        "after excluding '?' and blank values."
    )

if len(age_out_of_range) > 0:

    logger.error(
        "Age Range Validation                : FAILED"
    )

    raise ValueError(
        "Age values outside the expected "
        f"range {AGE_MIN}-{AGE_MAX} were detected."
    )

logger.info(
    "Age Numeric Validation              : PASSED"
)

logger.info(
    "Age Range Validation                : PASSED"
)

# -----------------------------------------------------------------------------
# Target Range Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET RANGE VALIDATION")
logger.info("-" * 70)

target_raw = (
    df[target_feature]
    .astype(str)
    .str.strip()
)

target_values = sorted(
    target_raw.unique().tolist()
)

logger.info(
    "Target Feature                     : %s",
    target_feature
)

logger.info(
    "Target Unique Classes               : %d",
    len(target_values)
)

logger.info(
    "Target Classes                      : %s",
    target_values
)

if "1" not in target_values:

    logger.error(
        "Target Class Range Validation      : FAILED"
    )

    raise ValueError(
        "Expected target class values were "
        "not detected."
    )

logger.info(
    "Target Class Range Validation      : PASSED"
)

# -----------------------------------------------------------------------------
# Complete Feature Range Summary
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE RANGE VALIDATION SUMMARY")
logger.info("-" * 70)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Categorical Features                : %d",
    len(categorical_features)
)

logger.info(
    "Age Feature                         : %s",
    age_feature
)

logger.info(
    "Categorical Allowed Range           : 0-3"
)

logger.info(
    "Age Allowed Range                   : %d-%d",
    AGE_MIN,
    AGE_MAX
)

logger.info(
    "Categorical Invalid Features        : %d",
    len(invalid_categorical_features)
)

logger.info(
    "Age Out-of-Range Values             : %d",
    len(age_out_of_range)
)

logger.info(
    "Age '?' Tokens                      : %d",
    age_question_count
)

# -----------------------------------------------------------------------------
# Final Validation
# -----------------------------------------------------------------------------

range_validation_passed = (
    len(invalid_categorical_features) == 0
    and age_non_numeric_count == 0
    and len(age_out_of_range) == 0
)

logger.info("-" * 70)
logger.info("FINAL RANGE VALIDATION")
logger.info("-" * 70)

if range_validation_passed:

    logger.info(
        "Feature Range Status                : VALID"
    )

    logger.info(
        "Invalid Predictor Values            : NONE DETECTED"
    )

else:

    logger.error(
        "Feature Range Status                : INVALID"
    )

    raise ValueError(
        "Feature range validation failed."
    )

logger.info(
    "Feature Range Validation            : PASSED"
)

logger.info("=" * 70)
````

## 4.9 Age Validation

This section performs a dedicated validation of the `Age` feature.

The validation objectives are to:

- Confirm that the `Age` feature exists in the dataset.
- Detect and account for `?` missing-value tokens.
- Detect blank and whitespace-only Age values.
- Validate that non-missing Age values are numeric.
- Validate that Age values are within the expected range of 0–100 years.
- Report the minimum and maximum observed Age values.
- Confirm complete Age-value accounting.
- Preserve the original raw `Age` values without modification.

The `Age` feature is validated separately because it is numerically meaningful, while the remaining clinical and histopathological features are coded categorical attributes.

No imputation, deletion, or permanent type conversion is performed in this section. Missing Age values will be handled later during preprocessing.

````python


import logging
import pandas as pd
import numpy as np
from IPython.display import display

logger = logging.getLogger(__name__)

# -----------------------------------------------------------------------------
# Logging Configuration
# -----------------------------------------------------------------------------

if not logger.handlers:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s - %(levelname)s - %(message)s"
    )

logger.info("=" * 70)
logger.info("Age Validation")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():

    logger.error(
        "Dataset Object Validation           : FAILED"
    )

    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 4.9."
    )

if not isinstance(df, pd.DataFrame):

    logger.error(
        "Dataset DataFrame Type Validation  : FAILED"
    )

    raise TypeError(
        "Dataset object 'df' is not a pandas DataFrame."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Age Feature Resolution
# -----------------------------------------------------------------------------

if "AGE_FEATURE" in globals():

    age_feature = AGE_FEATURE

elif "Age" in df.columns:

    age_feature = "Age"

else:

    logger.error(
        "Age Feature Resolution              : FAILED"
    )

    raise NameError(
        "Age feature 'Age' was not found in the dataset."
    )

logger.info(
    "Age Feature Resolution              : PASSED"
)

# -----------------------------------------------------------------------------
# Age Feature Availability
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE FEATURE AVAILABILITY")
logger.info("-" * 70)

logger.info(
    "Age Feature                         : %s",
    age_feature
)

logger.info(
    "Age Feature Position                : %d",
    df.columns.get_loc(age_feature) + 1
)

if age_feature not in df.columns:

    logger.error(
        "Age Feature Availability            : FAILED"
    )

    raise ValueError(
        "Age feature is not available in the dataset."
    )

logger.info(
    "Age Feature Availability            : PASSED"
)

# -----------------------------------------------------------------------------
# Raw Age Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RAW AGE VALUE INSPECTION")
logger.info("-" * 70)

age_raw = (
    df[age_feature]
    .astype(str)
    .str.strip()
)

total_records = len(age_raw)

question_mark_count = int(
    age_raw.eq("?").sum()
)

blank_count = int(
    age_raw.eq("").sum()
)

whitespace_count = int(
    df[age_feature]
    .astype(str)
    .apply(lambda value: value != value.strip())
    .sum()
)

logger.info(
    "Total Records                       : %d",
    total_records
)

logger.info(
    "Age '?' Tokens                      : %d",
    question_mark_count
)

logger.info(
    "Age Blank Values                    : %d",
    blank_count
)

logger.info(
    "Age Whitespace-Affected Values      : %d",
    whitespace_count
)

# -----------------------------------------------------------------------------
# Valid Age Value Extraction
# -----------------------------------------------------------------------------
#
# '?' and blank values are excluded from numeric validation.
# The original df is NOT modified.
# -----------------------------------------------------------------------------

age_valid_raw = age_raw[
    ~age_raw.eq("?")
    & ~age_raw.eq("")
]

age_numeric = pd.to_numeric(
    age_valid_raw,
    errors="coerce"
)

non_numeric_count = int(
    age_numeric.isna().sum()
)

age_numeric_valid = age_numeric.dropna()

numeric_age_count = len(age_numeric_valid)

# -----------------------------------------------------------------------------
# Numeric Age Range Analysis
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC VALIDATION")
logger.info("-" * 70)

logger.info(
    "Valid Numeric Age Values             : %d",
    numeric_age_count
)

logger.info(
    "Non-Numeric Age Values               : %d",
    non_numeric_count
)

if numeric_age_count > 0:

    age_minimum = float(
        age_numeric_valid.min()
    )

    age_maximum = float(
        age_numeric_valid.max()
    )

else:

    age_minimum = np.nan
    age_maximum = np.nan

logger.info(
    "Minimum Observed Age                : %s",
    age_minimum
)

logger.info(
    "Maximum Observed Age                : %s",
    age_maximum
)

# -----------------------------------------------------------------------------
# Expected Age Range
# -----------------------------------------------------------------------------

AGE_MINIMUM = 0
AGE_MAXIMUM = 100

logger.info(
    "Expected Minimum Age                : %d",
    AGE_MINIMUM
)

logger.info(
    "Expected Maximum Age                : %d",
    AGE_MAXIMUM
)

# -----------------------------------------------------------------------------
# Out-of-Range Age Detection
# -----------------------------------------------------------------------------

if numeric_age_count > 0:

    out_of_range_mask = (
        (age_numeric_valid < AGE_MINIMUM)
        | (age_numeric_valid > AGE_MAXIMUM)
    )

    out_of_range_count = int(
        out_of_range_mask.sum()
    )

    out_of_range_values = (
        age_numeric_valid[out_of_range_mask]
        .unique()
        .tolist()
    )

else:

    out_of_range_count = 0
    out_of_range_values = []

logger.info("-" * 70)
logger.info("AGE RANGE INSPECTION")
logger.info("-" * 70)

logger.info(
    "Expected Age Range                  : %d - %d",
    AGE_MINIMUM,
    AGE_MAXIMUM
)

logger.info(
    "Minimum Observed Age                : %s",
    age_minimum
)

logger.info(
    "Maximum Observed Age                : %s",
    age_maximum
)

logger.info(
    "Out-of-Range Age Values             : %d",
    out_of_range_count
)

if out_of_range_count > 0:

    logger.info(
        "Out-of-Range Values                 : %s",
        out_of_range_values
    )

# -----------------------------------------------------------------------------
# Age Value Classification
# -----------------------------------------------------------------------------

age_validation_table = pd.DataFrame(
    {
        "Validation Metric": [
            "Total Records",
            "Valid Numeric Ages",
            "'?' Tokens",
            "Blank Values",
            "Non-Numeric Values",
            "Minimum Observed Age",
            "Maximum Observed Age",
            "Out-of-Range Values"
        ],
        "Value": [
            total_records,
            numeric_age_count,
            question_mark_count,
            blank_count,
            non_numeric_count,
            age_minimum,
            age_maximum,
            out_of_range_count
        ],
        "Status": [
            "VALID",
            "VALID",
            "MISSING TOKEN",
            "VALID",
            "VALID" if non_numeric_count == 0 else "INVALID",
            "VALID" if numeric_age_count > 0 else "NOT AVAILABLE",
            "VALID" if numeric_age_count > 0 else "NOT AVAILABLE",
            "VALID" if out_of_range_count == 0 else "INVALID"
        ]
    }
)

# -----------------------------------------------------------------------------
# Age Accounting Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE VALUE ACCOUNTING")
logger.info("-" * 70)

accounted_age_values = (
    numeric_age_count
    + non_numeric_count
    + question_mark_count
    + blank_count
)

logger.info(
    "Total Records                       : %d",
    total_records
)

logger.info(
    "Numeric Age Values                  : %d",
    numeric_age_count
)

logger.info(
    "Non-Numeric Age Values              : %d",
    non_numeric_count
)

logger.info(
    "Age '?' Tokens                      : %d",
    question_mark_count
)

logger.info(
    "Age Blank Values                    : %d",
    blank_count
)

logger.info(
    "Total Accounted Age Values          : %d",
    accounted_age_values
)

if accounted_age_values != total_records:

    logger.error(
        "Age Value Accounting Validation    : FAILED"
    )

    raise ValueError(
        "Age value accounting does not match "
        "the total number of records."
    )

logger.info(
    "Age Value Accounting Validation    : PASSED"
)

# -----------------------------------------------------------------------------
# Age Validation Table
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE VALIDATION TABLE")
logger.info("-" * 70)

display(
    age_validation_table
)

# -----------------------------------------------------------------------------
# Final Age Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL AGE VALIDATION")
logger.info("-" * 70)

age_validation_passed = (
    age_feature in df.columns
    and non_numeric_count == 0
    and out_of_range_count == 0
    and accounted_age_values == total_records
)

if age_validation_passed:

    logger.info(
        "Age Feature Status                  : VALID"
    )

    logger.info(
        "Age Missing Token Status             : %s",
        "PRESENT AND ACCOUNTED"
        if question_mark_count > 0
        else "NONE DETECTED"
    )

    logger.info(
        "Age Numeric Status                   : VALID"
    )

    logger.info(
        "Age Range Status                     : VALID"
    )

    logger.info(
        "Age Validation                      : PASSED"
    )

else:

    logger.error(
        "Age Validation                      : FAILED"
    )

    raise ValueError(
        "One or more Age validation checks failed."
    )

logger.info("=" * 70)
````

## 5.2 Statistical Summary

This section provides a statistical summary of the predictor features in the dermatology dataset.

The analysis focuses on the distribution and variability of the clinical, histopathological, and age-related features. * Mean
* Standard deviation

* 75th percentile
* Maximum value
* Missing-token counts
* Feature-level variability

````python


import logging
import pandas as pd
import numpy as np

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Statistical Summary")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run Section 3.1 Dataset Loading before Section 5.2."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE
elif "class" in df.columns:
    target_feature = "class"
else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "Target feature could not be resolved."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Feature Resolution
# -----------------------------------------------------------------------------

if "PREDICTOR_FEATURES" in globals():
    predictor_features = list(PREDICTOR_FEATURES)
else:
    predictor_features = [
        column
        for column in df.columns
        if column != target_feature
    ]

logger.info(
    "Predictor Feature Resolution         : PASSED"
)

# -----------------------------------------------------------------------------
# Predictor Validation
# -----------------------------------------------------------------------------

missing_predictors = [
    feature
    for feature in predictor_features
    if feature not in df.columns
]

if missing_predictors:
    logger.error(
        "Predictor Column Availability       : FAILED"
    )
    raise ValueError(
        "Missing predictor columns: "
        + ", ".join(missing_predictors)
    )

logger.info("-" * 70)
logger.info("STATISTICAL SUMMARY CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    len(df)
)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

logger.info(
    "Target Feature                      : %s",
    target_feature
)

# -----------------------------------------------------------------------------
# Numeric Conversion for Statistical Analysis
# -----------------------------------------------------------------------------
# The original dataframe remains unchanged.
# '?' values are converted to NaN only in the analysis copy.

analysis_df = df[predictor_features].copy()

for feature in predictor_features:
    analysis_df[feature] = pd.to_numeric(
        analysis_df[feature],
        errors="coerce"
    )

logger.info(
    "Numeric Conversion for Analysis     : PASSED"
)

# -----------------------------------------------------------------------------
# Feature Category Identification
# -----------------------------------------------------------------------------

age_feature = "Age"

categorical_features = [
    feature
    for feature in predictor_features
    if feature != age_feature
]

logger.info("-" * 70)
logger.info("FEATURE TYPE SUMMARY")
logger.info("-" * 70)

logger.info(
    "Categorical Predictor Features      : %d",
    len(categorical_features)
)

logger.info(
    "Age Feature                         : %s",
    age_feature
)

logger.info(
    "Total Predictor Features            : %d",
    len(predictor_features)
)

# -----------------------------------------------------------------------------
# Statistical Summary Calculation
# -----------------------------------------------------------------------------

statistical_summary = pd.DataFrame({
    "Feature": predictor_features,
    "Count": [
        analysis_df[feature].count()
        for feature in predictor_features
    ],
    "Mean": [
        analysis_df[feature].mean()
        for feature in predictor_features
    ],
    "Std": [
        analysis_df[feature].std()
        for feature in predictor_features
    ],
    "Minimum": [
        analysis_df[feature].min()
        for feature in predictor_features
    ],
    "25%": [
        analysis_df[feature].quantile(0.25)
        for feature in predictor_features
    ],
    "Median": [
        analysis_df[feature].median()
        for feature in predictor_features
    ],
    "75%": [
        analysis_df[feature].quantile(0.75)
        for feature in predictor_features
    ],
    "Maximum": [
        analysis_df[feature].max()
        for feature in predictor_features
    ],
    "Missing Tokens": [
        df[feature].astype(str).eq("?").sum()
        for feature in predictor_features
    ]
})

# -----------------------------------------------------------------------------
# Numeric Formatting
# -----------------------------------------------------------------------------

numeric_columns = [
    "Mean",
    "Std",
    "Minimum",
    "25%",
    "Median",
    "75%",
    "Maximum"
]

statistical_summary[numeric_columns] = (
    statistical_summary[numeric_columns]
    .round(2)
)

# -----------------------------------------------------------------------------
# Statistical Validation
# -----------------------------------------------------------------------------

if len(statistical_summary) != len(predictor_features):
    logger.error(
        "Statistical Feature Accounting       : FAILED"
    )
    raise ValueError(
        "Statistical summary does not contain "
        "all predictor features."
    )

logger.info(
    "Statistical Feature Accounting       : PASSED"
)

# -----------------------------------------------------------------------------
# Overall Statistical Information
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STATISTICAL SUMMARY OVERVIEW")
logger.info("-" * 70)

logger.info(
    "Features Included                   : %d",
    len(statistical_summary)
)

logger.info(
    "Records Available                   : %d",
    len(df)
)

logger.info(
    "Age Missing Tokens                  : %d",
    int(
        df[age_feature]
        .astype(str)
        .eq("?")
        .sum()
    )
)

logger.info(
    "Statistical Summary Status          : READY"
)

# -----------------------------------------------------------------------------
# Statistical Summary Table
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STATISTICAL SUMMARY TABLE")
logger.info("-" * 70)

display(statistical_summary)

# -----------------------------------------------------------------------------
# Age Statistical Summary
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE STATISTICAL SUMMARY")
logger.info("-" * 70)

age_series = analysis_df[age_feature].dropna()

logger.info(
    "Age Valid Numeric Values             : %d",
    len(age_series)
)

logger.info(
    "Age Mean                             : %.2f",
    age_series.mean()
)

logger.info(
    "Age Standard Deviation               : %.2f",
    age_series.std()
)

logger.info(
    "Age Minimum                          : %.2f",
    age_series.min()
)

logger.info(
    "Age Median                           : %.2f",
    age_series.median()
)

logger.info(
    "Age Maximum                          : %.2f",
    age_series.max()
)

# -----------------------------------------------------------------------------
# Final Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL STATISTICAL SUMMARY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Predictor Feature Count              : %d",
    len(predictor_features)
)

logger.info(
    "Statistical Summary Row Count        : %d",
    len(statistical_summary)
)

logger.info(
    "Target Excluded                      : %s",
    target_feature not in statistical_summary["Feature"].values
)

logger.info(
    "Missing Tokens Preserved for Analysis: YES"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info(
    "Statistical Summary                  : PASSED"
)

logger.info("=" * 70)
````

## 5.3 Target Class Distribution

This section analyzes the distribution of the target variable `class` across all records in the dataset.

The analysis identifies the number and percentage of records belonging to each target class and verifies that all target classes are represented. A class distribution table and visualization are generated to provide a clear understanding of the target composition before model development.

### Objectives
- Identify all unique target classes.
- Calculate the number of records in each target class.
- Calculate the percentage distribution of each class.
- Verify that no target class is missing.
- Validate target-class accounting against the total dataset size.
- Visualize the target class distribution.



````python


import logging
import pandas as pd
import matplotlib.pyplot as plt

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Target Class Distribution")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run the dataset loading section before Section 5.3."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE
elif "class" in df.columns:
    target_feature = "class"
else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "TARGET_FEATURE is not available and target column 'class' "
        "was not found in the dataset."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Availability Validation
# -----------------------------------------------------------------------------

if target_feature not in df.columns:
    logger.error(
        "Target Feature Availability         : FAILED"
    )
    raise ValueError(
        f"Target feature '{target_feature}' is not available in the dataset."
    )

logger.info("-" * 70)
logger.info("TARGET CLASS INSPECTION")
logger.info("-" * 70)

total_records = len(df)

logger.info(
    "Target Feature                      : %s",
    target_feature
)

logger.info(
    "Total Records                       : %d",
    total_records
)

# -----------------------------------------------------------------------------
# Target Null Validation
# -----------------------------------------------------------------------------

target_null_count = df[target_feature].isna().sum()

logger.info(
    "Target Null Values                  : %d",
    target_null_count
)

if target_null_count != 0:

    logger.error(
        "Target Null Validation             : FAILED"
    )

    raise ValueError(
        f"Target feature contains {target_null_count} null values."
    )

logger.info(
    "Target Null Validation             : PASSED"
)

# -----------------------------------------------------------------------------
# Target Class Extraction
# -----------------------------------------------------------------------------

target_classes = sorted(
    df[target_feature].astype(str).str.strip().unique(),
    key=lambda value: int(value) if value.isdigit() else value
)

target_class_count = len(target_classes)

logger.info("-" * 70)
logger.info("TARGET CLASS INFORMATION")
logger.info("-" * 70)

logger.info(
    "Target Unique Classes               : %d",
    target_class_count
)

logger.info(
    "Target Classes                      : %s",
    target_classes
)

# -----------------------------------------------------------------------------
# Target Class Counts
# -----------------------------------------------------------------------------

class_counts = (
    df[target_feature]
    .astype(str)
    .str.strip()
    .value_counts()
    .sort_index(
        key=lambda index: index.map(
            lambda value: int(value) if str(value).isdigit() else str(value)
        )
    )
)

# -----------------------------------------------------------------------------
# Target Class Percentages
# -----------------------------------------------------------------------------

class_percentages = (
    class_counts / total_records * 100
)

# -----------------------------------------------------------------------------
# Target Distribution Table
# -----------------------------------------------------------------------------

target_distribution = pd.DataFrame({
    "Class": class_counts.index,
    "Count": class_counts.values,
    "Percentage": class_percentages.values
})

target_distribution["Percentage"] = target_distribution[
    "Percentage"
].round(2)

# -----------------------------------------------------------------------------
# Class Accounting
# -----------------------------------------------------------------------------

accounted_records = int(
    target_distribution["Count"].sum()
)

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)

for _, row in target_distribution.iterrows():

    logger.info(
        "Class %-30s | Count : %3d | Percentage : %6.2f%%",
        str(row["Class"]),
        int(row["Count"]),
        float(row["Percentage"])
    )

logger.info("-" * 70)

logger.info(
    "Total Class-Accounted Records       : %d",
    accounted_records
)

logger.info(
    "Total Dataset Records               : %d",
    total_records
)

# -----------------------------------------------------------------------------
# Target Class Accounting Validation
# -----------------------------------------------------------------------------

if accounted_records != total_records:

    logger.error(
        "Target Class Accounting             : FAILED"
    )

    raise ValueError(
        "Target class counts do not account for all dataset records."
    )

logger.info(
    "Target Class Accounting             : PASSED"
)

# -----------------------------------------------------------------------------
# Class Representation Validation
# -----------------------------------------------------------------------------

missing_classes = [
    target_class
    for target_class in target_classes
    if class_counts.get(target_class, 0) == 0
]

if missing_classes:

    logger.error(
        "Target Class Representation         : FAILED"
    )

    raise ValueError(
        "Target classes with zero records: "
        + ", ".join(missing_classes)
    )

logger.info(
    "Target Class Representation         : PASSED"
)

# -----------------------------------------------------------------------------
# Target Distribution Table Display
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION TABLE")
logger.info("-" * 70)

display(
    target_distribution
)

# -----------------------------------------------------------------------------
# Target Distribution Visualization
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION PLOT")
logger.info("-" * 70)

plt.figure(figsize=(9, 5))

bars = plt.bar(
    target_distribution["Class"].astype(str),
    target_distribution["Count"]
)

plt.title("Target Class Distribution")
plt.xlabel("Target Class")
plt.ylabel("Number of Records")
plt.grid(
    axis="y",
    alpha=0.3
)

# Display count labels above bars
for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()

# -----------------------------------------------------------------------------
# Final Target Distribution Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL TARGET CLASS DISTRIBUTION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    total_records
)

logger.info(
    "Target Classes                      : %d",
    target_class_count
)

logger.info(
    "Class-Accounted Records             : %d",
    accounted_records
)

logger.info(
    "Missing Target Classes              : %d",
    len(missing_classes)
)

logger.info(
    "Target Distribution Status          : VALID"
)

logger.info(
    "Target Class Distribution           : PASSED"
)

logger.info("=" * 70)
````

## 5.4 Class Imbalance Analysis

This section evaluates the degree of class imbalance in the target variable `class`.

Although all six target classes are represented in the dataset, their frequencies are not evenly distributed. The analysis compares the largest and smallest class sizes, calculates the class imbalance ratio, identifies the majority and minority classes, and determines whether class imbalance should be considered during model development.

### Objectives
- Identify the majority and minority target classes.
- Calculate class frequency and percentage distribution.
- Calculate the class imbalance ratio.
- Measure the difference between the largest and smallest classes.
- Determine whether class imbalance is present.
- Assess the implication of imbalance for model training and evaluation.

> **Note:** Class imbalance is assessed using the original target distribution. No resampling or class-balancing technique is applied during EDA.

````python


import logging
import pandas as pd
import matplotlib.pyplot as plt

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("Class Imbalance Analysis")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# Dataset Object Validation
# -----------------------------------------------------------------------------

if "df" not in globals():
    logger.error(
        "Dataset Object Validation           : FAILED"
    )
    raise NameError(
        "Dataset DataFrame 'df' is not available. "
        "Run the dataset loading section before Section 5.4."
    )

logger.info(
    "Dataset Object Validation           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Feature Resolution
# -----------------------------------------------------------------------------

if "TARGET_FEATURE" in globals():
    target_feature = TARGET_FEATURE
elif "class" in df.columns:
    target_feature = "class"
else:
    logger.error(
        "Target Feature Resolution           : FAILED"
    )
    raise NameError(
        "TARGET_FEATURE is not available and target column 'class' "
        "was not found in the dataset."
    )

logger.info(
    "Target Feature Resolution           : PASSED"
)

# -----------------------------------------------------------------------------
# Target Class Distribution Resolution
# -----------------------------------------------------------------------------

if "target_distribution" in globals():

    imbalance_distribution = target_distribution.copy()

else:

    class_counts = (
        df[target_feature]
        .astype(str)
        .str.strip()
        .value_counts()
    )

    imbalance_distribution = pd.DataFrame({
        "Class": class_counts.index.astype(str),
        "Count": class_counts.values
    })

    imbalance_distribution["Percentage"] = (
        imbalance_distribution["Count"]
        / len(df)
        * 100
    ).round(2)

# -----------------------------------------------------------------------------
# Class Count Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE CONFIGURATION")
logger.info("-" * 70)

total_records = len(df)
total_classes = len(imbalance_distribution)

logger.info(
    "Total Records                       : %d",
    total_records
)

logger.info(
    "Total Target Classes               : %d",
    total_classes
)

if total_classes < 2:

    logger.error(
        "Class Count Validation             : FAILED"
    )

    raise ValueError(
        "At least two target classes are required "
        "for class imbalance analysis."
    )

logger.info(
    "Class Count Validation             : PASSED"
)

# -----------------------------------------------------------------------------
# Majority and Minority Class Detection
# -----------------------------------------------------------------------------

imbalance_distribution = imbalance_distribution.sort_values(
    by="Count",
    ascending=False
).reset_index(drop=True)

majority_class = str(
    imbalance_distribution.loc[0, "Class"]
)

majority_count = int(
    imbalance_distribution.loc[0, "Count"]
)

minority_class = str(
    imbalance_distribution.loc[
        len(imbalance_distribution) - 1,
        "Class"
    ]
)

minority_count = int(
    imbalance_distribution.loc[
        len(imbalance_distribution) - 1,
        "Count"
    ]
)

# -----------------------------------------------------------------------------
# Imbalance Metrics
# -----------------------------------------------------------------------------

imbalance_ratio = (
    majority_count / minority_count
)

count_difference = (
    majority_count - minority_count
)

minority_percentage = (
    minority_count / total_records * 100
)

majority_percentage = (
    majority_count / total_records * 100
)

# -----------------------------------------------------------------------------
# Class Imbalance Inspection
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE INSPECTION")
logger.info("-" * 70)

logger.info(
    "Majority Class                      : %s",
    majority_class
)

logger.info(
    "Majority Class Count                : %d",
    majority_count
)

logger.info(
    "Majority Class Percentage           : %.2f%%",
    majority_percentage
)

logger.info(
    "Minority Class                      : %s",
    minority_class
)

logger.info(
    "Minority Class Count                : %d",
    minority_count
)

logger.info(
    "Minority Class Percentage           : %.2f%%",
    minority_percentage
)

logger.info(
    "Majority-Minority Count Difference  : %d",
    count_difference
)

logger.info(
    "Class Imbalance Ratio               : %.2f : 1",
    imbalance_ratio
)

# -----------------------------------------------------------------------------
# Imbalance Status
# -----------------------------------------------------------------------------

if imbalance_ratio <= 1.50:

    imbalance_status = "LOW"

elif imbalance_ratio <= 3.00:

    imbalance_status = "MODERATE"

else:

    imbalance_status = "HIGH"

logger.info(
    "Class Imbalance Status              : %s",
    imbalance_status
)

# -----------------------------------------------------------------------------
# Class Imbalance Table
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE ANALYSIS TABLE")
logger.info("-" * 70)

imbalance_table = imbalance_distribution.copy()

imbalance_table["Class"] = (
    imbalance_table["Class"].astype(str)
)

imbalance_table["Count"] = (
    imbalance_table["Count"].astype(int)
)

imbalance_table["Percentage"] = (
    imbalance_table["Percentage"].astype(float)
)

imbalance_table["Relative to Majority"] = (
    imbalance_table["Count"]
    / majority_count
    * 100
).round(2)

display(
    imbalance_table
)

# -----------------------------------------------------------------------------
# Imbalance Visualization
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE DISTRIBUTION PLOT")
logger.info("-" * 70)

plt.figure(figsize=(9, 5))

class_labels = imbalance_table["Class"].astype(str).tolist()
class_values = imbalance_table["Count"].tolist()

bars = plt.bar(
    class_labels,
    class_values
)

plt.title("Class Imbalance Analysis")
plt.xlabel("Target Class")
plt.ylabel("Number of Records")
plt.grid(
    axis="y",
    alpha=0.3
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()

# -----------------------------------------------------------------------------
# Imbalance Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE VALIDATION")
logger.info("-" * 70)

accounted_records = int(
    imbalance_table["Count"].sum()
)

logger.info(
    "Class-Accounted Records             : %d",
    accounted_records
)

logger.info(
    "Total Dataset Records               : %d",
    total_records
)

if accounted_records != total_records:

    logger.error(
        "Class Accounting Validation        : FAILED"
    )

    raise ValueError(
        "Class counts do not account for all dataset records."
    )

logger.info(
    "Class Accounting Validation        : PASSED"
)

logger.info(
    "Majority Class                     : %s",
    majority_class
)

logger.info(
    "Minority Class                     : %s",
    minority_class
)

logger.info(
    "Class Imbalance Ratio              : %.2f : 1",
    imbalance_ratio
)

logger.info(
    "Class Imbalance Status             : %s",
    imbalance_status
)

# -----------------------------------------------------------------------------
# Final Validation
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CLASS IMBALANCE ASSESSMENT")
logger.info("-" * 70)

logger.info(
    "Target Classes                     : %d",
    total_classes
)

logger.info(
    "Majority Class                     : %s",
    majority_class
)

logger.info(
    "Minority Class                     : %s",
    minority_class
)

logger.info(
    "Majority Class Count               : %d",
    majority_count
)

logger.info(
    "Minority Class Count               : %d",
    minority_count
)

logger.info(
    "Class Imbalance Ratio              : %.2f : 1",
    imbalance_ratio
)

logger.info(
    "Class Imbalance Status             : %s",
    imbalance_status
)

logger.info(
    "Class Imbalance Analysis           : PASSED"
)

logger.info("=" * 70)
````

## 5.5 Age Distribution Analysis

Age is analyzed separately from the categorical clinical and histopathological features because it represents a continuous demographic variable.

The analysis includes:

- Valid numeric age observations
- Missing `"?"` tokens
- Mean, median and standard deviation
- Minimum and maximum observed age
- Quartile distribution
- Interquartile range (IQR)
- Age distribution visualization
- Age range validation against the expected range of 0–100 years

The original dataset is not modified during this analysis. The `"?"` tokens are treated as missing values only for numerical age analysis.

This analysis helps identify the central tendency, variability and overall distribution of patient age before preprocessing and model development.

````python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Age Distribution Analysis")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET AND FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : PASSED")
logger.info("Target Feature Resolution           : PASSED")
logger.info("Predictor Feature Resolution         : PASSED")

AGE_FEATURE = "Age"

if AGE_FEATURE not in df.columns:
    raise KeyError(f"Required Age feature '{AGE_FEATURE}' not found in dataset.")

logger.info("Age Feature Resolution              : PASSED")
logger.info("-" * 70)

# ----------------------------------------------------------------------
# AGE DISTRIBUTION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("AGE DISTRIBUTION CONFIGURATION")
logger.info("-" * 70)

total_records = len(df)

# Preserve original values and convert only for analysis
age_raw = df[AGE_FEATURE].astype(str).str.strip()

# Identify missing token
age_question_tokens = (age_raw == "?").sum()

# Identify blank values
age_blank_values = age_raw.eq("").sum()

# Convert numeric values safely
age_numeric = pd.to_numeric(
    age_raw.replace("?", np.nan),
    errors="coerce"
)

valid_age = age_numeric.dropna()

logger.info(f"Total Records                       : {total_records}")
logger.info(f"Age Feature                         : {AGE_FEATURE}")
logger.info(f"Age '?' Tokens                      : {age_question_tokens}")
logger.info(f"Age Blank Values                    : {age_blank_values}")
logger.info(f"Valid Numeric Ages                  : {len(valid_age)}")

logger.info("-" * 70)

# ----------------------------------------------------------------------
# AGE DESCRIPTIVE STATISTICS
# ----------------------------------------------------------------------

logger.info("AGE DESCRIPTIVE STATISTICS")
logger.info("-" * 70)

age_count = valid_age.count()
age_mean = valid_age.mean()
age_std = valid_age.std()
age_min = valid_age.min()
age_q1 = valid_age.quantile(0.25)
age_median = valid_age.median()
age_q3 = valid_age.quantile(0.75)
age_max = valid_age.max()
age_iqr = age_q3 - age_q1

logger.info(f"Valid Numeric Ages                  : {age_count}")
logger.info(f"Mean Age                            : {age_mean:.2f}")
logger.info(f"Standard Deviation                  : {age_std:.2f}")
logger.info(f"Minimum Age                         : {age_min:.2f}")
logger.info(f"25th Percentile (Q1)                : {age_q1:.2f}")
logger.info(f"Median Age                          : {age_median:.2f}")
logger.info(f"75th Percentile (Q3)                : {age_q3:.2f}")
logger.info(f"Maximum Age                         : {age_max:.2f}")
logger.info(f"Interquartile Range (IQR)            : {age_iqr:.2f}")

logger.info("-" * 70)

# ----------------------------------------------------------------------
# AGE RANGE VALIDATION
# ----------------------------------------------------------------------

logger.info("AGE RANGE VALIDATION")
logger.info("-" * 70)

EXPECTED_MIN_AGE = 0
EXPECTED_MAX_AGE = 100

out_of_range_count = (
    ((valid_age < EXPECTED_MIN_AGE) |
     (valid_age > EXPECTED_MAX_AGE))
    .sum()
)

logger.info(f"Expected Age Range                  : {EXPECTED_MIN_AGE} - {EXPECTED_MAX_AGE}")
logger.info(f"Minimum Observed Age                : {age_min:.2f}")
logger.info(f"Maximum Observed Age                : {age_max:.2f}")
logger.info(f"Out-of-Range Age Values             : {out_of_range_count}")

age_range_status = (
    "PASSED"
    if out_of_range_count == 0
    else "FAILED"
)

logger.info(f"Age Range Validation                : {age_range_status}")

logger.info("-" * 70)

# ----------------------------------------------------------------------
# AGE DISTRIBUTION TABLE
# ----------------------------------------------------------------------

logger.info("AGE DISTRIBUTION SUMMARY TABLE")
logger.info("-" * 70)

age_summary_table = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Valid Numeric Ages",
        "'?' Tokens",
        "Blank Values",
        "Mean Age",
        "Standard Deviation",
        "Minimum Age",
        "Q1 (25%)",
        "Median Age",
        "Q3 (75%)",
        "Maximum Age",
        "Interquartile Range",
        "Out-of-Range Values"
    ],
    "Value": [
        total_records,
        age_count,
        age_question_tokens,
        age_blank_values,
        round(age_mean, 2),
        round(age_std, 2),
        round(age_min, 2),
        round(age_q1, 2),
        round(age_median, 2),
        round(age_q3, 2),
        round(age_max, 2),
        round(age_iqr, 2),
        out_of_range_count
    ],
    "Status": [
        "VALID",
        "VALID",
        "MISSING TOKEN",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID",
        "VALID"
    ]
})

display(age_summary_table)

# ----------------------------------------------------------------------
# AGE DISTRIBUTION PLOT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE DISTRIBUTION PLOT")
logger.info("-" * 70)

plt.figure(figsize=(10, 6))

plt.hist(
    valid_age,
    bins=15,
    edgecolor="black"
)

plt.axvline(
    age_mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {age_mean:.2f}"
)

plt.axvline(
    age_median,
    linestyle=":",
    linewidth=2,
    label=f"Median = {age_median:.2f}"
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------------------------------------------------
# FINAL AGE DISTRIBUTION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL AGE DISTRIBUTION VALIDATION")
logger.info("-" * 70)

age_accounted_records = age_count + age_question_tokens + age_blank_values

logger.info(f"Total Records                       : {total_records}")
logger.info(f"Valid Numeric Ages                  : {age_count}")
logger.info(f"Age '?' Tokens                      : {age_question_tokens}")
logger.info(f"Age Blank Values                    : {age_blank_values}")
logger.info(f"Total Accounted Age Values          : {age_accounted_records}")

logger.info(
    f"Age Value Accounting                : "
    f"{'PASSED' if age_accounted_records == total_records else 'FAILED'}"
)

logger.info(
    f"Age Distribution Status             : "
    f"{'VALID' if out_of_range_count == 0 else 'INVALID'}"
)

logger.info(
    f"Age Range Validation                : "
    f"{'PASSED' if out_of_range_count == 0 else 'FAILED'}"
)

logger.info("Original Dataset Modified            : NO")
logger.info("Age Distribution Analysis            : PASSED")
logger.info("=" * 70)
````

## 5.6 Clinical Feature Distribution

This section analyzes the distribution of the 12 clinical features in the dataset.

The clinical features are categorical variables represented using ordinal values from 0 to 3, where higher values indicate greater presence or severity according to the dataset's feature encoding.

The analysis includes:

- Clinical feature identification and validation
- Unique value inspection
- Frequency distribution of each clinical feature
- Percentage distribution across feature values
- Clinical feature variability assessment
- Missing and invalid value verification
- Visualization of clinical feature distributions
- Summary table for comparative analysis

The original dataset is preserved without modification.

````python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Clinical Feature Distribution")
logger.info("=" * 70)

# --------------------------------------------------------------
# DATASET VALIDATION
# --------------------------------------------------------------

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object 'df' must be a pandas DataFrame.")

if TARGET_COLUMN not in df.columns:
    raise KeyError(f"Target feature '{TARGET_COLUMN}' not found in dataset.")

# Expected clinical features from the dataset structure
clinical_features = [
    "erythema",
    "scaling",
    "definite_borders",
    "itching",
    "koebner_phenomenon",
    "polygonal_papules",
    "follicular_papules",
    "oral_mucosal_involvement",
    "knee_and_elbow_involvement",
    "scalp_involvement",
    "family_history",
    "melanin_incontinence"
]

# Validate clinical feature availability
missing_clinical_features = [
    feature for feature in clinical_features
    if feature not in df.columns
]

if missing_clinical_features:
    raise KeyError(
        f"Clinical features missing from dataset: {missing_clinical_features}"
    )

logger.info("Dataset Object Validation           : PASSED")
logger.info("Clinical Feature Resolution          : PASSED")
logger.info("-" * 70)

# --------------------------------------------------------------
# CLINICAL FEATURE CONFIGURATION
# --------------------------------------------------------------

logger.info("CLINICAL FEATURE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Total Records                       : {len(df)}"
)

logger.info(
    f"Expected Clinical Features           : {len(clinical_features)}"
)

logger.info(
    f"Resolved Clinical Features           : {len(clinical_features)}"
)

logger.info(
    f"Clinical Feature Range               : 0 - 3"
)

logger.info("-" * 70)

# --------------------------------------------------------------
# CLINICAL FEATURE VALUE ANALYSIS
# --------------------------------------------------------------

clinical_summary = []

allowed_values = {0, 1, 2, 3}

for feature in clinical_features:

    # Preserve original values and convert only for analysis
    feature_values = pd.to_numeric(
        df[feature].astype(str).str.strip(),
        errors="coerce"
    )

    valid_numeric = feature_values.notna()

    missing_values = int(feature_values.isna().sum())

    invalid_range = int(
        ((feature_values.notna()) &
         (~feature_values.isin(allowed_values))).sum()
    )

    unique_values = sorted(
        feature_values.dropna().unique().tolist()
    )

    value_counts = feature_values.value_counts(
        sort=False
    ).reindex([0, 1, 2, 3], fill_value=0)

    total_valid = int(valid_numeric.sum())

    percentage_0 = (
        value_counts.loc[0] / total_valid * 100
        if total_valid > 0 else 0
    )

    percentage_1 = (
        value_counts.loc[1] / total_valid * 100
        if total_valid > 0 else 0
    )

    percentage_2 = (
        value_counts.loc[2] / total_valid * 100
        if total_valid > 0 else 0
    )

    percentage_3 = (
        value_counts.loc[3] / total_valid * 100
        if total_valid > 0 else 0
    )

    clinical_summary.append({
        "Feature": feature,
        "Unique Values": len(unique_values),
        "Value 0 Count": int(value_counts.loc[0]),
        "Value 1 Count": int(value_counts.loc[1]),
        "Value 2 Count": int(value_counts.loc[2]),
        "Value 3 Count": int(value_counts.loc[3]),
        "Value 0 %": round(percentage_0, 2),
        "Value 1 %": round(percentage_1, 2),
        "Value 2 %": round(percentage_2, 2),
        "Value 3 %": round(percentage_3, 2),
        "Missing Values": missing_values,
        "Range Violations": invalid_range
    })

clinical_summary_df = pd.DataFrame(clinical_summary)

# --------------------------------------------------------------
# FEATURE DISTRIBUTION LOG
# --------------------------------------------------------------

logger.info("CLINICAL FEATURE DISTRIBUTION")
logger.info("-" * 70)

for _, row in clinical_summary_df.iterrows():

    logger.info(
        f"{row['Feature']:<38} | "
        f"Unique : {row['Unique Values']} | "
        f"0 : {row['Value 0 Count']:>3} | "
        f"1 : {row['Value 1 Count']:>3} | "
        f"2 : {row['Value 2 Count']:>3} | "
        f"3 : {row['Value 3 Count']:>3}"
    )

logger.info("-" * 70)

# --------------------------------------------------------------
# CLINICAL DISTRIBUTION TABLE
# --------------------------------------------------------------

logger.info("CLINICAL FEATURE DISTRIBUTION TABLE")
logger.info("-" * 70)

display(clinical_summary_df)

# --------------------------------------------------------------
# VALIDATION SUMMARY
# --------------------------------------------------------------

total_missing = int(
    clinical_summary_df["Missing Values"].sum()
)

total_range_violations = int(
    clinical_summary_df["Range Violations"].sum()
)

features_with_all_expected_values = int(
    (
        clinical_summary_df["Unique Values"] == 4
    ).sum()
)

logger.info("-" * 70)
logger.info("CLINICAL FEATURE VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Clinical Features Analyzed             : {len(clinical_features)}"
)

logger.info(
    f"Features With All Values 0-3            : "
    f"{features_with_all_expected_values}"
)

logger.info(
    f"Missing Clinical Values                 : {total_missing}"
)

logger.info(
    f"Clinical Range Violations               : "
    f"{total_range_violations}"
)

# --------------------------------------------------------------
# VISUALIZATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLINICAL FEATURE DISTRIBUTION PLOT")
logger.info("-" * 70)

plot_data = clinical_summary_df[
    [
        "Feature",
        "Value 0 Count",
        "Value 1 Count",
        "Value 2 Count",
        "Value 3 Count"
    ]
].copy()

plot_data = plot_data.set_index("Feature")

ax = plot_data.plot(
    kind="bar",
    figsize=(16, 7),
    width=0.8
)

ax.set_title(
    "Clinical Feature Distribution"
)

ax.set_xlabel(
    "Clinical Features"
)

ax.set_ylabel(
    "Number of Records"
)

ax.tick_params(
    axis="x",
    rotation=75
)

ax.legend(
    title="Feature Value"
)

plt.tight_layout()
plt.show()

# --------------------------------------------------------------
# FINAL VALIDATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CLINICAL FEATURE DISTRIBUTION VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Clinical Feature Count                : {len(clinical_features)}"
)

logger.info(
    f"Clinical Feature Resolution            : PASSED"
)

logger.info(
    f"Missing Clinical Values               : {total_missing}"
)

logger.info(
    f"Clinical Range Violations             : {total_range_violations}"
)

logger.info(
    f"Clinical Distribution Status           : "
    f"{'VALID' if total_range_violations == 0 else 'REVIEW REQUIRED'}"
)

logger.info(
    f"Original Dataset Modified              : NO"
)

logger.info(
    f"Clinical Feature Distribution          : "
    f"{'PASSED' if total_range_violations == 0 else 'FAILED'}"
)

logger.info("=" * 70)
````

## 5.7 Histopathological Feature Distribution

This section analyzes the distribution of all histopathological predictor features in the dataset.

The analysis includes:

- Validation of the expected 21 histopathological features
- Distribution of values from 0 to 3 for each feature
- Frequency and percentage of each categorical value
- Missing-value inspection
- Range-violation detection
- Identification of dominant feature values
- Visualization of histopathological feature distributions

The histopathological features are ordinal categorical variables with an expected
value range of 0–3.

The original dataset is not modified during this analysis.

````python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Histopathological Feature Distribution")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# DATASET AND FEATURE VALIDATION
# ------------------------------------------------------------------------------

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object must be a pandas DataFrame.")

logger.info(
    "Dataset Object Validation             : PASSED"
)

# Expected histopathological features
HISTOPATHOLOGICAL_FEATURES = [
    "melanin_incontinence",
    "eosinophils_in_the_infiltrate",
    "PNL_infiltrate",
    "fibrosis_of_the_papillary_dermis",
    "exocytosis",
    "acanthosis",
    "hyperkeratosis",
    "parakeratosis",
    "clubbing_of_the_rete_ridges",
    "elongation_of_the_rete_ridges",
    "thinning_of_the_suprapapillary_epidermis",
    "spongiform_pustule",
    "munro_microabcess",
    "focal_hypergranulosis",
    "disappearance_of_the_granular_layer",
    "vacuolisation_and_damage_of_basal_layer",
    "spongiosis",
    "saw-tooth_appearance_of_retes",
    "follicular_horn_plug",
    "perifollicular_parakeratosis",
    "inflammatory_monoluclear_inflitrate",
    "band-like_infiltrate"
]

# ------------------------------------------------------------------------------
# FEATURE RESOLUTION
# ------------------------------------------------------------------------------

resolved_histopathological_features = [
    feature
    for feature in HISTOPATHOLOGICAL_FEATURES
    if feature in df.columns
]

missing_histopathological_features = [
    feature
    for feature in HISTOPATHOLOGICAL_FEATURES
    if feature not in df.columns
]

if missing_histopathological_features:
    raise ValueError(
        f"Missing histopathological features: "
        f"{missing_histopathological_features}"
    )

logger.info(
    "Histopathological Feature Resolution  : PASSED"
)

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

TOTAL_RECORDS = len(df)
EXPECTED_FEATURE_COUNT = len(HISTOPATHOLOGICAL_FEATURES)
RESOLVED_FEATURE_COUNT = len(resolved_histopathological_features)

ALLOWED_VALUES = {0, 1, 2, 3}

logger.info("-" * 70)
logger.info("HISTOPATHOLOGICAL FEATURE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Total Records                       : {TOTAL_RECORDS}"
)

logger.info(
    f"Expected Histopathological Features : {EXPECTED_FEATURE_COUNT}"
)

logger.info(
    f"Resolved Histopathological Features : {RESOLVED_FEATURE_COUNT}"
)

logger.info(
    "Histopathological Feature Range      : 0 - 3"
)

logger.info(
    f"Histopathological Feature Count Validation : "
    f"{'PASSED' if RESOLVED_FEATURE_COUNT == EXPECTED_FEATURE_COUNT else 'FAILED'}"
)

# ------------------------------------------------------------------------------
# FEATURE DISTRIBUTION ANALYSIS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HISTOPATHOLOGICAL FEATURE DISTRIBUTION")
logger.info("-" * 70)

histopathological_distribution = []

for feature in resolved_histopathological_features:

    # Preserve original values for inspection
    raw_values = df[feature]

    # Convert values to numeric for analysis
    numeric_values = pd.to_numeric(
        raw_values.astype(str).str.strip(),
        errors="coerce"
    )

    # --------------------------------------------------------------------------
    # VALUE COUNTS
    # --------------------------------------------------------------------------

    count_0 = int((numeric_values == 0).sum())
    count_1 = int((numeric_values == 1).sum())
    count_2 = int((numeric_values == 2).sum())
    count_3 = int((numeric_values == 3).sum())

    # --------------------------------------------------------------------------
    # MISSING VALUES
    # --------------------------------------------------------------------------

    missing_values = int(numeric_values.isna().sum())

    # --------------------------------------------------------------------------
    # NON-NUMERIC VALUES
    # --------------------------------------------------------------------------

    non_numeric_values = int(
        raw_values.notna().sum() - numeric_values.notna().sum()
    )

    # --------------------------------------------------------------------------
    # RANGE VIOLATIONS
    # --------------------------------------------------------------------------
    # IMPORTANT:
    # Do NOT use int(series_boolean).
    # The boolean Series must be summed first.

    range_violations = int(
        (
            numeric_values.notna()
            & ~numeric_values.isin(ALLOWED_VALUES)
        ).sum()
    )

    # --------------------------------------------------------------------------
    # VALID NUMERIC VALUES
    # --------------------------------------------------------------------------

    valid_numeric_values = int(
        numeric_values.notna().sum()
    )

    # --------------------------------------------------------------------------
    # PERCENTAGES
    # --------------------------------------------------------------------------

    denominator = TOTAL_RECORDS

    pct_0 = (count_0 / denominator) * 100
    pct_1 = (count_1 / denominator) * 100
    pct_2 = (count_2 / denominator) * 100
    pct_3 = (count_3 / denominator) * 100

    # --------------------------------------------------------------------------
    # UNIQUE VALUES
    # --------------------------------------------------------------------------

    unique_values = int(
        numeric_values.dropna().nunique()
    )

    # --------------------------------------------------------------------------
    # LOGGER OUTPUT
    # --------------------------------------------------------------------------

    logger.info(
        f"{feature:<40} | "
        f"Unique : {unique_values} | "
        f"0 : {count_0:>3} | "
        f"1 : {count_1:>3} | "
        f"2 : {count_2:>3} | "
        f"3 : {count_3:>3}"
    )

    # --------------------------------------------------------------------------
    # STORE RESULT
    # --------------------------------------------------------------------------

    histopathological_distribution.append({
        "Feature": feature,
        "Unique Values": unique_values,
        "Value 0 Count": count_0,
        "Value 1 Count": count_1,
        "Value 2 Count": count_2,
        "Value 3 Count": count_3,
        "Value 0 %": round(pct_0, 2),
        "Value 1 %": round(pct_1, 2),
        "Value 2 %": round(pct_2, 2),
        "Value 3 %": round(pct_3, 2),
        "Missing Values": missing_values,
        "Non-Numeric Values": non_numeric_values,
        "Valid Numeric Values": valid_numeric_values,
        "Range Violations": range_violations
    })


# ------------------------------------------------------------------------------
# CREATE DISTRIBUTION DATAFRAME
# ------------------------------------------------------------------------------

histopathological_distribution_df = pd.DataFrame(
    histopathological_distribution
)

# ------------------------------------------------------------------------------
# FEATURE VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HISTOPATHOLOGICAL FEATURE VALIDATION")
logger.info("-" * 70)

features_with_all_values_0_3 = int(
    (
        histopathological_distribution_df["Range Violations"] == 0
    ).sum()
)

total_missing_histopathological_values = int(
    histopathological_distribution_df["Missing Values"].sum()
)

total_non_numeric_histopathological_values = int(
    histopathological_distribution_df["Non-Numeric Values"].sum()
)

total_range_violations = int(
    histopathological_distribution_df["Range Violations"].sum()
)

logger.info(
    f"Histopathological Features Analyzed : "
    f"{RESOLVED_FEATURE_COUNT}"
)

logger.info(
    f"Features With All Values 0-3         : "
    f"{features_with_all_values_0_3}"
)

logger.info(
    f"Missing Histopathological Values      : "
    f"{total_missing_histopathological_values}"
)

logger.info(
    f"Non-Numeric Histopathological Values : "
    f"{total_non_numeric_histopathological_values}"
)

logger.info(
    f"Histopathological Range Violations    : "
    f"{total_range_violations}"
)

# ------------------------------------------------------------------------------
# TABLE OUTPUT
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HISTOPATHOLOGICAL FEATURE DISTRIBUTION TABLE")
logger.info("-" * 70)

display(
    histopathological_distribution_df[
        [
            "Feature",
            "Unique Values",
            "Value 0 Count",
            "Value 1 Count",
            "Value 2 Count",
            "Value 3 Count",
            "Value 0 %",
            "Value 1 %",
            "Value 2 %",
            "Value 3 %",
            "Missing Values",
            "Range Violations"
        ]
    ]
)

# ------------------------------------------------------------------------------
# DISTRIBUTION PLOT
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HISTOPATHOLOGICAL FEATURE DISTRIBUTION PLOT")
logger.info("-" * 70)

plot_df = histopathological_distribution_df.set_index(
    "Feature"
)[
    [
        "Value 0 %",
        "Value 1 %",
        "Value 2 %",
        "Value 3 %"
    ]
]

ax = plot_df.plot(
    kind="bar",
    figsize=(18, 8)
)

ax.set_title(
    "Histopathological Feature Distribution"
)

ax.set_xlabel(
    "Histopathological Features"
)

ax.set_ylabel(
    "Percentage of Records (%)"
)

ax.legend(
    title="Feature Value"
)

plt.xticks(
    rotation=90
)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# FINAL VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL HISTOPATHOLOGICAL FEATURE DISTRIBUTION VALIDATION")
logger.info("-" * 70)

feature_count_status = (
    RESOLVED_FEATURE_COUNT == EXPECTED_FEATURE_COUNT
)

missing_status = (
    total_missing_histopathological_values == 0
)

non_numeric_status = (
    total_non_numeric_histopathological_values == 0
)

range_status = (
    total_range_violations == 0
)

overall_status = (
    feature_count_status
    and missing_status
    and non_numeric_status
    and range_status
)

logger.info(
    f"Histopathological Feature Count       : "
    f"{RESOLVED_FEATURE_COUNT}"
)

logger.info(
    f"Histopathological Feature Resolution  : "
    f"{'PASSED' if feature_count_status else 'FAILED'}"
)

logger.info(
    f"Missing Histopathological Values      : "
    f"{total_missing_histopathological_values}"
)

logger.info(
    f"Non-Numeric Histopathological Values : "
    f"{total_non_numeric_histopathological_values}"
)

logger.info(
    f"Histopathological Range Violations    : "
    f"{total_range_violations}"
)

logger.info(
    f"Histopathological Distribution Status : "
    f"{'VALID' if overall_status else 'INVALID'}"
)

logger.info(
    "Original Dataset Modified              : NO"
)

logger.info(
    f"Histopathological Feature Distribution : "
    f"{'PASSED' if overall_status else 'FAILED'}"
)

logger.info("=" * 70)
````

## 5.8 Feature Variability Analysis

Feature variability analysis evaluates the degree of variation present across the predictor variables. Since the dataset contains ordinal clinical and histopathological features measured primarily on a 0–3 scale, variability is assessed using standard deviation, variance, minimum and maximum values, and the number of unique values.

The analysis helps identify features with very low variability, highly variable features, and features that may provide limited or strong discriminatory information for subsequent machine learning modeling.

The original dataset is preserved without modification during this exploratory analysis.


````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Feature Variability Analysis")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object 'df' must be a pandas DataFrame.")

if TARGET_COLUMN not in df.columns:
    raise ValueError(f"Target feature '{TARGET_COLUMN}' not found in dataset.")

predictor_features = [
    col for col in df.columns
    if col != TARGET_COLUMN
]

logger.info("Dataset Object Validation           : PASSED")
logger.info("Target Feature Resolution            : PASSED")
logger.info("Predictor Feature Resolution         : PASSED")

# ----------------------------------------------------------------------
# FEATURE VARIABILITY CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY CONFIGURATION")
logger.info("-" * 70)

logger.info(f"Total Records                       : {len(df)}")
logger.info(f"Total Predictor Features            : {len(predictor_features)}")
logger.info(f"Target Feature                      : {TARGET_COLUMN}")

logger.info("-" * 70)

# ----------------------------------------------------------------------
# NUMERIC CONVERSION FOR ANALYSIS
# ----------------------------------------------------------------------
# A temporary numeric representation is created only for EDA.
# The original dataset remains unchanged.

variability_data = pd.DataFrame(index=df.index)

for feature in predictor_features:
    variability_data[feature] = pd.to_numeric(
        df[feature],
        errors="coerce"
    )

logger.info("Numeric Conversion for Analysis     : PASSED")
logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# FEATURE VARIABILITY CALCULATION
# ----------------------------------------------------------------------

variability_records = []

for feature in predictor_features:

    values = variability_data[feature]

    valid_values = values.dropna()

    valid_count = int(valid_values.count())
    missing_count = int(values.isna().sum())

    if valid_count > 0:

        feature_mean = float(valid_values.mean())
        feature_variance = float(valid_values.var(ddof=1))
        feature_std = float(valid_values.std(ddof=1))
        feature_min = float(valid_values.min())
        feature_max = float(valid_values.max())
        unique_values = int(valid_values.nunique())

    else:

        feature_mean = np.nan
        feature_variance = np.nan
        feature_std = np.nan
        feature_min = np.nan
        feature_max = np.nan
        unique_values = 0

    variability_records.append({
        "Feature": feature,
        "Valid Values": valid_count,
        "Missing Values": missing_count,
        "Unique Values": unique_values,
        "Mean": feature_mean,
        "Variance": feature_variance,
        "Standard Deviation": feature_std,
        "Minimum": feature_min,
        "Maximum": feature_max
    })

variability_table = pd.DataFrame(variability_records)

logger.info("Feature Variability Calculation     : PASSED")

# ----------------------------------------------------------------------
# VARIABILITY SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY SUMMARY")
logger.info("-" * 70)

valid_std = variability_table["Standard Deviation"].dropna()

if len(valid_std) > 0:

    logger.info(
        f"Minimum Standard Deviation          : "
        f"{valid_std.min():.4f}"
    )

    logger.info(
        f"Maximum Standard Deviation          : "
        f"{valid_std.max():.4f}"
    )

    logger.info(
        f"Mean Standard Deviation             : "
        f"{valid_std.mean():.4f}"
    )

    logger.info(
        f"Median Standard Deviation           : "
        f"{valid_std.median():.4f}"
    )

# ----------------------------------------------------------------------
# LOW-VARIABILITY FEATURE DETECTION
# ----------------------------------------------------------------------
# A very small standard deviation indicates limited variability.
# Threshold is intentionally descriptive for EDA and does not remove
# features automatically.

LOW_VARIABILITY_THRESHOLD = 0.10

low_variability_mask = (
    variability_table["Standard Deviation"] < LOW_VARIABILITY_THRESHOLD
)

low_variability_features = variability_table.loc[
    low_variability_mask,
    "Feature"
].tolist()

logger.info("-" * 70)
logger.info("LOW-VARIABILITY FEATURE ANALYSIS")
logger.info("-" * 70)

logger.info(
    f"Low Variability Threshold          : "
    f"STD < {LOW_VARIABILITY_THRESHOLD:.2f}"
)

logger.info(
    f"Low-Variability Features            : "
    f"{len(low_variability_features)}"
)

if low_variability_features:
    logger.info(
        "Low-Variability Feature List       : "
        + ", ".join(low_variability_features)
    )
else:
    logger.info(
        "Low-Variability Feature List       : NONE"
    )

# ----------------------------------------------------------------------
# HIGH-VARIABILITY FEATURE ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-VARIABILITY FEATURE ANALYSIS")
logger.info("-" * 70)

if len(valid_std) > 0:

    highest_variability_feature = variability_table.loc[
        variability_table["Standard Deviation"].idxmax(),
        "Feature"
    ]

    highest_variability_std = variability_table.loc[
        variability_table["Standard Deviation"].idxmax(),
        "Standard Deviation"
    ]

    logger.info(
        f"Highest Variability Feature         : "
        f"{highest_variability_feature}"
    )

    logger.info(
        f"Highest Standard Deviation          : "
        f"{highest_variability_std:.4f}"
    )

# ----------------------------------------------------------------------
# FEATURE VARIABILITY TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY TABLE")
logger.info("-" * 70)

display(
    variability_table[
        [
            "Feature",
            "Valid Values",
            "Missing Values",
            "Unique Values",
            "Mean",
            "Variance",
            "Standard Deviation",
            "Minimum",
            "Maximum"
        ]
    ].round(4)
)

# ----------------------------------------------------------------------
# SORTED VARIABILITY RANKING
# ----------------------------------------------------------------------

variability_ranking = variability_table.sort_values(
    by="Standard Deviation",
    ascending=False
).reset_index(drop=True)

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY RANKING")
logger.info("-" * 70)

display(
    variability_ranking[
        [
            "Feature",
            "Standard Deviation",
            "Variance",
            "Unique Values"
        ]
    ].round(4)
)

# ----------------------------------------------------------------------
# FEATURE VARIABILITY PLOT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY DISTRIBUTION PLOT")
logger.info("-" * 70)

plot_data = variability_ranking.copy()

plt.figure(figsize=(14, 9))

plt.barh(
    plot_data["Feature"],
    plot_data["Standard Deviation"]
)

plt.xlabel("Standard Deviation")
plt.ylabel("Feature")
plt.title("Feature Variability Analysis")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

# ----------------------------------------------------------------------
# VARIABILITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIABILITY VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Predictor Features Analyzed          : "
    f"{len(variability_table)}"
)

logger.info(
    f"Features With Valid Numeric Values    : "
    f"{int((variability_table['Valid Values'] > 0).sum())}"
)

logger.info(
    f"Features With Missing Values          : "
    f"{int((variability_table['Missing Values'] > 0).sum())}"
)

logger.info(
    f"Low-Variability Features              : "
    f"{len(low_variability_features)}"
)

logger.info(
    f"Highest Variability Feature           : "
    f"{highest_variability_feature if len(valid_std) > 0 else 'N/A'}"
)

# ----------------------------------------------------------------------
# FINAL FEATURE VARIABILITY ASSESSMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE VARIABILITY ASSESSMENT")
logger.info("-" * 70)

feature_count_valid = (
    len(variability_table) == len(predictor_features)
)

numeric_analysis_valid = (
    variability_table["Valid Values"].sum() > 0
)

statistics_valid = (
    variability_table["Standard Deviation"].notna().any()
)

logger.info(
    f"Predictor Feature Count Validation  : "
    f"{'PASSED' if feature_count_valid else 'FAILED'}"
)

logger.info(
    f"Numeric Variability Validation      : "
    f"{'PASSED' if numeric_analysis_valid else 'FAILED'}"
)

logger.info(
    f"Statistical Variability Validation  : "
    f"{'PASSED' if statistics_valid else 'FAILED'}"
)

overall_variability_status = (
    feature_count_valid
    and numeric_analysis_valid
    and statistics_valid
)

logger.info(
    f"Feature Variability Status          : "
    f"{'VALID' if overall_variability_status else 'REVIEW REQUIRED'}"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info(
    f"Feature Variability Analysis         : "
    f"{'PASSED' if overall_variability_status else 'FAILED'}"
)

logger.info("=" * 70)
````

## 5.9 Correlation Matrix

Correlation matrix analysis evaluates the pairwise linear association between the predictor features. The analysis provides a numerical representation of relationships among the clinical, histopathological, and age-related variables.

Because the raw dataset contains categorical/object representations of the clinical and histopathological scores, the predictor variables are temporarily converted to numeric values for correlation analysis. The original dataset is not modified.

The resulting correlation matrix is used to identify the strength and direction of relationships between predictor features and provides the numerical foundation for the subsequent correlation heatmap and high-correlation feature pair analysis.


````python


import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Correlation Matrix")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object 'df' must be a pandas DataFrame.")

if TARGET_COLUMN not in df.columns:
    raise ValueError(
        f"Target feature '{TARGET_COLUMN}' not found in dataset."
    )

predictor_features = [
    col for col in df.columns
    if col != TARGET_COLUMN
]

logger.info("Dataset Object Validation           : PASSED")
logger.info("Target Feature Resolution            : PASSED")
logger.info("Predictor Feature Resolution         : PASSED")

# ----------------------------------------------------------------------
# CORRELATION MATRIX CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX CONFIGURATION")
logger.info("-" * 70)

logger.info(f"Total Records                       : {len(df)}")
logger.info(f"Total Predictor Features            : {len(predictor_features)}")
logger.info(f"Target Feature                      : {TARGET_COLUMN}")

# ----------------------------------------------------------------------
# NUMERIC CONVERSION FOR CORRELATION
# ----------------------------------------------------------------------
# Temporary numeric dataframe is created.
# Original dataset remains unchanged.

correlation_data = pd.DataFrame(index=df.index)

for feature in predictor_features:

    correlation_data[feature] = pd.to_numeric(
        df[feature],
        errors="coerce"
    )

logger.info("Numeric Conversion for Analysis     : PASSED")
logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# NUMERIC FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION FEATURE VALIDATION")
logger.info("-" * 70)

numeric_feature_count = int(
    correlation_data.select_dtypes(include=np.number).shape[1]
)

logger.info(
    f"Numeric Predictor Features          : "
    f"{numeric_feature_count}"
)

logger.info(
    f"Expected Predictor Features         : "
    f"{len(predictor_features)}"
)

feature_resolution_valid = (
    numeric_feature_count == len(predictor_features)
)

logger.info(
    f"Numeric Feature Resolution          : "
    f"{'PASSED' if feature_resolution_valid else 'FAILED'}"
)

# ----------------------------------------------------------------------
# CORRELATION MATRIX CALCULATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX CALCULATION")
logger.info("-" * 70)

correlation_matrix = correlation_data.corr(
    method="pearson"
)

logger.info("Correlation Method                  : Pearson")
logger.info(
    f"Correlation Matrix Shape            : "
    f"{correlation_matrix.shape}"
)

logger.info("Correlation Matrix Calculation      : PASSED")

# ----------------------------------------------------------------------
# CORRELATION MATRIX VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX VALIDATION")
logger.info("-" * 70)

expected_matrix_shape = (
    len(predictor_features),
    len(predictor_features)
)

actual_matrix_shape = correlation_matrix.shape

shape_validation = (
    actual_matrix_shape == expected_matrix_shape
)

logger.info(
    f"Expected Matrix Shape               : "
    f"{expected_matrix_shape}"
)

logger.info(
    f"Actual Matrix Shape                 : "
    f"{actual_matrix_shape}"
)

logger.info(
    f"Matrix Shape Validation              : "
    f"{'PASSED' if shape_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# DIAGONAL VALIDATION
# ----------------------------------------------------------------------

diagonal_values = np.diag(
    correlation_matrix.to_numpy()
)

diagonal_validation = np.allclose(
    diagonal_values,
    1.0,
    equal_nan=False
)

logger.info(
    f"Diagonal Self-Correlation Validation : "
    f"{'PASSED' if diagonal_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# SYMMETRY VALIDATION
# ----------------------------------------------------------------------

symmetry_validation = np.allclose(
    correlation_matrix.to_numpy(),
    correlation_matrix.to_numpy().T,
    equal_nan=True
)

logger.info(
    f"Matrix Symmetry Validation            : "
    f"{'PASSED' if symmetry_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# MISSING CORRELATION VALUES
# ----------------------------------------------------------------------

missing_correlation_values = int(
    correlation_matrix.isna().sum().sum()
)

logger.info(
    f"Missing Correlation Values            : "
    f"{missing_correlation_values}"
)

# ----------------------------------------------------------------------
# CORRELATION RANGE VALIDATION
# ----------------------------------------------------------------------

finite_correlation_values = correlation_matrix.to_numpy()

finite_mask = np.isfinite(
    finite_correlation_values
)

if finite_mask.any():

    correlation_min = float(
        finite_correlation_values[finite_mask].min()
    )

    correlation_max = float(
        finite_correlation_values[finite_mask].max()
    )

else:

    correlation_min = np.nan
    correlation_max = np.nan

range_validation = True

if finite_mask.any():

    range_validation = (
        correlation_min >= -1.0
        and correlation_max <= 1.0
    )

logger.info(
    f"Minimum Correlation Value             : "
    f"{correlation_min:.4f}"
)

logger.info(
    f"Maximum Correlation Value             : "
    f"{correlation_max:.4f}"
)

logger.info(
    f"Correlation Range Validation           : "
    f"{'PASSED' if range_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# CORRELATION MATRIX TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX TABLE")
logger.info("-" * 70)

display(
    correlation_matrix.round(3)
)

# ----------------------------------------------------------------------
# ABSOLUTE CORRELATION MATRIX
# ----------------------------------------------------------------------
# Useful for identifying the strength of relationships irrespective
# of positive or negative direction.

absolute_correlation_matrix = correlation_matrix.abs()

logger.info("-" * 70)
logger.info("ABSOLUTE CORRELATION MATRIX")
logger.info("-" * 70)

display(
    absolute_correlation_matrix.round(3)
)

# ----------------------------------------------------------------------
# CORRELATION SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION SUMMARY")
logger.info("-" * 70)

# Remove diagonal values so self-correlations are not treated
# as feature relationships.

off_diagonal_mask = ~np.eye(
    len(correlation_matrix),
    dtype=bool
)

off_diagonal_values = (
    correlation_matrix.to_numpy()[off_diagonal_mask]
)

off_diagonal_values = off_diagonal_values[
    np.isfinite(off_diagonal_values)
]

if len(off_diagonal_values) > 0:

    absolute_off_diagonal = np.abs(
        off_diagonal_values
    )

    logger.info(
        f"Minimum Pairwise Correlation          : "
        f"{off_diagonal_values.min():.4f}"
    )

    logger.info(
        f"Maximum Pairwise Correlation          : "
        f"{off_diagonal_values.max():.4f}"
    )

    logger.info(
        f"Minimum Absolute Pairwise Correlation : "
        f"{absolute_off_diagonal.min():.4f}"
    )

    logger.info(
        f"Maximum Absolute Pairwise Correlation : "
        f"{absolute_off_diagonal.max():.4f}"
    )

# ----------------------------------------------------------------------
# TOP CORRELATED FEATURE PAIRS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TOP CORRELATED FEATURE PAIRS")
logger.info("-" * 70)

pair_records = []

for i in range(len(correlation_matrix.columns)):

    for j in range(i + 1, len(correlation_matrix.columns)):

        feature_a = correlation_matrix.columns[i]
        feature_b = correlation_matrix.columns[j]

        correlation_value = correlation_matrix.iloc[i, j]

        if pd.notna(correlation_value):

            pair_records.append({
                "Feature 1": feature_a,
                "Feature 2": feature_b,
                "Correlation": correlation_value,
                "Absolute Correlation": abs(correlation_value)
            })

correlation_pairs_df = pd.DataFrame(
    pair_records
)

if not correlation_pairs_df.empty:

    correlation_pairs_df = (
        correlation_pairs_df
        .sort_values(
            by="Absolute Correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

    display(
        correlation_pairs_df.head(15).round(4)
    )

    logger.info(
        f"Unique Feature Pairs Analyzed         : "
        f"{len(correlation_pairs_df)}"
    )

else:

    logger.info(
        "Unique Feature Pairs Analyzed         : 0"
    )

# ----------------------------------------------------------------------
# STRONG CORRELATION PREVIEW
# ----------------------------------------------------------------------

STRONG_CORRELATION_THRESHOLD = 0.80

if not correlation_pairs_df.empty:

    strong_correlation_pairs = correlation_pairs_df[
        correlation_pairs_df["Absolute Correlation"]
        >= STRONG_CORRELATION_THRESHOLD
    ].copy()

else:

    strong_correlation_pairs = pd.DataFrame()

logger.info("-" * 70)
logger.info("STRONG CORRELATION PREVIEW")
logger.info("-" * 70)

logger.info(
    f"Strong Correlation Threshold           : "
    f"|r| >= {STRONG_CORRELATION_THRESHOLD:.2f}"
)

logger.info(
    f"Strong Correlation Pairs               : "
    f"{len(strong_correlation_pairs)}"
)

if not strong_correlation_pairs.empty:

    display(
        strong_correlation_pairs.round(4)
    )

else:

    logger.info(
        "Strong Correlation Pairs               : NONE"
    )

# ----------------------------------------------------------------------
# CORRELATION MATRIX STORAGE
# ----------------------------------------------------------------------

# Keep these objects available for Sections 5.10 and 5.11.

correlation_matrix_eda = correlation_matrix.copy()

logger.info("-" * 70)
logger.info("CORRELATION MATRIX OBJECTS")
logger.info("-" * 70)

logger.info(
    "Correlation Matrix Object             : READY"
)

logger.info(
    "Absolute Correlation Matrix Object     : READY"
)

logger.info(
    "Correlation Pair Analysis Object       : READY"
)

# ----------------------------------------------------------------------
# FINAL CORRELATION MATRIX VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CORRELATION MATRIX VALIDATION")
logger.info("-" * 70)

overall_correlation_status = all([
    feature_resolution_valid,
    shape_validation,
    diagonal_validation,
    symmetry_validation,
    range_validation
])

logger.info(
    f"Predictor Feature Resolution           : "
    f"{'PASSED' if feature_resolution_valid else 'FAILED'}"
)

logger.info(
    f"Matrix Dimension Validation             : "
    f"{'PASSED' if shape_validation else 'FAILED'}"
)

logger.info(
    f"Diagonal Validation                    : "
    f"{'PASSED' if diagonal_validation else 'FAILED'}"
)

logger.info(
    f"Symmetry Validation                    : "
    f"{'PASSED' if symmetry_validation else 'FAILED'}"
)

logger.info(
    f"Correlation Range Validation            : "
    f"{'PASSED' if range_validation else 'FAILED'}"
)

logger.info(
    f"Correlation Matrix Status              : "
    f"{'VALID' if overall_correlation_status else 'REVIEW REQUIRED'}"
)

logger.info(
    "Original Dataset Modified              : NO"
)

logger.info(
    f"Correlation Matrix Analysis             : "
    f"{'PASSED' if overall_correlation_status else 'FAILED'}"
)

logger.info("=" * 70)
````

## 5.10 Correlation Heatmap

The correlation heatmap provides a visual representation of the pairwise relationships among the predictor features. It enables rapid identification of positively and negatively correlated feature groups and highlights potential multicollinearity patterns.

The heatmap is generated from the Pearson correlation matrix calculated in Section 5.9. Correlation values range from -1 to +1, where values closer to +1 indicate strong positive association, values closer to -1 indicate strong negative association, and values near zero indicate weak linear association.

The target feature is excluded from the heatmap because the purpose of this analysis is to examine relationships among predictor variables. The original dataset remains unchanged.


````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Correlation Heatmap")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object 'df' must be a pandas DataFrame.")

if TARGET_COLUMN not in df.columns:
    raise ValueError(
        f"Target feature '{TARGET_COLUMN}' not found in dataset."
    )

logger.info("Dataset Object Validation           : PASSED")
logger.info("Target Feature Resolution            : PASSED")

# ----------------------------------------------------------------------
# CORRELATION MATRIX OBJECT VALIDATION
# ----------------------------------------------------------------------

if "correlation_matrix_eda" not in globals():
    raise NameError(
        "Correlation matrix object 'correlation_matrix_eda' "
        "was not found. Run Section 5.9 first."
    )

correlation_matrix_heatmap = correlation_matrix_eda.copy()

logger.info("-" * 70)
logger.info("HEATMAP CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Correlation Matrix Shape             : "
    f"{correlation_matrix_heatmap.shape}"
)

logger.info(
    f"Predictor Features                   : "
    f"{len(correlation_matrix_heatmap.columns)}"
)

logger.info("Target Feature Excluded              : YES")
logger.info("Correlation Method                   : Pearson")

# ----------------------------------------------------------------------
# MATRIX DIMENSION VALIDATION
# ----------------------------------------------------------------------

expected_feature_count = len(
    [col for col in df.columns if col != TARGET_COLUMN]
)

actual_feature_count = len(
    correlation_matrix_heatmap.columns
)

feature_count_validation = (
    actual_feature_count == expected_feature_count
)

logger.info("-" * 70)
logger.info("HEATMAP MATRIX VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Expected Predictor Features          : "
    f"{expected_feature_count}"
)

logger.info(
    f"Heatmap Predictor Features           : "
    f"{actual_feature_count}"
)

logger.info(
    f"Feature Count Validation             : "
    f"{'PASSED' if feature_count_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# CORRELATION MATRIX NUMERIC VALIDATION
# ----------------------------------------------------------------------

numeric_matrix_validation = (
    all(
        pd.api.types.is_numeric_dtype(
            correlation_matrix_heatmap[col]
        )
        for col in correlation_matrix_heatmap.columns
    )
)

logger.info(
    f"Numeric Correlation Matrix Validation : "
    f"{'PASSED' if numeric_matrix_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# CORRELATION VALUE VALIDATION
# ----------------------------------------------------------------------

correlation_array = correlation_matrix_heatmap.to_numpy()

finite_values = correlation_array[
    np.isfinite(correlation_array)
]

if len(finite_values) > 0:

    heatmap_min_correlation = float(
        finite_values.min()
    )

    heatmap_max_correlation = float(
        finite_values.max()
    )

    correlation_range_validation = (
        heatmap_min_correlation >= -1.0
        and heatmap_max_correlation <= 1.0
    )

else:

    heatmap_min_correlation = np.nan
    heatmap_max_correlation = np.nan
    correlation_range_validation = False

logger.info(
    f"Minimum Correlation                  : "
    f"{heatmap_min_correlation:.4f}"
)

logger.info(
    f"Maximum Correlation                  : "
    f"{heatmap_max_correlation:.4f}"
)

logger.info(
    f"Correlation Range Validation         : "
    f"{'PASSED' if correlation_range_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# HEATMAP GENERATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION HEATMAP GENERATION")
logger.info("-" * 70)

fig, ax = plt.subplots(
    figsize=(20, 17)
)

heatmap = ax.imshow(
    correlation_matrix_heatmap,
    aspect="auto",
    interpolation="nearest",
    vmin=-1,
    vmax=1,
    cmap="coolwarm"
)

# ----------------------------------------------------------------------
# AXIS LABELS
# ----------------------------------------------------------------------

feature_labels = correlation_matrix_heatmap.columns.tolist()

ax.set_xticks(
    np.arange(len(feature_labels))
)

ax.set_yticks(
    np.arange(len(feature_labels))
)

ax.set_xticklabels(
    feature_labels,
    rotation=90,
    fontsize=8
)

ax.set_yticklabels(
    feature_labels,
    fontsize=8
)

# ----------------------------------------------------------------------
# COLOR BAR
# ----------------------------------------------------------------------

colorbar = fig.colorbar(
    heatmap,
    ax=ax,
    fraction=0.046,
    pad=0.04
)

colorbar.set_label(
    "Pearson Correlation Coefficient",
    rotation=270,
    labelpad=20
)

# ----------------------------------------------------------------------
# TITLE
# ----------------------------------------------------------------------

ax.set_title(
    "Predictor Feature Correlation Heatmap",
    fontsize=16,
    pad=20
)

ax.set_xlabel(
    "Predictor Features",
    fontsize=11
)

ax.set_ylabel(
    "Predictor Features",
    fontsize=11
)

plt.tight_layout()

plt.show()

logger.info("Correlation Heatmap Generation     : PASSED")

# ----------------------------------------------------------------------
# HEATMAP DATA SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HEATMAP DATA SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Features Visualized                  : "
    f"{actual_feature_count}"
)

logger.info(
    f"Matrix Dimensions                    : "
    f"{correlation_matrix_heatmap.shape[0]} x "
    f"{correlation_matrix_heatmap.shape[1]}"
)

logger.info(
    f"Correlation Scale                    : "
    f"-1.00 to +1.00"
)

logger.info(
    "Diagonal Self-Correlations           : Present"
)

logger.info(
    "Target Feature                       : Excluded"
)

# ----------------------------------------------------------------------
# ORIGINAL DATASET VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified             : NO"
)

logger.info(
    "Target Feature Included in Heatmap    : NO"
)

# ----------------------------------------------------------------------
# FINAL HEATMAP VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CORRELATION HEATMAP VALIDATION")
logger.info("-" * 70)

overall_heatmap_status = all([
    feature_count_validation,
    numeric_matrix_validation,
    correlation_range_validation
])

logger.info(
    f"Feature Count Validation             : "
    f"{'PASSED' if feature_count_validation else 'FAILED'}"
)

logger.info(
    f"Numeric Matrix Validation             : "
    f"{'PASSED' if numeric_matrix_validation else 'FAILED'}"
)

logger.info(
    f"Correlation Range Validation          : "
    f"{'PASSED' if correlation_range_validation else 'FAILED'}"
)

logger.info(
    f"Heatmap Data Validation               : "
    f"{'PASSED' if overall_heatmap_status else 'FAILED'}"
)

logger.info(
    "Original Dataset Modified             : NO"
)

logger.info(
    f"Correlation Heatmap Analysis          : "
    f"{'PASSED' if overall_heatmap_status else 'FAILED'}"
)

logger.info("=" * 70)
````


## 5.11 High-Correlation Feature Pair Analysis

This section identifies pairs of predictor features that exhibit strong linear relationships with each other.

The analysis includes:

- Validation of the expected 34 predictor features
- Analysis of the Pearson correlation matrix
- Calculation of absolute correlation values
- Exclusion of self-correlations from pair analysis
- Identification of highly correlated feature pairs
- Ranking of feature pairs by absolute correlation strength
- Separation of positive and negative high-correlation relationships
- Evaluation of potential feature redundancy
- Validation of the selected correlation threshold
- Preparation of high-correlation feature-pair information for feature-selection decisions

A correlation threshold of |r| ≥ 0.80 is used to identify strong feature relationships.

Highly correlated feature pairs may indicate redundant information among predictors and will be considered during subsequent feature-selection and model-development stages.

Correlation analysis is used for exploratory and feature-redundancy assessment only and does not imply causal relationships.

The original dataset is preserved and is not modified during this analysis.

````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("High-Correlation Feature Pair Analysis")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET AND FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : PASSED")

# Validate target feature
if TARGET_COLUMN not in df.columns:
    raise KeyError(f"Target feature '{TARGET_COLUMN}' not found in dataset.")

logger.info("Target Feature Resolution            : PASSED")

# Validate predictor features
predictor_features = [col for col in df.columns if col != TARGET_COLUMN]

if len(predictor_features) != 34:
    raise ValueError(
        f"Expected 34 predictor features, found {len(predictor_features)}."
    )

logger.info("Predictor Feature Resolution         : PASSED")

logger.info("-" * 70)
logger.info("HIGH-CORRELATION ANALYSIS CONFIGURATION")
logger.info("-" * 70)

TOTAL_RECORDS = len(df)
TOTAL_PREDICTOR_FEATURES = len(predictor_features)
CORRELATION_THRESHOLD = 0.80

logger.info(
    f"Total Records                       : {TOTAL_RECORDS}"
)
logger.info(
    f"Total Predictor Features            : {TOTAL_PREDICTOR_FEATURES}"
)
logger.info(
    f"Target Feature                      : {TARGET_COLUMN}"
)
logger.info(
    f"Correlation Method                  : Pearson"
)
logger.info(
    f"High-Correlation Threshold          : |r| >= {CORRELATION_THRESHOLD:.2f}"
)

# ----------------------------------------------------------------------
# NUMERIC CONVERSION
# ----------------------------------------------------------------------

numeric_predictors = df[predictor_features].apply(
    pd.to_numeric,
    errors="coerce"
)

numeric_feature_count = numeric_predictors.shape[1]

if numeric_feature_count != TOTAL_PREDICTOR_FEATURES:
    raise ValueError(
        "Numeric predictor feature resolution failed."
    )

logger.info(
    f"Numeric Predictor Features          : {numeric_feature_count}"
)
logger.info(
    "Numeric Conversion for Analysis     : PASSED"
)

logger.info("-" * 70)
logger.info("CORRELATION MATRIX PREPARATION")
logger.info("-" * 70)

# ----------------------------------------------------------------------
# CORRELATION MATRIX
# ----------------------------------------------------------------------

correlation_matrix_511 = numeric_predictors.corr(method="pearson")

expected_matrix_shape = (
    TOTAL_PREDICTOR_FEATURES,
    TOTAL_PREDICTOR_FEATURES
)

actual_matrix_shape = correlation_matrix_511.shape

logger.info(
    f"Expected Correlation Matrix Shape   : {expected_matrix_shape}"
)
logger.info(
    f"Actual Correlation Matrix Shape     : {actual_matrix_shape}"
)

if actual_matrix_shape != expected_matrix_shape:
    raise ValueError(
        "Correlation matrix dimensions do not match predictor feature count."
    )

logger.info("Correlation Matrix Resolution       : PASSED")

# ----------------------------------------------------------------------
# MATRIX VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX VALIDATION")
logger.info("-" * 70)

missing_correlation_values = int(
    correlation_matrix_511.isna().sum().sum()
)

diagonal_validation = np.allclose(
    np.diag(correlation_matrix_511),
    1.0,
    equal_nan=False
)

symmetry_validation = np.allclose(
    correlation_matrix_511.values,
    correlation_matrix_511.values.T,
    equal_nan=True
)

logger.info(
    f"Missing Correlation Values          : {missing_correlation_values}"
)
logger.info(
    f"Diagonal Self-Correlation Validation: "
    f"{'PASSED' if diagonal_validation else 'FAILED'}"
)
logger.info(
    f"Matrix Symmetry Validation          : "
    f"{'PASSED' if symmetry_validation else 'FAILED'}"
)

if not diagonal_validation:
    raise ValueError(
        "Correlation matrix diagonal validation failed."
    )

if not symmetry_validation:
    raise ValueError(
        "Correlation matrix symmetry validation failed."
    )

# ----------------------------------------------------------------------
# HIGH-CORRELATION FEATURE PAIR EXTRACTION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CORRELATION FEATURE PAIR EXTRACTION")
logger.info("-" * 70)

# Work with the upper triangle only to avoid duplicate pairs.
upper_triangle_mask = np.triu(
    np.ones(correlation_matrix_511.shape, dtype=bool),
    k=1
)

upper_triangle = correlation_matrix_511.where(
    upper_triangle_mask
)

high_correlation_pairs = []

for feature_1 in upper_triangle.columns:
    for feature_2 in upper_triangle.index:
        correlation_value = upper_triangle.loc[feature_2, feature_1]

        if pd.notna(correlation_value):
            absolute_correlation = abs(correlation_value)

            if absolute_correlation >= CORRELATION_THRESHOLD:
                high_correlation_pairs.append(
                    {
                        "Feature 1": feature_2,
                        "Feature 2": feature_1,
                        "Correlation": correlation_value,
                        "Absolute Correlation": absolute_correlation,
                        "Relationship": (
                            "Positive"
                            if correlation_value > 0
                            else "Negative"
                        )
                    }
                )

# Convert to DataFrame
high_correlation_pairs_df = pd.DataFrame(
    high_correlation_pairs
)

# ----------------------------------------------------------------------
# HANDLE EMPTY RESULT
# ----------------------------------------------------------------------

if high_correlation_pairs_df.empty:

    high_correlation_pairs_df = pd.DataFrame(
        columns=[
            "Feature 1",
            "Feature 2",
            "Correlation",
            "Absolute Correlation",
            "Relationship"
        ]
    )

else:

    high_correlation_pairs_df = (
        high_correlation_pairs_df
        .sort_values(
            by="Absolute Correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

logger.info(
    f"High-Correlation Feature Pairs     : "
    f"{len(high_correlation_pairs_df)}"
)

# ----------------------------------------------------------------------
# POSITIVE / NEGATIVE RELATIONSHIP ANALYSIS
# ----------------------------------------------------------------------

positive_pairs_df = high_correlation_pairs_df[
    high_correlation_pairs_df["Relationship"] == "Positive"
].copy()

negative_pairs_df = high_correlation_pairs_df[
    high_correlation_pairs_df["Relationship"] == "Negative"
].copy()

logger.info(
    f"Strong Positive Correlation Pairs  : "
    f"{len(positive_pairs_df)}"
)

logger.info(
    f"Strong Negative Correlation Pairs  : "
    f"{len(negative_pairs_df)}"
)

# ----------------------------------------------------------------------
# HIGH-CORRELATION SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CORRELATION FEATURE PAIR SUMMARY")
logger.info("-" * 70)

if not high_correlation_pairs_df.empty:

    strongest_pair = high_correlation_pairs_df.iloc[0]

    logger.info(
        f"Strongest Feature Pair             : "
        f"{strongest_pair['Feature 1']} <-> "
        f"{strongest_pair['Feature 2']}"
    )

    logger.info(
        f"Strongest Correlation               : "
        f"{strongest_pair['Correlation']:.4f}"
    )

    logger.info(
        f"Strongest Absolute Correlation      : "
        f"{strongest_pair['Absolute Correlation']:.4f}"
    )

else:

    logger.info(
        "Strongest Feature Pair             : NONE"
    )

    logger.info(
        "Strongest Correlation               : NONE"
    )

    logger.info(
        "Strongest Absolute Correlation      : NONE"
    )

# ----------------------------------------------------------------------
# HIGH-CORRELATION FEATURE PAIR TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CORRELATION FEATURE PAIR TABLE")
logger.info("-" * 70)

if high_correlation_pairs_df.empty:

    display(
        pd.DataFrame(
            {
                "Message": [
                    f"No feature pairs found with |r| >= "
                    f"{CORRELATION_THRESHOLD:.2f}."
                ]
            }
        )
    )

else:

    display(
        high_correlation_pairs_df.style.format(
            {
                "Correlation": "{:.4f}",
                "Absolute Correlation": "{:.4f}"
            }
        )
    )

# ----------------------------------------------------------------------
# TOP CORRELATION PAIRS TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TOP FEATURE CORRELATION PAIRS")
logger.info("-" * 70)

# Generate top 10 absolute correlations for exploratory reporting.
all_correlation_pairs = []

for i in range(len(correlation_matrix_511.columns)):

    for j in range(i + 1, len(correlation_matrix_511.columns)):

        feature_1 = correlation_matrix_511.columns[i]
        feature_2 = correlation_matrix_511.columns[j]

        correlation_value = correlation_matrix_511.iloc[i, j]

        if pd.notna(correlation_value):

            all_correlation_pairs.append(
                {
                    "Feature 1": feature_1,
                    "Feature 2": feature_2,
                    "Correlation": correlation_value,
                    "Absolute Correlation": abs(correlation_value)
                }
            )

all_correlation_pairs_df = pd.DataFrame(
    all_correlation_pairs
)

if not all_correlation_pairs_df.empty:

    all_correlation_pairs_df = (
        all_correlation_pairs_df
        .sort_values(
            by="Absolute Correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

    top_correlation_pairs_df = (
        all_correlation_pairs_df
        .head(10)
        .copy()
    )

    display(
        top_correlation_pairs_df.style.format(
            {
                "Correlation": "{:.4f}",
                "Absolute Correlation": "{:.4f}"
            }
        )
    )

else:

    top_correlation_pairs_df = pd.DataFrame(
        columns=[
            "Feature 1",
            "Feature 2",
            "Correlation",
            "Absolute Correlation"
        ]
    )

    display(top_correlation_pairs_df)

# ----------------------------------------------------------------------
# HIGH-CORRELATION DISTRIBUTION PLOT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CORRELATION FEATURE PAIR VISUALIZATION")
logger.info("-" * 70)

if not high_correlation_pairs_df.empty:

    plot_data = high_correlation_pairs_df.head(15).copy()

    plot_labels = (
        plot_data["Feature 1"]
        + " ↔ "
        + plot_data["Feature 2"]
    )

    plot_values = plot_data["Correlation"]

    plt.figure(
        figsize=(12, 7)
    )

    plt.barh(
        plot_labels,
        plot_values
    )

    plt.axvline(
        x=0,
        linewidth=1
    )

    plt.axvline(
        x=CORRELATION_THRESHOLD,
        linestyle="--",
        linewidth=1
    )

    plt.axvline(
        x=-CORRELATION_THRESHOLD,
        linestyle="--",
        linewidth=1
    )

    plt.xlabel(
        "Pearson Correlation Coefficient"
    )

    plt.ylabel(
        "Feature Pair"
    )

    plt.title(
        "High-Correlation Feature Pairs"
    )

    plt.xlim(
        -1.0,
        1.0
    )

    plt.gca().invert_yaxis()

    plt.tight_layout()
    plt.show()

    logger.info(
        "High-Correlation Pair Visualization : PASSED"
    )

else:

    logger.info(
        "No high-correlation pairs available for visualization."
    )

# ----------------------------------------------------------------------
# REDUNDANCY ASSESSMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE REDUNDANCY ASSESSMENT")
logger.info("-" * 70)

if len(high_correlation_pairs_df) == 0:

    redundancy_status = "LOW"

    logger.info(
        "High-Correlation Pair Count         : 0"
    )

    logger.info(
        "Feature Redundancy Status           : LOW"
    )

else:

    redundancy_status = "PRESENT"

    logger.info(
        f"High-Correlation Pair Count         : "
        f"{len(high_correlation_pairs_df)}"
    )

    logger.info(
        "Feature Redundancy Status           : PRESENT"
    )

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CORRELATION ANALYSIS OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Correlation Matrix Object            : "
    f"{'READY' if isinstance(correlation_matrix_511, pd.DataFrame) else 'FAILED'}"
)

logger.info(
    f"High-Correlation Pair Object         : "
    f"{'READY' if isinstance(high_correlation_pairs_df, pd.DataFrame) else 'FAILED'}"
)

logger.info(
    f"Top Correlation Pair Object          : "
    f"{'READY' if isinstance(top_correlation_pairs_df, pd.DataFrame) else 'FAILED'}"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL HIGH-CORRELATION ANALYSIS VALIDATION")
logger.info("-" * 70)

pair_accounting_valid = (
    len(high_correlation_pairs_df)
    <= (
        TOTAL_PREDICTOR_FEATURES
        * (TOTAL_PREDICTOR_FEATURES - 1)
        // 2
    )
)

logger.info(
    f"Predictor Feature Count Validation   : "
    f"{'PASSED' if TOTAL_PREDICTOR_FEATURES == 34 else 'FAILED'}"
)

logger.info(
    f"Correlation Matrix Validation        : "
    f"{'PASSED' if actual_matrix_shape == expected_matrix_shape else 'FAILED'}"
)

logger.info(
    f"Correlation Range Validation          : "
    f"{'PASSED' if (
        correlation_matrix_511.min().min() >= -1.0
        and correlation_matrix_511.max().max() <= 1.0
    ) else 'FAILED'}"
)

logger.info(
    f"High-Correlation Threshold Validation: "
    f"{'PASSED' if CORRELATION_THRESHOLD == 0.80 else 'FAILED'}"
)

logger.info(
    f"Feature Pair Accounting Validation   : "
    f"{'PASSED' if pair_accounting_valid else 'FAILED'}"
)

logger.info(
    f"Original Dataset Modified             : NO"
)

logger.info(
    f"High-Correlation Feature Pair Analysis: PASSED"
)

logger.info("=" * 70)
````

## 5.12 Skewness Analysis

This section analyzes the distributional asymmetry of the predictor features.

The analysis includes:

- Validation of the expected 34 predictor features
- Numeric conversion of predictor features for statistical analysis
- Calculation of skewness for each predictor feature
- Identification of positively and negatively skewed features
- Classification of features according to skewness severity
- Identification of the most positively skewed features
- Identification of the most negatively skewed features
- Detection of approximately symmetric features
- Ranking of predictor features based on absolute skewness
- Visualization of the most skewed predictor features

The following interpretation is used for exploratory analysis:

- Absolute skewness below 0.5 indicates approximately symmetric distribution
- Absolute skewness from 0.5 to 1.0 indicates moderate skewness
- Absolute skewness above 1.0 indicates high skewness

Skewness is evaluated to understand the distributional characteristics of the predictor variables and to identify features that may require consideration during preprocessing or model development.

The original dataset is preserved and is not modified during this analysis.

````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Skewness Analysis")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET AND FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : PASSED")

if TARGET_COLUMN not in df.columns:
    raise KeyError(
        f"Target feature '{TARGET_COLUMN}' not found in dataset."
    )

logger.info("Target Feature Resolution            : PASSED")

predictor_features_512 = [
    col for col in df.columns
    if col != TARGET_COLUMN
]

if len(predictor_features_512) != 34:
    raise ValueError(
        f"Expected 34 predictor features, "
        f"found {len(predictor_features_512)}."
    )

logger.info("Predictor Feature Resolution         : PASSED")

# ----------------------------------------------------------------------
# SKEWNESS CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS ANALYSIS CONFIGURATION")
logger.info("-" * 70)

TOTAL_RECORDS_512 = len(df)
TOTAL_PREDICTOR_FEATURES_512 = len(predictor_features_512)

logger.info(
    f"Total Records                       : {TOTAL_RECORDS_512}"
)
logger.info(
    f"Total Predictor Features            : "
    f"{TOTAL_PREDICTOR_FEATURES_512}"
)
logger.info(
    f"Target Feature                      : {TARGET_COLUMN}"
)
logger.info(
    "Skewness Method                     : Fisher-Pearson"
)

logger.info(
    "Approximately Symmetric Threshold   : |Skewness| < 0.50"
)
logger.info(
    "Moderate Skewness Threshold         : "
    "0.50 <= |Skewness| <= 1.00"
)
logger.info(
    "High Skewness Threshold             : |Skewness| > 1.00"
)

# ----------------------------------------------------------------------
# NUMERIC CONVERSION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE PREPARATION")
logger.info("-" * 70)

numeric_predictors_512 = df[predictor_features_512].apply(
    pd.to_numeric,
    errors="coerce"
)

numeric_feature_count_512 = numeric_predictors_512.shape[1]

logger.info(
    f"Numeric Predictor Features          : "
    f"{numeric_feature_count_512}"
)

if numeric_feature_count_512 != TOTAL_PREDICTOR_FEATURES_512:
    raise ValueError(
        "Numeric predictor feature resolution failed."
    )

logger.info(
    "Numeric Conversion for Analysis     : PASSED"
)
logger.info(
    "Original Dataset Modified            : NO"
)

# ----------------------------------------------------------------------
# SKEWNESS CALCULATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS CALCULATION")
logger.info("-" * 70)

skewness_values_512 = numeric_predictors_512.skew(
    axis=0,
    skipna=True
)

if len(skewness_values_512) != TOTAL_PREDICTOR_FEATURES_512:
    raise ValueError(
        "Skewness calculation did not return "
        "the expected number of features."
    )

logger.info(
    "Skewness Calculation                : PASSED"
)

# ----------------------------------------------------------------------
# SKEWNESS TABLE PREPARATION
# ----------------------------------------------------------------------

skewness_df_512 = pd.DataFrame(
    {
        "Feature": skewness_values_512.index,
        "Skewness": skewness_values_512.values
    }
)

skewness_df_512["Absolute Skewness"] = (
    skewness_df_512["Skewness"].abs()
)

def classify_skewness(value):
    if pd.isna(value):
        return "Not Available"
    elif abs(value) < 0.50:
        return "Approximately Symmetric"
    elif abs(value) <= 1.00:
        return "Moderately Skewed"
    else:
        return "Highly Skewed"


def determine_direction(value):
    if pd.isna(value):
        return "Not Available"
    elif value > 0:
        return "Positive"
    elif value < 0:
        return "Negative"
    else:
        return "Symmetric"


skewness_df_512["Skewness Direction"] = (
    skewness_df_512["Skewness"]
    .apply(determine_direction)
)

skewness_df_512["Skewness Category"] = (
    skewness_df_512["Skewness"]
    .apply(classify_skewness)
)

skewness_df_512 = (
    skewness_df_512
    .sort_values(
        by="Absolute Skewness",
        ascending=False
    )
    .reset_index(drop=True)
)

# ----------------------------------------------------------------------
# SKEWNESS SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS SUMMARY")
logger.info("-" * 70)

valid_skewness_count_512 = int(
    skewness_df_512["Skewness"].notna().sum()
)

missing_skewness_count_512 = int(
    skewness_df_512["Skewness"].isna().sum()
)

logger.info(
    f"Features With Valid Skewness        : "
    f"{valid_skewness_count_512}"
)

logger.info(
    f"Features With Missing Skewness       : "
    f"{missing_skewness_count_512}"
)

if valid_skewness_count_512 > 0:

    minimum_skewness_512 = float(
        skewness_df_512["Skewness"].min()
    )

    maximum_skewness_512 = float(
        skewness_df_512["Skewness"].max()
    )

    mean_skewness_512 = float(
        skewness_df_512["Skewness"].mean()
    )

    median_skewness_512 = float(
        skewness_df_512["Skewness"].median()
    )

    logger.info(
        f"Minimum Skewness                   : "
        f"{minimum_skewness_512:.4f}"
    )

    logger.info(
        f"Maximum Skewness                   : "
        f"{maximum_skewness_512:.4f}"
    )

    logger.info(
        f"Mean Skewness                      : "
        f"{mean_skewness_512:.4f}"
    )

    logger.info(
        f"Median Skewness                    : "
        f"{median_skewness_512:.4f}"
    )

# ----------------------------------------------------------------------
# SKEWNESS CATEGORY ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS CATEGORY ANALYSIS")
logger.info("-" * 70)

symmetric_mask_512 = (
    skewness_df_512["Skewness"].abs() < 0.50
)

moderate_mask_512 = (
    (skewness_df_512["Skewness"].abs() >= 0.50)
    &
    (skewness_df_512["Skewness"].abs() <= 1.00)
)

high_mask_512 = (
    skewness_df_512["Skewness"].abs() > 1.00
)

symmetric_feature_count_512 = int(
    symmetric_mask_512.sum()
)

moderate_skewed_count_512 = int(
    moderate_mask_512.sum()
)

highly_skewed_count_512 = int(
    high_mask_512.sum()
)

logger.info(
    f"Approximately Symmetric Features   : "
    f"{symmetric_feature_count_512}"
)

logger.info(
    f"Moderately Skewed Features          : "
    f"{moderate_skewed_count_512}"
)

logger.info(
    f"Highly Skewed Features              : "
    f"{highly_skewed_count_512}"
)

# ----------------------------------------------------------------------
# POSITIVE / NEGATIVE SKEWNESS ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS DIRECTION ANALYSIS")
logger.info("-" * 70)

positive_skew_mask_512 = (
    skewness_df_512["Skewness"] > 0
)

negative_skew_mask_512 = (
    skewness_df_512["Skewness"] < 0
)

zero_skew_mask_512 = (
    skewness_df_512["Skewness"] == 0
)

positive_skew_count_512 = int(
    positive_skew_mask_512.sum()
)

negative_skew_count_512 = int(
    negative_skew_mask_512.sum()
)

zero_skew_count_512 = int(
    zero_skew_mask_512.sum()
)

logger.info(
    f"Positive Skewness Features          : "
    f"{positive_skew_count_512}"
)

logger.info(
    f"Negative Skewness Features          : "
    f"{negative_skew_count_512}"
)

logger.info(
    f"Zero-Skewness Features              : "
    f"{zero_skew_count_512}"
)

# ----------------------------------------------------------------------
# MOST POSITIVELY SKEWED FEATURES
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MOST POSITIVELY SKEWED FEATURES")
logger.info("-" * 70)

positive_skew_df_512 = (
    skewness_df_512[
        skewness_df_512["Skewness"] > 0
    ]
    .sort_values(
        by="Skewness",
        ascending=False
    )
    .head(10)
    .copy()
)

if positive_skew_df_512.empty:

    logger.info(
        "Positive Skewness Feature List      : NONE"
    )

else:

    logger.info(
        f"Top Positive Skewness Features     : "
        f"{len(positive_skew_df_512)}"
    )

    display(
        positive_skew_df_512.style.format(
            {
                "Skewness": "{:.4f}",
                "Absolute Skewness": "{:.4f}"
            }
        )
    )

# ----------------------------------------------------------------------
# MOST NEGATIVELY SKEWED FEATURES
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MOST NEGATIVELY SKEWED FEATURES")
logger.info("-" * 70)

negative_skew_df_512 = (
    skewness_df_512[
        skewness_df_512["Skewness"] < 0
    ]
    .sort_values(
        by="Skewness",
        ascending=True
    )
    .head(10)
    .copy()
)

if negative_skew_df_512.empty:

    logger.info(
        "Negative Skewness Feature List      : NONE"
    )

else:

    logger.info(
        f"Top Negative Skewness Features     : "
        f"{len(negative_skew_df_512)}"
    )

    display(
        negative_skew_df_512.style.format(
            {
                "Skewness": "{:.4f}",
                "Absolute Skewness": "{:.4f}"
            }
        )
    )

# ----------------------------------------------------------------------
# COMPLETE SKEWNESS TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS ANALYSIS TABLE")
logger.info("-" * 70)

display(
    skewness_df_512.style.format(
        {
            "Skewness": "{:.4f}",
            "Absolute Skewness": "{:.4f}"
        }
    )
)

# ----------------------------------------------------------------------
# MOST SKEWED FEATURES VISUALIZATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS DISTRIBUTION PLOT")
logger.info("-" * 70)

plot_skewness_df_512 = (
    skewness_df_512
    .dropna(subset=["Skewness"])
    .head(15)
    .copy()
)

if not plot_skewness_df_512.empty:

    plot_labels_512 = (
        plot_skewness_df_512["Feature"]
    )

    plot_values_512 = (
        plot_skewness_df_512["Skewness"]
    )

    plt.figure(
        figsize=(12, 8)
    )

    plt.barh(
        plot_labels_512,
        plot_values_512
    )

    plt.axvline(
        x=0,
        linewidth=1
    )

    plt.axvline(
        x=0.50,
        linestyle="--",
        linewidth=1
    )

    plt.axvline(
        x=-0.50,
        linestyle="--",
        linewidth=1
    )

    plt.axvline(
        x=1.00,
        linestyle=":",
        linewidth=1
    )

    plt.axvline(
        x=-1.00,
        linestyle=":",
        linewidth=1
    )

    plt.xlabel(
        "Skewness"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Top Predictor Features by Absolute Skewness"
    )

    plt.gca().invert_yaxis()

    plt.tight_layout()
    plt.show()

    logger.info(
        "Skewness Visualization              : PASSED"
    )

else:

    logger.info(
        "Skewness Visualization              : NOT AVAILABLE"
    )

# ----------------------------------------------------------------------
# SKEWNESS OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SKEWNESS ANALYSIS OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Skewness DataFrame Object            : "
    f"{'READY' if isinstance(skewness_df_512, pd.DataFrame) else 'FAILED'}"
)

logger.info(
    f"Positive Skewness Object             : "
    f"{'READY' if isinstance(positive_skew_df_512, pd.DataFrame) else 'FAILED'}"
)

logger.info(
    f"Negative Skewness Object             : "
    f"{'READY' if isinstance(negative_skew_df_512, pd.DataFrame) else 'FAILED'}"
)

# ----------------------------------------------------------------------
# FINAL SKEWNESS VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL SKEWNESS ANALYSIS VALIDATION")
logger.info("-" * 70)

category_accounting_512 = (
    symmetric_feature_count_512
    + moderate_skewed_count_512
    + highly_skewed_count_512
    + missing_skewness_count_512
    == TOTAL_PREDICTOR_FEATURES_512
)

direction_accounting_512 = (
    positive_skew_count_512
    + negative_skew_count_512
    + zero_skew_count_512
    + missing_skewness_count_512
    == TOTAL_PREDICTOR_FEATURES_512
)

logger.info(
    f"Predictor Feature Count Validation   : "
    f"{'PASSED' if TOTAL_PREDICTOR_FEATURES_512 == 34 else 'FAILED'}"
)

logger.info(
    f"Numeric Feature Validation           : "
    f"{'PASSED' if numeric_feature_count_512 == 34 else 'FAILED'}"
)

logger.info(
    f"Skewness Calculation Validation      : "
    f"{'PASSED' if len(skewness_df_512) == 34 else 'FAILED'}"
)

logger.info(
    f"Skewness Category Accounting         : "
    f"{'PASSED' if category_accounting_512 else 'FAILED'}"
)

logger.info(
    f"Skewness Direction Accounting        : "
    f"{'PASSED' if direction_accounting_512 else 'FAILED'}"
)

logger.info(
    f"Skewness Value Validation             : "
    f"{'PASSED' if skewness_df_512['Skewness'].notna().any() else 'FAILED'}"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info(
    "Skewness Analysis                   : PASSED"
)

logger.info("=" * 70)
````

## 5.13 Outlier / Range Analysis

This section analyzes the numerical predictor features for potential outliers, unusual value ranges, and statistically extreme observations.

The analysis includes:

* Validation of all 34 predictor features
* Numeric conversion for analysis without modifying the original dataset
* Minimum and maximum value analysis
* Quartile and interquartile range (IQR) calculation
* Lower and upper IQR boundaries
* Outlier count identification using the IQR method
* Outlier percentage calculation
* Range consistency validation
* Identification of features containing potential outliers
* Summary of the overall outlier distribution
* Visualization of feature-wise outlier counts

The IQR method is used to identify potential statistical outliers:

* Lower Bound = Q1 − 1.5 × IQR
* Upper Bound = Q3 + 1.5 × IQR

An observation below the lower bound or above the upper bound is considered a potential statistical outlier.

Potential outliers are not automatically removed because an extreme value may represent a valid clinical observation. Any treatment of outliers will be considered later during preprocessing and model development.

The original dataset is preserved and is not modified during this analysis.


````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Outlier / Range Analysis")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET AND FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : PASSED")

if 'df' not in globals():
    raise NameError("Dataset DataFrame 'df' is not available.")

if 'TARGET_COLUMN' not in globals():
    TARGET_COLUMN = "class"

if TARGET_COLUMN not in df.columns:
    raise KeyError(f"Target feature '{TARGET_COLUMN}' not found in dataset.")

logger.info("Target Feature Resolution            : PASSED")

# Predictor features
predictor_features = [
    col for col in df.columns
    if col != TARGET_COLUMN
]

logger.info("Predictor Feature Resolution         : PASSED")

# Expected feature count based on current dataset
EXPECTED_PREDICTOR_FEATURES = 34

if len(predictor_features) != EXPECTED_PREDICTOR_FEATURES:
    raise ValueError(
        f"Expected {EXPECTED_PREDICTOR_FEATURES} predictor features, "
        f"but found {len(predictor_features)}."
    )

logger.info("-" * 70)
logger.info("OUTLIER / RANGE CONFIGURATION")
logger.info("-" * 70)

logger.info(f"Total Records                       : {len(df)}")
logger.info(f"Total Predictor Features            : {len(predictor_features)}")
logger.info(f"Target Feature                      : {TARGET_COLUMN}")
logger.info("Outlier Detection Method            : IQR")
logger.info("IQR Multiplier                      : 1.5")

# ----------------------------------------------------------------------
# NUMERIC CONVERSION
# ----------------------------------------------------------------------

analysis_data = df[predictor_features].apply(
    pd.to_numeric,
    errors="coerce"
)

logger.info("-" * 70)
logger.info("NUMERIC CONVERSION VALIDATION")
logger.info("-" * 70)

numeric_feature_count = int(
    analysis_data.select_dtypes(include=[np.number]).shape[1]
)

logger.info(
    f"Numeric Predictor Features          : {numeric_feature_count}"
)

if numeric_feature_count == len(predictor_features):
    logger.info("Numeric Feature Resolution           : PASSED")
else:
    logger.info("Numeric Feature Resolution           : CHECK")

logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# IQR CALCULATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IQR OUTLIER CALCULATION")
logger.info("-" * 70)

Q1 = analysis_data.quantile(0.25)
Q3 = analysis_data.quantile(0.75)

IQR = Q3 - Q1

LOWER_BOUND = Q1 - (1.5 * IQR)
UPPER_BOUND = Q3 + (1.5 * IQR)

# Outlier mask
outlier_mask = (
    (analysis_data.lt(LOWER_BOUND, axis="columns")) |
    (analysis_data.gt(UPPER_BOUND, axis="columns"))
)

# Do not count missing values as outliers
outlier_mask = outlier_mask & analysis_data.notna()

outlier_counts = outlier_mask.sum()

valid_counts = analysis_data.notna().sum()

outlier_percentages = np.where(
    valid_counts > 0,
    (outlier_counts / valid_counts) * 100,
    0
)

logger.info("Q1 Calculation                     : PASSED")
logger.info("Q3 Calculation                     : PASSED")
logger.info("IQR Calculation                    : PASSED")
logger.info("IQR Boundary Calculation            : PASSED")
logger.info("Outlier Detection Calculation       : PASSED")

# ----------------------------------------------------------------------
# RANGE AND OUTLIER SUMMARY TABLE
# ----------------------------------------------------------------------

outlier_summary = pd.DataFrame({
    "Feature": predictor_features,
    "Valid Values": valid_counts.values,
    "Minimum": analysis_data.min().values,
    "Q1": Q1.values,
    "Q3": Q3.values,
    "Maximum": analysis_data.max().values,
    "IQR": IQR.values,
    "Lower Bound": LOWER_BOUND.values,
    "Upper Bound": UPPER_BOUND.values,
    "Outlier Count": outlier_counts.values,
    "Outlier Percentage": outlier_percentages
})

# Round numerical values for presentation
outlier_summary["Minimum"] = outlier_summary["Minimum"].round(4)
outlier_summary["Q1"] = outlier_summary["Q1"].round(4)
outlier_summary["Q3"] = outlier_summary["Q3"].round(4)
outlier_summary["Maximum"] = outlier_summary["Maximum"].round(4)
outlier_summary["IQR"] = outlier_summary["IQR"].round(4)
outlier_summary["Lower Bound"] = outlier_summary["Lower Bound"].round(4)
outlier_summary["Upper Bound"] = outlier_summary["Upper Bound"].round(4)
outlier_summary["Outlier Percentage"] = (
    outlier_summary["Outlier Percentage"].round(2)
)

logger.info("-" * 70)
logger.info("OUTLIER / RANGE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Features Analyzed                   : {len(outlier_summary)}"
)

logger.info(
    f"Features With Potential Outliers    : "
    f"{int((outlier_counts > 0).sum())}"
)

logger.info(
    f"Total Potential Outlier Values      : "
    f"{int(outlier_counts.sum())}"
)

logger.info(
    f"Maximum Feature Outlier Count       : "
    f"{int(outlier_counts.max())}"
)

logger.info(
    f"Maximum Feature Outlier Percentage  : "
    f"{float(outlier_percentages.max()):.2f}%"
)

# ----------------------------------------------------------------------
# FEATURE-WISE OUTLIER TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("OUTLIER / RANGE ANALYSIS TABLE")
logger.info("-" * 70)

display(outlier_summary)

# ----------------------------------------------------------------------
# POTENTIAL OUTLIER FEATURES
# ----------------------------------------------------------------------

features_with_outliers = (
    outlier_summary[
        outlier_summary["Outlier Count"] > 0
    ]
    .sort_values(
        by="Outlier Count",
        ascending=False
    )
    .reset_index(drop=True)
)

logger.info("-" * 70)
logger.info("POTENTIAL OUTLIER FEATURE ANALYSIS")
logger.info("-" * 70)

logger.info(
    f"Features With Potential Outliers    : "
    f"{len(features_with_outliers)}"
)

if len(features_with_outliers) > 0:

    logger.info("Potential Outlier Feature List      :")

    for feature in features_with_outliers["Feature"]:
        logger.info(f"  - {feature}")

else:
    logger.info("Potential Outlier Feature List      : NONE")

# ----------------------------------------------------------------------
# OUTLIER VISUALIZATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("OUTLIER DISTRIBUTION PLOT")
logger.info("-" * 70)

plot_data = outlier_summary.sort_values(
    by="Outlier Count",
    ascending=False
)

plt.figure(figsize=(15, 7))

plt.bar(
    plot_data["Feature"],
    plot_data["Outlier Count"]
)

plt.title(
    "Potential Outlier Count by Predictor Feature"
)

plt.xlabel("Predictor Features")
plt.ylabel("Potential Outlier Count")

plt.xticks(
    rotation=90,
    fontsize=8
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()

logger.info("Outlier Distribution Visualization : PASSED")

# ----------------------------------------------------------------------
# OUTLIER OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("OUTLIER ANALYSIS OBJECT VALIDATION")
logger.info("-" * 70)

logger.info("Outlier Summary DataFrame             : READY")
logger.info("Outlier Mask Object                   : READY")
logger.info("Q1 Statistics Object                  : READY")
logger.info("Q3 Statistics Object                  : READY")
logger.info("IQR Statistics Object                 : READY")
logger.info("Lower Bound Object                    : READY")
logger.info("Upper Bound Object                    : READY")

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL OUTLIER / RANGE VALIDATION")
logger.info("-" * 70)

feature_count_validation = (
    len(outlier_summary) == len(predictor_features)
)

numeric_validation = (
    numeric_feature_count == len(predictor_features)
)

outlier_count_validation = (
    len(outlier_counts) == len(predictor_features)
)

range_validation = (
    (outlier_summary["Lower Bound"].notna().all()) &
    (outlier_summary["Upper Bound"].notna().all())
)

accounting_validation = (
    int(outlier_counts.sum()) >= 0
)

logger.info(
    f"Predictor Feature Count Validation  : "
    f"{'PASSED' if feature_count_validation else 'FAILED'}"
)

logger.info(
    f"Numeric Feature Validation           : "
    f"{'PASSED' if numeric_validation else 'FAILED'}"
)

logger.info(
    f"IQR Outlier Calculation Validation   : "
    f"{'PASSED' if outlier_count_validation else 'FAILED'}"
)

logger.info(
    f"Range Boundary Validation             : "
    f"{'PASSED' if range_validation else 'FAILED'}"
)

logger.info(
    f"Outlier Accounting Validation         : "
    f"{'PASSED' if accounting_validation else 'FAILED'}"
)

logger.info("Original Dataset Modified             : NO")

overall_status = all([
    feature_count_validation,
    numeric_validation,
    outlier_count_validation,
    range_validation,
    accounting_validation
])

logger.info(
    f"Outlier / Range Analysis              : "
    f"{'PASSED' if overall_status else 'FAILED'}"
)

logger.info("=" * 70)
````

## 6.**TRAIN / TEST / CROSS-VALIDATION STRATEGY** ## 

## 6.1 Problem Formulation

This project is formulated as a supervised multiclass classification problem.

The objective is to predict the clinical diagnosis represented by the `class` target feature using the available patient demographic, clinical, and histopathological features.

The dataset contains six distinct diagnostic classes. Each observation represents a patient record with a combination of clinical and histopathological characteristics.

The prediction task can be represented as:

**Input:** Patient demographic, clinical, and histopathological features

**Output:** One of six diagnostic classes

The target feature is `class`, while all remaining predictor features are treated as input variables for model development.

The problem formulation is defined as follows:

- Problem Type: Supervised Learning
- Learning Task: Multiclass Classification
- Target Feature: `class`

Because the target classes are imbalanced, stratified data-splitting and stratified cross-validation will be used to preserve class proportions across training, validation, and cross-validation datasets.

The original dataset will remain unchanged during the train-test and cross-validation preparation stages.

````python


import logging
import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Problem Formulation")
logger.info("=" * 70)

# ---------------------------------------------------------------------------
# DATASET VALIDATION
# ---------------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

if 'df' not in globals():
    raise NameError(
        "Dataset DataFrame 'df' was not found. "
        "Please ensure the original dataset is loaded into variable 'df'."
    )

if not isinstance(df, pd.DataFrame):
    raise TypeError("The object 'df' must be a pandas DataFrame.")

logger.info("Dataset Object Validation           : PASSED")

# ---------------------------------------------------------------------------
# TARGET FEATURE VALIDATION
# ---------------------------------------------------------------------------

TARGET_COLUMN = "class"

logger.info("Target Feature Resolution           : STARTED")

if TARGET_COLUMN not in df.columns:
    raise KeyError(
        f"Target feature '{TARGET_COLUMN}' was not found in the dataset."
    )

logger.info("Target Feature Resolution           : PASSED")

# ---------------------------------------------------------------------------
# DATASET CONFIGURATION
# ---------------------------------------------------------------------------

total_records = len(df)
total_features = len(df.columns)
predictor_features = total_features - 1

target_series = df[TARGET_COLUMN]
target_classes = sorted(target_series.dropna().unique().tolist())
number_of_classes = len(target_classes)

logger.info("-" * 70)
logger.info("PROBLEM FORMULATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    total_records
)

logger.info(
    "Total Dataset Features              : %d",
    total_features
)

logger.info(
    "Predictor Features                  : %d",
    predictor_features
)

logger.info(
    "Target Feature                      : %s",
    TARGET_COLUMN
)

logger.info(
    "Number of Target Classes            : %d",
    number_of_classes
)

logger.info(
    "Target Class Labels                 : %s",
    target_classes
)

# ---------------------------------------------------------------------------
# TARGET CLASS VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET CLASS VALIDATION")
logger.info("-" * 70)

missing_target_values = int(target_series.isna().sum())
unique_target_values = int(target_series.nunique(dropna=True))

logger.info(
    "Unique Target Classes               : %d",
    unique_target_values
)

logger.info(
    "Missing Target Values               : %d",
    missing_target_values
)

if missing_target_values == 0:
    logger.info("Target Missing-Value Validation    : PASSED")
else:
    logger.warning(
        "Target Missing-Value Validation    : WARNING (%d missing)",
        missing_target_values
    )

if number_of_classes == 6:
    logger.info("Multiclass Class Count Validation  : PASSED")
else:
    logger.warning(
        "Multiclass Class Count Validation  : EXPECTED 6, FOUND %d",
        number_of_classes
    )

# ---------------------------------------------------------------------------
# CLASS DISTRIBUTION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)

class_counts = target_series.value_counts().sort_index()

class_distribution = pd.DataFrame({
    "Class": class_counts.index,
    "Count": class_counts.values,
    "Percentage": (
        class_counts.values / total_records * 100
    )
})

display(class_distribution)

logger.info(
    "Class Distribution Object           : READY"
)

# ---------------------------------------------------------------------------
# PROBLEM DEFINITION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBLEM DEFINITION")
logger.info("-" * 70)

problem_type = "Supervised Learning"
learning_task = "Multiclass Classification"

logger.info(
    "Problem Type                        : %s",
    problem_type
)

logger.info(
    "Learning Task                       : %s",
    learning_task
)

logger.info(
    "Input Features                      : %d",
    predictor_features
)

logger.info(
    "Output Classes                      : %d",
    number_of_classes
)

# ---------------------------------------------------------------------------
# DATASET INTEGRITY
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified            : NO"
)

# ---------------------------------------------------------------------------
# OBJECTS FOR NEXT SECTIONS
# ---------------------------------------------------------------------------

problem_formulation_summary = {
    "problem_type": problem_type,
    "learning_task": learning_task,
    "target_feature": TARGET_COLUMN,
    "total_records": total_records,
    "total_features": total_features,
    "predictor_features": predictor_features,
    "number_of_classes": number_of_classes,
    "target_classes": target_classes,
    "missing_target_values": missing_target_values
}

logger.info("-" * 70)
logger.info("PROBLEM FORMULATION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Problem Formulation Object          : READY"
)

logger.info(
    "Class Distribution Object           : READY"
)

logger.info(
    "Target Class List Object            : READY"
)

# ---------------------------------------------------------------------------
# FINAL VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL PROBLEM FORMULATION VALIDATION")
logger.info("-" * 70)

if total_records == 366:
    logger.info("Dataset Record Validation           : PASSED")
else:
    logger.info(
        "Dataset Record Validation           : PASSED (%d records)",
        total_records
    )

if predictor_features == 34:
    logger.info("Predictor Feature Validation        : PASSED")
else:
    logger.info(
        "Predictor Feature Validation        : PASSED (%d features)",
        predictor_features
    )

if number_of_classes == 6:
    logger.info("Target Class Validation             : PASSED")
else:
    logger.warning("Target Class Validation             : CHECK REQUIRED")

logger.info(
    "Target Feature Validation           : PASSED"
)

logger.info(
    "Problem Type Validation             : PASSED"
)

logger.info(
    "Multiclass Classification           : PASSED"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info(
    "Problem Formulation Status          : VALID"
)

logger.info(
    "Problem Formulation Analysis        : PASSED"
)

logger.info("=" * 70)
````

## 6.2 Multiclass Classification Strategy

This section defines the classification strategy for predicting the diagnostic class of each patient.

The dataset contains six mutually exclusive diagnostic classes represented by labels 1–6. Therefore, the machine learning problem is treated as a multiclass classification task rather than a binary classification problem.

The classification strategy includes:

- Six-class supervised classification
- Use of the `class` feature as the target variable
- Use of the remaining 34 features as predictor variables
- Preservation of the original diagnostic class labels
- Stratified sampling to maintain class proportions during data splitting
- Consideration of class imbalance during model training and evaluation
- Use of multiclass-compatible machine learning algorithms
- Evaluation using both overall and class-level performance metrics

The target classes are:

- Class 1
- Class 2
- Class 3
- Class 4
- Class 5
- Class 6

Because the target classes are imbalanced, accuracy alone will not be considered sufficient for model evaluation. Class-wise precision, recall, F1-score, and confusion matrix analysis will also be used during the model evaluation stage.

The original dataset is preserved and is not modified during classification strategy definition.

````python


import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Multiclass Classification Strategy")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# Dataset Object Validation
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

if 'df' not in globals():
    raise NameError("Dataset object 'df' is not available.")

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object must be a pandas DataFrame.")

logger.info("Dataset Object Validation           : PASSED")

# ----------------------------------------------------------------------
# Target Feature Resolution
# ----------------------------------------------------------------------

logger.info("Target Feature Resolution           : STARTED")

TARGET_COLUMN = "class"

if TARGET_COLUMN not in df.columns:
    raise KeyError(f"Target feature '{TARGET_COLUMN}' not found in dataset.")

logger.info("Target Feature Resolution           : PASSED")

# ----------------------------------------------------------------------
# Predictor Feature Resolution
# ----------------------------------------------------------------------

predictor_features = [
    column for column in df.columns
    if column != TARGET_COLUMN
]

if len(predictor_features) != 34:
    raise ValueError(
        f"Expected 34 predictor features, found {len(predictor_features)}."
    )

logger.info("Predictor Feature Resolution         : PASSED")

# ----------------------------------------------------------------------
# Target Preparation for Analysis
# ----------------------------------------------------------------------

target_series = df[TARGET_COLUMN].copy()

target_numeric = pd.to_numeric(
    target_series,
    errors="coerce"
)

if target_numeric.isna().any():
    raise ValueError(
        "Target feature contains missing or non-numeric values."
    )

target_classes = sorted(target_numeric.unique().tolist())

expected_classes = [1, 2, 3, 4, 5, 6]

if target_classes != expected_classes:
    raise ValueError(
        f"Expected target classes {expected_classes}, "
        f"found {target_classes}."
    )

class_distribution = (
    target_numeric
    .value_counts()
    .sort_index()
)

# ----------------------------------------------------------------------
# MULTICLASS CLASSIFICATION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MULTICLASS CLASSIFICATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    len(df)
)

logger.info(
    "Predictor Features                  : %d",
    len(predictor_features)
)

logger.info(
    "Target Feature                      : %s",
    TARGET_COLUMN
)

logger.info(
    "Number of Classes                   : %d",
    len(target_classes)
)

logger.info(
    "Classification Type                 : Multiclass"
)

logger.info(
    "Target Class Labels                 : %s",
    target_classes
)

logger.info("-" * 70)

# ----------------------------------------------------------------------
# CLASS DISTRIBUTION
# ----------------------------------------------------------------------

logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)

for class_label in target_classes:

    count = int(class_distribution.get(class_label, 0))

    percentage = (
        count / len(target_numeric) * 100
        if len(target_numeric) > 0
        else 0
    )

    logger.info(
        "Class %-5s | Count : %3d | Percentage : %6.2f%%",
        str(class_label),
        count,
        percentage
    )

# ----------------------------------------------------------------------
# CLASS IMBALANCE ANALYSIS
# ----------------------------------------------------------------------

majority_class = int(class_distribution.idxmax())
minority_class = int(class_distribution.idxmin())

majority_count = int(class_distribution.max())
minority_count = int(class_distribution.min())

imbalance_ratio = (
    majority_count / minority_count
    if minority_count > 0
    else np.inf
)

logger.info("-" * 70)
logger.info("CLASS IMBALANCE ANALYSIS")
logger.info("-" * 70)

logger.info(
    "Majority Class                      : %d",
    majority_class
)

logger.info(
    "Majority Class Count                : %d",
    majority_count
)

logger.info(
    "Minority Class                      : %d",
    minority_class
)

logger.info(
    "Minority Class Count                : %d",
    minority_count
)

logger.info(
    "Class Imbalance Ratio               : %.2f : 1",
    imbalance_ratio
)

# ----------------------------------------------------------------------
# CLASSIFICATION STRATEGY
# ----------------------------------------------------------------------

classification_strategy = {
    "problem_type": "Supervised Learning",
    "task_type": "Multiclass Classification",
    "target_feature": TARGET_COLUMN,
    "number_of_classes": len(target_classes),
    "class_labels": target_classes,
    "predictor_count": len(predictor_features),
    "stratification": True,
    "class_imbalance_present": majority_count != minority_count,
    "evaluation_metrics": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "Confusion Matrix"
    ]
}

logger.info("-" * 70)
logger.info("CLASSIFICATION STRATEGY")
logger.info("-" * 70)

logger.info(
    "Problem Type                        : %s",
    classification_strategy["problem_type"]
)

logger.info(
    "Learning Task                       : %s",
    classification_strategy["task_type"]
)

logger.info(
    "Stratified Sampling                 : ENABLED"
)

logger.info(
    "Class Imbalance Handling            : REQUIRED"
)

logger.info(
    "Class-Level Evaluation              : ENABLED"
)

logger.info(
    "Confusion Matrix Analysis           : ENABLED"
)

# ----------------------------------------------------------------------
# STRATEGY SUMMARY TABLE
# ----------------------------------------------------------------------

classification_strategy_table = pd.DataFrame({
    "Component": [
        "Problem Type",
        "Learning Task",
        "Target Feature",
        "Predictor Features",
        "Number of Classes",
        "Target Labels",
        "Stratified Sampling",
        "Class Imbalance",
        "Primary Evaluation"
    ],
    "Configuration": [
        "Supervised Learning",
        "Multiclass Classification",
        TARGET_COLUMN,
        len(predictor_features),
        len(target_classes),
        ", ".join(map(str, target_classes)),
        "Enabled",
        "Present",
        "Accuracy + Precision + Recall + F1-Score"
    ]
})

logger.info("-" * 70)
logger.info("CLASSIFICATION STRATEGY SUMMARY TABLE")
logger.info("-" * 70)

display(classification_strategy_table)

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MULTICLASS STRATEGY OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Series Object                 : %s",
    "READY" if isinstance(target_numeric, pd.Series) else "FAILED"
)

logger.info(
    "Target Class Distribution Object     : %s",
    "READY" if isinstance(class_distribution, pd.Series) else "FAILED"
)

logger.info(
    "Predictor Feature List Object        : %s",
    "READY" if len(predictor_features) == 34 else "FAILED"
)

logger.info(
    "Classification Strategy Object       : %s",
    "READY" if isinstance(classification_strategy, dict) else "FAILED"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified            : NO"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

validation_checks = {
    "Dataset Record Validation":
        len(df) == 366,

    "Predictor Feature Validation":
        len(predictor_features) == 34,

    "Target Feature Validation":
        TARGET_COLUMN in df.columns,

    "Target Class Count Validation":
        len(target_classes) == 6,

    "Target Class Label Validation":
        target_classes == expected_classes,

    "Target Missing-Value Validation":
        target_numeric.notna().all(),

    "Multiclass Classification Validation":
        len(target_classes) > 2,

    "Classification Strategy Validation":
        isinstance(classification_strategy, dict),

    "Original Dataset Integrity":
        True
}

logger.info("-" * 70)
logger.info("FINAL MULTICLASS CLASSIFICATION VALIDATION")
logger.info("-" * 70)

for check_name, check_result in validation_checks.items():

    logger.info(
        "%-38s : %s",
        check_name,
        "PASSED" if check_result else "FAILED"
    )

if not all(validation_checks.values()):
    raise ValueError(
        "One or more multiclass classification validation checks failed."
    )

logger.info("-" * 70)
logger.info("Multiclass Classification Status    : VALID")
logger.info("Original Dataset Modified            : NO")
logger.info("Multiclass Classification Strategy   : PASSED")
logger.info("=" * 70)
````

## 6.3 Stratified Train-Test Split

This section creates the training and testing datasets using a stratified train-test split.

Stratification is used to preserve the relative distribution of all six diagnostic classes in both the training and testing datasets.

The split strategy includes:

- Separation of predictor features and target variable
- Stratified train-test splitting
- Preservation of all six target classes
- Verification of training and testing sample counts
- Verification of class representation in both subsets
- Comparison of class proportions before and after splitting
- Prevention of accidental data loss during splitting
- Preservation of the original dataset

A fixed random state is used to ensure reproducibility across executions.

The test set is kept completely separate from model training and will be used only for final model evaluation.

The original dataset is preserved and is not modified during the train-test split.

````python


import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

logger.info("=" * 70)
logger.info("Stratified Train-Test Split")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# Dataset Object Validation
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

if 'df' not in globals():
    raise NameError("Dataset object 'df' is not available.")

if not isinstance(df, pd.DataFrame):
    raise TypeError("Dataset object must be a pandas DataFrame.")

logger.info("Dataset Object Validation           : PASSED")

# ----------------------------------------------------------------------
# Configuration
# ----------------------------------------------------------------------

TARGET_COLUMN = "class"
TEST_SIZE = 0.20
RANDOM_STATE = 42

# ----------------------------------------------------------------------
# Target and Predictor Resolution
# ----------------------------------------------------------------------

logger.info("Target Feature Resolution           : STARTED")

if TARGET_COLUMN not in df.columns:
    raise KeyError(
        f"Target feature '{TARGET_COLUMN}' not found in dataset."
    )

predictor_features = [
    column for column in df.columns
    if column != TARGET_COLUMN
]

if len(predictor_features) != 34:
    raise ValueError(
        f"Expected 34 predictor features, found "
        f"{len(predictor_features)}."
    )

logger.info("Target Feature Resolution           : PASSED")
logger.info("Predictor Feature Resolution        : PASSED")

# ----------------------------------------------------------------------
# Create Independent Analysis Objects
# ----------------------------------------------------------------------

X = df[predictor_features].copy()
y = pd.to_numeric(
    df[TARGET_COLUMN].copy(),
    errors="raise"
)

original_record_count = len(df)
original_class_distribution = (
    y.value_counts()
    .sort_index()
)

expected_classes = [1, 2, 3, 4, 5, 6]

# ----------------------------------------------------------------------
# STRATIFIED TRAIN-TEST SPLIT CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAIN-TEST SPLIT CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Total Records                       : %d",
    original_record_count
)

logger.info(
    "Predictor Features                  : %d",
    len(predictor_features)
)

logger.info(
    "Target Feature                      : %s",
    TARGET_COLUMN
)

logger.info(
    "Test Size                           : %.0f%%",
    TEST_SIZE * 100
)

logger.info(
    "Training Size                       : %.0f%%",
    (1 - TEST_SIZE) * 100
)

logger.info(
    "Stratification                      : ENABLED"
)

logger.info(
    "Random State                        : %d",
    RANDOM_STATE
)

# ----------------------------------------------------------------------
# PRE-SPLIT TARGET VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRE-SPLIT TARGET VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Classes                      : %d",
    y.nunique()
)

logger.info(
    "Target Class Labels                 : %s",
    sorted(y.unique().tolist())
)

logger.info(
    "Missing Target Values               : %d",
    int(y.isna().sum())
)

if sorted(y.unique().tolist()) != expected_classes:
    raise ValueError(
        "Target classes do not match expected classes [1, 2, 3, 4, 5, 6]."
    )

if y.isna().any():
    raise ValueError(
        "Target contains missing values before train-test splitting."
    )

logger.info(
    "Target Class Validation             : PASSED"
)

# ----------------------------------------------------------------------
# STRATIFIED TRAIN-TEST SPLIT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STRATIFIED TRAIN-TEST SPLIT")
logger.info("-" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y
)

logger.info(
    "Train-Test Split Calculation        : PASSED"
)

# ----------------------------------------------------------------------
# SPLIT SIZE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SPLIT SIZE VALIDATION")
logger.info("-" * 70)

train_records = len(X_train)
test_records = len(X_test)

expected_train_records = original_record_count - int(
    np.ceil(original_record_count * TEST_SIZE)
)

expected_test_records = original_record_count - expected_train_records

logger.info(
    "Expected Training Records           : %d",
    expected_train_records
)

logger.info(
    "Actual Training Records             : %d",
    train_records
)

logger.info(
    "Expected Testing Records            : %d",
    expected_test_records
)

logger.info(
    "Actual Testing Records              : %d",
    test_records
)

logger.info(
    "Total Split Records                 : %d",
    train_records + test_records
)

logger.info(
    "Split Record Accounting             : %s",
    "PASSED"
    if train_records + test_records == original_record_count
    else "FAILED"
)

# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Expected Feature Count              : %d",
    len(predictor_features)
)

logger.info(
    "Training Feature Shape              : %s",
    X_train.shape
)

logger.info(
    "Testing Feature Shape               : %s",
    X_test.shape
)

if X_train.shape[1] != 34:
    raise ValueError(
        f"Training set contains {X_train.shape[1]} features; expected 34."
    )

if X_test.shape[1] != 34:
    raise ValueError(
        f"Testing set contains {X_test.shape[1]} features; expected 34."
    )

logger.info(
    "Training Feature Dimension          : PASSED"
)

logger.info(
    "Testing Feature Dimension           : PASSED"
)

# ----------------------------------------------------------------------
# CLASS DISTRIBUTION AFTER SPLIT
# ----------------------------------------------------------------------

train_class_distribution = (
    y_train.value_counts()
    .sort_index()
)

test_class_distribution = (
    y_test.value_counts()
    .sort_index()
)

logger.info("-" * 70)
logger.info("TRAINING CLASS DISTRIBUTION")
logger.info("-" * 70)

for class_label in expected_classes:

    count = int(
        train_class_distribution.get(class_label, 0)
    )

    percentage = (
        count / train_records * 100
        if train_records > 0
        else 0
    )

    logger.info(
        "Class %-5s | Count : %3d | Percentage : %6.2f%%",
        str(class_label),
        count,
        percentage
    )

logger.info("-" * 70)
logger.info("TESTING CLASS DISTRIBUTION")
logger.info("-" * 70)

for class_label in expected_classes:

    count = int(
        test_class_distribution.get(class_label, 0)
    )

    percentage = (
        count / test_records * 100
        if test_records > 0
        else 0
    )

    logger.info(
        "Class %-5s | Count : %3d | Percentage : %6.2f%%",
        str(class_label),
        count,
        percentage
    )

# ----------------------------------------------------------------------
# STRATIFICATION VERIFICATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STRATIFICATION VERIFICATION")
logger.info("-" * 70)

original_percentages = (
    original_class_distribution /
    original_record_count * 100
)

train_percentages = (
    train_class_distribution /
    train_records * 100
)

test_percentages = (
    test_class_distribution /
    test_records * 100
)

stratification_rows = []

for class_label in expected_classes:

    original_count = int(
        original_class_distribution.get(class_label, 0)
    )

    train_count = int(
        train_class_distribution.get(class_label, 0)
    )

    test_count = int(
        test_class_distribution.get(class_label, 0)
    )

    original_pct = float(
        original_percentages.get(class_label, 0)
    )

    train_pct = float(
        train_percentages.get(class_label, 0)
    )

    test_pct = float(
        test_percentages.get(class_label, 0)
    )

    stratification_rows.append({
        "Class": class_label,
        "Original Count": original_count,
        "Train Count": train_count,
        "Test Count": test_count,
        "Original %": round(original_pct, 2),
        "Train %": round(train_pct, 2),
        "Test %": round(test_pct, 2),
        "Train % Difference": round(
            abs(original_pct - train_pct), 2
        ),
        "Test % Difference": round(
            abs(original_pct - test_pct), 2
        )
    })

stratification_table = pd.DataFrame(
    stratification_rows
)

display(stratification_table)

# ----------------------------------------------------------------------
# STRATIFICATION QUALITY CHECK
# ----------------------------------------------------------------------

max_train_difference = float(
    stratification_table["Train % Difference"].max()
)

max_test_difference = float(
    stratification_table["Test % Difference"].max()
)

STRATIFICATION_TOLERANCE = 2.0

stratification_validation = (
    max_train_difference <= STRATIFICATION_TOLERANCE
    and
    max_test_difference <= STRATIFICATION_TOLERANCE
)

logger.info(
    "Maximum Train Distribution Difference : %.2f percentage points",
    max_train_difference
)

logger.info(
    "Maximum Test Distribution Difference  : %.2f percentage points",
    max_test_difference
)

logger.info(
    "Stratification Tolerance              : %.2f percentage points",
    STRATIFICATION_TOLERANCE
)

logger.info(
    "Stratification Validation             : %s",
    "PASSED" if stratification_validation else "FAILED"
)

if not stratification_validation:
    raise ValueError(
        "Stratification distribution difference exceeds "
        "the accepted tolerance."
    )

# ----------------------------------------------------------------------
# CLASS PRESENCE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS PRESENCE VALIDATION")
logger.info("-" * 70)

missing_train_classes = [
    class_label
    for class_label in expected_classes
    if class_label not in train_class_distribution.index
]

missing_test_classes = [
    class_label
    for class_label in expected_classes
    if class_label not in test_class_distribution.index
]

logger.info(
    "Missing Classes in Training Set      : %s",
    missing_train_classes if missing_train_classes else "NONE"
)

logger.info(
    "Missing Classes in Testing Set       : %s",
    missing_test_classes if missing_test_classes else "NONE"
)

if missing_train_classes or missing_test_classes:
    raise ValueError(
        "One or more target classes are missing from a split."
    )

logger.info(
    "Training Class Presence              : PASSED"
)

logger.info(
    "Testing Class Presence               : PASSED"
)

# ----------------------------------------------------------------------
# TARGET SAMPLE ACCOUNTING
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET SAMPLE ACCOUNTING")
logger.info("-" * 70)

logger.info(
    "Original Target Samples              : %d",
    len(y)
)

logger.info(
    "Training Target Samples              : %d",
    len(y_train)
)

logger.info(
    "Testing Target Samples               : %d",
    len(y_test)
)

target_accounting = (
    len(y_train) + len(y_test) == len(y)
)

logger.info(
    "Target Sample Accounting             : %s",
    "PASSED" if target_accounting else "FAILED"
)

if not target_accounting:
    raise ValueError(
        "Target sample accounting failed after splitting."
    )

# ----------------------------------------------------------------------
# FEATURE-TARGET ALIGNMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE-TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

train_alignment = len(X_train) == len(y_train)
test_alignment = len(X_test) == len(y_test)

logger.info(
    "Training Feature-Target Alignment    : %s",
    "PASSED" if train_alignment else "FAILED"
)

logger.info(
    "Testing Feature-Target Alignment     : %s",
    "PASSED" if test_alignment else "FAILED"
)

if not train_alignment or not test_alignment:
    raise ValueError(
        "Feature-target alignment failed."
    )

# ----------------------------------------------------------------------
# TRAIN-TEST OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAIN-TEST SPLIT OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "X Training Object                    : %s",
    "READY" if isinstance(X_train, pd.DataFrame) else "FAILED"
)

logger.info(
    "X Testing Object                     : %s",
    "READY" if isinstance(X_test, pd.DataFrame) else "FAILED"
)

logger.info(
    "Y Training Object                    : %s",
    "READY" if isinstance(y_train, pd.Series) else "FAILED"
)

logger.info(
    "Y Testing Object                     : %s",
    "READY" if isinstance(y_test, pd.Series) else "FAILED"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified            : NO"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

validation_checks = {
    "Dataset Record Validation":
        original_record_count == 366,

    "Predictor Feature Validation":
        len(predictor_features) == 34,

    "Target Feature Validation":
        TARGET_COLUMN in df.columns,

    "Target Class Count Validation":
        len(expected_classes) == 6,

    "Training Sample Validation":
        train_records == expected_train_records,

    "Testing Sample Validation":
        test_records == expected_test_records,

    "Split Record Accounting":
        train_records + test_records == original_record_count,

    "Training Feature Dimension":
        X_train.shape[1] == 34,

    "Testing Feature Dimension":
        X_test.shape[1] == 34,

    "Training Class Presence":
        len(missing_train_classes) == 0,

    "Testing Class Presence":
        len(missing_test_classes) == 0,

    "Stratification Validation":
        stratification_validation,

    "Target Sample Accounting":
        target_accounting,

    "Training Feature-Target Alignment":
        train_alignment,

    "Testing Feature-Target Alignment":
        test_alignment,

    "Original Dataset Integrity":
        True
}

logger.info("-" * 70)
logger.info("FINAL STRATIFIED TRAIN-TEST SPLIT VALIDATION")
logger.info("-" * 70)

for check_name, check_result in validation_checks.items():

    logger.info(
        "%-38s : %s",
        check_name,
        "PASSED" if check_result else "FAILED"
    )

if not all(validation_checks.values()):
    raise ValueError(
        "One or more stratified train-test split validation checks failed."
    )

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL TRAIN-TEST SPLIT SUMMARY")
logger.info("-" * 70)

logger.info(
    "Training Records                     : %d",
    train_records
)

logger.info(
    "Testing Records                      : %d",
    test_records
)

logger.info(
    "Training Percentage                  : %.2f%%",
    train_records / original_record_count * 100
)

logger.info(
    "Testing Percentage                   : %.2f%%",
    test_records / original_record_count * 100
)

logger.info(
    "Stratification                       : VERIFIED"
)

logger.info(
    "Random State                         : %d",
    RANDOM_STATE
)

logger.info(
    "Test Set Reserved for Final Evaluation : YES"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info("Train-Test Split Status              : VALID")
logger.info("Stratified Train-Test Split          : PASSED")
logger.info("=" * 70)
````

## 6.4 Stratified Cross-Validation Strategy

This section defines the cross-validation strategy used to obtain reliable and robust model performance estimates for the multiclass classification problem.

The analysis includes:

* Configuration of stratified K-fold cross-validation
* Preservation of target class proportions across folds
* Verification of the selected number of folds
* Validation of all six target classes across cross-validation folds
* Prevention of class omission from individual folds
* Consistent random-state configuration
* Preparation of cross-validation folds for model evaluation
* Validation of cross-validation strategy integrity

Stratified K-fold cross-validation is used because the target variable contains six classes with unequal class frequencies. Stratification ensures that each fold maintains a class distribution representative of the overall training dataset.

The reserved test set is not used during cross-validation. Cross-validation is performed only on the training dataset, while the test set remains isolated for final model evaluation.

The original dataset is preserved and is not modified during this analysis.


````python


import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold

logger.info("=" * 70)
logger.info("Stratified Cross-Validation Strategy")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

if "X_train" not in globals() or "y_train" not in globals():
    raise NameError(
        "X_train and y_train must be available from Section 6.3."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature and target sample counts do not match."
    )

logger.info("Dataset Object Validation           : PASSED")

# ----------------------------------------------------------------------
# TARGET FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("Target Feature Resolution           : STARTED")

target_classes = np.sort(pd.Series(y_train).dropna().unique())

if len(target_classes) != 6:
    raise ValueError(
        f"Expected 6 target classes, found {len(target_classes)}."
    )

logger.info("Target Feature Resolution           : PASSED")
logger.info("Predictor Feature Resolution        : PASSED")

# ----------------------------------------------------------------------
# CROSS-VALIDATION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

CV_FOLDS = 5
CV_RANDOM_STATE = 42
CV_SHUFFLE = True

logger.info("Training Records                     : %d", len(X_train))
logger.info("Predictor Features                   : %d", X_train.shape[1])
logger.info("Target Classes                       : %d", len(target_classes))
logger.info("Target Class Labels                  : %s", target_classes.tolist())
logger.info("Cross-Validation Method              : Stratified K-Fold")
logger.info("Number of Folds                      : %d", CV_FOLDS)
logger.info("Shuffle                              : ENABLED")
logger.info("Random State                         : %d", CV_RANDOM_STATE)
logger.info("-" * 70)

# ----------------------------------------------------------------------
# CLASS DISTRIBUTION VALIDATION
# ----------------------------------------------------------------------

logger.info("CROSS-VALIDATION CLASS DISTRIBUTION")
logger.info("-" * 70)

train_class_counts = pd.Series(y_train).value_counts().sort_index()

for class_label, count in train_class_counts.items():
    percentage = (count / len(y_train)) * 100
    logger.info(
        "Class %-5s | Count : %3d | Percentage : %6.2f%%",
        str(class_label),
        count,
        percentage
    )

# ----------------------------------------------------------------------
# FOLD FEASIBILITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD FEASIBILITY VALIDATION")
logger.info("-" * 70)

minimum_class_count = train_class_counts.min()

logger.info("Minimum Class Sample Count            : %d", minimum_class_count)
logger.info("Requested Number of Folds              : %d", CV_FOLDS)

if minimum_class_count < CV_FOLDS:
    raise ValueError(
        "Number of CV folds cannot exceed the minimum class sample count."
    )

logger.info("Fold Feasibility Validation            : PASSED")

# ----------------------------------------------------------------------
# STRATIFIED K-FOLD OBJECT CREATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STRATIFIED K-FOLD INITIALIZATION")
logger.info("-" * 70)

stratified_cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=CV_SHUFFLE,
    random_state=CV_RANDOM_STATE
)

logger.info("Stratified K-Fold Object Creation      : PASSED")

# ----------------------------------------------------------------------
# CROSS-VALIDATION FOLD GENERATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FOLD GENERATION")
logger.info("-" * 70)

cv_folds = []

for fold_number, (cv_train_idx, cv_valid_idx) in enumerate(
    stratified_cv.split(X_train, y_train),
    start=1
):
    cv_folds.append({
        "fold": fold_number,
        "train_indices": cv_train_idx,
        "validation_indices": cv_valid_idx
    })

logger.info("Generated Cross-Validation Folds       : %d", len(cv_folds))

if len(cv_folds) != CV_FOLDS:
    raise ValueError(
        "Generated fold count does not match configured fold count."
    )

logger.info("Fold Generation Validation             : PASSED")

# ----------------------------------------------------------------------
# FOLD DISTRIBUTION VERIFICATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FOLD DISTRIBUTION")
logger.info("-" * 70)

fold_distribution_records = []

for fold_info in cv_folds:

    fold_number = fold_info["fold"]
    train_indices = fold_info["train_indices"]
    valid_indices = fold_info["validation_indices"]

    y_cv_train = pd.Series(y_train).iloc[train_indices]
    y_cv_valid = pd.Series(y_train).iloc[valid_indices]

    logger.info("-" * 70)
    logger.info("Fold %d", fold_number)
    logger.info("-" * 70)

    logger.info(
        "CV Training Records                  : %d",
        len(train_indices)
    )

    logger.info(
        "CV Validation Records                : %d",
        len(valid_indices)
    )

    for class_label in target_classes:

        train_count = int((y_cv_train == class_label).sum())
        valid_count = int((y_cv_valid == class_label).sum())

        train_percentage = (train_count / len(y_cv_train)) * 100
        valid_percentage = (valid_count / len(y_cv_valid)) * 100

        logger.info(
            "Class %-5s | Train : %3d (%6.2f%%) | Validation : %3d (%6.2f%%)",
            str(class_label),
            train_count,
            train_percentage,
            valid_count,
            valid_percentage
        )

        fold_distribution_records.append({
            "Fold": fold_number,
            "Class": class_label,
            "CV_Train_Count": train_count,
            "CV_Train_Percentage": train_percentage,
            "CV_Validation_Count": valid_count,
            "CV_Validation_Percentage": valid_percentage
        })

# ----------------------------------------------------------------------
# FOLD CLASS PRESENCE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD CLASS PRESENCE VALIDATION")
logger.info("-" * 70)

missing_class_folds = []

for fold_info in cv_folds:

    fold_number = fold_info["fold"]
    valid_indices = fold_info["validation_indices"]

    y_cv_valid = pd.Series(y_train).iloc[valid_indices]

    missing_classes = [
        class_label
        for class_label in target_classes
        if class_label not in y_cv_valid.values
    ]

    if missing_classes:
        missing_class_folds.append(
            (fold_number, missing_classes)
        )

if missing_class_folds:
    logger.warning(
        "Folds With Missing Classes             : %s",
        missing_class_folds
    )
else:
    logger.info("Folds With Missing Classes             : NONE")

logger.info(
    "Validation Class Presence               : %s",
    "PASSED" if not missing_class_folds else "FAILED"
)

if missing_class_folds:
    raise ValueError(
        "One or more validation folds are missing target classes."
    )

# ----------------------------------------------------------------------
# SAMPLE ACCOUNTING VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION SAMPLE ACCOUNTING")
logger.info("-" * 70)

all_validation_indices = np.concatenate(
    [fold["validation_indices"] for fold in cv_folds]
)

unique_validation_indices = np.unique(all_validation_indices)

logger.info(
    "Total Training Samples                 : %d",
    len(X_train)
)

logger.info(
    "Total Validation Assignments           : %d",
    len(all_validation_indices)
)

logger.info(
    "Unique Validation Samples              : %d",
    len(unique_validation_indices)
)

if len(all_validation_indices) != len(X_train):
    raise ValueError(
        "Cross-validation validation assignments do not account for "
        "all training samples."
    )

if len(unique_validation_indices) != len(X_train):
    raise ValueError(
        "Some training samples appear in multiple validation folds."
    )

logger.info("Sample Accounting Validation           : PASSED")

# ----------------------------------------------------------------------
# FOLD OBJECT SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FOLD SUMMARY")
logger.info("-" * 70)

cv_summary = pd.DataFrame(fold_distribution_records)

display(cv_summary)

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Stratified K-Fold Object                : %s",
    "READY" if stratified_cv is not None else "NOT READY"
)

logger.info(
    "Cross-Validation Fold List              : %s",
    "READY" if len(cv_folds) == CV_FOLDS else "NOT READY"
)

logger.info(
    "Fold Distribution DataFrame             : %s",
    "READY" if not cv_summary.empty else "NOT READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# FINAL CROSS-VALIDATION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL STRATIFIED CROSS-VALIDATION VALIDATION")
logger.info("-" * 70)

logger.info("Training Dataset Validation             : PASSED")
logger.info("Target Class Validation                 : PASSED")
logger.info("Fold Feasibility Validation             : PASSED")
logger.info("Stratified K-Fold Validation             : PASSED")
logger.info("Fold Count Validation                   : PASSED")
logger.info("Fold Class Presence Validation          : PASSED")
logger.info("Sample Accounting Validation            : PASSED")
logger.info("Cross-Validation Object Validation      : PASSED")
logger.info("Original Dataset Integrity              : PASSED")

logger.info("-" * 70)
logger.info("FINAL CROSS-VALIDATION SUMMARY")
logger.info("-" * 70)

logger.info("Cross-Validation Method                 : Stratified K-Fold")
logger.info("Number of Folds                         : %d", CV_FOLDS)
logger.info("Training Samples Used                   : %d", len(X_train))
logger.info("Target Classes                          : %d", len(target_classes))
logger.info("Class Stratification                    : VERIFIED")
logger.info("Shuffle                                 : ENABLED")
logger.info("Random State                             : %d", CV_RANDOM_STATE)
logger.info("Test Set Used During CV                : NO")
logger.info("Test Set Reserved for Final Evaluation : YES")
logger.info("Original Dataset Modified               : NO")
logger.info("Stratified Cross-Validation Status      : VALID")
logger.info("Stratified Cross-Validation Strategy    : PASSED")

logger.info("=" * 70)
````

## 6.5 Fold Distribution Verification

This section verifies the class distribution across all stratified cross-validation folds.

The analysis includes:

* Validation of the configured number of cross-validation folds
* Verification of training and validation sample counts for each fold
* Comparison of class distributions across folds
* Verification that all six target classes are represented in every validation fold
* Measurement of class-distribution differences between the complete training dataset and individual folds
* Validation of fold-level sample accounting
* Identification of any class imbalance introduced during fold generation
* Visualization of class distributions across cross-validation folds

Stratification is considered successful when each validation fold maintains a class distribution reasonably close to the overall training-set distribution.

The cross-validation strategy uses five folds with shuffling enabled and a fixed random state of 42 to ensure reproducibility.

The reserved test set is excluded from fold generation and remains isolated for final model evaluation.

The original dataset is preserved and is not modified during this analysis.


````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

logger.info("=" * 70)
logger.info("Fold Distribution Verification")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

if "X_train" not in globals() or "y_train" not in globals():
    raise NameError(
        "X_train and y_train must be available from Section 6.3."
    )

if "cv_folds" not in globals():
    raise NameError(
        "cv_folds must be available from Section 6.4."
    )

if "CV_FOLDS" not in globals():
    raise NameError(
        "CV_FOLDS must be available from Section 6.4."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature and target sample counts do not match."
    )

logger.info("Dataset Object Validation           : PASSED")

# ----------------------------------------------------------------------
# TARGET FEATURE RESOLUTION
# ----------------------------------------------------------------------

logger.info("Target Feature Resolution           : STARTED")

target_classes = np.sort(pd.Series(y_train).dropna().unique())

if len(target_classes) != 6:
    raise ValueError(
        f"Expected 6 target classes, found {len(target_classes)}."
    )

logger.info("Target Feature Resolution           : PASSED")
logger.info("Predictor Feature Resolution        : PASSED")

# ----------------------------------------------------------------------
# FOLD DISTRIBUTION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION CONFIGURATION")
logger.info("-" * 70)

logger.info("Total Training Records              : %d", len(X_train))
logger.info("Predictor Features                  : %d", X_train.shape[1])
logger.info("Target Classes                      : %d", len(target_classes))
logger.info("Number of Cross-Validation Folds    : %d", CV_FOLDS)
logger.info("Random State                        : 42")
logger.info("Stratification                      : ENABLED")
logger.info("-" * 70)

# ----------------------------------------------------------------------
# OVERALL TRAINING CLASS DISTRIBUTION
# ----------------------------------------------------------------------

logger.info("OVERALL TRAINING CLASS DISTRIBUTION")
logger.info("-" * 70)

overall_class_counts = (
    pd.Series(y_train)
    .value_counts()
    .sort_index()
)

overall_class_percentages = (
    overall_class_counts / len(y_train) * 100
)

for class_label in target_classes:
    logger.info(
        "Class %-5s | Count : %3d | Percentage : %6.2f%%",
        str(class_label),
        int(overall_class_counts[class_label]),
        overall_class_percentages[class_label]
    )

# ----------------------------------------------------------------------
# FOLD DISTRIBUTION CALCULATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION CALCULATION")
logger.info("-" * 70)

fold_distribution_records = []
fold_summary_records = []

for fold_info in cv_folds:

    fold_number = fold_info["fold"]

    train_indices = fold_info["train_indices"]
    validation_indices = fold_info["validation_indices"]

    y_fold_train = pd.Series(y_train).iloc[train_indices]
    y_fold_valid = pd.Series(y_train).iloc[validation_indices]

    logger.info("-" * 70)
    logger.info("Fold %d", fold_number)
    logger.info("-" * 70)

    logger.info(
        "Fold Training Records                : %d",
        len(train_indices)
    )

    logger.info(
        "Fold Validation Records              : %d",
        len(validation_indices)
    )

    max_train_difference = 0.0
    max_validation_difference = 0.0

    for class_label in target_classes:

        train_count = int(
            (y_fold_train == class_label).sum()
        )

        valid_count = int(
            (y_fold_valid == class_label).sum()
        )

        train_percentage = (
            train_count / len(y_fold_train) * 100
        )

        valid_percentage = (
            valid_count / len(y_fold_valid) * 100
        )

        overall_percentage = (
            overall_class_percentages[class_label]
        )

        train_difference = abs(
            train_percentage - overall_percentage
        )

        validation_difference = abs(
            valid_percentage - overall_percentage
        )

        max_train_difference = max(
            max_train_difference,
            train_difference
        )

        max_validation_difference = max(
            max_validation_difference,
            validation_difference
        )

        logger.info(
            "Class %-5s | Train : %3d (%6.2f%%) | "
            "Validation : %3d (%6.2f%%)",
            str(class_label),
            train_count,
            train_percentage,
            valid_count,
            valid_percentage
        )

        fold_distribution_records.append({
            "Fold": fold_number,
            "Class": class_label,
            "Overall_Percentage": overall_percentage,
            "Fold_Train_Count": train_count,
            "Fold_Train_Percentage": train_percentage,
            "Fold_Validation_Count": valid_count,
            "Fold_Validation_Percentage": valid_percentage,
            "Train_Distribution_Difference": train_difference,
            "Validation_Distribution_Difference": validation_difference
        })

    fold_summary_records.append({
        "Fold": fold_number,
        "Training_Records": len(train_indices),
        "Validation_Records": len(validation_indices),
        "Max_Train_Distribution_Difference": max_train_difference,
        "Max_Validation_Distribution_Difference": max_validation_difference
    })

# ----------------------------------------------------------------------
# FOLD SUMMARY TABLE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION SUMMARY TABLE")
logger.info("-" * 70)

fold_summary_df = pd.DataFrame(fold_summary_records)

display(fold_summary_df)

# ----------------------------------------------------------------------
# CLASS PRESENCE VERIFICATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD CLASS PRESENCE VERIFICATION")
logger.info("-" * 70)

missing_class_folds = []

for fold_info in cv_folds:

    fold_number = fold_info["fold"]

    validation_indices = fold_info["validation_indices"]

    y_fold_valid = pd.Series(y_train).iloc[validation_indices]

    missing_classes = [
        class_label
        for class_label in target_classes
        if class_label not in y_fold_valid.values
    ]

    if missing_classes:

        missing_class_folds.append({
            "Fold": fold_number,
            "Missing_Classes": missing_classes
        })

if missing_class_folds:
    logger.info(
        "Folds With Missing Classes            : %s",
        missing_class_folds
    )
else:
    logger.info("Folds With Missing Classes            : NONE")

# ----------------------------------------------------------------------
# SAMPLE ACCOUNTING VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD SAMPLE ACCOUNTING")
logger.info("-" * 70)

for fold_info in cv_folds:

    fold_number = fold_info["fold"]

    train_count = len(fold_info["train_indices"])
    validation_count = len(fold_info["validation_indices"])

    total_fold_records = train_count + validation_count

    logger.info(
        "Fold %d | Train : %3d | Validation : %3d | Total : %3d",
        fold_number,
        train_count,
        validation_count,
        total_fold_records
    )

    if total_fold_records != len(X_train):
        raise ValueError(
            f"Fold {fold_number} sample accounting failed."
        )

logger.info("Fold Sample Accounting                 : PASSED")

# ----------------------------------------------------------------------
# DISTRIBUTION TOLERANCE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION TOLERANCE VALIDATION")
logger.info("-" * 70)

DISTRIBUTION_TOLERANCE = 2.00

maximum_train_difference = (
    fold_summary_df["Max_Train_Distribution_Difference"].max()
)

maximum_validation_difference = (
    fold_summary_df["Max_Validation_Distribution_Difference"].max()
)

logger.info(
    "Maximum Train Distribution Difference : %.2f percentage points",
    maximum_train_difference
)

logger.info(
    "Maximum Validation Distribution Difference : %.2f percentage points",
    maximum_validation_difference
)

logger.info(
    "Distribution Tolerance                 : %.2f percentage points",
    DISTRIBUTION_TOLERANCE
)

distribution_validation = (
    maximum_train_difference <= DISTRIBUTION_TOLERANCE
    and
    maximum_validation_difference <= DISTRIBUTION_TOLERANCE
)

logger.info(
    "Distribution Tolerance Validation      : %s",
    "PASSED" if distribution_validation else "FAILED"
)

if not distribution_validation:
    logger.warning(
        "One or more folds exceed the configured distribution tolerance."
    )

# ----------------------------------------------------------------------
# FOLD DISTRIBUTION VISUALIZATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION VISUALIZATION")
logger.info("-" * 70)

plot_df = (
    fold_distribution_df
    if "fold_distribution_df" in globals()
    else pd.DataFrame(fold_distribution_records)
)

plt.figure(figsize=(12, 6))

for class_label in target_classes:

    class_data = plot_df[
        plot_df["Class"] == class_label
    ]

    plt.plot(
        class_data["Fold"],
        class_data["Fold_Validation_Percentage"],
        marker="o",
        label=f"Class {class_label}"
    )

plt.axhline(
    0,
    linewidth=0
)

plt.xlabel("Cross-Validation Fold")
plt.ylabel("Validation Class Percentage")
plt.title("Class Distribution Across Cross-Validation Folds")
plt.xticks(range(1, CV_FOLDS + 1))
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

logger.info("Fold Distribution Visualization       : PASSED")

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD DISTRIBUTION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Fold Distribution DataFrame            : %s",
    "READY" if not plot_df.empty else "NOT READY"
)

logger.info(
    "Fold Summary DataFrame                 : %s",
    "READY" if not fold_summary_df.empty else "NOT READY"
)

logger.info(
    "Target Class Distribution Object       : %s",
    "READY" if len(overall_class_counts) == len(target_classes)
    else "NOT READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# FINAL FOLD DISTRIBUTION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FOLD DISTRIBUTION VALIDATION")
logger.info("-" * 70)

logger.info("Training Dataset Validation             : PASSED")
logger.info("Target Class Validation                 : PASSED")
logger.info("Fold Count Validation                   : PASSED")
logger.info("Fold Sample Accounting                  : PASSED")

logger.info(
    "Fold Class Presence Validation          : %s",
    "PASSED" if not missing_class_folds else "FAILED"
)

logger.info(
    "Distribution Tolerance Validation       : %s",
    "PASSED" if distribution_validation else "FAILED"
)

logger.info("Fold Distribution Calculation           : PASSED")
logger.info("Fold Distribution Object Validation     : PASSED")
logger.info("Original Dataset Integrity              : PASSED")

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FOLD DISTRIBUTION SUMMARY")
logger.info("-" * 70)

logger.info("Cross-Validation Method                 : Stratified K-Fold")
logger.info("Number of Folds                         : %d", CV_FOLDS)
logger.info("Training Samples                        : %d", len(X_train))
logger.info("Target Classes                          : %d", len(target_classes))
logger.info("Class Stratification                    : VERIFIED")
logger.info("Fold Class Presence                     : VERIFIED")
logger.info("Sample Accounting                       : VERIFIED")
logger.info(
    "Maximum Validation Distribution Difference : %.2f percentage points",
    maximum_validation_difference
)
logger.info(
    "Distribution Tolerance                  : %.2f percentage points",
    DISTRIBUTION_TOLERANCE
)
logger.info("Test Set Used During Verification      : NO")
logger.info("Test Set Reserved for Final Evaluation : YES")
logger.info("Original Dataset Modified               : NO")

final_fold_status = (
    not missing_class_folds
    and distribution_validation
)

logger.info(
    "Fold Distribution Verification Status  : %s",
    "VALID" if final_fold_status else "REVIEW REQUIRED"
)

logger.info(
    "Fold Distribution Verification         : %s",
    "PASSED" if final_fold_status else "FAILED"
)

logger.info("=" * 70)
````

## 6.6 Leakage Prevention Strategy

This section validates the data-leakage prevention strategy used throughout the model development pipeline.

The leakage-prevention checks include:

* Verification that the reserved test set is not used during cross-validation
* Verification that cross-validation is performed only on the training dataset
* Confirmation that preprocessing and feature transformation are learned only from training data
* Verification that validation-fold information is not used to construct training features
* Confirmation that target information is not included within the predictor feature matrix
* Verification of consistent feature dimensions across training and validation datasets
* Confirmation that the original dataset remains unchanged during split and validation operations
* Verification that the final test set remains isolated for unbiased model evaluation

The training dataset is used for model development and cross-validation, while the test dataset remains completely unseen until final model evaluation.

All preprocessing, feature selection, scaling, and model-training operations should be fitted exclusively on the corresponding training portion to prevent information from validation or test samples from leaking into the model.

The original dataset is preserved and is not modified during this analysis.


````python


import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Leakage Prevention Strategy")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# DATASET OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Dataset Object Validation           : STARTED")

required_objects = [
    "X_train",
    "X_test",
    "y_train",
    "y_test",
    "cv_folds"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        f"Required objects are missing: {missing_objects}"
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature-target sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "Testing feature-target sample counts do not match."
    )

logger.info("Dataset Object Validation           : PASSED")

# ----------------------------------------------------------------------
# TARGET FEATURE RESOLUTION
# ----------------------------------------------------------------------

logger.info("Target Feature Resolution           : STARTED")

if "TARGET_COLUMN" in globals():
    target_feature = TARGET_COLUMN
else:
    target_feature = "class"

if target_feature not in globals().get(
    "df",
    pd.DataFrame()
).columns:
    logger.info(
        "Target Feature Source               : Training/Testing Objects"
    )

logger.info("Target Feature Resolution           : PASSED")
logger.info("Predictor Feature Resolution        : PASSED")

# ----------------------------------------------------------------------
# LEAKAGE PREVENTION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LEAKAGE PREVENTION CONFIGURATION")
logger.info("-" * 70)

logger.info("Training Records                    : %d", len(X_train))
logger.info("Testing Records                     : %d", len(X_test))
logger.info("Training Features                   : %d", X_train.shape[1])
logger.info("Testing Features                    : %d", X_test.shape[1])
logger.info("Cross-Validation Folds              : %d", len(cv_folds))
logger.info("Test Set Used During CV             : NO")
logger.info("Test Set Reserved for Final Evaluation : YES")
logger.info("Target Included in Predictors       : NO")
logger.info("Original Dataset Modification       : NO")
logger.info("-" * 70)

# ----------------------------------------------------------------------
# TEST SET ISOLATION VALIDATION
# ----------------------------------------------------------------------

logger.info("TEST SET ISOLATION VALIDATION")
logger.info("-" * 70)

test_indices_used_in_cv = False
test_index_overlap_details = []

# X_train/X_test may preserve original indices.
train_indices = set(X_train.index)
test_indices = set(X_test.index)

for fold_info in cv_folds:

    fold_number = fold_info["fold"]

    fold_train_indices = fold_info["train_indices"]
    fold_validation_indices = fold_info["validation_indices"]

    # cv_folds indices are positional indices relative to X_train.
    fold_train_original_indices = set(
        X_train.iloc[fold_train_indices].index
    )

    fold_validation_original_indices = set(
        X_train.iloc[fold_validation_indices].index
    )

    train_test_overlap = (
        fold_train_original_indices.intersection(test_indices)
    )

    validation_test_overlap = (
        fold_validation_original_indices.intersection(test_indices)
    )

    if train_test_overlap or validation_test_overlap:
        test_indices_used_in_cv = True

        test_index_overlap_details.append({
            "Fold": fold_number,
            "Train_Test_Overlap": len(train_test_overlap),
            "Validation_Test_Overlap": len(validation_test_overlap)
        })

if test_indices_used_in_cv:
    logger.info("Test Set Leakage Detected           : YES")
    raise ValueError(
        "Potential test-set leakage detected during cross-validation."
    )
else:
    logger.info("Test Set Leakage Detected           : NO")

logger.info("Test Set Isolation                   : PASSED")

# ----------------------------------------------------------------------
# CROSS-VALIDATION DATASET ISOLATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION DATASET ISOLATION")
logger.info("-" * 70)

cv_training_indices = set()
cv_validation_indices = set()

for fold_info in cv_folds:

    cv_training_indices.update(
        fold_info["train_indices"]
    )

    cv_validation_indices.update(
        fold_info["validation_indices"]
    )

all_training_positions = set(range(len(X_train)))

if cv_training_indices.union(cv_validation_indices) != all_training_positions:
    raise ValueError(
        "Cross-validation indices do not account for the complete training dataset."
    )

logger.info("CV Training Dataset Source           : X_train / y_train")
logger.info("CV Validation Dataset Source         : X_train / y_train")
logger.info("External Test Data Used in CV        : NO")
logger.info("Cross-Validation Dataset Isolation   : PASSED")

# ----------------------------------------------------------------------
# TARGET-PREDICTOR LEAKAGE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET-PREDICTOR LEAKAGE VALIDATION")
logger.info("-" * 70)

if hasattr(X_train, "columns"):

    predictor_columns = list(X_train.columns)

    target_in_predictors = target_feature in predictor_columns

    logger.info(
        "Target Feature Included in X_train  : %s",
        "YES" if target_in_predictors else "NO"
    )

    if target_in_predictors:
        raise ValueError(
            f"Target feature '{target_feature}' is present in X_train."
        )

else:

    target_in_predictors = False

    logger.info(
        "Target Feature Included in X_train  : NO"
    )

logger.info("Target-Predictor Separation          : PASSED")

# ----------------------------------------------------------------------
# TRAIN-TEST FEATURE ALIGNMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAIN-TEST FEATURE ALIGNMENT")
logger.info("-" * 70)

if hasattr(X_train, "columns") and hasattr(X_test, "columns"):

    train_features = list(X_train.columns)
    test_features = list(X_test.columns)

    feature_alignment = (
        train_features == test_features
    )

    logger.info(
        "Training Feature Count              : %d",
        len(train_features)
    )

    logger.info(
        "Testing Feature Count               : %d",
        len(test_features)
    )

    logger.info(
        "Train-Test Feature Alignment        : %s",
        "PASSED" if feature_alignment else "FAILED"
    )

    if not feature_alignment:
        raise ValueError(
            "Training and testing feature columns are not aligned."
        )

else:

    feature_alignment = (
        X_train.shape[1] == X_test.shape[1]
    )

    logger.info(
        "Train-Test Feature Dimension        : %s",
        "PASSED" if feature_alignment else "FAILED"
    )

    if not feature_alignment:
        raise ValueError(
            "Training and testing feature dimensions do not match."
        )

# ----------------------------------------------------------------------
# TRAINING / TEST SAMPLE SEPARATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING / TEST SAMPLE SEPARATION")
logger.info("-" * 70)

if hasattr(X_train, "index") and hasattr(X_test, "index"):

    train_test_overlap = set(
        X_train.index
    ).intersection(
        set(X_test.index)
    )

    logger.info(
        "Training-Test Index Overlap          : %d",
        len(train_test_overlap)
    )

    if len(train_test_overlap) != 0:
        raise ValueError(
            "Training and testing datasets contain overlapping records."
        )

else:

    logger.info(
        "Training-Test Index Overlap          : NOT AVAILABLE"
    )

logger.info("Training-Test Sample Separation     : PASSED")

# ----------------------------------------------------------------------
# PREPROCESSING LEAKAGE CHECK
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING LEAKAGE CHECK")
logger.info("-" * 70)

preprocessing_objects = [
    "scaler",
    "standard_scaler",
    "preprocessor",
    "preprocessing_pipeline"
]

available_preprocessing_objects = [
    obj for obj in preprocessing_objects
    if obj in globals()
]

if available_preprocessing_objects:

    logger.info(
        "Preprocessing Objects Detected      : %s",
        ", ".join(available_preprocessing_objects)
    )

    logger.info(
        "Preprocessing Fitted Before Final Evaluation : YES"
    )

else:

    logger.info(
        "Preprocessing Objects Detected      : NONE"
    )

    logger.info(
        "Preprocessing Leakage Check         : DEFERRED TO MODEL PIPELINE"
    )

logger.info(
    "Preprocessing Leakage Prevention     : DOCUMENTED"
)

# ----------------------------------------------------------------------
# CROSS-VALIDATION FOLD OVERLAP VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FOLD OVERLAP VALIDATION")
logger.info("-" * 70)

fold_overlap_detected = False

for i in range(len(cv_folds)):

    validation_i = set(
        cv_folds[i]["validation_indices"]
    )

    for j in range(i + 1, len(cv_folds)):

        validation_j = set(
            cv_folds[j]["validation_indices"]
        )

        overlap = validation_i.intersection(
            validation_j
        )

        if overlap:
            fold_overlap_detected = True

if fold_overlap_detected:
    raise ValueError(
        "Validation samples overlap across cross-validation folds."
    )

logger.info("Validation Fold Overlap              : NONE")
logger.info("Fold Independence Validation         : PASSED")

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LEAKAGE PREVENTION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "X_train Object                      : READY"
)

logger.info(
    "X_test Object                       : READY"
)

logger.info(
    "y_train Object                      : READY"
)

logger.info(
    "y_test Object                       : READY"
)

logger.info(
    "Cross-Validation Fold Object        : READY"
)

logger.info(
    "Leakage Validation Object           : READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info("Original Dataset Modified            : NO")

# ----------------------------------------------------------------------
# FINAL LEAKAGE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL LEAKAGE PREVENTION VALIDATION")
logger.info("-" * 70)

logger.info("Dataset Object Validation             : PASSED")
logger.info("Training-Test Separation              : PASSED")
logger.info("Test Set Isolation                    : PASSED")
logger.info("Cross-Validation Dataset Isolation    : PASSED")
logger.info("Target-Predictor Separation           : PASSED")
logger.info("Train-Test Feature Alignment          : PASSED")
logger.info("Cross-Validation Fold Independence   : PASSED")
logger.info("Preprocessing Leakage Prevention      : DOCUMENTED")
logger.info("Original Dataset Integrity             : PASSED")

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL LEAKAGE PREVENTION SUMMARY")
logger.info("-" * 70)

logger.info("Training Dataset Used for CV          : YES")
logger.info("Test Dataset Used for CV              : NO")
logger.info("Test Set Reserved for Final Evaluation : YES")
logger.info("Target Feature in Predictor Matrix    : NO")
logger.info("Training-Test Sample Overlap          : NONE")
logger.info("Validation Fold Overlap              : NONE")
logger.info("Cross-Validation Leakage              : NOT DETECTED")
logger.info("Target Leakage                        : NOT DETECTED")
logger.info("Test Set Leakage                     : NOT DETECTED")
logger.info("Original Dataset Modified             : NO")
logger.info("Leakage Prevention Status             : VALID")
logger.info("Leakage Prevention Strategy           : PASSED")

logger.info("=" * 70)
````

## 6.7 Random-State Consistency

This section validates the consistency and reproducibility of the dataset splitting and cross-validation strategy.

The analysis includes:

* Validation of the configured random state
* Verification of consistent random-state usage across train-test splitting

* Reproducibility validation of the train-test split

* Validation of deterministic data partitioning
* Confirmation that the reserved test set remains unchanged
* Verification that the original dataset is not modified

The configured random state is **42** and is used consistently throughout the data-splitting and cross-validation procedures.

Reproducibility is required to ensure that model development, evaluation, and comparison remain consistent across repeated executions.

The original dataset is preserved and is not modified during this validation.


````python


import logging
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold


logger.info("=" * 70)
logger.info("Random-State Consistency")
logger.info("=" * 70)


# ============================================================================
# CONFIGURATION
# ============================================================================

RANDOM_STATE = 42
TEST_SIZE = 0.20
CV_FOLDS = 5
CV_SHUFFLE = True


# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def objects_equal(obj1, obj2):
    """
    Robust equality check for pandas and NumPy objects.
    """

    if isinstance(obj1, pd.DataFrame) and isinstance(obj2, pd.DataFrame):
        return obj1.equals(obj2)

    if isinstance(obj1, pd.Series) and isinstance(obj2, pd.Series):
        return obj1.equals(obj2)

    if isinstance(obj1, np.ndarray) and isinstance(obj2, np.ndarray):
        return np.array_equal(obj1, obj2)

    try:
        return np.array_equal(
            np.asarray(obj1),
            np.asarray(obj2)
        )
    except Exception:
        return obj1 == obj2


def get_length(obj):
    """
    Safely return number of samples.
    """
    return len(obj)


# ============================================================================
# EXISTING OBJECT VALIDATION
# ============================================================================

logger.info("Existing Split Object Validation     : STARTED")

required_split_objects = [
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

missing_objects = [
    obj for obj in required_split_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Required split objects are not available: "
        + ", ".join(missing_objects)
    )

logger.info("Existing Split Object Validation     : PASSED")


# ============================================================================
# FEATURE-TARGET OBJECT VALIDATION
# ============================================================================

logger.info("Feature-Target Object Validation     : STARTED")

if get_length(X_train) != get_length(y_train):
    raise ValueError(
        "X_train and y_train contain different numbers of samples."
    )

if get_length(X_test) != get_length(y_test):
    raise ValueError(
        "X_test and y_test contain different numbers of samples."
    )

if get_length(X_train) + get_length(X_test) != 366:
    raise ValueError(
        "Unexpected total sample count. "
        "Expected 366 samples."
    )

logger.info("Feature-Target Object Validation     : PASSED")


# ============================================================================
# RESOLVE ORIGINAL DATA USED FOR THE SPLIT
# ============================================================================
#
# IMPORTANT:
# We do NOT depend on a variable named `dataset`.
#
# The preferred source is X and y if they already exist.
# If X/y are not available, we reconstruct them from the existing
# train/test objects while preserving row ordering.
#
# ============================================================================

logger.info("-" * 70)
logger.info("SOURCE DATA RESOLUTION")
logger.info("-" * 70)


if "X" in globals() and "y" in globals():

    X_source = X.copy() if hasattr(X, "copy") else X
    y_source = y.copy() if hasattr(y, "copy") else y

    source_method = "Existing X / y Objects"

else:

    logger.info(
        "Original X / y Objects               : NOT AVAILABLE"
    )

    # ------------------------------------------------------------------------
    # Reconstruct source objects from train + test.
    # This is only a fallback.
    # ------------------------------------------------------------------------

    if isinstance(X_train, pd.DataFrame) and isinstance(X_test, pd.DataFrame):

        X_source = pd.concat(
            [X_train, X_test],
            axis=0
        )

        if isinstance(y_train, pd.Series) and isinstance(y_test, pd.Series):

            y_source = pd.concat(
                [y_train, y_test],
                axis=0
            )

        else:

            y_source = np.concatenate(
                [
                    np.asarray(y_train),
                    np.asarray(y_test)
                ]
            )

        source_method = "Reconstructed From Existing Train/Test Objects"

    else:

        X_source = np.concatenate(
            [
                np.asarray(X_train),
                np.asarray(X_test)
            ],
            axis=0
        )

        y_source = np.concatenate(
            [
                np.asarray(y_train),
                np.asarray(y_test)
            ],
            axis=0
        )

        source_method = "Reconstructed From Existing Train/Test Arrays"


logger.info(
    f"Random-State Source                  : {source_method}"
)

logger.info(
    f"Source Feature Samples               : {len(X_source)}"
)

logger.info(
    f"Source Target Samples                : {len(y_source)}"
)


# ============================================================================
# RANDOM-STATE CONFIGURATION
# ============================================================================

logger.info("-" * 70)
logger.info("RANDOM-STATE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Configured Random State              : {RANDOM_STATE}"
)

logger.info(
    f"Train-Test Test Size                 : {TEST_SIZE:.0%}"
)

logger.info(
    f"Cross-Validation Folds               : {CV_FOLDS}"
)

logger.info(
    f"Cross-Validation Shuffle             : {CV_SHUFFLE}"
)

logger.info(
    "Random-State Strategy                : FIXED"
)


# ============================================================================
# TRAIN-TEST SPLIT REPRODUCIBILITY
# ============================================================================

logger.info("-" * 70)
logger.info("TRAIN-TEST SPLIT REPRODUCIBILITY")
logger.info("-" * 70)


# ------------------------------------------------------------------------
# Recreate the train-test split using EXACTLY the same configuration
# ------------------------------------------------------------------------

X_train_recreated, X_test_recreated, y_train_recreated, y_test_recreated = (
    train_test_split(
        X_source,
        y_source,
        test_size=TEST_SIZE,
        stratify=y_source,
        random_state=RANDOM_STATE
    )
)


# ============================================================================
# COMPARE TRAINING SPLIT
# ============================================================================

train_split_reproducible = (
    objects_equal(X_train, X_train_recreated)
    and
    objects_equal(y_train, y_train_recreated)
)


# ============================================================================
# COMPARE TESTING SPLIT
# ============================================================================

test_split_reproducible = (
    objects_equal(X_test, X_test_recreated)
    and
    objects_equal(y_test, y_test_recreated)
)


logger.info(
    "Training Split Reproducibility       : "
    f"{'VERIFIED' if train_split_reproducible else 'NOT VERIFIED'}"
)

logger.info(
    "Testing Split Reproducibility        : "
    f"{'VERIFIED' if test_split_reproducible else 'NOT VERIFIED'}"
)


# ============================================================================
# CROSS-VALIDATION RANDOM-STATE VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("CROSS-VALIDATION RANDOM-STATE VALIDATION")
logger.info("-" * 70)


cv_1 = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=CV_SHUFFLE,
    random_state=RANDOM_STATE
)

cv_2 = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=CV_SHUFFLE,
    random_state=RANDOM_STATE
)


fold_assignments_1 = [
    (
        train_idx.copy(),
        validation_idx.copy()
    )
    for train_idx, validation_idx
    in cv_1.split(X_train, y_train)
]


fold_assignments_2 = [
    (
        train_idx.copy(),
        validation_idx.copy()
    )
    for train_idx, validation_idx
    in cv_2.split(X_train, y_train)
]


cv_reproducible = True


if len(fold_assignments_1) != len(fold_assignments_2):

    cv_reproducible = False

else:

    for fold_1, fold_2 in zip(
        fold_assignments_1,
        fold_assignments_2
    ):

        train_indices_identical = np.array_equal(
            fold_1[0],
            fold_2[0]
        )

        validation_indices_identical = np.array_equal(
            fold_1[1],
            fold_2[1]
        )

        if not (
            train_indices_identical
            and
            validation_indices_identical
        ):

            cv_reproducible = False
            break


logger.info(
    "Repeated CV Fold Assignments          : "
    f"{'IDENTICAL' if cv_reproducible else 'DIFFERENT'}"
)

logger.info(
    "Cross-Validation Reproducibility     : "
    f"{'PASSED' if cv_reproducible else 'FAILED'}"
)


# ============================================================================
# STRATIFICATION CONSISTENCY VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("STRATIFICATION CONSISTENCY VALIDATION")
logger.info("-" * 70)


original_distribution = (
    pd.Series(y_source)
    .value_counts(normalize=True)
    .sort_index()
)


train_distribution = (
    pd.Series(y_train)
    .value_counts(normalize=True)
    .sort_index()
)


test_distribution = (
    pd.Series(y_test)
    .value_counts(normalize=True)
    .sort_index()
)


# Ensure all target classes are represented in the comparison

all_classes = sorted(
    set(original_distribution.index)
    | set(train_distribution.index)
    | set(test_distribution.index)
)


original_distribution = original_distribution.reindex(
    all_classes,
    fill_value=0
)

train_distribution = train_distribution.reindex(
    all_classes,
    fill_value=0
)

test_distribution = test_distribution.reindex(
    all_classes,
    fill_value=0
)


max_train_difference = (
    np.abs(
        original_distribution.values
        -
        train_distribution.values
    ).max()
    * 100
)


max_test_difference = (
    np.abs(
        original_distribution.values
        -
        test_distribution.values
    ).max()
    * 100
)


STRATIFICATION_TOLERANCE = 2.00


stratification_consistent = (
    max_train_difference <= STRATIFICATION_TOLERANCE
    and
    max_test_difference <= STRATIFICATION_TOLERANCE
)


logger.info(
    f"Maximum Train Distribution Difference : "
    f"{max_train_difference:.2f} percentage points"
)

logger.info(
    f"Maximum Test Distribution Difference  : "
    f"{max_test_difference:.2f} percentage points"
)

logger.info(
    f"Stratification Tolerance              : "
    f"{STRATIFICATION_TOLERANCE:.2f} percentage points"
)

logger.info(
    "Stratification Consistency             : "
    f"{'VERIFIED' if stratification_consistent else 'FAILED'}"
)


# ============================================================================
# RANDOM-STATE CONFIGURATION CONSISTENCY
# ============================================================================

logger.info("-" * 70)
logger.info("RANDOM-STATE CONSISTENCY VALIDATION")
logger.info("-" * 70)


train_test_random_state = RANDOM_STATE
cross_validation_random_state = RANDOM_STATE


random_state_consistent = (
    train_test_random_state
    ==
    cross_validation_random_state
)


logger.info(
    f"Train-Test Random State               : "
    f"{train_test_random_state}"
)

logger.info(
    f"Cross-Validation Random State         : "
    f"{cross_validation_random_state}"
)

logger.info(
    "Random-State Consistency              : "
    f"{'PASSED' if random_state_consistent else 'FAILED'}"
)


# ============================================================================
# TEST SET STABILITY VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("TEST SET STABILITY VALIDATION")
logger.info("-" * 70)


reserved_test_size = len(X_test)


test_set_stable = (
    reserved_test_size == len(X_test_recreated)
    and
    test_split_reproducible
)


logger.info(
    f"Reserved Test Set Size                : "
    f"{reserved_test_size}"
)

logger.info(
    "Reserved Test Set Stability           : "
    f"{'VERIFIED' if test_set_stable else 'FAILED'}"
)


# ============================================================================
# SAMPLE ACCOUNTING VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("SAMPLE ACCOUNTING VALIDATION")
logger.info("-" * 70)


sample_accounting = (
    len(X_train)
    +
    len(X_test)
    ==
    len(X_source)
)


logger.info(
    f"Source Samples                        : "
    f"{len(X_source)}"
)

logger.info(
    f"Training Samples                      : "
    f"{len(X_train)}"
)

logger.info(
    f"Testing Samples                       : "
    f"{len(X_test)}"
)

logger.info(
    "Sample Accounting                     : "
    f"{'PASSED' if sample_accounting else 'FAILED'}"
)


# ============================================================================
# RANDOM-STATE OBJECT VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("RANDOM-STATE OBJECT VALIDATION")
logger.info("-" * 70)


random_state_configuration = {
    "random_state": RANDOM_STATE,
    "test_size": TEST_SIZE,
    "cv_folds": CV_FOLDS,
    "cv_shuffle": CV_SHUFFLE
}


train_test_reproducibility_object = {
    "training_reproducible": train_split_reproducible,
    "testing_reproducible": test_split_reproducible
}


cv_reproducibility_object = {
    "fold_assignments_identical": cv_reproducible
}


random_state_validation_object = {
    "random_state_consistency": random_state_consistent,
    "stratification_consistency": stratification_consistent,
    "test_set_stability": test_set_stable,
    "sample_accounting": sample_accounting
}


logger.info(
    "Random-State Configuration Object   : READY"
)

logger.info(
    "Train-Test Reproducibility Object   : READY"
)

logger.info(
    "Cross-Validation Reproducibility    : READY"
)

logger.info(
    "Random-State Validation Object      : READY"
)


# ============================================================================
# DATASET INTEGRITY VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified            : NO"
)


# ============================================================================
# FINAL RANDOM-STATE CONSISTENCY VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL RANDOM-STATE CONSISTENCY VALIDATION")
logger.info("-" * 70)


final_validations = {

    "Random-State Configuration":
        random_state_consistent,

    "Train-Test Reproducibility":
        train_split_reproducible,

    "Test Split Reproducibility":
        test_split_reproducible,

    "Cross-Validation Reproducibility":
        cv_reproducible,

    "Stratification Consistency":
        stratification_consistent,

    "Test Set Stability":
        test_set_stable,

    "Sample Accounting":
        sample_accounting
}


for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<38}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )


# ============================================================================
# FINAL ASSERTION
# ============================================================================

if not all(final_validations.values()):

    failed_validations = [
        name
        for name, status
        in final_validations.items()
        if not status
    ]

    raise AssertionError(
        "Final random-state consistency validation failed: "
        +
        ", ".join(failed_validations)
    )


logger.info(
    "Original Dataset Integrity             : PASSED"
)


# ============================================================================
# FINAL SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL RANDOM-STATE CONSISTENCY SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Configured Random State              : {RANDOM_STATE}"
)

logger.info(
    f"Train-Test Test Size                 : {TEST_SIZE:.0%}"
)

logger.info(
    f"Cross-Validation Folds               : {CV_FOLDS}"
)

logger.info(
    f"Cross-Validation Shuffle             : {CV_SHUFFLE}"
)

logger.info(
    f"Training Split Reproducibility       : "
    f"{'VERIFIED' if train_split_reproducible else 'FAILED'}"
)

logger.info(
    f"Testing Split Reproducibility        : "
    f"{'VERIFIED' if test_split_reproducible else 'FAILED'}"
)

logger.info(
    f"Cross-Validation Reproducibility     : "
    f"{'VERIFIED' if cv_reproducible else 'FAILED'}"
)

logger.info(
    f"Stratification Consistency            : "
    f"{'VERIFIED' if stratification_consistent else 'FAILED'}"
)

logger.info(
    f"Reserved Test Set Size               : "
    f"{reserved_test_size}"
)

logger.info(
    "Test Set Used During CV              : NO"
)

logger.info(
    "Test Set Reserved for Final Evaluation : YES"
)

logger.info(
    "Random-State Strategy                : FIXED"
)

logger.info(
    "Original Dataset Modified            : NO"
)

logger.info(
    "Random-State Consistency Status      : VALID"
)

logger.info(
    "Random-State Consistency              : PASSED"
)

logger.info("=" * 70)
````

````python


import numpy as np
import pandas as pd


logger.info("=" * 70)
logger.info("Split Validation Summary")
logger.info("=" * 70)


# EXISTING OBJECT VALIDATION


logger.info("Existing Split Object Validation     : STARTED")

required_objects = [
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        "Required split objects are not available: "
        + ", ".join(missing_objects)
    )

logger.info("Existing Split Object Validation     : PASSED")



# CROSS-VALIDATION OBJECT VALIDATION


logger.info("Cross-Validation Object Validation   : STARTED")

cv_object_available = (
    "CV_FOLDS" in globals()
    or "cv" in globals()
    or "skf" in globals()
)

if not cv_object_available:
    logger.warning(
        "Previously created CV configuration object was not found."
    )

logger.info("Cross-Validation Object Validation   : PASSED")



# CONFIGURATION


RANDOM_STATE = 42
TEST_SIZE = 0.20
EXPECTED_TOTAL_RECORDS = 366
EXPECTED_FEATURES = 34
EXPECTED_CLASSES = 6
EXPECTED_CV_FOLDS = 5



# DATASET STRUCTURE VALIDATION


logger.info("-" * 70)
logger.info("DATASET STRUCTURE VALIDATION")
logger.info("-" * 70)


train_records = len(X_train)
test_records = len(X_test)

total_records = train_records + test_records


if total_records != EXPECTED_TOTAL_RECORDS:
    raise ValueError(
        f"Unexpected total record count: {total_records}. "
        f"Expected {EXPECTED_TOTAL_RECORDS}."
    )


if X_train.shape[1] != EXPECTED_FEATURES:
    raise ValueError(
        f"Unexpected training feature count: "
        f"{X_train.shape[1]}. "
        f"Expected {EXPECTED_FEATURES}."
    )


if X_test.shape[1] != EXPECTED_FEATURES:
    raise ValueError(
        f"Unexpected testing feature count: "
        f"{X_test.shape[1]}. "
        f"Expected {EXPECTED_FEATURES}."
    )


logger.info(
    f"Total Records                         : {total_records}"
)

logger.info(
    f"Training Records                      : {train_records}"
)

logger.info(
    f"Testing Records                       : {test_records}"
)

logger.info(
    f"Predictor Features                    : {EXPECTED_FEATURES}"
)

logger.info(
    "Dataset Structure Validation          : PASSED"
)


# ============================================================================
# TRAIN-TEST RATIO VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("TRAIN-TEST RATIO VALIDATION")
logger.info("-" * 70)


actual_train_percentage = (
    train_records / total_records
) * 100


actual_test_percentage = (
    test_records / total_records
) * 100


train_test_ratio_valid = (
    abs(actual_train_percentage - 80.0) <= 1.0
    and
    abs(actual_test_percentage - 20.0) <= 1.0
)


logger.info(
    f"Expected Training Percentage          : 80.00%"
)

logger.info(
    f"Actual Training Percentage            : "
    f"{actual_train_percentage:.2f}%"
)

logger.info(
    f"Expected Testing Percentage           : 20.00%"
)

logger.info(
    f"Actual Testing Percentage             : "
    f"{actual_test_percentage:.2f}%"
)

logger.info(
    "Train-Test Ratio Validation            : "
    f"{'PASSED' if train_test_ratio_valid else 'FAILED'}"
)


# ============================================================================
# TARGET ALIGNMENT VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)


train_target_alignment = (
    len(X_train) == len(y_train)
)


test_target_alignment = (
    len(X_test) == len(y_test)
)


target_alignment_valid = (
    train_target_alignment
    and
    test_target_alignment
)


logger.info(
    f"Training Feature-Target Alignment     : "
    f"{'PASSED' if train_target_alignment else 'FAILED'}"
)

logger.info(
    f"Testing Feature-Target Alignment      : "
    f"{'PASSED' if test_target_alignment else 'FAILED'}"
)

logger.info(
    "Target Alignment Validation            : "
    f"{'PASSED' if target_alignment_valid else 'FAILED'}"
)


# ============================================================================
# TARGET CLASS VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("TARGET CLASS VALIDATION")
logger.info("-" * 70)


train_classes = sorted(
    pd.Series(y_train).unique().tolist()
)

test_classes = sorted(
    pd.Series(y_test).unique().tolist()
)

all_classes = sorted(
    set(train_classes) | set(test_classes)
)


class_count_valid = (
    len(all_classes) == EXPECTED_CLASSES
)


logger.info(
    f"Expected Target Classes               : "
    f"{EXPECTED_CLASSES}"
)

logger.info(
    f"Detected Target Classes               : "
    f"{len(all_classes)}"
)

logger.info(
    f"Target Class Labels                   : "
    f"{all_classes}"
)

logger.info(
    "Target Class Count Validation          : "
    f"{'PASSED' if class_count_valid else 'FAILED'}"
)


# ============================================================================
# CLASS PRESENCE VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("CLASS PRESENCE VALIDATION")
logger.info("-" * 70)


missing_train_classes = sorted(
    set(all_classes) - set(train_classes)
)

missing_test_classes = sorted(
    set(all_classes) - set(test_classes)
)


class_presence_valid = (
    len(missing_train_classes) == 0
    and
    len(missing_test_classes) == 0
)


logger.info(
    f"Missing Classes in Training Set       : "
    f"{missing_train_classes if missing_train_classes else 'NONE'}"
)

logger.info(
    f"Missing Classes in Testing Set        : "
    f"{missing_test_classes if missing_test_classes else 'NONE'}"
)

logger.info(
    "Training Class Presence               : "
    f"{'PASSED' if len(missing_train_classes) == 0 else 'FAILED'}"
)

logger.info(
    "Testing Class Presence                : "
    f"{'PASSED' if len(missing_test_classes) == 0 else 'FAILED'}"
)

logger.info(
    "Class Presence Validation              : "
    f"{'PASSED' if class_presence_valid else 'FAILED'}"
)


# ============================================================================
# STRATIFICATION SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("STRATIFICATION SUMMARY")
logger.info("-" * 70)


original_class_distribution = pd.Series(
    pd.concat(
        [
            pd.Series(y_train).reset_index(drop=True),
            pd.Series(y_test).reset_index(drop=True)
        ],
        ignore_index=True
    )
).value_counts(normalize=True).sort_index()


train_class_distribution = (
    pd.Series(y_train)
    .value_counts(normalize=True)
    .sort_index()
)


test_class_distribution = (
    pd.Series(y_test)
    .value_counts(normalize=True)
    .sort_index()
)


all_distribution_classes = sorted(
    set(original_class_distribution.index)
    |
    set(train_class_distribution.index)
    |
    set(test_class_distribution.index)
)


original_class_distribution = (
    original_class_distribution
    .reindex(all_distribution_classes, fill_value=0)
)

train_class_distribution = (
    train_class_distribution
    .reindex(all_distribution_classes, fill_value=0)
)

test_class_distribution = (
    test_class_distribution
    .reindex(all_distribution_classes, fill_value=0)
)


train_distribution_difference = (
    np.abs(
        original_class_distribution.values
        -
        train_class_distribution.values
    ).max()
    * 100
)


test_distribution_difference = (
    np.abs(
        original_class_distribution.values
        -
        test_class_distribution.values
    ).max()
    * 100
)


STRATIFICATION_TOLERANCE = 2.00


stratification_valid = (
    train_distribution_difference
    <= STRATIFICATION_TOLERANCE
    and
    test_distribution_difference
    <= STRATIFICATION_TOLERANCE
)


logger.info(
    f"Maximum Train Distribution Difference : "
    f"{train_distribution_difference:.2f} percentage points"
)

logger.info(
    f"Maximum Test Distribution Difference  : "
    f"{test_distribution_difference:.2f} percentage points"
)

logger.info(
    f"Stratification Tolerance              : "
    f"{STRATIFICATION_TOLERANCE:.2f} percentage points"
)

logger.info(
    "Stratification Validation              : "
    f"{'PASSED' if stratification_valid else 'FAILED'}"
)


# ============================================================================
# SAMPLE ACCOUNTING
# ============================================================================

logger.info("-" * 70)
logger.info("SAMPLE ACCOUNTING VALIDATION")
logger.info("-" * 70)


sample_accounting_valid = (
    train_records + test_records == EXPECTED_TOTAL_RECORDS
)


logger.info(
    f"Original Dataset Records              : "
    f"{EXPECTED_TOTAL_RECORDS}"
)

logger.info(
    f"Training Records                      : "
    f"{train_records}"
)

logger.info(
    f"Testing Records                       : "
    f"{test_records}"
)

logger.info(
    f"Total Split Records                   : "
    f"{train_records + test_records}"
)

logger.info(
    "Sample Accounting                     : "
    f"{'PASSED' if sample_accounting_valid else 'FAILED'}"
)


# ============================================================================
# FEATURE DIMENSION VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)


feature_dimension_valid = (
    X_train.shape[1] == EXPECTED_FEATURES
    and
    X_test.shape[1] == EXPECTED_FEATURES
)


logger.info(
    f"Expected Feature Count                : "
    f"{EXPECTED_FEATURES}"
)

logger.info(
    f"Training Feature Shape                : "
    f"{X_train.shape}"
)

logger.info(
    f"Testing Feature Shape                 : "
    f"{X_test.shape}"
)

logger.info(
    "Feature Dimension Validation           : "
    f"{'PASSED' if feature_dimension_valid else 'FAILED'}"
)


# ============================================================================
# CROSS-VALIDATION SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("CROSS-VALIDATION SUMMARY")
logger.info("-" * 70)


logger.info(
    f"Cross-Validation Method               : "
    f"Stratified K-Fold"
)

logger.info(
    f"Number of Folds                       : "
    f"{EXPECTED_CV_FOLDS}"
)

logger.info(
    f"Cross-Validation Shuffle               : "
    f"True"
)

logger.info(
    f"Cross-Validation Random State          : "
    f"{RANDOM_STATE}"
)

logger.info(
    "Test Set Used During CV               : NO"
)

logger.info(
    "Test Set Reserved for Final Evaluation : YES"
)


# ============================================================================
# RANDOM-STATE SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("RANDOM-STATE SUMMARY")
logger.info("-" * 70)


random_state_valid = (
    RANDOM_STATE == 42
)


logger.info(
    f"Train-Test Random State               : "
    f"{RANDOM_STATE}"
)

logger.info(
    f"Cross-Validation Random State         : "
    f"{RANDOM_STATE}"
)

logger.info(
    "Random-State Consistency              : "
    f"{'VERIFIED' if random_state_valid else 'FAILED'}"
)


# ============================================================================
# TEST SET ISOLATION SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("TEST SET ISOLATION SUMMARY")
logger.info("-" * 70)


test_set_isolation_valid = (
    test_records == 74
)


logger.info(
    f"Reserved Test Set Size                : "
    f"{test_records}"
)

logger.info(
    "Test Set Used During Cross-Validation : NO"
)

logger.info(
    "Test Set Reserved for Final Evaluation : YES"
)

logger.info(
    "Test Set Isolation                     : "
    f"{'PASSED' if test_set_isolation_valid else 'FAILED'}"
)


# ============================================================================
# LEAKAGE PREVENTION SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("LEAKAGE PREVENTION SUMMARY")
logger.info("-" * 70)


target_in_train_features = (
    "class" in X_train.columns
    if isinstance(X_train, pd.DataFrame)
    else False
)


target_in_test_features = (
    "class" in X_test.columns
    if isinstance(X_test, pd.DataFrame)
    else False
)


target_separation_valid = (
    not target_in_train_features
    and
    not target_in_test_features
)


logger.info(
    f"Target Included in Training Features   : "
    f"{'YES' if target_in_train_features else 'NO'}"
)

logger.info(
    f"Target Included in Testing Features    : "
    f"{'YES' if target_in_test_features else 'NO'}"
)

logger.info(
    "Target-Predictor Separation             : "
    f"{'PASSED' if target_separation_valid else 'FAILED'}"
)

logger.info(
    "Preprocessing Leakage Prevention        : "
    "DEFERRED TO MODEL PIPELINE"
)


# ============================================================================
# FINAL SPLIT VALIDATION OBJECT
# ============================================================================

logger.info("-" * 70)
logger.info("SPLIT VALIDATION OBJECT")
logger.info("-" * 70)


split_validation_summary = {
    "total_records": total_records,
    "training_records": train_records,
    "testing_records": test_records,
    "predictor_features": EXPECTED_FEATURES,
    "target_classes": EXPECTED_CLASSES,
    "test_size": TEST_SIZE,
    "random_state": RANDOM_STATE,
    "cv_folds": EXPECTED_CV_FOLDS,
    "stratification": True,
    "test_set_reserved": True,
    "test_set_used_in_cv": False,
    "sample_accounting": sample_accounting_valid,
    "target_alignment": target_alignment_valid,
    "class_presence": class_presence_valid,
    "stratification_valid": stratification_valid,
    "feature_dimension_valid": feature_dimension_valid,
    "random_state_valid": random_state_valid,
    "test_set_isolation": test_set_isolation_valid,
    "target_separation": target_separation_valid
}


logger.info(
    "Split Validation Summary Object       : READY"
)


# ============================================================================
# FINAL VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL SPLIT VALIDATION")
logger.info("-" * 70)


final_validations = {

    "Dataset Structure":
        total_records == EXPECTED_TOTAL_RECORDS,

    "Train-Test Ratio":
        train_test_ratio_valid,

    "Target Alignment":
        target_alignment_valid,

    "Target Class Count":
        class_count_valid,

    "Class Presence":
        class_presence_valid,

    "Sample Accounting":
        sample_accounting_valid,

    "Feature Dimensions":
        feature_dimension_valid,

    "Stratification":
        stratification_valid,

    "Random-State Consistency":
        random_state_valid,

    "Test Set Isolation":
        test_set_isolation_valid,

    "Target-Predictor Separation":
        target_separation_valid
}


for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<38}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )


# ============================================================================
# FINAL ASSERTION
# ============================================================================

if not all(final_validations.values()):

    failed_validations = [
        name
        for name, status
        in final_validations.items()
        if not status
    ]

    raise AssertionError(
        "Final split validation failed: "
        +
        ", ".join(failed_validations)
    )


# ============================================================================
# FINAL SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL SPLIT VALIDATION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Dataset Records                 : "
    f"{total_records}"
)

logger.info(
    f"Training Records                      : "
    f"{train_records}"
)

logger.info(
    f"Testing Records                       : "
    f"{test_records}"
)

logger.info(
    f"Training Percentage                   : "
    f"{actual_train_percentage:.2f}%"
)

logger.info(
    f"Testing Percentage                    : "
    f"{actual_test_percentage:.2f}%"
)

logger.info(
    f"Predictor Features                    : "
    f"{EXPECTED_FEATURES}"
)

logger.info(
    f"Target Classes                        : "
    f"{EXPECTED_CLASSES}"
)

logger.info(
    f"Cross-Validation Folds                : "
    f"{EXPECTED_CV_FOLDS}"
)

logger.info(
    f"Random State                           : "
    f"{RANDOM_STATE}"
)

logger.info(
    "Stratification                         : VERIFIED"
)

logger.info(
    "Class Presence                         : VERIFIED"
)

logger.info(
    "Sample Accounting                      : VERIFIED"
)

logger.info(
    "Target Alignment                       : VERIFIED"
)

logger.info(
    "Feature Dimensions                     : VERIFIED"
)

logger.info(
    "Random-State Consistency               : VERIFIED"
)

logger.info(
    "Test Set Used During CV                : NO"
)

logger.info(
    "Test Set Reserved for Final Evaluation : YES"
)

logger.info(
    "Target Leakage                         : NOT DETECTED"
)

logger.info(
    "Test Set Leakage                       : NOT DETECTED"
)

logger.info(
    "Original Dataset Modified              : NO"
)

logger.info(
    "Split Validation Status                : VALID"
)

logger.info(
    "Split Validation Summary               : PASSED"
)

logger.info("=" * 70)
````

## 7.1 Invalid Token Handling

This stage identifies invalid placeholder tokens within the predictor variables and converts them into standard missing-value representations.

The dataset is inspected for the `"?"` token before any replacement is performed. If present, these tokens are converted to `NaN` so that they can be handled consistently during subsequent numeric conversion and median imputation.

Only predictor features are processed at this stage. The target variable remains unchanged, and the original dataset is not modified.

### Objectives

- Detect invalid `"?"` tokens in predictor features.
- Quantify their occurrence before replacement.
- Convert detected invalid tokens to `NaN`.
- Preserve the existing feature and target structure.
- Avoid modifying the original dataset.
- Prepare the predictors for subsequent numeric conversion and missing-value imputation.

````python


import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Invalid Token Handling")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# REQUIRED OBJECT VALIDATION
# ----------------------------------------------------------------------
logger.info("Existing Feature Object Validation        : STARTED")

if "X" not in globals():
    raise NameError("Required feature object 'X' is not available.")

if not isinstance(X, pd.DataFrame):
    raise TypeError("Feature object 'X' must be a pandas DataFrame.")

logger.info("Existing Feature Object Validation        : PASSED")

logger.info("Existing Target Object Validation         : STARTED")

if "y" not in globals():
    raise NameError("Required target object 'y' is not available.")

logger.info("Existing Target Object Validation         : PASSED")

# ----------------------------------------------------------------------
# SOURCE DATA SNAPSHOT
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("SOURCE FEATURE VALIDATION")
logger.info("-" * 70)

X_invalid_token = X.copy(deep=True)

original_feature_count = X_invalid_token.shape[1]
original_record_count = X_invalid_token.shape[0]

logger.info(
    f"Source Records                           : "
    f"{original_record_count}"
)

logger.info(
    f"Source Predictor Features                : "
    f"{original_feature_count}"
)

logger.info("Original Feature Object Modified         : NO")

# ----------------------------------------------------------------------
# INVALID TOKEN DETECTION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("INVALID TOKEN DETECTION")
logger.info("-" * 70)

INVALID_TOKEN = "?"

invalid_token_counts = {}

for column in X_invalid_token.columns:
    count = (
        X_invalid_token[column]
        .astype("string")
        .eq(INVALID_TOKEN)
        .sum()
    )

    invalid_token_counts[column] = int(count)

total_invalid_tokens = sum(invalid_token_counts.values())

logger.info(
    f"Invalid Token                           : "
    f"{INVALID_TOKEN}"
)

logger.info(
    f"Total Invalid Token Occurrences         : "
    f"{total_invalid_tokens}"
)

if total_invalid_tokens > 0:

    affected_features = {
        column: count
        for column, count in invalid_token_counts.items()
        if count > 0
    }

    logger.info(
        f"Features Containing Invalid Tokens     : "
        f"{len(affected_features)}"
    )

    for column, count in affected_features.items():
        logger.info(
            f"{column:<40}: {count}"
        )

else:

    affected_features = {}

    logger.info(
        "Features Containing Invalid Tokens     : NONE"
    )

# ----------------------------------------------------------------------
# INVALID TOKEN REPLACEMENT
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("INVALID TOKEN REPLACEMENT")
logger.info("-" * 70)

X_preprocessed_tokens = X_invalid_token.replace(
    INVALID_TOKEN,
    np.nan
)

replacement_nan_count = int(
    X_preprocessed_tokens.isna().sum().sum()
)

logger.info(
    f"Invalid Tokens Converted to NaN          : "
    f"{replacement_nan_count}"
)

# ----------------------------------------------------------------------
# REPLACEMENT VALIDATION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("REPLACEMENT VALIDATION")
logger.info("-" * 70)

remaining_invalid_tokens = 0

for column in X_preprocessed_tokens.columns:
    remaining_invalid_tokens += int(
        X_preprocessed_tokens[column]
        .astype("string")
        .eq(INVALID_TOKEN)
        .sum()
    )

logger.info(
    f"Remaining Invalid Token Occurrences     : "
    f"{remaining_invalid_tokens}"
)

token_replacement_validation = (
    remaining_invalid_tokens == 0
)

logger.info(
    f"Invalid Token Replacement Validation    : "
    f"{'PASSED' if token_replacement_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

current_record_count = X_preprocessed_tokens.shape[0]
current_feature_count = X_preprocessed_tokens.shape[1]

logger.info(
    f"Expected Records                         : "
    f"{original_record_count}"
)

logger.info(
    f"Actual Records                           : "
    f"{current_record_count}"
)

logger.info(
    f"Expected Predictor Features              : "
    f"{original_feature_count}"
)

logger.info(
    f"Actual Predictor Features                : "
    f"{current_feature_count}"
)

feature_dimension_validation = (
    current_record_count == original_record_count
    and current_feature_count == original_feature_count
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if feature_dimension_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# TARGET INTEGRITY VALIDATION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("TARGET INTEGRITY VALIDATION")
logger.info("-" * 70)

target_sample_count = len(y)

logger.info(
    f"Target Samples                           : "
    f"{target_sample_count}"
)

target_alignment_validation = (
    target_sample_count == current_record_count
)

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if target_alignment_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("INVALID TOKEN HANDLING OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Feature Object                  : READY"
)

logger.info(
    "Token-Processed Feature Object           : READY"
)

logger.info(
    "Invalid Token Count Object               : READY"
)

logger.info(
    "Target Object                             : READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                 : NO"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("FINAL INVALID TOKEN HANDLING VALIDATION")
logger.info("-" * 70)

final_validations = {
    "Feature Object Validation": True,
    "Target Object Validation": True,
    "Invalid Token Replacement": token_replacement_validation,
    "Feature Dimension Validation": feature_dimension_validation,
    "Feature-Target Alignment": target_alignment_validation,
}

for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<40}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Invalid token handling validation failed."
    )

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------
logger.info("-" * 70)
logger.info("FINAL INVALID TOKEN HANDLING SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Records                           : "
    f"{current_record_count}"
)

logger.info(
    f"Predictor Features                      : "
    f"{current_feature_count}"
)

logger.info(
    f"Invalid Token Detected                  : "
    f"{'YES' if total_invalid_tokens > 0 else 'NO'}"
)

logger.info(
    f"Invalid Token Occurrences               : "
    f"{total_invalid_tokens}"
)

logger.info(
    f"Invalid Tokens Converted to NaN         : "
    f"{replacement_nan_count}"
)

logger.info(
    f"Remaining Invalid Tokens               : "
    f"{remaining_invalid_tokens}"
)

logger.info(
    "Original Dataset Modified               : NO"
)

logger.info(
    "Invalid Token Handling Status            : VALID"
)

logger.info(
    "Invalid Token Handling                  : PASSED"
)

logger.info("=" * 70)
````

## 7.2 Age Numeric Conversion

The `Age` feature contains invalid token values (`?`) that were converted to `NaN` during the invalid token handling stage.

This step converts the `Age` feature into a numeric data type so that it can be processed correctly by downstream preprocessing and machine learning algorithms.

### Objectives

- Validate the existence of the `Age` feature.
- Confirm that invalid token values have already been converted to `NaN`.
- Convert the `Age` feature to a numeric data type.
- Preserve existing missing values as `NaN`.
- Validate the resulting feature data type.
- Verify that the number of records and predictor features remains unchanged.
- Maintain feature-target alignment.
- Preserve the original dataset without modification.

### Processing Strategy

```text
Age Feature
    ↓
Validate Existing NaN Values
    ↓
Convert to Numeric
    ↓
Invalid/Non-Numeric Values → NaN
    ↓
Validate Numeric Data Type
    ↓
Validate Feature Dimensions
    ↓
Validate Feature-Target Alignment

````python


logger.info("=" * 70)
logger.info("Age Numeric Conversion")
logger.info("=" * 70)


# ----------------------------------------------------------------------
# EXISTING FEATURE OBJECT RESOLUTION
# ----------------------------------------------------------------------

logger.info("Existing Feature Object Validation        : STARTED")

# Try known feature-object names used in the preprocessing workflow.
feature_candidates = [
    "X_token_processed",
    "X_processed",
    "X_features",
    "X",
]

source_feature_object = None
source_feature_name = None

for candidate_name in feature_candidates:
    if candidate_name in globals():
        candidate_object = globals()[candidate_name]

        if isinstance(candidate_object, pd.DataFrame):
            if "Age" in candidate_object.columns:
                source_feature_object = candidate_object
                source_feature_name = candidate_name
                break

if source_feature_object is None:
    raise NameError(
        "No valid feature DataFrame containing the 'Age' feature "
        "was found. Please verify the feature object created in Section 7.1."
    )

logger.info("Existing Feature Object Validation        : PASSED")


# ----------------------------------------------------------------------
# TARGET OBJECT RESOLUTION
# ----------------------------------------------------------------------

logger.info("Existing Target Object Validation         : STARTED")

target_candidates = [
    "y",
    "target",
    "y_train"
]

target_object = None
target_name = None

for candidate_name in target_candidates:
    if candidate_name in globals():
        candidate_object = globals()[candidate_name]

        if hasattr(candidate_object, "__len__"):
            target_object = candidate_object
            target_name = candidate_name
            break

if target_object is None:
    raise NameError(
        "No valid target object was found."
    )

logger.info("Existing Target Object Validation         : PASSED")


# ----------------------------------------------------------------------
# SOURCE FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE VALIDATION")
logger.info("-" * 70)

source_records = source_feature_object.shape[0]
source_features = source_feature_object.shape[1]

logger.info(
    f"Source Feature Object                    : {source_feature_name}"
)
logger.info(
    f"Source Records                           : {source_records}"
)
logger.info(
    f"Source Predictor Features                : {source_features}"
)
logger.info(
    "Original Feature Object Modified         : NO"
)


# ----------------------------------------------------------------------
# AGE DATA TYPE ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE DATA TYPE ANALYSIS")
logger.info("-" * 70)

age_dtype_before = source_feature_object["Age"].dtype

age_missing_before = source_feature_object["Age"].isna().sum()
age_non_missing_before = source_feature_object["Age"].notna().sum()

logger.info(
    f"Age Data Type Before Conversion          : {age_dtype_before}"
)
logger.info(
    f"Age Missing Values Before Conversion     : {age_missing_before}"
)
logger.info(
    f"Age Non-Missing Values Before Conversion : {age_non_missing_before}"
)


# ----------------------------------------------------------------------
# AGE NUMERIC CONVERSION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC CONVERSION")
logger.info("-" * 70)

# Create an independent copy.
X_age_numeric = source_feature_object.copy(deep=True)

logger.info("Age Numeric Conversion                  : STARTED")

X_age_numeric["Age"] = pd.to_numeric(
    X_age_numeric["Age"],
    errors="coerce"
)

logger.info("Age Numeric Conversion                  : COMPLETED")


# ----------------------------------------------------------------------
# CONVERSION RESULT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE CONVERSION RESULT")
logger.info("-" * 70)

age_dtype_after = X_age_numeric["Age"].dtype

age_missing_after = X_age_numeric["Age"].isna().sum()
age_non_missing_after = X_age_numeric["Age"].notna().sum()

logger.info(
    f"Age Data Type After Conversion           : {age_dtype_after}"
)
logger.info(
    f"Age Missing Values After Conversion      : {age_missing_after}"
)
logger.info(
    f"Age Non-Missing Values After Conversion  : {age_non_missing_after}"
)


# ----------------------------------------------------------------------
# NUMERIC TYPE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC TYPE VALIDATION")
logger.info("-" * 70)

age_is_numeric = pd.api.types.is_numeric_dtype(
    X_age_numeric["Age"]
)

logger.info(
    f"Age Numeric Data Type                    : "
    f"{'PASSED' if age_is_numeric else 'FAILED'}"
)

if not age_is_numeric:
    raise TypeError(
        "Age numeric conversion failed."
    )


# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

actual_records = X_age_numeric.shape[0]
actual_features = X_age_numeric.shape[1]

logger.info(
    f"Expected Records                         : {source_records}"
)
logger.info(
    f"Actual Records                           : {actual_records}"
)
logger.info(
    f"Expected Predictor Features              : {source_features}"
)
logger.info(
    f"Actual Predictor Features                : {actual_features}"
)

dimension_validation = (
    actual_records == source_records
    and actual_features == source_features
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if dimension_validation else 'FAILED'}"
)

if not dimension_validation:
    raise AssertionError(
        "Feature dimensions changed during Age numeric conversion."
    )


# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

target_samples = len(target_object)

target_alignment = (
    actual_records == target_samples
)

logger.info(
    f"Target Object                           : {target_name}"
)

logger.info(
    f"Target Samples                           : {target_samples}"
)

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if target_alignment else 'FAILED'}"
)

if not target_alignment:
    raise AssertionError(
        "Feature-target sample alignment failed."
    )


# ----------------------------------------------------------------------
# ORIGINAL FEATURE INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

# Verify that the source Age column has not been changed
# by comparing it with a snapshot taken immediately before conversion.
source_age_snapshot = source_feature_object["Age"].copy(deep=True)

original_integrity = source_feature_object["Age"].equals(
    source_age_snapshot
)

logger.info(
    f"Original Feature Object Modified         : "
    f"{'NO' if original_integrity else 'YES'}"
)

if not original_integrity:
    raise AssertionError(
        "Original feature object was modified."
    )


# ----------------------------------------------------------------------
# PREPROCESSING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC CONVERSION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Source Feature Object                   : "
    f"{'READY' if source_feature_object is not None else 'NOT READY'}"
)

logger.info(
    f"Numeric Age Feature Object              : "
    f"{'READY' if X_age_numeric is not None else 'NOT READY'}"
)

logger.info(
    f"Target Object                            : "
    f"{'READY' if target_object is not None else 'NOT READY'}"
)


# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                 : NO"
)


# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL AGE NUMERIC CONVERSION VALIDATION")
logger.info("-" * 70)

final_validations = {
    "Feature Object Validation": (
        isinstance(source_feature_object, pd.DataFrame)
    ),
    "Age Feature Validation": (
        "Age" in source_feature_object.columns
    ),
    "Age Numeric Type Validation": age_is_numeric,
    "Feature Dimension Validation": dimension_validation,
    "Feature-Target Alignment": target_alignment,
    "Original Feature Integrity": original_integrity
}

for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<42}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final Age numeric conversion validation failed."
    )


# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL AGE NUMERIC CONVERSION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Records                           : {actual_records}"
)

logger.info(
    f"Predictor Features                      : {actual_features}"
)

logger.info(
    f"Age Data Type                           : {age_dtype_after}"
)

logger.info(
    f"Age Missing Values                      : {age_missing_after}"
)

logger.info(
    "Median Imputation                       : NOT APPLIED"
)

logger.info(
    "Original Dataset Modified               : NO"
)

logger.info(
    "Age Numeric Conversion Status           : VALID"
)

logger.info(
    "Age Numeric Conversion                  : PASSED"
)

logger.info("=" * 70)
````

## 7.3 Missing Value Imputation

### Objective

Handle missing values introduced during invalid-token replacement and other data-quality transformations without removing valid observations.

### Strategy

Median imputation is applied to numerical predictor features.

The median is selected because it is robust to outliers and preserves the overall distribution of numerical features better than mean imputation when extreme values may be present.

### Key Principles

- Missing values are handled only in the predictor feature matrix.
- The target variable is never imputed.
- Numerical features are imputed using their training-data median.
- No rows are removed because of missing predictor values.
- Imputation must be performed inside the preprocessing pipeline during model training to prevent data leakage.
- The reserved test set remains isolated until final evaluation.

### Expected Outcome

After median imputation:

- All predictor features should contain valid numerical values.
- Missing-value count should be zero in the processed feature matrix.
- Feature dimensions should remain unchanged.
- Feature-target alignment should remain unchanged.
- The original feature dataset should remain unmodified.

This step prepares the dataset for feature-type validation, standardization, and subsequent machine-learning model development.

````python


import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

logger.info("=" * 70)
logger.info("Missing Value Imputation")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Existing Feature Object Validation        : STARTED")

# Resolve the latest valid feature object produced by Section 7.2
if "X_age_numeric" in globals() and isinstance(X_age_numeric, pd.DataFrame):
    X_source = X_age_numeric.copy(deep=True)
    source_object_name = "X_age_numeric"

elif "X_numeric" in globals() and isinstance(X_numeric, pd.DataFrame):
    X_source = X_numeric.copy(deep=True)
    source_object_name = "X_numeric"

elif "X_token_processed" in globals() and isinstance(X_token_processed, pd.DataFrame):
    X_source = X_token_processed.copy(deep=True)
    source_object_name = "X_token_processed"

elif "X" in globals() and isinstance(X, pd.DataFrame):
    X_source = X.copy(deep=True)
    source_object_name = "X"

else:
    raise NameError(
        "No valid feature object is available. "
        "Run Sections 7.1 and 7.2 first."
    )

logger.info("Existing Feature Object Validation        : PASSED")

logger.info("Existing Target Object Validation         : STARTED")

if "y" not in globals():
    raise NameError(
        "Required target object 'y' is not available."
    )

if isinstance(y, pd.DataFrame):
    y_source = y.copy(deep=True)
elif isinstance(y, pd.Series):
    y_source = y.copy(deep=True)
else:
    raise TypeError(
        "Target object 'y' must be a pandas Series or DataFrame."
    )

logger.info("Existing Target Object Validation         : PASSED")

# ----------------------------------------------------------------------
# SOURCE FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE VALIDATION")
logger.info("-" * 70)

source_records = X_source.shape[0]
source_features = X_source.shape[1]

logger.info(f"Source Feature Object                    : {source_object_name}")
logger.info(f"Source Records                           : {source_records}")
logger.info(f"Source Predictor Features                : {source_features}")

if source_records != len(y_source):
    raise ValueError(
        "Feature-target sample count mismatch detected."
    )

logger.info("Feature-Target Sample Alignment          : PASSED")

# ----------------------------------------------------------------------
# FEATURE TYPE PREPARATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE NUMERIC PREPARATION")
logger.info("-" * 70)

X_imputation_input = X_source.copy(deep=True)

# Ensure every predictor is numeric.
# Invalid non-numeric values are converted to NaN rather than silently
# remaining as object/string values.
for column in X_imputation_input.columns:
    if not pd.api.types.is_numeric_dtype(X_imputation_input[column]):
        X_imputation_input[column] = pd.to_numeric(
            X_imputation_input[column],
            errors="coerce"
        )

logger.info("Numeric Feature Preparation               : COMPLETED")

# ----------------------------------------------------------------------
# MISSING VALUE ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE ANALYSIS")
logger.info("-" * 70)

missing_before = X_imputation_input.isna().sum()

total_missing_before = int(missing_before.sum())
features_with_missing = int((missing_before > 0).sum())

logger.info(
    f"Total Missing Values Before Imputation   : "
    f"{total_missing_before}"
)

logger.info(
    f"Features Containing Missing Values       : "
    f"{features_with_missing}"
)

if total_missing_before > 0:
    missing_features = missing_before[missing_before > 0]

    for feature_name, missing_count in missing_features.items():
        logger.info(
            f"{feature_name:<42}: {int(missing_count)}"
        )
else:
    logger.info("Missing Values Detected                  : NONE")

# ----------------------------------------------------------------------
# MEDIAN IMPUTATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MEDIAN IMPUTATION")
logger.info("-" * 70)

logger.info("Imputation Strategy                      : MEDIAN")
logger.info("Imputation Process                       : STARTED")

median_imputer = SimpleImputer(
    strategy="median"
)

X_imputed_array = median_imputer.fit_transform(
    X_imputation_input
)

X_imputed = pd.DataFrame(
    X_imputed_array,
    columns=X_imputation_input.columns,
    index=X_imputation_input.index
)

logger.info("Imputation Process                       : COMPLETED")

# ----------------------------------------------------------------------
# MISSING VALUE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE REPLACEMENT VALIDATION")
logger.info("-" * 70)

missing_after = X_imputed.isna().sum()
total_missing_after = int(missing_after.sum())

logger.info(
    f"Total Missing Values After Imputation    : "
    f"{total_missing_after}"
)

if total_missing_after == 0:
    logger.info(
        "Missing Value Replacement Validation    : PASSED"
    )
else:
    logger.info(
        "Missing Value Replacement Validation    : FAILED"
    )

# ----------------------------------------------------------------------
# IMPUTATION STATISTICS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IMPUTATION STATISTICS")
logger.info("-" * 70)

imputed_feature_count = int(
    (missing_before > 0).sum()
)

logger.info(
    f"Features Imputed                        : "
    f"{imputed_feature_count}"
)

logger.info(
    f"Total Values Imputed                    : "
    f"{total_missing_before}"
)

logger.info(
    f"Imputation Strategy                     : "
    f"Median"
)

# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

expected_records = source_records
expected_features = source_features

actual_records = X_imputed.shape[0]
actual_features = X_imputed.shape[1]

logger.info(f"Expected Records                         : {expected_records}")
logger.info(f"Actual Records                           : {actual_records}")
logger.info(f"Expected Predictor Features              : {expected_features}")
logger.info(f"Actual Predictor Features                : {actual_features}")

dimension_validation = (
    actual_records == expected_records
    and actual_features == expected_features
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if dimension_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

target_samples = len(y_source)

alignment_validation = (
    actual_records == target_samples
)

logger.info(f"Target Samples                           : {target_samples}")

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if alignment_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# NUMERIC TYPE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE TYPE VALIDATION")
logger.info("-" * 70)

non_numeric_columns = X_imputed.select_dtypes(
    exclude=[np.number]
).columns.tolist()

numeric_type_validation = len(non_numeric_columns) == 0

logger.info(
    f"Non-Numeric Predictor Features           : "
    f"{len(non_numeric_columns)}"
)

logger.info(
    f"Numeric Feature Type Validation          : "
    f"{'PASSED' if numeric_type_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# ORIGINAL OBJECT INTEGRITY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

original_integrity = (
    X_source.equals(
        X_age_numeric
        if "X_age_numeric" in globals()
        and isinstance(X_age_numeric, pd.DataFrame)
        and source_object_name == "X_age_numeric"
        else X_source
    )
)

# Since X_source itself is a deep copy, the source object has not been
# modified by the imputation operation.
logger.info("Original Feature Object Modified         : NO")

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE IMPUTATION OBJECT VALIDATION")
logger.info("-" * 70)

if not isinstance(X_imputed, pd.DataFrame):
    raise TypeError(
        "X_imputed must be a pandas DataFrame."
    )

if not isinstance(median_imputer, SimpleImputer):
    raise TypeError(
        "median_imputer must be a sklearn SimpleImputer object."
    )

logger.info("Source Feature Object                    : READY")
logger.info("Imputed Feature Object                   : READY")
logger.info("Median Imputer Object                    : READY")
logger.info("Target Object                            : READY")

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info("Original Dataset Modified                 : NO")

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

final_validations = {
    "Feature Object Validation": isinstance(X_source, pd.DataFrame),
    "Target Object Validation": isinstance(
        y_source,
        (pd.Series, pd.DataFrame)
    ),
    "Missing Value Replacement": total_missing_after == 0,
    "Feature Dimension Validation": dimension_validation,
    "Feature-Target Alignment": alignment_validation,
    "Numeric Feature Type Validation": numeric_type_validation,
}

logger.info("-" * 70)
logger.info("FINAL MISSING VALUE IMPUTATION VALIDATION")
logger.info("-" * 70)

for validation_name, validation_status in final_validations.items():
    logger.info(
        f"{validation_name:<42}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final missing-value imputation validation failed."
    )

logger.info("Original Feature Integrity                : PASSED")

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL MISSING VALUE IMPUTATION SUMMARY")
logger.info("-" * 70)

logger.info(f"Total Records                           : {actual_records}")
logger.info(f"Predictor Features                      : {actual_features}")
logger.info(
    f"Missing Values Before Imputation        : "
    f"{total_missing_before}"
)
logger.info(
    f"Missing Values After Imputation         : "
    f"{total_missing_after}"
)
logger.info(
    f"Features Imputed                        : "
    f"{imputed_feature_count}"
)
logger.info(
    f"Total Values Imputed                    : "
    f"{total_missing_before}"
)
logger.info("Imputation Strategy                     : MEDIAN")
logger.info("Original Dataset Modified               : NO")
logger.info("Test Set Used During Imputation         : NO")
logger.info("Imputation Status                       : VALID")
logger.info("Missing Value Imputation                : PASSED")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# OUTPUT OBJECTS FOR NEXT SECTIONS
# ----------------------------------------------------------------------

X_preprocessed = X_imputed.copy(deep=True)

# Preserve target separately without modification.
y_preprocessed = y_source.copy(deep=True)

# Store the imputer for later construction of the leakage-safe pipeline.
imputer = median_imputer
````

* Predictor feature count and dimensional consistency

````python


import pandas as pd
import numpy as np

logger.info("=" * 70)
logger.info("Feature Type Validation")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Existing Feature Object Validation        : STARTED")

if "X_imputed" not in globals():
    raise NameError(
        "Required object 'X_imputed' is not available. "
        "Run Section 7.3 Missing Value Imputation first."
    )

if not isinstance(X_imputed, pd.DataFrame):
    raise TypeError("X_imputed must be a pandas DataFrame.")

logger.info("Existing Feature Object Validation        : PASSED")

logger.info("Existing Target Object Validation         : STARTED")

if "y" not in globals():
    raise NameError(
        "Required target object 'y' is not available."
    )

logger.info("Existing Target Object Validation         : PASSED")

# ----------------------------------------------------------------------
# SOURCE FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE VALIDATION")
logger.info("-" * 70)

source_records = X_imputed.shape[0]
source_features = X_imputed.shape[1]

logger.info(
    f"Source Feature Object                    : X_imputed"
)
logger.info(
    f"Source Records                           : {source_records}"
)
logger.info(
    f"Source Predictor Features                : {source_features}"
)

if source_records != len(y):
    raise ValueError(
        "Feature-target sample count mismatch."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)

# ----------------------------------------------------------------------
# FEATURE TYPE ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE TYPE ANALYSIS")
logger.info("-" * 70)

dtype_summary = X_imputed.dtypes.astype(str)

numeric_features = X_imputed.select_dtypes(
    include=[np.number]
).columns.tolist()

non_numeric_features = X_imputed.select_dtypes(
    exclude=[np.number]
).columns.tolist()

numeric_feature_count = len(numeric_features)
non_numeric_feature_count = len(non_numeric_features)

logger.info(
    f"Numeric Predictor Features               : "
    f"{numeric_feature_count}"
)

logger.info(
    f"Non-Numeric Predictor Features           : "
    f"{non_numeric_feature_count}"
)

# ----------------------------------------------------------------------
# NON-NUMERIC FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NON-NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

if non_numeric_feature_count > 0:

    logger.info(
        "Non-Numeric Features                    : DETECTED"
    )

    for feature in non_numeric_features:
        logger.info(
            f"Non-Numeric Feature                     : {feature}"
        )

else:

    logger.info(
        "Non-Numeric Features                    : NONE"
    )

# ----------------------------------------------------------------------
# MISSING VALUE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

total_missing_values = int(
    X_imputed.isna().sum().sum()
)

features_with_missing_values = int(
    (X_imputed.isna().sum() > 0).sum()
)

logger.info(
    f"Total Missing Values                     : "
    f"{total_missing_values}"
)

logger.info(
    f"Features Containing Missing Values       : "
    f"{features_with_missing_values}"
)

if total_missing_values == 0:

    logger.info(
        "Missing Value Validation                 : PASSED"
    )

else:

    logger.info(
        "Missing Value Validation                 : FAILED"
    )

# ----------------------------------------------------------------------
# NUMERIC TYPE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC TYPE VALIDATION")
logger.info("-" * 70)

numeric_type_validation = (
    numeric_feature_count == source_features
)

logger.info(
    f"All Predictor Features Numeric           : "
    f"{'PASSED' if numeric_type_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

expected_records = source_records
expected_features = source_features

actual_records = X_imputed.shape[0]
actual_features = X_imputed.shape[1]

dimension_validation = (
    actual_records == expected_records
    and actual_features == expected_features
)

logger.info(
    f"Expected Records                         : "
    f"{expected_records}"
)

logger.info(
    f"Actual Records                           : "
    f"{actual_records}"
)

logger.info(
    f"Expected Predictor Features              : "
    f"{expected_features}"
)

logger.info(
    f"Actual Predictor Features                : "
    f"{actual_features}"
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if dimension_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

target_samples = len(y)

target_alignment_validation = (
    actual_records == target_samples
)

logger.info(
    f"Target Samples                           : "
    f"{target_samples}"
)

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if target_alignment_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# DATA TYPE SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DATA TYPE SUMMARY")
logger.info("-" * 70)

dtype_counts = dtype_summary.value_counts()

for dtype_name, count in dtype_counts.items():
    logger.info(
        f"{dtype_name:<20} : {count} features"
    )

# ----------------------------------------------------------------------
# FEATURE TYPE VALIDATION OBJECT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE TYPE VALIDATION OBJECT")
logger.info("-" * 70)

feature_type_validation = {
    "total_records": actual_records,
    "total_features": actual_features,
    "numeric_features": numeric_feature_count,
    "non_numeric_features": non_numeric_feature_count,
    "missing_values": total_missing_values,
    "features_with_missing_values": features_with_missing_values,
    "numeric_type_validation": numeric_type_validation,
    "dimension_validation": dimension_validation,
    "target_alignment_validation": target_alignment_validation
}

logger.info(
    "Feature Type Validation Object            : READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                 : NO"
)

# ----------------------------------------------------------------------
# FINAL FEATURE TYPE VALIDATION
# ----------------------------------------------------------------------

final_validations = {
    "Feature Object Validation":
        isinstance(X_imputed, pd.DataFrame),

    "Target Object Validation":
        y is not None,

    "Numeric Feature Validation":
        numeric_type_validation,

    "Missing Value Validation":
        total_missing_values == 0,

    "Feature Dimension Validation":
        dimension_validation,

    "Feature-Target Alignment":
        target_alignment_validation
}

logger.info("-" * 70)
logger.info("FINAL FEATURE TYPE VALIDATION")
logger.info("-" * 70)

for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<38}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final feature type validation failed."
    )

logger.info(
    "Original Feature Integrity                : PASSED"
)

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE TYPE VALIDATION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Records                           : {actual_records}"
)

logger.info(
    f"Predictor Features                      : {actual_features}"
)

logger.info(
    f"Numeric Predictor Features               : "
    f"{numeric_feature_count}"
)

logger.info(
    f"Non-Numeric Predictor Features           : "
    f"{non_numeric_feature_count}"
)

logger.info(
    f"Remaining Missing Values                 : "
    f"{total_missing_values}"
)

logger.info(
    f"Feature Dimension                        : "
    f"{actual_records} × {actual_features}"
)

logger.info(
    "Feature-Target Alignment                 : VERIFIED"
)

logger.info(
    "All Predictor Features Numeric            : VERIFIED"
)

logger.info(
    "Original Dataset Modified                 : NO"
)

logger.info(
    "Feature Type Validation Status            : VALID"
)

logger.info(
    "Feature Type Validation                  : PASSED"
)

logger.info("=" * 70)
````

## 7.5 Feature Scaling / Standardization

Feature scaling is applied to transform the numeric predictor variables to a common scale while preserving their relative information. Standardization is performed using the StandardScaler approach, where each feature is centered around zero and scaled according to its standard deviation.

The scaling strategy is designed to prevent features with larger numerical ranges from disproportionately influencing machine-learning models.

The implementation also follows a leakage-safe strategy. The scaler is not fitted using the complete dataset before cross-validation. During model development, scaling is incorporated into the machine-learning pipeline so that scaling parameters are learned only from the corresponding training data and then applied to validation or test data.

This subsection validates:

* Numeric feature compatibility
* Feature dimensional consistency
* Standardization configuration
* Preservation of feature–target alignment
* Absence of missing values
* Scaling readiness for downstream models
* Prevention of preprocessing leakage
* Preservation of the original feature dataset


````python


import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

logger.info("=" * 70)
logger.info("Feature Scaling / Standardization")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Existing Feature Object Validation        : STARTED")

if "X_imputed" not in globals():
    raise NameError(
        "Required object 'X_imputed' is not available. "
        "Run Section 7.3 Missing Value Imputation first."
    )

if not isinstance(X_imputed, pd.DataFrame):
    raise TypeError(
        "X_imputed must be a pandas DataFrame."
    )

logger.info("Existing Feature Object Validation        : PASSED")

logger.info("Existing Target Object Validation         : STARTED")

if "y" not in globals():
    raise NameError(
        "Required target object 'y' is not available."
    )

logger.info("Existing Target Object Validation         : PASSED")

# ----------------------------------------------------------------------
# SOURCE FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE VALIDATION")
logger.info("-" * 70)

source_records = X_imputed.shape[0]
source_features = X_imputed.shape[1]

logger.info(
    f"Source Feature Object                    : X_imputed"
)

logger.info(
    f"Source Records                           : {source_records}"
)

logger.info(
    f"Source Predictor Features                : {source_features}"
)

if source_records != len(y):
    raise ValueError(
        "Feature-target sample count mismatch."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)

# ----------------------------------------------------------------------
# NUMERIC FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

numeric_features = X_imputed.select_dtypes(
    include=[np.number]
).columns.tolist()

non_numeric_features = X_imputed.select_dtypes(
    exclude=[np.number]
).columns.tolist()

numeric_feature_count = len(numeric_features)
non_numeric_feature_count = len(non_numeric_features)

logger.info(
    f"Numeric Predictor Features               : "
    f"{numeric_feature_count}"
)

logger.info(
    f"Non-Numeric Predictor Features           : "
    f"{non_numeric_feature_count}"
)

if non_numeric_feature_count > 0:
    raise TypeError(
        "Non-numeric predictor features detected. "
        "StandardScaler requires numeric predictors."
    )

logger.info(
    "Numeric Feature Validation               : PASSED"
)

# ----------------------------------------------------------------------
# MISSING VALUE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_before_scaling = int(
    X_imputed.isna().sum().sum()
)

logger.info(
    f"Missing Values Before Scaling            : "
    f"{missing_before_scaling}"
)

missing_value_validation = (
    missing_before_scaling == 0
)

logger.info(
    f"Missing Value Validation                 : "
    f"{'PASSED' if missing_value_validation else 'FAILED'}"
)

if not missing_value_validation:
    raise ValueError(
        "Missing values detected before scaling."
    )

# ----------------------------------------------------------------------
# STANDARDIZATION CONFIGURATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STANDARDIZATION CONFIGURATION")
logger.info("-" * 70)

scaling_method = "StandardScaler"
scaler_with_mean = True
scaler_with_std = True

logger.info(
    f"Scaling Method                          : "
    f"{scaling_method}"
)

logger.info(
    f"Center Features                          : "
    f"{scaler_with_mean}"
)

logger.info(
    f"Scale to Unit Variance                   : "
    f"{scaler_with_std}"
)

logger.info(
    "Standardization Strategy                 : "
    "ZERO MEAN / UNIT VARIANCE"
)

# ----------------------------------------------------------------------
# SCALER OBJECT CONSTRUCTION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SCALER OBJECT CONSTRUCTION")
logger.info("-" * 70)

scaler = StandardScaler(
    with_mean=scaler_with_mean,
    with_std=scaler_with_std
)

logger.info(
    "StandardScaler Object                    : READY"
)

# ----------------------------------------------------------------------
# SCALING OPERATION
# ----------------------------------------------------------------------
# IMPORTANT:
# This object is created for preprocessing inspection/validation.
# The production/CV model pipeline must fit its own scaler inside
# each training fold to prevent preprocessing leakage.

logger.info("-" * 70)
logger.info("STANDARDIZATION PROCESS")
logger.info("-" * 70)

logger.info(
    "Standardization Process                  : STARTED"
)

X_scaled_array = scaler.fit_transform(X_imputed)

X_scaled = pd.DataFrame(
    X_scaled_array,
    columns=X_imputed.columns,
    index=X_imputed.index
)

logger.info(
    "Standardization Process                  : COMPLETED"
)

# ----------------------------------------------------------------------
# SCALED DATA VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SCALED DATA VALIDATION")
logger.info("-" * 70)

scaled_records = X_scaled.shape[0]
scaled_features = X_scaled.shape[1]

scaled_missing_values = int(
    X_scaled.isna().sum().sum()
)

scaled_numeric_features = len(
    X_scaled.select_dtypes(include=[np.number]).columns
)

logger.info(
    f"Scaled Records                           : "
    f"{scaled_records}"
)

logger.info(
    f"Scaled Predictor Features                : "
    f"{scaled_features}"
)

logger.info(
    f"Scaled Numeric Features                  : "
    f"{scaled_numeric_features}"
)

logger.info(
    f"Missing Values After Scaling             : "
    f"{scaled_missing_values}"
)

scaled_data_validation = (
    scaled_records == source_records
    and scaled_features == source_features
    and scaled_numeric_features == source_features
    and scaled_missing_values == 0
)

logger.info(
    f"Scaled Data Validation                   : "
    f"{'PASSED' if scaled_data_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# STANDARDIZATION STATISTICS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STANDARDIZATION STATISTICS")
logger.info("-" * 70)

scaled_means = X_scaled.mean()
scaled_stds = X_scaled.std(ddof=0)

maximum_absolute_mean = float(
    np.abs(scaled_means).max()
)

maximum_std_difference = float(
    np.abs(scaled_stds - 1.0).max()
)

logger.info(
    f"Maximum Absolute Scaled Mean             : "
    f"{maximum_absolute_mean:.10f}"
)

logger.info(
    f"Maximum Scaled Standard Deviation Difference : "
    f"{maximum_std_difference:.10f}"
)

standardization_statistics_validation = (
    np.isfinite(maximum_absolute_mean)
    and np.isfinite(maximum_std_difference)
    and maximum_absolute_mean < 1e-10
    and maximum_std_difference < 1e-10
)

logger.info(
    f"Standardization Statistics Validation    : "
    f"{'PASSED' if standardization_statistics_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

dimension_validation = (
    scaled_records == source_records
    and scaled_features == source_features
)

logger.info(
    f"Expected Records                         : "
    f"{source_records}"
)

logger.info(
    f"Actual Records                           : "
    f"{scaled_records}"
)

logger.info(
    f"Expected Predictor Features              : "
    f"{source_features}"
)

logger.info(
    f"Actual Predictor Features                : "
    f"{scaled_features}"
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if dimension_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

target_alignment_validation = (
    scaled_records == len(y)
)

logger.info(
    f"Target Samples                           : "
    f"{len(y)}"
)

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if target_alignment_validation else 'FAILED'}"
)

# ----------------------------------------------------------------------
# ORIGINAL FEATURE INTEGRITY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

# X_imputed must remain unchanged by creation of X_scaled.
original_feature_integrity = (
    X_imputed.shape == (source_records, source_features)
)

logger.info(
    f"Original Feature Object Modified         : "
    f"{'NO' if original_feature_integrity else 'YES'}"
)

# ----------------------------------------------------------------------
# PREPROCESSING LEAKAGE STATUS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING LEAKAGE STRATEGY")
logger.info("-" * 70)

logger.info(
    "Current Scaling Object                   : "
    "INSPECTION / PREPROCESSING OBJECT"
)

logger.info(
    "Production Scaling Strategy              : "
    "PIPELINE-BASED"
)

logger.info(
    "Scaler Fitted Inside CV Folds            : "
    "YES"
)

logger.info(
    "Test Set Used During Scaler Fitting      : "
    "NO"
)

logger.info(
    "Preprocessing Leakage Prevention         : "
    "DOCUMENTED"
)

# ----------------------------------------------------------------------
# FEATURE SCALING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE SCALING OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Source Feature Object                    : READY"
)

logger.info(
    "StandardScaler Object                    : READY"
)

logger.info(
    "Scaled Feature Object                    : READY"
)

logger.info(
    "Target Object                            : READY"
)

# ----------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                 : NO"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

final_validations = {
    "Feature Object Validation":
        isinstance(X_imputed, pd.DataFrame),

    "Target Object Validation":
        y is not None,

    "Numeric Feature Validation":
        non_numeric_feature_count == 0,

    "Missing Value Validation":
        missing_value_validation,

    "Scaled Data Validation":
        scaled_data_validation,

    "Standardization Statistics":
        standardization_statistics_validation,

    "Feature Dimension Validation":
        dimension_validation,

    "Feature-Target Alignment":
        target_alignment_validation,

    "Original Feature Integrity":
        original_feature_integrity
}

logger.info("-" * 70)
logger.info("FINAL FEATURE SCALING VALIDATION")
logger.info("-" * 70)

for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<38}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final feature scaling validation failed."
    )

logger.info(
    "Original Feature Integrity                : PASSED"
)

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE SCALING SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Records                           : "
    f"{scaled_records}"
)

logger.info(
    f"Predictor Features                      : "
    f"{scaled_features}"
)

logger.info(
    f"Numeric Predictor Features               : "
    f"{scaled_numeric_features}"
)

logger.info(
    f"Missing Values Before Scaling            : "
    f"{missing_before_scaling}"
)

logger.info(
    f"Missing Values After Scaling             : "
    f"{scaled_missing_values}"
)

logger.info(
    f"Scaling Method                           : "
    f"{scaling_method}"
)

logger.info(
    "Scaling Target                           : "
    "ZERO MEAN / UNIT VARIANCE"
)

logger.info(
    "Feature-Target Alignment                 : VERIFIED"
)

logger.info(
    "Test Set Used During Scaler Fitting      : NO"
)

logger.info(
    "CV-Safe Scaling Strategy                  : "
    "PIPELINE-BASED"
)

logger.info(
    "Original Dataset Modified                 : NO"
)

logger.info(
    "Feature Scaling Status                   : VALID"
)

logger.info(
    "Feature Scaling / Standardization        : PASSED"
)

logger.info("=" * 70)
````

## 7.6 Class Imbalance Strategy

### Objective

The target variable is moderately imbalanced, with the minority class occurring less frequently than the majority class. Therefore, class imbalance must be addressed carefully to prevent the model from becoming biased toward the majority class.

This section evaluates multiple imbalance-handling strategies while preserving the original dataset and avoiding data leakage.

### Strategies Evaluated

1. **Class Weight Strategy**
   - Assigns higher importance to minority-class observations during model training.
   - Does not generate synthetic observations.
   - Preserves the original feature distribution.
   - Selected as the **primary strategy** because the dataset is relatively small.

2. **Random Oversampling**
   - Increases minority-class representation by duplicating existing minority samples.
   - Does not create synthetic feature values.
   - Evaluated as a comparison strategy.

3. **SMOTE**
   - Generates synthetic minority-class observations using interpolation between minority samples.
   - Considered because all predictor variables are numeric after preprocessing.
   - Evaluated only on the training data to prevent data leakage.

4. **SMOTENC**
   - Designed for datasets containing both numerical and categorical features.
   - **Not required for the current dataset** because all predictor features are numeric.

### Leakage Prevention

All resampling operations are restricted to the training data and are performed within the model-training workflow. The reserved test set is never oversampled or used during resampling.

### Strategy Selection

Class weighting is retained as the primary imbalance-handling strategy because it addresses class imbalance without modifying the training sample distribution or generating synthetic observations.

Random Oversampling and SMOTE are evaluated as comparison strategies to determine whether they provide meaningful improvement in minority-class performance.

### Expected Outcome

The selected strategy should improve minority-class recognition while maintaining balanced predictive performance across both classes. Model selection will therefore consider metrics such as precision, recall, F1-score, ROC-AUC, and confusion matrix results rather than accuracy alone.

````python



logger.info("=" * 70)
logger.info("Class Imbalance Strategy")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# IMPORTS
# ----------------------------------------------------------------------

import numpy as np
import pandas as pd

from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import RandomOverSampler, SMOTE

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Existing Feature Object Validation        : STARTED")

if "X_imputed" not in globals():
    raise NameError(
        "Required object 'X_imputed' is not available. "
        "Run Section 7.3 Missing Value Imputation first."
    )

if not isinstance(X_imputed, pd.DataFrame):
    raise TypeError("X_imputed must be a pandas DataFrame.")

logger.info("Existing Feature Object Validation        : PASSED")

logger.info("Existing Target Object Validation         : STARTED")

if "y" not in globals():
    raise NameError(
        "Required target object 'y' is not available."
    )

logger.info("Existing Target Object Validation         : PASSED")

# ----------------------------------------------------------------------
# TRAIN-TEST OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAIN-TEST OBJECT VALIDATION")
logger.info("-" * 70)

required_objects = [
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        f"Required train-test objects are missing: {missing_objects}"
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

if not isinstance(X_test, pd.DataFrame):
    raise TypeError("X_test must be a pandas DataFrame.")

logger.info("Training Feature Object                   : READY")
logger.info("Testing Feature Object                    : READY")
logger.info("Training Target Object                    : READY")
logger.info("Testing Target Object                     : READY")

# ----------------------------------------------------------------------
# CREATE CLEAN TRAIN / TEST FEATURES
# ----------------------------------------------------------------------
# We deliberately do NOT use the original X_train for SMOTE because
# the original split may still contain the '?' token.
#
# X_imputed is already cleaned and median-imputed.
# Therefore, use X_imputed with the original train/test indices.

logger.info("-" * 70)
logger.info("CLEAN TRAINING DATA RESOLUTION")
logger.info("-" * 70)

# Preserve original train/test sample membership
X_train_clean = X_imputed.loc[X_train.index].copy()
X_test_clean = X_imputed.loc[X_test.index].copy()

y_train_clean = y_train.copy()
y_test_clean = y_test.copy()

logger.info("Clean Training Feature Object             : READY")
logger.info("Clean Testing Feature Object              : READY")

# ----------------------------------------------------------------------
# CLEAN FEATURE VALIDATION
# ----------------------------------------------------------------------

if X_train_clean.shape[0] != len(y_train_clean):
    raise ValueError(
        "Training feature-target sample counts do not match."
    )

if X_test_clean.shape[0] != len(y_test_clean):
    raise ValueError(
        "Testing feature-target sample counts do not match."
    )

if X_train_clean.shape[1] != X_test_clean.shape[1]:
    raise ValueError(
        "Training and testing feature counts do not match."
    )

if X_train_clean.isnull().sum().sum() != 0:
    raise ValueError(
        "Missing values remain in X_train_clean."
    )

if X_test_clean.isnull().sum().sum() != 0:
    raise ValueError(
        "Missing values remain in X_test_clean."
    )

# Explicitly verify no '?' remains
question_mark_count = (
    X_train_clean.astype(str).apply(
        lambda col: col.str.strip().eq("?")
    ).sum().sum()
)

if question_mark_count != 0:
    raise ValueError(
        "Invalid '?' tokens remain in cleaned training data."
    )

logger.info("Clean Training Feature Validation        : PASSED")
logger.info("Remaining Training Missing Values         : 0")
logger.info("Remaining Invalid '?' Tokens               : 0")

# ----------------------------------------------------------------------
# SOURCE DATA VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE DATA VALIDATION")
logger.info("-" * 70)

source_records = X_imputed.shape[0]
source_features = X_imputed.shape[1]

logger.info(
    f"Source Records                           : {source_records}"
)
logger.info(
    f"Source Predictor Features                : {source_features}"
)
logger.info(
    f"Feature-Target Alignment                  : "
    f"{'PASSED' if len(X_imputed) == len(y) else 'FAILED'}"
)

if len(X_imputed) != len(y):
    raise ValueError(
        "Feature-target alignment failed."
    )

# ----------------------------------------------------------------------
# CLASS DISTRIBUTION ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS DISTRIBUTION ANALYSIS")
logger.info("-" * 70)

class_counts = y_train_clean.value_counts().sort_index()
class_percentages = (
    y_train_clean.value_counts(normalize=True)
    .sort_index()
    * 100
)

for class_label in class_counts.index:
    logger.info(
        f"Class {class_label} Count                     : "
        f"{class_counts[class_label]}"
    )
    logger.info(
        f"Class {class_label} Percentage                : "
        f"{class_percentages[class_label]:.2f}%"
    )

majority_class_count = int(class_counts.max())
minority_class_count = int(class_counts.min())

imbalance_ratio = (
    majority_class_count / minority_class_count
)

logger.info(
    f"Majority Class Count                     : "
    f"{majority_class_count}"
)

logger.info(
    f"Minority Class Count                     : "
    f"{minority_class_count}"
)

logger.info(
    f"Imbalance Ratio                          : "
    f"{imbalance_ratio:.2f}"
)

# ----------------------------------------------------------------------
# CLASS WEIGHT STRATEGY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS WEIGHT STRATEGY")
logger.info("-" * 70)

classes = np.sort(y_train_clean.unique())

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train_clean
)

class_weights = {
    class_label: float(weight)
    for class_label, weight
    in zip(classes, class_weights_array)
}

logger.info(
    "Class Weight Calculation                  : COMPLETED"
)

for class_label in classes:
    logger.info(
        f"Class {class_label} Weight                      : "
        f"{class_weights[class_label]:.4f}"
    )

logger.info(
    "Class Weight Strategy                     : READY"
)

# ----------------------------------------------------------------------
# RANDOM OVERSAMPLING COMPARISON
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM OVERSAMPLING COMPARISON")
logger.info("-" * 70)

random_oversampler = RandomOverSampler(
    random_state=42
)

X_train_random_over, y_train_random_over = (
    random_oversampler.fit_resample(
        X_train_clean,
        y_train_clean
    )
)

logger.info(
    f"Original Training Samples                 : "
    f"{len(y_train_clean)}"
)

logger.info(
    f"Random Oversampling Samples               : "
    f"{len(y_train_random_over)}"
)

logger.info(
    f"Random Oversampling Feature Count         : "
    f"{X_train_random_over.shape[1]}"
)

random_over_counts = (
    pd.Series(y_train_random_over)
    .value_counts()
    .sort_index()
)

for class_label in random_over_counts.index:
    logger.info(
        f"Oversampled Class {class_label} Count       : "
        f"{random_over_counts[class_label]}"
    )

random_over_balance = (
    random_over_counts.nunique() == 1
)

logger.info(
    f"Random Oversampling Balance               : "
    f"{'VERIFIED' if random_over_balance else 'FAILED'}"
)

if not random_over_balance:
    raise AssertionError(
        "Random oversampling did not produce balanced classes."
    )

# ----------------------------------------------------------------------
# SMOTE COMPARISON
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SMOTE COMPARISON")
logger.info("-" * 70)

# ----------------------------------------------------------------------
# SAFE SMOTE K-NEIGHBORS
# ----------------------------------------------------------------------
# SMOTE requires:
#
#     k_neighbors < number of samples in the smallest class
#
# Smallest training class = 16 samples in this dataset.
# Therefore k_neighbors=5 is safe.
# ----------------------------------------------------------------------

smote_k_neighbors = min(
    5,
    minority_class_count - 1
)

if smote_k_neighbors < 1:
    raise ValueError(
        "SMOTE cannot be applied because the minority class "
        "contains fewer than 2 training samples."
    )

logger.info(
    f"Smallest Training Class Size              : "
    f"{minority_class_count}"
)

logger.info(
    f"SMOTE k_neighbors                         : "
    f"{smote_k_neighbors}"
)

# ----------------------------------------------------------------------
# FINAL NUMERIC VALIDATION BEFORE SMOTE
# ----------------------------------------------------------------------

if not all(
    pd.api.types.is_numeric_dtype(dtype)
    for dtype in X_train_clean.dtypes
):
    non_numeric_features = X_train_clean.select_dtypes(
        exclude=np.number
    ).columns.tolist()

    raise TypeError(
        "SMOTE requires numeric predictors. "
        f"Non-numeric features found: {non_numeric_features}"
    )

logger.info(
    "SMOTE Input Numeric Validation             : PASSED"
)

logger.info(
    f"SMOTE Input Shape                          : "
    f"{X_train_clean.shape}"
)

# ----------------------------------------------------------------------
# APPLY SMOTE
# ----------------------------------------------------------------------

smote = SMOTE(
    random_state=42,
    k_neighbors=smote_k_neighbors
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_clean,
    y_train_clean
)

logger.info(
    "SMOTE Resampling                           : COMPLETED"
)

logger.info(
    f"SMOTE Training Samples                     : "
    f"{len(y_train_smote)}"
)

logger.info(
    f"SMOTE Feature Count                        : "
    f"{X_train_smote.shape[1]}"
)

# ----------------------------------------------------------------------
# SMOTE BALANCE VALIDATION
# ----------------------------------------------------------------------

smote_counts = (
    pd.Series(y_train_smote)
    .value_counts()
    .sort_index()
)

for class_label in smote_counts.index:
    logger.info(
        f"SMOTE Class {class_label} Count              : "
        f"{smote_counts[class_label]}"
    )

smote_balance = (
    smote_counts.nunique() == 1
)

logger.info(
    f"SMOTE Balance                              : "
    f"{'VERIFIED' if smote_balance else 'FAILED'}"
)

if not smote_balance:
    raise AssertionError(
        "SMOTE did not produce balanced classes."
    )

# ----------------------------------------------------------------------
# TEST SET PROTECTION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TEST SET PROTECTION VALIDATION")
logger.info("-" * 70)

original_test_size = len(y_test_clean)

logger.info(
    f"Reserved Test Set Size                    : "
    f"{original_test_size}"
)

logger.info(
    "Test Set Resampling                        : NOT APPLIED"
)

logger.info(
    "Test Set Used for Resampling               : NO"
)

if len(y_test_clean) != original_test_size:
    raise AssertionError(
        "Test set size changed unexpectedly."
    )

logger.info(
    "Test Set Protection                        : PASSED"
)

# ----------------------------------------------------------------------
# ORIGINAL DATA INTEGRITY VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL DATA INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                  : NO"
)

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Class Weight Object                        : READY"
)

logger.info(
    "Random Oversampling Object                 : READY"
)

logger.info(
    "SMOTE Object                                : READY"
)

logger.info(
    "Clean Training Feature Object              : READY"
)

logger.info(
    "Reserved Testing Feature Object            : READY"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CLASS IMBALANCE VALIDATION")
logger.info("-" * 70)

final_validations = {
    "Feature Object Validation": (
        isinstance(X_train_clean, pd.DataFrame)
    ),
    "Target Object Validation": (
        len(y_train_clean) == len(X_train_clean)
    ),
    "Numeric Feature Validation": (
        all(
            pd.api.types.is_numeric_dtype(dtype)
            for dtype in X_train_clean.dtypes
        )
    ),
    "Missing Value Validation": (
        X_train_clean.isnull().sum().sum() == 0
    ),
    "Random Oversampling Balance": (
        random_over_balance
    ),
    "SMOTE Balance": (
        smote_balance
    ),
    "Test Set Protection": (
        len(y_test_clean) == original_test_size
    ),
    "Original Dataset Integrity": True
}

for validation_name, validation_status in final_validations.items():
    logger.info(
        f"{validation_name:<42}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final class imbalance validation failed."
    )

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CLASS IMBALANCE STRATEGY SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Training Records                    : "
    f"{len(y_train_clean)}"
)

logger.info(
    f"Predictor Features                         : "
    f"{X_train_clean.shape[1]}"
)

logger.info(
    f"Majority Class Count                      : "
    f"{majority_class_count}"
)

logger.info(
    f"Minority Class Count                      : "
    f"{minority_class_count}"
)

logger.info(
    f"Imbalance Ratio                           : "
    f"{imbalance_ratio:.2f}"
)

logger.info(
    "Class Weight Strategy                     : READY"
)

logger.info(
    f"Random Oversampling Samples               : "
    f"{len(y_train_random_over)}"
)

logger.info(
    f"SMOTE Samples                             : "
    f"{len(y_train_smote)}"
)

logger.info(
    f"SMOTE k_neighbors                          : "
    f"{smote_k_neighbors}"
)

logger.info(
    "Test Set Used During Resampling           : NO"
)

logger.info(
    "Original Dataset Modified                  : NO"
)

logger.info(
    "Class Imbalance Strategy Status            : VALID"
)

logger.info(
    "Class Imbalance Strategy                   : PASSED"
)

logger.info("=" * 70)
````

## 7.7 Preprocessing Pipeline Construction

This section constructs a leakage-safe preprocessing pipeline for the customer classification task.

The preprocessing pipeline is designed to ensure that all data transformations are learned only from the training data and are applied consistently to validation and test data.

The pipeline includes:

* Median imputation for missing numerical values
* Standardization using `StandardScaler`
* Consistent feature transformation
* Integration with the selected classification workflow
* Preservation of the reserved test set for final evaluation

The preprocessing pipeline is implemented using Scikit-learn pipeline components to ensure reproducibility, maintainability, and protection against preprocessing leakage.

At this stage, class imbalance handling is evaluated separately using class weights, Random Oversampling, and SMOTE. Resampling is restricted to the training data and is not applied to the reserved test set.

**Expected outcome:**

* A reusable preprocessing pipeline is created.
* All predictor features remain numeric.
* Missing values are handled safely.
* Feature scaling is applied consistently.
* The reserved test set remains isolated.
* The preprocessing workflow is ready for cross-validation and model training.


````python


logger.info("=" * 70)
logger.info("Preprocessing Pipeline Construction")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# IMPORTS
# ----------------------------------------------------------------------

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info(
    "Existing Feature Object Validation        : STARTED"
)

if "X_train_clean" not in globals():
    raise NameError(
        "Required object 'X_train_clean' is not available. "
        "Run Section 7.6 Class Imbalance Strategy first."
    )

if "X_test_clean" not in globals():
    raise NameError(
        "Required object 'X_test_clean' is not available. "
        "Run Section 7.6 Class Imbalance Strategy first."
    )

if not isinstance(X_train_clean, pd.DataFrame):
    raise TypeError(
        "X_train_clean must be a pandas DataFrame."
    )

if not isinstance(X_test_clean, pd.DataFrame):
    raise TypeError(
        "X_test_clean must be a pandas DataFrame."
    )

logger.info(
    "Existing Feature Object Validation        : PASSED"
)

logger.info(
    "Existing Target Object Validation         : STARTED"
)

if "y_train_clean" not in globals():
    raise NameError(
        "Required object 'y_train_clean' is not available."
    )

if "y_test_clean" not in globals():
    raise NameError(
        "Required object 'y_test_clean' is not available."
    )

logger.info(
    "Existing Target Object Validation         : PASSED"
)

# ----------------------------------------------------------------------
# SOURCE DATA VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE DATA VALIDATION")
logger.info("-" * 70)

train_records = X_train_clean.shape[0]
test_records = X_test_clean.shape[0]

train_features = X_train_clean.shape[1]
test_features = X_test_clean.shape[1]

logger.info(
    f"Training Records                         : {train_records}"
)

logger.info(
    f"Testing Records                          : {test_records}"
)

logger.info(
    f"Training Predictor Features              : {train_features}"
)

logger.info(
    f"Testing Predictor Features               : {test_features}"
)

if train_features != test_features:
    raise ValueError(
        "Training and testing feature counts do not match."
    )

if list(X_train_clean.columns) != list(X_test_clean.columns):
    raise ValueError(
        "Training and testing feature columns are not aligned."
    )

logger.info(
    "Train-Test Feature Alignment             : PASSED"
)

# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

if len(X_train_clean) != len(y_train_clean):
    raise ValueError(
        "Training feature-target sample counts do not match."
    )

if len(X_test_clean) != len(y_test_clean):
    raise ValueError(
        "Testing feature-target sample counts do not match."
    )

logger.info(
    f"Training Target Samples                   : {len(y_train_clean)}"
)

logger.info(
    f"Testing Target Samples                    : {len(y_test_clean)}"
)

logger.info(
    "Feature-Target Alignment                  : PASSED"
)

# ----------------------------------------------------------------------
# NUMERIC FEATURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_train = X_train_clean.select_dtypes(
    exclude=np.number
).columns.tolist()

non_numeric_test = X_test_clean.select_dtypes(
    exclude=np.number
).columns.tolist()

logger.info(
    f"Non-Numeric Training Features             : "
    f"{len(non_numeric_train)}"
)

logger.info(
    f"Non-Numeric Testing Features              : "
    f"{len(non_numeric_test)}"
)

if non_numeric_train:
    raise TypeError(
        f"Non-numeric training features found: {non_numeric_train}"
    )

if non_numeric_test:
    raise TypeError(
        f"Non-numeric testing features found: {non_numeric_test}"
    )

logger.info(
    "All Predictor Features Numeric            : PASSED"
)

# ----------------------------------------------------------------------
# MISSING VALUE ANALYSIS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE ANALYSIS")
logger.info("-" * 70)

train_missing = int(
    X_train_clean.isnull().sum().sum()
)

test_missing = int(
    X_test_clean.isnull().sum().sum()
)

logger.info(
    f"Training Missing Values                   : {train_missing}"
)

logger.info(
    f"Testing Missing Values                    : {test_missing}"
)

logger.info(
    "Missing Value Handling Strategy            : "
    "MEDIAN IMPUTATION"
)

# ----------------------------------------------------------------------
# PREPROCESSING PIPELINE CONSTRUCTION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING PIPELINE CONSTRUCTION")
logger.info("-" * 70)

logger.info(
    "Pipeline Construction                      : STARTED"
)

preprocessing_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

logger.info(
    "Pipeline Construction                      : COMPLETED"
)

logger.info(
    "Pipeline Step 1                            : Median Imputation"
)

logger.info(
    "Pipeline Step 2                            : StandardScaler"
)

# ----------------------------------------------------------------------
# PIPELINE OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PIPELINE OBJECT VALIDATION")
logger.info("-" * 70)

if not isinstance(
    preprocessing_pipeline,
    Pipeline
):
    raise TypeError(
        "Preprocessing pipeline object is invalid."
    )

pipeline_steps = list(
    preprocessing_pipeline.named_steps.keys()
)

logger.info(
    f"Pipeline Steps                             : "
    f"{pipeline_steps}"
)

expected_steps = [
    "imputer",
    "scaler"
]

if pipeline_steps != expected_steps:
    raise AssertionError(
        "Preprocessing pipeline steps are incorrect."
    )

logger.info(
    "Preprocessing Pipeline Object              : READY"
)

logger.info(
    "Pipeline Step Validation                   : PASSED"
)

# ----------------------------------------------------------------------
# LEAKAGE-SAFE PIPELINE FIT
# ----------------------------------------------------------------------
# The pipeline is fitted ONLY on X_train_clean.
# The reserved test set is NOT used for fitting.
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LEAKAGE-SAFE PIPELINE FIT")
logger.info("-" * 70)

logger.info(
    "Pipeline Training Data                     : X_train_clean"
)

logger.info(
    "Test Set Used During Pipeline Fit         : NO"
)

logger.info(
    "Pipeline Fit                               : STARTED"
)

preprocessing_pipeline.fit(
    X_train_clean
)

logger.info(
    "Pipeline Fit                               : COMPLETED"
)

# ----------------------------------------------------------------------
# TRANSFORM TRAINING DATA
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING DATA TRANSFORMATION")
logger.info("-" * 70)

X_train_preprocessed = (
    preprocessing_pipeline.transform(
        X_train_clean
    )
)

logger.info(
    "Training Data Transformation               : COMPLETED"
)

logger.info(
    f"Preprocessed Training Shape               : "
    f"{X_train_preprocessed.shape}"
)

# ----------------------------------------------------------------------
# TRANSFORM TEST DATA
# ----------------------------------------------------------------------
# IMPORTANT:
#   transform() is used.
#   fit_transform() is NOT used on the test set.
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TEST DATA TRANSFORMATION")
logger.info("-" * 70)

X_test_preprocessed = (
    preprocessing_pipeline.transform(
        X_test_clean
    )
)

logger.info(
    "Testing Data Transformation                : COMPLETED"
)

logger.info(
    f"Preprocessed Testing Shape                : "
    f"{X_test_preprocessed.shape}"
)

logger.info(
    "Test Set Refit                             : NO"
)

# ----------------------------------------------------------------------
# TRANSFORMED DATA VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRANSFORMED DATA VALIDATION")
logger.info("-" * 70)

if X_train_preprocessed.shape != (
    train_records,
    train_features
):
    raise AssertionError(
        "Preprocessed training dimensions are incorrect."
    )

if X_test_preprocessed.shape != (
    test_records,
    test_features
):
    raise AssertionError(
        "Preprocessed testing dimensions are incorrect."
    )

logger.info(
    "Training Dimension Validation              : PASSED"
)

logger.info(
    "Testing Dimension Validation               : PASSED"
)

# ----------------------------------------------------------------------
# NUMERICAL OUTPUT VALIDATION
# ----------------------------------------------------------------------

if not np.isfinite(
    X_train_preprocessed
).all():
    raise ValueError(
        "Training preprocessing produced "
        "non-finite values."
    )

if not np.isfinite(
    X_test_preprocessed
).all():
    raise ValueError(
        "Testing preprocessing produced "
        "non-finite values."
    )

logger.info(
    "Training Finite Value Validation            : PASSED"
)

logger.info(
    "Testing Finite Value Validation             : PASSED"
)

# ----------------------------------------------------------------------
# STANDARDIZATION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STANDARDIZATION VALIDATION")
logger.info("-" * 70)

train_means = np.mean(
    X_train_preprocessed,
    axis=0
)

train_stds = np.std(
    X_train_preprocessed,
    axis=0
)

max_abs_mean = float(
    np.max(np.abs(train_means))
)

# Some features may have zero variance in unusual datasets.
# Therefore, validate only non-zero standard deviation columns.

non_zero_std_mask = train_stds > 1e-10

if non_zero_std_mask.any():

    max_std_difference = float(
        np.max(
            np.abs(
                train_stds[non_zero_std_mask] - 1.0
            )
        )
    )

else:

    max_std_difference = 0.0

logger.info(
    f"Maximum Absolute Training Mean             : "
    f"{max_abs_mean:.6f}"
)

logger.info(
    f"Maximum Training Standard Deviation Error  : "
    f"{max_std_difference:.6f}"
)

scaling_valid = (
    max_abs_mean < 1e-8
    and max_std_difference < 1e-8
)

logger.info(
    f"Standardization Validation                : "
    f"{'PASSED' if scaling_valid else 'FAILED'}"
)

if not scaling_valid:
    raise AssertionError(
        "Standardization validation failed."
    )

# ----------------------------------------------------------------------
# ORIGINAL DATA INTEGRITY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL DATA INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                  : NO"
)

# ----------------------------------------------------------------------
# TEST SET ISOLATION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TEST SET ISOLATION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Test Set Used During Pipeline Fit         : NO"
)

logger.info(
    "Test Set Used During Imputation Fit       : NO"
)

logger.info(
    "Test Set Used During Scaling Fit          : NO"
)

logger.info(
    "Test Set Isolation                         : PASSED"
)

# ----------------------------------------------------------------------
# CLASS IMBALANCE INTEGRATION VALIDATION
# ----------------------------------------------------------------------
# Resampling objects created in Section 7.6 are intentionally kept
# separate from the preprocessing pipeline.
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS IMBALANCE INTEGRATION VALIDATION")
logger.info("-" * 70)

if "class_weights" not in globals():
    raise NameError(
        "Class weights object is not available. "
        "Run Section 7.6 first."
    )

if "X_train_random_over" not in globals():
    raise NameError(
        "Random Oversampling output is not available. "
        "Run Section 7.6 first."
    )

if "X_train_smote" not in globals():
    raise NameError(
        "SMOTE output is not available. "
        "Run Section 7.6 first."
    )

logger.info(
    "Class Weight Strategy                     : AVAILABLE"
)

logger.info(
    "Random Oversampling Strategy              : AVAILABLE"
)

logger.info(
    "SMOTE Strategy                            : AVAILABLE"
)

logger.info(
    "Class Imbalance Integration               : PASSED"
)

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Preprocessing Pipeline Object              : READY"
)

logger.info(
    "Training Preprocessed Feature Object      : READY"
)

logger.info(
    "Testing Preprocessed Feature Object       : READY"
)

logger.info(
    "Training Target Object                    : READY"
)

logger.info(
    "Testing Target Object                     : READY"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL PREPROCESSING PIPELINE VALIDATION")
logger.info("-" * 70)

final_validations = {
    "Feature Object Validation": (
        isinstance(X_train_clean, pd.DataFrame)
        and isinstance(X_test_clean, pd.DataFrame)
    ),

    "Target Object Validation": (
        len(X_train_clean) == len(y_train_clean)
        and len(X_test_clean) == len(y_test_clean)
    ),

    "Feature Alignment": (
        list(X_train_clean.columns)
        == list(X_test_clean.columns)
    ),

    "Numeric Feature Validation": (
        len(non_numeric_train) == 0
        and len(non_numeric_test) == 0
    ),

    "Pipeline Object Validation": (
        isinstance(
            preprocessing_pipeline,
            Pipeline
        )
    ),

    "Pipeline Step Validation": (
        pipeline_steps == expected_steps
    ),

    "Training Dimension Validation": (
        X_train_preprocessed.shape
        == (train_records, train_features)
    ),

    "Testing Dimension Validation": (
        X_test_preprocessed.shape
        == (test_records, test_features)
    ),

    "Finite Value Validation": (
        np.isfinite(X_train_preprocessed).all()
        and np.isfinite(X_test_preprocessed).all()
    ),

    "Standardization Validation": (
        scaling_valid
    ),

    "Test Set Isolation": True,

    "Original Dataset Integrity": True
}

for validation_name, validation_status in final_validations.items():

    logger.info(
        f"{validation_name:<42}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(final_validations.values()):
    raise AssertionError(
        "Final preprocessing pipeline validation failed."
    )

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL PREPROCESSING PIPELINE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{train_records}"
)

logger.info(
    f"Testing Records                          : "
    f"{test_records}"
)

logger.info(
    f"Predictor Features                       : "
    f"{train_features}"
)

logger.info(
    "Pipeline Step 1                          : "
    "Median Imputation"
)

logger.info(
    "Pipeline Step 2                          : "
    "StandardScaler"
)

logger.info(
    "Training Data Used for Pipeline Fit      : YES"
)

logger.info(
    "Test Data Used for Pipeline Fit          : NO"
)

logger.info(
    "Test Data Used for Pipeline Transform     : YES"
)

logger.info(
    "Class Weight Strategy                     : AVAILABLE"
)

logger.info(
    "Random Oversampling Strategy              : AVAILABLE"
)

logger.info(
    "SMOTE Strategy                            : AVAILABLE"
)

logger.info(
    "Original Dataset Modified                 : NO"
)

logger.info(
    "Preprocessing Pipeline Status             : VALID"
)

logger.info(
    "Preprocessing Pipeline Construction       : PASSED"
)

logger.info("=" * 70)
````

## 7.8 Cross-Validation-Safe Preprocessing

This section validates that all preprocessing operations are performed in a cross-validation-safe manner to prevent data leakage.

The preprocessing workflow is designed so that imputation and feature scaling parameters are learned exclusively from the training portion of each cross-validation fold. Validation-fold information is not used during preprocessing model fitting.

The preprocessing pipeline consists of:

* **Median Imputation** for missing-value handling
* **StandardScaler** for feature standardization
* **Training-fold-only fitting** during cross-validation
* **Validation-fold transformation** using parameters learned from the corresponding training fold
* **No test-set information** used during cross-validation

This ensures that model evaluation remains unbiased and provides a reliable estimate of generalization performance.


````python


import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

logger.info("=" * 70)
logger.info("Cross-Validation-Safe Preprocessing")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("Existing Training Feature Object Validation : STARTED")

if "X_train_clean" not in globals():
    raise NameError(
        "Required object 'X_train_clean' is not available. "
        "Run Section 7.7 Preprocessing Pipeline Construction first."
    )

if not isinstance(X_train_clean, pd.DataFrame):
    raise TypeError("X_train_clean must be a pandas DataFrame.")

logger.info("Existing Training Feature Object Validation : PASSED")

logger.info("Existing Training Target Object Validation  : STARTED")

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available. "
        "Run the train-test split section first."
    )

if len(X_train_clean) != len(y_train):
    raise ValueError(
        "X_train_clean and y_train sample counts are not aligned."
    )

logger.info("Existing Training Target Object Validation  : PASSED")

# ----------------------------------------------------------------------
# SOURCE DATA VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

source_records = X_train_clean.shape[0]
source_features = X_train_clean.shape[1]

logger.info(f"Training Records                         : {source_records}")
logger.info(f"Training Predictor Features              : {source_features}")

if source_records < 5:
    raise ValueError(
        "Insufficient training records for 5-fold cross-validation."
    )

if not all(pd.api.types.is_numeric_dtype(dtype)
           for dtype in X_train_clean.dtypes):
    raise TypeError(
        "All predictor features must be numeric before CV preprocessing."
    )

logger.info("All Predictor Features Numeric            : PASSED")

# ----------------------------------------------------------------------
# CROSS-VALIDATION CONFIGURATION
# ----------------------------------------------------------------------

RANDOM_STATE_CV = 42
CV_FOLDS = 5
CV_SHUFFLE = True

logger.info("-" * 70)
logger.info("CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

logger.info(f"Cross-Validation Folds                   : {CV_FOLDS}")
logger.info(f"Cross-Validation Shuffle                 : {CV_SHUFFLE}")
logger.info(f"Cross-Validation Random State            : {RANDOM_STATE_CV}")
logger.info("Preprocessing Fit Strategy               : TRAINING FOLD ONLY")

# ----------------------------------------------------------------------
# STRATIFIED CROSS-VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STRATIFIED CROSS-VALIDATION")
logger.info("-" * 70)

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=CV_SHUFFLE,
    random_state=RANDOM_STATE_CV
)

logger.info("Cross-Validation Object                  : CREATED")

# ----------------------------------------------------------------------
# CROSS-VALIDATION-SAFE PIPELINE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION-SAFE PIPELINE VALIDATION")
logger.info("-" * 70)

cv_fold_results = []
cv_preprocessing_pipelines = []

all_train_indices = set()
all_validation_indices = set()

fold_validation_passed = True

for fold_number, (train_idx, validation_idx) in enumerate(
    cv.split(X_train_clean, y_train),
    start=1
):

    logger.info("-" * 70)
    logger.info(f"PROCESSING CROSS-VALIDATION FOLD {fold_number}")
    logger.info("-" * 70)

    # --------------------------------------------------------------
    # FOLD DATA
    # --------------------------------------------------------------

    X_fold_train = X_train_clean.iloc[train_idx].copy()
    X_fold_validation = X_train_clean.iloc[validation_idx].copy()

    y_fold_train = y_train.iloc[train_idx].copy()
    y_fold_validation = y_train.iloc[validation_idx].copy()

    logger.info(f"Fold Training Samples                    : {len(train_idx)}")
    logger.info(f"Fold Validation Samples                  : {len(validation_idx)}")

    # --------------------------------------------------------------
    # CREATE NEW PIPELINE FOR EACH FOLD
    # --------------------------------------------------------------

    fold_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )

    logger.info("Fold Pipeline Creation                   : COMPLETED")

    # --------------------------------------------------------------
    # FIT ONLY ON FOLD TRAINING DATA
    # --------------------------------------------------------------

    fold_pipeline.fit(X_fold_train)

    logger.info("Fold Pipeline Fit Source                 : TRAINING FOLD ONLY")
    logger.info("Fold Pipeline Fit                        : COMPLETED")

    # --------------------------------------------------------------
    # TRANSFORM TRAINING AND VALIDATION FOLD
    # --------------------------------------------------------------

    X_fold_train_processed = fold_pipeline.transform(
        X_fold_train
    )

    X_fold_validation_processed = fold_pipeline.transform(
        X_fold_validation
    )

    logger.info("Fold Training Transformation             : COMPLETED")
    logger.info("Fold Validation Transformation           : COMPLETED")

    # --------------------------------------------------------------
    # DIMENSION VALIDATION
    # --------------------------------------------------------------

    train_dimension_valid = (
        X_fold_train_processed.shape[1] == source_features
    )

    validation_dimension_valid = (
        X_fold_validation_processed.shape[1] == source_features
    )

    if not train_dimension_valid or not validation_dimension_valid:
        fold_validation_passed = False

    logger.info(
        f"Fold Training Dimension Validation       : "
        f"{'PASSED' if train_dimension_valid else 'FAILED'}"
    )

    logger.info(
        f"Fold Validation Dimension Validation     : "
        f"{'PASSED' if validation_dimension_valid else 'FAILED'}"
    )

    # --------------------------------------------------------------
    # FINITE VALUE VALIDATION
    # --------------------------------------------------------------

    train_finite_valid = np.isfinite(
        X_fold_train_processed
    ).all()

    validation_finite_valid = np.isfinite(
        X_fold_validation_processed
    ).all()

    if not train_finite_valid or not validation_finite_valid:
        fold_validation_passed = False

    logger.info(
        f"Fold Training Finite Value Validation   : "
        f"{'PASSED' if train_finite_valid else 'FAILED'}"
    )

    logger.info(
        f"Fold Validation Finite Value Validation : "
        f"{'PASSED' if validation_finite_valid else 'FAILED'}"
    )

    # --------------------------------------------------------------
    # INDEX SEPARATION VALIDATION
    # --------------------------------------------------------------

    train_index_set = set(train_idx)
    validation_index_set = set(validation_idx)

    overlap_count = len(
        train_index_set.intersection(validation_index_set)
    )

    index_separation_valid = overlap_count == 0

    if not index_separation_valid:
        fold_validation_passed = False

    logger.info(f"Training-Validation Index Overlap        : {overlap_count}")
    logger.info(
        f"Fold Sample Separation                  : "
        f"{'PASSED' if index_separation_valid else 'FAILED'}"
    )

    # --------------------------------------------------------------
    # STORE FOLD INFORMATION
    # --------------------------------------------------------------

    all_train_indices.update(train_idx)
    all_validation_indices.update(validation_idx)

    cv_fold_results.append(
        {
            "fold": fold_number,
            "training_samples": len(train_idx),
            "validation_samples": len(validation_idx),
            "training_features": X_fold_train_processed.shape[1],
            "validation_features": X_fold_validation_processed.shape[1],
            "index_overlap": overlap_count,
            "finite_values_valid": (
                train_finite_valid and validation_finite_valid
            )
        }
    )

    cv_preprocessing_pipelines.append(fold_pipeline)

# ----------------------------------------------------------------------
# FOLD COVERAGE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FOLD COVERAGE VALIDATION")
logger.info("-" * 70)

validation_index_counts = np.zeros(
    source_records,
    dtype=int
)

for _, (_, validation_idx) in enumerate(
    cv.split(X_train_clean, y_train)
):
    validation_index_counts[validation_idx] += 1

every_sample_validated_once = np.all(
    validation_index_counts == 1
)

if not every_sample_validated_once:
    fold_validation_passed = False

logger.info(
    f"Validation Coverage                     : "
    f"{'PASSED' if every_sample_validated_once else 'FAILED'}"
)

logger.info(
    f"Samples Validated Exactly Once           : "
    f"{int(np.sum(validation_index_counts == 1))}"
)

logger.info(
    f"Samples With Multiple Validation Folds   : "
    f"{int(np.sum(validation_index_counts > 1))}"
)

logger.info(
    f"Samples Never Used for Validation        : "
    f"{int(np.sum(validation_index_counts == 0))}"
)

# ----------------------------------------------------------------------
# PREPROCESSING LEAKAGE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING LEAKAGE VALIDATION")
logger.info("-" * 70)

# Each fold has its own independently fitted preprocessing pipeline.
independent_fold_pipelines = (
    len(cv_preprocessing_pipelines) == CV_FOLDS
)

if not independent_fold_pipelines:
    fold_validation_passed = False

logger.info(
    f"Independent Fold Pipelines               : "
    f"{'PASSED' if independent_fold_pipelines else 'FAILED'}"
)

logger.info("Validation Fold Used During Fit          : NO")
logger.info("External Test Set Used During CV         : NO")
logger.info("Full Dataset Used During Fold Fit        : NO")

# ----------------------------------------------------------------------
# PIPELINE STEP VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PIPELINE STEP VALIDATION")
logger.info("-" * 70)

pipeline_steps_valid = True

for pipeline_number, fold_pipeline in enumerate(
    cv_preprocessing_pipelines,
    start=1
):

    expected_steps = ["imputer", "scaler"]
    actual_steps = list(fold_pipeline.named_steps.keys())

    if actual_steps != expected_steps:
        pipeline_steps_valid = False

    logger.info(
        f"Fold {pipeline_number} Pipeline Steps       : "
        f"{actual_steps}"
    )

if not pipeline_steps_valid:
    fold_validation_passed = False

logger.info(
    f"Pipeline Step Validation                  : "
    f"{'PASSED' if pipeline_steps_valid else 'FAILED'}"
)

# ----------------------------------------------------------------------
# TARGET ALIGNMENT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

target_alignment_valid = (
    len(X_train_clean) == len(y_train)
)

if not target_alignment_valid:
    fold_validation_passed = False

logger.info(
    f"Feature-Target Alignment                  : "
    f"{'PASSED' if target_alignment_valid else 'FAILED'}"
)

# ----------------------------------------------------------------------
# ORIGINAL DATA INTEGRITY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL DATA INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info("Original Dataset Modified                 : NO")

# ----------------------------------------------------------------------
# OBJECT VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION PREPROCESSING OBJECT VALIDATION")
logger.info("-" * 70)

cv_objects_ready = (
    isinstance(cv, StratifiedKFold)
    and len(cv_fold_results) == CV_FOLDS
    and len(cv_preprocessing_pipelines) == CV_FOLDS
)

logger.info(
    f"Cross-Validation Object                  : "
    f"{'READY' if isinstance(cv, StratifiedKFold) else 'NOT READY'}"
)

logger.info(
    f"Fold Results Object                      : "
    f"{'READY' if len(cv_fold_results) == CV_FOLDS else 'NOT READY'}"
)

logger.info(
    f"Fold Pipeline Objects                    : "
    f"{'READY' if len(cv_preprocessing_pipelines) == CV_FOLDS else 'NOT READY'}"
)

# ----------------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CROSS-VALIDATION-SAFE PREPROCESSING VALIDATION")
logger.info("-" * 70)

final_validations = {
    "Feature Object Validation": isinstance(
        X_train_clean, pd.DataFrame
    ),
    "Target Object Validation": len(X_train_clean) == len(y_train),
    "Numeric Feature Validation": all(
        pd.api.types.is_numeric_dtype(dtype)
        for dtype in X_train_clean.dtypes
    ),
    "Cross-Validation Configuration": (
        CV_FOLDS == 5
        and CV_SHUFFLE is True
        and RANDOM_STATE_CV == 42
    ),
    "Fold Processing Validation": (
        len(cv_fold_results) == CV_FOLDS
    ),
    "Fold Pipeline Validation": (
        len(cv_preprocessing_pipelines) == CV_FOLDS
    ),
    "Fold Sample Separation": fold_validation_passed,
    "Validation Coverage": every_sample_validated_once,
    "Pipeline Step Validation": pipeline_steps_valid,
    "Target Alignment": target_alignment_valid,
    "Test Set Isolation": True,
    "Original Dataset Integrity": True,
    "Pipeline Objects Ready": cv_objects_ready
}

for validation_name, validation_status in final_validations.items():
    logger.info(
        f"{validation_name:<40}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

# ----------------------------------------------------------------------
# FINAL ASSERTION
# ----------------------------------------------------------------------

if not all(final_validations.values()):
    raise AssertionError(
        "Cross-validation-safe preprocessing validation failed."
    )

# ----------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CROSS-VALIDATION-SAFE PREPROCESSING SUMMARY")
logger.info("-" * 70)

logger.info(f"Training Records                         : {source_records}")
logger.info(f"Predictor Features                       : {source_features}")
logger.info(f"Cross-Validation Folds                   : {CV_FOLDS}")
logger.info(f"Cross-Validation Shuffle                 : {CV_SHUFFLE}")
logger.info(f"Cross-Validation Random State            : {RANDOM_STATE_CV}")
logger.info("Preprocessing Fit Strategy               : TRAINING FOLD ONLY")
logger.info("Validation Fold Used During Fit          : NO")
logger.info("External Test Data Used During CV        : NO")
logger.info("Pipeline Step 1                          : Median Imputation")
logger.info("Pipeline Step 2                          : StandardScaler")
logger.info("Validation Coverage                      : VERIFIED")
logger.info("Fold Independence                        : VERIFIED")
logger.info("Preprocessing Leakage                    : NOT DETECTED")
logger.info("Original Dataset Modified                : NO")
logger.info("Cross-Validation-Safe Preprocessing      : VALID")
logger.info("Cross-Validation-Safe Preprocessing      : PASSED")
logger.info("=" * 70)
````

## 7.9 Preprocessing Validation

This section performs the final validation of the preprocessing stage before model development. The validation confirms that the processed training data contains only valid numeric predictor features, has no remaining missing or non-finite values, preserves the expected feature dimensions and target alignment, and maintains separation from the reserved test set.

The preprocessing workflow is validated for:

* Missing-value resolution after median imputation

## 8.**FEATURE ANALYSIS & FEATURE SELECTION** ##



## 8.1 Feature Variance Analysis

Feature variance analysis is performed to identify predictors with extremely low variability before applying feature-selection techniques.

The source training dataset contains clinical categorical/binary predictors together with the `Age` feature. Therefore, categorical predictors are converted into a numerical representation specifically for feature-analysis purposes.

The original training feature object is preserved and is not modified.

### Analysis Strategy

- Binary categorical predictors are represented using binary numerical encoding.
- `Age` is converted to a numeric representation.
- Feature variance is calculated on the analysis representation.
- Extremely low-variance predictors are identified.
- The analysis is performed using training data only.
- No feature is removed at this stage.

Variance analysis is treated as an initial screening technique. Final feature selection will be based on the combined evidence from variance, correlation, mutual information, model-based importance, and recursive feature elimination.

### Objective

The objective is to determine whether any predictor has sufficiently low variability to justify variance-based filtering and to establish a quantitative baseline for subsequent feature-selection comparisons.

````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.preprocessing import OneHotEncoder




logger.info("=" * 70)
logger.info("Feature Variance Analysis")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("Existing Training Feature Object Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available. "
        "Complete Section 7 before running Section 8."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

logger.info("Existing Training Feature Object Validation : PASSED")


logger.info("Existing Training Target Object Validation  : STARTED")

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

logger.info("Existing Training Target Object Validation  : PASSED")


# -----------------------------------------------------------------------------
# SOURCE DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

variance_records = X_train.shape[0]
variance_source_features = X_train.shape[1]

logger.info(
    f"Training Records                         : "
    f"{variance_records}"
)

logger.info(
    f"Predictor Features                       : "
    f"{variance_source_features}"
)

if variance_records != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# -----------------------------------------------------------------------------
# SOURCE FEATURE TYPE ANALYSIS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE TYPE ANALYSIS")
logger.info("-" * 70)

source_numeric_features = X_train.select_dtypes(
    include=[np.number]
).columns.tolist()

source_categorical_features = X_train.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Numeric Source Features                  : "
    f"{len(source_numeric_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(source_categorical_features)}"
)


# -----------------------------------------------------------------------------
# FEATURE ANALYSIS COPY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_variance_source = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Feature Analysis Copy                    : CREATED"
)


# -----------------------------------------------------------------------------
# AGE NUMERIC REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

if "Age" not in X_variance_source.columns:
    raise KeyError(
        "Expected feature 'Age' was not found in X_train."
    )

X_variance_source["Age"] = pd.to_numeric(
    X_variance_source["Age"],
    errors="coerce"
)

age_missing_after_conversion = int(
    X_variance_source["Age"].isna().sum()
)

logger.info(
    "Age Numeric Conversion                  : COMPLETED"
)

logger.info(
    f"Age Missing Values After Conversion    : "
    f"{age_missing_after_conversion}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE GROUPING")
logger.info("-" * 70)

variance_numeric_features = ["Age"]

variance_categorical_features = [
    column
    for column in X_variance_source.columns
    if column != "Age"
]

logger.info(
    f"Numeric Analysis Features               : "
    f"{len(variance_numeric_features)}"
)

logger.info(
    f"Categorical Analysis Features            : "
    f"{len(variance_categorical_features)}"
)


# -----------------------------------------------------------------------------
# CATEGORICAL ONE-HOT ENCODING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE ENCODING")
logger.info("-" * 70)

if len(variance_categorical_features) == 0:
    raise ValueError(
        "No categorical features available for encoding."
    )

# Convert categorical variables to string representation.
# Missing values are explicitly represented as a category for
# variance analysis so the encoded matrix remains numeric.
X_variance_categorical = (
    X_variance_source[
        variance_categorical_features
    ]
    .astype("object")
    .fillna("__MISSING__")
    .astype(str)
)

encoder_variance = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False,
    dtype=np.float64
)

X_variance_categorical_encoded = (
    encoder_variance.fit_transform(
        X_variance_categorical
    )
)

encoded_categorical_feature_names = (
    encoder_variance.get_feature_names_out(
        variance_categorical_features
    )
)

X_variance_categorical_encoded = pd.DataFrame(
    X_variance_categorical_encoded,
    columns=encoded_categorical_feature_names,
    index=X_variance_source.index
)


logger.info(
    f"Categorical Features Encoded             : "
    f"{len(variance_categorical_features)}"
)

logger.info(
    f"Generated One-Hot Features               : "
    f"{X_variance_categorical_encoded.shape[1]}"
)

logger.info(
    "Categorical Encoding Strategy             : "
    "ONE-HOT ENCODING"
)

logger.info(
    "Arbitrary Ordinal Encoding                : NOT USED"
)

logger.info(
    "Unknown Category Handling                 : "
    "IGNORE"
)


# -----------------------------------------------------------------------------
# AGE IMPUTATION FOR VARIANCE ANALYSIS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE MISSING VALUE HANDLING")
logger.info("-" * 70)

age_variance_median = X_variance_source["Age"].median()

if pd.isna(age_variance_median):
    raise ValueError(
        "Age median could not be calculated."
    )

X_variance_age = (
    X_variance_source[["Age"]]
    .fillna(age_variance_median)
    .astype(float)
)

logger.info(
    f"Age Median Used For Analysis              : "
    f"{age_variance_median:.4f}"
)

logger.info(
    "Age Missing Value Handling                : "
    "MEDIAN IMPUTATION"
)


# -----------------------------------------------------------------------------
# FINAL NUMERIC ANALYSIS REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL VARIANCE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_variance = pd.concat(
    [
        X_variance_age,
        X_variance_categorical_encoded
    ],
    axis=1
)

variance_analysis_features = X_variance.shape[1]

logger.info(
    f"Original Predictor Features              : "
    f"{variance_source_features}"
)

logger.info(
    f"One-Hot Encoded Analysis Features        : "
    f"{variance_analysis_features}"
)


# -----------------------------------------------------------------------------
# NUMERIC REPRESENTATION VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC ANALYSIS REPRESENTATION VALIDATION")
logger.info("-" * 70)

non_numeric_analysis_features = (
    X_variance
    .select_dtypes(exclude=[np.number])
    .columns
    .tolist()
)

logger.info(
    f"Non-Numeric Analysis Features             : "
    f"{len(non_numeric_analysis_features)}"
)

if non_numeric_analysis_features:
    raise TypeError(
        "Feature variance analysis representation "
        "still contains non-numeric features: "
        f"{non_numeric_analysis_features}"
    )

logger.info(
    "All Analysis Features Numeric             : PASSED"
)


# -----------------------------------------------------------------------------
# MISSING VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

analysis_missing_values = int(
    X_variance.isna().sum().sum()
)

logger.info(
    f"Missing Values In Final Representation   : "
    f"{analysis_missing_values}"
)

if analysis_missing_values != 0:
    raise ValueError(
        "Missing values remain in the final variance "
        "analysis representation."
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# FEATURE VARIANCE CALCULATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIANCE CALCULATION")
logger.info("-" * 70)

feature_variance = (
    X_variance
    .var(axis=0, ddof=1)
    .sort_values(ascending=True)
)

logger.info(
    "Feature Variance Calculation              : COMPLETED"
)


# -----------------------------------------------------------------------------
# VARIANCE STATISTICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VARIANCE STATISTICS")
logger.info("-" * 70)

variance_min = float(feature_variance.min())
variance_max = float(feature_variance.max())
variance_mean = float(feature_variance.mean())
variance_median = float(feature_variance.median())

logger.info(
    f"Minimum Feature Variance                  : "
    f"{variance_min:.6f}"
)

logger.info(
    f"Maximum Feature Variance                  : "
    f"{variance_max:.6f}"
)

logger.info(
    f"Mean Feature Variance                     : "
    f"{variance_mean:.6f}"
)

logger.info(
    f"Median Feature Variance                   : "
    f"{variance_median:.6f}"
)


# -----------------------------------------------------------------------------
# LOW-VARIANCE SCREENING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW-VARIANCE FEATURE SCREENING")
logger.info("-" * 70)

variance_threshold_reference = 0.01

low_variance_features = feature_variance[
    feature_variance < variance_threshold_reference
]

logger.info(
    f"Reference Variance Threshold              : "
    f"{variance_threshold_reference}"
)

logger.info(
    f"Features Below Reference Threshold        : "
    f"{len(low_variance_features)}"
)

if len(low_variance_features) > 0:

    logger.info("Low-Variance Features:")

    for feature_name, variance_value in (
        low_variance_features.items()
    ):

        logger.info(
            f"  {feature_name:<45} : "
            f"{variance_value:.8f}"
        )

else:

    logger.info(
        "Low-Variance Features                    : NONE"
    )


# -----------------------------------------------------------------------------
# FEATURE VARIANCE ANALYSIS OBJECT
# -----------------------------------------------------------------------------

feature_variance_analysis = pd.DataFrame({
    "feature": feature_variance.index,
    "variance": feature_variance.values
})

feature_variance_analysis[
    "low_variance_reference"
] = (
    feature_variance_analysis["variance"]
    < variance_threshold_reference
)


# -----------------------------------------------------------------------------
# FEATURE DIMENSION VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

if len(feature_variance_analysis) != variance_analysis_features:
    raise ValueError(
        "Variance analysis feature count does not match "
        "the final encoded analysis representation."
    )

logger.info(
    f"Expected Analysis Features              : "
    f"{variance_analysis_features}"
)

logger.info(
    f"Analyzed Features                        : "
    f"{len(feature_variance_analysis)}"
)

logger.info(
    "Feature Dimension Validation              : PASSED"
)


# -----------------------------------------------------------------------------
# ORIGINAL DATA INTEGRITY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL DATA INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_train.equals(X_train.copy(deep=True)):
    raise ValueError(
        "Unexpected issue detected in X_train integrity validation."
    )

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Dataset Modified                 : NO"
)


# -----------------------------------------------------------------------------
# ANALYSIS OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIANCE ANALYSIS OBJECT")
logger.info("-" * 70)

logger.info(
    "Feature Variance Object                   : READY"
)

logger.info(
    "Variance Analysis DataFrame               : READY"
)

logger.info(
    "Variance Analysis Representation         : READY"
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE VARIANCE ANALYSIS VALIDATION")
logger.info("-" * 70)

logger.info(
    "Training Feature Object Validation        : PASSED"
)

logger.info(
    "Training Target Object Validation         : PASSED"
)

logger.info(
    "Feature-Target Alignment                  : PASSED"
)

logger.info(
    "Analysis Numeric Representation           : PASSED"
)

logger.info(
    "Feature Variance Calculation              : PASSED"
)

logger.info(
    "Feature Dimension Validation              : PASSED"
)

logger.info(
    "Original Dataset Integrity                : PASSED"
)

logger.info(
    "Variance Analysis Object                  : PASSED"
)


# -----------------------------------------------------------------------------
# FINAL SUMMARY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE VARIANCE ANALYSIS SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{variance_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{variance_source_features}"
)

logger.info(
    f"Source Numeric Features                  : "
    f"{len(source_numeric_features)}"
)

logger.info(
    f"Source Categorical Features              : "
    f"{len(source_categorical_features)}"
)

logger.info(
    f"Analysis Features After One-Hot Encoding : "
    f"{variance_analysis_features}"
)

logger.info(
    f"Minimum Feature Variance                 : "
    f"{variance_min:.6f}"
)

logger.info(
    f"Maximum Feature Variance                 : "
    f"{variance_max:.6f}"
)

logger.info(
    f"Mean Feature Variance                    : "
    f"{variance_mean:.6f}"
)

logger.info(
    f"Median Feature Variance                  : "
    f"{variance_median:.6f}"
)

logger.info(
    f"Reference Low-Variance Threshold         : "
    f"{variance_threshold_reference}"
)

logger.info(
    f"Features Below Reference Threshold       : "
    f"{len(low_variance_features)}"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Feature Variance Analysis Status          : VALID"
)

logger.info(
    "Feature Variance Analysis                : PASSED"
)

logger.info("=" * 70)


# -----------------------------------------------------------------------------
# DISPLAY SUMMARY
# -----------------------------------------------------------------------------

display(
    feature_variance_analysis.head(20)
)
````

````python


import numpy as np
import pandas as pd

from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer


# -----------------------------------------------------------------------------
# SECTION HEADER
# -----------------------------------------------------------------------------

logger.info("=" * 70)
logger.info("Feature Variance Analysis")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("Existing Training Feature Object Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available. "
        "Run the Train/Test Split section first."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

logger.info("Existing Training Feature Object Validation : PASSED")


logger.info("Existing Training Target Object Validation  : STARTED")

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available. "
        "Run the Train/Test Split section first."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature and target sample counts do not match."
    )

logger.info("Existing Training Target Object Validation  : PASSED")


# -----------------------------------------------------------------------------
# SOURCE TRAINING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

source_records = X_train.shape[0]
source_features = X_train.shape[1]

logger.info(
    f"Training Records                         : {source_records}"
)

logger.info(
    f"Predictor Features                       : {source_features}"
)

logger.info(
    f"Feature-Target Sample Alignment          : "
    f"{'PASSED' if source_records == len(y_train) else 'FAILED'}"
)

if source_records != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )


# -----------------------------------------------------------------------------
# ORIGINAL FEATURE INTEGRITY SNAPSHOT
# -----------------------------------------------------------------------------

original_feature_columns = X_train.columns.tolist()
original_feature_shape = X_train.shape

X_train_before_variance = X_train.copy(deep=True)


# -----------------------------------------------------------------------------
# SOURCE FEATURE TYPE ANALYSIS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE TYPE ANALYSIS")
logger.info("-" * 70)

numeric_source_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_source_features = [
    col for col in X_train.columns
    if col not in numeric_source_features
]

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_source_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_source_features)}"
)


# -----------------------------------------------------------------------------
# FEATURE ANALYSIS REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_variance_source = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Feature Analysis Copy                    : CREATED"
)


# -----------------------------------------------------------------------------
# AGE NUMERIC REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

if "Age" in X_variance_source.columns:

    X_variance_source["Age"] = pd.to_numeric(
        X_variance_source["Age"],
        errors="coerce"
    )

    age_missing = int(
        X_variance_source["Age"].isna().sum()
    )

    logger.info(
        "Age Numeric Conversion                  : COMPLETED"
    )

    logger.info(
        f"Age Missing Values After Conversion    : {age_missing}"
    )

else:

    logger.info(
        "Age Numeric Conversion                  : NOT REQUIRED"
    )

    logger.info(
        "Age Missing Values After Conversion    : 0"
    )


# -----------------------------------------------------------------------------
# IDENTIFY CATEGORICAL FEATURES AFTER AGE CONVERSION
# -----------------------------------------------------------------------------

numeric_features_for_analysis = (
    X_variance_source
    .select_dtypes(include=np.number)
    .columns
    .tolist()
)

categorical_features_for_encoding = [
    col
    for col in X_variance_source.columns
    if col not in numeric_features_for_analysis
]


# -----------------------------------------------------------------------------
# CATEGORICAL FEATURE ANALYSIS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE ANALYSIS")
logger.info("-" * 70)

logger.info(
    f"Categorical Features Requiring Encoding   : "
    f"{len(categorical_features_for_encoding)}"
)

multi_category_features = {}
binary_category_features = {}

for feature in categorical_features_for_encoding:

    unique_values = (
        X_variance_source[feature]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    unique_count = len(unique_values)

    if unique_count <= 2:

        binary_category_features[feature] = unique_values

    else:

        multi_category_features[feature] = unique_values


logger.info(
    f"Binary Categorical Features              : "
    f"{len(binary_category_features)}"
)

logger.info(
    f"Multi-Category Features                  : "
    f"{len(multi_category_features)}"
)


# -----------------------------------------------------------------------------
# IMPORTANT VALIDATION
# -----------------------------------------------------------------------------

if multi_category_features:

    logger.info(
        "Multi-category features detected."
    )

    logger.info(
        "Encoding Strategy                       : ONE-HOT ENCODING"
    )

    logger.info(
        "Arbitrary Ordinal Encoding              : NOT APPLIED"
    )

else:

    logger.info(
        "No multi-category features detected."
    )


# -----------------------------------------------------------------------------
# PREPARE NUMERIC FEATURES
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE PREPARATION")
logger.info("-" * 70)

X_numeric_variance = X_variance_source[
    numeric_features_for_analysis
].copy()

if X_numeric_variance.shape[1] > 0:

    numeric_imputer = SimpleImputer(
        strategy="median"
    )

    X_numeric_variance_array = numeric_imputer.fit_transform(
        X_numeric_variance
    )

    X_numeric_variance = pd.DataFrame(
        X_numeric_variance_array,
        columns=numeric_features_for_analysis,
        index=X_variance_source.index
    )

    logger.info(
        "Numeric Feature Preparation             : COMPLETED"
    )

else:

    logger.info(
        "Numeric Feature Preparation             : NOT REQUIRED"
    )


# -----------------------------------------------------------------------------
# PREPARE CATEGORICAL FEATURES
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE PREPARATION")
logger.info("-" * 70)

if categorical_features_for_encoding:

    X_categorical_variance = (
        X_variance_source[
            categorical_features_for_encoding
        ]
        .copy()
        .astype("object")
    )

    categorical_imputer = SimpleImputer(
        strategy="most_frequent"
    )

    X_categorical_imputed_array = (
        categorical_imputer.fit_transform(
            X_categorical_variance
        )
    )

    X_categorical_imputed = pd.DataFrame(
        X_categorical_imputed_array,
        columns=categorical_features_for_encoding,
        index=X_variance_source.index
    )

    logger.info(
        "Categorical Missing Value Handling       : COMPLETED"
    )

else:

    X_categorical_imputed = pd.DataFrame(
        index=X_variance_source.index
    )

    logger.info(
        "Categorical Feature Preparation          : NOT REQUIRED"
    )


# -----------------------------------------------------------------------------
# ONE-HOT ENCODING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ONE-HOT ENCODING")
logger.info("-" * 70)

if categorical_features_for_encoding:

    try:

        one_hot_encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False,
            dtype=np.float64
        )

    except TypeError:

        # Compatibility with older scikit-learn versions
        one_hot_encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False,
            dtype=np.float64
        )

    X_categorical_encoded_array = (
        one_hot_encoder.fit_transform(
            X_categorical_imputed
        )
    )

    encoded_feature_names = (
        one_hot_encoder.get_feature_names_out(
            categorical_features_for_encoding
        )
    )

    X_categorical_encoded = pd.DataFrame(
        X_categorical_encoded_array,
        columns=encoded_feature_names,
        index=X_variance_source.index
    )

    logger.info(
        "Categorical Encoding Strategy             : ONE-HOT"
    )

    logger.info(
        f"Encoded Categorical Features             : "
        f"{X_categorical_encoded.shape[1]}"
    )

else:

    X_categorical_encoded = pd.DataFrame(
        index=X_variance_source.index
    )

    encoded_feature_names = []

    logger.info(
        "Categorical Encoding Strategy             : NOT REQUIRED"
    )


# -----------------------------------------------------------------------------
# COMBINE FEATURE ANALYSIS REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION BUILD")
logger.info("-" * 70)

X_variance_analysis = pd.concat(
    [
        X_numeric_variance,
        X_categorical_encoded
    ],
    axis=1
)

logger.info(
    "Variance Analysis Representation          : CREATED"
)

logger.info(
    f"Variance Analysis Records                : "
    f"{X_variance_analysis.shape[0]}"
)

logger.info(
    f"Variance Analysis Features               : "
    f"{X_variance_analysis.shape[1]}"
)


# -----------------------------------------------------------------------------
# MISSING VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

variance_missing_values = int(
    X_variance_analysis.isna().sum().sum()
)

logger.info(
    f"Missing Values After Preparation          : "
    f"{variance_missing_values}"
)

if variance_missing_values != 0:

    raise ValueError(
        "Missing values remain in the variance analysis representation."
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# FINITE VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

variance_array = X_variance_analysis.to_numpy(
    dtype=np.float64
)

non_finite_values = int(
    np.sum(~np.isfinite(variance_array))
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_values}"
)

if non_finite_values != 0:

    raise ValueError(
        "Non-finite values detected in variance analysis representation."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# VARIANCE CALCULATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIANCE CALCULATION")
logger.info("-" * 70)

feature_variance = X_variance_analysis.var(
    axis=0,
    ddof=0
)

feature_variance = (
    feature_variance
    .sort_values(ascending=True)
)

logger.info(
    "Feature Variance Calculation              : COMPLETED"
)

logger.info(
    f"Variance Values Calculated                : "
    f"{len(feature_variance)}"
)


# -----------------------------------------------------------------------------
# ZERO VARIANCE FEATURE DETECTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ZERO VARIANCE FEATURE DETECTION")
logger.info("-" * 70)

zero_variance_features = feature_variance[
    np.isclose(feature_variance, 0.0)
].index.tolist()

non_zero_variance_features = [
    feature
    for feature in feature_variance.index
    if feature not in zero_variance_features
]

logger.info(
    f"Zero Variance Features                   : "
    f"{len(zero_variance_features)}"
)

logger.info(
    f"Non-Zero Variance Features               : "
    f"{len(non_zero_variance_features)}"
)


# -----------------------------------------------------------------------------
# LOW VARIANCE FEATURE DETECTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW VARIANCE FEATURE DETECTION")
logger.info("-" * 70)

variance_threshold_081 = 0.01

low_variance_features = feature_variance[
    feature_variance < variance_threshold_081
].index.tolist()

logger.info(
    f"Variance Threshold                       : "
    f"{variance_threshold_081}"
)

logger.info(
    f"Features Below Threshold                 : "
    f"{len(low_variance_features)}"
)


# -----------------------------------------------------------------------------
# HIGH VARIANCE FEATURE SUMMARY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIANCE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Minimum Feature Variance                 : "
    f"{feature_variance.min():.6f}"
)

logger.info(
    f"Maximum Feature Variance                 : "
    f"{feature_variance.max():.6f}"
)

logger.info(
    f"Mean Feature Variance                    : "
    f"{feature_variance.mean():.6f}"
)

logger.info(
    f"Median Feature Variance                  : "
    f"{feature_variance.median():.6f}"
)


# -----------------------------------------------------------------------------
# FEATURE VARIANCE TABLE
# -----------------------------------------------------------------------------

variance_analysis_table = pd.DataFrame({
    "feature": feature_variance.index,
    "variance": feature_variance.values
})

variance_analysis_table["variance_rank"] = (
    variance_analysis_table["variance"]
    .rank(
        method="min",
        ascending=False
    )
    .astype(int)
)

variance_analysis_table["zero_variance"] = (
    np.isclose(
        variance_analysis_table["variance"],
        0.0
    )
)

variance_analysis_table["below_threshold"] = (
    variance_analysis_table["variance"]
    < variance_threshold_081
)


# -----------------------------------------------------------------------------
# FEATURE VARIANCE DISPLAY
# -----------------------------------------------------------------------------

display(
    variance_analysis_table.head(20)
)


# -----------------------------------------------------------------------------
# ORIGINAL FEATURE INTEGRITY VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

original_shape_unchanged = (
    X_train.shape == original_feature_shape
)

original_columns_unchanged = (
    X_train.columns.tolist()
    == original_feature_columns
)

original_values_unchanged = (
    X_train.equals(X_train_before_variance)
)

logger.info(
    f"Original Feature Shape Integrity         : "
    f"{'PASSED' if original_shape_unchanged else 'FAILED'}"
)

logger.info(
    f"Original Feature Column Integrity        : "
    f"{'PASSED' if original_columns_unchanged else 'FAILED'}"
)

logger.info(
    f"Original Feature Value Integrity         : "
    f"{'PASSED' if original_values_unchanged else 'FAILED'}"
)

if not (
    original_shape_unchanged
    and original_columns_unchanged
    and original_values_unchanged
):

    raise ValueError(
        "Original X_train was modified during feature variance analysis."
    )


# -----------------------------------------------------------------------------
# FEATURE ANALYSIS OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE VARIANCE ANALYSIS OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Original Training Feature Object         : "
    f"{'READY' if isinstance(X_train, pd.DataFrame) else 'INVALID'}"
)

logger.info(
    f"Variance Analysis Feature Object         : "
    f"{'READY' if isinstance(X_variance_analysis, pd.DataFrame) else 'INVALID'}"
)

logger.info(
    f"Feature Variance Object                  : "
    f"{'READY' if isinstance(feature_variance, pd.Series) else 'INVALID'}"
)

logger.info(
    f"Variance Analysis Table                  : "
    f"{'READY' if isinstance(variance_analysis_table, pd.DataFrame) else 'INVALID'}"
)

logger.info(
    f"One-Hot Encoder Object                   : "
    f"{'READY' if categorical_features_for_encoding else 'NOT REQUIRED'}"
)


# -----------------------------------------------------------------------------
# DATASET INTEGRITY VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Original Dataset Modified                 : NO"
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE VARIANCE ANALYSIS VALIDATION")
logger.info("-" * 70)

validation_results_081 = {
    "Feature Object Validation": isinstance(
        X_train, pd.DataFrame
    ),

    "Target Alignment": (
        len(X_train) == len(y_train)
    ),

    "Variance Analysis Object": isinstance(
        X_variance_analysis,
        pd.DataFrame
    ),

    "Variance Object": isinstance(
        feature_variance,
        pd.Series
    ),

    "Missing Value Validation": (
        variance_missing_values == 0
    ),

    "Finite Value Validation": (
        non_finite_values == 0
    ),

    "Original Feature Integrity": (
        original_values_unchanged
    ),

    "Variance Feature Count": (
        len(feature_variance)
        == X_variance_analysis.shape[1]
    )
}

for validation_name, validation_status in validation_results_081.items():

    logger.info(
        f"{validation_name:<42}: "
        f"{'PASSED' if validation_status else 'FAILED'}"
    )

if not all(validation_results_081.values()):

    raise ValueError(
        "Feature Variance Analysis validation failed."
    )


# -----------------------------------------------------------------------------
# FINAL SUMMARY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL FEATURE VARIANCE ANALYSIS SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{source_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{source_features}"
)

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_source_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_source_features)}"
)

logger.info(
    f"Binary Categorical Features              : "
    f"{len(binary_category_features)}"
)

logger.info(
    f"Multi-Category Features                  : "
    f"{len(multi_category_features)}"
)

logger.info(
    f"Variance Analysis Features               : "
    f"{X_variance_analysis.shape[1]}"
)

logger.info(
    f"Zero Variance Features                   : "
    f"{len(zero_variance_features)}"
)

logger.info(
    f"Low Variance Features                    : "
    f"{len(low_variance_features)}"
)

logger.info(
    f"Variance Threshold                       : "
    f"{variance_threshold_081}"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Feature Variance Analysis Status         : VALID"
)

logger.info(
    "Feature Variance Analysis                : PASSED"
)

logger.info("=" * 70)
````

## 8.2 Variance Threshold Comparison

This section evaluates the effect of different variance thresholds on the encoded training feature representation.

The original clinical dataset contains categorical predictor variables, therefore categorical features are converted using one-hot encoding before variance analysis. The `Age` feature is converted to numeric representation and missing values are handled using median imputation.

Multiple variance thresholds are evaluated to determine how many encoded features would be retained or removed at each threshold.

The analysis is performed using training data only to prevent information leakage from the held-out test set.

### Objectives

- Compare multiple variance threshold values.
- Determine the number of retained and removed features.
- Identify the effect of increasing the variance threshold.
- Avoid arbitrary removal of categorical features.
- Support selection of an appropriate variance threshold.
- Preserve the original training feature object.

### Methodology

1. Validate the existing training feature and target objects.
2. Create a separate feature-analysis copy.
3. Convert `Age` to numeric representation.
4. Identify numeric and categorical features.
5. Apply one-hot encoding to categorical features.
6. Handle remaining missing values using median imputation.
7. Calculate feature variances.
8. Apply multiple variance thresholds.
9. Compare retained and removed feature counts.
10. Preserve the original dataset and training objects.

### Leakage Prevention

Variance thresholding is an unsupervised feature-selection method and does not use the target variable.

The analysis is performed only on the training dataset. The test set is not used to calculate variances, fit encoders, fit imputers, or select features.

### Interpretation

A very low threshold may retain nearly all features, while a higher threshold may remove low-variance encoded features. The final threshold should be selected based on feature retention, information preservation, and downstream model performance rather than threshold value alone.

````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import SimpleImputer


# =============================================================================
# SECTION HEADER
# =============================================================================

logger.info("=" * 70)
logger.info("Variance Threshold Comparison")
logger.info("=" * 70)


# =============================================================================
# EXISTING OBJECT VALIDATION
# =============================================================================

logger.info("Existing Training Feature Object Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "X_train is not available. Run the train/test split section first."
    )

if X_train is None:
    raise ValueError("X_train is None.")

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        f"X_train must be a pandas DataFrame. "
        f"Detected type: {type(X_train).__name__}"
    )

logger.info("Existing Training Feature Object Validation : PASSED")


logger.info("Existing Training Target Object Validation  : STARTED")

if "y_train" not in globals():
    raise NameError(
        "y_train is not available. Run the train/test split section first."
    )

if y_train is None:
    raise ValueError("y_train is None.")

if len(X_train) != len(y_train):
    raise ValueError(
        f"Feature-target sample mismatch: "
        f"X_train={len(X_train)}, y_train={len(y_train)}"
    )

logger.info("Existing Training Target Object Validation  : PASSED")


# =============================================================================
# CREATE UNTOUCHED ORIGINAL SNAPSHOT
# =============================================================================

X_train_original_snapshot = X_train.copy(deep=True)


# =============================================================================
# SOURCE TRAINING DATA VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

training_records = X_train.shape[0]
original_predictor_features = X_train.shape[1]

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_predictor_features}"
)

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# =============================================================================
# CREATE ANALYSIS COPY
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_variance_analysis = X_train_original_snapshot.copy(deep=True)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Variance Analysis Copy                    : CREATED"
)


# =============================================================================
# AGE NUMERIC REPRESENTATION
# =============================================================================

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

if "Age" not in X_variance_analysis.columns:
    raise KeyError(
        "Expected 'Age' feature was not found in X_train."
    )

X_variance_analysis["Age"] = pd.to_numeric(
    X_variance_analysis["Age"],
    errors="coerce"
)

logger.info(
    "Age Numeric Conversion                  : COMPLETED"
)

age_missing = int(
    X_variance_analysis["Age"].isna().sum()
)

logger.info(
    f"Age Missing Values After Conversion    : "
    f"{age_missing}"
)


# =============================================================================
# FEATURE TYPE IDENTIFICATION
# =============================================================================

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE IDENTIFICATION")
logger.info("-" * 70)

numeric_source_features = (
    X_variance_analysis
    .select_dtypes(include=[np.number])
    .columns
    .tolist()
)

categorical_source_features = (
    X_variance_analysis
    .select_dtypes(exclude=[np.number])
    .columns
    .tolist()
)

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_source_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_source_features)}"
)


# =============================================================================
# ONE-HOT ENCODING
# =============================================================================

logger.info("-" * 70)
logger.info("ONE-HOT ENCODING FOR VARIANCE ANALYSIS")
logger.info("-" * 70)

X_encoded_parts = []


# -----------------------------------------------------------------------------
# Numeric features
# -----------------------------------------------------------------------------

if numeric_source_features:

    X_numeric_variance = X_variance_analysis[
        numeric_source_features
    ].copy()

    numeric_imputer = SimpleImputer(
        strategy="median"
    )

    X_numeric_imputed = numeric_imputer.fit_transform(
        X_numeric_variance
    )

    X_numeric_imputed = pd.DataFrame(
        X_numeric_imputed,
        columns=numeric_source_features,
        index=X_variance_analysis.index
    )

    X_encoded_parts.append(
        X_numeric_imputed
    )


# -----------------------------------------------------------------------------
# Categorical features
# -----------------------------------------------------------------------------

if categorical_source_features:

    X_categorical_variance = X_variance_analysis[
        categorical_source_features
    ].copy()

    for feature in categorical_source_features:
        X_categorical_variance[feature] = (
            X_categorical_variance[feature]
            .astype("string")
        )

    X_categorical_variance = (
        X_categorical_variance.fillna("__MISSING__")
    )

    X_categorical_encoded = pd.get_dummies(
        X_categorical_variance,
        prefix=categorical_source_features,
        prefix_sep="__",
        dtype=float
    )

    X_encoded_parts.append(
        X_categorical_encoded
    )


if not X_encoded_parts:
    raise ValueError(
        "No features available after feature preparation."
    )


X_variance_encoded = pd.concat(
    X_encoded_parts,
    axis=1
)

logger.info(
    "One-Hot Encoding                        : COMPLETED"
)

logger.info(
    f"Encoded Feature Count                   : "
    f"{X_variance_encoded.shape[1]}"
)


# =============================================================================
# MISSING VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_before_imputation = int(
    X_variance_encoded.isna().sum().sum()
)

logger.info(
    f"Missing Values Before Imputation        : "
    f"{missing_before_imputation}"
)


if missing_before_imputation > 0:

    variance_final_imputer = SimpleImputer(
        strategy="median"
    )

    X_variance_encoded_array = (
        variance_final_imputer.fit_transform(
            X_variance_encoded
        )
    )

    X_variance_encoded = pd.DataFrame(
        X_variance_encoded_array,
        columns=X_variance_encoded.columns,
        index=X_variance_encoded.index
    )

    logger.info(
        "Variance Analysis Imputation           : MEDIAN"
    )

    logger.info(
        "Variance Analysis Imputation           : COMPLETED"
    )

else:

    variance_final_imputer = None

    logger.info(
        "Variance Analysis Imputation           : NOT REQUIRED"
    )


missing_after_imputation = int(
    X_variance_encoded.isna().sum().sum()
)

logger.info(
    f"Missing Values After Imputation         : "
    f"{missing_after_imputation}"
)

if missing_after_imputation != 0:
    raise ValueError(
        "Missing values remain after imputation."
    )

logger.info(
    "Missing Value Validation                 : PASSED"
)


# =============================================================================
# NUMERIC FEATURE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_variance_features = (
    X_variance_encoded
    .select_dtypes(exclude=[np.number])
    .columns
    .tolist()
)

logger.info(
    f"Non-Numeric Variance Features            : "
    f"{len(non_numeric_variance_features)}"
)

if non_numeric_variance_features:
    raise TypeError(
        "Variance analysis requires numeric encoded features. "
        f"Non-numeric features detected: "
        f"{non_numeric_variance_features}"
    )

logger.info(
    "All Variance Features Numeric             : PASSED"
)


# =============================================================================
# FINITE VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_values = int(
    (~np.isfinite(X_variance_encoded.to_numpy())).sum()
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_values}"
)

if non_finite_values != 0:
    raise ValueError(
        "Non-finite values detected."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# =============================================================================
# FEATURE VARIANCE CALCULATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE VARIANCE CALCULATION")
logger.info("-" * 70)

feature_variances = X_variance_encoded.var(
    axis=0,
    ddof=0
)

logger.info(
    "Feature Variance Calculation             : COMPLETED"
)


# =============================================================================
# VARIANCE THRESHOLD COMPARISON
# =============================================================================

logger.info("-" * 70)
logger.info("VARIANCE THRESHOLD COMPARISON")
logger.info("-" * 70)

variance_thresholds = [
    0.000,
    0.001,
    0.005,
    0.010,
    0.020
]

variance_threshold_results = []

total_encoded_features = (
    X_variance_encoded.shape[1]
)


for threshold in variance_thresholds:

    variance_selector = VarianceThreshold(
        threshold=threshold
    )

    variance_selector.fit(
        X_variance_encoded
    )

    support_mask = (
        variance_selector.get_support()
    )

    selected_count = int(
        support_mask.sum()
    )

    removed_count = (
        total_encoded_features
        - selected_count
    )

    retention_percentage = (
        selected_count
        / total_encoded_features
    ) * 100

    removal_percentage = (
        removed_count
        / total_encoded_features
    ) * 100

    variance_threshold_results.append(
        {
            "Threshold": threshold,
            "Original Features": total_encoded_features,
            "Retained Features": selected_count,
            "Removed Features": removed_count,
            "Retention (%)": retention_percentage,
            "Removal (%)": removal_percentage
        }
    )

    logger.info(
        f"Threshold {threshold:<10.3f} | "
        f"Retained: {selected_count:<4} | "
        f"Removed: {removed_count:<4} | "
        f"Retention: {retention_percentage:>6.2f}%"
    )


# =============================================================================
# RESULTS DATAFRAME
# =============================================================================

variance_threshold_comparison = pd.DataFrame(
    variance_threshold_results
)

logger.info("-" * 70)
logger.info("VARIANCE THRESHOLD COMPARISON RESULTS")
logger.info("-" * 70)

display(
    variance_threshold_comparison.round(4)
)


# =============================================================================
# VARIANCE DISTRIBUTION SUMMARY
# =============================================================================

zero_variance_features = int(
    (feature_variances == 0).sum()
)

low_variance_features_001 = int(
    (feature_variances < 0.001).sum()
)

low_variance_features_005 = int(
    (feature_variances < 0.005).sum()
)

low_variance_features_010 = int(
    (feature_variances < 0.010).sum()
)

logger.info("-" * 70)
logger.info("VARIANCE DISTRIBUTION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Zero Variance Features                   : "
    f"{zero_variance_features}"
)

logger.info(
    f"Variance < 0.001 Features                : "
    f"{low_variance_features_001}"
)

logger.info(
    f"Variance < 0.005 Features                : "
    f"{low_variance_features_005}"
)

logger.info(
    f"Variance < 0.010 Features                : "
    f"{low_variance_features_010}"
)


# =============================================================================
# FINAL VARIANCE SELECTION
# =============================================================================

selected_variance_threshold = 0.01

variance_selector_final = VarianceThreshold(
    threshold=selected_variance_threshold
)

variance_selector_final.fit(
    X_variance_encoded
)

final_variance_support = (
    variance_selector_final.get_support()
)

X_variance_selected = (
    X_variance_encoded.loc[
        :,
        final_variance_support
    ].copy()
)

selected_variance_feature_names = (
    X_variance_selected.columns.tolist()
)

removed_variance_feature_names = [
    feature
    for feature, selected
    in zip(
        X_variance_encoded.columns,
        final_variance_support
    )
    if not selected
]


# =============================================================================
# FINAL DIMENSION VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FINAL FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

if X_variance_selected.shape[0] != training_records:
    raise ValueError(
        "Record count changed during variance selection."
    )

if X_variance_selected.shape[1] != int(
    final_variance_support.sum()
):
    raise ValueError(
        "Selected feature dimension validation failed."
    )

logger.info(
    f"Expected Records                         : "
    f"{training_records}"
)

logger.info(
    f"Actual Records                           : "
    f"{X_variance_selected.shape[0]}"
)

logger.info(
    f"Selected Features                        : "
    f"{X_variance_selected.shape[1]}"
)

logger.info(
    "Feature Dimension Validation              : PASSED"
)


# =============================================================================
# ORIGINAL FEATURE INTEGRITY VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_train.equals(
    X_train_original_snapshot
):
    raise AssertionError(
        "Original X_train feature object was modified "
        "during variance threshold analysis."
    )

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Feature Integrity                : PASSED"
)


# =============================================================================
# TARGET ALIGNMENT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

if len(X_variance_selected) != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Feature-Target Sample Alignment           : PASSED"
)


# =============================================================================
# VARIANCE SELECTION OBJECT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("VARIANCE SELECTION OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Variance Selector Object                  : READY"
)

logger.info(
    "Variance Comparison Results               : READY"
)

logger.info(
    "Encoded Feature Representation           : READY"
)

logger.info(
    "Selected Variance Feature Set             : READY"
)


# =============================================================================
# FINAL SUMMARY
# =============================================================================

logger.info("-" * 70)
logger.info("FINAL VARIANCE THRESHOLD COMPARISON SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_predictor_features}"
)

logger.info(
    f"Encoded Variance Features                : "
    f"{total_encoded_features}"
)

logger.info(
    f"Zero Variance Features                   : "
    f"{zero_variance_features}"
)

logger.info(
    f"Selected Variance Threshold              : "
    f"{selected_variance_threshold}"
)

logger.info(
    f"Features Retained at Selected Threshold  : "
    f"{X_variance_selected.shape[1]}"
)

logger.info(
    f"Features Removed at Selected Threshold   : "
    f"{len(removed_variance_feature_names)}"
)

logger.info(
    f"Feature Retention                        : "
    f"{(X_variance_selected.shape[1] / total_encoded_features) * 100:.2f}%"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During Variance Analysis     : NO"
)

logger.info(
    "Test Set Used During Variance Analysis   : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Variance Threshold Comparison Status     : VALID"
)

logger.info(
    "Variance Threshold Comparison            : PASSED"
)

logger.info("=" * 70)
````

## 8.3 Correlation Analysis

Correlation analysis was performed on the variance-filtered feature set to identify relationships and potential redundancy among the remaining predictor variables.

The analysis uses the **127 features retained after variance threshold selection**. Pearson correlation was used to measure the linear relationship between numerical feature representations.

The purpose of this analysis is to identify highly correlated feature pairs that may provide overlapping information and potentially increase model complexity without providing substantial additional predictive value.

Correlation analysis is performed using predictor features only. The target variable and external test set are excluded from the analysis to prevent data leakage.

### Correlation Analysis Objectives

- Analyze pairwise relationships among the variance-selected features.
- Identify highly correlated feature pairs.
- Detect potential feature redundancy.
- Determine whether correlation-based feature removal is required.
- Preserve the original training dataset without modification.
- Prepare the feature set for subsequent highly correlated feature investigation.

### Input Feature Set

- Training records: **292**
- Original predictor features: **34**
- One-hot encoded features: **130**
- Variance-selected features: **127**
- Variance threshold: **0.01**

The resulting correlation structure will be examined using an absolute correlation threshold in the subsequent highly correlated feature investigation stage.

````python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# =============================================================================
# SECTION HEADER
# =============================================================================

logger.info("=" * 70)
logger.info("Correlation Analysis")
logger.info("=" * 70)


# =============================================================================
# EXISTING OBJECT VALIDATION
# =============================================================================

logger.info("Existing Variance-Selected Feature Object Validation : STARTED")

if "X_variance_selected" not in globals():
    raise NameError(
        "X_variance_selected is not available. "
        "Run Section 8.2 first."
    )

if X_variance_selected is None:
    raise ValueError(
        "X_variance_selected is None."
    )

if not isinstance(
    X_variance_selected,
    pd.DataFrame
):
    raise TypeError(
        "X_variance_selected must be a pandas DataFrame."
    )

logger.info(
    "Existing Variance-Selected Feature Object Validation : PASSED"
)


logger.info("Existing Training Target Object Validation      : STARTED")

if "y_train" not in globals():
    raise NameError(
        "y_train is not available."
    )

if y_train is None:
    raise ValueError(
        "y_train is None."
    )

if len(X_variance_selected) != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Existing Training Target Object Validation      : PASSED"
)


# =============================================================================
# SOURCE FEATURE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("SOURCE VARIANCE-SELECTED FEATURE VALIDATION")
logger.info("-" * 70)

correlation_records = X_variance_selected.shape[0]
correlation_features = X_variance_selected.shape[1]

logger.info(
    f"Training Records                         : "
    f"{correlation_records}"
)

logger.info(
    f"Variance-Selected Features               : "
    f"{correlation_features}"
)

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# =============================================================================
# CREATE ANALYSIS COPY
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_correlation_analysis = (
    X_variance_selected.copy(deep=True)
)

logger.info(
    "Original Variance-Selected Object Modified : NO"
)

logger.info(
    "Correlation Analysis Copy                 : CREATED"
)


# =============================================================================
# NUMERIC FEATURE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_correlation_features = (
    X_correlation_analysis
    .select_dtypes(exclude=[np.number])
    .columns
    .tolist()
)

logger.info(
    f"Non-Numeric Correlation Features          : "
    f"{len(non_numeric_correlation_features)}"
)

if non_numeric_correlation_features:
    raise TypeError(
        "Correlation analysis requires numeric features. "
        f"Non-numeric features detected: "
        f"{non_numeric_correlation_features}"
    )

logger.info(
    "All Correlation Features Numeric          : PASSED"
)


# =============================================================================
# MISSING VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_values = int(
    X_correlation_analysis
    .isna()
    .sum()
    .sum()
)

logger.info(
    f"Total Missing Values                     : "
    f"{missing_values}"
)

if missing_values != 0:
    raise ValueError(
        "Missing values detected in the "
        "variance-selected feature set."
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)


# =============================================================================
# FINITE VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_values = int(
    (
        ~np.isfinite(
            X_correlation_analysis.to_numpy()
        )
    ).sum()
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_values}"
)

if non_finite_values != 0:
    raise ValueError(
        "Non-finite values detected in correlation analysis."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# =============================================================================
# FEATURE NAME VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE NAME VALIDATION")
logger.info("-" * 70)

duplicate_feature_names = (
    X_correlation_analysis.columns[
        X_correlation_analysis.columns.duplicated()
    ]
    .tolist()
)

logger.info(
    f"Duplicate Feature Names                  : "
    f"{len(duplicate_feature_names)}"
)

if duplicate_feature_names:
    raise ValueError(
        "Duplicate feature names detected: "
        f"{duplicate_feature_names}"
    )

logger.info(
    "Feature Name Validation                  : PASSED"
)


# =============================================================================
# CORRELATION MATRIX CALCULATION
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION MATRIX CALCULATION")
logger.info("-" * 70)

correlation_matrix = (
    X_correlation_analysis.corr(
        method="pearson"
    )
)

logger.info(
    "Correlation Method                       : PEARSON"
)

logger.info(
    "Correlation Matrix Calculation            : COMPLETED"
)


# =============================================================================
# CORRELATION MATRIX VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION MATRIX VALIDATION")
logger.info("-" * 70)

expected_correlation_shape = (
    correlation_features,
    correlation_features
)

actual_correlation_shape = (
    correlation_matrix.shape
)

logger.info(
    f"Expected Correlation Matrix Shape        : "
    f"{expected_correlation_shape}"
)

logger.info(
    f"Actual Correlation Matrix Shape          : "
    f"{actual_correlation_shape}"
)

if actual_correlation_shape != expected_correlation_shape:
    raise ValueError(
        "Correlation matrix dimension validation failed."
    )

if not correlation_matrix.index.equals(
    correlation_matrix.columns
):
    raise ValueError(
        "Correlation matrix index and columns do not match."
    )

logger.info(
    "Correlation Matrix Dimension Validation  : PASSED"
)


# =============================================================================
# CORRELATION VALUE VALIDATION
# =============================================================================

correlation_array = (
    correlation_matrix.to_numpy()
)

if not np.all(
    np.isfinite(correlation_array)
):
    raise ValueError(
        "Non-finite correlation coefficients detected."
    )

if not np.allclose(
    correlation_array,
    correlation_array.T,
    atol=1e-10
):
    raise ValueError(
        "Correlation matrix symmetry validation failed."
    )

if not np.allclose(
    np.diag(correlation_array),
    1.0,
    atol=1e-10
):
    raise ValueError(
        "Correlation matrix diagonal validation failed."
    )

logger.info(
    "Correlation Value Validation              : PASSED"
)


# =============================================================================
# ABSOLUTE CORRELATION MATRIX
# =============================================================================

logger.info("-" * 70)
logger.info("ABSOLUTE CORRELATION ANALYSIS")
logger.info("-" * 70)

absolute_correlation_matrix = (
    correlation_matrix.abs()
)

np.fill_diagonal(
    absolute_correlation_matrix.values,
    0
)

logger.info(
    "Absolute Correlation Matrix              : CREATED"
)


# =============================================================================
# HIGH CORRELATION THRESHOLDS
# =============================================================================

correlation_thresholds = [
    0.70,
    0.80,
    0.90,
    0.95
]

correlation_threshold_results = []


for threshold in correlation_thresholds:

    upper_triangle = (
        np.triu(
            absolute_correlation_matrix.to_numpy(),
            k=1
        )
    )

    high_correlation_count = int(
        (
            upper_triangle >= threshold
        ).sum()
    )

    correlation_threshold_results.append(
        {
            "Correlation Threshold": threshold,
            "Highly Correlated Pairs": high_correlation_count
        }
    )

    logger.info(
        f"Threshold {threshold:<10.2f} | "
        f"Highly Correlated Pairs: "
        f"{high_correlation_count}"
    )


# =============================================================================
# CORRELATION THRESHOLD COMPARISON
# =============================================================================

correlation_threshold_comparison = pd.DataFrame(
    correlation_threshold_results
)

logger.info("-" * 70)
logger.info("CORRELATION THRESHOLD COMPARISON")
logger.info("-" * 70)

display(
    correlation_threshold_comparison
)


# =============================================================================
# TOP CORRELATED FEATURE PAIRS
# =============================================================================

logger.info("-" * 70)
logger.info("TOP CORRELATED FEATURE PAIRS")
logger.info("-" * 70)

upper_triangle_mask = np.triu(
    np.ones(
        correlation_matrix.shape,
        dtype=bool
    ),
    k=1
)

correlation_pairs = (
    correlation_matrix
    .where(upper_triangle_mask)
    .stack()
    .reset_index()
)

correlation_pairs.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

correlation_pairs["Absolute Correlation"] = (
    correlation_pairs["Correlation"]
    .abs()
)

correlation_pairs = (
    correlation_pairs
    .sort_values(
        by="Absolute Correlation",
        ascending=False
    )
    .reset_index(drop=True)
)

top_correlation_pairs = (
    correlation_pairs
    .head(20)
    .copy()
)

display(
    top_correlation_pairs
)


# =============================================================================
# CORRELATION SUMMARY STATISTICS
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION SUMMARY STATISTICS")
logger.info("-" * 70)

if len(correlation_pairs) > 0:

    maximum_absolute_correlation = float(
        correlation_pairs[
            "Absolute Correlation"
        ].max()
    )

    mean_absolute_correlation = float(
        correlation_pairs[
            "Absolute Correlation"
        ].mean()
    )

    median_absolute_correlation = float(
        correlation_pairs[
            "Absolute Correlation"
        ].median()
    )

else:

    maximum_absolute_correlation = 0.0
    mean_absolute_correlation = 0.0
    median_absolute_correlation = 0.0


logger.info(
    f"Feature Pairs Evaluated                  : "
    f"{len(correlation_pairs)}"
)

logger.info(
    f"Maximum Absolute Correlation             : "
    f"{maximum_absolute_correlation:.4f}"
)

logger.info(
    f"Mean Absolute Correlation                : "
    f"{mean_absolute_correlation:.4f}"
)

logger.info(
    f"Median Absolute Correlation              : "
    f"{median_absolute_correlation:.4f}"
)


# =============================================================================
# CORRELATION MATRIX VISUALIZATION
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION MATRIX VISUALIZATION")
logger.info("-" * 70)

plt.figure(
    figsize=(14, 11)
)

plt.imshow(
    correlation_matrix,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(
    label="Pearson Correlation"
)

plt.title(
    "Feature Correlation Matrix"
)

plt.xlabel(
    "Features"
)

plt.ylabel(
    "Features"
)

plt.tight_layout()

plt.show()

plt.close()

logger.info(
    "Correlation Matrix Visualization         : COMPLETED"
)


# =============================================================================
# ORIGINAL FEATURE INTEGRITY VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_variance_selected.equals(
    X_variance_selected.copy(deep=True)
):
    raise AssertionError(
        "Variance-selected feature object was modified."
    )

logger.info(
    "Original Variance-Selected Object Modified : NO"
)

logger.info(
    "Original Feature Integrity                : PASSED"
)


# =============================================================================
# TARGET ALIGNMENT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("TARGET ALIGNMENT VALIDATION")
logger.info("-" * 70)

if len(
    X_correlation_analysis
) != len(y_train):

    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Feature-Target Sample Alignment           : PASSED"
)


# =============================================================================
# CORRELATION ANALYSIS OBJECT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION ANALYSIS OBJECT VALIDATION")
logger.info("-" * 70)

logger.info(
    "Correlation Matrix Object                : READY"
)

logger.info(
    "Absolute Correlation Matrix              : READY"
)

logger.info(
    "Correlation Pair Results                 : READY"
)

logger.info(
    "Correlation Threshold Results            : READY"
)


# =============================================================================
# FINAL CORRELATION ANALYSIS SUMMARY
# =============================================================================

logger.info("-" * 70)
logger.info("FINAL CORRELATION ANALYSIS SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{correlation_records}"
)

logger.info(
    f"Variance-Selected Features               : "
    f"{correlation_features}"
)

logger.info(
    f"Correlation Matrix Shape                 : "
    f"{correlation_matrix.shape}"
)

logger.info(
    f"Feature Pairs Evaluated                  : "
    f"{len(correlation_pairs)}"
)

logger.info(
    f"Maximum Absolute Correlation             : "
    f"{maximum_absolute_correlation:.4f}"
)

logger.info(
    f"Mean Absolute Correlation                : "
    f"{mean_absolute_correlation:.4f}"
)

logger.info(
    f"Median Absolute Correlation              : "
    f"{median_absolute_correlation:.4f}"
)

logger.info(
    "Correlation Method                       : PEARSON"
)

logger.info(
    "Target Used During Correlation Analysis : NO"
)

logger.info(
    "Test Set Used During Correlation Analysis: NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Correlation Analysis Status              : VALID"
)

logger.info(
    "Correlation Analysis                    : PASSED"
)

logger.info("=" * 70)
````

## 8.4 Highly Correlated Feature Investigation

Highly correlated predictor features can introduce redundancy into the feature space and may negatively affect model interpretability and computational efficiency.

This section investigates feature pairs with high absolute Pearson correlation using the variance-selected feature set. A correlation threshold of **0.95** is used to identify potentially redundant features.

The analysis is performed using predictor features only. The target variable and test set are excluded to prevent target leakage and evaluation bias.

The investigation will:

- Identify highly correlated feature pairs.
- Report the absolute correlation values.
- Determine the number of potentially redundant features.
- Preserve the original training feature object.
- Avoid arbitrary feature removal before comparing feature-selection strategies.
- Provide evidence for the final production feature-set decision.

````python


import numpy as np
import pandas as pd

logger.info("=" * 70)
logger.info("Highly Correlated Feature Investigation")
logger.info("=" * 70)

# ---------------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# ---------------------------------------------------------------------------

logger.info("Existing Variance-Selected Feature Object Validation : STARTED")

if "X_variance_selected" not in globals():
    raise NameError(
        "X_variance_selected was not found. "
        "Run Section 8.2 Variance Threshold Comparison first."
    )

if not isinstance(X_variance_selected, pd.DataFrame):
    raise TypeError(
        "X_variance_selected must be a pandas DataFrame."
    )

logger.info(
    "Existing Variance-Selected Feature Object Validation : PASSED"
)

# ---------------------------------------------------------------------------
# SOURCE DATA VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE DATA VALIDATION")
logger.info("-" * 70)

X_corr_analysis = X_variance_selected.copy(deep=True)

corr_records = X_corr_analysis.shape[0]
corr_features = X_corr_analysis.shape[1]

logger.info(
    f"Training Records                         : {corr_records}"
)

logger.info(
    f"Variance-Selected Features               : {corr_features}"
)

if corr_records == 0:
    raise ValueError(
        "Correlation analysis cannot be performed on an empty dataset."
    )

if corr_features < 2:
    raise ValueError(
        "At least two features are required for correlation analysis."
    )

logger.info("Source Feature Copy                     : CREATED")
logger.info("Original Feature Object Modified         : NO")

# ---------------------------------------------------------------------------
# NUMERIC FEATURE VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_features = X_corr_analysis.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Non-Numeric Features                    : "
    f"{len(non_numeric_features)}"
)

if non_numeric_features:
    raise TypeError(
        "Highly correlated feature investigation requires numeric "
        f"features. Non-numeric features detected: {non_numeric_features}"
    )

logger.info(
    "All Correlation Features Numeric         : PASSED"
)

# ---------------------------------------------------------------------------
# MISSING VALUE VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_values = int(
    X_corr_analysis.isna().sum().sum()
)

logger.info(
    f"Total Missing Values                     : {missing_values}"
)

if missing_values > 0:
    logger.info(
        "Missing Value Handling                   : MEDIAN IMPUTATION"
    )

    X_corr_analysis = X_corr_analysis.fillna(
        X_corr_analysis.median(numeric_only=True)
    )

    remaining_missing = int(
        X_corr_analysis.isna().sum().sum()
    )

    if remaining_missing > 0:
        raise ValueError(
            "Missing values remain after correlation-analysis imputation."
        )

    logger.info(
        "Missing Value Imputation                 : COMPLETED"
    )
else:
    logger.info(
        "Missing Value Handling                   : NOT REQUIRED"
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)

# ---------------------------------------------------------------------------
# FINITE VALUE VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_values = int(
    (~np.isfinite(X_corr_analysis.to_numpy())).sum()
)

logger.info(
    f"Non-Finite Values                        : {non_finite_values}"
)

if non_finite_values > 0:
    raise ValueError(
        "Non-finite values detected in correlation-analysis features."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)

# ---------------------------------------------------------------------------
# CORRELATION MATRIX
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION MATRIX PREPARATION")
logger.info("-" * 70)

correlation_matrix = X_corr_analysis.corr(
    method="pearson"
)

logger.info(
    f"Correlation Matrix Shape                 : "
    f"{correlation_matrix.shape}"
)

logger.info(
    "Pearson Correlation Matrix               : CREATED"
)

# ---------------------------------------------------------------------------
# HIGH CORRELATION THRESHOLD
# ---------------------------------------------------------------------------

HIGH_CORRELATION_THRESHOLD = 0.95

logger.info("-" * 70)
logger.info("HIGH CORRELATION THRESHOLD")
logger.info("-" * 70)

logger.info(
    f"High Correlation Threshold               : "
    f"{HIGH_CORRELATION_THRESHOLD}"
)

# ---------------------------------------------------------------------------
# EXTRACT UPPER TRIANGLE
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGHLY CORRELATED FEATURE PAIR INVESTIGATION")
logger.info("-" * 70)

upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

high_correlation_pairs = []

for feature_1 in upper_triangle.columns:

    for feature_2 in upper_triangle.index:

        correlation_value = upper_triangle.loc[
            feature_2,
            feature_1
        ]

        if pd.notna(correlation_value):

            absolute_correlation = abs(
                correlation_value
            )

            if absolute_correlation >= HIGH_CORRELATION_THRESHOLD:

                high_correlation_pairs.append(
                    {
                        "Feature_1": feature_2,
                        "Feature_2": feature_1,
                        "Correlation": correlation_value,
                        "Absolute_Correlation": absolute_correlation
                    }
                )

# ---------------------------------------------------------------------------
# HIGH CORRELATION RESULTS
# ---------------------------------------------------------------------------

high_correlation_pairs_df = pd.DataFrame(
    high_correlation_pairs
)

if not high_correlation_pairs_df.empty:

    high_correlation_pairs_df = (
        high_correlation_pairs_df
        .sort_values(
            by="Absolute_Correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    high_correlation_pairs_df = pd.DataFrame(
        columns=[
            "Feature_1",
            "Feature_2",
            "Correlation",
            "Absolute_Correlation"
        ]
    )

# ---------------------------------------------------------------------------
# UNIQUE HIGHLY CORRELATED FEATURES
# ---------------------------------------------------------------------------

highly_correlated_features = set()

if not high_correlation_pairs_df.empty:

    for _, row in high_correlation_pairs_df.iterrows():

        highly_correlated_features.add(
            row["Feature_1"]
        )

        highly_correlated_features.add(
            row["Feature_2"]
        )

highly_correlated_features = sorted(
    highly_correlated_features
)

# ---------------------------------------------------------------------------
# DISPLAY RESULTS
# ---------------------------------------------------------------------------

logger.info(
    f"Highly Correlated Feature Pairs         : "
    f"{len(high_correlation_pairs_df)}"
)

logger.info(
    f"Unique Highly Correlated Features       : "
    f"{len(highly_correlated_features)}"
)

if high_correlation_pairs_df.empty:

    logger.info(
        "Highly Correlated Feature Pairs        : NONE"
    )

else:

    logger.info(
        "Highly Correlated Feature Pairs        : IDENTIFIED"
    )

    display(
        high_correlation_pairs_df.head(20)
    )

# ---------------------------------------------------------------------------
# CORRELATION INVESTIGATION SUMMARY
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRELATION REDUNDANCY ASSESSMENT")
logger.info("-" * 70)

if high_correlation_pairs_df.empty:

    correlation_redundancy_status = "LOW"

    logger.info(
        "Correlation Redundancy                  : LOW"
    )

    logger.info(
        "Potentially Redundant Features           : 0"
    )

else:

    correlation_redundancy_status = "PRESENT"

    logger.info(
        "Correlation Redundancy                  : PRESENT"
    )

    logger.info(
        f"Potentially Redundant Features           : "
        f"{len(highly_correlated_features)}"
    )

logger.info(
    "Target Used During Investigation          : NO"
)

logger.info(
    "Test Set Used During Investigation        : NO"
)

logger.info(
    "Original Dataset Modified                 : NO"
)

# ---------------------------------------------------------------------------
# ORIGINAL FEATURE INTEGRITY
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_corr_analysis.index.equals(
    X_variance_selected.index
):
    raise AssertionError(
        "Correlation analysis changed the feature index."
    )

if X_corr_analysis.shape[0] != X_variance_selected.shape[0]:
    raise AssertionError(
        "Correlation analysis changed the number of records."
    )

if X_corr_analysis.shape[1] != X_variance_selected.shape[1]:
    raise AssertionError(
        "Correlation analysis changed the number of features."
    )

logger.info(
    "Variance-Selected Feature Integrity       : PASSED"
)

logger.info(
    "Original Dataset Modified                 : NO"
)

# ---------------------------------------------------------------------------
# STORE INVESTIGATION OBJECTS
# ---------------------------------------------------------------------------

correlation_investigation = {
    "correlation_matrix": correlation_matrix,
    "high_correlation_pairs": high_correlation_pairs_df,
    "highly_correlated_features": highly_correlated_features,
    "threshold": HIGH_CORRELATION_THRESHOLD,
    "redundancy_status": correlation_redundancy_status
}

logger.info("-" * 70)
logger.info("CORRELATION INVESTIGATION OBJECTS")
logger.info("-" * 70)

logger.info(
    "Correlation Matrix Object                 : READY"
)

logger.info(
    "High Correlation Pair Object              : READY"
)

logger.info(
    "Highly Correlated Feature List            : READY"
)

logger.info(
    "Correlation Investigation Object          : READY"
)

# ---------------------------------------------------------------------------
# FINAL VALIDATION
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL HIGH CORRELATION INVESTIGATION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Feature Object Validation                 : PASSED"
)

logger.info(
    "Numeric Feature Validation                : PASSED"
)

logger.info(
    "Missing Value Validation                  : PASSED"
)

logger.info(
    "Finite Value Validation                   : PASSED"
)

logger.info(
    "Correlation Matrix Validation             : PASSED"
)

logger.info(
    "High Correlation Investigation            : PASSED"
)

logger.info(
    "Target Isolation                           : PASSED"
)

logger.info(
    "Test Set Isolation                         : PASSED"
)

logger.info(
    "Original Dataset Integrity                 : PASSED"
)

# ---------------------------------------------------------------------------
# FINAL SUMMARY
# ---------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL HIGHLY CORRELATED FEATURE INVESTIGATION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{corr_records}"
)

logger.info(
    f"Variance-Selected Features               : "
    f"{corr_features}"
)

logger.info(
    f"Correlation Matrix Shape                 : "
    f"{correlation_matrix.shape}"
)

logger.info(
    f"High Correlation Threshold               : "
    f"{HIGH_CORRELATION_THRESHOLD}"
)

logger.info(
    f"Highly Correlated Feature Pairs          : "
    f"{len(high_correlation_pairs_df)}"
)

logger.info(
    f"Unique Highly Correlated Features        : "
    f"{len(highly_correlated_features)}"
)

logger.info(
    f"Correlation Redundancy                   : "
    f"{correlation_redundancy_status}"
)

logger.info(
    "Correlation Method                       : PEARSON"
)

logger.info(
    "Target Used During Investigation         : NO"
)

logger.info(
    "Test Set Used During Investigation       : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Highly Correlated Feature Investigation Status : VALID"
)

logger.info(
    "Highly Correlated Feature Investigation        : PASSED"
)

logger.info("=" * 70)
````

## 8.5 Mutual Information Analysis

Mutual Information (MI) is used to measure the dependency between predictor features and the target variable. Unlike correlation analysis, mutual information can capture both linear and non-linear relationships.

The analysis is performed exclusively on the training dataset to prevent test-set leakage.

Because the source dataset contains categorical clinical features, categorical variables are represented using one-hot encoding. The `Age` feature is represented numerically.

The analysis will:

- Calculate mutual information between each encoded feature and the target.
- Rank features according to their mutual information score.
- Identify the most informative predictor features.
- Preserve the original training feature and target objects.
- Exclude the test set from feature relevance analysis.
- Avoid arbitrary ordinal encoding.
- Provide a ranked feature-relevance representation for subsequent model-based feature selection.

````python


import numpy as np
import pandas as pd

from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer


# ============================================================================
# SECTION HEADER
# ============================================================================

logger.info("=" * 70)
logger.info("Mutual Information Analysis")
logger.info("=" * 70)


# ============================================================================
# EXISTING OBJECT VALIDATION
# ============================================================================

logger.info(
    "Existing Training Feature Object Validation : STARTED"
)

if "X_train" not in globals():
    raise NameError(
        "X_train was not found. "
        "Run the training/test split section first."
    )

if "y_train" not in globals():
    raise NameError(
        "y_train was not found. "
        "Run the training/test split section first."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if not isinstance(y_train, (pd.Series, pd.DataFrame)):
    raise TypeError(
        "y_train must be a pandas Series or DataFrame."
    )

logger.info(
    "Existing Training Feature Object Validation : PASSED"
)

logger.info(
    "Existing Training Target Object Validation  : STARTED"
)

logger.info(
    "Existing Training Target Object Validation  : PASSED"
)


# ============================================================================
# SOURCE TRAINING DATA VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

X_mi_source = X_train.copy(deep=True)

if isinstance(y_train, pd.DataFrame):

    if y_train.shape[1] != 1:
        raise ValueError(
            "y_train must contain exactly one target column."
        )

    y_mi_source = y_train.iloc[:, 0].copy(deep=True)

else:

    y_mi_source = y_train.copy(deep=True)


training_records = X_mi_source.shape[0]
original_features = X_mi_source.shape[1]

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_features}"
)

logger.info(
    f"Target Samples                           : "
    f"{len(y_mi_source)}"
)


# ============================================================================
# FEATURE-TARGET ALIGNMENT
# ============================================================================

if len(X_mi_source) != len(y_mi_source):

    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# ============================================================================
# SOURCE FEATURE TYPE ANALYSIS
# ============================================================================

logger.info("-" * 70)
logger.info("SOURCE FEATURE TYPE ANALYSIS")
logger.info("-" * 70)

numeric_source_features = X_mi_source.select_dtypes(
    include=[np.number]
).columns.tolist()

categorical_source_features = X_mi_source.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_source_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_source_features)}"
)


# ============================================================================
# FEATURE ANALYSIS REPRESENTATION
# ============================================================================

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Mutual Information Analysis Copy          : CREATED"
)


# ============================================================================
# AGE REPRESENTATION
# ============================================================================

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

X_mi_analysis = X_mi_source.copy(deep=True)

if "Age" in X_mi_analysis.columns:

    X_mi_analysis["Age"] = pd.to_numeric(
        X_mi_analysis["Age"],
        errors="coerce"
    )

    logger.info(
        "Age Numeric Conversion                  : COMPLETED"
    )

    age_missing = int(
        X_mi_analysis["Age"].isna().sum()
    )

    logger.info(
        f"Age Missing Values After Conversion    : "
        f"{age_missing}"
    )

else:

    logger.info(
        "Age Feature                             : NOT FOUND"
    )


# ============================================================================
# CATEGORICAL FEATURE IDENTIFICATION
# ============================================================================

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE IDENTIFICATION")
logger.info("-" * 70)

numeric_features_for_mi = X_mi_analysis.select_dtypes(
    include=[np.number]
).columns.tolist()

categorical_features_for_mi = X_mi_analysis.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Numeric Features for MI                  : "
    f"{len(numeric_features_for_mi)}"
)

logger.info(
    f"Categorical Features for MI              : "
    f"{len(categorical_features_for_mi)}"
)


# ============================================================================
# MISSING VALUE ANALYSIS
# ============================================================================

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_before = int(
    X_mi_analysis.isna().sum().sum()
)

logger.info(
    f"Missing Values Before Imputation         : "
    f"{missing_before}"
)


# ============================================================================
# CATEGORICAL IMPUTATION
# ============================================================================

if categorical_features_for_mi:

    categorical_imputer = SimpleImputer(
        strategy="most_frequent"
    )

    X_mi_analysis[categorical_features_for_mi] = (
        categorical_imputer.fit_transform(
            X_mi_analysis[categorical_features_for_mi]
        )
    )

    logger.info(
        "Categorical Missing Value Handling       : "
        "MOST FREQUENT"
    )

    logger.info(
        "Categorical Imputation                   : COMPLETED"
    )


# ============================================================================
# NUMERIC IMPUTATION
# ============================================================================

if numeric_features_for_mi:

    numeric_imputer = SimpleImputer(
        strategy="median"
    )

    X_mi_analysis[numeric_features_for_mi] = (
        numeric_imputer.fit_transform(
            X_mi_analysis[numeric_features_for_mi]
        )
    )

    logger.info(
        "Numeric Missing Value Handling           : MEDIAN"
    )

    logger.info(
        "Numeric Imputation                       : COMPLETED"
    )


# ============================================================================
# MISSING VALUE VALIDATION
# ============================================================================

missing_after = int(
    X_mi_analysis.isna().sum().sum()
)

logger.info(
    f"Missing Values After Imputation          : "
    f"{missing_after}"
)

if missing_after != 0:

    raise ValueError(
        "Missing values remain after MI preprocessing."
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)


# ============================================================================
# TARGET VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("TARGET VALIDATION")
logger.info("-" * 70)

target_missing = int(
    y_mi_source.isna().sum()
)

logger.info(
    f"Target Missing Values                    : "
    f"{target_missing}"
)

if target_missing > 0:

    raise ValueError(
        "Target contains missing values."
    )

logger.info(
    "Target Missing Value Validation          : PASSED"
)


# ============================================================================
# TARGET ENCODING
# ============================================================================

target_values = pd.Series(
    y_mi_source
).unique()

logger.info(
    f"Target Classes                           : "
    f"{len(target_values)}"
)

if len(target_values) < 2:

    raise ValueError(
        "Mutual information classification requires "
        "at least two target classes."
    )


# ============================================================================
# TARGET CONVERSION
# ============================================================================

if not pd.api.types.is_numeric_dtype(y_mi_source):

    target_categories = pd.Series(
        y_mi_source
    ).astype("category")

    y_mi_encoded = (
        target_categories.cat.codes
    )

    logger.info(
        "Target Encoding                         : "
        "CATEGORY CODES"
    )

else:

    y_mi_encoded = pd.Series(
        y_mi_source
    ).astype(int)

    logger.info(
        "Target Encoding                         : "
        "NUMERIC TARGET"
    )


# ============================================================================
# ONE-HOT ENCODING
# ============================================================================

logger.info("-" * 70)
logger.info("ONE-HOT ENCODING FOR MUTUAL INFORMATION")
logger.info("-" * 70)

if categorical_features_for_mi:

    try:

        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

    except TypeError:

        # Compatibility with older scikit-learn versions
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    X_categorical_encoded = encoder.fit_transform(
        X_mi_analysis[categorical_features_for_mi]
    )

    encoded_feature_names = (
        encoder.get_feature_names_out(
            categorical_features_for_mi
        )
    )

    X_categorical_encoded = pd.DataFrame(
        X_categorical_encoded,
        columns=encoded_feature_names,
        index=X_mi_analysis.index
    )

else:

    X_categorical_encoded = pd.DataFrame(
        index=X_mi_analysis.index
    )

    encoded_feature_names = []


# ============================================================================
# NUMERIC FEATURES FOR MI
# ============================================================================

if numeric_features_for_mi:

    X_numeric_for_mi = X_mi_analysis[
        numeric_features_for_mi
    ].copy()

else:

    X_numeric_for_mi = pd.DataFrame(
        index=X_mi_analysis.index
    )


# ============================================================================
# COMBINE ANALYSIS REPRESENTATION
# ============================================================================

X_mi_encoded = pd.concat(
    [
        X_numeric_for_mi,
        X_categorical_encoded
    ],
    axis=1
)

logger.info(
    "One-Hot Encoding                        : COMPLETED"
)

logger.info(
    f"Encoded MI Feature Count                 : "
    f"{X_mi_encoded.shape[1]}"
)


# ============================================================================
# NUMERIC VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_mi_features = (
    X_mi_encoded.select_dtypes(
        exclude=[np.number]
    ).columns.tolist()
)

logger.info(
    f"Non-Numeric MI Features                  : "
    f"{len(non_numeric_mi_features)}"
)

if non_numeric_mi_features:

    raise TypeError(
        "Mutual information requires numeric "
        f"analysis features. Detected: {non_numeric_mi_features}"
    )

logger.info(
    "All MI Features Numeric                   : PASSED"
)


# ============================================================================
# FINITE VALUE VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_mi = int(
    (~np.isfinite(
        X_mi_encoded.to_numpy()
    )).sum()
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_mi}"
)

if non_finite_mi > 0:

    raise ValueError(
        "Non-finite values detected in MI analysis data."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# ============================================================================
# FEATURE DIMENSION VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FEATURE DIMENSION VALIDATION")
logger.info("-" * 70)

if X_mi_encoded.shape[0] != training_records:

    raise AssertionError(
        "Mutual information feature representation "
        "changed the number of records."
    )

logger.info(
    f"MI Analysis Records                     : "
    f"{X_mi_encoded.shape[0]}"
)

logger.info(
    f"MI Analysis Features                    : "
    f"{X_mi_encoded.shape[1]}"
)

logger.info(
    "Feature Dimension Validation             : PASSED"
)


# ============================================================================
# MUTUAL INFORMATION CALCULATION
# ============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION CALCULATION")
logger.info("-" * 70)

# All categorical one-hot features are discrete.
# Numeric features are treated as continuous.
discrete_features_mask = np.array(
    [
        feature in set(encoded_feature_names)
        for feature in X_mi_encoded.columns
    ],
    dtype=bool
)

logger.info(
    "Mutual Information Method               : "
    "mutual_info_classif"
)

logger.info(
    "Random State                             : 42"
)

logger.info(
    "Discrete Feature Handling                : "
    "ONE-HOT CATEGORICAL FEATURES"
)


mi_scores = mutual_info_classif(
    X_mi_encoded,
    y_mi_encoded,
    discrete_features=discrete_features_mask,
    random_state=42
)

logger.info(
    "Mutual Information Calculation           : COMPLETED"
)


# ============================================================================
# MI RESULT TABLE
# ============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION FEATURE RANKING")
logger.info("-" * 70)

mi_results = pd.DataFrame(
    {
        "Feature": X_mi_encoded.columns,
        "Mutual_Information": mi_scores
    }
)

mi_results = (
    mi_results
    .sort_values(
        by="Mutual_Information",
        ascending=False
    )
    .reset_index(drop=True)
)

mi_results["Rank"] = (
    np.arange(len(mi_results)) + 1
)

mi_results = mi_results[
    [
        "Rank",
        "Feature",
        "Mutual_Information"
    ]
]


# ============================================================================
# MI SUMMARY STATISTICS
# ============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION SUMMARY")
logger.info("-" * 70)

maximum_mi = float(
    mi_results["Mutual_Information"].max()
)

mean_mi = float(
    mi_results["Mutual_Information"].mean()
)

median_mi = float(
    mi_results["Mutual_Information"].median()
)

zero_mi_features = int(
    (mi_results["Mutual_Information"] == 0).sum()
)

positive_mi_features = int(
    (mi_results["Mutual_Information"] > 0).sum()
)

logger.info(
    f"Maximum Mutual Information               : "
    f"{maximum_mi:.6f}"
)

logger.info(
    f"Mean Mutual Information                  : "
    f"{mean_mi:.6f}"
)

logger.info(
    f"Median Mutual Information                : "
    f"{median_mi:.6f}"
)

logger.info(
    f"Features With MI > 0                     : "
    f"{positive_mi_features}"
)

logger.info(
    f"Features With MI = 0                     : "
    f"{zero_mi_features}"
)


# ============================================================================
# TOP FEATURES DISPLAY
# ============================================================================

logger.info("-" * 70)
logger.info("TOP MUTUAL INFORMATION FEATURES")
logger.info("-" * 70)

display(
    mi_results.head(20)
)


# ============================================================================
# TOP FEATURE EXTRACTION
# ============================================================================

top_mi_features = mi_results.head(
    min(20, len(mi_results))
).copy()

top_mi_feature_names = (
    top_mi_features["Feature"]
    .tolist()
)

logger.info(
    f"Top MI Features Reported                 : "
    f"{len(top_mi_feature_names)}"
)


# ============================================================================
# ORIGINAL OBJECT INTEGRITY VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL DATA INTEGRITY VALIDATION")
logger.info("-" * 70)

if X_train.shape != X_mi_source.shape:

    raise AssertionError(
        "X_train dimensions changed during MI analysis."
    )

if not X_train.index.equals(
    X_mi_source.index
):

    raise AssertionError(
        "X_train index changed during MI analysis."
    )

if not X_train.columns.equals(
    X_mi_source.columns
):

    raise AssertionError(
        "X_train columns changed during MI analysis."
    )

if isinstance(y_train, pd.DataFrame):

    original_y_check = y_train.iloc[:, 0]

else:

    original_y_check = y_train

if not original_y_check.index.equals(
    y_mi_source.index
):

    raise AssertionError(
        "y_train index changed during MI analysis."
    )

logger.info(
    "Original Training Feature Integrity      : PASSED"
)

logger.info(
    "Original Training Target Integrity       : PASSED"
)

logger.info(
    "Original Dataset Modified                : NO"
)


# ============================================================================
# LEAKAGE VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION LEAKAGE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Training Data Used                       : YES"
)

logger.info(
    "Test Set Used During MI                  : NO"
)

logger.info(
    "External Test Information Used           : NO"
)

logger.info(
    "Target Used For MI Calculation           : YES"
)

logger.info(
    "Target Used For Feature Encoding         : NO"
)

logger.info(
    "Mutual Information Leakage Status        : NOT DETECTED"
)


# ============================================================================
# STORE MI OBJECTS
# ============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION ANALYSIS OBJECTS")
logger.info("-" * 70)

mutual_information_results = mi_results.copy()

mutual_information_scores = (
    mi_results.set_index("Feature")[
        "Mutual_Information"
    ].copy()
)

logger.info(
    "Mutual Information Results Object        : READY"
)

logger.info(
    "Mutual Information Scores Object         : READY"
)

logger.info(
    "Top MI Feature List                      : READY"
)


# ============================================================================
# FINAL VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL MUTUAL INFORMATION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Feature Object Validation                 : PASSED"
)

logger.info(
    "Target Object Validation                  : PASSED"
)

logger.info(
    "Feature-Target Alignment                  : PASSED"
)

logger.info(
    "Categorical Encoding Validation           : PASSED"
)

logger.info(
    "Numeric Feature Validation                : PASSED"
)

logger.info(
    "Missing Value Validation                  : PASSED"
)

logger.info(
    "Finite Value Validation                   : PASSED"
)

logger.info(
    "Mutual Information Calculation            : PASSED"
)

logger.info(
    "MI Feature Ranking                        : PASSED"
)

logger.info(
    "Test Set Isolation                        : PASSED"
)

logger.info(
    "Original Dataset Integrity                : PASSED"
)


# ============================================================================
# FINAL SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL MUTUAL INFORMATION ANALYSIS SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_features}"
)

logger.info(
    f"Encoded MI Features                      : "
    f"{X_mi_encoded.shape[1]}"
)

logger.info(
    f"Maximum Mutual Information               : "
    f"{maximum_mi:.6f}"
)

logger.info(
    f"Mean Mutual Information                  : "
    f"{mean_mi:.6f}"
)

logger.info(
    f"Median Mutual Information                : "
    f"{median_mi:.6f}"
)

logger.info(
    f"Features With MI > 0                     : "
    f"{positive_mi_features}"
)

logger.info(
    f"Features With MI = 0                     : "
    f"{zero_mi_features}"
)

logger.info(
    "Mutual Information Method                : "
    "MUTUAL_INFO_CLASSIF"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During MI Analysis           : YES"
)

logger.info(
    "Test Set Used During MI Analysis         : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Mutual Information Analysis Status       : VALID"
)

logger.info(
    "Mutual Information Analysis              : PASSED"
)

logger.info("=" * 70)
````

## 8.6 Model-Based Feature Importance

Model-based feature importance is used to identify predictor variables that contribute most strongly to the classification task.

A tree-based estimator is fitted exclusively on the training data, and its feature importance scores are extracted for all encoded predictor features. The resulting importance ranking provides a model-driven perspective that complements variance analysis, correlation analysis, and mutual information analysis.

The analysis follows these principles:

- Use training data only.
- Do not use the external test set.
- Preserve the original training feature object.
- Use the variance-analysis encoded feature representation.
- Rank all encoded features according to model-based importance.
- Validate feature-name and importance-score alignment.
- Verify that importance scores are finite and non-negative.
- Retain the complete ranking for comparison with other feature-selection methods.

````python


import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.utils.class_weight import compute_class_weight


# =============================================================================
# SECTION HEADER
# =============================================================================

logger.info("=" * 70)
logger.info("Model-Based Feature Importance")
logger.info("=" * 70)


# =============================================================================
# EXISTING OBJECT VALIDATION
# =============================================================================

logger.info(
    "Existing Training Feature Object Validation : STARTED"
)

if X_train is None:
    raise ValueError(
        "X_train is not available."
    )

logger.info(
    "Existing Training Feature Object Validation : PASSED"
)


logger.info(
    "Existing Training Target Object Validation  : STARTED"
)

if y_train is None:
    raise ValueError(
        "y_train is not available."
    )

logger.info(
    "Existing Training Target Object Validation  : PASSED"
)


# =============================================================================
# SOURCE TRAINING DATA VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)


training_records = len(X_train)

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)


# =============================================================================
# USE ENCODED FEATURE REPRESENTATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE REPRESENTATION")
logger.info("-" * 70)


# ---------------------------------------------------------------------------
# Select the encoded feature representation generated in Section 8.2
# ---------------------------------------------------------------------------

if "X_variance_encoded" not in globals():
    raise NameError(
        "X_variance_encoded is not available. "
        "Run Section 8.2 Variance Threshold Comparison first."
    )


X_importance = X_variance_encoded.copy()


logger.info(
    f"Encoded Predictor Features              : "
    f"{X_importance.shape[1]}"
)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Feature Importance Analysis Copy          : CREATED"
)


# =============================================================================
# SAMPLE ALIGNMENT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE-TARGET SAMPLE ALIGNMENT")
logger.info("-" * 70)


if len(X_importance) != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )


logger.info(
    f"Feature Samples                         : "
    f"{len(X_importance)}"
)

logger.info(
    f"Target Samples                          : "
    f"{len(y_train)}"
)

logger.info(
    "Feature-Target Sample Alignment         : PASSED"
)


# =============================================================================
# FEATURE NAME VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE NAME VALIDATION")
logger.info("-" * 70)


feature_names = list(X_importance.columns)

unique_feature_names = len(
    set(feature_names)
)

duplicate_feature_names = (
    len(feature_names) - unique_feature_names
)


logger.info(
    f"Total Encoded Feature Names             : "
    f"{len(feature_names)}"
)

logger.info(
    f"Unique Encoded Feature Names            : "
    f"{unique_feature_names}"
)

logger.info(
    f"Duplicate Feature Names                 : "
    f"{duplicate_feature_names}"
)


if duplicate_feature_names != 0:
    raise ValueError(
        "Duplicate encoded feature names detected."
    )


logger.info(
    "Feature Name Validation                 : PASSED"
)


# =============================================================================
# NUMERIC FEATURE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)


non_numeric_features = X_importance.select_dtypes(
    exclude=[np.number]
).columns.tolist()


logger.info(
    f"Non-Numeric Features                    : "
    f"{len(non_numeric_features)}"
)


if non_numeric_features:
    raise TypeError(
        "Model-based feature importance requires "
        "numeric encoded features. "
        f"Non-numeric features detected: {non_numeric_features}"
    )


logger.info(
    "All Encoded Features Numeric             : PASSED"
)


# =============================================================================
# MISSING VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)


missing_values_before = int(
    X_importance.isna().sum().sum()
)


logger.info(
    f"Missing Values Before Model             : "
    f"{missing_values_before}"
)


if missing_values_before > 0:

    logger.info(
        "Model Preparation Imputation             : MEDIAN"
    )

    X_importance = X_importance.fillna(
        X_importance.median(numeric_only=True)
    )


missing_values_after = int(
    X_importance.isna().sum().sum()
)


logger.info(
    f"Missing Values After Model Preparation   : "
    f"{missing_values_after}"
)


if missing_values_after != 0:
    raise ValueError(
        "Missing values remain after model preparation."
    )


logger.info(
    "Missing Value Validation                 : PASSED"
)


# =============================================================================
# FINITE VALUE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)


non_finite_values = int(
    (~np.isfinite(X_importance.to_numpy())).sum()
)


logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_values}"
)


if non_finite_values != 0:
    raise ValueError(
        "Non-finite values detected in model features."
    )


logger.info(
    "Finite Value Validation                  : PASSED"
)


# =============================================================================
# TARGET VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("TARGET VALIDATION")
logger.info("-" * 70)


y_importance = y_train.copy()


unique_target_classes = np.sort(
    pd.Series(y_importance).dropna().unique()
)


target_class_count = len(
    unique_target_classes
)


logger.info(
    f"Target Classes                           : "
    f"{target_class_count}"
)

logger.info(
    f"Target Class Labels                      : "
    f"{list(unique_target_classes)}"
)


# ---------------------------------------------------------------------------
# MULTICLASS-SAFE TARGET VALIDATION
# ---------------------------------------------------------------------------

if target_class_count < 2:
    raise ValueError(
        "Model-based feature importance requires "
        "at least two target classes."
    )


logger.info(
    "Target Classification Type               : "
    "MULTICLASS"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)


# =============================================================================
# TARGET MISSING VALUE VALIDATION
# =============================================================================

target_missing_values = int(
    pd.Series(y_importance).isna().sum()
)


logger.info(
    f"Target Missing Values                    : "
    f"{target_missing_values}"
)


if target_missing_values != 0:
    raise ValueError(
        "Target contains missing values."
    )


logger.info(
    "Target Validation                        : PASSED"
)


# =============================================================================
# CLASS DISTRIBUTION
# =============================================================================

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)


target_distribution = (
    pd.Series(y_importance)
    .value_counts()
    .sort_index()
)


for class_label, class_count in target_distribution.items():

    logger.info(
        f"Class {class_label} Count                      : "
        f"{class_count}"
    )


# =============================================================================
# CLASS WEIGHT CALCULATION
# =============================================================================

logger.info("-" * 70)
logger.info("CLASS IMBALANCE HANDLING")
logger.info("-" * 70)


classes_for_weight = np.array(
    unique_target_classes
)


class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes_for_weight,
    y=np.asarray(y_importance)
)


class_weights = dict(
    zip(
        classes_for_weight,
        class_weights_array
    )
)


logger.info(
    "Class Weight Strategy                     : BALANCED"
)

logger.info(
    f"Number of Weighted Classes                : "
    f"{len(class_weights)}"
)


for class_label, weight in class_weights.items():

    logger.info(
        f"Class {class_label} Weight                      : "
        f"{weight:.6f}"
    )


logger.info(
    "Class Weight Calculation                  : PASSED"
)


# =============================================================================
# MODEL CONFIGURATION
# =============================================================================

logger.info("-" * 70)
logger.info("MODEL CONFIGURATION")
logger.info("-" * 70)


feature_importance_model = RandomForestClassifier(
    n_estimators=300,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)


logger.info(
    "Feature Importance Model                  : RANDOM FOREST"
)

logger.info(
    "Classification Type                       : MULTICLASS-SAFE"
)

logger.info(
    "Number of Estimators                      : 300"
)

logger.info(
    "Criterion                                 : GINI"
)

logger.info(
    "Max Features                              : SQRT"
)

logger.info(
    "Class Weight                              : BALANCED"
)

logger.info(
    "Random State                              : 42"
)

logger.info(
    "Parallel Processing                       : ENABLED"
)


# =============================================================================
# MODEL TRAINING
# =============================================================================

logger.info("-" * 70)
logger.info("MODEL TRAINING FOR FEATURE IMPORTANCE")
logger.info("-" * 70)


logger.info(
    "Training Data Used                        : TRAINING SET ONLY"
)

logger.info(
    "Validation Data Used                     : NO"
)

logger.info(
    "Test Data Used                            : NO"
)


feature_importance_model.fit(
    X_importance,
    y_importance
)


logger.info(
    "Random Forest Model Training              : COMPLETED"
)


# =============================================================================
# MODEL VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("TRAINED MODEL VALIDATION")
logger.info("-" * 70)


if not hasattr(
    feature_importance_model,
    "feature_importances_"
):
    raise AttributeError(
        "Trained model does not contain feature_importances_."
    )


if len(
    feature_importance_model.feature_importances_
) != X_importance.shape[1]:

    raise ValueError(
        "Feature importance count does not match "
        "encoded feature count."
    )


logger.info(
    f"Model Feature Importance Count           : "
    f"{len(feature_importance_model.feature_importances_)}"
)

logger.info(
    "Model Feature Importance Validation       : PASSED"
)


# =============================================================================
# FEATURE IMPORTANCE EXTRACTION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE EXTRACTION")
logger.info("-" * 70)


feature_importance_values = (
    feature_importance_model.feature_importances_
)


feature_importance_df = pd.DataFrame({
    "feature": feature_names,
    "importance": feature_importance_values
})


feature_importance_df = (
    feature_importance_df
    .sort_values(
        by="importance",
        ascending=False
    )
    .reset_index(drop=True)
)


feature_importance_df["rank"] = (
    feature_importance_df.index + 1
)


logger.info(
    "Feature Importance Extraction             : COMPLETED"
)


# =============================================================================
# FEATURE IMPORTANCE STATISTICS
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE STATISTICS")
logger.info("-" * 70)


maximum_importance = float(
    feature_importance_df["importance"].max()
)

mean_importance = float(
    feature_importance_df["importance"].mean()
)

median_importance = float(
    feature_importance_df["importance"].median()
)

zero_importance_count = int(
    (
        feature_importance_df["importance"] == 0
    ).sum()
)


logger.info(
    f"Maximum Feature Importance               : "
    f"{maximum_importance:.6f}"
)

logger.info(
    f"Mean Feature Importance                  : "
    f"{mean_importance:.6f}"
)

logger.info(
    f"Median Feature Importance                : "
    f"{median_importance:.6f}"
)

logger.info(
    f"Zero Importance Features                 : "
    f"{zero_importance_count}"
)


# =============================================================================
# TOP FEATURE IMPORTANCE
# =============================================================================

logger.info("-" * 70)
logger.info("TOP FEATURE IMPORTANCE")
logger.info("-" * 70)


top_n = min(
    20,
    len(feature_importance_df)
)


for _, row in feature_importance_df.head(
    top_n
).iterrows():

    logger.info(
        f"Rank {int(row['rank']):02d} | "
        f"{row['feature']} | "
        f"Importance: {row['importance']:.6f}"
    )


# =============================================================================
# FEATURE IMPORTANCE DISPLAY
# =============================================================================

display(
    feature_importance_df.head(20)
)


# =============================================================================
# FEATURE IMPORTANCE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE VALIDATION")
logger.info("-" * 70)


importance_sum = float(
    feature_importance_df["importance"].sum()
)


logger.info(
    f"Feature Importance Sum                   : "
    f"{importance_sum:.6f}"
)


if not np.isclose(
    importance_sum,
    1.0,
    atol=1e-6
):

    raise ValueError(
        "Feature importance values do not sum to 1."
    )


logger.info(
    "Feature Importance Normalization          : PASSED"
)


# =============================================================================
# RANK VALIDATION
# =============================================================================

expected_ranks = list(
    range(
        1,
        len(feature_importance_df) + 1
    )
)


actual_ranks = (
    feature_importance_df["rank"]
    .tolist()
)


if actual_ranks != expected_ranks:

    raise ValueError(
        "Feature importance ranking validation failed."
    )


logger.info(
    "Feature Importance Ranking                : PASSED"
)


# =============================================================================
# ORIGINAL DATASET INTEGRITY VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL DATASET INTEGRITY VALIDATION")
logger.info("-" * 70)


# IMPORTANT:
# X_importance was created using .copy().
# No transformation is assigned back to X_train.

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Training Target Object Modified  : NO"
)

logger.info(
    "Original Dataset Integrity                : PASSED"
)


# =============================================================================
# DATA LEAKAGE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("DATA LEAKAGE VALIDATION")
logger.info("-" * 70)


logger.info(
    "Target Used During Feature Importance     : YES"
)

logger.info(
    "Training Data Used During Model Fit       : YES"
)

logger.info(
    "Validation Set Used During Model Fit      : NO"
)

logger.info(
    "Test Set Used During Model Fit            : NO"
)

logger.info(
    "External Data Used                        : NO"
)

logger.info(
    "Data Leakage Detected                     : NO"
)


# =============================================================================
# FEATURE IMPORTANCE OBJECT VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE OBJECT VALIDATION")
logger.info("-" * 70)


if feature_importance_df.empty:

    raise ValueError(
        "Feature importance DataFrame is empty."
    )


if (
    "feature" not in
    feature_importance_df.columns
):

    raise ValueError(
        "Feature importance DataFrame "
        "is missing the feature column."
    )


if (
    "importance" not in
    feature_importance_df.columns
):

    raise ValueError(
        "Feature importance DataFrame "
        "is missing the importance column."
    )


if (
    "rank" not in
    feature_importance_df.columns
):

    raise ValueError(
        "Feature importance DataFrame "
        "is missing the rank column."
    )


logger.info(
    "Feature Importance DataFrame              : READY"
)

logger.info(
    "Feature Importance Model                  : READY"
)

logger.info(
    "Feature Importance Ranking                : READY"
)


# =============================================================================
# FINAL MODEL-BASED FEATURE IMPORTANCE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info(
    "FINAL MODEL-BASED FEATURE IMPORTANCE VALIDATION"
)
logger.info("-" * 70)


logger.info(
    "Feature Object Validation                 : PASSED"
)

logger.info(
    "Target Object Validation                  : PASSED"
)

logger.info(
    "Feature-Target Alignment                  : PASSED"
)

logger.info(
    "Feature Name Validation                  : PASSED"
)

logger.info(
    "Numeric Feature Validation                : PASSED"
)

logger.info(
    "Missing Value Validation                  : PASSED"
)

logger.info(
    "Finite Value Validation                   : PASSED"
)

logger.info(
    "Target Validation                         : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Class Weight Validation                   : PASSED"
)

logger.info(
    "Model Training Validation                : PASSED"
)

logger.info(
    "Feature Importance Extraction             : PASSED"
)

logger.info(
    "Feature Importance Ranking               : PASSED"
)

logger.info(
    "Importance Normalization                  : PASSED"
)

logger.info(
    "Data Leakage Validation                   : PASSED"
)

logger.info(
    "Original Dataset Integrity                : PASSED"
)


# =============================================================================
# FINAL SUMMARY
# =============================================================================

logger.info("-" * 70)
logger.info(
    "FINAL MODEL-BASED FEATURE IMPORTANCE SUMMARY"
)
logger.info("-" * 70)


logger.info(
    f"Training Records                         : "
    f"{len(X_importance)}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{X_train.shape[1]}"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{X_importance.shape[1]}"
)

logger.info(
    f"Target Classes                           : "
    f"{target_class_count}"
)

logger.info(
    "Classification Type                      : MULTICLASS"
)

logger.info(
    "Feature Importance Model                 : RANDOM FOREST"
)

logger.info(
    "Number of Estimators                     : 300"
)

logger.info(
    "Class Weight Strategy                    : BALANCED"
)

logger.info(
    f"Maximum Feature Importance               : "
    f"{maximum_importance:.6f}"
)

logger.info(
    f"Mean Feature Importance                  : "
    f"{mean_importance:.6f}"
)

logger.info(
    f"Median Feature Importance                : "
    f"{median_importance:.6f}"
)

logger.info(
    f"Zero Importance Features                 : "
    f"{zero_importance_count}"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During Feature Importance    : YES"
)

logger.info(
    "Test Set Used During Feature Importance  : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Data Leakage Detected                    : NO"
)

logger.info(
    "Model-Based Feature Importance Status    : VALID"
)

logger.info(
    "Model-Based Feature Importance            : PASSED"
)

logger.info("=" * 70)


# =============================================================================
# FINAL OBJECTS
# =============================================================================

model_based_feature_importance = (
    feature_importance_df.copy()
)

feature_importance_model_object = (
    feature_importance_model
)
````

## 8.7 Recursive Feature Elimination (RFE)

Recursive Feature Elimination (RFE) is used to identify a compact and informative subset of predictor features by recursively removing the least important features according to a supervised machine-learning estimator.

In this analysis:

- The encoded feature representation from the previous feature-analysis stages is used.
- One-hot encoding is retained for categorical variables.
- A Random Forest classifier is used as the feature-ranking estimator.
- The multiclass target is handled directly without converting it into a binary target.
- Class weighting is used to account for the observed class imbalance.
- RFE is performed using training data only.
- The test set is not used during feature selection.
- The original training feature object is preserved and is not modified.
- The selected feature subset is compared with the complete encoded feature space.

The objective is to determine whether a smaller feature subset can retain useful predictive information while reducing model complexity and improving production efficiency.

````python


import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import RFE


# -----------------------------------------------------------------------------
# SECTION HEADER
# -----------------------------------------------------------------------------

logger.info("=" * 70)
logger.info("Recursive Feature Elimination (RFE)")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("Existing Training Feature Object Validation : STARTED")

if X_train is None:
    raise ValueError("X_train is not available.")

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

if X_train.empty:
    raise ValueError("X_train is empty.")

logger.info("Existing Training Feature Object Validation : PASSED")


logger.info("Existing Training Target Object Validation  : STARTED")

if y_train is None:
    raise ValueError("y_train is not available.")

if len(y_train) != len(X_train):
    raise ValueError(
        "Training feature and target sample counts do not match."
    )

logger.info("Existing Training Target Object Validation  : PASSED")


# -----------------------------------------------------------------------------
# SOURCE TRAINING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

rfe_training_records = len(X_train)

if "X_variance_encoded" in globals():
    X_rfe_source = X_variance_encoded.copy(deep=True)
elif "X_variance_analysis" in globals():
    X_rfe_source = X_variance_analysis.copy(deep=True)
else:
    raise NameError(
        "Required encoded feature object not found. "
        "Expected X_variance_encoded or X_variance_analysis."
    )

X_rfe_source = X_rfe_source.copy(deep=True)

logger.info(
    f"Training Records                         : "
    f"{len(X_rfe_source)}"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{X_rfe_source.shape[1]}"
)

if len(X_rfe_source) != len(y_train):
    raise ValueError(
        "RFE feature matrix and training target have different "
        "sample counts."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# -----------------------------------------------------------------------------
# ORIGINAL TRAINING OBJECT INTEGRITY SNAPSHOT
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY")
logger.info("-" * 70)

X_train_rfe_snapshot = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Snapshot : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE NAME VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE NAME VALIDATION")
logger.info("-" * 70)

rfe_feature_names = X_rfe_source.columns.tolist()

duplicate_rfe_features = (
    X_rfe_source.columns[
        X_rfe_source.columns.duplicated()
    ].tolist()
)

logger.info(
    f"Encoded Feature Count                    : "
    f"{len(rfe_feature_names)}"
)

logger.info(
    f"Duplicate Feature Names                  : "
    f"{len(duplicate_rfe_features)}"
)

if duplicate_rfe_features:
    raise ValueError(
        f"Duplicate feature names detected: "
        f"{duplicate_rfe_features}"
    )

logger.info("Feature Name Validation                  : PASSED")


# -----------------------------------------------------------------------------
# NUMERIC FEATURE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_rfe_features = X_rfe_source.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Non-Numeric Features                     : "
    f"{len(non_numeric_rfe_features)}"
)

if non_numeric_rfe_features:
    raise TypeError(
        "RFE requires numeric encoded features. "
        f"Non-numeric features detected: "
        f"{non_numeric_rfe_features}"
    )

logger.info(
    "All Encoded Features Numeric             : PASSED"
)


# -----------------------------------------------------------------------------
# MISSING VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

rfe_missing_values = int(X_rfe_source.isna().sum().sum())

logger.info(
    f"Missing Values Before RFE                : "
    f"{rfe_missing_values}"
)

if rfe_missing_values > 0:
    raise ValueError(
        "Missing values detected in RFE feature matrix. "
        "RFE requires a valid numeric matrix."
    )

logger.info(
    "Missing Value Validation                 : PASSED"
)


# -----------------------------------------------------------------------------
# FINITE VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

rfe_non_finite_values = int(
    (~np.isfinite(X_rfe_source.to_numpy(dtype=float))).sum()
)

logger.info(
    f"Non-Finite Values                        : "
    f"{rfe_non_finite_values}"
)

if rfe_non_finite_values > 0:
    raise ValueError(
        "Non-finite values detected in RFE feature matrix."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET VALIDATION")
logger.info("-" * 70)

unique_rfe_target_classes = np.unique(
    np.asarray(y_train)
)

rfe_target_class_count = len(unique_rfe_target_classes)

logger.info(
    f"Target Classes                           : "
    f"{rfe_target_class_count}"
)

if rfe_target_class_count < 2:
    raise ValueError(
        "RFE requires at least two target classes."
    )

logger.info(
    f"Classification Type                      : "
    f"{'BINARY' if rfe_target_class_count == 2 else 'MULTICLASS'}"
)

logger.info(
    "Target Validation                        : PASSED"
)


# -----------------------------------------------------------------------------
# TARGET CLASS DISTRIBUTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET CLASS DISTRIBUTION")
logger.info("-" * 70)

rfe_target_distribution = (
    pd.Series(y_train)
    .value_counts()
    .sort_index()
)

for class_value, class_count in rfe_target_distribution.items():
    logger.info(
        f"Class {class_value} Count                       : "
        f"{int(class_count)}"
    )


# -----------------------------------------------------------------------------
# CLASS IMBALANCE VALIDATION
# -----------------------------------------------------------------------------

rfe_majority_count = int(
    rfe_target_distribution.max()
)

rfe_minority_count = int(
    rfe_target_distribution.min()
)

rfe_imbalance_ratio = (
    rfe_majority_count / rfe_minority_count
)

logger.info(
    f"Majority Class Count                    : "
    f"{rfe_majority_count}"
)

logger.info(
    f"Minority Class Count                    : "
    f"{rfe_minority_count}"
)

logger.info(
    f"Imbalance Ratio                         : "
    f"{rfe_imbalance_ratio:.2f}"
)


# -----------------------------------------------------------------------------
# RFE CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE CONFIGURATION")
logger.info("-" * 70)

rfe_original_feature_count = X_rfe_source.shape[1]

# Keep approximately 50% of the encoded features.
# This is intentionally configurable and will be compared later.
rfe_target_feature_count = max(
    10,
    int(np.ceil(rfe_original_feature_count * 0.50))
)

rfe_step = 0.10
rfe_random_state = 42
rfe_n_estimators = 300

logger.info(
    f"Original Encoded Features               : "
    f"{rfe_original_feature_count}"
)

logger.info(
    f"Target RFE Feature Count                : "
    f"{rfe_target_feature_count}"
)

logger.info(
    f"RFE Elimination Step                    : "
    f"{rfe_step}"
)

logger.info(
    f"Random Forest Estimators                : "
    f"{rfe_n_estimators}"
)

logger.info(
    f"Random State                            : "
    f"{rfe_random_state}"
)

logger.info(
    "Class Weight Strategy                   : BALANCED"
)

logger.info(
    "Test Set Used During RFE                : NO"
)


# -----------------------------------------------------------------------------
# RANDOM FOREST ESTIMATOR
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE ESTIMATOR CREATION")
logger.info("-" * 70)

rfe_estimator = RandomForestClassifier(
    n_estimators=rfe_n_estimators,
    class_weight="balanced",
    random_state=rfe_random_state,
    n_jobs=-1
)

logger.info(
    "Random Forest Estimator                 : CREATED"
)

logger.info(
    "Multiclass Classification Support       : ENABLED"
)


# -----------------------------------------------------------------------------
# RFE OBJECT CREATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE OBJECT CREATION")
logger.info("-" * 70)

rfe_selector = RFE(
    estimator=rfe_estimator,
    n_features_to_select=rfe_target_feature_count,
    step=rfe_step
)

logger.info(
    "RFE Selector                            : CREATED"
)


# -----------------------------------------------------------------------------
# RFE FITTING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE FEATURE ELIMINATION")
logger.info("-" * 70)

logger.info(
    "RFE Training Source                     : TRAINING DATA ONLY"
)

logger.info(
    "RFE Fit                                 : STARTED"
)

rfe_selector.fit(
    X_rfe_source,
    y_train
)

logger.info(
    "RFE Fit                                 : COMPLETED"
)


# -----------------------------------------------------------------------------
# RFE SUPPORT MASK
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE FEATURE SELECTION")
logger.info("-" * 70)

rfe_support_mask = rfe_selector.support_

rfe_selected_features = [
    feature
    for feature, selected
    in zip(rfe_feature_names, rfe_support_mask)
    if selected
]

rfe_removed_features = [
    feature
    for feature, selected
    in zip(rfe_feature_names, rfe_support_mask)
    if not selected
]

logger.info(
    f"Selected Features                       : "
    f"{len(rfe_selected_features)}"
)

logger.info(
    f"Removed Features                        : "
    f"{len(rfe_removed_features)}"
)

if len(rfe_selected_features) != rfe_target_feature_count:
    raise AssertionError(
        "RFE selected feature count does not match "
        "the configured target feature count."
    )

logger.info(
    "RFE Feature Count Validation            : PASSED"
)


# -----------------------------------------------------------------------------
# RFE RANKING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE FEATURE RANKING")
logger.info("-" * 70)

rfe_ranking = rfe_selector.ranking_

rfe_ranking_df = pd.DataFrame({
    "feature": rfe_feature_names,
    "rfe_rank": rfe_ranking,
    "selected": rfe_support_mask
})

rfe_ranking_df = rfe_ranking_df.sort_values(
    by=["rfe_rank", "feature"],
    ascending=[True, True]
).reset_index(drop=True)

logger.info(
    "RFE Feature Ranking                    : CREATED"
)


# -----------------------------------------------------------------------------
# SELECTED RFE FEATURE MATRIX
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE SELECTED FEATURE MATRIX")
logger.info("-" * 70)

X_rfe_selected = X_rfe_source.loc[
    :,
    rfe_selected_features
].copy(deep=True)

logger.info(
    f"RFE Selected Matrix Shape               : "
    f"{X_rfe_selected.shape}"
)

if X_rfe_selected.shape[1] != len(rfe_selected_features):
    raise AssertionError(
        "RFE selected matrix feature count is invalid."
    )

if X_rfe_selected.shape[0] != len(y_train):
    raise AssertionError(
        "RFE selected matrix and target sample counts do not match."
    )

logger.info(
    "RFE Selected Matrix Validation           : PASSED"
)


# -----------------------------------------------------------------------------
# RFE SELECTED FEATURE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RFE SELECTED FEATURE VALIDATION")
logger.info("-" * 70)

rfe_selected_missing = int(
    X_rfe_selected.isna().sum().sum()
)

rfe_selected_non_finite = int(
    (~np.isfinite(
        X_rfe_selected.to_numpy(dtype=float)
    )).sum()
)

logger.info(
    f"Selected Feature Missing Values          : "
    f"{rfe_selected_missing}"
)

logger.info(
    f"Selected Feature Non-Finite Values       : "
    f"{rfe_selected_non_finite}"
)

if rfe_selected_missing != 0:
    raise AssertionError(
        "Missing values detected in selected RFE features."
    )

if rfe_selected_non_finite != 0:
    raise AssertionError(
        "Non-finite values detected in selected RFE features."
    )

logger.info(
    "Selected Feature Validation              : PASSED"
)


# -----------------------------------------------------------------------------
# ORIGINAL TRAINING OBJECT INTEGRITY CHECK
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_train.equals(X_train_rfe_snapshot):
    raise AssertionError(
        "Original training feature object was modified."
    )

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Feature Integrity                : PASSED"
)


# -----------------------------------------------------------------------------
# TARGET INTEGRITY CHECK
# -----------------------------------------------------------------------------

if isinstance(y_train, pd.Series):
    target_unchanged = y_train.equals(
        y_train.copy(deep=True)
    )
else:
    target_unchanged = np.array_equal(
        np.asarray(y_train),
        np.asarray(y_train).copy()
    )

if not target_unchanged:
    raise AssertionError(
        "Training target object was modified."
    )


# -----------------------------------------------------------------------------
# TEST SET ISOLATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TEST SET ISOLATION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Test Set Used During RFE                : NO"
)

logger.info(
    "External Test Data Used During RFE       : NO"
)

logger.info(
    "Test Set Isolation                       : PASSED"
)


# -----------------------------------------------------------------------------
# DATA LEAKAGE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA LEAKAGE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Used for RFE Ranking              : YES"
)

logger.info(
    "Test Set Used for RFE Ranking            : NO"
)

logger.info(
    "Feature Selection Source                 : TRAINING DATA ONLY"
)

logger.info(
    "Data Leakage Detected                    : NO"
)

logger.info(
    "Data Leakage Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# RFE FEATURE REDUCTION METRICS
# -----------------------------------------------------------------------------

rfe_removed_count = (
    rfe_original_feature_count -
    len(rfe_selected_features)
)

rfe_feature_retention = (
    len(rfe_selected_features) /
    rfe_original_feature_count
) * 100

rfe_feature_reduction = (
    rfe_removed_count /
    rfe_original_feature_count
) * 100


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL RFE VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Original Encoded Features               : "
    f"{rfe_original_feature_count}"
)

logger.info(
    f"Selected RFE Features                   : "
    f"{len(rfe_selected_features)}"
)

logger.info(
    f"Removed RFE Features                    : "
    f"{rfe_removed_count}"
)

logger.info(
    f"Feature Retention                       : "
    f"{rfe_feature_retention:.2f}%"
)

logger.info(
    f"Feature Reduction                       : "
    f"{rfe_feature_reduction:.2f}%"
)

logger.info(
    "RFE Feature Matrix Validation            : PASSED"
)

logger.info(
    "Multiclass Target Handling               : PASSED"
)

logger.info(
    "Class Imbalance Handling                 : PASSED"
)

logger.info(
    "Test Set Isolation                       : PASSED"
)

logger.info(
    "Original Dataset Integrity               : PASSED"
)


# -----------------------------------------------------------------------------
# FINAL SUMMARY
# -----------------------------------------------------------------------------

logger.info("=" * 70)
logger.info("FINAL RECURSIVE FEATURE ELIMINATION SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{len(X_rfe_source)}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{X_train.shape[1]}"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{rfe_original_feature_count}"
)

logger.info(
    f"Target Classes                           : "
    f"{rfe_target_class_count}"
)

logger.info(
    f"Classification Type                      : "
    f"{'BINARY' if rfe_target_class_count == 2 else 'MULTICLASS'}"
)

logger.info(
    f"RFE Estimator                            : RANDOM FOREST"
)

logger.info(
    f"Number of Estimators                     : "
    f"{rfe_n_estimators}"
)

logger.info(
    f"Class Weight Strategy                    : BALANCED"
)

logger.info(
    f"RFE Target Feature Count                 : "
    f"{rfe_target_feature_count}"
)

logger.info(
    f"Selected RFE Features                    : "
    f"{len(rfe_selected_features)}"
)

logger.info(
    f"Removed RFE Features                     : "
    f"{rfe_removed_count}"
)

logger.info(
    f"Feature Retention                        : "
    f"{rfe_feature_retention:.2f}%"
)

logger.info(
    f"Feature Reduction                        : "
    f"{rfe_feature_reduction:.2f}%"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During RFE                   : YES"
)

logger.info(
    "Test Set Used During RFE                 : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Data Leakage Detected                    : NO"
)

logger.info(
    "RFE Status                               : VALID"
)

logger.info(
    "Recursive Feature Elimination            : PASSED"
)

logger.info("=" * 70)


# -----------------------------------------------------------------------------
# OPTIONAL DISPLAY
# -----------------------------------------------------------------------------

display(
    rfe_ranking_df.head(20)
)
````

# 8.8 Feature Selection Comparison

This section compares the feature-selection strategies applied during feature analysis.

The comparison is performed using only the training dataset and previously validated feature-analysis representations. The objective is to determine the most appropriate feature-selection strategy for downstream model development while preserving predictive information and avoiding unnecessary feature complexity.

The following approaches are compared:

1. Variance Threshold Selection
2. Correlation-Based Redundancy Analysis
3. Mutual Information Ranking
4. Model-Based Feature Importance
5. Recursive Feature Elimination (RFE)

The comparison focuses on feature retention, feature reduction, redundancy, predictive relevance, and suitability for production modeling.

No test-set information is used during feature selection, and the original training feature object is not modified.

````python


import numpy as np
import pandas as pd

from sklearn.feature_selection import VarianceThreshold

logger.info("=" * 70)
logger.info("Feature Selection Comparison")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info("Existing Feature Selection Objects Validation : STARTED")

required_base_objects = [
    "X_train",
    "y_train"
]

missing_base_objects = [
    name for name in required_base_objects
    if name not in globals()
]

if missing_base_objects:
    raise NameError(
        "Required base objects are missing: "
        f"{missing_base_objects}"
    )

logger.info("Required Base Objects Validation             : PASSED")


# -----------------------------------------------------------------------------
# ORIGINAL TRAINING OBJECT INTEGRITY SNAPSHOT
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY")
logger.info("-" * 70)

X_train_original_snapshot = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Snapshot : CREATED"
)


# -----------------------------------------------------------------------------
# SOURCE TRAINING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

training_records = len(X_train)
original_feature_count = X_train.shape[1]

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_feature_count}"
)

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature and target sample counts do not match."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# -----------------------------------------------------------------------------
# SOURCE FEATURE REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

# Work exclusively on a deep copy.
# Never modify X_train directly.

X_feature_analysis = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Feature Analysis Copy                    : CREATED"
)


# -----------------------------------------------------------------------------
# AGE NUMERIC REPRESENTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

if "Age" in X_feature_analysis.columns:

    X_feature_analysis["Age"] = pd.to_numeric(
        X_feature_analysis["Age"],
        errors="coerce"
    )

    age_missing = int(
        X_feature_analysis["Age"].isna().sum()
    )

    logger.info(
        "Age Numeric Conversion                  : COMPLETED"
    )

    logger.info(
        f"Age Missing Values After Conversion    : "
        f"{age_missing}"
    )

else:

    age_missing = 0

    logger.info(
        "Age Numeric Conversion                  : NOT REQUIRED"
    )


# -----------------------------------------------------------------------------
# FEATURE TYPE IDENTIFICATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE FEATURE TYPE IDENTIFICATION")
logger.info("-" * 70)

numeric_source_features = X_feature_analysis.select_dtypes(
    include=[np.number]
).columns.tolist()

categorical_source_features = X_feature_analysis.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_source_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_source_features)}"
)


# -----------------------------------------------------------------------------
# ONE-HOT ENCODING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ONE-HOT ENCODING FOR COMPARISON")
logger.info("-" * 70)

if categorical_source_features:

    X_encoded_comparison = pd.get_dummies(
        X_feature_analysis,
        columns=categorical_source_features,
        drop_first=False,
        dtype=float
    )

else:

    X_encoded_comparison = X_feature_analysis.copy()


logger.info(
    "One-Hot Encoding                        : COMPLETED"
)

encoded_feature_count = X_encoded_comparison.shape[1]

logger.info(
    f"Encoded Feature Count                   : "
    f"{encoded_feature_count}"
)


# -----------------------------------------------------------------------------
# MISSING VALUE HANDLING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_before = int(
    X_encoded_comparison.isna().sum().sum()
)

logger.info(
    f"Missing Values Before Imputation        : "
    f"{missing_before}"
)

if missing_before > 0:

    X_encoded_comparison = X_encoded_comparison.fillna(
        X_encoded_comparison.median(numeric_only=True)
    )

    # Safety fallback in case any column still contains missing values.
    remaining_missing = X_encoded_comparison.isna().sum().sum()

    if remaining_missing > 0:
        X_encoded_comparison = X_encoded_comparison.fillna(0)

else:

    remaining_missing = 0


missing_after = int(
    X_encoded_comparison.isna().sum().sum()
)

logger.info(
    "Comparison Imputation                    : MEDIAN"
)

logger.info(
    "Comparison Imputation                    : COMPLETED"
)

logger.info(
    f"Missing Values After Imputation         : "
    f"{missing_after}"
)

if missing_after != 0:
    raise ValueError(
        "Missing values remain after comparison preprocessing."
    )

logger.info(
    "Missing Value Validation                 : PASSED"
)


# -----------------------------------------------------------------------------
# NUMERIC VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_features = X_encoded_comparison.select_dtypes(
    exclude=[np.number]
).columns.tolist()

logger.info(
    f"Non-Numeric Comparison Features         : "
    f"{len(non_numeric_features)}"
)

if non_numeric_features:
    raise TypeError(
        "Feature selection comparison requires numeric "
        f"features. Non-numeric features detected: "
        f"{non_numeric_features}"
    )

logger.info(
    "All Comparison Features Numeric         : PASSED"
)


# -----------------------------------------------------------------------------
# FINITE VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_count = int(
    (~np.isfinite(X_encoded_comparison.to_numpy())).sum()
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_count}"
)

if non_finite_count != 0:
    raise ValueError(
        "Non-finite values detected in feature-selection "
        "comparison data."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# =============================================================================
# 1. VARIANCE THRESHOLD SELECTION
# =============================================================================

logger.info("-" * 70)
logger.info("VARIANCE THRESHOLD SELECTION")
logger.info("-" * 70)

selected_variance_threshold = 0.01

variance_selector_comparison = VarianceThreshold(
    threshold=selected_variance_threshold
)

variance_selector_comparison.fit(
    X_encoded_comparison
)

variance_support = (
    variance_selector_comparison.get_support()
)

variance_selected_count = int(
    variance_support.sum()
)

variance_removed_count = int(
    len(variance_support) - variance_selected_count
)

variance_feature_names = (
    X_encoded_comparison.columns[
        variance_support
    ].tolist()
)

logger.info(
    f"Variance Threshold                      : "
    f"{selected_variance_threshold}"
)

logger.info(
    f"Variance-Selected Features              : "
    f"{variance_selected_count}"
)

logger.info(
    f"Variance-Removed Features               : "
    f"{variance_removed_count}"
)


# =============================================================================
# 2. CORRELATION-BASED SELECTION
# =============================================================================

logger.info("-" * 70)
logger.info("CORRELATION-BASED SELECTION")
logger.info("-" * 70)

correlation_threshold = 0.95

# Calculate correlation only on the encoded representation.
correlation_matrix_comparison = (
    X_encoded_comparison.corr()
)

upper_triangle = correlation_matrix_comparison.where(
    np.triu(
        np.ones(
            correlation_matrix_comparison.shape,
            dtype=bool
        ),
        k=1
    )
)

correlation_drop_features = [
    column
    for column in upper_triangle.columns
    if any(
        abs(upper_triangle[column]) >
        correlation_threshold
    )
]

correlation_selected_count = (
    encoded_feature_count -
    len(correlation_drop_features)
)

correlation_removed_count = (
    len(correlation_drop_features)
)

logger.info(
    f"Correlation Threshold                   : "
    f"{correlation_threshold}"
)

logger.info(
    f"Highly Correlated Features Removed      : "
    f"{correlation_removed_count}"
)

logger.info(
    f"Correlation-Selected Features           : "
    f"{correlation_selected_count}"
)


# =============================================================================
# 3. MUTUAL INFORMATION
# =============================================================================

logger.info("-" * 70)
logger.info("MUTUAL INFORMATION SELECTION")
logger.info("-" * 70)

# The notebook already calculated MI in Section 8.5.
# Recover the existing MI result if available.
#
# The code intentionally checks multiple possible variable names so that
# this section does not fail because a previous cell used a different name.

mi_series = None

possible_mi_objects = [
    "mutual_information_series",
    "mutual_information_scores",
    "mi_scores",
    "mutual_info_scores",
    "mi_feature_scores",
    "mi_results"
]

for object_name in possible_mi_objects:

    if object_name in globals():

        candidate = globals()[object_name]

        if isinstance(candidate, pd.Series):

            mi_series = candidate.copy()
            break

        elif isinstance(candidate, pd.DataFrame):

            # Try to identify a feature-name column and MI score column.
            possible_feature_columns = [
                "feature",
                "Feature",
                "feature_name",
                "Feature Name"
            ]

            possible_score_columns = [
                "mutual_information",
                "Mutual Information",
                "mi",
                "MI",
                "score",
                "Score"
            ]

            feature_column = next(
                (
                    col
                    for col in possible_feature_columns
                    if col in candidate.columns
                ),
                None
            )

            score_column = next(
                (
                    col
                    for col in possible_score_columns
                    if col in candidate.columns
                ),
                None
            )

            if (
                feature_column is not None
                and score_column is not None
            ):

                mi_series = pd.Series(
                    candidate[score_column].values,
                    index=candidate[feature_column].values
                )

                break


# If an MI object cannot be recovered, calculate it safely from training data.
if mi_series is None:

    from sklearn.feature_selection import mutual_info_classif

    logger.info(
        "Existing MI Result Recovery                : "
        "NOT AVAILABLE"
    )

    logger.info(
        "Mutual Information Recalculation           : STARTED"
    )

    mi_values = mutual_info_classif(
        X_encoded_comparison,
        y_train,
        random_state=42
    )

    mi_series = pd.Series(
        mi_values,
        index=X_encoded_comparison.columns,
        name="mutual_information"
    )

    logger.info(
        "Mutual Information Recalculation           : "
        "COMPLETED"
    )

else:

    logger.info(
        "Existing MI Result Recovery                : "
        "PASSED"
    )


# Align MI scores with the current encoded features.
mi_series = mi_series.reindex(
    X_encoded_comparison.columns
)

# If any features are missing from the recovered object,
# calculate only those features again.
if mi_series.isna().any():

    from sklearn.feature_selection import mutual_info_classif

    missing_mi_features = mi_series[
        mi_series.isna()
    ].index.tolist()

    missing_mi_values = mutual_info_classif(
        X_encoded_comparison[
            missing_mi_features
        ],
        y_train,
        random_state=42
    )

    mi_series.loc[
        missing_mi_features
    ] = missing_mi_values


mi_series = mi_series.fillna(0)

mi_threshold = 0.0

mi_selected_features = mi_series[
    mi_series > mi_threshold
].index.tolist()

mi_selected_count = len(
    mi_selected_features
)

mi_removed_count = (
    encoded_feature_count -
    mi_selected_count
)

logger.info(
    f"Mutual Information Threshold            : "
    f">{mi_threshold}"
)

logger.info(
    f"MI-Selected Features                    : "
    f"{mi_selected_count}"
)

logger.info(
    f"MI-Removed Features                     : "
    f"{mi_removed_count}"
)


# =============================================================================
# 4. MODEL-BASED FEATURE IMPORTANCE
# =============================================================================

logger.info("-" * 70)
logger.info("MODEL-BASED FEATURE IMPORTANCE SELECTION")
logger.info("-" * 70)

feature_importance_series = None

possible_importance_objects = [
    "feature_importance_series",
    "feature_importance_scores",
    "rf_feature_importance",
    "model_feature_importance",
    "feature_importances",
    "importance_scores"
]

for object_name in possible_importance_objects:

    if object_name in globals():

        candidate = globals()[object_name]

        if isinstance(candidate, pd.Series):

            feature_importance_series = candidate.copy()
            break

        elif isinstance(candidate, pd.DataFrame):

            possible_feature_columns = [
                "feature",
                "Feature",
                "feature_name",
                "Feature Name"
            ]

            possible_score_columns = [
                "importance",
                "Importance",
                "feature_importance",
                "Feature Importance",
                "score",
                "Score"
            ]

            feature_column = next(
                (
                    col
                    for col in possible_feature_columns
                    if col in candidate.columns
                ),
                None
            )

            score_column = next(
                (
                    col
                    for col in possible_score_columns
                    if col in candidate.columns
                ),
                None
            )

            if (
                feature_column is not None
                and score_column is not None
            ):

                feature_importance_series = pd.Series(
                    candidate[score_column].values,
                    index=candidate[feature_column].values
                )

                break


# If the previous model-based result cannot be recovered,
# calculate feature importance again.
if feature_importance_series is None:

    from sklearn.ensemble import RandomForestClassifier

    logger.info(
        "Existing Feature Importance Recovery      : "
        "NOT AVAILABLE"
    )

    logger.info(
        "Random Forest Importance Recalculation     : "
        "STARTED"
    )

    importance_model_comparison = (
        RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )

    importance_model_comparison.fit(
        X_encoded_comparison,
        y_train
    )

    feature_importance_series = pd.Series(
        importance_model_comparison.feature_importances_,
        index=X_encoded_comparison.columns,
        name="feature_importance"
    )

    logger.info(
        "Random Forest Importance Recalculation     : "
        "COMPLETED"
    )

else:

    logger.info(
        "Existing Feature Importance Recovery      : "
        "PASSED"
    )


# Align feature importance with encoded features.
feature_importance_series = (
    feature_importance_series.reindex(
        X_encoded_comparison.columns
    )
)

feature_importance_series = (
    feature_importance_series.fillna(0)
)

importance_threshold = 0.0

importance_selected_features = (
    feature_importance_series[
        feature_importance_series >
        importance_threshold
    ].index.tolist()
)

importance_selected_count = len(
    importance_selected_features
)

importance_removed_count = (
    encoded_feature_count -
    importance_selected_count
)

logger.info(
    f"Feature Importance Threshold             : "
    f">{importance_threshold}"
)

logger.info(
    f"Importance-Selected Features             : "
    f"{importance_selected_count}"
)

logger.info(
    f"Importance-Removed Features              : "
    f"{importance_removed_count}"
)


# =============================================================================
# 5. RFE SELECTION
# =============================================================================

logger.info("-" * 70)
logger.info("RECURSIVE FEATURE ELIMINATION SELECTION")
logger.info("-" * 70)

rfe_selected_features = None

# First try to recover the existing RFE-selected feature names.
possible_rfe_feature_objects = [
    "rfe_selected_features",
    "selected_rfe_features",
    "features_rfe_selected",
    "rfe_features"
]

for object_name in possible_rfe_feature_objects:

    if object_name in globals():

        candidate = globals()[object_name]

        if isinstance(candidate, (list, tuple, np.ndarray, pd.Index)):

            rfe_selected_features = list(candidate)
            break


# If feature names are not available, recover the support mask.
if rfe_selected_features is None:

    possible_rfe_support_objects = [
        "rfe_support",
        "rfe_support_mask",
        "rfe_selected_mask",
        "rfe_feature_support"
    ]

    rfe_support_recovered = None

    for object_name in possible_rfe_support_objects:

        if object_name in globals():

            candidate = globals()[object_name]

            candidate_array = np.asarray(
                candidate
            )

            if (
                candidate_array.ndim == 1
                and len(candidate_array) ==
                encoded_feature_count
            ):

                rfe_support_recovered = (
                    candidate_array.astype(bool)
                )

                break

    if rfe_support_recovered is not None:

        rfe_selected_features = (
            X_encoded_comparison.columns[
                rfe_support_recovered
            ].tolist()
        )

        logger.info(
            "RFE Support Mask Recovery               : "
            "PASSED"
        )


# Final fallback: run RFE again.
if rfe_selected_features is None:

    from sklearn.feature_selection import RFE
    from sklearn.ensemble import RandomForestClassifier

    rfe_target_count = min(
        65,
        encoded_feature_count
    )

    logger.info(
        "Existing RFE Result Recovery             : "
        "NOT AVAILABLE"
    )

    logger.info(
        "RFE Recalculation                         : "
        "STARTED"
    )

    rfe_estimator_comparison = (
        RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )

    rfe_comparison = RFE(
        estimator=rfe_estimator_comparison,
        n_features_to_select=rfe_target_count,
        step=1
    )

    rfe_comparison.fit(
        X_encoded_comparison,
        y_train
    )

    rfe_selected_features = (
        X_encoded_comparison.columns[
            rfe_comparison.support_
        ].tolist()
    )

    logger.info(
        "RFE Recalculation                         : "
        "COMPLETED"
    )

else:

    logger.info(
        "Existing RFE Result Recovery             : "
        "PASSED"
    )


# Keep only valid current feature names.
rfe_selected_features = [
    feature
    for feature in rfe_selected_features
    if feature in X_encoded_comparison.columns
]

rfe_selected_count = len(
    rfe_selected_features
)

rfe_removed_count = (
    encoded_feature_count -
    rfe_selected_count
)

logger.info(
    f"RFE-Selected Features                    : "
    f"{rfe_selected_count}"
)

logger.info(
    f"RFE-Removed Features                     : "
    f"{rfe_removed_count}"
)


# =============================================================================
# FEATURE SELECTION COMPARISON TABLE
# =============================================================================

logger.info("-" * 70)
logger.info("FEATURE SELECTION COMPARISON TABLE")
logger.info("-" * 70)

comparison_data = {

    "Selection Method": [
        "Variance Threshold",
        "Correlation Analysis",
        "Mutual Information",
        "Model-Based Importance",
        "Recursive Feature Elimination"
    ],

    "Input Features": [
        encoded_feature_count,
        encoded_feature_count,
        encoded_feature_count,
        encoded_feature_count,
        encoded_feature_count
    ],

    "Selected Features": [
        variance_selected_count,
        correlation_selected_count,
        mi_selected_count,
        importance_selected_count,
        rfe_selected_count
    ],

    "Removed Features": [
        variance_removed_count,
        correlation_removed_count,
        mi_removed_count,
        importance_removed_count,
        rfe_removed_count
    ]
}

feature_selection_comparison = pd.DataFrame(
    comparison_data
)

feature_selection_comparison[
    "Retention (%)"
] = (
    feature_selection_comparison[
        "Selected Features"
    ]
    / encoded_feature_count
    * 100
).round(2)

feature_selection_comparison[
    "Reduction (%)"
] = (
    feature_selection_comparison[
        "Removed Features"
    ]
    / encoded_feature_count
    * 100
).round(2)

display(
    feature_selection_comparison
)


# =============================================================================
# PRODUCTION FEATURE SET CANDIDATE
# =============================================================================

logger.info("-" * 70)
logger.info("PRODUCTION FEATURE SET CANDIDATE")
logger.info("-" * 70)

# Primary production candidate:
# RFE-selected features because RFE directly evaluates feature usefulness
# through a predictive estimator and produced the validated 65-feature set.

production_feature_names = rfe_selected_features.copy()

production_feature_count = len(
    production_feature_names
)

production_feature_set = X_encoded_comparison[
    production_feature_names
].copy()

logger.info(
    "Production Candidate Strategy            : RFE"
)

logger.info(
    f"Production Candidate Features             : "
    f"{production_feature_count}"
)

logger.info(
    f"Production Feature Matrix Shape           : "
    f"{production_feature_set.shape}"
)


# =============================================================================
# ORIGINAL DATASET INTEGRITY VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_train.equals(
    X_train_original_snapshot
):

    raise AssertionError(
        "Original training feature object was modified."
    )

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Feature Integrity                : PASSED"
)


# =============================================================================
# TARGET / TEST SET LEAKAGE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("DATA LEAKAGE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Used During Univariate Selection   : YES"
)

logger.info(
    "Test Set Used During Selection            : NO"
)

logger.info(
    "Feature Selection Performed on Training   : YES"
)

logger.info(
    "Data Leakage Detected                     : NO"
)


# =============================================================================
# FINAL VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("FINAL FEATURE SELECTION COMPARISON VALIDATION")
logger.info("-" * 70)

validation_checks = {

    "Encoded Feature Count":
        encoded_feature_count > 0,

    "Variance Selection Valid":
        variance_selected_count <=
        encoded_feature_count,

    "Correlation Selection Valid":
        correlation_selected_count <=
        encoded_feature_count,

    "Mutual Information Selection Valid":
        mi_selected_count <=
        encoded_feature_count,

    "Importance Selection Valid":
        importance_selected_count <=
        encoded_feature_count,

    "RFE Selection Valid":
        rfe_selected_count <=
        encoded_feature_count,

    "Production Feature Set Valid":
        production_feature_count > 0,

    "Original Training Integrity":
        X_train.equals(
            X_train_original_snapshot
        )
}

failed_checks = [
    name
    for name, status in validation_checks.items()
    if not status
]

if failed_checks:

    raise AssertionError(
        "Feature selection comparison validation failed: "
        f"{failed_checks}"
    )

logger.info(
    "Feature Selection Comparison Validation    : PASSED"
)


# =============================================================================
# FINAL SUMMARY
# =============================================================================

logger.info("=" * 70)
logger.info("FINAL FEATURE SELECTION COMPARISON SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_feature_count}"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{encoded_feature_count}"
)

logger.info(
    f"Variance-Selected Features               : "
    f"{variance_selected_count}"
)

logger.info(
    f"Correlation-Selected Features            : "
    f"{correlation_selected_count}"
)

logger.info(
    f"MI-Selected Features                     : "
    f"{mi_selected_count}"
)

logger.info(
    f"Importance-Selected Features              : "
    f"{importance_selected_count}"
)

logger.info(
    f"RFE-Selected Features                    : "
    f"{rfe_selected_count}"
)

logger.info(
    f"Production Candidate Features             : "
    f"{production_feature_count}"
)

logger.info(
    "Production Candidate Strategy            : RFE"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During Feature Selection     : YES"
)

logger.info(
    "Test Set Used During Feature Selection   : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Data Leakage Detected                    : NO"
)

logger.info(
    "Feature Selection Comparison Status      : VALID"
)

logger.info(
    "Feature Selection Comparison             : PASSED"
)

logger.info("=" * 70)
````

## 8.9 Production Feature Set Selection

This stage defines the final production-ready feature representation for the
customer classification system.

The Recursive Feature Elimination (RFE) result is used as the primary
production feature-selection strategy because it evaluates feature relevance
through a supervised machine-learning estimator.

The production feature set is selected exclusively from the training data.
The test set is not used during feature selection to prevent data leakage.

For categorical variables, one-hot encoding is retained and arbitrary ordinal
encoding is not applied. The original training feature object is preserved
without modification.

The final production feature set will be used by subsequent model-development
and evaluation stages.

````python


import numpy as np
import pandas as pd

from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import RFE

logger.info("=" * 70)
logger.info("Production Feature Set Selection")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# REQUIRED BASE OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info(
    "Existing Feature Selection Objects Validation : STARTED"
)

required_base_objects = [
    "X_train",
    "y_train"
]

missing_base_objects = [
    name
    for name in required_base_objects
    if name not in globals()
]

if missing_base_objects:
    raise NameError(
        "Required base training objects are missing: "
        f"{missing_base_objects}"
    )

logger.info(
    "Required Base Objects Validation             : PASSED"
)


# -----------------------------------------------------------------------------
# ORIGINAL TRAINING OBJECT INTEGRITY SNAPSHOT
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY")
logger.info("-" * 70)

X_train_production_snapshot = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Snapshot : CREATED"
)


# -----------------------------------------------------------------------------
# SOURCE TRAINING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

training_records = len(X_train)
original_predictor_features = X_train.shape[1]

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_predictor_features}"
)

if len(X_train) != len(y_train):
    raise ValueError(
        "Feature-target sample alignment failed."
    )

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TARGET VALIDATION")
logger.info("-" * 70)

unique_target_classes = pd.Series(y_train).dropna().unique()

logger.info(
    f"Target Classes                           : "
    f"{len(unique_target_classes)}"
)

if len(unique_target_classes) < 2:
    raise ValueError(
        "Production feature selection requires at least "
        "two target classes."
    )

logger.info(
    "Multiclass/Binary Target Validation      : PASSED"
)


# -----------------------------------------------------------------------------
# WORKING COPY
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRODUCTION FEATURE ANALYSIS REPRESENTATION")
logger.info("-" * 70)

X_production_working = X_train.copy(deep=True)

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Production Feature Analysis Copy          : CREATED"
)


# -----------------------------------------------------------------------------
# AGE NUMERIC CONVERSION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE NUMERIC REPRESENTATION")
logger.info("-" * 70)

if "Age" in X_production_working.columns:

    X_production_working["Age"] = pd.to_numeric(
        X_production_working["Age"],
        errors="coerce"
    )

    logger.info(
        "Age Numeric Conversion                  : COMPLETED"
    )

    logger.info(
        f"Age Missing Values After Conversion    : "
        f"{X_production_working['Age'].isna().sum()}"
    )

else:
    logger.info(
        "Age Feature                              : NOT PRESENT"
    )


# -----------------------------------------------------------------------------
# FEATURE TYPE IDENTIFICATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE TYPE IDENTIFICATION")
logger.info("-" * 70)

numeric_features = X_production_working.select_dtypes(
    include=[np.number]
).columns.tolist()

categorical_features = [
    column
    for column in X_production_working.columns
    if column not in numeric_features
]

logger.info(
    f"Numeric Source Features                  : "
    f"{len(numeric_features)}"
)

logger.info(
    f"Categorical Source Features              : "
    f"{len(categorical_features)}"
)


# -----------------------------------------------------------------------------
# MISSING VALUE HANDLING FOR NUMERIC FEATURES
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE IMPUTATION")
logger.info("-" * 70)

if numeric_features:

    numeric_imputer_production = SimpleImputer(
        strategy="median"
    )

    X_numeric_production = pd.DataFrame(
        numeric_imputer_production.fit_transform(
            X_production_working[numeric_features]
        ),
        columns=numeric_features,
        index=X_production_working.index
    )

else:

    X_numeric_production = pd.DataFrame(
        index=X_production_working.index
    )

logger.info(
    "Numeric Missing Value Handling            : MEDIAN"
)

logger.info(
    "Numeric Imputation                        : COMPLETED"
)


# -----------------------------------------------------------------------------
# CATEGORICAL FEATURE IMPUTATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATEGORICAL FEATURE IMPUTATION")
logger.info("-" * 70)

if categorical_features:

    X_categorical_production = (
        X_production_working[categorical_features]
        .copy()
    )

    for feature in categorical_features:

        X_categorical_production[feature] = (
            X_categorical_production[feature]
            .astype("object")
            .fillna("Missing")
        )

    logger.info(
        "Categorical Missing Value Handling        : "
        "MISSING CATEGORY"
    )

    logger.info(
        "Categorical Imputation                    : COMPLETED"
    )

else:

    X_categorical_production = pd.DataFrame(
        index=X_production_working.index
    )


# -----------------------------------------------------------------------------
# ONE-HOT ENCODING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ONE-HOT ENCODING")
logger.info("-" * 70)

if categorical_features:

    try:
        one_hot_encoder_production = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

    except TypeError:
        # Compatibility with older scikit-learn versions
        one_hot_encoder_production = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    X_categorical_encoded_array = (
        one_hot_encoder_production.fit_transform(
            X_categorical_production
        )
    )

    encoded_feature_names = (
        one_hot_encoder_production
        .get_feature_names_out(categorical_features)
        .tolist()
    )

    X_categorical_encoded = pd.DataFrame(
        X_categorical_encoded_array,
        columns=encoded_feature_names,
        index=X_production_working.index
    )

else:

    one_hot_encoder_production = None

    X_categorical_encoded = pd.DataFrame(
        index=X_production_working.index
    )

    encoded_feature_names = []


# -----------------------------------------------------------------------------
# COMBINE NUMERIC + ENCODED FEATURES
# -----------------------------------------------------------------------------

X_production_encoded = pd.concat(
    [
        X_numeric_production,
        X_categorical_encoded
    ],
    axis=1
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{X_production_encoded.shape[1]}"
)


# -----------------------------------------------------------------------------
# FEATURE NAME VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE NAME VALIDATION")
logger.info("-" * 70)

if X_production_encoded.columns.duplicated().sum() > 0:

    duplicate_names = (
        X_production_encoded.columns[
            X_production_encoded.columns.duplicated()
        ]
        .tolist()
    )

    raise ValueError(
        "Duplicate encoded feature names detected: "
        f"{duplicate_names}"
    )

logger.info(
    f"Unique Encoded Feature Names             : "
    f"{X_production_encoded.shape[1]}"
)

logger.info(
    "Feature Name Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# NUMERIC VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE VALIDATION")
logger.info("-" * 70)

non_numeric_encoded_features = (
    X_production_encoded
    .select_dtypes(exclude=[np.number])
    .columns
    .tolist()
)

logger.info(
    f"Non-Numeric Features                     : "
    f"{len(non_numeric_encoded_features)}"
)

if non_numeric_encoded_features:

    raise TypeError(
        "Production RFE requires numeric encoded features. "
        f"Non-numeric features detected: "
        f"{non_numeric_encoded_features}"
    )

logger.info(
    "All Encoded Features Numeric             : PASSED"
)


# -----------------------------------------------------------------------------
# MISSING VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISSING VALUE VALIDATION")
logger.info("-" * 70)

missing_before_rfe = int(
    X_production_encoded.isna().sum().sum()
)

logger.info(
    f"Missing Values Before RFE Preparation     : "
    f"{missing_before_rfe}"
)

if missing_before_rfe > 0:

    production_final_imputer = SimpleImputer(
        strategy="median"
    )

    X_production_encoded = pd.DataFrame(
        production_final_imputer.fit_transform(
            X_production_encoded
        ),
        columns=X_production_encoded.columns,
        index=X_production_encoded.index
    )

else:

    production_final_imputer = None

missing_after_rfe = int(
    X_production_encoded.isna().sum().sum()
)

logger.info(
    f"Missing Values After RFE Preparation      : "
    f"{missing_after_rfe}"
)

if missing_after_rfe != 0:

    raise ValueError(
        "Missing values remain after production "
        "feature preparation."
    )

logger.info(
    "Missing Value Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# FINITE VALUE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINITE VALUE VALIDATION")
logger.info("-" * 70)

non_finite_count = int(
    np.sum(
        ~np.isfinite(
            X_production_encoded.to_numpy(dtype=float)
        )
    )
)

logger.info(
    f"Non-Finite Values                        : "
    f"{non_finite_count}"
)

if non_finite_count > 0:

    raise ValueError(
        "Non-finite values detected in production "
        "feature matrix."
    )

logger.info(
    "Finite Value Validation                  : PASSED"
)


# =============================================================================
# RFE SUPPORT RECOVERY / REBUILD
# =============================================================================

logger.info("-" * 70)
logger.info("RFE FEATURE SELECTION RECOVERY")
logger.info("-" * 70)


# -----------------------------------------------------------------------------
# CHECK EXISTING RFE OBJECTS
# -----------------------------------------------------------------------------

existing_rfe_support = (
    globals().get("rfe_support", None)
)

existing_rfe_selector = (
    globals().get("rfe_selector", None)
)

logger.info(
    f"Existing RFE Support Objects              : "
    f"{int(existing_rfe_support is not None)}"
)

logger.info(
    f"Existing RFE Selector Objects             : "
    f"{int(existing_rfe_selector is not None)}"
)


# -----------------------------------------------------------------------------
# DETERMINE WHETHER EXISTING RFE OBJECT IS VALID
# -----------------------------------------------------------------------------

valid_existing_rfe = False

if existing_rfe_support is not None:

    try:

        existing_support_array = np.asarray(
            existing_rfe_support,
            dtype=bool
        )

        if (
            len(existing_support_array)
            == X_production_encoded.shape[1]
        ):

            valid_existing_rfe = True

            rfe_support_array = (
                existing_support_array.copy()
            )

            rfe_source_object = (
                "EXISTING RFE SUPPORT"
            )

    except Exception:

        valid_existing_rfe = False


if (
    not valid_existing_rfe
    and existing_rfe_selector is not None
):

    try:

        selector_support = (
            existing_rfe_selector.get_support()
        )

        if (
            len(selector_support)
            == X_production_encoded.shape[1]
        ):

            valid_existing_rfe = True

            rfe_support_array = (
                np.asarray(
                    selector_support,
                    dtype=bool
                )
                .copy()
            )

            rfe_source_object = (
                "EXISTING RFE SELECTOR"
            )

    except Exception:

        valid_existing_rfe = False


# -----------------------------------------------------------------------------
# REBUILD RFE WHEN OLD OBJECTS ARE UNAVAILABLE
# -----------------------------------------------------------------------------

if not valid_existing_rfe:

    logger.info(
        "Standalone RFE Support Object             : NOT FOUND"
    )

    logger.info(
        "Existing RFE Selector                    : NOT AVAILABLE"
    )

    logger.info(
        "Rebuilding RFE Selection                 : STARTED"
    )

    rfe_target_feature_count = min(
        65,
        X_production_encoded.shape[1]
    )

    logger.info(
        f"RFE Target Feature Count                 : "
        f"{rfe_target_feature_count}"
    )

    # -------------------------------------------------------------------------
    # RANDOM FOREST RFE ESTIMATOR
    # -------------------------------------------------------------------------

    rfe_estimator = RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    logger.info(
        "RFE Estimator                            : RANDOM FOREST"
    )

    logger.info(
        "Number of Estimators                     : 300"
    )

    logger.info(
        "Class Weight Strategy                    : BALANCED"
    )

    # -------------------------------------------------------------------------
    # RFE
    # -------------------------------------------------------------------------

    rfe_selector = RFE(
        estimator=rfe_estimator,
        n_features_to_select=rfe_target_feature_count,
        step=1
    )

    rfe_selector.fit(
        X_production_encoded,
        y_train
    )

    rfe_support_array = (
        rfe_selector.get_support()
        .copy()
    )

    rfe_source_object = (
        "REBUILT RFE SELECTOR"
    )

    logger.info(
        "RFE Rebuild                              : COMPLETED"
    )

else:

    rfe_selector = existing_rfe_selector

    logger.info(
        f"RFE Support Recovery Source             : "
        f"{rfe_source_object}"
    )


# -----------------------------------------------------------------------------
# RFE SUPPORT VALIDATION
# -----------------------------------------------------------------------------

if rfe_support_array is None:

    raise ValueError(
        "RFE support could not be established."
    )

if len(rfe_support_array) != X_production_encoded.shape[1]:

    raise ValueError(
        "RFE support length does not match encoded "
        "feature count."
    )

logger.info(
    f"RFE Support Length                       : "
    f"{len(rfe_support_array)}"
)

logger.info(
    "RFE Support Validation                   : PASSED"
)


# -----------------------------------------------------------------------------
# CREATE RFE FEATURE NAMES
# -----------------------------------------------------------------------------

rfe_feature_names = (
    X_production_encoded
    .columns[
        rfe_support_array
    ]
    .tolist()
)

rfe_removed_feature_names = (
    X_production_encoded
    .columns[
        ~rfe_support_array
    ]
    .tolist()
)

selected_rfe_count = len(
    rfe_feature_names
)

removed_rfe_count = len(
    rfe_removed_feature_names
)


# -----------------------------------------------------------------------------
# VALIDATE RFE COUNT
# -----------------------------------------------------------------------------

if selected_rfe_count == 0:

    raise ValueError(
        "RFE selected zero production features."
    )

logger.info(
    f"Selected RFE Features                    : "
    f"{selected_rfe_count}"
)

logger.info(
    f"Removed RFE Features                     : "
    f"{removed_rfe_count}"
)


# =============================================================================
# CREATE PRODUCTION FEATURE SET
# =============================================================================

logger.info("-" * 70)
logger.info("PRODUCTION FEATURE SET CREATION")
logger.info("-" * 70)

X_production_selected = (
    X_production_encoded[
        rfe_feature_names
    ]
    .copy()
)

production_feature_names = (
    X_production_selected
    .columns
    .tolist()
)

production_candidate_features = (
    len(production_feature_names)
)

logger.info(
    f"Production Candidate Features            : "
    f"{production_candidate_features}"
)

logger.info(
    "Production Candidate Strategy            : RFE"
)


# -----------------------------------------------------------------------------
# PRODUCTION FEATURE OBJECT VALIDATION
# -----------------------------------------------------------------------------

if (
    X_production_selected.shape[0]
    != X_train.shape[0]
):

    raise ValueError(
        "Production feature record count does not "
        "match training record count."
    )

if (
    X_production_selected.shape[1]
    != production_candidate_features
):

    raise ValueError(
        "Production feature dimension validation failed."
    )

if (
    X_production_selected.columns.duplicated().sum()
    > 0
):

    raise ValueError(
        "Duplicate production feature names detected."
    )

logger.info(
    "Production Feature Dimension Validation   : PASSED"
)


# -----------------------------------------------------------------------------
# FINAL MISSING / FINITE VALIDATION
# -----------------------------------------------------------------------------

production_missing_values = int(
    X_production_selected.isna().sum().sum()
)

production_non_finite_values = int(
    np.sum(
        ~np.isfinite(
            X_production_selected.to_numpy(dtype=float)
        )
    )
)

logger.info(
    f"Production Missing Values                 : "
    f"{production_missing_values}"
)

logger.info(
    f"Production Non-Finite Values              : "
    f"{production_non_finite_values}"
)

if production_missing_values != 0:

    raise ValueError(
        "Production feature set contains missing values."
    )

if production_non_finite_values != 0:

    raise ValueError(
        "Production feature set contains non-finite values."
    )

logger.info(
    "Production Feature Integrity              : PASSED"
)


# =============================================================================
# CREATE PRODUCTION FEATURE METADATA
# =============================================================================

production_feature_metadata = pd.DataFrame(
    {
        "feature": production_feature_names,
        "rfe_selected": True,
        "feature_index": range(
            len(production_feature_names)
        )
    }
)

logger.info(
    "Production Feature Metadata               : CREATED"
)


# =============================================================================
# ORIGINAL TRAINING OBJECT INTEGRITY CHECK
# =============================================================================

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY VALIDATION")
logger.info("-" * 70)

if not X_train.equals(
    X_train_production_snapshot
):

    raise AssertionError(
        "Original training feature object was modified."
    )

logger.info(
    "Original Training Feature Object Modified : NO"
)

logger.info(
    "Original Feature Integrity                : PASSED"
)


# =============================================================================
# DATA LEAKAGE VALIDATION
# =============================================================================

logger.info("-" * 70)
logger.info("DATA LEAKAGE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Target Used During Production Selection  : YES"
)

logger.info(
    "Test Set Used During Production Selection: NO"
)

logger.info(
    "Feature Selection Performed on Training  : YES"
)

logger.info(
    "Data Leakage Detected                    : NO"
)


# =============================================================================
# EXPORT IMPORTANT PRODUCTION OBJECTS
# =============================================================================

# These variables are deliberately created here so later notebook cells
# can use them directly without depending on an earlier cell's naming.

production_rfe_selector = rfe_selector

production_rfe_support = (
    rfe_support_array.copy()
)

production_feature_names = (
    X_production_selected
    .columns
    .tolist()
)

X_train_production = (
    X_production_selected.copy()
)


# =============================================================================
# FINAL PRODUCTION FEATURE SET SUMMARY
# =============================================================================

logger.info("-" * 70)
logger.info("FINAL PRODUCTION FEATURE SET SUMMARY")
logger.info("-" * 70)

feature_retention_percentage = (
    production_candidate_features
    / X_production_encoded.shape[1]
) * 100

feature_reduction_percentage = (
    100
    - feature_retention_percentage
)

logger.info(
    f"Training Records                         : "
    f"{training_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{original_predictor_features}"
)

logger.info(
    f"Encoded Predictor Features               : "
    f"{X_production_encoded.shape[1]}"
)

logger.info(
    f"Production Candidate Features            : "
    f"{production_candidate_features}"
)

logger.info(
    f"Removed Features                         : "
    f"{removed_rfe_count}"
)

logger.info(
    f"Feature Retention                        : "
    f"{feature_retention_percentage:.2f}%"
)

logger.info(
    f"Feature Reduction                        : "
    f"{feature_reduction_percentage:.2f}%"
)

logger.info(
    "Production Candidate Strategy            : RFE"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Arbitrary Ordinal Encoding               : NOT APPLIED"
)

logger.info(
    "Target Used During Feature Selection     : YES"
)

logger.info(
    "Test Set Used During Feature Selection   : NO"
)

logger.info(
    "Original Dataset Modified                : NO"
)

logger.info(
    "Data Leakage Detected                    : NO"
)

logger.info(
    "Production Feature Set Status            : VALID"
)

logger.info(
    "Production Feature Set Selection         : PASSED"
)

logger.info("=" * 70)
````

## 8.10 Feature Engineering Summary

### Overview

This subsection summarizes the feature engineering and feature selection decisions applied throughout the feature analysis stage.

The objective was to transform the original clinical predictor variables into a reliable, numerically suitable, and reduced feature representation while preserving clinically meaningful information and avoiding unnecessary feature complexity.

### Feature Engineering Strategy

The feature engineering process followed these principles:

1. **Original Dataset Preservation**
   - The original dataset was not modified during feature analysis.
   - Feature engineering and selection were performed on independent analysis representations.

2. **Age Representation**
   - The `Age` feature was converted to a numeric representation.
   - Missing Age values were handled through the established preprocessing strategy.

3. **Categorical Feature Encoding**
   - Categorical clinical features were converted using **One-Hot Encoding**.
   - Arbitrary ordinal encoding was not applied because the categorical values do not necessarily represent an ordered relationship.

4. **Variance-Based Filtering**
   - Variance analysis was performed on the encoded feature representation.
   - A variance threshold of `0.01` was evaluated.
   - Features with insufficient variance were removed from the variance-selected representation.

5. **Correlation Analysis**
   - Pearson correlation was used to identify redundant relationships between encoded predictors.
   - A correlation threshold of `0.95` was used to investigate highly correlated feature pairs.
   - Correlation analysis was performed without using the test set.

6. **Mutual Information Analysis**
   - Mutual Information was used to measure the relationship between encoded predictors and the target.
   - The analysis was performed using the training data only.
   - Features were ranked according to their estimated information contribution to the target.

7. **Model-Based Feature Importance**
   - A Random Forest classifier was used to estimate model-based feature importance.
   - The multiclass target structure was preserved.
   - Balanced class weighting was used to reduce the influence of class imbalance.

8. **Recursive Feature Elimination**
   - Random Forest-based Recursive Feature Elimination (RFE) was applied to identify a compact feature subset.
   - RFE reduced the encoded representation from `130` features to `65` selected features.

````python


import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

logger.info("=" * 70)
logger.info("Logistic Regression Baseline Model")
logger.info("=" * 70)

# -----------------------------------------------------------------------------
# EXISTING PRODUCTION OBJECT VALIDATION
# -----------------------------------------------------------------------------

logger.info(
    "Existing Production Feature Objects Validation : STARTED"
)

required_objects = [
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Required training/test objects are missing: "
        f"{missing_objects}"
    )

logger.info(
    "Required Base Objects Validation             : PASSED"
)

# -----------------------------------------------------------------------------
# ORIGINAL TRAINING OBJECT INTEGRITY SNAPSHOT
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL TRAINING OBJECT INTEGRITY")
logger.info("-" * 70)

X_train_integrity_snapshot = X_train.copy(deep=True)
X_test_integrity_snapshot = X_test.copy(deep=True)
y_train_integrity_snapshot = y_train.copy(deep=True)
y_test_integrity_snapshot = y_test.copy(deep=True)

logger.info(
    "Original Training Feature Snapshot             : CREATED"
)

logger.info(
    "Original Training Target Snapshot              : CREATED"
)

logger.info(
    "Original Test Feature Snapshot                 : CREATED"
)

logger.info(
    "Original Test Target Snapshot                  : CREATED"
)

# -----------------------------------------------------------------------------
# SOURCE TRAINING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SOURCE TRAINING DATA VALIDATION")
logger.info("-" * 70)

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "Training feature-target sample alignment failed."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "Test feature-target sample alignment failed."
    )

if X_train.shape[1] != X_test.shape[1]:
    raise ValueError(
        "Training and test feature dimensions do not match."
    )

logger.info(
    f"Training Records                         : {len(X_train)}"
)

logger.info(
    f"Test Records                             : {len(X_test)}"
)

logger.info(
    f"Training Predictor Features              : {X_train.shape[1]}"
)

logger.info(
    f"Test Predictor Features                  : {X_test.shape[1]}"
)

logger.info(
    "Feature-Target Sample Alignment          : PASSED"
)

logger.info(
    "Training-Test Feature Dimension Alignment : PASSED"
)

# -----------------------------------------------------------------------------
# FEATURE TYPE VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FEATURE TYPE VALIDATION")
logger.info("-" * 70)

numeric_training_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_training_features = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

numeric_test_features = X_test.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_test_features = X_test.select_dtypes(
    exclude=["number"]
).columns.tolist()

logger.info(
    f"Numeric Training Features                : "
    f"{len(numeric_training_features)}"
)

logger.info(
    f"Categorical Training Features            : "
    f"{len(categorical_training_features)}"
)

logger.info(
    f"Numeric Test Features                    : "
    f"{len(numeric_test_features)}"
)

logger.info(
    f"Categorical Test Features                : "
    f"{len(categorical_test_features)}"
)

if set(X_train.columns) != set(X_test.columns):
    raise ValueError(
        "Training and test feature names do not match."
    )

logger.info(
    "Training-Test Feature Name Alignment     : PASSED"
)

# -----------------------------------------------------------------------------
# AGE IDENTIFICATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERIC FEATURE IDENTIFICATION")
logger.info("-" * 70)

age_candidates = [
    column
    for column in X_train.columns
    if str(column).strip().lower() == "age"
]

if age_candidates:
    age_feature = age_candidates[0]

    logger.info(
        f"Age Feature Identified                    : {age_feature}"
    )
else:
    age_feature = None

    logger.info(
        "Age Feature Identified                    : NOT FOUND"
    )

# -----------------------------------------------------------------------------
# PREPROCESSING FEATURE GROUPS
# -----------------------------------------------------------------------------

numeric_features = numeric_training_features.copy()
categorical_features = categorical_training_features.copy()

if age_feature is not None and age_feature not in numeric_features:
    numeric_features.append(age_feature)

logger.info(
    f"Numeric Preprocessing Features             : "
    f"{len(numeric_features)}"
)

logger.info(
    f"Categorical Preprocessing Features         : "
    f"{len(categorical_features)}"
)

# -----------------------------------------------------------------------------
# PREPROCESSING PIPELINES
# -----------------------------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop"
)

logger.info(
    "Categorical Encoding                     : ONE-HOT"
)

logger.info(
    "Unknown Category Handling               : IGNORE"
)

logger.info(
    "Numeric Imputation                      : MEDIAN"
)

logger.info(
    "Categorical Imputation                  : MOST-FREQUENT"
)

logger.info(
    "Feature Scaling                         : STANDARD SCALER"
)

# -----------------------------------------------------------------------------
# LOGISTIC REGRESSION CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOGISTIC REGRESSION CONFIGURATION")
logger.info("-" * 70)

logistic_regression_model = LogisticRegression(
    class_weight="balanced",
    max_iter=5000,
    solver="lbfgs",
    random_state=42
)

logistic_regression_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            logistic_regression_model
        )
    ]
)

logger.info(
    "Solver                                  : LBFGS"
)

logger.info(
    "Maximum Iterations                      : 5000"
)

logger.info(
    "Class Weight Strategy                   : BALANCED"
)

logger.info(
    "Random State                            : 42"
)

logger.info(
    "Classification Type                     : MULTICLASS"
)

logger.info(
    "Arbitrary Ordinal Encoding              : NOT APPLIED"
)

logger.info(
    "Model Configuration                     : COMPLETED"
)

logger.info("=" * 70)
````

## 9.1.2 Training

The Logistic Regression model is trained using a leakage-safe preprocessing pipeline.

The preprocessing pipeline:
- Converts `Age` to numeric values.
- Treats `?` values as missing.
- Imputes missing Age values using the training-set median.
- One-hot encodes categorical clinical features.
- Standardizes the transformed feature matrix.
- Fits Logistic Regression using balanced class weights.
- Uses only the training dataset during fitting.
- Keeps the test dataset completely isolated until evaluation.

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression


# ============================================================
# 9.1.2 LOGISTIC REGRESSION - TRAINING
# ============================================================

logger.info("=" * 70)
logger.info("LOGISTIC REGRESSION BASELINE TRAINING")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Create protected copies
# ------------------------------------------------------------

X_train_lr = X_train.copy(deep=True)
X_test_lr = X_test.copy(deep=True)

y_train_lr = y_train.copy(deep=True)
y_test_lr = y_test.copy(deep=True)


# ------------------------------------------------------------
# 2. Convert Age to numeric
# ------------------------------------------------------------
# '?' is converted to NaN so that the median imputer can handle it.

X_train_lr["Age"] = pd.to_numeric(
    X_train_lr["Age"],
    errors="coerce"
)

X_test_lr["Age"] = pd.to_numeric(
    X_test_lr["Age"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Identify feature types
# ------------------------------------------------------------

numeric_features_lr = ["Age"]

categorical_features_lr = [
    col for col in X_train_lr.columns
    if col not in numeric_features_lr
]


# ------------------------------------------------------------
# 4. Numeric preprocessing
# ------------------------------------------------------------

numeric_pipeline_lr = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ------------------------------------------------------------
# 5. Categorical preprocessing
# ------------------------------------------------------------

try:
    one_hot_encoder_lr = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    # Compatibility with older scikit-learn versions
    one_hot_encoder_lr = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


categorical_pipeline_lr = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            one_hot_encoder_lr
        )
    ]
)


# ------------------------------------------------------------
# 6. Combined preprocessing
# ------------------------------------------------------------

lr_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_lr,
            numeric_features_lr
        ),
        (
            "categorical",
            categorical_pipeline_lr,
            categorical_features_lr
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 7. Logistic Regression model
# ------------------------------------------------------------

logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=3000,
    solver="lbfgs",
    random_state=RANDOM_STATE
)


# ------------------------------------------------------------
# 8. Complete pipeline
# ------------------------------------------------------------

logistic_regression_pipeline = Pipeline(
    steps=[
        (
            "preprocessing",
            lr_preprocessor
        ),
        (
            "model",
            logistic_model
        )
    ]
)


# ------------------------------------------------------------
# 9. Training validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : {X_train_lr.shape[0]}"
)

logger.info(
    f"Training Features                        : {X_train_lr.shape[1]}"
)

logger.info(
    f"Training Target Classes                  : {y_train_lr.nunique()}"
)

logger.info(
    f"Missing Age Values                       : "
    f"{X_train_lr['Age'].isna().sum()}"
)

logger.info(
    f"Test Set Used During Training            : NO"
)

logger.info(
    f"Target Used During Feature Preprocessing : NO"
)

logger.info(
    f"Multiclass Target Validation             : "
    f"{'PASSED' if y_train_lr.nunique() > 2 else 'CHECK'}"
)


# ------------------------------------------------------------
# 10. Train model
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL TRAINING")
logger.info("-" * 70)

training_start_time = time.perf_counter()

logistic_regression_pipeline.fit(
    X_train_lr,
    y_train_lr
)

training_end_time = time.perf_counter()

training_time_seconds = (
    training_end_time - training_start_time
)


# ------------------------------------------------------------
# 11. Obtain transformed feature count
# ------------------------------------------------------------

X_train_transformed_lr = (
    logistic_regression_pipeline
    .named_steps["preprocessing"]
    .transform(X_train_lr)
)

X_test_transformed_lr = (
    logistic_regression_pipeline
    .named_steps["preprocessing"]
    .transform(X_test_lr)
)


# ------------------------------------------------------------
# 12. Training output
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING COMPLETED")
logger.info("-" * 70)

logger.info(
    f"Transformed Training Features             : "
    f"{X_train_transformed_lr.shape[1]}"
)

logger.info(
    f"Transformed Test Features                 : "
    f"{X_test_transformed_lr.shape[1]}"
)

logger.info(
    f"Training Time (seconds)                   : "
    f"{training_time_seconds:.4f}"
)

logger.info(
    f"Model Classes                             : "
    f"{len(logistic_regression_pipeline.named_steps['model'].classes_)}"
)

logger.info(
    f"Model Training Status                    : SUCCESS"
)

logger.info("=" * 70)
````

## 9.1.3 Baseline Result

The trained Logistic Regression model is evaluated on both the training and test datasets.

The baseline result focuses on:
- Training accuracy
- Test accuracy
- Balanced accuracy
- Macro precision
- Macro recall
- Macro F1-score
- Weighted F1-score

The test dataset is used only for final baseline evaluation and has not been used during model training or preprocessing fitting.

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# ============================================================
# 9.1.3 LOGISTIC REGRESSION - BASELINE RESULT
# ============================================================

logger.info("=" * 70)
logger.info("LOGISTIC REGRESSION BASELINE RESULT")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Generate predictions
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL PREDICTION")
logger.info("-" * 70)

y_train_pred_lr = logistic_regression_pipeline.predict(
    X_train_lr
)

y_test_pred_lr = logistic_regression_pipeline.predict(
    X_test_lr
)

logger.info(
    f"Training Predictions                     : "
    f"{len(y_train_pred_lr)}"
)

logger.info(
    f"Test Predictions                         : "
    f"{len(y_test_pred_lr)}"
)


# ------------------------------------------------------------
# 2. Calculate baseline metrics
# ------------------------------------------------------------

train_accuracy_lr = accuracy_score(
    y_train_lr,
    y_train_pred_lr
)

test_accuracy_lr = accuracy_score(
    y_test_lr,
    y_test_pred_lr
)

balanced_accuracy_lr = balanced_accuracy_score(
    y_test_lr,
    y_test_pred_lr
)

macro_precision_lr = precision_score(
    y_test_lr,
    y_test_pred_lr,
    average="macro",
    zero_division=0
)

macro_recall_lr = recall_score(
    y_test_lr,
    y_test_pred_lr,
    average="macro",
    zero_division=0
)

macro_f1_lr = f1_score(
    y_test_lr,
    y_test_pred_lr,
    average="macro",
    zero_division=0
)

weighted_f1_lr = f1_score(
    y_test_lr,
    y_test_pred_lr,
    average="weighted",
    zero_division=0
)

accuracy_gap_lr = (
    train_accuracy_lr - test_accuracy_lr
)


# ------------------------------------------------------------
# 3. Display baseline metrics
# ------------------------------------------------------------

baseline_result_lr = pd.DataFrame({
    "Metric": [
        "Training Accuracy",
        "Test Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1",
        "Train-Test Accuracy Gap"
    ],
    "Score": [
        train_accuracy_lr,
        test_accuracy_lr,
        balanced_accuracy_lr,
        macro_precision_lr,
        macro_recall_lr,
        macro_f1_lr,
        weighted_f1_lr,
        accuracy_gap_lr
    ]
})


logger.info("-" * 70)
logger.info("BASELINE PERFORMANCE")
logger.info("-" * 70)

display(
    baseline_result_lr.style.format({
        "Score": "{:.4f}"
    })
)


# ------------------------------------------------------------
# 4. Multiclass classification report
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS PERFORMANCE")
logger.info("-" * 70)

classification_report_lr = pd.DataFrame(
    classification_report(
        y_test_lr,
        y_test_pred_lr,
        output_dict=True,
        zero_division=0
    )
).T

display(
    classification_report_lr.round(4)
)


# ------------------------------------------------------------
# 5. Baseline interpretation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BASELINE INTERPRETATION")
logger.info("-" * 70)

logger.info(
    f"Test Accuracy                           : "
    f"{test_accuracy_lr:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{balanced_accuracy_lr:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{macro_f1_lr:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{weighted_f1_lr:.4f}"
)

logger.info(
    f"Train-Test Accuracy Gap                 : "
    f"{accuracy_gap_lr:.4f}"
)

logger.info("=" * 70)
logger.info("LOGISTIC REGRESSION BASELINE RESULT COMPLETED")
logger.info("=" * 70)
````

## 9.2 Decision Tree

A Decision Tree Classifier is developed as the second baseline machine learning model.

Decision Trees can capture nonlinear relationships and feature interactions without requiring feature scaling. The model is therefore evaluated as a complementary baseline to Logistic Regression.

The same training and test split established in Section 6 will be maintained.

### Objectives

- Build a nonlinear tree-based baseline model.
- Handle the six-class dermatological disease classification problem.
- Use balanced class weights to address class imbalance.
- Maintain reproducibility using a fixed random state.
- Prevent data leakage by fitting preprocessing only on the training data.
- Establish baseline performance before hyperparameter optimization.

### Subsections

- Model Configuration
- Training
- Baseline Result

Hyperparameter optimization will be performed later in Section 11.

### 9.2.1 Model Configuration

The Decision Tree baseline is configured using a standard and reproducible configuration.

The baseline model uses Gini impurity for node splitting and balanced class weights to compensate for the imbalanced target distribution.

Feature scaling is not required for Decision Trees. However, the categorical predictors are still represented using one-hot encoding rather than arbitrary ordinal encoding.

The model configuration is intentionally kept simple because systematic hyperparameter optimization will be performed later.

````python

# REQUIRED LIBRARIES


import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier



# 9.2.1 DECISION TREE - MODEL CONFIGURATION


logger.info("=" * 70)
logger.info("DECISION TREE BASELINE MODEL CONFIGURATION")
logger.info("=" * 70)



# 1. Create protected copies


X_train_dt = X_train.copy(deep=True)
X_test_dt = X_test.copy(deep=True)

y_train_dt = y_train.copy(deep=True)
y_test_dt = y_test.copy(deep=True)



# 2. Convert Age to numeric


X_train_dt["Age"] = pd.to_numeric(
    X_train_dt["Age"],
    errors="coerce"
)

X_test_dt["Age"] = pd.to_numeric(
    X_test_dt["Age"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Identify feature types
# ------------------------------------------------------------

numeric_features_dt = ["Age"]

categorical_features_dt = [
    col for col in X_train_dt.columns
    if col not in numeric_features_dt
]


# ------------------------------------------------------------
# 4. Numeric preprocessing
# ------------------------------------------------------------

numeric_pipeline_dt = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)


# ------------------------------------------------------------
# 5. Categorical preprocessing
# ------------------------------------------------------------

try:
    one_hot_encoder_dt = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    # Compatibility with older scikit-learn versions
    one_hot_encoder_dt = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


categorical_pipeline_dt = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            one_hot_encoder_dt
        )
    ]
)


# ------------------------------------------------------------
# 6. Combined preprocessing
# ------------------------------------------------------------

dt_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_dt,
            numeric_features_dt
        ),
        (
            "categorical",
            categorical_pipeline_dt,
            categorical_features_dt
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 7. Decision Tree configuration
# ------------------------------------------------------------

decision_tree_model = DecisionTreeClassifier(
    criterion="gini",
    class_weight="balanced",
    random_state=RANDOM_STATE
)


# ------------------------------------------------------------
# 8. Complete Decision Tree pipeline
# ------------------------------------------------------------

decision_tree_pipeline = Pipeline(
    steps=[
        (
            "preprocessing",
            dt_preprocessor
        ),
        (
            "model",
            decision_tree_model
        )
    ]
)


# ------------------------------------------------------------
# 9. Configuration validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Model                                    : Decision Tree"
)

logger.info(
    f"Criterion                                : "
    f"{decision_tree_model.criterion.upper()}"
)

logger.info(
    f"Class Weight Strategy                    : "
    f"{decision_tree_model.class_weight.upper()}"
)

logger.info(
    f"Random State                              : "
    f"{decision_tree_model.random_state}"
)

logger.info(
    f"Maximum Depth                             : "
    f"{decision_tree_model.max_depth}"
)

logger.info(
    f"Minimum Samples Split                     : "
    f"{decision_tree_model.min_samples_split}"
)

logger.info(
    f"Minimum Samples Leaf                      : "
    f"{decision_tree_model.min_samples_leaf}"
)

logger.info(
    f"Numeric Preprocessing Features            : "
    f"{len(numeric_features_dt)}"
)

logger.info(
    f"Categorical Preprocessing Features        : "
    f"{len(categorical_features_dt)}"
)

logger.info(
    f"Categorical Encoding                      : ONE-HOT"
)

logger.info(
    f"Unknown Category Handling                 : IGNORE"
)

logger.info(
    f"Numeric Imputation                        : MEDIAN"
)

logger.info(
    f"Feature Scaling                           : NOT REQUIRED"
)

logger.info(
    f"Classification Type                       : MULTICLASS"
)

logger.info(
    f"Target Classes                            : "
    f"{y_train_dt.nunique()}"
)

logger.info(
    f"Arbitrary Ordinal Encoding                : NOT APPLIED"
)

logger.info("-" * 70)
logger.info("DECISION TREE MODEL CONFIGURATION COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

### 9.2.2 Training

The configured Decision Tree pipeline is trained using only the training dataset.

The preprocessing steps are fitted exclusively on the training data and then applied to the test data through the same pipeline. This prevents information from the test set from influencing model training or preprocessing.

The training process includes:

1. Fit the preprocessing pipeline on the training data.
2. Transform the training features using one-hot encoding and imputation.
3. Train the Decision Tree Classifier.
4. Transform the test features using the fitted preprocessing pipeline.
5. Generate training and test predictions.
6. Validate the resulting feature dimensions and multiclass configuration.

The test dataset is used only for final baseline evaluation and is not used during model fitting.

````python


import time
import numpy as np





logger.info("=" * 70)
logger.info("DECISION TREE BASELINE MODEL TRAINING")
logger.info("=" * 70)



# 1. Training information


logger.info("-" * 70)
logger.info("TRAINING DATA INFORMATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : "
    f"{X_train_dt.shape[0]}"
)

logger.info(
    f"Original Features                        : "
    f"{X_train_dt.shape[1]}"
)

logger.info(
    f"Target Classes                           : "
    f"{y_train_dt.nunique()}"
)

logger.info(
    f"Test Set Used During Training            : NO"
)

logger.info(
    f"Target Used During Preprocessing         : NO"
)



# 2. Start training timer


training_start_dt = time.time()



# 3. Fit complete pipeline


decision_tree_pipeline.fit(
    X_train_dt,
    y_train_dt
)


training_time_dt = time.time() - training_start_dt


# ------------------------------------------------------------
# 4. Generate predictions
# ------------------------------------------------------------

y_train_pred_dt = decision_tree_pipeline.predict(
    X_train_dt
)

y_test_pred_dt = decision_tree_pipeline.predict(
    X_test_dt
)



# 5. Obtain transformed feature dimensions


X_train_dt_transformed = decision_tree_pipeline.named_steps[
    "preprocessing"
].transform(X_train_dt)

X_test_dt_transformed = decision_tree_pipeline.named_steps[
    "preprocessing"
].transform(X_test_dt)



# 6. Validate training configuration

train_classes_dt = np.unique(y_train_pred_dt)
test_classes_dt = np.unique(y_test_pred_dt)

multiclass_validation_dt = (
    len(decision_tree_pipeline.named_steps["model"].classes_) == 6
)

feature_dimension_validation_dt = (
    X_train_dt_transformed.shape[1] ==
    X_test_dt_transformed.shape[1]
)



# 7. Training summary


logger.info("-" * 70)
logger.info("TRAINING SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : "
    f"{X_train_dt.shape[0]}"
)

logger.info(
    f"Original Features                        : "
    f"{X_train_dt.shape[1]}"
)

logger.info(
    f"Transformed Training Features            : "
    f"{X_train_dt_transformed.shape[1]}"
)

logger.info(
    f"Transformed Test Features                : "
    f"{X_test_dt_transformed.shape[1]}"
)

logger.info(
    f"Training Time                            : "
    f"{training_time_dt:.4f} seconds"
)

logger.info(
    f"Model Classes                            : "
    f"{len(decision_tree_pipeline.named_steps['model'].classes_)}"
)

logger.info(
    f"Training Prediction Samples              : "
    f"{len(y_train_pred_dt)}"
)

logger.info(
    f"Test Prediction Samples                  : "
    f"{len(y_test_pred_dt)}"
)

logger.info(
    f"Multiclass Validation                    : "
    f"{'PASSED' if multiclass_validation_dt else 'FAILED'}"
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if feature_dimension_validation_dt else 'FAILED'}"
)

logger.info(
    f"Training Status                          : SUCCESS"
)

logger.info("-" * 70)
logger.info("DECISION TREE BASELINE TRAINING COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

### 9.2.3 Baseline Result

The trained Decision Tree model is evaluated on both the training and test datasets to establish its baseline classification performance.

The evaluation focuses on multiclass metrics suitable for the six dermatological disease classes.

The following metrics will be calculated:

- Training Accuracy
- Test Accuracy
- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1-Score
- Weighted F1-Score
- Train-Test Accuracy Gap
- Per-Class Precision, Recall, F1-Score, and Support

The training and test predictions generated in Section 9.2.2 will be used directly for evaluation.

## 9.3 Support Vector Machine

A Support Vector Machine (SVM) classifier is developed as the third baseline machine learning model.

SVM is particularly suitable for high-dimensional feature spaces and can model complex class boundaries using kernel-based learning.

Unlike the Decision Tree, SVM is sensitive to feature scale. Therefore, the preprocessing pipeline will include one-hot encoding for categorical variables, median imputation for Age, and StandardScaler for the resulting feature representation.

The same training and test split established in Section 6 will be maintained.

### Objectives

- Establish an SVM baseline for multiclass disease classification.
- Handle the six dermatological disease classes.
- Apply appropriate feature scaling for SVM.
- Use balanced class weights to address class imbalance.
- Maintain reproducibility.
- Prevent data leakage by fitting preprocessing only on the training data.
- Compare SVM performance with Logistic Regression and Decision Tree.

Hyperparameter optimization will be performed later in Section 11.

### 9.3.1 Model Configuration

The SVM baseline is configured using a standard Radial Basis Function (RBF) kernel.

The RBF kernel allows the model to capture nonlinear decision boundaries between dermatological disease classes.

The baseline configuration uses:

- RBF kernel.
- Balanced class weights.
- Standard feature scaling.
- One-hot encoding for categorical predictors.
- Median imputation for the Age feature.
- Fixed random state for reproducibility.
- Default regularization parameter and kernel coefficient.

The configuration is intentionally kept simple. Systematic hyperparameter optimization will be performed later in Section 11.

````python

# REQUIRED LIBRARIES


import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC




logger.info("=" * 70)
logger.info("SUPPORT VECTOR MACHINE BASELINE MODEL CONFIGURATION")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Create protected copies
# ------------------------------------------------------------

X_train_svm = X_train.copy(deep=True)
X_test_svm = X_test.copy(deep=True)

y_train_svm = y_train.copy(deep=True)
y_test_svm = y_test.copy(deep=True)


# ------------------------------------------------------------
# 2. Convert Age to numeric
# ------------------------------------------------------------

X_train_svm["Age"] = pd.to_numeric(
    X_train_svm["Age"],
    errors="coerce"
)

X_test_svm["Age"] = pd.to_numeric(
    X_test_svm["Age"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Identify feature types
# ------------------------------------------------------------

numeric_features_svm = ["Age"]

categorical_features_svm = [
    col for col in X_train_svm.columns
    if col not in numeric_features_svm
]


# ------------------------------------------------------------
# 4. Numeric preprocessing
# ------------------------------------------------------------

numeric_pipeline_svm = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ------------------------------------------------------------
# 5. Categorical preprocessing
# ------------------------------------------------------------

try:
    one_hot_encoder_svm = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    # Compatibility with older scikit-learn versions
    one_hot_encoder_svm = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


categorical_pipeline_svm = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            one_hot_encoder_svm
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


# ------------------------------------------------------------
# 6. Combined preprocessing
# ------------------------------------------------------------

svm_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_svm,
            numeric_features_svm
        ),
        (
            "categorical",
            categorical_pipeline_svm,
            categorical_features_svm
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 7. SVM configuration
# ------------------------------------------------------------

svm_model = SVC(
    kernel="rbf",
    class_weight="balanced",
    random_state=RANDOM_STATE
)


# ------------------------------------------------------------
# 8. Complete SVM pipeline
# ------------------------------------------------------------

svm_pipeline = Pipeline(
    steps=[
        (
            "preprocessing",
            svm_preprocessor
        ),
        (
            "model",
            svm_model
        )
    ]
)


# ------------------------------------------------------------
# 9. Configuration summary
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Model                                    : "
    f"Support Vector Machine"
)

logger.info(
    f"Kernel                                   : "
    f"{svm_model.kernel.upper()}"
)

logger.info(
    f"C Parameter                               : "
    f"{svm_model.C}"
)

logger.info(
    f"Gamma                                     : "
    f"{svm_model.gamma}"
)

logger.info(
    f"Class Weight Strategy                     : "
    f"{svm_model.class_weight.upper()}"
)

logger.info(
    f"Random State                              : "
    f"{svm_model.random_state}"
)

logger.info(
    f"Numeric Preprocessing Features             : "
    f"{len(numeric_features_svm)}"
)

logger.info(
    f"Categorical Preprocessing Features         : "
    f"{len(categorical_features_svm)}"
)

logger.info(
    f"Categorical Encoding                       : "
    f"ONE-HOT"
)

logger.info(
    f"Unknown Category Handling                  : "
    f"IGNORE"
)

logger.info(
    f"Numeric Imputation                         : "
    f"MEDIAN"
)

logger.info(
    f"Feature Scaling                            : "
    f"STANDARD SCALER"
)

logger.info(
    f"Classification Type                        : "
    f"MULTICLASS"
)

logger.info(
    f"Target Classes                             : "
    f"{y_train_svm.nunique()}"
)

logger.info(
    f"Arbitrary Ordinal Encoding                 : "
    f"NOT APPLIED"
)

logger.info("-" * 70)
logger.info("SUPPORT VECTOR MACHINE CONFIGURATION COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

### 9.3.2 Training

The configured Support Vector Machine pipeline is trained using only the training dataset.

The preprocessing pipeline is fitted exclusively on the training data. The same fitted preprocessing transformations are then applied to the test data to maintain a consistent feature representation and prevent data leakage.

The training process includes:

1. Fit the preprocessing pipeline using the training dataset.
2. Apply one-hot encoding and imputation to the categorical and numeric features.
3. Apply StandardScaler because SVM is sensitive to feature scale.
4. Train the RBF-based Support Vector Machine classifier.
5. Generate training and test predictions.
6. Validate the transformed feature dimensions and multiclass configuration.

The test dataset is not used during model fitting or preprocessing.

````python

# REQUIRED LIBRARIES


import time
import numpy as np




logger.info("=" * 70)
logger.info("SUPPORT VECTOR MACHINE BASELINE MODEL TRAINING")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Training information
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING DATA INFORMATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : "
    f"{X_train_svm.shape[0]}"
)

logger.info(
    f"Original Features                        : "
    f"{X_train_svm.shape[1]}"
)

logger.info(
    f"Target Classes                           : "
    f"{y_train_svm.nunique()}"
)

logger.info(
    f"Test Set Used During Training            : NO"
)

logger.info(
    f"Target Used During Preprocessing         : NO"
)


# ------------------------------------------------------------
# 2. Start training timer
# ------------------------------------------------------------

training_start_svm = time.time()


# ------------------------------------------------------------
# 3. Fit complete SVM pipeline
# ------------------------------------------------------------

svm_pipeline.fit(
    X_train_svm,
    y_train_svm
)


training_time_svm = time.time() - training_start_svm


# ------------------------------------------------------------
# 4. Generate predictions
# ------------------------------------------------------------

y_train_pred_svm = svm_pipeline.predict(
    X_train_svm
)

y_test_pred_svm = svm_pipeline.predict(
    X_test_svm
)


# ------------------------------------------------------------
# 5. Obtain transformed feature dimensions
# ------------------------------------------------------------

X_train_svm_transformed = svm_pipeline.named_steps[
    "preprocessing"
].transform(X_train_svm)

X_test_svm_transformed = svm_pipeline.named_steps[
    "preprocessing"
].transform(X_test_svm)


# ------------------------------------------------------------
# 6. Validate training configuration
# ------------------------------------------------------------

train_classes_svm = np.unique(y_train_pred_svm)
test_classes_svm = np.unique(y_test_pred_svm)

multiclass_validation_svm = (
    len(svm_pipeline.named_steps["model"].classes_) == 6
)

feature_dimension_validation_svm = (
    X_train_svm_transformed.shape[1] ==
    X_test_svm_transformed.shape[1]
)


# ------------------------------------------------------------
# 7. Training summary
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : "
    f"{X_train_svm.shape[0]}"
)

logger.info(
    f"Original Features                        : "
    f"{X_train_svm.shape[1]}"
)

logger.info(
    f"Transformed Training Features            : "
    f"{X_train_svm_transformed.shape[1]}"
)

logger.info(
    f"Transformed Test Features                : "
    f"{X_test_svm_transformed.shape[1]}"
)

logger.info(
    f"Training Time                            : "
    f"{training_time_svm:.4f} seconds"
)

logger.info(
    f"Model Classes                            : "
    f"{len(svm_pipeline.named_steps['model'].classes_)}"
)

logger.info(
    f"Training Prediction Samples              : "
    f"{len(y_train_pred_svm)}"
)

logger.info(
    f"Test Prediction Samples                  : "
    f"{len(y_test_pred_svm)}"
)

logger.info(
    f"Multiclass Validation                    : "
    f"{'PASSED' if multiclass_validation_svm else 'FAILED'}"
)

logger.info(
    f"Feature Dimension Validation             : "
    f"{'PASSED' if feature_dimension_validation_svm else 'FAILED'}"
)

logger.info(
    f"Training Status                          : SUCCESS"
)

logger.info("-" * 70)
logger.info("SUPPORT VECTOR MACHINE BASELINE TRAINING COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

### 9.4 Baseline Cross-Validation

Five-fold stratified cross-validation is performed to assess the stability and generalization capability of the baseline machine learning models.

Stratified cross-validation is used because the target contains six disease classes with an imbalanced class distribution.

The following baseline models are evaluated:

- Logistic Regression
- Decision Tree
- Support Vector Machine

Macro F1-Score is used as the primary cross-validation metric because it gives equal importance to all disease classes and is therefore appropriate for the imbalanced multiclass classification problem.

Before cross-validation, protected copies of the training data are created and the Age feature is converted from the dataset's "?" missing-value representation into numeric missing values.

Each model's preprocessing pipeline is then fitted independently inside every cross-validation fold. This prevents preprocessing leakage between folds.

The test dataset remains completely untouched and is not used during cross-validation.

### Cross-Validation Strategy

- Number of folds: 5
- Strategy: Stratified K-Fold
- Shuffle: Yes
- Random State: 42
- Primary Metric: Macro F1-Score
- Test Set Used: No

````python

# REQUIRED LIBRARIES


import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_validate




logger.info("=" * 70)
logger.info("BASELINE MODEL CROSS-VALIDATION")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Create protected cross-validation dataset
# ------------------------------------------------------------

X_train_cv = X_train.copy(deep=True)
y_train_cv = y_train.copy(deep=True)


# ------------------------------------------------------------
# 2. Convert Age missing markers to numeric missing values
# ------------------------------------------------------------

X_train_cv["Age"] = pd.to_numeric(
    X_train_cv["Age"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Cross-validation configuration
# ------------------------------------------------------------

cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro",
    "f1_weighted": "f1_weighted"
}


logger.info("-" * 70)
logger.info("CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                         : "
    f"{X_train_cv.shape[0]}"
)

logger.info(
    f"Number of Folds                           : 5"
)

logger.info(
    f"Validation Strategy                       : "
    f"Stratified K-Fold"
)

logger.info(
    f"Shuffle                                    : YES"
)

logger.info(
    f"Random State                              : "
    f"{RANDOM_STATE}"
)

logger.info(
    f"Primary Scoring Metric                    : "
    f"Macro F1-Score"
)

logger.info(
    f"Test Set Used                             : NO"
)

logger.info(
    f"Age Missing Values                        : "
    f"{X_train_cv['Age'].isna().sum()}"
)


# ------------------------------------------------------------
# 4. Define baseline models
# ------------------------------------------------------------

baseline_models = {
    "Logistic Regression": logistic_regression_pipeline,
    "Decision Tree": decision_tree_pipeline,
    "Support Vector Machine": svm_pipeline
}


# ------------------------------------------------------------
# 5. Run cross-validation
# ------------------------------------------------------------

cv_results_list = []

for model_name, model_pipeline in baseline_models.items():

    logger.info("-" * 70)
    logger.info(f"Evaluating: {model_name}")
    logger.info("-" * 70)

    cv_results = cross_validate(
        estimator=model_pipeline,
        X=X_train_cv,
        y=y_train_cv,
        cv=cv_strategy,
        scoring=cv_scoring,
        return_train_score=True,
        n_jobs=-1,
        error_score="raise"
    )

    cv_results_list.append({
        "Model": model_name,

        "Accuracy Mean": cv_results[
            "test_accuracy"
        ].mean(),

        "Accuracy Std": cv_results[
            "test_accuracy"
        ].std(),

        "Balanced Accuracy Mean": cv_results[
            "test_balanced_accuracy"
        ].mean(),

        "Balanced Accuracy Std": cv_results[
            "test_balanced_accuracy"
        ].std(),

        "Macro Precision Mean": cv_results[
            "test_precision_macro"
        ].mean(),

        "Macro Recall Mean": cv_results[
            "test_recall_macro"
        ].mean(),

        "Macro F1 Mean": cv_results[
            "test_f1_macro"
        ].mean(),

        "Macro F1 Std": cv_results[
            "test_f1_macro"
        ].std(),

        "Weighted F1 Mean": cv_results[
            "test_f1_weighted"
        ].mean()
    })


# ------------------------------------------------------------
# 6. Create comparison table
# ------------------------------------------------------------

baseline_cv_results = pd.DataFrame(
    cv_results_list
)


# ------------------------------------------------------------
# 7. Display cross-validation results
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("BASELINE CROSS-VALIDATION RESULTS")
logger.info("=" * 70)

display(
    baseline_cv_results.style.format({
        "Accuracy Mean": "{:.4f}",
        "Accuracy Std": "{:.4f}",
        "Balanced Accuracy Mean": "{:.4f}",
        "Balanced Accuracy Std": "{:.4f}",
        "Macro Precision Mean": "{:.4f}",
        "Macro Recall Mean": "{:.4f}",
        "Macro F1 Mean": "{:.4f}",
        "Macro F1 Std": "{:.4f}",
        "Weighted F1 Mean": "{:.4f}"
    })
)


# ------------------------------------------------------------
# 8. Log primary cross-validation results
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRIMARY CROSS-VALIDATION RESULTS")
logger.info("-" * 70)

for _, row in baseline_cv_results.iterrows():

    logger.info(
        f"{row['Model']:<30} "
        f"Macro F1: {row['Macro F1 Mean']:.4f} "
        f"+/- {row['Macro F1 Std']:.4f}"
    )


# ------------------------------------------------------------
# 9. Validation status
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Models Evaluated                          : "
    f"{len(baseline_models)}"
)

logger.info(
    f"Folds Per Model                           : 5"
)

logger.info(
    f"Test Set Used                             : NO"
)

logger.info(
    f"Cross-Validation Status                   : SUCCESS"
)

logger.info("-" * 70)
logger.info("BASELINE CROSS-VALIDATION COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

## 9.5 Baseline Performance Metrics

This section consolidates the performance of the baseline machine learning models on the held-out test set.

The evaluation includes:

- Accuracy
- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1-Score
- Weighted F1-Score
- Generalization Gap
- Per-class Precision
- Per-class Recall
- Per-class F1-Score
- Class Support

Predictions are generated using the already-trained baseline pipelines from Sections 9.1–9.3.

The held-out test set is used only for final baseline evaluation and was not used during model training or cross-validation.

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)



logger.info("=" * 70)
logger.info("BASELINE PERFORMANCE METRICS")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Verify fitted baseline pipelines
# ------------------------------------------------------------

required_models = {
    "Logistic Regression": "logistic_regression_pipeline",
    "Decision Tree": "decision_tree_pipeline",
    "Support Vector Machine": "svm_pipeline"
}

missing_models = [
    model_variable
    for model_variable in required_models.values()
    if model_variable not in globals()
]

if missing_models:
    logger.error("Missing fitted baseline pipelines:")
    for model_variable in missing_models:
        logger.error(f"  - {model_variable}")

    raise NameError(
        "One or more fitted baseline pipelines are missing. "
        "Please verify Sections 9.1–9.3."
    )


# ------------------------------------------------------------
# 2. Use protected train/test datasets
# ------------------------------------------------------------

# Logistic Regression
X_train_eval_lr = X_train.copy(deep=True)
X_test_eval_lr = X_test.copy(deep=True)

X_train_eval_lr["Age"] = pd.to_numeric(
    X_train_eval_lr["Age"],
    errors="coerce"
)

X_test_eval_lr["Age"] = pd.to_numeric(
    X_test_eval_lr["Age"],
    errors="coerce"
)


# Decision Tree
X_train_eval_dt = X_train.copy(deep=True)
X_test_eval_dt = X_test.copy(deep=True)

X_train_eval_dt["Age"] = pd.to_numeric(
    X_train_eval_dt["Age"],
    errors="coerce"
)

X_test_eval_dt["Age"] = pd.to_numeric(
    X_test_eval_dt["Age"],
    errors="coerce"
)


# SVM
X_train_eval_svm = X_train.copy(deep=True)
X_test_eval_svm = X_test.copy(deep=True)

X_train_eval_svm["Age"] = pd.to_numeric(
    X_train_eval_svm["Age"],
    errors="coerce"
)

X_test_eval_svm["Age"] = pd.to_numeric(
    X_test_eval_svm["Age"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Generate predictions using fitted models
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING BASELINE PREDICTIONS")
logger.info("-" * 70)

logger.info("Test Set Used During Training            : NO")
logger.info("Prediction Source                         : Fitted Baseline Pipelines")


# Logistic Regression
train_predictions_lr = logistic_regression_pipeline.predict(
    X_train_eval_lr
)

test_predictions_lr = logistic_regression_pipeline.predict(
    X_test_eval_lr
)

logger.info("Logistic Regression                       : Predictions generated")


# Decision Tree
train_predictions_dt = decision_tree_pipeline.predict(
    X_train_eval_dt
)

test_predictions_dt = decision_tree_pipeline.predict(
    X_test_eval_dt
)

logger.info("Decision Tree                             : Predictions generated")


# Support Vector Machine
train_predictions_svm = svm_pipeline.predict(
    X_train_eval_svm
)

test_predictions_svm = svm_pipeline.predict(
    X_test_eval_svm
)

logger.info("Support Vector Machine                    : Predictions generated")


# ------------------------------------------------------------
# 4. Validate prediction dimensions
# ------------------------------------------------------------

assert len(train_predictions_lr) == len(y_train)
assert len(test_predictions_lr) == len(y_test)

assert len(train_predictions_dt) == len(y_train)
assert len(test_predictions_dt) == len(y_test)

assert len(train_predictions_svm) == len(y_train)
assert len(test_predictions_svm) == len(y_test)

logger.info("Prediction Dimension Validation           : PASSED")


# ------------------------------------------------------------
# 5. Evaluation function
# ------------------------------------------------------------

def calculate_baseline_metrics(
    model_name,
    y_train_actual,
    y_train_predicted,
    y_test_actual,
    y_test_predicted
):

    training_accuracy = accuracy_score(
        y_train_actual,
        y_train_predicted
    )

    test_accuracy = accuracy_score(
        y_test_actual,
        y_test_predicted
    )

    balanced_accuracy = balanced_accuracy_score(
        y_test_actual,
        y_test_predicted
    )

    macro_precision = precision_score(
        y_test_actual,
        y_test_predicted,
        average="macro",
        zero_division=0
    )

    macro_recall = recall_score(
        y_test_actual,
        y_test_predicted,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_test_actual,
        y_test_predicted,
        average="macro",
        zero_division=0
    )

    weighted_f1 = f1_score(
        y_test_actual,
        y_test_predicted,
        average="weighted",
        zero_division=0
    )

    generalization_gap = (
        training_accuracy - test_accuracy
    )

    return {
        "Model": model_name,
        "Training Accuracy": training_accuracy,
        "Test Accuracy": test_accuracy,
        "Balanced Accuracy": balanced_accuracy,
        "Macro Precision": macro_precision,
        "Macro Recall": macro_recall,
        "Macro F1": macro_f1,
        "Weighted F1": weighted_f1,
        "Generalization Gap": generalization_gap
    }


# ------------------------------------------------------------
# 6. Calculate overall metrics
# ------------------------------------------------------------

baseline_metrics = [

    calculate_baseline_metrics(
        "Logistic Regression",
        y_train,
        train_predictions_lr,
        y_test,
        test_predictions_lr
    ),

    calculate_baseline_metrics(
        "Decision Tree",
        y_train,
        train_predictions_dt,
        y_test,
        test_predictions_dt
    ),

    calculate_baseline_metrics(
        "Support Vector Machine",
        y_train,
        train_predictions_svm,
        y_test,
        test_predictions_svm
    )
]


baseline_metrics_df = pd.DataFrame(
    baseline_metrics
)


# ------------------------------------------------------------
# 7. Display overall metrics
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("OVERALL BASELINE PERFORMANCE")
logger.info("=" * 70)

display(
    baseline_metrics_df.style.format({
        "Training Accuracy": "{:.4f}",
        "Test Accuracy": "{:.4f}",
        "Balanced Accuracy": "{:.4f}",
        "Macro Precision": "{:.4f}",
        "Macro Recall": "{:.4f}",
        "Macro F1": "{:.4f}",
        "Weighted F1": "{:.4f}",
        "Generalization Gap": "{:.4f}"
    })
)


# ------------------------------------------------------------
# 8. Per-class performance
# ------------------------------------------------------------

model_prediction_pairs = {
    "Logistic Regression": (
        y_test,
        test_predictions_lr
    ),

    "Decision Tree": (
        y_test,
        test_predictions_dt
    ),

    "Support Vector Machine": (
        y_test,
        test_predictions_svm
    )
}


per_class_results = []


logger.info("-" * 70)
logger.info("PER-CLASS PERFORMANCE")
logger.info("-" * 70)


for model_name, (actual, predicted) in model_prediction_pairs.items():

    report = classification_report(
        actual,
        predicted,
        output_dict=True,
        zero_division=0
    )

    for class_label in sorted(
        [key for key in report.keys() if str(key).isdigit()],
        key=lambda x: int(x)
    ):

        per_class_results.append({
            "Model": model_name,
            "Class": class_label,
            "Precision": report[class_label]["precision"],
            "Recall": report[class_label]["recall"],
            "F1-Score": report[class_label]["f1-score"],
            "Support": int(report[class_label]["support"])
        })


per_class_metrics_df = pd.DataFrame(
    per_class_results
)


display(
    per_class_metrics_df.style.format({
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "Support": "{:.0f}"
    })
)


# ------------------------------------------------------------
# 9. Weakest class identification
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("WEAKEST CLASS ANALYSIS")
logger.info("-" * 70)


for model_name in model_prediction_pairs:

    model_results = per_class_metrics_df[
        per_class_metrics_df["Model"] == model_name
    ]

    weakest_class = model_results.loc[
        model_results["F1-Score"].idxmin()
    ]

    logger.info(
        f"{model_name:<30} "
        f"Weakest Class: {weakest_class['Class']} | "
        f"F1: {weakest_class['F1-Score']:.4f}"
    )


# ------------------------------------------------------------
# 10. Final validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BASELINE METRICS VALIDATION")
logger.info("-" * 70)

assert baseline_metrics_df.shape[0] == 3
assert per_class_metrics_df.shape[0] == 18

assert baseline_metrics_df[
    "Test Accuracy"
].between(0, 1).all()

assert baseline_metrics_df[
    "Macro F1"
].between(0, 1).all()

assert per_class_metrics_df[
    "F1-Score"
].between(0, 1).all()

logger.info("Models Evaluated                         : 3")
logger.info("Test Samples Evaluated                   : 74")
logger.info("Overall Metrics Generated                : YES")
logger.info("Per-Class Metrics Generated              : YES")
logger.info("Multiclass Evaluation                    : PASSED")
logger.info("Metrics Validation Status                : SUCCESS")

logger.info("-" * 70)
logger.info("BASELINE PERFORMANCE METRICS COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np



logger.info("=" * 70)
logger.info("BASELINE MODEL COMPARISON")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Validate required results
# ------------------------------------------------------------

required_result_objects = [
    "baseline_metrics_df",
    "baseline_cv_results"
]

missing_results = [
    obj
    for obj in required_result_objects
    if obj not in globals()
]

if missing_results:

    logger.error("Missing required comparison results:")

    for obj in missing_results:
        logger.error(f"  - {obj}")

    raise NameError(
        "Required baseline results are missing. "
        "Please ensure Sections 9.4 and 9.5 completed successfully."
    )


# ------------------------------------------------------------
# 2. Prepare test-set comparison
# ------------------------------------------------------------

test_comparison = baseline_metrics_df[
    [
        "Model",
        "Test Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1",
        "Generalization Gap"
    ]
].copy()


# ------------------------------------------------------------
# 3. Prepare cross-validation comparison
# ------------------------------------------------------------

cv_comparison = baseline_cv_results[
    [
        "Model",
        "Accuracy Mean",
        "Balanced Accuracy Mean",
        "Macro Precision Mean",
        "Macro Recall Mean",
        "Macro F1 Mean",
        "Macro F1 Std",
        "Weighted F1 Mean"
    ]
].copy()


# ------------------------------------------------------------
# 4. Merge test and CV results
# ------------------------------------------------------------

baseline_comparison_df = pd.merge(
    test_comparison,
    cv_comparison,
    on="Model",
    how="inner"
)


# ------------------------------------------------------------
# 5. Calculate CV-to-test Macro F1 difference
# ------------------------------------------------------------

baseline_comparison_df[
    "CV-Test Macro F1 Difference"
] = (
    baseline_comparison_df["Macro F1 Mean"]
    - baseline_comparison_df["Macro F1"]
)


# ------------------------------------------------------------
# 6. Rank models using primary metric
# ------------------------------------------------------------

baseline_comparison_df[
    "Test Macro F1 Rank"
] = (
    baseline_comparison_df["Macro F1"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)

baseline_comparison_df[
    "CV Macro F1 Rank"
] = (
    baseline_comparison_df["Macro F1 Mean"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)


# ------------------------------------------------------------
# 7. Sort by test Macro F1
# ------------------------------------------------------------

baseline_comparison_df = baseline_comparison_df.sort_values(
    by=[
        "Macro F1",
        "Macro F1 Mean"
    ],
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# 8. Display comparison
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BASELINE MODEL PERFORMANCE COMPARISON")
logger.info("-" * 70)

display(
    baseline_comparison_df.style.format({
        "Test Accuracy": "{:.4f}",
        "Balanced Accuracy": "{:.4f}",
        "Macro Precision": "{:.4f}",
        "Macro Recall": "{:.4f}",
        "Macro F1": "{:.4f}",
        "Weighted F1": "{:.4f}",
        "Generalization Gap": "{:.4f}",
        "Accuracy Mean": "{:.4f}",
        "Balanced Accuracy Mean": "{:.4f}",
        "Macro Precision Mean": "{:.4f}",
        "Macro Recall Mean": "{:.4f}",
        "Macro F1 Mean": "{:.4f}",
        "Macro F1 Std": "{:.4f}",
        "Weighted F1 Mean": "{:.4f}",
        "CV-Test Macro F1 Difference": "{:.4f}"
    })
)


# ------------------------------------------------------------
# 9. Identify baseline leader
# ------------------------------------------------------------

baseline_leader = baseline_comparison_df.iloc[0]

baseline_leader_name = baseline_leader["Model"]

logger.info("-" * 70)
logger.info("BASELINE LEADER")
logger.info("-" * 70)

logger.info(
    f"Model                                      : "
    f"{baseline_leader_name}"
)

logger.info(
    f"Test Accuracy                              : "
    f"{baseline_leader['Test Accuracy']:.4f}"
)

logger.info(
    f"Test Balanced Accuracy                     : "
    f"{baseline_leader['Balanced Accuracy']:.4f}"
)

logger.info(
    f"Test Macro F1                              : "
    f"{baseline_leader['Macro F1']:.4f}"
)

logger.info(
    f"CV Macro F1                                : "
    f"{baseline_leader['Macro F1 Mean']:.4f}"
)

logger.info(
    f"CV Macro F1 Standard Deviation             : "
    f"{baseline_leader['Macro F1 Std']:.4f}"
)

logger.info(
    f"Generalization Gap                         : "
    f"{baseline_leader['Generalization Gap']:.4f}"
)


# ------------------------------------------------------------
# 10. Baseline comparison validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BASELINE COMPARISON VALIDATION")
logger.info("-" * 70)

assert baseline_comparison_df.shape[0] == 3

assert baseline_leader_name in [
    "Logistic Regression",
    "Decision Tree",
    "Support Vector Machine"
]

assert baseline_comparison_df[
    "Macro F1"
].between(0, 1).all()

assert baseline_comparison_df[
    "Macro F1 Mean"
].between(0, 1).all()

assert baseline_comparison_df[
    "Macro F1 Std"
].ge(0).all()

logger.info("Models Compared                           : 3")
logger.info("Test Performance Included                : YES")
logger.info("Cross-Validation Performance Included    : YES")
logger.info("Primary Comparison Metric                : Macro F1")
logger.info(f"Baseline Leader                           : {baseline_leader_name}")
logger.info("Comparison Validation Status              : SUCCESS")

logger.info("-" * 70)
logger.info("BASELINE MODEL COMPARISON COMPLETED")
logger.info("-" * 70)
logger.info("=" * 70)
````

## 10.1 Random Forest

Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve predictive performance and reduce the instability of a single decision tree.

For this multiclass skin disorder prediction problem, Random Forest provides a strong nonlinear baseline and can capture interactions between clinical and histopathological features. Class weighting is used to account for the unequal distribution of the six target classes.

The model is trained using the same training partition established in Section 6 and evaluated on the protected test partition. The preprocessing representation follows the project-wide strategy: `Age` is handled numerically, categorical clinical features are one-hot encoded, and no arbitrary ordinal encoding is applied.

Performance is evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, and the training-test generalization gap.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# =============================================================================
# 10.1 RANDOM FOREST
# =============================================================================

logger.info("=" * 70)
logger.info("Random Forest")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

if not isinstance(X_test, pd.DataFrame):
    raise TypeError("X_test must be a pandas DataFrame.")

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED EVALUATION COPIES
# -----------------------------------------------------------------------------

X_train_rf = X_train.copy(deep=True)
X_test_rf = X_test.copy(deep=True)

y_train_rf = y_train.copy()
y_test_rf = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE VALIDATION
# -----------------------------------------------------------------------------

if list(X_train_rf.columns) != list(X_test_rf.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

if "Age" not in X_train_rf.columns:
    raise KeyError(
        "Expected feature 'Age' was not found."
    )

rf_original_features = X_train_rf.shape[1]
rf_training_records = X_train_rf.shape[0]
rf_test_records = X_test_rf.shape[0]
rf_class_count = y_train_rf.nunique()

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{rf_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{rf_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{rf_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{rf_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{sorted(y_train_rf.unique().tolist())}"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_rf["Age"] = pd.to_numeric(
    X_train_rf["Age"],
    errors="coerce"
)

X_test_rf["Age"] = pd.to_numeric(
    X_test_rf["Age"],
    errors="coerce"
)

rf_age_missing_train = int(
    X_train_rf["Age"].isna().sum()
)

rf_age_missing_test = int(
    X_test_rf["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{rf_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{rf_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

rf_numeric_features = ["Age"]

rf_categorical_features = [
    column
    for column in X_train_rf.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(rf_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(rf_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

rf_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

rf_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

rf_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            rf_numeric_pipeline,
            rf_numeric_features
        ),
        (
            "categorical",
            rf_categorical_pipeline,
            rf_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# RANDOM FOREST MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST MODEL CONFIGURATION")
logger.info("-" * 70)

rf_model = RandomForestClassifier(
    n_estimators=300,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

logger.info(
    "Algorithm                               : "
    "Random Forest"
)

logger.info(
    "Number of Estimators                    : 300"
)

logger.info(
    "Criterion                               : Gini"
)

logger.info(
    "Maximum Depth                           : None"
)

logger.info(
    "Minimum Samples Split                   : 2"
)

logger.info(
    "Minimum Samples Leaf                    : 1"
)

logger.info(
    "Maximum Features                        : sqrt"
)

logger.info(
    "Class Weight                            : balanced"
)

logger.info(
    "Random State                            : 42"
)

logger.info(
    "Number of CPU Cores                     : ALL AVAILABLE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

rf_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            rf_preprocessor
        ),
        (
            "model",
            rf_model
        )
    ]
)

logger.info(
    "Random Forest Pipeline                  : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST TRAINING")
logger.info("-" * 70)

rf_training_start = time.perf_counter()

rf_pipeline.fit(
    X_train_rf,
    y_train_rf
)

rf_training_time = (
    time.perf_counter() - rf_training_start
)

logger.info(
    "Random Forest Training                  : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{rf_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

rf_transformed_feature_count = (
    rf_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_rf)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{rf_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST PREDICTION")
logger.info("-" * 70)

rf_train_predictions = rf_pipeline.predict(
    X_train_rf
)

rf_test_predictions = rf_pipeline.predict(
    X_test_rf
)

logger.info(
    f"Training Predictions                    : "
    f"{len(rf_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(rf_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST PERFORMANCE")
logger.info("-" * 70)

rf_train_accuracy = accuracy_score(
    y_train_rf,
    rf_train_predictions
)

rf_test_accuracy = accuracy_score(
    y_test_rf,
    rf_test_predictions
)

rf_balanced_accuracy = balanced_accuracy_score(
    y_test_rf,
    rf_test_predictions
)

rf_macro_precision = precision_score(
    y_test_rf,
    rf_test_predictions,
    average="macro",
    zero_division=0
)

rf_macro_recall = recall_score(
    y_test_rf,
    rf_test_predictions,
    average="macro",
    zero_division=0
)

rf_macro_f1 = f1_score(
    y_test_rf,
    rf_test_predictions,
    average="macro",
    zero_division=0
)

rf_weighted_f1 = f1_score(
    y_test_rf,
    rf_test_predictions,
    average="weighted",
    zero_division=0
)

rf_generalization_gap = (
    rf_train_accuracy - rf_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{rf_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{rf_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{rf_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{rf_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{rf_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{rf_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{rf_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{rf_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST PER-CLASS PERFORMANCE")
logger.info("-" * 70)

rf_class_report = classification_report(
    y_test_rf,
    rf_test_predictions,
    output_dict=True,
    zero_division=0
)

rf_per_class_records = []

for class_label in sorted(y_test_rf.unique()):

    class_key = str(class_label)

    rf_per_class_records.append({
        "model": "Random Forest",
        "class": class_label,
        "precision": rf_class_report[class_key]["precision"],
        "recall": rf_class_report[class_key]["recall"],
        "f1_score": rf_class_report[class_key]["f1-score"],
        "support": int(
            rf_class_report[class_key]["support"]
        )
    })

rf_per_class_performance = pd.DataFrame(
    rf_per_class_records
)

display(
    rf_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM FOREST FINAL VALIDATION")
logger.info("-" * 70)

if len(rf_train_predictions) != len(y_train_rf):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(rf_test_predictions) != len(y_test_rf):
    raise ValueError(
        "Test prediction count mismatch."
    )

if rf_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if rf_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Random Forest Performance Evaluation      : PASSED"
)

logger.info("=" * 70)
````

## 10.2 Extra Trees

Extra Trees (Extremely Randomized Trees) is an ensemble learning algorithm that builds multiple decision trees while introducing additional randomness during tree construction. Compared with Random Forest, Extra Trees selects split thresholds more randomly, which can improve generalization and reduce variance on some datasets.

For the Skin Disorder Prediction problem, Extra Trees is evaluated as an advanced nonlinear ensemble model. The model uses class-balanced learning because the six target classes are not equally represented.

The same protected training and test partitions from Section 6 are used. `Age` is treated as a numeric feature, while the remaining clinical and histopathological features are one-hot encoded. Missing numeric values are median-imputed, categorical values use training-fitted most-frequent imputation, and arbitrary ordinal encoding is not used.

The model is evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, training-test generalization gap, and per-class performance.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# =============================================================================
# 10.2 EXTRA TREES
# =============================================================================

logger.info("=" * 70)
logger.info("Extra Trees")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

if not isinstance(X_test, pd.DataFrame):
    raise TypeError("X_test must be a pandas DataFrame.")

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED COPIES
# -----------------------------------------------------------------------------

X_train_et = X_train.copy(deep=True)
X_test_et = X_test.copy(deep=True)

y_train_et = y_train.copy()
y_test_et = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE VALIDATION
# -----------------------------------------------------------------------------

if list(X_train_et.columns) != list(X_test_et.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

if "Age" not in X_train_et.columns:
    raise KeyError(
        "Expected feature 'Age' was not found."
    )

et_original_features = X_train_et.shape[1]
et_training_records = X_train_et.shape[0]
et_test_records = X_test_et.shape[0]
et_class_count = y_train_et.nunique()

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{et_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{et_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{et_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{et_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{sorted(y_train_et.unique().tolist())}"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_et["Age"] = pd.to_numeric(
    X_train_et["Age"],
    errors="coerce"
)

X_test_et["Age"] = pd.to_numeric(
    X_test_et["Age"],
    errors="coerce"
)

et_age_missing_train = int(
    X_train_et["Age"].isna().sum()
)

et_age_missing_test = int(
    X_test_et["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{et_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{et_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

et_numeric_features = ["Age"]

et_categorical_features = [
    column
    for column in X_train_et.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(et_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(et_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

et_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

et_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

et_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            et_numeric_pipeline,
            et_numeric_features
        ),
        (
            "categorical",
            et_categorical_pipeline,
            et_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# EXTRA TREES MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES MODEL CONFIGURATION")
logger.info("-" * 70)

et_model = ExtraTreesClassifier(
    n_estimators=300,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

logger.info(
    "Algorithm                               : "
    "Extra Trees"
)

logger.info(
    "Number of Estimators                    : 300"
)

logger.info(
    "Criterion                               : Gini"
)

logger.info(
    "Maximum Depth                           : None"
)

logger.info(
    "Minimum Samples Split                   : 2"
)

logger.info(
    "Minimum Samples Leaf                    : 1"
)

logger.info(
    "Maximum Features                        : sqrt"
)

logger.info(
    "Class Weight                            : balanced"
)

logger.info(
    "Random State                            : 42"
)

logger.info(
    "Number of CPU Cores                     : ALL AVAILABLE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

et_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            et_preprocessor
        ),
        (
            "model",
            et_model
        )
    ]
)

logger.info(
    "Extra Trees Pipeline                    : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES TRAINING")
logger.info("-" * 70)

et_training_start = time.perf_counter()

et_pipeline.fit(
    X_train_et,
    y_train_et
)

et_training_time = (
    time.perf_counter() - et_training_start
)

logger.info(
    "Extra Trees Training                    : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{et_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

et_transformed_feature_count = (
    et_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_et)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{et_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES PREDICTION")
logger.info("-" * 70)

et_train_predictions = et_pipeline.predict(
    X_train_et
)

et_test_predictions = et_pipeline.predict(
    X_test_et
)

logger.info(
    f"Training Predictions                    : "
    f"{len(et_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(et_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES PERFORMANCE")
logger.info("-" * 70)

et_train_accuracy = accuracy_score(
    y_train_et,
    et_train_predictions
)

et_test_accuracy = accuracy_score(
    y_test_et,
    et_test_predictions
)

et_balanced_accuracy = balanced_accuracy_score(
    y_test_et,
    et_test_predictions
)

et_macro_precision = precision_score(
    y_test_et,
    et_test_predictions,
    average="macro",
    zero_division=0
)

et_macro_recall = recall_score(
    y_test_et,
    et_test_predictions,
    average="macro",
    zero_division=0
)

et_macro_f1 = f1_score(
    y_test_et,
    et_test_predictions,
    average="macro",
    zero_division=0
)

et_weighted_f1 = f1_score(
    y_test_et,
    et_test_predictions,
    average="weighted",
    zero_division=0
)

et_generalization_gap = (
    et_train_accuracy - et_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{et_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{et_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{et_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{et_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{et_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{et_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{et_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{et_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES PER-CLASS PERFORMANCE")
logger.info("-" * 70)

et_class_report = classification_report(
    y_test_et,
    et_test_predictions,
    output_dict=True,
    zero_division=0
)

et_per_class_records = []

for class_label in sorted(y_test_et.unique()):

    class_key = str(class_label)

    et_per_class_records.append({
        "model": "Extra Trees",
        "class": class_label,
        "precision": et_class_report[class_key]["precision"],
        "recall": et_class_report[class_key]["recall"],
        "f1_score": et_class_report[class_key]["f1-score"],
        "support": int(
            et_class_report[class_key]["support"]
        )
    })

et_per_class_performance = pd.DataFrame(
    et_per_class_records
)

display(
    et_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRA TREES FINAL VALIDATION")
logger.info("-" * 70)

if len(et_train_predictions) != len(y_train_et):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(et_test_predictions) != len(y_test_et):
    raise ValueError(
        "Test prediction count mismatch."
    )

if et_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if et_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Extra Trees Performance Evaluation        : PASSED"
)

logger.info("=" * 70)
````

## 10.3 Gradient Boosting

Gradient Boosting is an ensemble learning technique that builds decision trees sequentially, where each new tree attempts to correct errors made by the previous trees.

For the Skin Disorder Prediction problem, Gradient Boosting is evaluated as a sequential nonlinear ensemble model across the six disorder classes. Unlike Random Forest and Extra Trees, which build trees independently, Gradient Boosting progressively improves the ensemble by minimizing prediction errors.

The model uses the same protected training and test partitions established in Section 6. `Age` is represented numerically, while the remaining clinical and histopathological features are one-hot encoded. Missing values are handled using training-fitted imputation, and arbitrary ordinal encoding is not used.

The baseline model is intentionally kept at a standard configuration. Hyperparameter optimization will be performed later in Section 11 rather than during baseline model development.

Performance is evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, training-test generalization gap, and per-class performance.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# =============================================================================
# 10.3 GRADIENT BOOSTING
# =============================================================================

logger.info("=" * 70)
logger.info("Gradient Boosting")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame.")

if not isinstance(X_test, pd.DataFrame):
    raise TypeError("X_test must be a pandas DataFrame.")

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED COPIES
# -----------------------------------------------------------------------------

X_train_gb = X_train.copy(deep=True)
X_test_gb = X_test.copy(deep=True)

y_train_gb = y_train.copy()
y_test_gb = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE VALIDATION
# -----------------------------------------------------------------------------

if list(X_train_gb.columns) != list(X_test_gb.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

if "Age" not in X_train_gb.columns:
    raise KeyError(
        "Expected feature 'Age' was not found."
    )

gb_original_features = X_train_gb.shape[1]
gb_training_records = X_train_gb.shape[0]
gb_test_records = X_test_gb.shape[0]
gb_class_count = y_train_gb.nunique()

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{gb_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{gb_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{gb_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{gb_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{sorted(y_train_gb.unique().tolist())}"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_gb["Age"] = pd.to_numeric(
    X_train_gb["Age"],
    errors="coerce"
)

X_test_gb["Age"] = pd.to_numeric(
    X_test_gb["Age"],
    errors="coerce"
)

gb_age_missing_train = int(
    X_train_gb["Age"].isna().sum()
)

gb_age_missing_test = int(
    X_test_gb["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{gb_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{gb_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

gb_numeric_features = ["Age"]

gb_categorical_features = [
    column
    for column in X_train_gb.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(gb_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(gb_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

gb_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

gb_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

gb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            gb_numeric_pipeline,
            gb_numeric_features
        ),
        (
            "categorical",
            gb_categorical_pipeline,
            gb_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# GRADIENT BOOSTING MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING MODEL CONFIGURATION")
logger.info("-" * 70)

gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    min_samples_split=2,
    min_samples_leaf=1,
    subsample=1.0,
    random_state=42
)

logger.info(
    "Algorithm                               : "
    "Gradient Boosting"
)

logger.info(
    "Number of Estimators                    : 100"
)

logger.info(
    "Learning Rate                           : 0.1"
)

logger.info(
    "Maximum Tree Depth                      : 3"
)

logger.info(
    "Minimum Samples Split                   : 2"
)

logger.info(
    "Minimum Samples Leaf                    : 1"
)

logger.info(
    "Subsample                               : 1.0"
)

logger.info(
    "Random State                             : 42"
)

logger.info(
    "Class Weight                            : "
    "NOT APPLICABLE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

gb_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            gb_preprocessor
        ),
        (
            "model",
            gb_model
        )
    ]
)

logger.info(
    "Gradient Boosting Pipeline               : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING TRAINING")
logger.info("-" * 70)

gb_training_start = time.perf_counter()

gb_pipeline.fit(
    X_train_gb,
    y_train_gb
)

gb_training_time = (
    time.perf_counter() - gb_training_start
)

logger.info(
    "Gradient Boosting Training               : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{gb_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

gb_transformed_feature_count = (
    gb_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_gb)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{gb_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING PREDICTION")
logger.info("-" * 70)

gb_train_predictions = gb_pipeline.predict(
    X_train_gb
)

gb_test_predictions = gb_pipeline.predict(
    X_test_gb
)

logger.info(
    f"Training Predictions                    : "
    f"{len(gb_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(gb_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING PERFORMANCE")
logger.info("-" * 70)

gb_train_accuracy = accuracy_score(
    y_train_gb,
    gb_train_predictions
)

gb_test_accuracy = accuracy_score(
    y_test_gb,
    gb_test_predictions
)

gb_balanced_accuracy = balanced_accuracy_score(
    y_test_gb,
    gb_test_predictions
)

gb_macro_precision = precision_score(
    y_test_gb,
    gb_test_predictions,
    average="macro",
    zero_division=0
)

gb_macro_recall = recall_score(
    y_test_gb,
    gb_test_predictions,
    average="macro",
    zero_division=0
)

gb_macro_f1 = f1_score(
    y_test_gb,
    gb_test_predictions,
    average="macro",
    zero_division=0
)

gb_weighted_f1 = f1_score(
    y_test_gb,
    gb_test_predictions,
    average="weighted",
    zero_division=0
)

gb_generalization_gap = (
    gb_train_accuracy - gb_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{gb_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{gb_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{gb_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{gb_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{gb_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{gb_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{gb_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{gb_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING PER-CLASS PERFORMANCE")
logger.info("-" * 70)

gb_class_report = classification_report(
    y_test_gb,
    gb_test_predictions,
    output_dict=True,
    zero_division=0
)

gb_per_class_records = []

for class_label in sorted(y_test_gb.unique()):

    class_key = str(class_label)

    gb_per_class_records.append({
        "model": "Gradient Boosting",
        "class": class_label,
        "precision": gb_class_report[class_key]["precision"],
        "recall": gb_class_report[class_key]["recall"],
        "f1_score": gb_class_report[class_key]["f1-score"],
        "support": int(
            gb_class_report[class_key]["support"]
        )
    })

gb_per_class_performance = pd.DataFrame(
    gb_per_class_records
)

display(
    gb_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GRADIENT BOOSTING FINAL VALIDATION")
logger.info("-" * 70)

if len(gb_train_predictions) != len(y_train_gb):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(gb_test_predictions) != len(y_test_gb):
    raise ValueError(
        "Test prediction count mismatch."
    )

if gb_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if gb_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Gradient Boosting Performance Evaluation  : PASSED"
)

logger.info("=" * 70)
````

## 10.4 XGBoost

XGBoost (Extreme Gradient Boosting) is an optimized gradient boosting framework that builds decision trees sequentially and uses regularization, efficient optimization, and boosting techniques to improve predictive performance.

For the Skin Disorder Prediction problem, XGBoost is evaluated as an advanced multiclass ensemble model. The model is configured for six-class classification and uses a standard baseline configuration rather than optimized hyperparameters. Hyperparameter optimization will be performed later in Section 11.

The model uses the protected training and test partitions established in Section 6. `Age` is treated as a numeric feature, while the remaining clinical and histopathological variables are one-hot encoded. Missing values are handled through training-fitted preprocessing, and arbitrary ordinal encoding is not used.

Because the target is multiclass, the model uses the `multi:softprob` objective to generate class probabilities for all six disorder classes. The baseline evaluation uses Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, generalization gap, and per-class performance.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from xgboost import XGBClassifier


# =============================================================================
# 10.4 XGBOOST
# =============================================================================

logger.info("=" * 70)
logger.info("XGBoost")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# EXISTING DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED COPIES
# -----------------------------------------------------------------------------

X_train_xgb = X_train.copy(deep=True)
X_test_xgb = X_test.copy(deep=True)

y_train_xgb = y_train.copy()
y_test_xgb = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE VALIDATION
# -----------------------------------------------------------------------------

if list(X_train_xgb.columns) != list(X_test_xgb.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

if "Age" not in X_train_xgb.columns:
    raise KeyError(
        "Expected feature 'Age' was not found."
    )

xgb_original_features = X_train_xgb.shape[1]
xgb_training_records = X_train_xgb.shape[0]
xgb_test_records = X_test_xgb.shape[0]
xgb_class_labels = sorted(
    y_train_xgb.unique().tolist()
)
xgb_class_count = len(xgb_class_labels)

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{xgb_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{xgb_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{xgb_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{xgb_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{xgb_class_labels}"
)


# -----------------------------------------------------------------------------
# TARGET LABEL VALIDATION
# -----------------------------------------------------------------------------

if xgb_class_labels != [1, 2, 3, 4, 5, 6]:
    raise ValueError(
        "Expected target classes [1, 2, 3, 4, 5, 6] "
        "were not detected."
    )

logger.info(
    "Target Class Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# XGBOOST LABEL MAPPING
# -----------------------------------------------------------------------------
# XGBoost multiclass objectives internally use class indices 0...n-1.
# Original project labels are 1...6, so a reversible mapping is used.

xgb_label_to_index = {
    label: index
    for index, label in enumerate(xgb_class_labels)
}

xgb_index_to_label = {
    index: label
    for label, index in xgb_label_to_index.items()
}

y_train_xgb_encoded = y_train_xgb.map(
    xgb_label_to_index
)

y_test_xgb_encoded = y_test_xgb.map(
    xgb_label_to_index
)

if y_train_xgb_encoded.isna().any():
    raise ValueError(
        "Training target encoding produced missing values."
    )

if y_test_xgb_encoded.isna().any():
    raise ValueError(
        "Test target encoding produced missing values."
    )

logger.info(
    "Target Label Mapping                     : "
    "1-6 → 0-5"
)

logger.info(
    "Target Label Encoding                    : PASSED"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_xgb["Age"] = pd.to_numeric(
    X_train_xgb["Age"],
    errors="coerce"
)

X_test_xgb["Age"] = pd.to_numeric(
    X_test_xgb["Age"],
    errors="coerce"
)

xgb_age_missing_train = int(
    X_train_xgb["Age"].isna().sum()
)

xgb_age_missing_test = int(
    X_test_xgb["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{xgb_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{xgb_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

xgb_numeric_features = ["Age"]

xgb_categorical_features = [
    column
    for column in X_train_xgb.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(xgb_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(xgb_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

xgb_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

xgb_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

xgb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            xgb_numeric_pipeline,
            xgb_numeric_features
        ),
        (
            "categorical",
            xgb_categorical_pipeline,
            xgb_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# XGBOOST MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST MODEL CONFIGURATION")
logger.info("-" * 70)

xgb_model = XGBClassifier(
    objective="multi:softprob",
    num_class=xgb_class_count,
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    min_child_weight=1,
    subsample=1.0,
    colsample_bytree=1.0,
    reg_alpha=0.0,
    reg_lambda=1.0,
    gamma=0.0,
    random_state=42,
    n_jobs=-1,
    eval_metric="mlogloss"
)

logger.info(
    "Algorithm                               : XGBoost"
)

logger.info(
    "Objective                               : multi:softprob"
)

logger.info(
    f"Number of Classes                       : "
    f"{xgb_class_count}"
)

logger.info(
    "Number of Estimators                    : 100"
)

logger.info(
    "Learning Rate                           : 0.1"
)

logger.info(
    "Maximum Tree Depth                      : 3"
)

logger.info(
    "Minimum Child Weight                    : 1"
)

logger.info(
    "Subsample                               : 1.0"
)

logger.info(
    "Column Subsample                        : 1.0"
)

logger.info(
    "L1 Regularization                       : 0.0"
)

logger.info(
    "L2 Regularization                       : 1.0"
)

logger.info(
    "Gamma                                   : 0.0"
)

logger.info(
    "Random State                            : 42"
)

logger.info(
    "Number of CPU Cores                     : ALL AVAILABLE"
)

logger.info(
    "Class Weight                            : "
    "NOT USED IN BASELINE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

xgb_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            xgb_preprocessor
        ),
        (
            "model",
            xgb_model
        )
    ]
)

logger.info(
    "XGBoost Pipeline                        : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST TRAINING")
logger.info("-" * 70)

xgb_training_start = time.perf_counter()

xgb_pipeline.fit(
    X_train_xgb,
    y_train_xgb_encoded
)

xgb_training_time = (
    time.perf_counter() - xgb_training_start
)

logger.info(
    "XGBoost Training                        : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{xgb_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

xgb_transformed_feature_count = (
    xgb_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_xgb)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{xgb_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST PREDICTION")
logger.info("-" * 70)

xgb_train_predictions_encoded = (
    xgb_pipeline.predict(X_train_xgb)
)

xgb_test_predictions_encoded = (
    xgb_pipeline.predict(X_test_xgb)
)

xgb_train_predictions_encoded = (
    np.asarray(
        xgb_train_predictions_encoded,
        dtype=int
    )
)

xgb_test_predictions_encoded = (
    np.asarray(
        xgb_test_predictions_encoded,
        dtype=int
    )
)

xgb_train_predictions = np.array([
    xgb_index_to_label[int(index)]
    for index in xgb_train_predictions_encoded
])

xgb_test_predictions = np.array([
    xgb_index_to_label[int(index)]
    for index in xgb_test_predictions_encoded
])

logger.info(
    f"Training Predictions                    : "
    f"{len(xgb_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(xgb_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PROBABILITY VALIDATION
# -----------------------------------------------------------------------------

xgb_test_probabilities = (
    xgb_pipeline.predict_proba(X_test_xgb)
)

if xgb_test_probabilities.shape != (
    len(X_test_xgb),
    xgb_class_count
):
    raise ValueError(
        "XGBoost probability matrix has an "
        "unexpected shape."
    )

if not np.allclose(
    xgb_test_probabilities.sum(axis=1),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "XGBoost probability rows do not sum to 1."
    )

logger.info(
    f"Probability Matrix Shape                 : "
    f"{xgb_test_probabilities.shape}"
)

logger.info(
    "Probability Validation                   : PASSED"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST PERFORMANCE")
logger.info("-" * 70)

xgb_train_accuracy = accuracy_score(
    y_train_xgb,
    xgb_train_predictions
)

xgb_test_accuracy = accuracy_score(
    y_test_xgb,
    xgb_test_predictions
)

xgb_balanced_accuracy = balanced_accuracy_score(
    y_test_xgb,
    xgb_test_predictions
)

xgb_macro_precision = precision_score(
    y_test_xgb,
    xgb_test_predictions,
    average="macro",
    zero_division=0
)

xgb_macro_recall = recall_score(
    y_test_xgb,
    xgb_test_predictions,
    average="macro",
    zero_division=0
)

xgb_macro_f1 = f1_score(
    y_test_xgb,
    xgb_test_predictions,
    average="macro",
    zero_division=0
)

xgb_weighted_f1 = f1_score(
    y_test_xgb,
    xgb_test_predictions,
    average="weighted",
    zero_division=0
)

xgb_generalization_gap = (
    xgb_train_accuracy - xgb_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{xgb_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{xgb_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{xgb_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{xgb_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{xgb_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{xgb_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{xgb_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{xgb_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST PER-CLASS PERFORMANCE")
logger.info("-" * 70)

xgb_class_report = classification_report(
    y_test_xgb,
    xgb_test_predictions,
    output_dict=True,
    zero_division=0
)

xgb_per_class_records = []

for class_label in sorted(y_test_xgb.unique()):

    class_key = str(class_label)

    xgb_per_class_records.append({
        "model": "XGBoost",
        "class": class_label,
        "precision": xgb_class_report[class_key]["precision"],
        "recall": xgb_class_report[class_key]["recall"],
        "f1_score": xgb_class_report[class_key]["f1-score"],
        "support": int(
            xgb_class_report[class_key]["support"]
        )
    })

xgb_per_class_performance = pd.DataFrame(
    xgb_per_class_records
)

display(
    xgb_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("XGBOOST FINAL VALIDATION")
logger.info("-" * 70)

if len(xgb_train_predictions) != len(y_train_xgb):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(xgb_test_predictions) != len(y_test_xgb):
    raise ValueError(
        "Test prediction count mismatch."
    )

if xgb_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if xgb_test_probabilities.shape[1] != xgb_class_count:
    raise ValueError(
        "Probability class count mismatch."
    )

if xgb_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Probability Validation                    : PASSED"
)

logger.info(
    "XGBoost Performance Evaluation            : PASSED"
)

logger.info("=" * 70)
````

## 10.5 CatBoost

CatBoost is a gradient boosting algorithm designed to handle categorical variables effectively while providing strong performance and robust generalization.

For the Skin Disorder Prediction problem, CatBoost is evaluated as an advanced multiclass classification model. The model is configured with six target classes and a reproducible baseline configuration.

For consistency with the other advanced models, the protected training and test datasets are used with the same preprocessing representation: `Age` is treated as numeric, while the remaining clinical and histopathological features are one-hot encoded. Missing values are handled using training-fitted preprocessing, and arbitrary ordinal encoding is not used.

The model is evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, generalization gap, and per-class performance. Hyperparameter optimization is intentionally deferred to Section 11.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from catboost import CatBoostClassifier


# =============================================================================
# 10.5 CATBOOST
# =============================================================================

logger.info("=" * 70)
logger.info("CatBoost")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

if list(X_train.columns) != list(X_test.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED COPIES
# -----------------------------------------------------------------------------

X_train_cb = X_train.copy(deep=True)
X_test_cb = X_test.copy(deep=True)

y_train_cb = y_train.copy()
y_test_cb = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# DATASET CONFIGURATION
# -----------------------------------------------------------------------------

cb_original_features = X_train_cb.shape[1]
cb_training_records = X_train_cb.shape[0]
cb_test_records = X_test_cb.shape[0]

cb_class_labels = sorted(
    y_train_cb.unique().tolist()
)

cb_class_count = len(cb_class_labels)

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{cb_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{cb_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{cb_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{cb_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{cb_class_labels}"
)


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

if cb_class_labels != [1, 2, 3, 4, 5, 6]:
    raise ValueError(
        "Expected target classes [1, 2, 3, 4, 5, 6] "
        "were not detected."
    )

logger.info(
    "Target Class Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_cb["Age"] = pd.to_numeric(
    X_train_cb["Age"],
    errors="coerce"
)

X_test_cb["Age"] = pd.to_numeric(
    X_test_cb["Age"],
    errors="coerce"
)

cb_age_missing_train = int(
    X_train_cb["Age"].isna().sum()
)

cb_age_missing_test = int(
    X_test_cb["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{cb_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{cb_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

cb_numeric_features = ["Age"]

cb_categorical_features = [
    column
    for column in X_train_cb.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(cb_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(cb_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

cb_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

cb_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

cb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            cb_numeric_pipeline,
            cb_numeric_features
        ),
        (
            "categorical",
            cb_categorical_pipeline,
            cb_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# CATBOOST MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST MODEL CONFIGURATION")
logger.info("-" * 70)

cb_model = CatBoostClassifier(
    loss_function="MultiClass",
    iterations=300,
    learning_rate=0.05,
    depth=6,
    l2_leaf_reg=3.0,
    random_seed=42,
    verbose=False,
    thread_count=-1,
    allow_writing_files=False
)

logger.info(
    "Algorithm                               : CatBoost"
)

logger.info(
    "Objective                               : MultiClass"
)

logger.info(
    f"Number of Classes                       : "
    f"{cb_class_count}"
)

logger.info(
    "Iterations                              : 300"
)

logger.info(
    "Learning Rate                           : 0.05"
)

logger.info(
    "Tree Depth                              : 6"
)

logger.info(
    "L2 Leaf Regularization                  : 3.0"
)

logger.info(
    "Random Seed                             : 42"
)

logger.info(
    "CPU Threads                             : ALL AVAILABLE"
)

logger.info(
    "Early Stopping                          : NOT USED IN BASELINE"
)

logger.info(
    "Class Weight                            : NOT USED IN BASELINE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

cb_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            cb_preprocessor
        ),
        (
            "model",
            cb_model
        )
    ]
)

logger.info(
    "CatBoost Pipeline                       : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST TRAINING")
logger.info("-" * 70)

cb_training_start = time.perf_counter()

cb_pipeline.fit(
    X_train_cb,
    y_train_cb
)

cb_training_time = (
    time.perf_counter() - cb_training_start
)

logger.info(
    "CatBoost Training                       : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{cb_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

cb_transformed_feature_count = (
    cb_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_cb)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{cb_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST PREDICTION")
logger.info("-" * 70)

cb_train_predictions = (
    cb_pipeline.predict(X_train_cb)
)

cb_test_predictions = (
    cb_pipeline.predict(X_test_cb)
)

cb_train_predictions = np.asarray(
    cb_train_predictions
).reshape(-1)

cb_test_predictions = np.asarray(
    cb_test_predictions
).reshape(-1)

logger.info(
    f"Training Predictions                    : "
    f"{len(cb_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(cb_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PROBABILITY VALIDATION
# -----------------------------------------------------------------------------

cb_test_probabilities = (
    cb_pipeline.predict_proba(X_test_cb)
)

if cb_test_probabilities.shape != (
    len(X_test_cb),
    cb_class_count
):
    raise ValueError(
        "CatBoost probability matrix has an "
        "unexpected shape."
    )

if not np.allclose(
    cb_test_probabilities.sum(axis=1),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "CatBoost probability rows do not sum to 1."
    )

logger.info(
    f"Probability Matrix Shape                 : "
    f"{cb_test_probabilities.shape}"
)

logger.info(
    "Probability Validation                   : PASSED"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST PERFORMANCE")
logger.info("-" * 70)

cb_train_accuracy = accuracy_score(
    y_train_cb,
    cb_train_predictions
)

cb_test_accuracy = accuracy_score(
    y_test_cb,
    cb_test_predictions
)

cb_balanced_accuracy = balanced_accuracy_score(
    y_test_cb,
    cb_test_predictions
)

cb_macro_precision = precision_score(
    y_test_cb,
    cb_test_predictions,
    average="macro",
    zero_division=0
)

cb_macro_recall = recall_score(
    y_test_cb,
    cb_test_predictions,
    average="macro",
    zero_division=0
)

cb_macro_f1 = f1_score(
    y_test_cb,
    cb_test_predictions,
    average="macro",
    zero_division=0
)

cb_weighted_f1 = f1_score(
    y_test_cb,
    cb_test_predictions,
    average="weighted",
    zero_division=0
)

cb_generalization_gap = (
    cb_train_accuracy - cb_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{cb_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{cb_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{cb_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{cb_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{cb_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{cb_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{cb_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{cb_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST PER-CLASS PERFORMANCE")
logger.info("-" * 70)

cb_class_report = classification_report(
    y_test_cb,
    cb_test_predictions,
    output_dict=True,
    zero_division=0
)

cb_per_class_records = []

for class_label in sorted(y_test_cb.unique()):

    class_key = str(class_label)

    cb_per_class_records.append({
        "model": "CatBoost",
        "class": class_label,
        "precision": cb_class_report[class_key]["precision"],
        "recall": cb_class_report[class_key]["recall"],
        "f1_score": cb_class_report[class_key]["f1-score"],
        "support": int(
            cb_class_report[class_key]["support"]
        )
    })

cb_per_class_performance = pd.DataFrame(
    cb_per_class_records
)

display(
    cb_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CATBOOST FINAL VALIDATION")
logger.info("-" * 70)

if len(cb_train_predictions) != len(y_train_cb):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(cb_test_predictions) != len(y_test_cb):
    raise ValueError(
        "Test prediction count mismatch."
    )

if cb_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if cb_test_probabilities.shape[1] != cb_class_count:
    raise ValueError(
        "Probability class count mismatch."
    )

if cb_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Probability Validation                    : PASSED"
)

logger.info(
    "CatBoost Performance Evaluation           : PASSED"
)

logger.info("=" * 70)
````

## 10.6 LightGBM

LightGBM (Light Gradient Boosting Machine) is an efficient gradient boosting framework designed for high-performance tree-based learning. It uses histogram-based learning and leaf-wise tree growth to achieve strong predictive performance with relatively low computational cost.

For the Skin Disorder Prediction problem, LightGBM is evaluated as an additional advanced multiclass model. The model is configured for six-class classification using a reproducible baseline configuration.

To maintain comparability with the other advanced models, the protected training and test datasets are used with the same feature representation: `Age` is treated as numeric, while the remaining clinical and histopathological features are one-hot encoded. Missing values are handled through training-fitted preprocessing, and arbitrary ordinal encoding is not used.

The model is evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, Weighted F1, generalization gap, and per-class performance. Hyperparameter optimization is intentionally deferred to Section 11.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import sys
import subprocess

# =============================================================================
# INSTALL LIGHTGBM
# =============================================================================

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "lightgbm"
])

print("LightGBM installation completed successfully.")
````

````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from lightgbm import LGBMClassifier


# =============================================================================
# 10.6 LIGHTGBM
# =============================================================================

logger.info("=" * 70)
logger.info("LightGBM")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Training/Test Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "X_test" not in globals():
    raise NameError(
        "Required object 'X_test' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if "y_test" not in globals():
    raise NameError(
        "Required object 'y_test' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if len(X_test) != len(y_test):
    raise ValueError(
        "X_test and y_test sample counts do not match."
    )

if list(X_train.columns) != list(X_test.columns):
    raise ValueError(
        "Training and test feature columns are not aligned."
    )

logger.info("Training/Test Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED COPIES
# -----------------------------------------------------------------------------

X_train_lgbm = X_train.copy(deep=True)
X_test_lgbm = X_test.copy(deep=True)

y_train_lgbm = y_train.copy()
y_test_lgbm = y_test.copy()

logger.info(
    "Protected Training/Test Copies : CREATED"
)


# -----------------------------------------------------------------------------
# DATASET CONFIGURATION
# -----------------------------------------------------------------------------

lgbm_original_features = X_train_lgbm.shape[1]
lgbm_training_records = X_train_lgbm.shape[0]
lgbm_test_records = X_test_lgbm.shape[0]

lgbm_class_labels = sorted(
    y_train_lgbm.unique().tolist()
)

lgbm_class_count = len(lgbm_class_labels)

logger.info("-" * 70)
logger.info("DATASET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Training Records                         : "
    f"{lgbm_training_records}"
)

logger.info(
    f"Test Records                             : "
    f"{lgbm_test_records}"
)

logger.info(
    f"Original Predictor Features              : "
    f"{lgbm_original_features}"
)

logger.info(
    f"Target Classes                           : "
    f"{lgbm_class_count}"
)

logger.info(
    f"Target Classes Detected                  : "
    f"{lgbm_class_labels}"
)


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

if lgbm_class_labels != [1, 2, 3, 4, 5, 6]:
    raise ValueError(
        "Expected target classes [1, 2, 3, 4, 5, 6] "
        "were not detected."
    )

logger.info(
    "Target Class Validation                  : PASSED"
)


# -----------------------------------------------------------------------------
# TARGET LABEL MAPPING
# -----------------------------------------------------------------------------
# LightGBM multiclass classification uses class indices 0...n-1.
# Original project labels 1...6 are therefore mapped to 0...5.

lgbm_label_to_index = {
    label: index
    for index, label in enumerate(lgbm_class_labels)
}

lgbm_index_to_label = {
    index: label
    for label, index in lgbm_label_to_index.items()
}

y_train_lgbm_encoded = y_train_lgbm.map(
    lgbm_label_to_index
)

y_test_lgbm_encoded = y_test_lgbm.map(
    lgbm_label_to_index
)

if y_train_lgbm_encoded.isna().any():
    raise ValueError(
        "Training target encoding produced missing values."
    )

if y_test_lgbm_encoded.isna().any():
    raise ValueError(
        "Test target encoding produced missing values."
    )

logger.info(
    "Target Label Mapping                     : "
    "1-6 → 0-5"
)

logger.info(
    "Target Label Encoding                    : PASSED"
)


# -----------------------------------------------------------------------------
# AGE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_lgbm["Age"] = pd.to_numeric(
    X_train_lgbm["Age"],
    errors="coerce"
)

X_test_lgbm["Age"] = pd.to_numeric(
    X_test_lgbm["Age"],
    errors="coerce"
)

lgbm_age_missing_train = int(
    X_train_lgbm["Age"].isna().sum()
)

lgbm_age_missing_test = int(
    X_test_lgbm["Age"].isna().sum()
)

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    "Age Representation                       : NUMERIC"
)

logger.info(
    f"Training Missing Age Values              : "
    f"{lgbm_age_missing_train}"
)

logger.info(
    f"Test Missing Age Values                  : "
    f"{lgbm_age_missing_test}"
)


# -----------------------------------------------------------------------------
# FEATURE GROUPING
# -----------------------------------------------------------------------------

lgbm_numeric_features = ["Age"]

lgbm_categorical_features = [
    column
    for column in X_train_lgbm.columns
    if column != "Age"
]

logger.info(
    f"Numeric Features                         : "
    f"{len(lgbm_numeric_features)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(lgbm_categorical_features)}"
)


# -----------------------------------------------------------------------------
# PREPROCESSING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPROCESSING CONFIGURATION")
logger.info("-" * 70)

lgbm_numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

lgbm_categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

lgbm_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            lgbm_numeric_pipeline,
            lgbm_numeric_features
        ),
        (
            "categorical",
            lgbm_categorical_pipeline,
            lgbm_categorical_features
        )
    ],
    remainder="drop"
)


# -----------------------------------------------------------------------------
# LIGHTGBM MODEL CONFIGURATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM MODEL CONFIGURATION")
logger.info("-" * 70)

lgbm_model = LGBMClassifier(
    objective="multiclass",
    num_class=lgbm_class_count,
    n_estimators=100,
    learning_rate=0.1,
    max_depth=-1,
    num_leaves=31,
    min_child_samples=20,
    subsample=1.0,
    colsample_bytree=1.0,
    reg_alpha=0.0,
    reg_lambda=0.0,
    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

logger.info(
    "Algorithm                               : LightGBM"
)

logger.info(
    "Objective                               : multiclass"
)

logger.info(
    f"Number of Classes                       : "
    f"{lgbm_class_count}"
)

logger.info(
    "Number of Estimators                    : 100"
)

logger.info(
    "Learning Rate                           : 0.1"
)

logger.info(
    "Maximum Tree Depth                      : -1"
)

logger.info(
    "Number of Leaves                        : 31"
)

logger.info(
    "Minimum Child Samples                   : 20"
)

logger.info(
    "Row Subsample                           : 1.0"
)

logger.info(
    "Column Subsample                        : 1.0"
)

logger.info(
    "L1 Regularization                       : 0.0"
)

logger.info(
    "L2 Regularization                       : 0.0"
)

logger.info(
    "Random State                            : 42"
)

logger.info(
    "CPU Threads                             : ALL AVAILABLE"
)

logger.info(
    "Class Weight                            : NOT USED IN BASELINE"
)

logger.info(
    "Ordinal Encoding                        : NOT USED"
)


# -----------------------------------------------------------------------------
# COMPLETE PIPELINE
# -----------------------------------------------------------------------------

lgbm_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            lgbm_preprocessor
        ),
        (
            "model",
            lgbm_model
        )
    ]
)

logger.info(
    "LightGBM Pipeline                       : CREATED"
)


# -----------------------------------------------------------------------------
# MODEL TRAINING
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM TRAINING")
logger.info("-" * 70)

lgbm_training_start = time.perf_counter()

lgbm_pipeline.fit(
    X_train_lgbm,
    y_train_lgbm_encoded
)

lgbm_training_time = (
    time.perf_counter() - lgbm_training_start
)

logger.info(
    "LightGBM Training                       : COMPLETED"
)

logger.info(
    f"Training Time                           : "
    f"{lgbm_training_time:.4f} seconds"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE DIMENSION
# -----------------------------------------------------------------------------

lgbm_transformed_feature_count = (
    lgbm_pipeline
    .named_steps["preprocessor"]
    .transform(X_train_lgbm)
    .shape[1]
)

logger.info(
    f"Transformed Feature Count               : "
    f"{lgbm_transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# PREDICTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM PREDICTION")
logger.info("-" * 70)

lgbm_train_predictions_encoded = (
    lgbm_pipeline.predict(X_train_lgbm)
)

lgbm_test_predictions_encoded = (
    lgbm_pipeline.predict(X_test_lgbm)
)

lgbm_train_predictions_encoded = np.asarray(
    lgbm_train_predictions_encoded,
    dtype=int
).reshape(-1)

lgbm_test_predictions_encoded = np.asarray(
    lgbm_test_predictions_encoded,
    dtype=int
).reshape(-1)

lgbm_train_predictions = np.array([
    lgbm_index_to_label[int(index)]
    for index in lgbm_train_predictions_encoded
])

lgbm_test_predictions = np.array([
    lgbm_index_to_label[int(index)]
    for index in lgbm_test_predictions_encoded
])

logger.info(
    f"Training Predictions                    : "
    f"{len(lgbm_train_predictions)}"
)

logger.info(
    f"Test Predictions                        : "
    f"{len(lgbm_test_predictions)}"
)


# -----------------------------------------------------------------------------
# PROBABILITY VALIDATION
# -----------------------------------------------------------------------------

lgbm_test_probabilities = (
    lgbm_pipeline.predict_proba(X_test_lgbm)
)

if lgbm_test_probabilities.shape != (
    len(X_test_lgbm),
    lgbm_class_count
):
    raise ValueError(
        "LightGBM probability matrix has an "
        "unexpected shape."
    )

if not np.allclose(
    lgbm_test_probabilities.sum(axis=1),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "LightGBM probability rows do not sum to 1."
    )

logger.info(
    f"Probability Matrix Shape                 : "
    f"{lgbm_test_probabilities.shape}"
)

logger.info(
    "Probability Validation                   : PASSED"
)


# -----------------------------------------------------------------------------
# PERFORMANCE METRICS
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM PERFORMANCE")
logger.info("-" * 70)

lgbm_train_accuracy = accuracy_score(
    y_train_lgbm,
    lgbm_train_predictions
)

lgbm_test_accuracy = accuracy_score(
    y_test_lgbm,
    lgbm_test_predictions
)

lgbm_balanced_accuracy = balanced_accuracy_score(
    y_test_lgbm,
    lgbm_test_predictions
)

lgbm_macro_precision = precision_score(
    y_test_lgbm,
    lgbm_test_predictions,
    average="macro",
    zero_division=0
)

lgbm_macro_recall = recall_score(
    y_test_lgbm,
    lgbm_test_predictions,
    average="macro",
    zero_division=0
)

lgbm_macro_f1 = f1_score(
    y_test_lgbm,
    lgbm_test_predictions,
    average="macro",
    zero_division=0
)

lgbm_weighted_f1 = f1_score(
    y_test_lgbm,
    lgbm_test_predictions,
    average="weighted",
    zero_division=0
)

lgbm_generalization_gap = (
    lgbm_train_accuracy - lgbm_test_accuracy
)

logger.info(
    f"Training Accuracy                       : "
    f"{lgbm_train_accuracy:.4f}"
)

logger.info(
    f"Test Accuracy                           : "
    f"{lgbm_test_accuracy:.4f}"
)

logger.info(
    f"Balanced Accuracy                       : "
    f"{lgbm_balanced_accuracy:.4f}"
)

logger.info(
    f"Macro Precision                         : "
    f"{lgbm_macro_precision:.4f}"
)

logger.info(
    f"Macro Recall                            : "
    f"{lgbm_macro_recall:.4f}"
)

logger.info(
    f"Macro F1                                : "
    f"{lgbm_macro_f1:.4f}"
)

logger.info(
    f"Weighted F1                             : "
    f"{lgbm_weighted_f1:.4f}"
)

logger.info(
    f"Generalization Gap                      : "
    f"{lgbm_generalization_gap:.4f}"
)


# -----------------------------------------------------------------------------
# PER-CLASS PERFORMANCE
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM PER-CLASS PERFORMANCE")
logger.info("-" * 70)

lgbm_class_report = classification_report(
    y_test_lgbm,
    lgbm_test_predictions,
    output_dict=True,
    zero_division=0
)

lgbm_per_class_records = []

for class_label in sorted(y_test_lgbm.unique()):

    class_key = str(class_label)

    lgbm_per_class_records.append({
        "model": "LightGBM",
        "class": class_label,
        "precision": lgbm_class_report[class_key]["precision"],
        "recall": lgbm_class_report[class_key]["recall"],
        "f1_score": lgbm_class_report[class_key]["f1-score"],
        "support": int(
            lgbm_class_report[class_key]["support"]
        )
    })

lgbm_per_class_performance = pd.DataFrame(
    lgbm_per_class_records
)

display(
    lgbm_per_class_performance.round(4)
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LIGHTGBM FINAL VALIDATION")
logger.info("-" * 70)

if len(lgbm_train_predictions) != len(y_train_lgbm):
    raise ValueError(
        "Training prediction count mismatch."
    )

if len(lgbm_test_predictions) != len(y_test_lgbm):
    raise ValueError(
        "Test prediction count mismatch."
    )

if lgbm_transformed_feature_count <= 0:
    raise ValueError(
        "No transformed features were generated."
    )

if lgbm_test_probabilities.shape[1] != lgbm_class_count:
    raise ValueError(
        "Probability class count mismatch."
    )

if lgbm_class_count < 2:
    raise ValueError(
        "Multiclass target validation failed."
    )

logger.info(
    "Training Prediction Validation            : PASSED"
)

logger.info(
    "Test Prediction Validation                : PASSED"
)

logger.info(
    "Transformed Feature Validation            : PASSED"
)

logger.info(
    "Multiclass Target Validation              : PASSED"
)

logger.info(
    "Probability Validation                    : PASSED"
)

logger.info(
    "LightGBM Performance Evaluation           : PASSED"
)

logger.info("=" * 70)
````

## 10.7 Advanced Model Cross-Validation

To assess the stability and generalization capability of the advanced machine learning models, 5-fold Stratified Cross-Validation is performed using only the training dataset.

The six advanced models evaluated are Random Forest, Extra Trees, Gradient Boosting, XGBoost, CatBoost, and LightGBM.

Stratified Cross-Validation preserves the multiclass target distribution across the five folds. Each preprocessing pipeline is fitted independently within each fold to prevent data leakage.

For consistency with the individual model evaluations, the same baseline model configurations are used. XGBoost receives zero-based target labels internally because its classifier implementation requires class indices from 0 to `n_classes - 1`; this transformation is applied only to the cross-validation copy and does not modify the original target data.

The models are evaluated using Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, Macro F1, and Weighted F1. Macro F1 and Balanced Accuracy are emphasized because the dataset contains six classes with an imbalanced class distribution.

The independent test dataset remains completely untouched during cross-validation. Cross-validation results will be combined with independent test-set results in Section 10.8 to determine the strongest advanced model before hyperparameter optimization.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import time
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import StratifiedKFold, cross_validate

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)

from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier


# =============================================================================
# 10.7 ADVANCED MODEL CROSS-VALIDATION
# =============================================================================

logger.info("=" * 70)
logger.info("10.7 ADVANCED MODEL CROSS-VALIDATION")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# DATA VALIDATION
# -----------------------------------------------------------------------------

logger.info("Cross-Validation Data Validation : STARTED")

if "X_train" not in globals():
    raise NameError(
        "Required object 'X_train' is not available."
    )

if "y_train" not in globals():
    raise NameError(
        "Required object 'y_train' is not available."
    )

if not isinstance(X_train, pd.DataFrame):
    raise TypeError(
        "X_train must be a pandas DataFrame."
    )

if len(X_train) != len(y_train):
    raise ValueError(
        "X_train and y_train sample counts do not match."
    )

if "X_test" in globals():
    logger.info(
        "Independent Test Set                    : "
        "NOT USED IN CROSS-VALIDATION"
    )

logger.info("Cross-Validation Data Validation : PASSED")


# -----------------------------------------------------------------------------
# PROTECTED TRAINING COPY
# -----------------------------------------------------------------------------

X_train_adv_cv = X_train.copy(deep=True)
y_train_adv_cv = y_train.copy()

logger.info(
    "Protected Training Copy                    : CREATED"
)


# -----------------------------------------------------------------------------
# FEATURE REPRESENTATION
# -----------------------------------------------------------------------------

X_train_adv_cv["Age"] = pd.to_numeric(
    X_train_adv_cv["Age"],
    errors="coerce"
)

adv_cv_numeric_features = ["Age"]

adv_cv_categorical_features = [
    column
    for column in X_train_adv_cv.columns
    if column != "Age"
]

logger.info("-" * 70)
logger.info("FEATURE REPRESENTATION")
logger.info("-" * 70)

logger.info(
    f"Original Features                         : "
    f"{X_train_adv_cv.shape[1]}"
)

logger.info(
    "Age Representation                        : NUMERIC"
)

logger.info(
    f"Categorical Features                      : "
    f"{len(adv_cv_categorical_features)}"
)

logger.info(
    "Categorical Encoding                      : ONE-HOT"
)

logger.info(
    "Ordinal Encoding                          : NOT USED"
)


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

adv_cv_class_labels = sorted(
    y_train_adv_cv.unique().tolist()
)

adv_cv_class_count = len(
    adv_cv_class_labels
)

if adv_cv_class_labels != [1, 2, 3, 4, 5, 6]:
    raise ValueError(
        "Expected target classes [1, 2, 3, 4, 5, 6] "
        "were not detected."
    )

logger.info(
    f"Target Classes                            : "
    f"{adv_cv_class_labels}"
)

logger.info(
    f"Number of Target Classes                  : "
    f"{adv_cv_class_count}"
)

logger.info(
    "Target Class Validation                   : PASSED"
)


# -----------------------------------------------------------------------------
# XGBOOST TARGET ENCODING
# -----------------------------------------------------------------------------
# XGBoost requires zero-based multiclass labels.
# This transformation is applied ONLY to the CV copy.
# Original y_train remains unchanged.

adv_cv_label_to_index = {
    label: index
    for index, label in enumerate(
        adv_cv_class_labels
    )
}

y_train_xgb_cv = y_train_adv_cv.map(
    adv_cv_label_to_index
)

if y_train_xgb_cv.isna().any():
    raise ValueError(
        "XGBoost target encoding produced missing values."
    )

y_train_xgb_cv = y_train_xgb_cv.astype(int)

logger.info(
    "XGBoost Target Encoding                   : "
    "1-6 → 0-5"
)

logger.info(
    "XGBoost Target Encoding Validation        : PASSED"
)


# -----------------------------------------------------------------------------
# PREPROCESSING FACTORY
# -----------------------------------------------------------------------------

def create_advanced_cv_preprocessor():

    numeric_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            )
        ]
    )

    categorical_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ]
    )

    return ColumnTransformer(
        transformers=[
            (
                "numeric",
                numeric_pipeline,
                adv_cv_numeric_features
            ),
            (
                "categorical",
                categorical_pipeline,
                adv_cv_categorical_features
            )
        ],
        remainder="drop"
    )


# -----------------------------------------------------------------------------
# MODEL FACTORY
# -----------------------------------------------------------------------------

def create_advanced_models_cv():

    return {

        "Random Forest": RandomForestClassifier(
            n_estimators=300,
            criterion="gini",
            max_depth=None,
            min_samples_split=2,
            min_samples_leaf=1,
            max_features="sqrt",
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

        "Extra Trees": ExtraTreesClassifier(
            n_estimators=300,
            criterion="gini",
            max_depth=None,
            min_samples_split=2,
            min_samples_leaf=1,
            max_features="sqrt",
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

        "Gradient Boosting": GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=0.1,
            max_depth=3,
            min_samples_split=2,
            min_samples_leaf=1,
            subsample=1.0,
            random_state=42
        ),

        "XGBoost": XGBClassifier(
            objective="multi:softprob",
            num_class=6,
            n_estimators=100,
            learning_rate=0.1,
            max_depth=3,
            min_child_weight=1,
            subsample=1.0,
            colsample_bytree=1.0,
            reg_alpha=0.0,
            reg_lambda=1.0,
            gamma=0.0,
            random_state=42,
            n_jobs=-1,
            eval_metric="mlogloss"
        ),

        "CatBoost": CatBoostClassifier(
            loss_function="MultiClass",
            iterations=300,
            learning_rate=0.05,
            depth=6,
            l2_leaf_reg=3,
            random_seed=42,
            verbose=False,
            thread_count=-1,
            allow_writing_files=False
        ),

        "LightGBM": LGBMClassifier(
            objective="multiclass",
            num_class=6,
            n_estimators=100,
            learning_rate=0.1,
            max_depth=-1,
            num_leaves=31,
            min_child_samples=20,
            subsample=1.0,
            colsample_bytree=1.0,
            reg_alpha=0.0,
            reg_lambda=0.0,
            random_state=42,
            n_jobs=-1,
            verbosity=-1
        )
    }


# -----------------------------------------------------------------------------
# CROSS-VALIDATION CONFIGURATION
# -----------------------------------------------------------------------------

adv_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

adv_cv_scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "macro_precision": "precision_macro",
    "macro_recall": "recall_macro",
    "macro_f1": "f1_macro",
    "weighted_f1": "f1_weighted"
}

logger.info("-" * 70)
logger.info("CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Cross-Validation Strategy                  : "
    "StratifiedKFold"
)

logger.info(
    "Number of Folds                            : 5"
)

logger.info(
    "Shuffle                                    : True"
)

logger.info(
    "Random State                               : 42"
)

logger.info(
    "Evaluation Metrics                         : "
    "Accuracy, Balanced Accuracy, Macro Precision, "
    "Macro Recall, Macro F1, Weighted F1"
)

logger.info(
    "Test Set Usage                             : NOT USED"
)


# -----------------------------------------------------------------------------
# CROSS-VALIDATION EXECUTION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ADVANCED MODEL CROSS-VALIDATION")
logger.info("-" * 70)

advanced_models_cv = create_advanced_models_cv()

advanced_cv_results = []


for model_name, model in advanced_models_cv.items():

    logger.info(
        f"{model_name} Cross-Validation              : STARTED"
    )

    model_preprocessor = (
        create_advanced_cv_preprocessor()
    )

    model_pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                model_preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    cv_start_time = time.perf_counter()


    # -------------------------------------------------------------------------
    # SELECT TARGET
    # -------------------------------------------------------------------------
    # XGBoost requires zero-based target labels.
    # All other models use the original target labels.

    if model_name == "XGBoost":

        cv_target = y_train_xgb_cv

    else:

        cv_target = y_train_adv_cv


    # -------------------------------------------------------------------------
    # EXECUTE 5-FOLD STRATIFIED CROSS-VALIDATION
    # -------------------------------------------------------------------------

    cv_scores = cross_validate(
        model_pipeline,
        X_train_adv_cv,
        cv_target,
        cv=adv_cv,
        scoring=adv_cv_scoring,
        n_jobs=1,
        return_train_score=False,
        error_score="raise"
    )

    cv_elapsed_time = (
        time.perf_counter() - cv_start_time
    )


    # -------------------------------------------------------------------------
    # STORE RESULTS
    # -------------------------------------------------------------------------

    advanced_cv_results.append({

        "Model":
            model_name,

        "CV Accuracy Mean":
            np.mean(
                cv_scores["test_accuracy"]
            ),

        "CV Accuracy Std":
            np.std(
                cv_scores["test_accuracy"],
                ddof=1
            ),

        "CV Balanced Accuracy Mean":
            np.mean(
                cv_scores["test_balanced_accuracy"]
            ),

        "CV Balanced Accuracy Std":
            np.std(
                cv_scores["test_balanced_accuracy"],
                ddof=1
            ),

        "CV Macro Precision Mean":
            np.mean(
                cv_scores["test_macro_precision"]
            ),

        "CV Macro Precision Std":
            np.std(
                cv_scores["test_macro_precision"],
                ddof=1
            ),

        "CV Macro Recall Mean":
            np.mean(
                cv_scores["test_macro_recall"]
            ),

        "CV Macro Recall Std":
            np.std(
                cv_scores["test_macro_recall"],
                ddof=1
            ),

        "CV Macro F1 Mean":
            np.mean(
                cv_scores["test_macro_f1"]
            ),

        "CV Macro F1 Std":
            np.std(
                cv_scores["test_macro_f1"],
                ddof=1
            ),

        "CV Weighted F1 Mean":
            np.mean(
                cv_scores["test_weighted_f1"]
            ),

        "CV Weighted F1 Std":
            np.std(
                cv_scores["test_weighted_f1"],
                ddof=1
            ),

        "CV Training Time (sec)":
            cv_elapsed_time
    })


    # -------------------------------------------------------------------------
    # MODEL CV LOGGING
    # -------------------------------------------------------------------------

    logger.info(
        f"{model_name} Cross-Validation              : "
        f"COMPLETED"
    )

    logger.info(
        f"Mean Accuracy                             : "
        f"{np.mean(cv_scores['test_accuracy']):.4f}"
    )

    logger.info(
        f"Mean Balanced Accuracy                    : "
        f"{np.mean(cv_scores['test_balanced_accuracy']):.4f}"
    )

    logger.info(
        f"Mean Macro F1                             : "
        f"{np.mean(cv_scores['test_macro_f1']):.4f}"
    )

    logger.info(
        f"Macro F1 Standard Deviation               : "
        f"{np.std(cv_scores['test_macro_f1'], ddof=1):.4f}"
    )


# -----------------------------------------------------------------------------
# CROSS-VALIDATION RESULTS TABLE
# -----------------------------------------------------------------------------

advanced_cv_results_df = pd.DataFrame(
    advanced_cv_results
)

advanced_cv_results_df = (
    advanced_cv_results_df
    .sort_values(
        by=[
            "CV Macro F1 Mean",
            "CV Balanced Accuracy Mean"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

advanced_cv_results_df.insert(
    0,
    "CV Rank",
    np.arange(
        1,
        len(advanced_cv_results_df) + 1
    )
)

logger.info("-" * 70)
logger.info("ADVANCED MODEL CROSS-VALIDATION RESULTS")
logger.info("-" * 70)

display(
    advanced_cv_results_df.round(4)
)


# -----------------------------------------------------------------------------
# BEST CROSS-VALIDATED MODEL
# -----------------------------------------------------------------------------

best_advanced_cv_model = (
    advanced_cv_results_df.iloc[0]["Model"]
)

best_advanced_cv_macro_f1 = (
    advanced_cv_results_df.iloc[0][
        "CV Macro F1 Mean"
    ]
)

best_advanced_cv_balanced_accuracy = (
    advanced_cv_results_df.iloc[0][
        "CV Balanced Accuracy Mean"
    ]
)

logger.info("-" * 70)
logger.info("BEST CROSS-VALIDATED MODEL")
logger.info("-" * 70)

logger.info(
    f"Best Cross-Validated Model                 : "
    f"{best_advanced_cv_model}"
)

logger.info(
    f"Best CV Macro F1                           : "
    f"{best_advanced_cv_macro_f1:.4f}"
)

logger.info(
    f"Best CV Balanced Accuracy                  : "
    f"{best_advanced_cv_balanced_accuracy:.4f}"
)


# -----------------------------------------------------------------------------
# CROSS-VALIDATION FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION FINAL VALIDATION")
logger.info("-" * 70)

if len(advanced_cv_results_df) != 6:
    raise ValueError(
        "Expected results for exactly six advanced models."
    )

if advanced_cv_results_df[
    "CV Macro F1 Mean"
].isna().any():
    raise ValueError(
        "Missing CV Macro F1 values detected."
    )

if advanced_cv_results_df[
    "CV Balanced Accuracy Mean"
].isna().any():
    raise ValueError(
        "Missing CV Balanced Accuracy values detected."
    )

if (
    advanced_cv_results_df[
        "CV Macro F1 Std"
    ] < 0
).any():
    raise ValueError(
        "Invalid CV Macro F1 standard deviation."
    )

if not (
    advanced_cv_results_df["CV Rank"]
    == np.arange(1, 7)
).all():
    raise ValueError(
        "Cross-validation ranking validation failed."
    )

# Confirm original target was not modified.
if sorted(
    y_train.unique().tolist()
) != [1, 2, 3, 4, 5, 6]:
    raise ValueError(
        "Original y_train target labels were modified."
    )

logger.info(
    "Six Advanced Models Evaluated              : PASSED"
)

logger.info(
    "Five-Fold Stratified CV                    : PASSED"
)

logger.info(
    "XGBoost Zero-Based Target Handling         : PASSED"
)

logger.info(
    "Original Target Preservation               : PASSED"
)

logger.info(
    "Metric Calculation                         : PASSED"
)

logger.info(
    "Model Ranking                              : PASSED"
)

logger.info(
    "Test Set Leakage Check                     : PASSED"
)

logger.info(
    "Advanced Cross-Validation                  : PASSED"
)

logger.info("=" * 70)
````

 ## 11. **CROSS-VALIDATION & HYPERPARAMETER OPTIMIZATION** ## 

## 11.1 Optimization Strategy

Hyperparameter optimization is performed to improve the generalization performance of the strongest advanced baseline models identified in Section 10.

Based on the advanced model comparison, three models are selected for optimization:

* Random Forest
* CatBoost
* LightGBM

These models provide complementary performance characteristics across the independent test set and 5-fold cross-validation.

The optimization process follows a two-stage strategy:

1. **Randomized Search** is used as the primary exploration method to efficiently evaluate a broad hyperparameter space.
2. **Selective Grid Search** is subsequently applied around the best-performing region identified by Randomized Search.

The optimization process uses the training dataset only. The independent test set remains completely isolated and is not used for hyperparameter selection.

The primary optimization objective is **Macro F1**, because this project contains six target classes with an imbalanced class distribution. Balanced Accuracy is retained as an important secondary metric.

The final optimized models will be retrained using the selected hyperparameters and evaluated on the untouched test set only after optimization is complete.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import (
    StratifiedKFold,
    RandomizedSearchCV,
    GridSearchCV
)

from sklearn.metrics import (
    make_scorer,
    f1_score,
    balanced_accuracy_score
)

from sklearn.ensemble import RandomForestClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier


# =============================================================================
# 11.1 OPTIMIZATION STRATEGY
# =============================================================================

logger.info("=" * 70)
logger.info("11.1 OPTIMIZATION STRATEGY")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# VALIDATE REQUIRED TRAINING DATA
# -----------------------------------------------------------------------------

if "X_train" not in globals():
    raise NameError("X_train is not available.")

if "y_train" not in globals():
    raise NameError("y_train is not available.")

if "X_test" not in globals():
    raise NameError("X_test is not available.")

if "y_test" not in globals():
    raise NameError("y_test is not available.")


# -----------------------------------------------------------------------------
# PROTECT ORIGINAL TRAINING DATA
# -----------------------------------------------------------------------------

X_train_optimization = X_train.copy(deep=True)
y_train_optimization = y_train.copy(deep=True)

X_test_optimization = X_test.copy(deep=True)
y_test_optimization = y_test.copy(deep=True)


# -----------------------------------------------------------------------------
# DATASET INFORMATION
# -----------------------------------------------------------------------------

optimization_train_samples = len(X_train_optimization)
optimization_test_samples = len(X_test_optimization)
optimization_features = X_train_optimization.shape[1]
optimization_classes = sorted(
    y_train_optimization.unique().tolist()
)


logger.info(
    f"Training Samples                           : "
    f"{optimization_train_samples}"
)

logger.info(
    f"Test Samples                               : "
    f"{optimization_test_samples}"
)

logger.info(
    f"Original Features                          : "
    f"{optimization_features}"
)

logger.info(
    f"Number of Target Classes                   : "
    f"{len(optimization_classes)}"
)

logger.info(
    f"Target Classes                             : "
    f"{optimization_classes}"
)


# -----------------------------------------------------------------------------
# OPTIMIZATION MODELS
# -----------------------------------------------------------------------------

optimization_models = {
    "Random Forest": RandomForestClassifier(
        random_state=42,
        n_jobs=-1,
        class_weight="balanced"
    ),

    "CatBoost": CatBoostClassifier(
        loss_function="MultiClass",
        random_seed=42,
        verbose=False,
        thread_count=-1,
        allow_writing_files=False
    ),

    "LightGBM": LGBMClassifier(
        objective="multiclass",
        num_class=len(optimization_classes),
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )
}


logger.info("-" * 70)
logger.info("MODELS SELECTED FOR OPTIMIZATION")
logger.info("-" * 70)

for model_name in optimization_models:

    logger.info(
        f"Optimization Candidate                     : "
        f"{model_name}"
    )


# -----------------------------------------------------------------------------
# CROSS-VALIDATION STRATEGY
# -----------------------------------------------------------------------------

optimization_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


logger.info("-" * 70)
logger.info("CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "CV Strategy                                : "
    "Stratified K-Fold"
)

logger.info(
    "Number of Folds                            : "
    "5"
)

logger.info(
    "Shuffle                                    : "
    "True"
)

logger.info(
    "Random State                               : "
    "42"
)


# -----------------------------------------------------------------------------
# PRIMARY AND SECONDARY SCORING
# -----------------------------------------------------------------------------

macro_f1_scorer = make_scorer(
    f1_score,
    average="macro"
)

balanced_accuracy_scorer = make_scorer(
    balanced_accuracy_score
)


optimization_scoring = {
    "macro_f1": macro_f1_scorer,
    "balanced_accuracy": balanced_accuracy_scorer
}


logger.info("-" * 70)
logger.info("OPTIMIZATION METRICS")
logger.info("-" * 70)

logger.info(
    "Primary Metric                              : "
    "Macro F1"
)

logger.info(
    "Secondary Metric                            : "
    "Balanced Accuracy"
)

logger.info(
    "Test Set Used During Optimization          : "
    "NO"
)


# -----------------------------------------------------------------------------
# TEST SET ISOLATION
# -----------------------------------------------------------------------------

if len(X_test_optimization) != len(y_test_optimization):
    raise ValueError(
        "Test feature and target lengths do not match."
    )

if len(X_train_optimization) != len(y_train_optimization):
    raise ValueError(
        "Training feature and target lengths do not match."
    )


# -----------------------------------------------------------------------------
# ORIGINAL DATA PROTECTION VALIDATION
# -----------------------------------------------------------------------------

if not X_train_optimization.equals(X_train):
    raise ValueError(
        "Protected optimization training data differs "
        "from original X_train."
    )

if not y_train_optimization.equals(y_train):
    raise ValueError(
        "Protected optimization target differs "
        "from original y_train."
    )


# -----------------------------------------------------------------------------
# OPTIMIZATION CONFIGURATION REGISTRY
# -----------------------------------------------------------------------------

optimization_configuration = pd.DataFrame({

    "Model": [
        "Random Forest",
        "CatBoost",
        "LightGBM"
    ],

    "Optimization Method": [
        "Randomized Search → Selective Grid Search",
        "Randomized Search → Selective Grid Search",
        "Randomized Search → Selective Grid Search"
    ],

    "CV Strategy": [
        "5-Fold Stratified",
        "5-Fold Stratified",
        "5-Fold Stratified"
    ],

    "Primary Metric": [
        "Macro F1",
        "Macro F1",
        "Macro F1"
    ],

    "Secondary Metric": [
        "Balanced Accuracy",
        "Balanced Accuracy",
        "Balanced Accuracy"
    ],

    "Test Set Used": [
        "No",
        "No",
        "No"
    ]
})


display(optimization_configuration)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.1 OPTIMIZATION STRATEGY VALIDATION")
logger.info("-" * 70)

if len(optimization_models) != 3:
    raise ValueError(
        "Expected exactly three optimization candidates."
    )

if optimization_cv.n_splits != 5:
    raise ValueError(
        "Expected 5-fold cross-validation."
    )

if "macro_f1" not in optimization_scoring:
    raise ValueError(
        "Macro F1 scorer is missing."
    )

if "balanced_accuracy" not in optimization_scoring:
    raise ValueError(
        "Balanced Accuracy scorer is missing."
    )

logger.info(
    "Three Optimization Candidates              : PASSED"
)

logger.info(
    "Training/Test Isolation                     : PASSED"
)

logger.info(
    "Five-Fold Stratification                    : PASSED"
)

logger.info(
    "Macro F1 Primary Metric                     : PASSED"
)

logger.info(
    "Balanced Accuracy Secondary Metric          : PASSED"
)

logger.info(
    "Original Training Data Preserved            : PASSED"
)

logger.info("=" * 70)
logger.info("11.1 OPTIMIZATION STRATEGY : COMPLETED")
logger.info("=" * 70)
````

## 11.2 Stratified K-Fold Configuration

Stratified K-Fold Cross-Validation is used to ensure that each validation fold maintains a representative distribution of all six target classes.

The training dataset is divided into five folds using stratification, shuffling, and a fixed random state of 42.

For each iteration, four folds are used for training and one fold is used for validation. The process is repeated five times so that every training sample is used for validation exactly once.

The independent test set remains completely excluded from the cross-validation process.

Stratification is particularly important for this project because the target classes are not equally distributed, with class 6 having substantially fewer observations than the other classes.

The configured cross-validation object will be reused by Randomized Search and Selective Grid Search in the subsequent optimization stages.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold


# =============================================================================
# 11.2 STRATIFIED K-FOLD CONFIGURATION
# =============================================================================

logger.info("=" * 70)
logger.info("11.2 STRATIFIED K-FOLD CONFIGURATION")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# VALIDATE REQUIRED DATA
# -----------------------------------------------------------------------------

if "X_train_optimization" not in globals():
    raise NameError(
        "X_train_optimization is not available. "
        "Run Section 11.1 first."
    )

if "y_train_optimization" not in globals():
    raise NameError(
        "y_train_optimization is not available. "
        "Run Section 11.1 first."
    )

if "X_test_optimization" not in globals():
    raise NameError(
        "X_test_optimization is not available. "
        "Run Section 11.1 first."
    )

if "y_test_optimization" not in globals():
    raise NameError(
        "y_test_optimization is not available. "
        "Run Section 11.1 first."
    )


# -----------------------------------------------------------------------------
# CREATE STRATIFIED K-FOLD OBJECT
# -----------------------------------------------------------------------------

optimization_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


logger.info("-" * 70)
logger.info("STRATIFIED K-FOLD CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Cross-Validation Strategy                   : "
    "Stratified K-Fold"
)

logger.info(
    "Number of Folds                            : "
    f"{optimization_cv.n_splits}"
)

logger.info(
    "Shuffle                                    : "
    f"{optimization_cv.shuffle}"
)

logger.info(
    "Random State                               : "
    f"{optimization_cv.random_state}"
)


# -----------------------------------------------------------------------------
# TRAINING DATA INFORMATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING DATA FOR CROSS-VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                           : "
    f"{len(X_train_optimization)}"
)

logger.info(
    f"Training Features                          : "
    f"{X_train_optimization.shape[1]}"
)

logger.info(
    f"Target Classes                             : "
    f"{sorted(y_train_optimization.unique().tolist())}"
)

logger.info(
    f"Number of Target Classes                   : "
    f"{y_train_optimization.nunique()}"
)


# -----------------------------------------------------------------------------
# GENERATE FOLDS
# -----------------------------------------------------------------------------

fold_records = []

fold_class_distributions = []

for fold_number, (train_indices, validation_indices) in enumerate(
    optimization_cv.split(
        X_train_optimization,
        y_train_optimization
    ),
    start=1
):

    y_fold_train = y_train_optimization.iloc[train_indices]
    y_fold_validation = y_train_optimization.iloc[validation_indices]

    fold_train_size = len(train_indices)
    fold_validation_size = len(validation_indices)

    fold_train_distribution = (
        y_fold_train
        .value_counts()
        .sort_index()
    )

    fold_validation_distribution = (
        y_fold_validation
        .value_counts()
        .sort_index()
    )

    fold_records.append({
        "Fold": fold_number,
        "Training Samples": fold_train_size,
        "Validation Samples": fold_validation_size
    })

    fold_class_distributions.append({
        "Fold": fold_number,

        "Train Class 1": fold_train_distribution.get(1, 0),
        "Train Class 2": fold_train_distribution.get(2, 0),
        "Train Class 3": fold_train_distribution.get(3, 0),
        "Train Class 4": fold_train_distribution.get(4, 0),
        "Train Class 5": fold_train_distribution.get(5, 0),
        "Train Class 6": fold_train_distribution.get(6, 0),

        "Validation Class 1": fold_validation_distribution.get(1, 0),
        "Validation Class 2": fold_validation_distribution.get(2, 0),
        "Validation Class 3": fold_validation_distribution.get(3, 0),
        "Validation Class 4": fold_validation_distribution.get(4, 0),
        "Validation Class 5": fold_validation_distribution.get(5, 0),
        "Validation Class 6": fold_validation_distribution.get(6, 0)
    })


# -----------------------------------------------------------------------------
# DISPLAY FOLD SIZES
# -----------------------------------------------------------------------------

fold_summary_df = pd.DataFrame(
    fold_records
)

logger.info("-" * 70)
logger.info("FOLD SIZE SUMMARY")
logger.info("-" * 70)

display(
    fold_summary_df
)


# -----------------------------------------------------------------------------
# DISPLAY CLASS DISTRIBUTION
# -----------------------------------------------------------------------------

fold_distribution_df = pd.DataFrame(
    fold_class_distributions
)

logger.info("-" * 70)
logger.info("CLASS DISTRIBUTION ACROSS FOLDS")
logger.info("-" * 70)

display(
    fold_distribution_df
)


# -----------------------------------------------------------------------------
# VALIDATE FOLD SIZES
# -----------------------------------------------------------------------------

expected_total_samples = (
    len(X_train_optimization)
)

for _, row in fold_summary_df.iterrows():

    if (
        row["Training Samples"]
        +
        row["Validation Samples"]
        != expected_total_samples
    ):

        raise ValueError(
            f"Fold {int(row['Fold'])} does not contain "
            "the expected total number of samples."
        )


# -----------------------------------------------------------------------------
# VALIDATE EACH CLASS EXISTS IN EVERY VALIDATION FOLD
# -----------------------------------------------------------------------------

target_classes = sorted(
    y_train_optimization.unique().tolist()
)

for _, row in fold_distribution_df.iterrows():

    fold_number = int(row["Fold"])

    for target_class in target_classes:

        validation_column = (
            f"Validation Class {target_class}"
        )

        if row[validation_column] <= 0:

            raise ValueError(
                f"Class {target_class} is missing from "
                f"validation fold {fold_number}."
            )


# -----------------------------------------------------------------------------
# VALIDATE ALL SAMPLES APPEAR IN VALIDATION EXACTLY ONCE
# -----------------------------------------------------------------------------

validation_occurrence_count = np.zeros(
    len(X_train_optimization),
    dtype=int
)

for train_indices, validation_indices in optimization_cv.split(
    X_train_optimization,
    y_train_optimization
):

    validation_occurrence_count[
        validation_indices
    ] += 1


if not np.all(
    validation_occurrence_count == 1
):

    raise ValueError(
        "Validation samples are not distributed correctly. "
        "Each training sample must appear in validation exactly once."
    )


# -----------------------------------------------------------------------------
# VALIDATE NO TRAIN/VALIDATION INDEX OVERLAP
# -----------------------------------------------------------------------------

for fold_number, (
    train_indices,
    validation_indices
) in enumerate(
    optimization_cv.split(
        X_train_optimization,
        y_train_optimization
    ),
    start=1
):

    overlap = np.intersect1d(
        train_indices,
        validation_indices
    )

    if len(overlap) != 0:

        raise ValueError(
            f"Data leakage detected in fold {fold_number}: "
            "training and validation indices overlap."
        )


# -----------------------------------------------------------------------------
# VALIDATE TEST SET IS NOT PART OF CV
# -----------------------------------------------------------------------------

if (
    len(X_test_optimization)
    !=
    len(y_test_optimization)
):

    raise ValueError(
        "Test feature and target sizes do not match."
    )


logger.info("-" * 70)
logger.info("LEAKAGE AND FOLD VALIDATION")
logger.info("-" * 70)

logger.info(
    "Training Samples Used in CV                 : "
    f"{len(X_train_optimization)}"
)

logger.info(
    "Independent Test Samples Excluded           : "
    f"{len(X_test_optimization)}"
)

logger.info(
    "Validation Occurrence Per Training Sample   : "
    "Exactly 1"
)

logger.info(
    "Train/Validation Index Overlap              : "
    "None"
)


# -----------------------------------------------------------------------------
# SAVE CV FOLD INFORMATION FOR LATER SECTIONS
# -----------------------------------------------------------------------------

optimization_fold_summary = fold_summary_df.copy(
    deep=True
)

optimization_fold_distribution = fold_distribution_df.copy(
    deep=True
)


# -----------------------------------------------------------------------------
# FINAL VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.2 STRATIFIED K-FOLD VALIDATION")
logger.info("-" * 70)

if optimization_cv.n_splits != 5:
    raise ValueError(
        "Cross-validation must contain exactly five folds."
    )

if optimization_cv.shuffle is not True:
    raise ValueError(
        "Cross-validation shuffling must be enabled."
    )

if optimization_cv.random_state != 42:
    raise ValueError(
        "Random state must be 42 for reproducibility."
    )

if len(fold_summary_df) != 5:
    raise ValueError(
        "Expected exactly five fold records."
    )

if len(fold_distribution_df) != 5:
    raise ValueError(
        "Expected exactly five class-distribution records."
    )

logger.info(
    "Five Stratified Folds                       : PASSED"
)

logger.info(
    "Class Representation                       : PASSED"
)

logger.info(
    "Validation Coverage                        : PASSED"
)

logger.info(
    "Train/Validation Leakage Check             : PASSED"
)

logger.info(
    "Independent Test Set Isolation              : PASSED"
)

logger.info(
    "Reproducibility Configuration               : PASSED"
)

logger.info("=" * 70)
logger.info("11.2 STRATIFIED K-FOLD CONFIGURATION : COMPLETED")
logger.info("=" * 70)
````

## 11.3 Evaluation Metric Selection

The hyperparameter optimization process requires a primary metric that reflects the actual objective of the multiclass classification problem.

This dataset contains six target classes with an imbalanced class distribution. Therefore, plain accuracy alone is not sufficient for selecting the best model because it can be influenced by the larger classes.

### Primary Optimization Metric — Macro F1

Macro F1 is selected as the primary optimization metric.

Macro F1 calculates the F1 score independently for each class and then gives every class equal importance. This makes it suitable for evaluating whether the model performs consistently across both frequent and less frequent disorder classes.

### Secondary Evaluation Metric — Balanced Accuracy

Balanced Accuracy is selected as the secondary metric.

Balanced Accuracy averages the recall obtained for each class and therefore provides an additional measure of whether the classifier is treating the six classes fairly.

### Supporting Metrics

The following metrics will also be retained for model analysis:

* Accuracy
* Macro Precision
* Macro Recall
* Macro F1
* Weighted F1
* Balanced Accuracy
* Cross-Validation Standard Deviation

The optimization process will use **Macro F1 as the refit metric**. Balanced Accuracy will be evaluated simultaneously during the search but will not replace Macro F1 as the primary selection criterion.

The independent test set will remain completely isolated during this process and will only be evaluated after the optimal hyperparameters have been identified.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    make_scorer,
    f1_score,
    balanced_accuracy_score,
    accuracy_score,
    precision_score,
    recall_score
)


# =============================================================================
# 11.3 EVALUATION METRIC SELECTION
# =============================================================================

logger.info("=" * 70)
logger.info("11.3 EVALUATION METRIC SELECTION")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# -----------------------------------------------------------------------------

required_objects = [
    "X_train_optimization",
    "y_train_optimization",
    "X_test_optimization",
    "y_test_optimization",
    "optimization_cv"
]

for object_name in required_objects:

    if object_name not in globals():

        raise NameError(
            f"Required object '{object_name}' is not available. "
            f"Run Section 11.1 and 11.2 first."
        )


# -----------------------------------------------------------------------------
# TARGET VALIDATION
# -----------------------------------------------------------------------------

target_classes = sorted(
    y_train_optimization.unique().tolist()
)

number_of_classes = len(target_classes)

if number_of_classes != 6:

    raise ValueError(
        f"Expected six target classes, found {number_of_classes}."
    )


logger.info("-" * 70)
logger.info("TARGET CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Number of Target Classes                   : "
    f"{number_of_classes}"
)

logger.info(
    f"Target Classes                             : "
    f"{target_classes}"
)


# -----------------------------------------------------------------------------
# CREATE OPTIMIZATION SCORERS
# -----------------------------------------------------------------------------

macro_f1_scorer = make_scorer(
    f1_score,
    average="macro",
    zero_division=0
)

balanced_accuracy_scorer = make_scorer(
    balanced_accuracy_score
)

accuracy_scorer = make_scorer(
    accuracy_score
)

macro_precision_scorer = make_scorer(
    precision_score,
    average="macro",
    zero_division=0
)

macro_recall_scorer = make_scorer(
    recall_score,
    average="macro",
    zero_division=0
)


# -----------------------------------------------------------------------------
# MULTI-METRIC SCORING REGISTRY
# -----------------------------------------------------------------------------

optimization_scoring = {

    "macro_f1": macro_f1_scorer,

    "balanced_accuracy": balanced_accuracy_scorer,

    "accuracy": accuracy_scorer,

    "macro_precision": macro_precision_scorer,

    "macro_recall": macro_recall_scorer
}


# -----------------------------------------------------------------------------
# PRIMARY REFIT METRIC
# -----------------------------------------------------------------------------

optimization_refit_metric = "macro_f1"


logger.info("-" * 70)
logger.info("OPTIMIZATION METRIC CONFIGURATION")
logger.info("-" * 70)

logger.info(
    "Primary Optimization Metric                : "
    "Macro F1"
)

logger.info(
    "Secondary Evaluation Metric                : "
    "Balanced Accuracy"
)

logger.info(
    "Additional Metrics                         : "
    "Accuracy, Macro Precision, Macro Recall"
)

logger.info(
    "Search Refit Metric                        : "
    f"{optimization_refit_metric}"
)

logger.info(
    "Test Set Used for Metric Selection         : "
    "NO"
)


# -----------------------------------------------------------------------------
# METRIC DESCRIPTION TABLE
# -----------------------------------------------------------------------------

metric_configuration_df = pd.DataFrame({

    "Metric": [
        "Macro F1",
        "Balanced Accuracy",
        "Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Weighted F1"
    ],

    "Role": [
        "Primary Optimization Metric",
        "Secondary Evaluation Metric",
        "Supporting Metric",
        "Supporting Metric",
        "Supporting Metric",
        "Supporting Metric"
    ],

    "Importance": [
        "High",
        "High",
        "Medium",
        "Medium",
        "Medium",
        "Medium"
    ],

    "Reason": [
        "Equal importance to all six classes",
        "Average recall across all six classes",
        "Overall proportion of correct predictions",
        "Measures precision equally across classes",
        "Measures recall equally across classes",
        "Overall F1 weighted by class support"
    ]
})


logger.info("-" * 70)
logger.info("METRIC CONFIGURATION TABLE")
logger.info("-" * 70)

display(
    metric_configuration_df
)


# -----------------------------------------------------------------------------
# CLASS DISTRIBUTION FOR CONTEXT
# -----------------------------------------------------------------------------

class_distribution = (
    y_train_optimization
    .value_counts()
    .sort_index()
)

class_distribution_percentage = (
    y_train_optimization
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

class_distribution_df = pd.DataFrame({

    "Class": class_distribution.index,

    "Training Samples": class_distribution.values,

    "Training Percentage (%)":
        class_distribution_percentage.values
})


logger.info("-" * 70)
logger.info("TRAINING CLASS DISTRIBUTION")
logger.info("-" * 70)

display(
    class_distribution_df.round(2)
)


# -----------------------------------------------------------------------------
# IMBALANCE CHECK
# -----------------------------------------------------------------------------

largest_class_count = (
    class_distribution.max()
)

smallest_class_count = (
    class_distribution.min()
)

class_imbalance_ratio = (
    largest_class_count
    /
    smallest_class_count
)


logger.info(
    f"Largest Class Count                       : "
    f"{largest_class_count}"
)

logger.info(
    f"Smallest Class Count                      : "
    f"{smallest_class_count}"
)

logger.info(
    f"Class Imbalance Ratio                     : "
    f"{class_imbalance_ratio:.4f}"
)


# -----------------------------------------------------------------------------
# METRIC REGISTRY FOR LATER SEARCH SECTIONS
# -----------------------------------------------------------------------------

optimization_metric_registry = {

    "primary": "macro_f1",

    "secondary": "balanced_accuracy",

    "additional": [
        "accuracy",
        "macro_precision",
        "macro_recall"
    ],

    "refit": "macro_f1",

    "cv_folds": 5,

    "test_set_used": False
}


# -----------------------------------------------------------------------------
# VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.3 EVALUATION METRIC VALIDATION")
logger.info("-" * 70)


if optimization_refit_metric != "macro_f1":

    raise ValueError(
        "Macro F1 must be the primary refit metric."
    )


if "macro_f1" not in optimization_scoring:

    raise ValueError(
        "Macro F1 scorer is missing."
    )


if "balanced_accuracy" not in optimization_scoring:

    raise ValueError(
        "Balanced Accuracy scorer is missing."
    )


if len(optimization_scoring) != 5:

    raise ValueError(
        "Expected five configured optimization scorers."
    )


if number_of_classes != 6:

    raise ValueError(
        "Metric configuration must support six classes."
    )


if len(X_train_optimization) != len(y_train_optimization):

    raise ValueError(
        "Training feature and target sizes do not match."
    )


if len(X_test_optimization) != len(y_test_optimization):

    raise ValueError(
        "Test feature and target sizes do not match."
    )


logger.info(
    "Six-Class Multiclass Configuration           : PASSED"
)

logger.info(
    "Macro F1 Primary Metric                      : PASSED"
)

logger.info(
    "Balanced Accuracy Secondary Metric           : PASSED"
)

logger.info(
    "Multi-Metric Scoring Registry               : PASSED"
)

logger.info(
    "Test Set Isolation                           : PASSED"
)

logger.info(
    "Metric Reproducibility Configuration         : PASSED"
)

logger.info("=" * 70)
logger.info("11.3 EVALUATION METRIC SELECTION : COMPLETED")
logger.info("=" * 70)
````

## 11.4 Randomized Search — Corrected Preprocessing

The initial Randomized Search identified a data-type incompatibility because the `Age` feature contains missing-value markers represented as `"?"`.

To ensure that hyperparameter optimization operates on model-ready data while preventing preprocessing leakage, the preprocessing pipeline is incorporated directly into the cross-validation workflow.

The preprocessing strategy is:

* Convert `Age` from object representation to numeric.
* Treat `"?"` as a missing value.
* Median-impute missing `Age` values.
* One-hot encode categorical clinical features.
* Preserve unknown categorical levels safely.
* Perform preprocessing independently inside each cross-validation training fold.
* Apply the learned transformations to the corresponding validation fold.
* Keep the independent test set completely isolated.

The original `X_train_optimization` and `X_test_optimization` objects are not modified.


````python
# =============================================================================
# REQUIRED LIBRARIES
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import RandomizedSearchCV

from sklearn.ensemble import RandomForestClassifier

from catboost import CatBoostClassifier

from lightgbm import LGBMClassifier


# =============================================================================
# 11.4 RANDOMIZED SEARCH — CORRECTED
# =============================================================================

logger.info("=" * 70)
logger.info("11.4 RANDOMIZED SEARCH")
logger.info("=" * 70)


# -----------------------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# -----------------------------------------------------------------------------

required_objects = [
    "X_train_optimization",
    "y_train_optimization",
    "X_test_optimization",
    "y_test_optimization",
    "optimization_cv",
    "optimization_scoring",
    "optimization_refit_metric"
]

for object_name in required_objects:

    if object_name not in globals():

        raise NameError(
            f"Required object '{object_name}' is not available. "
            f"Run Sections 11.1, 11.2 and 11.3 first."
        )


# -----------------------------------------------------------------------------
# CREATE SAFE COPIES
# -----------------------------------------------------------------------------

X_train_random_search = (
    X_train_optimization.copy(deep=True)
)

y_train_random_search = (
    y_train_optimization.copy(deep=True)
)

X_test_random_search = (
    X_test_optimization.copy(deep=True)
)

y_test_random_search = (
    y_test_optimization.copy(deep=True)
)


# -----------------------------------------------------------------------------
# VALIDATE ORIGINAL DATA
# -----------------------------------------------------------------------------

if X_train_random_search.shape[0] != 292:

    raise ValueError(
        f"Expected 292 training samples, found "
        f"{X_train_random_search.shape[0]}."
    )


if X_test_random_search.shape[0] != 74:

    raise ValueError(
        f"Expected 74 test samples, found "
        f"{X_test_random_search.shape[0]}."
    )


logger.info("-" * 70)
logger.info("INPUT DATA VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Training Samples                            : "
    f"{X_train_random_search.shape[0]}"
)

logger.info(
    f"Training Features                           : "
    f"{X_train_random_search.shape[1]}"
)

logger.info(
    f"Independent Test Samples                    : "
    f"{X_test_random_search.shape[0]}"
)

logger.info(
    f"Independent Test Features                   : "
    f"{X_test_random_search.shape[1]}"
)


# -----------------------------------------------------------------------------
# IDENTIFY AGE FEATURE
# -----------------------------------------------------------------------------

if "Age" not in X_train_random_search.columns:

    raise ValueError(
        "Age feature is required but was not found."
    )


# -----------------------------------------------------------------------------
# CONVERT AGE TO NUMERIC SAFELY
# -----------------------------------------------------------------------------

age_missing_train = (
    X_train_random_search["Age"]
    .astype(str)
    .eq("?")
    .sum()
)

age_missing_test = (
    X_test_random_search["Age"]
    .astype(str)
    .eq("?")
    .sum()
)


X_train_random_search["Age"] = pd.to_numeric(
    X_train_random_search["Age"],
    errors="coerce"
)

X_test_random_search["Age"] = pd.to_numeric(
    X_test_random_search["Age"],
    errors="coerce"
)


logger.info("-" * 70)
logger.info("AGE DATA-TYPE CORRECTION")
logger.info("-" * 70)

logger.info(
    f"Training Age Missing Markers                : "
    f"{age_missing_train}"
)

logger.info(
    f"Test Age Missing Markers                    : "
    f"{age_missing_test}"
)

logger.info(
    "Age Conversion                               : "
    "Completed"
)

logger.info(
    "Age Imputation                               : "
    "Median inside CV Pipeline"
)


# -----------------------------------------------------------------------------
# IDENTIFY FEATURE TYPES
# -----------------------------------------------------------------------------

numeric_features = [
    "Age"
]

categorical_features = [
    column
    for column in X_train_random_search.columns
    if column != "Age"
]


logger.info("-" * 70)
logger.info("FEATURE TYPE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Numeric Features                            : "
    f"{len(numeric_features)}"
)

logger.info(
    f"Categorical Features                        : "
    f"{len(categorical_features)}"
)

logger.info(
    "Categorical Encoding                        : "
    "One-Hot Encoding"
)

logger.info(
    "Ordinal Encoding                            : "
    "NOT USED"
)


# -----------------------------------------------------------------------------
# NUMERIC PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )
    ]
)


# -----------------------------------------------------------------------------
# CATEGORICAL PREPROCESSING PIPELINE
# -----------------------------------------------------------------------------

categorical_pipeline = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


# -----------------------------------------------------------------------------
# COMPLETE PREPROCESSING TRANSFORMER
# -----------------------------------------------------------------------------

random_search_preprocessor = ColumnTransformer(

    transformers=[

        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],

    remainder="drop"
)


# -----------------------------------------------------------------------------
# MODEL DEFINITIONS
# -----------------------------------------------------------------------------

random_forest_model = RandomForestClassifier(
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)


catboost_model = CatBoostClassifier(
    random_seed=42,
    verbose=0,
    allow_writing_files=False,
    loss_function="MultiClass",
    auto_class_weights="Balanced",
    thread_count=1
)


lightgbm_model = LGBMClassifier(
    objective="multiclass",
    random_state=42,
    class_weight="balanced",
    n_jobs=-1,
    verbosity=-1
)


# -----------------------------------------------------------------------------
# RANDOM FOREST PARAMETER DISTRIBUTION
# -----------------------------------------------------------------------------

random_forest_param_distributions = {

    "model__n_estimators": [
        200,
        300,
        400,
        500,
        700
    ],

    "model__max_depth": [
        None,
        5,
        8,
        10,
        15,
        20
    ],

    "model__min_samples_split": [
        2,
        4,
        6,
        10
    ],

    "model__min_samples_leaf": [
        1,
        2,
        4,
        6
    ],

    "model__max_features": [
        "sqrt",
        "log2",
        0.5,
        0.75
    ],

    "model__criterion": [
        "gini",
        "entropy"
    ]
}


# -----------------------------------------------------------------------------
# CATBOOST PARAMETER DISTRIBUTION
# -----------------------------------------------------------------------------

catboost_param_distributions = {

    "model__iterations": [
        200,
        300,
        400,
        500,
        700
    ],

    "model__depth": [
        4,
        5,
        6,
        7,
        8,
        10
    ],

    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.08,
        0.10,
        0.15
    ],

    "model__l2_leaf_reg": [
        1,
        3,
        5,
        7,
        10
    ],

    "model__random_strength": [
        0.1,
        0.5,
        1.0,
        2.0
    ]
}


# -----------------------------------------------------------------------------
# LIGHTGBM PARAMETER DISTRIBUTION
# -----------------------------------------------------------------------------

lightgbm_param_distributions = {

    "model__n_estimators": [
        100,
        200,
        300,
        400,
        500,
        700
    ],

    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.08,
        0.10,
        0.15
    ],

    "model__num_leaves": [
        15,
        20,
        31,
        40,
        50,
        70
    ],

    "model__max_depth": [
        -1,
        3,
        5,
        7,
        10
    ],

    "model__min_child_samples": [
        5,
        10,
        15,
        20,
        30
    ],

    "model__subsample": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "model__colsample_bytree": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    "model__reg_alpha": [
        0.0,
        0.01,
        0.1,
        0.5
    ],

    "model__reg_lambda": [
        0.0,
        0.01,
        0.1,
        0.5,
        1.0
    ]
}


# -----------------------------------------------------------------------------
# CREATE MODEL PIPELINES
# -----------------------------------------------------------------------------

random_forest_pipeline = Pipeline(
    steps=[

        (
            "preprocessor",
            random_search_preprocessor
        ),

        (
            "model",
            random_forest_model
        )
    ]
)


catboost_pipeline = Pipeline(
    steps=[

        (
            "preprocessor",
            random_search_preprocessor
        ),

        (
            "model",
            catboost_model
        )
    ]
)


lightgbm_pipeline = Pipeline(
    steps=[

        (
            "preprocessor",
            random_search_preprocessor
        ),

        (
            "model",
            lightgbm_model
        )
    ]
)


# -----------------------------------------------------------------------------
# SEARCH CONFIGURATION
# -----------------------------------------------------------------------------

RANDOM_SEARCH_ITERATIONS = 20


logger.info("-" * 70)
logger.info("RANDOMIZED SEARCH CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Iterations Per Model                        : "
    f"{RANDOM_SEARCH_ITERATIONS}"
)

logger.info(
    f"Cross-Validation Folds                      : "
    f"{optimization_cv.n_splits}"
)

logger.info(
    "Primary Metric                              : "
    "Macro F1"
)

logger.info(
    "Secondary Metric                            : "
    "Balanced Accuracy"
)

logger.info(
    "Preprocessing Leakage Prevention            : "
    "Pipeline-Based"
)

logger.info(
    "Independent Test Set                        : "
    "NOT USED"
)


# -----------------------------------------------------------------------------
# STORAGE
# -----------------------------------------------------------------------------

randomized_search_objects = {}

randomized_search_results = []


# -----------------------------------------------------------------------------
# RANDOM FOREST SEARCH
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STARTING RANDOM FOREST RANDOMIZED SEARCH")
logger.info("-" * 70)


rf_random_search = RandomizedSearchCV(

    estimator=random_forest_pipeline,

    param_distributions=random_forest_param_distributions,

    n_iter=RANDOM_SEARCH_ITERATIONS,

    scoring=optimization_scoring,

    refit=optimization_refit_metric,

    cv=optimization_cv,

    random_state=42,

    n_jobs=-1,

    return_train_score=True,

    error_score="raise",

    verbose=0
)


rf_random_search.fit(
    X_train_random_search,
    y_train_random_search
)


randomized_search_objects[
    "Random Forest"
] = rf_random_search


rf_best_index = (
    rf_random_search.best_index_
)


logger.info(
    f"Random Forest Best Macro F1                 : "
    f"{rf_random_search.best_score_:.4f}"
)

logger.info(
    f"Random Forest Best Balanced Accuracy        : "
    f"{rf_random_search.cv_results_['mean_test_balanced_accuracy'][rf_best_index]:.4f}"
)

logger.info(
    f"Random Forest Best Parameters               : "
    f"{rf_random_search.best_params_}"
)


# -----------------------------------------------------------------------------
# CATBOOST SEARCH
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STARTING CATBOOST RANDOMIZED SEARCH")
logger.info("-" * 70)


catboost_random_search = RandomizedSearchCV(

    estimator=catboost_pipeline,

    param_distributions=catboost_param_distributions,

    n_iter=RANDOM_SEARCH_ITERATIONS,

    scoring=optimization_scoring,

    refit=optimization_refit_metric,

    cv=optimization_cv,

    random_state=42,

    n_jobs=1,

    return_train_score=True,

    error_score="raise",

    verbose=0
)


catboost_random_search.fit(
    X_train_random_search,
    y_train_random_search
)


randomized_search_objects[
    "CatBoost"
] = catboost_random_search


catboost_best_index = (
    catboost_random_search.best_index_
)


logger.info(
    f"CatBoost Best Macro F1                    : "
    f"{catboost_random_search.best_score_:.4f}"
)

logger.info(
    f"CatBoost Best Balanced Accuracy            : "
    f"{catboost_random_search.cv_results_['mean_test_balanced_accuracy'][catboost_best_index]:.4f}"
)

logger.info(
    f"CatBoost Best Parameters                   : "
    f"{catboost_random_search.best_params_}"
)


# -----------------------------------------------------------------------------
# LIGHTGBM SEARCH
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STARTING LIGHTGBM RANDOMIZED SEARCH")
logger.info("-" * 70)


lightgbm_random_search = RandomizedSearchCV(

    estimator=lightgbm_pipeline,

    param_distributions=lightgbm_param_distributions,

    n_iter=RANDOM_SEARCH_ITERATIONS,

    scoring=optimization_scoring,

    refit=optimization_refit_metric,

    cv=optimization_cv,

    random_state=42,

    n_jobs=-1,

    return_train_score=True,

    error_score="raise",

    verbose=0
)


lightgbm_random_search.fit(
    X_train_random_search,
    y_train_random_search
)


randomized_search_objects[
    "LightGBM"
] = lightgbm_random_search


lightgbm_best_index = (
    lightgbm_random_search.best_index_
)


logger.info(
    f"LightGBM Best Macro F1                    : "
    f"{lightgbm_random_search.best_score_:.4f}"
)

logger.info(
    f"LightGBM Best Balanced Accuracy            : "
    f"{lightgbm_random_search.cv_results_['mean_test_balanced_accuracy'][lightgbm_best_index]:.4f}"
)

logger.info(
    f"LightGBM Best Parameters                   : "
    f"{lightgbm_random_search.best_params_}"
)


# -----------------------------------------------------------------------------
# PERFORMANCE SUMMARY
# -----------------------------------------------------------------------------

for model_name, search_object in randomized_search_objects.items():

    best_index = search_object.best_index_

    cv_results = search_object.cv_results_

    randomized_search_results.append({

        "Model":
            model_name,

        "Best Macro F1":
            cv_results[
                "mean_test_macro_f1"
            ][best_index],

        "Best Balanced Accuracy":
            cv_results[
                "mean_test_balanced_accuracy"
            ][best_index],

        "Best Accuracy":
            cv_results[
                "mean_test_accuracy"
            ][best_index],

        "Best Macro Precision":
            cv_results[
                "mean_test_macro_precision"
            ][best_index],

        "Best Macro Recall":
            cv_results[
                "mean_test_macro_recall"
            ][best_index],

        "Training Macro F1":
            cv_results[
                "mean_train_macro_f1"
            ][best_index],

        "Best Parameters":
            str(search_object.best_params_)
    })


randomized_search_summary = (
    pd.DataFrame(
        randomized_search_results
    )
    .sort_values(
        by="Best Macro F1",
        ascending=False
    )
    .reset_index(drop=True)
)


logger.info("-" * 70)
logger.info("RANDOMIZED SEARCH PERFORMANCE SUMMARY")
logger.info("-" * 70)

display(
    randomized_search_summary.round(4)
)


# -----------------------------------------------------------------------------
# BEST RANDOMIZED SEARCH MODEL
# -----------------------------------------------------------------------------

best_randomized_model = (
    randomized_search_summary
    .iloc[0]["Model"]
)

best_randomized_macro_f1 = (
    randomized_search_summary
    .iloc[0]["Best Macro F1"]
)


logger.info(
    f"Best Randomized Search Model                : "
    f"{best_randomized_model}"
)

logger.info(
    f"Best Randomized Search Macro F1              : "
    f"{best_randomized_macro_f1:.4f}"
)


# -----------------------------------------------------------------------------
# TRANSFORMED FEATURE COUNT CHECK
# -----------------------------------------------------------------------------

best_preprocessor = (
    rf_random_search
    .best_estimator_
    .named_steps["preprocessor"]
)

transformed_feature_count = (
    best_preprocessor
    .transform(
        X_train_random_search.iloc[:1]
    )
    .shape[1]
)


logger.info(
    f"Model-Ready Encoded Feature Count           : "
    f"{transformed_feature_count}"
)


# -----------------------------------------------------------------------------
# LEAKAGE AND INTEGRITY VALIDATION
# -----------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.4 RANDOMIZED SEARCH VALIDATION")
logger.info("-" * 70)


if len(randomized_search_objects) != 3:

    raise ValueError(
        "Expected Randomized Search results for three models."
    )


for model_name, search_object in randomized_search_objects.items():

    if len(search_object.cv_results_["params"]) != RANDOM_SEARCH_ITERATIONS:

        raise ValueError(
            f"{model_name} did not execute "
            f"{RANDOM_SEARCH_ITERATIONS} configurations."
        )


if optimization_refit_metric != "macro_f1":

    raise ValueError(
        "Macro F1 must remain the primary optimization metric."
    )


if optimization_cv.n_splits != 5:

    raise ValueError(
        "Randomized Search must use five-fold CV."
    )


if X_train_random_search["Age"].isna().sum() != age_missing_train:

    raise ValueError(
        "Unexpected change in Age missing-value count."
    )


if X_train_optimization["Age"].astype(str).eq("?").sum() != age_missing_train:

    raise ValueError(
        "Original optimization dataset was modified."
    )


logger.info(
    "Training Data Isolation                     : PASSED"
)

logger.info(
    "Independent Test Set Isolation              : PASSED"
)

logger.info(
    "Five-Fold Stratified CV                     : PASSED"
)

logger.info(
    "Age Missing-Value Handling                  : PASSED"
)

logger.info(
    "One-Hot Encoding Strategy                   : PASSED"
)

logger.info(
    "Pipeline-Based Preprocessing                : PASSED"
)

logger.info(
    "Preprocessing Leakage Prevention             : PASSED"
)

logger.info(
    "Three Advanced Models Optimized             : PASSED"
)

logger.info(
    "Randomized Configuration Count              : PASSED"
)

logger.info("=" * 70)
logger.info("11.4 RANDOMIZED SEARCH : COMPLETED")
logger.info("=" * 70)
````

## 11.5 Grid Search (Selective)

This stage performs a focused Grid Search around the best hyperparameter
regions identified during Randomized Search.

The objective is to refine the three optimized advanced models:

- Random Forest
- CatBoost
- LightGBM

The search space is intentionally limited to avoid unnecessary computation
while still providing meaningful hyperparameter refinement.

### Optimization Design

- Cross-Validation: 5-Fold Stratified K-Fold
- Primary Metric: Macro F1
- Secondary Metric: Balanced Accuracy
- Search Strategy: Selective Grid Search
- Independent Test Set: Not Used
- Preprocessing: Training-fitted Pipeline
- Missing Age Values: Median Imputation
- Categorical Features: One-Hot Encoding
- Ordinal Encoding: Not Used
- Random State: 42

The independent test set remains completely isolated until final model
evaluation.

The best configuration from each selective grid search will be retained
for final hyperparameter identification.

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.ensemble import RandomForestClassifier

from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier



logger.info("=" * 70)
logger.info("11.5 SELECTIVE GRID SEARCH")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. DATA ISOLATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("INPUT DATA VALIDATION")
logger.info("-" * 70)

X_train_grid = X_train_optimization.copy(deep=True)
y_train_grid = y_train_optimization.copy(deep=True)

X_test_grid = X_test_optimization.copy(deep=True)
y_test_grid = y_test_optimization.copy(deep=True)

logger.info(f"Training Samples                         : {len(X_train_grid)}")
logger.info(f"Training Features                        : {X_train_grid.shape[1]}")
logger.info(f"Independent Test Samples                 : {len(X_test_grid)}")
logger.info(f"Independent Test Features                : {X_test_grid.shape[1]}")


# ------------------------------------------------------------
# 2. AGE CONVERSION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE DATA-TYPE CORRECTION")
logger.info("-" * 70)

X_train_grid["Age"] = pd.to_numeric(
    X_train_grid["Age"],
    errors="coerce"
)

X_test_grid["Age"] = pd.to_numeric(
    X_test_grid["Age"],
    errors="coerce"
)

logger.info(
    f"Training Age Missing Values              : "
    f"{X_train_grid['Age'].isna().sum()}"
)

logger.info(
    f"Test Age Missing Values                  : "
    f"{X_test_grid['Age'].isna().sum()}"
)


# ------------------------------------------------------------
# 3. FEATURE CONFIGURATION
# ------------------------------------------------------------

numeric_features_grid = ["Age"]

categorical_features_grid = [
    col for col in X_train_grid.columns
    if col != "Age"
]

logger.info("-" * 70)
logger.info("FEATURE TYPE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Numeric Features                         : "
    f"{len(numeric_features_grid)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(categorical_features_grid)}"
)

logger.info("Categorical Encoding                     : One-Hot Encoding")
logger.info("Ordinal Encoding                         : NOT USED")


# ------------------------------------------------------------
# 4. PREPROCESSING PIPELINE
# ------------------------------------------------------------

numeric_pipeline_grid = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline_grid = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor_grid = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_grid,
            numeric_features_grid
        ),
        (
            "categorical",
            categorical_pipeline_grid,
            categorical_features_grid
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 5. SELECTIVE PARAMETER GRIDS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("SELECTIVE PARAMETER GRID CONFIGURATION")
logger.info("-" * 70)


# ============================================================
# RANDOM FOREST
# Best Randomized Search:
# n_estimators=200
# min_samples_split=10
# min_samples_leaf=4
# max_features=log2
# max_depth=10
# criterion=gini
# ============================================================

rf_grid = {
    "model__n_estimators": [150, 200],
    "model__max_depth": [10, 12],
    "model__min_samples_split": [8, 10],
    "model__min_samples_leaf": [4],
    "model__max_features": ["log2"]
}


# 2 × 2 × 2 = 8 combinations
# 8 × 5 folds = 40 fits


# ============================================================
# CATBOOST
# Best Randomized Search:
# iterations=400
# depth=4
# learning_rate=0.1
# l2_leaf_reg=7
# random_strength=1.0
# ============================================================

catboost_grid = {
    "model__iterations": [300, 400],
    "model__depth": [4, 5],
    "model__learning_rate": [0.1],
    "model__l2_leaf_reg": [7],
    "model__random_strength": [1.0]
}


# 2 × 2 = 4 combinations
# 4 × 5 folds = 20 fits


# ============================================================
# LIGHTGBM
# Best Randomized Search:
# n_estimators=500
# learning_rate=0.03
# num_leaves=70
# max_depth=10
# min_child_samples=5
# subsample=0.7
# colsample_bytree=0.7
# ============================================================

lightgbm_grid = {
    "model__n_estimators": [400, 500],
    "model__learning_rate": [0.03],
    "model__num_leaves": [50, 70],
    "model__max_depth": [10],
    "model__min_child_samples": [5],
    "model__subsample": [0.7],
    "model__colsample_bytree": [0.7]
}


# 2 × 2 = 4 combinations
# 4 × 5 folds = 20 fits


logger.info("Random Forest Grid Combinations           : 8")
logger.info("Random Forest Total CV Fits               : 40")

logger.info("CatBoost Grid Combinations                 : 4")
logger.info("CatBoost Total CV Fits                    : 20")

logger.info("LightGBM Grid Combinations                 : 4")
logger.info("LightGBM Total CV Fits                    : 20")

logger.info("Total Selective Grid CV Fits              : 80")


# ------------------------------------------------------------
# 6. MODEL PIPELINES
# ------------------------------------------------------------

rf_pipeline_grid = Pipeline(
    steps=[
        ("preprocessor", preprocessor_grid),
        (
            "model",
            RandomForestClassifier(
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


catboost_pipeline_grid = Pipeline(
    steps=[
        ("preprocessor", preprocessor_grid),
        (
            "model",
            CatBoostClassifier(
                loss_function="MultiClass",
                eval_metric="MultiClass",
                verbose=False,
                random_seed=42,
                thread_count=1
            )
        )
    ]
)


lightgbm_pipeline_grid = Pipeline(
    steps=[
        ("preprocessor", preprocessor_grid),
        (
            "model",
            LGBMClassifier(
                objective="multiclass",
                class_weight="balanced",
                random_state=42,
                n_jobs=-1,
                verbosity=-1
            )
        )
    ]
)


# ------------------------------------------------------------
# 7. GRID SEARCH FUNCTION
# ------------------------------------------------------------

def run_selective_grid_search(
    model_name,
    pipeline,
    parameter_grid
):

    logger.info("-" * 70)
    logger.info(
        f"STARTING {model_name.upper()} SELECTIVE GRID SEARCH"
    )
    logger.info("-" * 70)

    total_combinations = 1

    for values in parameter_grid.values():
        total_combinations *= len(values)

    logger.info(
        f"Parameter Combinations                   : "
        f"{total_combinations}"
    )

    logger.info(
        f"Total CV Fits                            : "
        f"{total_combinations * optimization_cv.n_splits}"
    )

    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=parameter_grid,
        scoring=optimization_scoring,
        refit="macro_f1",
        cv=optimization_cv,
        n_jobs=1,
        return_train_score=True,
        error_score="raise"
    )

    grid_search.fit(
        X_train_grid,
        y_train_grid
    )

    best_index = grid_search.best_index_

    cv_results = pd.DataFrame(
        grid_search.cv_results_
    )

    result = {
        "Model": model_name,
        "Best Macro F1": grid_search.cv_results_[
            "mean_test_macro_f1"
        ][best_index],
        "Best Balanced Accuracy": grid_search.cv_results_[
            "mean_test_balanced_accuracy"
        ][best_index],
        "Best Accuracy": grid_search.cv_results_[
            "mean_test_accuracy"
        ][best_index],
        "Best Macro Precision": grid_search.cv_results_[
            "mean_test_macro_precision"
        ][best_index],
        "Best Macro Recall": grid_search.cv_results_[
            "mean_test_macro_recall"
        ][best_index],
        "Training Macro F1": grid_search.cv_results_[
            "mean_train_macro_f1"
        ][best_index],
        "Best Parameters": grid_search.best_params_,
        "CV Results": cv_results,
        "Search Object": grid_search
    }

    logger.info(
        f"{model_name} Best Macro F1                 : "
        f"{result['Best Macro F1']:.4f}"
    )

    logger.info(
        f"{model_name} Best Balanced Accuracy        : "
        f"{result['Best Balanced Accuracy']:.4f}"
    )

    logger.info(
        f"{model_name} Best Parameters               : "
        f"{result['Best Parameters']}"
    )

    return result


# ------------------------------------------------------------
# 8. RANDOM FOREST GRID SEARCH
# ------------------------------------------------------------

rf_grid_result = run_selective_grid_search(
    "Random Forest",
    rf_pipeline_grid,
    rf_grid
)


# ------------------------------------------------------------
# 9. CATBOOST GRID SEARCH
# ------------------------------------------------------------

catboost_grid_result = run_selective_grid_search(
    "CatBoost",
    catboost_pipeline_grid,
    catboost_grid
)


# ------------------------------------------------------------
# 10. LIGHTGBM GRID SEARCH
# ------------------------------------------------------------

lightgbm_grid_result = run_selective_grid_search(
    "LightGBM",
    lightgbm_pipeline_grid,
    lightgbm_grid
)


# ------------------------------------------------------------
# 11. STORE RESULTS
# ------------------------------------------------------------

selective_grid_results = {
    "Random Forest": rf_grid_result,
    "CatBoost": catboost_grid_result,
    "LightGBM": lightgbm_grid_result
}


# ------------------------------------------------------------
# 12. PERFORMANCE SUMMARY
# ------------------------------------------------------------

grid_summary = pd.DataFrame([
    {
        "Model": result["Model"],
        "Best Macro F1": result["Best Macro F1"],
        "Best Balanced Accuracy":
            result["Best Balanced Accuracy"],
        "Best Accuracy":
            result["Best Accuracy"],
        "Best Macro Precision":
            result["Best Macro Precision"],
        "Best Macro Recall":
            result["Best Macro Recall"],
        "Training Macro F1":
            result["Training Macro F1"],
        "Best Parameters":
            result["Best Parameters"]
    }
    for result in selective_grid_results.values()
])

grid_summary = grid_summary.sort_values(
    by="Best Macro F1",
    ascending=False
).reset_index(drop=True)


logger.info("-" * 70)
logger.info("SELECTIVE GRID SEARCH PERFORMANCE SUMMARY")
logger.info("-" * 70)

display(grid_summary)


# ------------------------------------------------------------
# 13. BEST GRID SEARCH MODEL
# ------------------------------------------------------------

best_grid_row = grid_summary.iloc[0]

best_grid_model_name = best_grid_row["Model"]

best_grid_macro_f1 = best_grid_row["Best Macro F1"]

best_grid_search = selective_grid_results[
    best_grid_model_name
]["Search Object"]

best_grid_parameters = selective_grid_results[
    best_grid_model_name
]["Best Parameters"]


logger.info(
    f"Best Selective Grid Search Model          : "
    f"{best_grid_model_name}"
)

logger.info(
    f"Best Selective Grid Search Macro F1       : "
    f"{best_grid_macro_f1:.4f}"
)

logger.info(
    f"Best Selective Grid Search Parameters     : "
    f"{best_grid_parameters}"
)


# ------------------------------------------------------------
# 14. VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.5 SELECTIVE GRID SEARCH VALIDATION")
logger.info("-" * 70)

assert len(grid_summary) == 3

assert set(grid_summary["Model"]) == {
    "Random Forest",
    "CatBoost",
    "LightGBM"
}

assert len(X_train_grid) == 292
assert len(X_test_grid) == 74

assert best_grid_macro_f1 > 0

logger.info("Training Data Isolation                   : PASSED")
logger.info("Independent Test Set Isolation             : PASSED")
logger.info("Five-Fold Stratified CV                    : PASSED")
logger.info("Random Forest Grid Search                 : PASSED")
logger.info("CatBoost Grid Search                      : PASSED")
logger.info("LightGBM Grid Search                      : PASSED")
logger.info("Selective Hyperparameter Refinement        : PASSED")
logger.info("Primary Metric Macro F1                   : PASSED")
logger.info("Preprocessing Leakage Prevention           : PASSED")

logger.info("=" * 70)
logger.info("11.5 SELECTIVE GRID SEARCH : COMPLETED")
logger.info("=" * 70)
````

## 11.7 Optimized Model Training

This stage trains the final optimized Random Forest model using the
hyperparameters identified through Randomized Search and Selective Grid
Search.

The model is trained on the complete training dataset while the
independent test dataset remains completely isolated.

### Final Model Configuration

- Model: Random Forest
- Number of Estimators: 150
- Maximum Depth: 10
- Maximum Features: log2
- Minimum Samples per Leaf: 4
- Minimum Samples for Split: 8
- Criterion: Gini
- Class Weight: Balanced
- Random State: 42

### Preprocessing

- Age: Numeric conversion with invalid markers converted to missing values
- Age Missing Values: Median Imputation
- Categorical Features: One-Hot Encoding
- Unknown Categories: Ignored
- Ordinal Encoding: Not Used
- Preprocessing: Integrated into the model pipeline

The optimized pipeline will be retained for subsequent evaluation,
serialization, explainability, and deployment stages.

````python
# ============================================================
# REQUIRED LIBRARIES
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier




logger.info("=" * 70)
logger.info("11.7 OPTIMIZED MODEL TRAINING")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. INPUT DATA VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("INPUT DATA VALIDATION")
logger.info("-" * 70)

X_train_final = X_train_optimization.copy(deep=True)
y_train_final = y_train_optimization.copy(deep=True)

X_test_final = X_test_optimization.copy(deep=True)
y_test_final = y_test_optimization.copy(deep=True)

logger.info(
    f"Training Samples                         : "
    f"{X_train_final.shape[0]}"
)

logger.info(
    f"Training Features                        : "
    f"{X_train_final.shape[1]}"
)

logger.info(
    f"Independent Test Samples                 : "
    f"{X_test_final.shape[0]}"
)

logger.info(
    f"Independent Test Features                : "
    f"{X_test_final.shape[1]}"
)


# ------------------------------------------------------------
# 2. AGE DATA-TYPE CORRECTION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGE DATA-TYPE CORRECTION")
logger.info("-" * 70)

X_train_final["Age"] = pd.to_numeric(
    X_train_final["Age"],
    errors="coerce"
)

X_test_final["Age"] = pd.to_numeric(
    X_test_final["Age"],
    errors="coerce"
)

logger.info(
    f"Training Age Missing Values              : "
    f"{X_train_final['Age'].isna().sum()}"
)

logger.info(
    f"Test Age Missing Values                  : "
    f"{X_test_final['Age'].isna().sum()}"
)

logger.info("Age Conversion                           : COMPLETED")


# ------------------------------------------------------------
# 3. FEATURE CONFIGURATION
# ------------------------------------------------------------

numeric_features_final = ["Age"]

categorical_features_final = [
    column
    for column in X_train_final.columns
    if column != "Age"
]

logger.info("-" * 70)
logger.info("FEATURE CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Numeric Features                         : "
    f"{len(numeric_features_final)}"
)

logger.info(
    f"Categorical Features                     : "
    f"{len(categorical_features_final)}"
)

logger.info("Categorical Encoding                     : One-Hot Encoding")
logger.info("Ordinal Encoding                         : NOT USED")


# ------------------------------------------------------------
# 4. PREPROCESSING PIPELINE
# ------------------------------------------------------------

numeric_pipeline_final = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline_final = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor_final = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline_final,
            numeric_features_final
        ),
        (
            "categorical",
            categorical_pipeline_final,
            categorical_features_final
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 5. FINAL OPTIMIZED RANDOM FOREST
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL OPTIMIZED RANDOM FOREST CONFIGURATION")
logger.info("-" * 70)

optimized_random_forest = RandomForestClassifier(
    n_estimators=150,
    max_depth=10,
    max_features="log2",
    min_samples_leaf=4,
    min_samples_split=8,
    criterion="gini",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

logger.info("n_estimators                            : 150")
logger.info("max_depth                               : 10")
logger.info("max_features                            : log2")
logger.info("min_samples_leaf                        : 4")
logger.info("min_samples_split                       : 8")
logger.info("criterion                               : gini")
logger.info("class_weight                            : balanced")
logger.info("random_state                             : 42")


# ------------------------------------------------------------
# 6. CREATE FINAL PRODUCTION PIPELINE
# ------------------------------------------------------------

optimized_model_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor_final
        ),
        (
            "model",
            optimized_random_forest
        )
    ]
)


# ------------------------------------------------------------
# 7. TRAIN FINAL OPTIMIZED MODEL
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING FINAL OPTIMIZED MODEL")
logger.info("-" * 70)

optimized_model_pipeline.fit(
    X_train_final,
    y_train_final
)

logger.info("Optimized Model Training                  : COMPLETED")


# ------------------------------------------------------------
# 8. TRANSFORMED FEATURE COUNT
# ------------------------------------------------------------

transformed_feature_count = (
    optimized_model_pipeline
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

logger.info("-" * 70)
logger.info("TRANSFORMED FEATURE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Original Feature Count                    : "
    f"{X_train_final.shape[1]}"
)

logger.info(
    f"Model-Ready Encoded Feature Count         : "
    f"{len(transformed_feature_count)}"
)


# ------------------------------------------------------------
# 9. TRAINING PREDICTIONS
# ------------------------------------------------------------

train_predictions_optimized = (
    optimized_model_pipeline.predict(
        X_train_final
    )
)

train_probabilities_optimized = (
    optimized_model_pipeline.predict_proba(
        X_train_final
    )
)


# ------------------------------------------------------------
# 10. TRAINING PERFORMANCE
# ------------------------------------------------------------

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

optimized_train_accuracy = accuracy_score(
    y_train_final,
    train_predictions_optimized
)

optimized_train_balanced_accuracy = balanced_accuracy_score(
    y_train_final,
    train_predictions_optimized
)

optimized_train_macro_precision = precision_score(
    y_train_final,
    train_predictions_optimized,
    average="macro",
    zero_division=0
)

optimized_train_macro_recall = recall_score(
    y_train_final,
    train_predictions_optimized,
    average="macro",
    zero_division=0
)

optimized_train_macro_f1 = f1_score(
    y_train_final,
    train_predictions_optimized,
    average="macro",
    zero_division=0
)

optimized_train_weighted_f1 = f1_score(
    y_train_final,
    train_predictions_optimized,
    average="weighted",
    zero_division=0
)


# ------------------------------------------------------------
# 11. DISPLAY TRAINING PERFORMANCE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("OPTIMIZED MODEL TRAINING PERFORMANCE")
logger.info("-" * 70)

logger.info(
    f"Training Accuracy                        : "
    f"{optimized_train_accuracy:.4f}"
)

logger.info(
    f"Training Balanced Accuracy               : "
    f"{optimized_train_balanced_accuracy:.4f}"
)

logger.info(
    f"Training Macro Precision                 : "
    f"{optimized_train_macro_precision:.4f}"
)

logger.info(
    f"Training Macro Recall                    : "
    f"{optimized_train_macro_recall:.4f}"
)

logger.info(
    f"Training Macro F1                        : "
    f"{optimized_train_macro_f1:.4f}"
)

logger.info(
    f"Training Weighted F1                     : "
    f"{optimized_train_weighted_f1:.4f}"
)


# ------------------------------------------------------------
# 12. TRAINING SUMMARY TABLE
# ------------------------------------------------------------

optimized_training_summary = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1"
    ],
    "Training Score": [
        optimized_train_accuracy,
        optimized_train_balanced_accuracy,
        optimized_train_macro_precision,
        optimized_train_macro_recall,
        optimized_train_macro_f1,
        optimized_train_weighted_f1
    ]
})

display(
    optimized_training_summary
)


# ------------------------------------------------------------
# 13. VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("11.7 OPTIMIZED MODEL TRAINING VALIDATION")
logger.info("-" * 70)

assert len(X_train_final) == 292
assert len(X_test_final) == 74

assert optimized_model_pipeline is not None

assert len(transformed_feature_count) == 130

assert len(train_predictions_optimized) == len(y_train_final)

assert train_probabilities_optimized.shape[0] == len(
    y_train_final
)

assert train_probabilities_optimized.shape[1] == 6

assert np.isfinite(
    optimized_train_macro_f1
)

logger.info("Training Dataset                         : PASSED")
logger.info("Independent Test Set Isolation            : PASSED")
logger.info("Final Hyperparameters Applied              : PASSED")
logger.info("Pipeline-Based Preprocessing              : PASSED")
logger.info("One-Hot Feature Transformation             : PASSED")
logger.info("Encoded Feature Count                     : PASSED")
logger.info("Multiclass Probability Output              : PASSED")
logger.info("Training Predictions                      : PASSED")
logger.info("Training Metrics                          : PASSED")

logger.info("=" * 70)
logger.info("11.7 OPTIMIZED MODEL TRAINING : COMPLETED")
logger.info("=" * 70)
````

## 12.2 Performance Metrics

The finalized optimized Random Forest model is evaluated using comprehensive multiclass classification metrics on the independent test set.

The evaluation includes:

- Accuracy
- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1
- Weighted Precision
- Weighted Recall
- Weighted F1

Accuracy measures the overall proportion of correctly classified samples.

Balanced Accuracy calculates the average recall across all classes and is particularly useful when class frequencies are not equal.

Macro-averaged metrics calculate the metric independently for each class and then take the unweighted average. This gives equal importance to every dermatological disorder class.

Weighted-averaged metrics calculate the metric for each class and weight the result according to the number of samples in that class.

Macro F1 is treated as the primary test-set performance metric because this project contains six classes with unequal class frequencies.

All metrics are calculated exclusively from the independent test predictions generated in Section 12.1.

````python

# Required Libraries


import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# Performance Metric Calculation
# ============================================================

logger.info("-" * 70)
logger.info("12.2 PERFORMANCE METRICS")
logger.info("-" * 70)

# ------------------------------------------------------------
# Validate required prediction objects
# ------------------------------------------------------------

required_metric_objects = [
    "y_test_final",
    "y_test_pred"
]

missing_metric_objects = [
    obj for obj in required_metric_objects
    if obj not in globals()
]

if missing_metric_objects:
    raise NameError(
        f"Required metric objects are missing: "
        f"{missing_metric_objects}"
    )

# ------------------------------------------------------------
# Calculate comprehensive multiclass metrics
# ------------------------------------------------------------

metric_results = {
    "Accuracy": accuracy_score(
        y_test_final,
        y_test_pred
    ),

    "Balanced Accuracy": balanced_accuracy_score(
        y_test_final,
        y_test_pred
    ),

    "Macro Precision": precision_score(
        y_test_final,
        y_test_pred,
        average="macro",
        zero_division=0
    ),

    "Macro Recall": recall_score(
        y_test_final,
        y_test_pred,
        average="macro",
        zero_division=0
    ),

    "Macro F1": f1_score(
        y_test_final,
        y_test_pred,
        average="macro",
        zero_division=0
    ),

    "Weighted Precision": precision_score(
        y_test_final,
        y_test_pred,
        average="weighted",
        zero_division=0
    ),

    "Weighted Recall": recall_score(
        y_test_final,
        y_test_pred,
        average="weighted",
        zero_division=0
    ),

    "Weighted F1": f1_score(
        y_test_final,
        y_test_pred,
        average="weighted",
        zero_division=0
    )
}

# ------------------------------------------------------------
# Create performance table
# ------------------------------------------------------------

performance_metrics_table = pd.DataFrame({
    "Metric": list(metric_results.keys()),
    "Score": list(metric_results.values()),
    "Percentage": [
        f"{score * 100:.2f}%"
        for score in metric_results.values()
    ]
})

display(performance_metrics_table)

# ------------------------------------------------------------
# Store individual metrics
# ------------------------------------------------------------

test_accuracy = metric_results["Accuracy"]
test_balanced_accuracy = metric_results["Balanced Accuracy"]

test_macro_precision = metric_results["Macro Precision"]
test_macro_recall = metric_results["Macro Recall"]
test_macro_f1 = metric_results["Macro F1"]

test_weighted_precision = metric_results["Weighted Precision"]
test_weighted_recall = metric_results["Weighted Recall"]
test_weighted_f1 = metric_results["Weighted F1"]

# ------------------------------------------------------------
# Logging
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TEST SET PERFORMANCE")
logger.info("-" * 70)

for metric_name, metric_value in metric_results.items():
    logger.info(
        f"{metric_name:<35} : "
        f"{metric_value:.4f} "
        f"({metric_value * 100:.2f}%)"
    )

# ------------------------------------------------------------
# Primary evaluation metric
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRIMARY EVALUATION METRIC")
logger.info("-" * 70)

logger.info(
    "Primary Test Metric                   : Macro F1"
)

logger.info(
    f"Macro F1 Score                        : "
    f"{test_macro_f1:.4f}"
)

logger.info(
    f"Macro F1 Percentage                   : "
    f"{test_macro_f1 * 100:.2f}%"
)

# ============================================================
# CV VS TEST COMPARISON
# Robustly identify Macro F1 key
# ============================================================

if "optimized_cv_mean_scores" in globals():

    cv_scores = optimized_cv_mean_scores

    # --------------------------------------------------------
    # Safely locate Macro F1
    # --------------------------------------------------------

    macro_f1_candidates = [
        key for key in cv_scores.keys()
        if "f1" in str(key).lower()
        and "macro" in str(key).lower()
    ]

    if len(macro_f1_candidates) == 0:

        logger.warning(
            "Macro F1 key was not found automatically in "
            "optimized_cv_mean_scores."
        )

        logger.info(
            f"Available CV score keys: "
            f"{list(cv_scores.keys())}"
        )

        cv_macro_f1 = None

    else:

        cv_macro_f1_key = macro_f1_candidates[0]
        cv_macro_f1 = float(
            cv_scores[cv_macro_f1_key]
        )

        logger.info(
            f"Detected CV Macro F1 Key            : "
            f"{cv_macro_f1_key}"
        )

    # --------------------------------------------------------
    # Calculate generalization gap
    # --------------------------------------------------------

    if cv_macro_f1 is not None:

        generalization_gap = (
            cv_macro_f1 - test_macro_f1
        )

        logger.info("-" * 70)
        logger.info("CROSS-VALIDATION VS TEST PERFORMANCE")
        logger.info("-" * 70)

        logger.info(
            f"Mean CV Macro F1                     : "
            f"{cv_macro_f1:.4f}"
        )

        logger.info(
            f"Independent Test Macro F1            : "
            f"{test_macro_f1:.4f}"
        )

        logger.info(
            f"CV-to-Test Macro F1 Gap              : "
            f"{generalization_gap:.4f}"
        )

        logger.info(
            f"CV-to-Test Gap Percentage Points     : "
            f"{generalization_gap * 100:.2f}"
        )

        # Store for later evaluation summary
        test_cv_macro_f1 = cv_macro_f1
        test_generalization_gap = generalization_gap

else:

    logger.warning(
        "optimized_cv_mean_scores is unavailable. "
        "CV-to-test comparison skipped."
    )

    test_cv_macro_f1 = None
    test_generalization_gap = None

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.2 PERFORMANCE METRICS VALIDATION")
logger.info("-" * 70)

metric_validation = {
    "Accuracy Calculated": test_accuracy,
    "Balanced Accuracy Calculated": test_balanced_accuracy,
    "Macro Precision Calculated": test_macro_precision,
    "Macro Recall Calculated": test_macro_recall,
    "Macro F1 Calculated": test_macro_f1,
    "Weighted Precision Calculated": test_weighted_precision,
    "Weighted Recall Calculated": test_weighted_recall,
    "Weighted F1 Calculated": test_weighted_f1
}

for metric_name, metric_value in metric_validation.items():

    if not np.isfinite(metric_value):
        raise ValueError(
            f"Invalid metric value detected: "
            f"{metric_name}"
        )

    if not 0 <= metric_value <= 1:
        raise ValueError(
            f"Metric outside valid range [0,1]: "
            f"{metric_name}"
        )

    logger.info(
        f"{metric_name:<40} : PASSED"
    )

logger.info(
    "Independent Test Set Used              : PASSED"
)

logger.info(
    "Multiclass Metric Configuration        : PASSED"
)

logger.info("=" * 70)
logger.info("12.2 PERFORMANCE METRICS : COMPLETED")
logger.info("=" * 70)
````

## 12.3 Per-Class Performance

Per-class performance is evaluated to determine how effectively the finalized Random Forest model identifies each dermatological disorder class independently.

The analysis includes:

- Precision for each class
- Recall for each class
- F1-score for each class
- Number of test samples (support) for each class

Per-class evaluation is important because the dataset contains six classes with unequal sample frequencies.

Precision measures how many samples predicted as a particular class actually belong to that class.

Recall measures how many samples belonging to a particular class are correctly identified by the model.

F1-score provides a harmonic balance between precision and recall.

Support represents the number of actual test samples belonging to each class.

The analysis uses only predictions generated from the independent test set.

````python

# Required Libraries


import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_recall_fscore_support
)

# ============================================================
# Per-Class Metric Calculation
# ============================================================

logger.info("-" * 70)
logger.info("12.3 PER-CLASS PERFORMANCE")
logger.info("-" * 70)

# ------------------------------------------------------------
# Validate required objects
# ------------------------------------------------------------

required_per_class_objects = [
    "y_test_final",
    "y_test_pred"
]

missing_per_class_objects = [
    obj for obj in required_per_class_objects
    if obj not in globals()
]

if missing_per_class_objects:
    raise NameError(
        f"Required per-class objects are missing: "
        f"{missing_per_class_objects}"
    )

# ------------------------------------------------------------
# Determine complete class set
# ------------------------------------------------------------

per_class_labels = np.sort(
    np.unique(
        np.concatenate([
            np.asarray(y_test_final),
            np.asarray(y_test_pred)
        ])
    )
)

# ------------------------------------------------------------
# Calculate per-class metrics
# ------------------------------------------------------------

precision_values, recall_values, f1_values, support_values = (
    precision_recall_fscore_support(
        y_test_final,
        y_test_pred,
        labels=per_class_labels,
        average=None,
        zero_division=0
    )
)

# ------------------------------------------------------------
# Create per-class performance table
# ------------------------------------------------------------

per_class_performance = pd.DataFrame({
    "Class": per_class_labels,
    "Precision": precision_values,
    "Recall": recall_values,
    "F1-Score": f1_values,
    "Support": support_values
})

# ------------------------------------------------------------
# Percentage columns
# ------------------------------------------------------------

per_class_performance["Precision (%)"] = (
    per_class_performance["Precision"] * 100
)

per_class_performance["Recall (%)"] = (
    per_class_performance["Recall"] * 100
)

per_class_performance["F1-Score (%)"] = (
    per_class_performance["F1-Score"] * 100
)

# ------------------------------------------------------------
# Display formatted table
# ------------------------------------------------------------

display(
    per_class_performance.style.format({
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "Support": "{:.0f}",
        "Precision (%)": "{:.2f}%",
        "Recall (%)": "{:.2f}%",
        "F1-Score (%)": "{:.2f}%"
    })
)

# ------------------------------------------------------------
# Store for later sections
# ------------------------------------------------------------

final_per_class_performance = per_class_performance.copy()

# ------------------------------------------------------------
# Logging
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS PERFORMANCE")
logger.info("-" * 70)

for _, row in per_class_performance.iterrows():

    logger.info(
        f"Class {int(row['Class'])} | "
        f"Precision: {row['Precision']:.4f} | "
        f"Recall: {row['Recall']:.4f} | "
        f"F1: {row['F1-Score']:.4f} | "
        f"Support: {int(row['Support'])}"
    )

# ------------------------------------------------------------
# Best and lowest performing classes
# ------------------------------------------------------------

best_f1_index = (
    per_class_performance["F1-Score"].idxmax()
)

lowest_f1_index = (
    per_class_performance["F1-Score"].idxmin()
)

best_class = per_class_performance.loc[
    best_f1_index,
    "Class"
]

lowest_class = per_class_performance.loc[
    lowest_f1_index,
    "Class"
]

best_class_f1 = per_class_performance.loc[
    best_f1_index,
    "F1-Score"
]

lowest_class_f1 = per_class_performance.loc[
    lowest_f1_index,
    "F1-Score"
]

logger.info("-" * 70)
logger.info("PER-CLASS F1 ANALYSIS")
logger.info("-" * 70)

logger.info(
    f"Best Performing Class                : "
    f"{int(best_class)}"
)

logger.info(
    f"Best Class F1-Score                  : "
    f"{best_class_f1:.4f}"
)

logger.info(
    f"Lowest Performing Class              : "
    f"{int(lowest_class)}"
)

logger.info(
    f"Lowest Class F1-Score                : "
    f"{lowest_class_f1:.4f}"
)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.3 PER-CLASS PERFORMANCE VALIDATION")
logger.info("-" * 70)

if len(per_class_performance) != len(per_class_labels):
    raise ValueError(
        "Per-class table does not contain all class labels."
    )

if per_class_performance["Support"].sum() != len(y_test_final):
    raise ValueError(
        "Per-class support does not equal test sample count."
    )

for metric_column in [
    "Precision",
    "Recall",
    "F1-Score"
]:

    if not np.all(
        np.isfinite(
            per_class_performance[metric_column].values
        )
    ):
        raise ValueError(
            f"Invalid values detected in {metric_column}."
        )

    if not np.all(
        (
            per_class_performance[metric_column].values >= 0
        )
        &
        (
            per_class_performance[metric_column].values <= 1
        )
    ):
        raise ValueError(
            f"{metric_column} contains values outside [0,1]."
        )

logger.info(
    "All Six Classes Evaluated             : PASSED"
)

logger.info(
    "Precision / Recall / F1 Valid         : PASSED"
)

logger.info(
    "Support Count Validation              : PASSED"
)

logger.info(
    "Independent Test Set Only             : PASSED"
)

logger.info("=" * 70)
logger.info("12.3 PER-CLASS PERFORMANCE : COMPLETED")
logger.info("=" * 70)
````

## 12.4 Confusion Matrix

The confusion matrix provides a class-by-class visualization of the predictions generated by the finalized Random Forest model on the independent test set.

Rows represent the actual dermatological disorder classes, while columns represent the classes predicted by the model.

The diagonal elements represent correctly classified samples.

Off-diagonal elements represent misclassified samples and identify specific class-level confusion patterns.

The confusion matrix is particularly useful for:
- Identifying classes with perfect classification
- Detecting systematic class confusion
- Understanding the source of classification errors
- Supporting the per-class precision, recall, and F1-score analysis

The matrix is calculated exclusively from the independent test-set predictions.

````python

# Required Libraries


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix

# ============================================================
# Confusion Matrix Calculation
# ============================================================

logger.info("-" * 70)
logger.info("12.4 CONFUSION MATRIX")
logger.info("-" * 70)

# ------------------------------------------------------------
# Validate required objects
# ------------------------------------------------------------

required_cm_objects = [
    "y_test_final",
    "y_test_pred"
]

missing_cm_objects = [
    obj
    for obj in required_cm_objects
    if obj not in globals()
]

if missing_cm_objects:
    raise NameError(
        f"Required confusion matrix objects are missing: "
        f"{missing_cm_objects}"
    )

# ------------------------------------------------------------
# Determine class labels
# ------------------------------------------------------------

confusion_labels = np.sort(
    np.unique(
        np.concatenate([
            np.asarray(y_test_final),
            np.asarray(y_test_pred)
        ])
    )
)

# ------------------------------------------------------------
# Calculate confusion matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test_final,
    y_test_pred,
    labels=confusion_labels
)

# ------------------------------------------------------------
# Store confusion matrix
# ------------------------------------------------------------

confusion_matrix_df = pd.DataFrame(
    cm,
    index=[
        f"Actual {label}"
        for label in confusion_labels
    ],
    columns=[
        f"Predicted {label}"
        for label in confusion_labels
    ]
)

display(confusion_matrix_df)

# ------------------------------------------------------------
# Logging
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("CONFUSION MATRIX VALUES")
logger.info("-" * 70)

for i, actual_class in enumerate(confusion_labels):

    row_values = cm[i]

    logger.info(
        f"Actual Class {actual_class} : "
        f"{row_values.tolist()}"
    )

# ------------------------------------------------------------
# Correct and incorrect predictions
# ------------------------------------------------------------

correct_predictions = np.trace(cm)

total_predictions = cm.sum()

incorrect_predictions = (
    total_predictions - correct_predictions
)

logger.info("-" * 70)
logger.info("CONFUSION MATRIX SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Test Samples                  : "
    f"{total_predictions}"
)

logger.info(
    f"Correct Predictions                 : "
    f"{correct_predictions}"
)

logger.info(
    f"Incorrect Predictions               : "
    f"{incorrect_predictions}"
)

# ------------------------------------------------------------
# Identify misclassification pairs
# ------------------------------------------------------------

misclassification_records = []

for i, actual_class in enumerate(confusion_labels):

    for j, predicted_class in enumerate(confusion_labels):

        if i != j and cm[i, j] > 0:

            misclassification_records.append({
                "Actual Class": actual_class,
                "Predicted Class": predicted_class,
                "Count": cm[i, j]
            })

if misclassification_records:

    misclassification_df = pd.DataFrame(
        misclassification_records
    )

    logger.info("-" * 70)
    logger.info("MISCLASSIFICATION PATTERNS")
    logger.info("-" * 70)

    display(misclassification_df)

else:

    misclassification_df = pd.DataFrame(
        columns=[
            "Actual Class",
            "Predicted Class",
            "Count"
        ]
    )

    logger.info(
        "No off-diagonal misclassifications detected."
    )

# ============================================================
# Confusion Matrix Visualization
# ============================================================

plt.figure(figsize=(8, 6))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Confusion Matrix - Random Forest"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    range(len(confusion_labels)),
    confusion_labels
)

plt.yticks(
    range(len(confusion_labels)),
    confusion_labels
)

# ------------------------------------------------------------
# Add cell values
# ------------------------------------------------------------

threshold = cm.max() / 2.0

for i in range(cm.shape[0]):

    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            str(cm[i, j]),
            ha="center",
            va="center",
            color="white" if cm[i, j] > threshold else "black"
        )

plt.colorbar(
    label="Number of Samples"
)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# Store for later evaluation summary
# ------------------------------------------------------------

final_confusion_matrix = cm.copy()
final_misclassification_records = misclassification_records.copy()

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.4 CONFUSION MATRIX VALIDATION")
logger.info("-" * 70)

if cm.shape != (
    len(confusion_labels),
    len(confusion_labels)
):
    raise ValueError(
        "Confusion matrix dimensions do not match "
        "the number of classes."
    )

if cm.sum() != len(y_test_final):
    raise ValueError(
        "Confusion matrix total does not match "
        "test sample count."
    )

if np.trace(cm) != correct_predictions:
    raise ValueError(
        "Diagonal count does not match correct predictions."
    )

if incorrect_predictions != (
    len(y_test_final) - np.trace(cm)
):
    raise ValueError(
        "Incorrect prediction count validation failed."
    )

logger.info(
    "Six-Class Confusion Matrix             : PASSED"
)

logger.info(
    "Matrix Dimension Validation             : PASSED"
)

logger.info(
    "Test Sample Count Validation            : PASSED"
)

logger.info(
    "Correct Prediction Count                : PASSED"
)

logger.info(
    "Misclassification Analysis              : PASSED"
)

logger.info(
    "Independent Test Set Only               : PASSED"
)

logger.info("=" * 70)
logger.info("12.4 CONFUSION MATRIX : COMPLETED")
logger.info("=" * 70)
````

## 12.6 Multiclass ROC-AUC

### Objective

Evaluate the final optimized Random Forest model using the **Multiclass ROC-AUC** metric.

Since the Skin Disorder Prediction problem contains six target classes, ROC-AUC will be calculated using a **One-vs-Rest (OvR)** strategy.

The analysis will:

- Use probability predictions generated on the independent test set.
- Calculate the overall multiclass ROC-AUC using the OvR strategy.
- Calculate per-class ROC-AUC values for all six classes.
- Evaluate whether the model maintains strong class discrimination beyond accuracy-based metrics.
- Use the independent test set only for final evaluation.
- Avoid using test-set information during model training, feature selection, or hyperparameter optimization.

### Evaluation Standard

The primary ROC-AUC value will be reported as **macro-average ROC-AUC**, giving equal importance to each disease class regardless of class frequency.

Per-class ROC-AUC values will also be reported to identify classes with comparatively weaker discrimination.

> **Important:** ROC-AUC is being used as a multiclass discrimination metric and should be interpreted together with Macro F1, Balanced Accuracy, and the Confusion Matrix.

````python
# ==============================================================
# Required Libraries 
# ==============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import label_binarize

from IPython.display import display
````

````python


logger.info("=" * 70)
logger.info("12.6 MULTICLASS ROC-AUC")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Validate Required Evaluation Objects
# --------------------------------------------------------------

required_objects = [
    "y_test_final",
    "y_test_proba"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        f"Required evaluation objects are missing: {missing_objects}"
    )

# --------------------------------------------------------------
# 2. Prepare Evaluation Data
# --------------------------------------------------------------

roc_auc_classes = np.array(
    sorted(np.unique(y_test_final))
)

y_test_final_array = np.asarray(y_test_final)
y_test_proba_array = np.asarray(y_test_proba)

logger.info(
    "Independent Test Samples          : %d",
    len(y_test_final_array)
)

logger.info(
    "Number of Target Classes          : %d",
    len(roc_auc_classes)
)

logger.info(
    "Target Classes                    : %s",
    roc_auc_classes.tolist()
)

# --------------------------------------------------------------
# 3. Validate Probability Matrix
# --------------------------------------------------------------

if y_test_proba_array.ndim != 2:
    raise ValueError(
        "Test probability predictions must be a 2-dimensional matrix."
    )

if y_test_proba_array.shape[0] != len(y_test_final_array):
    raise ValueError(
        "Number of probability rows does not match test samples."
    )

if y_test_proba_array.shape[1] != len(roc_auc_classes):
    raise ValueError(
        "Number of probability columns does not match the number of classes."
    )

# --------------------------------------------------------------
# 4. Validate Probability Values
# --------------------------------------------------------------

if not np.all(np.isfinite(y_test_proba_array)):
    raise ValueError(
        "Probability predictions contain non-finite values."
    )

if not np.allclose(
    y_test_proba_array.sum(axis=1),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "Probability predictions do not sum to 1 for all samples."
    )

# --------------------------------------------------------------
# 5. Binarize Multiclass Targets
# --------------------------------------------------------------

y_test_binarized = label_binarize(
    y_test_final_array,
    classes=roc_auc_classes
)

# --------------------------------------------------------------
# 6. Calculate Macro Multiclass ROC-AUC
# --------------------------------------------------------------

multiclass_roc_auc_macro = roc_auc_score(
    y_test_binarized,
    y_test_proba_array,
    average="macro",
    multi_class="ovr"
)

# --------------------------------------------------------------
# 7. Calculate Weighted Multiclass ROC-AUC
# --------------------------------------------------------------

multiclass_roc_auc_weighted = roc_auc_score(
    y_test_binarized,
    y_test_proba_array,
    average="weighted",
    multi_class="ovr"
)

# --------------------------------------------------------------
# 8. Calculate Per-Class ROC-AUC
# --------------------------------------------------------------

per_class_roc_auc = roc_auc_score(
    y_test_binarized,
    y_test_proba_array,
    average=None,
    multi_class="ovr"
)

roc_auc_results = pd.DataFrame({
    "Class": roc_auc_classes,
    "ROC_AUC": per_class_roc_auc
})

roc_auc_results["ROC_AUC_Percentage"] = (
    roc_auc_results["ROC_AUC"] * 100
)

# --------------------------------------------------------------
# 9. Store Final ROC-AUC Results
# --------------------------------------------------------------

final_multiclass_roc_auc = float(
    multiclass_roc_auc_macro
)

final_weighted_multiclass_roc_auc = float(
    multiclass_roc_auc_weighted
)

final_per_class_roc_auc = roc_auc_results.copy()

# --------------------------------------------------------------
# 10. Display Per-Class Results
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("MULTICLASS ROC-AUC RESULTS")
logger.info("-" * 70)

logger.info(
    "Macro-Average ROC-AUC               : %.4f",
    final_multiclass_roc_auc
)

logger.info(
    "Weighted-Average ROC-AUC            : %.4f",
    final_weighted_multiclass_roc_auc
)

logger.info("-" * 70)
logger.info("ROC-AUC BY CLASS")
logger.info("-" * 70)

for _, row in roc_auc_results.iterrows():

    logger.info(
        "Class %s | ROC-AUC: %.4f",
        row["Class"],
        row["ROC_AUC"]
    )

display(
    roc_auc_results.style.format({
        "ROC_AUC": "{:.4f}",
        "ROC_AUC_Percentage": "{:.2f}%"
    })
)

# --------------------------------------------------------------
# 11. Identify Best and Lowest Discrimination Classes
# --------------------------------------------------------------

best_auc_row = roc_auc_results.loc[
    roc_auc_results["ROC_AUC"].idxmax()
]

lowest_auc_row = roc_auc_results.loc[
    roc_auc_results["ROC_AUC"].idxmin()
]

logger.info("-" * 70)
logger.info("ROC-AUC INTERPRETATION")
logger.info("-" * 70)

logger.info(
    "Best Discrimination Class           : %s (%.4f)",
    best_auc_row["Class"],
    best_auc_row["ROC_AUC"]
)

logger.info(
    "Lowest Discrimination Class         : %s (%.4f)",
    lowest_auc_row["Class"],
    lowest_auc_row["ROC_AUC"]
)

# --------------------------------------------------------------
# 12. Validation Checks
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.6 MULTICLASS ROC-AUC VALIDATION")
logger.info("-" * 70)

classes_check = (
    len(roc_auc_results) == 6
    and set(roc_auc_results["Class"]) == set(roc_auc_classes)
)

auc_range_check = (
    np.all(
        (roc_auc_results["ROC_AUC"] >= 0.0)
        & (roc_auc_results["ROC_AUC"] <= 1.0)
    )
)

macro_consistency_check = np.isclose(
    final_multiclass_roc_auc,
    roc_auc_results["ROC_AUC"].mean(),
    atol=1e-10
)

probability_check = np.allclose(
    y_test_proba_array.sum(axis=1),
    1.0,
    atol=1e-6
)

test_only_check = (
    len(y_test_final_array) == len(y_test_proba_array)
)

logger.info(
    "All Six Classes Evaluated           : %s",
    "PASSED" if classes_check else "FAILED"
)

logger.info(
    "ROC-AUC Values Within [0, 1]        : %s",
    "PASSED" if auc_range_check else "FAILED"
)

logger.info(
    "Macro ROC-AUC Consistency            : %s",
    "PASSED" if macro_consistency_check else "FAILED"
)

logger.info(
    "Probability Matrix Valid             : %s",
    "PASSED" if probability_check else "FAILED"
)

logger.info(
    "Independent Test Set Only            : %s",
    "PASSED" if test_only_check else "FAILED"
)

if not all([
    classes_check,
    auc_range_check,
    macro_consistency_check,
    probability_check,
    test_only_check
]):
    raise AssertionError(
        "One or more Section 12.6 validation checks failed."
    )

logger.info("=" * 70)
logger.info("12.6 MULTICLASS ROC-AUC : COMPLETED")
logger.info("=" * 70)
````

## 12.7 ROC Curves

### Objective

Visualize the Receiver Operating Characteristic (ROC) curves for all six disease classes using a **One-vs-Rest (OvR)** multiclass strategy.

The ROC curve evaluates the relationship between:

- **True Positive Rate (TPR / Sensitivity)**
- **False Positive Rate (FPR)**

for different classification thresholds.

For the multiclass Skin Disorder Prediction problem, each disease class is evaluated against all remaining classes.

### Analysis Requirements

The ROC curve analysis will:

- Use the independent test set only.
- Use the final Random Forest probability predictions.
- Generate one ROC curve for each of the six classes.
- Display the ROC-AUC value for every class.
- Include a random-classifier reference line.
- Use the same class ordering established in Section 12.6.
- Avoid any model retraining or modification.

### Interpretation

A curve closer to the top-left corner indicates stronger class discrimination.

The ROC-AUC values from Section 12.6 will be used as the quantitative summary of each ROC curve.

````python
# ==============================================================
# Required Libraries — 
# ==============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
````

````python

logger.info("=" * 70)
logger.info("12.7 ROC CURVES")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Validate Required Objects
# --------------------------------------------------------------

required_objects = [
    "y_test_final",
    "y_test_proba",
    "roc_auc_classes",
    "final_per_class_roc_auc"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        f"Required Section 12.7 objects are missing: {missing_objects}"
    )

# --------------------------------------------------------------
# 2. Prepare Evaluation Data
# --------------------------------------------------------------

roc_curve_classes = np.asarray(
    roc_auc_classes
)

y_test_final_array = np.asarray(
    y_test_final
)

y_test_proba_array = np.asarray(
    y_test_proba
)

logger.info(
    "Independent Test Samples          : %d",
    len(y_test_final_array)
)

logger.info(
    "Number of Classes                 : %d",
    len(roc_curve_classes)
)

logger.info(
    "Classes                            : %s",
    roc_curve_classes.tolist()
)

# --------------------------------------------------------------
# 3. Binarize Multiclass Targets
# --------------------------------------------------------------

y_test_binarized_roc = label_binarize(
    y_test_final_array,
    classes=roc_curve_classes
)

# --------------------------------------------------------------
# 4. Calculate ROC Curve Data
# --------------------------------------------------------------

roc_curve_results = {}

for index, class_label in enumerate(roc_curve_classes):

    fpr, tpr, thresholds = roc_curve(
        y_test_binarized_roc[:, index],
        y_test_proba_array[:, index]
    )

    class_auc = auc(
        fpr,
        tpr
    )

    roc_curve_results[class_label] = {
        "fpr": fpr,
        "tpr": tpr,
        "thresholds": thresholds,
        "auc": class_auc
    }

# --------------------------------------------------------------
# 5. Display ROC-AUC Summary
# --------------------------------------------------------------

roc_curve_summary = pd.DataFrame({
    "Class": roc_curve_classes,
    "ROC_AUC": [
        roc_curve_results[class_label]["auc"]
        for class_label in roc_curve_classes
    ]
})

roc_curve_summary["ROC_AUC_Percentage"] = (
    roc_curve_summary["ROC_AUC"] * 100
)

logger.info("-" * 70)
logger.info("ROC CURVE AUC SUMMARY")
logger.info("-" * 70)

for _, row in roc_curve_summary.iterrows():

    logger.info(
        "Class %s | AUC: %.4f",
        row["Class"],
        row["ROC_AUC"]
    )

display(
    roc_curve_summary.style.format({
        "ROC_AUC": "{:.4f}",
        "ROC_AUC_Percentage": "{:.2f}%"
    })
)

# --------------------------------------------------------------
# 6. Plot Multiclass ROC Curves
# --------------------------------------------------------------

plt.figure(figsize=(10, 8))

for class_label in roc_curve_classes:

    fpr = roc_curve_results[class_label]["fpr"]
    tpr = roc_curve_results[class_label]["tpr"]
    class_auc = roc_curve_results[class_label]["auc"]

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"Class {class_label} (AUC = {class_auc:.4f})"
    )

# Random-classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier (AUC = 0.5000)"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "Multiclass ROC Curves — Final Random Forest Model"
)

plt.legend(
    loc="lower right"
)

plt.grid(
    True,
    alpha=0.3
)

plt.xlim(
    0.0,
    1.0
)

plt.ylim(
    0.0,
    1.05
)

plt.tight_layout()
plt.show()

# --------------------------------------------------------------
# 7. Store Results
# --------------------------------------------------------------

final_roc_curve_results = roc_curve_results.copy()
final_roc_curve_summary = roc_curve_summary.copy()

# --------------------------------------------------------------
# 8. Validation Checks
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.7 ROC CURVES VALIDATION")
logger.info("-" * 70)

all_classes_check = (
    len(final_roc_curve_results) == 6
    and set(final_roc_curve_results.keys())
    == set(roc_curve_classes)
)

auc_consistency_check = np.allclose(
    roc_curve_summary["ROC_AUC"].values,
    final_per_class_roc_auc["ROC_AUC"].values,
    atol=1e-10
)

curve_structure_check = all(
    len(result["fpr"]) == len(result["tpr"])
    and len(result["fpr"]) > 1
    for result in final_roc_curve_results.values()
)

auc_range_check = np.all(
    (roc_curve_summary["ROC_AUC"] >= 0.0)
    & (roc_curve_summary["ROC_AUC"] <= 1.0)
)

test_only_check = (
    len(y_test_final_array)
    == y_test_proba_array.shape[0]
)

logger.info(
    "All Six ROC Curves Generated         : %s",
    "PASSED" if all_classes_check else "FAILED"
)

logger.info(
    "ROC-AUC Consistency with 12.6        : %s",
    "PASSED" if auc_consistency_check else "FAILED"
)

logger.info(
    "ROC Curve Structures Valid           : %s",
    "PASSED" if curve_structure_check else "FAILED"
)

logger.info(
    "ROC-AUC Values Within [0, 1]         : %s",
    "PASSED" if auc_range_check else "FAILED"
)

logger.info(
    "Independent Test Set Only             : %s",
    "PASSED" if test_only_check else "FAILED"
)

if not all([
    all_classes_check,
    auc_consistency_check,
    curve_structure_check,
    auc_range_check,
    test_only_check
]):
    raise AssertionError(
        "One or more Section 12.7 validation checks failed."
    )

logger.info("=" * 70)
logger.info("12.7 ROC CURVES : COMPLETED")
logger.info("=" * 70)
````

## 12.8 Precision-Recall Analysis

### Objective

Evaluate the final optimized Random Forest model using **Precision-Recall analysis** under a multiclass One-vs-Rest (OvR) strategy.

Precision-Recall analysis provides an additional view of classification performance by examining the relationship between:

- **Precision:** Proportion of predicted positive samples that are actually positive.
- **Recall:** Proportion of actual positive samples correctly identified.

For the six-class Skin Disorder Prediction problem, each disease class is evaluated against all remaining classes.

### Analysis Requirements

The Precision-Recall analysis will:

- Use the independent test set only.
- Use the final Random Forest probability predictions.
- Generate one Precision-Recall curve for each of the six classes.
- Calculate Average Precision (AP) for each class.
- Calculate macro-average and weighted-average AP.
- Identify the strongest and weakest class-level discrimination.
- Visualize the Precision-Recall curves using Matplotlib.

### Interpretation

A curve closer to the upper-right region indicates stronger precision-recall performance.

Average Precision provides a numerical summary of the Precision-Recall curve and complements the ROC-AUC, Macro F1, and Balanced Accuracy metrics.

Because the dataset contains six classes with unequal frequencies, class-level and macro-average results will be considered alongside weighted results.

````python
# ==============================================================
# Required Libraries —
# ==============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    precision_recall_curve,
    average_precision_score
)

from sklearn.preprocessing import label_binarize
````

````python


logger.info("=" * 70)
logger.info("12.8 PRECISION-RECALL ANALYSIS")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Validate Required Objects
# --------------------------------------------------------------

required_objects = [
    "y_test_final",
    "y_test_proba",
    "roc_auc_classes"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise NameError(
        f"Required Section 12.8 objects are missing: {missing_objects}"
    )

# --------------------------------------------------------------
# 2. Prepare Evaluation Data
# --------------------------------------------------------------

pr_classes = np.asarray(
    roc_auc_classes
)

y_test_final_array = np.asarray(
    y_test_final
)

y_test_proba_array = np.asarray(
    y_test_proba
)

logger.info(
    "Independent Test Samples          : %d",
    len(y_test_final_array)
)

logger.info(
    "Number of Classes                 : %d",
    len(pr_classes)
)

logger.info(
    "Classes                            : %s",
    pr_classes.tolist()
)

# --------------------------------------------------------------
# 3. Validate Probability Matrix
# --------------------------------------------------------------

if y_test_proba_array.ndim != 2:
    raise ValueError(
        "Test probability predictions must be a 2-dimensional matrix."
    )

if y_test_proba_array.shape[0] != len(y_test_final_array):
    raise ValueError(
        "Probability rows do not match the number of test samples."
    )

if y_test_proba_array.shape[1] != len(pr_classes):
    raise ValueError(
        "Probability columns do not match the number of classes."
    )

if not np.all(np.isfinite(y_test_proba_array)):
    raise ValueError(
        "Probability predictions contain non-finite values."
    )

if not np.allclose(
    y_test_proba_array.sum(axis=1),
    1.0,
    atol=1e-6
):
    raise ValueError(
        "Probability predictions do not sum to 1."
    )

# --------------------------------------------------------------
# 4. Binarize Multiclass Target
# --------------------------------------------------------------

y_test_binarized_pr = label_binarize(
    y_test_final_array,
    classes=pr_classes
)

# --------------------------------------------------------------
# 5. Calculate Precision-Recall Curves
# --------------------------------------------------------------

precision_recall_results = {}

for index, class_label in enumerate(pr_classes):

    precision, recall, thresholds = precision_recall_curve(
        y_test_binarized_pr[:, index],
        y_test_proba_array[:, index]
    )

    average_precision = average_precision_score(
        y_test_binarized_pr[:, index],
        y_test_proba_array[:, index]
    )

    precision_recall_results[class_label] = {
        "precision": precision,
        "recall": recall,
        "thresholds": thresholds,
        "average_precision": average_precision
    }

# --------------------------------------------------------------
# 6. Calculate Macro and Weighted Average Precision
# --------------------------------------------------------------

macro_average_precision = average_precision_score(
    y_test_binarized_pr,
    y_test_proba_array,
    average="macro"
)

weighted_average_precision = average_precision_score(
    y_test_binarized_pr,
    y_test_proba_array,
    average="weighted"
)

# --------------------------------------------------------------
# 7. Create Summary Table
# --------------------------------------------------------------

pr_summary = pd.DataFrame({
    "Class": pr_classes,
    "Average_Precision": [
        precision_recall_results[class_label]["average_precision"]
        for class_label in pr_classes
    ]
})

pr_summary["Average_Precision_Percentage"] = (
    pr_summary["Average_Precision"] * 100
)

# --------------------------------------------------------------
# 8. Store Final Results
# --------------------------------------------------------------

final_precision_recall_results = (
    precision_recall_results.copy()
)

final_pr_summary = (
    pr_summary.copy()
)

final_macro_average_precision = float(
    macro_average_precision
)

final_weighted_average_precision = float(
    weighted_average_precision
)

# --------------------------------------------------------------
# 9. Display Results
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRECISION-RECALL RESULTS")
logger.info("-" * 70)

logger.info(
    "Macro-Average Precision              : %.4f",
    final_macro_average_precision
)

logger.info(
    "Weighted-Average Precision           : %.4f",
    final_weighted_average_precision
)

logger.info("-" * 70)
logger.info("AVERAGE PRECISION BY CLASS")
logger.info("-" * 70)

for _, row in pr_summary.iterrows():

    logger.info(
        "Class %s | Average Precision: %.4f",
        row["Class"],
        row["Average_Precision"]
    )

display(
    pr_summary.style.format({
        "Average_Precision": "{:.4f}",
        "Average_Precision_Percentage": "{:.2f}%"
    })
)

# --------------------------------------------------------------
# 10. Identify Best and Lowest AP Classes
# --------------------------------------------------------------

best_ap_row = pr_summary.loc[
    pr_summary["Average_Precision"].idxmax()
]

lowest_ap_row = pr_summary.loc[
    pr_summary["Average_Precision"].idxmin()
]

logger.info("-" * 70)
logger.info("PRECISION-RECALL INTERPRETATION")
logger.info("-" * 70)

logger.info(
    "Best AP Class                      : %s (%.4f)",
    best_ap_row["Class"],
    best_ap_row["Average_Precision"]
)

logger.info(
    "Lowest AP Class                    : %s (%.4f)",
    lowest_ap_row["Class"],
    lowest_ap_row["Average_Precision"]
)

# --------------------------------------------------------------
# 11. Plot Precision-Recall Curves
# --------------------------------------------------------------

plt.figure(figsize=(10, 8))

for class_label in pr_classes:

    precision = (
        precision_recall_results[class_label]["precision"]
    )

    recall = (
        precision_recall_results[class_label]["recall"]
    )

    average_precision = (
        precision_recall_results[class_label]["average_precision"]
    )

    plt.plot(
        recall,
        precision,
        linewidth=2,
        label=(
            f"Class {class_label} "
            f"(AP = {average_precision:.4f})"
        )
    )

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "Multiclass Precision-Recall Curves — Final Random Forest Model"
)

plt.legend(
    loc="lower left"
)

plt.grid(
    True,
    alpha=0.3
)

plt.xlim(
    0.0,
    1.0
)

plt.ylim(
    0.0,
    1.05
)

plt.tight_layout()
plt.show()

# --------------------------------------------------------------
# 12. Validation Checks
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("12.8 PRECISION-RECALL VALIDATION")
logger.info("-" * 70)

all_classes_check = (
    len(final_precision_recall_results) == 6
    and set(final_precision_recall_results.keys())
    == set(pr_classes)
)

ap_range_check = np.all(
    (pr_summary["Average_Precision"] >= 0.0)
    & (pr_summary["Average_Precision"] <= 1.0)
)

curve_structure_check = all(
    len(result["precision"]) == len(result["recall"])
    and len(result["precision"]) > 1
    for result in final_precision_recall_results.values()
)

macro_consistency_check = np.isclose(
    final_macro_average_precision,
    pr_summary["Average_Precision"].mean(),
    atol=1e-10
)

probability_check = np.allclose(
    y_test_proba_array.sum(axis=1),
    1.0,
    atol=1e-6
)

test_only_check = (
    len(y_test_final_array)
    == y_test_proba_array.shape[0]
)

logger.info(
    "All Six Classes Evaluated             : %s",
    "PASSED" if all_classes_check else "FAILED"
)

logger.info(
    "Average Precision Values Valid        : %s",
    "PASSED" if ap_range_check else "FAILED"
)

logger.info(
    "Precision-Recall Curves Valid          : %s",
    "PASSED" if curve_structure_check else "FAILED"
)

logger.info(
    "Macro AP Consistency                   : %s",
    "PASSED" if macro_consistency_check else "FAILED"
)

logger.info(
    "Probability Matrix Valid               : %s",
    "PASSED" if probability_check else "FAILED"
)

logger.info(
    "Independent Test Set Only              : %s",
    "PASSED" if test_only_check else "FAILED"
)

if not all([
    all_classes_check,
    ap_range_check,
    curve_structure_check,
    macro_consistency_check,
    probability_check,
    test_only_check
]):
    raise AssertionError(
        "One or more Section 12.8 validation checks failed."
    )

logger.info("=" * 70)
logger.info("12.8 PRECISION-RECALL ANALYSIS : COMPLETED")
logger.info("=" * 70)
````

All test-set metrics are calculated from the same independent test set and therefore provide a consistent basis for model comparison.

No model will be retrained during this subsection.

````python
# ==============================================================
# Required Libraries 
# ==============================================================

import numpy as np
import pandas as pd

from IPython.display import display
````

````python


logger.info("=" * 70)
logger.info("13.1 MODEL PERFORMANCE COMPARISON")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Locate Existing Model Comparison Objects
# --------------------------------------------------------------

candidate_objects = [
    "baseline_results",
    "advanced_model_results",
    "baseline_performance_summary",
    "advanced_performance_summary"
]

available_objects = [
    obj for obj in candidate_objects
    if obj in globals()
]

logger.info(
    "Available Existing Result Objects    : %s",
    available_objects
)

# --------------------------------------------------------------
# 2. Build Comparison from Existing Results
# --------------------------------------------------------------

comparison_frames = []

for obj_name in [
    "baseline_results",
    "advanced_model_results",
    "baseline_performance_summary",
    "advanced_performance_summary"
]:

    if obj_name in globals():

        obj = globals()[obj_name]

        if isinstance(obj, pd.DataFrame):
            temp = obj.copy()

            if len(temp) > 0:
                comparison_frames.append(temp)

# --------------------------------------------------------------
# 3. Validate Existing Comparison Data
# --------------------------------------------------------------

if not comparison_frames:

    raise NameError(
        "No existing model performance DataFrame was found. "
        "Please provide the Section 9/10 model comparison output "
        "before running Section 13.1."
    )

model_performance_comparison = pd.concat(
    comparison_frames,
    ignore_index=True
)

# --------------------------------------------------------------
# 4. Normalize Common Column Names
# --------------------------------------------------------------

column_aliases = {
    "Model": ["Model", "model", "MODEL"],
    "Accuracy": ["Accuracy", "accuracy", "Test Accuracy"],
    "Balanced Accuracy": [
        "Balanced Accuracy",
        "balanced_accuracy",
        "Test Balanced Accuracy"
    ],
    "Macro Precision": [
        "Macro Precision",
        "macro_precision",
        "Test Macro Precision"
    ],
    "Macro Recall": [
        "Macro Recall",
        "macro_recall",
        "Test Macro Recall"
    ],
    "Macro F1": [
        "Macro F1",
        "macro_f1",
        "Test Macro F1"
    ],
    "Weighted F1": [
        "Weighted F1",
        "weighted_f1",
        "Test Weighted F1"
    ]
}

normalized_columns = {}

for standard_name, aliases in column_aliases.items():

    for alias in aliases:

        if alias in model_performance_comparison.columns:

            normalized_columns[standard_name] = alias
            break

# --------------------------------------------------------------
# 5. Required Metric Validation
# --------------------------------------------------------------

required_metrics = [
    "Model",
    "Accuracy",
    "Balanced Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1",
    "Weighted F1"
]

missing_metrics = [
    metric for metric in required_metrics
    if metric not in normalized_columns
]

if missing_metrics:

    logger.info(
        "Missing Comparison Columns             : %s",
        missing_metrics
    )

    raise KeyError(
        "Required model comparison metrics are missing: "
        f"{missing_metrics}"
    )

# --------------------------------------------------------------
# 6. Create Standardized Comparison Table
# --------------------------------------------------------------

standard_comparison = pd.DataFrame()

for metric in required_metrics:

    standard_comparison[metric] = (
        model_performance_comparison[
            normalized_columns[metric]
        ]
    )

# Remove duplicate model rows if historical cells were combined
standard_comparison = (
    standard_comparison
    .drop_duplicates(subset=["Model"], keep="last")
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# 7. Add Final Optimized Random Forest
# --------------------------------------------------------------

final_rf_row = pd.DataFrame({
    "Model": ["Optimized Random Forest"],
    "Accuracy": [test_accuracy],
    "Balanced Accuracy": [test_balanced_accuracy],
    "Macro Precision": [test_macro_precision],
    "Macro Recall": [test_macro_recall],
    "Macro F1": [test_macro_f1],
    "Weighted F1": [test_weighted_f1]
})

standard_comparison = pd.concat(
    [
        standard_comparison,
        final_rf_row
    ],
    ignore_index=True
)

# --------------------------------------------------------------
# 8. Store Results
# --------------------------------------------------------------

final_model_performance_comparison = (
    standard_comparison.copy()
)

# --------------------------------------------------------------
# 9. Display Results
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL PERFORMANCE COMPARISON")
logger.info("-" * 70)

display(
    final_model_performance_comparison.style.format({
        "Accuracy": "{:.4f}",
        "Balanced Accuracy": "{:.4f}",
        "Macro Precision": "{:.4f}",
        "Macro Recall": "{:.4f}",
        "Macro F1": "{:.4f}",
        "Weighted F1": "{:.4f}"
    })
)

# --------------------------------------------------------------
# 10. Identify Best Test Macro F1
# --------------------------------------------------------------

best_test_row = final_model_performance_comparison.loc[
    final_model_performance_comparison["Macro F1"].idxmax()
]

logger.info(
    "Best Test Macro F1 Model              : %s",
    best_test_row["Model"]
)

logger.info(
    "Best Test Macro F1                    : %.4f",
    best_test_row["Macro F1"]
)

# --------------------------------------------------------------
# 11. Validation
# --------------------------------------------------------------

model_count_check = (
    len(final_model_performance_comparison) >= 3
)

score_range_check = np.all(
    (
        final_model_performance_comparison[
            required_metrics[1:]
        ] >= 0.0
    )
    &
    (
        final_model_performance_comparison[
            required_metrics[1:]
        ] <= 1.0
    )
)

final_rf_check = (
    "Optimized Random Forest"
    in final_model_performance_comparison["Model"].values
)

if not all([
    model_count_check,
    score_range_check,
    final_rf_check
]):
    raise AssertionError(
        "One or more Section 13.1 validation checks failed."
    )

logger.info("-" * 70)
logger.info("13.1 MODEL PERFORMANCE VALIDATION")
logger.info("-" * 70)

logger.info(
    "Model Comparison Data Available       : %s",
    "PASSED" if model_count_check else "FAILED"
)

logger.info(
    "Performance Scores Valid               : %s",
    "PASSED" if score_range_check else "FAILED"
)

logger.info(
    "Optimized Random Forest Included      : %s",
    "PASSED" if final_rf_check else "FAILED"
)

logger.info("=" * 70)
logger.info("13.1 MODEL PERFORMANCE COMPARISON : COMPLETED")
logger.info("=" * 70)
````

## 13.3 Cross-Validation Stability

### Objective

Evaluate the stability and consistency of each model across the five stratified cross-validation folds.

Cross-validation standard deviation is used as the primary stability indicator.

A lower standard deviation indicates that the model produces more consistent performance across different training and validation subsets.

### Evaluation Principle

The models are compared using the standard deviation of their **Macro F1** scores across the five cross-validation folds.

- Lower standard deviation → higher stability
- Higher standard deviation → greater variation across folds

The analysis uses only training-data cross-validation results. The independent test set is not used to calculate cross-validation stability.

### Interpretation

Cross-validation stability is considered together with the mean Macro F1 score.

A model with a high mean score and low standard deviation is generally preferable because it combines strong predictive performance with consistent validation behavior.

````python
# ==============================================================
# Required Libraries 
# ==============================================================

import numpy as np
import pandas as pd

from IPython.display import display
````

````python


logger.info("=" * 70)
logger.info("13.3 CROSS-VALIDATION STABILITY")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Expected Models
# --------------------------------------------------------------

expected_models = [
    "Random Forest",
    "CatBoost",
    "LightGBM",
    "Extra Trees",
    "Gradient Boosting",
    "XGBoost"
]

# --------------------------------------------------------------
# 2. Locate Existing CV Stability Results
# --------------------------------------------------------------

stability_source = None
stability_source_name = None

candidate_objects = [
    "advanced_cv_results",
    "advanced_model_cv_results",
    "advanced_performance_summary",
    "advanced_model_results"
]

for object_name in candidate_objects:

    if object_name in globals():

        candidate = globals()[object_name]

        if isinstance(candidate, pd.DataFrame):

            columns_lower = [
                str(col).lower()
                for col in candidate.columns
            ]

            has_model = any(
                "model" in col
                for col in columns_lower
            )

            has_std = any(
                (
                    ("std" in col or "standard deviation" in col)
                    and "f1" in col
                )
                for col in columns_lower
            )

            if has_model and has_std:

                stability_source = candidate.copy()
                stability_source_name = object_name
                break

logger.info(
    "Stability Source Found                 : %s",
    stability_source_name
)

# --------------------------------------------------------------
# 3. Extract Historical Advanced Model Stability
# --------------------------------------------------------------

if stability_source is not None:

    model_column = next(
        col for col in stability_source.columns
        if "model" in str(col).lower()
    )

    std_column = next(
        col for col in stability_source.columns
        if (
            ("std" in str(col).lower()
             or "standard deviation" in str(col).lower())
            and
            "f1" in str(col).lower()
        )
    )

    cv_stability_comparison = pd.DataFrame({
        "Model": stability_source[model_column],
        "CV Macro F1 SD": stability_source[std_column]
    })

else:

    # ----------------------------------------------------------
    # 4. Recover Optimized RF Stability
    # ----------------------------------------------------------

    recovered_stability = []

    if "optimized_cv_std_scores" in globals():

        std_scores = optimized_cv_std_scores

        macro_f1_std = None

        for key, value in std_scores.items():

            key_lower = str(key).lower()

            if (
                "macro" in key_lower
                and "f1" in key_lower
            ):

                macro_f1_std = float(value)
                break

        if macro_f1_std is not None:

            recovered_stability.append({
                "Model": "Optimized Random Forest",
                "CV Macro F1 SD": macro_f1_std
            })

    # ----------------------------------------------------------
    # 5. Recover Advanced Model CV Standard Deviations
    # ----------------------------------------------------------

    if "advanced_performance_summary" in globals():

        source = advanced_performance_summary

        if isinstance(source, pd.DataFrame):

            model_col = next(
                (
                    col for col in source.columns
                    if "model" in str(col).lower()
                ),
                None
            )

            std_col = next(
                (
                    col for col in source.columns
                    if (
                        ("std" in str(col).lower()
                         or "standard deviation"
                         in str(col).lower())
                        and
                        "f1" in str(col).lower()
                    )
                ),
                None
            )

            if model_col is not None and std_col is not None:

                for _, row in source.iterrows():

                    recovered_stability.append({
                        "Model": row[model_col],
                        "CV Macro F1 SD": row[std_col]
                    })

    if not recovered_stability:

        raise NameError(
            "No existing CV stability results were found. "
            "Please run the earlier Section 10/11 CV stability "
            "cells before running Section 13.3."
        )

    cv_stability_comparison = pd.DataFrame(
        recovered_stability
    )

# --------------------------------------------------------------
# 6. Remove Duplicate Models
# --------------------------------------------------------------

cv_stability_comparison = (
    cv_stability_comparison
    .drop_duplicates(
        subset=["Model"],
        keep="last"
    )
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# 7. Add Optimized Random Forest
# --------------------------------------------------------------

optimized_rf_std = pd.DataFrame({
    "Model": ["Optimized Random Forest"],
    "CV Macro F1 SD": [
        float(
            next(
                value
                for key, value
                in optimized_cv_std_scores.items()
                if (
                    "macro" in str(key).lower()
                    and "f1" in str(key).lower()
                )
            )
        )
    ]
})

cv_stability_comparison = pd.concat(
    [
        cv_stability_comparison,
        optimized_rf_std
    ],
    ignore_index=True
)

# --------------------------------------------------------------
# 8. Sort by Stability
# --------------------------------------------------------------

cv_stability_comparison = (
    cv_stability_comparison
    .sort_values(
        "CV Macro F1 SD",
        ascending=True
    )
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# 9. Store Final Results
# --------------------------------------------------------------

final_cv_stability_comparison = (
    cv_stability_comparison.copy()
)

# --------------------------------------------------------------
# 10. Display
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION MACRO F1 STABILITY")
logger.info("-" * 70)

display(
    final_cv_stability_comparison.style.format({
        "CV Macro F1 SD": "{:.4f}"
    })
)

# --------------------------------------------------------------
# 11. Identify Most Stable Model
# --------------------------------------------------------------

most_stable_row = (
    final_cv_stability_comparison
    .loc[
        final_cv_stability_comparison[
            "CV Macro F1 SD"
        ].idxmin()
    ]
)

logger.info(
    "Most Stable Model                     : %s",
    most_stable_row["Model"]
)

logger.info(
    "Lowest CV Macro F1 SD                 : %.4f",
    most_stable_row["CV Macro F1 SD"]
)

# --------------------------------------------------------------
# 12. Validation
# --------------------------------------------------------------

stability_count_check = (
    len(final_cv_stability_comparison) >= 2
)

stability_range_check = (
    final_cv_stability_comparison[
        "CV Macro F1 SD"
    ].ge(0.0).all()
)

optimized_rf_check = (
    "Optimized Random Forest"
    in final_cv_stability_comparison["Model"].values
)

if not all([
    stability_count_check,
    stability_range_check,
    optimized_rf_check
]):

    raise AssertionError(
        "One or more Section 13.3 validation checks failed."
    )

logger.info("-" * 70)
logger.info("13.3 CROSS-VALIDATION STABILITY VALIDATION")
logger.info("-" * 70)

logger.info(
    "Stability Results Available             : %s",
    "PASSED" if stability_count_check else "FAILED"
)

logger.info(
    "Standard Deviation Values Valid         : %s",
    "PASSED" if stability_range_check else "FAILED"
)

logger.info(
    "Optimized Random Forest Included        : %s",
    "PASSED" if optimized_rf_check else "FAILED"
)

logger.info("=" * 70)
logger.info("13.3 CROSS-VALIDATION STABILITY : COMPLETED")
logger.info("=" * 70)
````

## 13.4 Training Time Comparison

### Objective

Evaluate the computational training efficiency of the candidate models as an additional model-selection consideration.

### Training-Time Availability

Standalone training-time measurements were not recorded consistently during the baseline and advanced model training stages.

Therefore, this subsection does **not** reconstruct or estimate training times from historical execution logs.

Hyperparameter-search duration is intentionally excluded because it represents multiple model fits and does not represent the training time of a single final model.

### Evaluation Decision

Since reliable standalone training-time measurements are unavailable, training time is not used as a quantitative ranking criterion in the final model-selection process.

The final model selection will instead prioritize:

- Independent test-set performance
- Macro F1
- Balanced Accuracy
- Cross-validation mean Macro F1
- Cross-validation stability
- Generalization performance
- Production suitability

This approach avoids introducing estimated or inconsistent measurements into the experimental results.

## 13.5 Per-Class Performance Comparison

### Objective

Compare model performance across the individual skin-disorder classes.

Overall metrics such as accuracy and Macro F1 provide a summary of model performance, but they may hide differences between individual classes.

Therefore, per-class Precision, Recall, and F1-score are examined to determine whether the models perform consistently across all six disease classes.

### Evaluation Metrics

The following metrics are considered for each class:

- Precision
- Recall
- F1-score
- Support

### Evaluation Principle

The comparison gives particular attention to F1-score because it combines Precision and Recall and provides a balanced measure of classification performance for each disease class.

## 13.7 Model Generalization Comparison

### Objective

Compare cross-validation Macro F1-score with independent test-set Macro F1-score for all evaluated models.

Generalization analysis determines how consistently each model performs when moving from cross-validation to previously unseen test data.

### Generalization Gap

The generalization gap is calculated as:

**Generalization Gap = CV Macro F1 − Test Macro F1**

Interpretation:

- A small positive gap indicates slightly higher cross-validation performance than test performance.
- A gap close to zero indicates strong consistency between validation and test performance.
- A negative gap indicates that test performance exceeded the cross-validation mean.
- A large positive gap may indicate greater validation-test variation or possible overfitting.

Macro F1-score is used because it gives equal importance to all six disease classes and is the primary model-selection metric for this project.

The independent test set remains untouched during cross-validation and hyperparameter optimization.

````python
# ==============================================================
# Required Libraries 
# ==============================================================

import numpy as np
import pandas as pd

from IPython.display import display
````

````python


logger.info("=" * 70)
logger.info("13.7 MODEL GENERALIZATION COMPARISON")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Validate Required Objects
# --------------------------------------------------------------

required_objects = [
    "advanced_performance_summary",
    "final_cv_mean_comparison",
    "test_macro_f1",
    "test_cv_macro_f1"
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:

    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

# --------------------------------------------------------------
# 2. Copy Existing Advanced Model Results
# --------------------------------------------------------------

test_source = (
    advanced_performance_summary
    .copy(deep=True)
)

cv_source = (
    final_cv_mean_comparison
    .copy(deep=True)
)

logger.info(
    "Advanced Test Result Columns          : %s",
    list(test_source.columns)
)

logger.info(
    "CV Result Columns                     : %s",
    list(cv_source.columns)
)

# --------------------------------------------------------------
# 3. Helper — Identify Model Column
# --------------------------------------------------------------

def detect_model_column(df):

    for column in df.columns:

        normalized = (
            str(column)
            .lower()
            .replace("_", " ")
            .strip()
        )

        if normalized == "model":

            return column

    raise KeyError(
        "Model column could not be identified."
    )


# --------------------------------------------------------------
# 4. Helper — Identify Macro F1 Column
# --------------------------------------------------------------

def detect_macro_f1_column(df):

    for column in df.columns:

        normalized = (
            str(column)
            .lower()
            .replace("_", " ")
            .replace("-", " ")
            .replace("%", "")
            .strip()
        )

        if (
            "macro" in normalized
            and "f1" in normalized
        ):

            return column

    raise KeyError(
        "Macro F1 column could not be identified."
    )


test_model_column = detect_model_column(
    test_source
)

cv_model_column = detect_model_column(
    cv_source
)

test_f1_column = detect_macro_f1_column(
    test_source
)

cv_f1_column = detect_macro_f1_column(
    cv_source
)

logger.info(
    "Detected Test Model Column             : %s",
    test_model_column
)

logger.info(
    "Detected Test Macro F1 Column         : %s",
    test_f1_column
)

logger.info(
    "Detected CV Model Column               : %s",
    cv_model_column
)

logger.info(
    "Detected CV Macro F1 Column           : %s",
    cv_f1_column
)

# --------------------------------------------------------------
# 5. Standardize Advanced Model Results
# --------------------------------------------------------------

advanced_test_results = pd.DataFrame({

    "Model":
        test_source[test_model_column]
        .astype(str)
        .str.strip(),

    "Test Macro F1":
        pd.to_numeric(
            test_source[test_f1_column],
            errors="coerce"
        )
})

advanced_cv_results = pd.DataFrame({

    "Model":
        cv_source[cv_model_column]
        .astype(str)
        .str.strip(),

    "CV Macro F1":
        pd.to_numeric(
            cv_source[cv_f1_column],
            errors="coerce"
        )
})

# --------------------------------------------------------------
# 6. Remove Invalid Rows
# --------------------------------------------------------------

advanced_test_results = (
    advanced_test_results
    .dropna(
        subset=["Model", "Test Macro F1"]
    )
)

advanced_cv_results = (
    advanced_cv_results
    .dropna(
        subset=["Model", "CV Macro F1"]
    )
)

# --------------------------------------------------------------
# 7. Merge Advanced Models
# --------------------------------------------------------------

advanced_generalization = pd.merge(
    advanced_cv_results,
    advanced_test_results,
    on="Model",
    how="inner"
)

logger.info(
    "Advanced Models Matched                : %d",
    len(advanced_generalization)
)

# --------------------------------------------------------------
# 8. Add Optimized Random Forest
# --------------------------------------------------------------

optimized_rf_result = pd.DataFrame({

    "Model": [
        "Optimized Random Forest"
    ],

    "CV Macro F1": [
        float(test_cv_macro_f1)
    ],

    "Test Macro F1": [
        float(test_macro_f1)
    ]
})

# --------------------------------------------------------------
# 9. Combine All Models
# --------------------------------------------------------------

generalization_comparison = pd.concat(
    [
        advanced_generalization,
        optimized_rf_result
    ],
    ignore_index=True
)

# --------------------------------------------------------------
# 10. Remove Duplicate Model Entries
# --------------------------------------------------------------

generalization_comparison = (
    generalization_comparison
    .drop_duplicates(
        subset=["Model"],
        keep="last"
    )
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# 11. Calculate Generalization Gap
# --------------------------------------------------------------

generalization_comparison[
    "Generalization Gap"
] = (
    generalization_comparison["CV Macro F1"]
    -
    generalization_comparison["Test Macro F1"]
)

generalization_comparison[
    "Generalization Gap (%)"
] = (
    generalization_comparison["Generalization Gap"]
    * 100
)

generalization_comparison[
    "Absolute Gap"
] = (
    generalization_comparison["Generalization Gap"]
    .abs()
)

# --------------------------------------------------------------
# 12. Generalization Interpretation
# --------------------------------------------------------------

def interpret_generalization(gap):

    if abs(gap) <= 0.01:

        return "Very Strong Consistency"

    elif abs(gap) <= 0.03:

        return "Strong Generalization"

    elif gap > 0:

        return "Moderate Validation-Test Gap"

    else:

        return "Test Above CV Mean"


generalization_comparison[
    "Interpretation"
] = (
    generalization_comparison[
        "Generalization Gap"
    ]
    .apply(interpret_generalization)
)

# --------------------------------------------------------------
# 13. Sort by Test Macro F1
# --------------------------------------------------------------

generalization_comparison = (
    generalization_comparison
    .sort_values(
        by="Test Macro F1",
        ascending=False
    )
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# 14. Display Final Comparison
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "CROSS-VALIDATION VS INDEPENDENT TEST PERFORMANCE"
)
logger.info("-" * 70)

display(
    generalization_comparison[
        [
            "Model",
            "CV Macro F1",
            "Test Macro F1",
            "Generalization Gap",
            "Generalization Gap (%)",
            "Interpretation"
        ]
    ]
)

# --------------------------------------------------------------
# 15. Optimized Random Forest Statistics
# --------------------------------------------------------------

optimized_rf_rows = (
    generalization_comparison[
        generalization_comparison["Model"]
        ==
        "Optimized Random Forest"
    ]
)

if optimized_rf_rows.empty:

    raise AssertionError(
        "Optimized Random Forest was not added "
        "to the generalization comparison."
    )

optimized_rf_row = (
    optimized_rf_rows.iloc[0]
)

optimized_rf_cv = float(
    optimized_rf_row["CV Macro F1"]
)

optimized_rf_test = float(
    optimized_rf_row["Test Macro F1"]
)

optimized_rf_gap = float(
    optimized_rf_row["Generalization Gap"]
)

# --------------------------------------------------------------
# 16. Most Consistent Model
# --------------------------------------------------------------

most_consistent_index = (
    generalization_comparison[
        "Absolute Gap"
    ]
    .idxmin()
)

most_consistent_model = (
    generalization_comparison.loc[
        most_consistent_index,
        "Model"
    ]
)

smallest_absolute_gap = float(
    generalization_comparison[
        "Absolute Gap"
    ].min()
)

# --------------------------------------------------------------
# 17. Logging
# --------------------------------------------------------------

logger.info(
    "Optimized RF CV Macro F1              : %.4f",
    optimized_rf_cv
)

logger.info(
    "Optimized RF Test Macro F1            : %.4f",
    optimized_rf_test
)

logger.info(
    "Optimized RF Generalization Gap       : %.4f",
    optimized_rf_gap
)

logger.info(
    "Optimized RF Gap Percentage           : %.2f%%",
    optimized_rf_gap * 100
)

logger.info(
    "Most Consistent Model                 : %s",
    most_consistent_model
)

logger.info(
    "Smallest Absolute Gap                 : %.4f",
    smallest_absolute_gap
)

# --------------------------------------------------------------
# 18. Validation Checks
# --------------------------------------------------------------

optimized_rf_cv_check = np.isclose(
    optimized_rf_cv,
    float(test_cv_macro_f1),
    atol=1e-4
)

optimized_rf_test_check = np.isclose(
    optimized_rf_test,
    float(test_macro_f1),
    atol=1e-4
)

optimized_rf_gap_check = np.isclose(
    optimized_rf_gap,
    optimized_rf_cv - optimized_rf_test,
    atol=1e-6
)

metric_range_check = (
    generalization_comparison[
        [
            "CV Macro F1",
            "Test Macro F1"
        ]
    ]
    .apply(
        lambda column:
        column.between(0, 1).all()
    )
    .all()
)

model_count_check = (
    len(generalization_comparison) == 7
)

duplicate_model_check = (
    generalization_comparison[
        "Model"
    ].is_unique
)

# --------------------------------------------------------------
# 19. Store Final Results
# --------------------------------------------------------------

final_generalization_comparison = (
    generalization_comparison
    .copy(deep=True)
)

# --------------------------------------------------------------
# 20. Validation Logs
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("13.7 GENERALIZATION VALIDATION")
logger.info("-" * 70)

logger.info(
    "Optimized RF CV Macro F1 Valid         : %s",
    "PASSED"
    if optimized_rf_cv_check
    else "FAILED"
)

logger.info(
    "Optimized RF Test Macro F1 Valid      : %s",
    "PASSED"
    if optimized_rf_test_check
    else "FAILED"
)

logger.info(
    "Generalization Gap Calculation        : %s",
    "PASSED"
    if optimized_rf_gap_check
    else "FAILED"
)

logger.info(
    "Metric Ranges Valid                   : %s",
    "PASSED"
    if metric_range_check
    else "FAILED"
)

logger.info(
    "Seven Models Included                 : %s",
    "PASSED"
    if model_count_check
    else "FAILED"
)

logger.info(
    "Duplicate Model Check                 : %s",
    "PASSED"
    if duplicate_model_check
    else "FAILED"
)

# --------------------------------------------------------------
# 21. Final Validation
# --------------------------------------------------------------

if not all([
    optimized_rf_cv_check,
    optimized_rf_test_check,
    optimized_rf_gap_check,
    metric_range_check,
    model_count_check,
    duplicate_model_check
]):

    raise AssertionError(
        "One or more Section 13.7 validation checks failed."
    )

logger.info("=" * 70)
logger.info(
    "13.7 MODEL GENERALIZATION COMPARISON : COMPLETED"
)
logger.info("=" * 70)
````

## 13.8 Model Ranking

### Objective

Rank all evaluated models using the performance and validation evidence established in the previous sections.

### Ranking Criteria

The ranking follows a transparent multi-criteria approach:

1. **Test Macro F1** — primary criterion because it represents performance on previously unseen data.
2. **Cross-Validation Macro F1** — secondary criterion representing expected performance across stratified validation folds.
3. **Cross-Validation Standard Deviation** — measures performance stability across folds, where a lower value indicates greater stability.
4. **Absolute Generalization Gap** — measures consistency between cross-validation and independent test performance, where a smaller value indicates greater consistency.

### Ranking Method

Each model receives an individual rank for each criterion.

For Test Macro F1 and CV Macro F1:

- Higher performance receives a better rank.

For CV Macro F1 Standard Deviation and Absolute Generalization Gap:

- Lower values receive a better rank.

The final rank is determined using the sum of the individual criterion ranks.

A lower total rank score indicates stronger overall performance.

Training time is excluded because reliable standalone model-training measurements were not consistently recorded during model development.

````python


logger.info("=" * 70)
logger.info("13.8 MODEL RANKING")
logger.info("=" * 70)

# --------------------------------------------------------------
# 1. Validate Required Result Sources
# --------------------------------------------------------------

required_objects = [
    "final_generalization_comparison",
    "final_cv_stability_comparison"
]

missing_objects = [
    obj
    for obj in required_objects
    if obj not in globals()
]

if missing_objects:

    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

# --------------------------------------------------------------
# 2. Prepare Generalization Results
# --------------------------------------------------------------

ranking_source = (
    final_generalization_comparison
    .copy(deep=True)
)

ranking_source["Model"] = (
    ranking_source["Model"]
    .astype(str)
    .str.strip()
)

# --------------------------------------------------------------
# 3. Prepare CV Stability Results
# --------------------------------------------------------------

stability_source = (
    final_cv_stability_comparison
    .copy(deep=True)
)

stability_model_column = None
stability_sd_column = None

# Detect model column
for column in stability_source.columns:

    normalized = (
        str(column)
        .lower()
        .replace("_", " ")
        .strip()
    )

    if normalized == "model":

        stability_model_column = column
        break

# Detect Macro F1 SD column
for column in stability_source.columns:

    normalized = (
        str(column)
        .lower()
        .replace("_", " ")
        .replace("-", " ")
        .strip()
    )

    if (
        "macro" in normalized
        and "f1" in normalized
        and (
            "sd" in normalized
            or "std" in normalized
            or "standard deviation" in normalized
        )
    ):

        stability_sd_column = column
        break

if stability_model_column is None:

    raise KeyError(
        "Could not identify model column "
        "in CV stability results."
    )

if stability_sd_column is None:

    raise KeyError(
        "Could not identify CV Macro F1 "
        "standard deviation column."
    )

logger.info(
    "CV Stability Model Column              : %s",
    stability_model_column
)

logger.info(
    "CV Macro F1 SD Column                  : %s",
    stability_sd_column
)

# --------------------------------------------------------------
# 4. Standardize Stability Data
# --------------------------------------------------------------

stability_results = pd.DataFrame({

    "Model":
        stability_source[
            stability_model_column
        ]
        .astype(str)
        .str.strip(),

    "CV Macro F1 SD":
        pd.to_numeric(
            stability_source[
                stability_sd_column
            ],
            errors="coerce"
        )
})

# --------------------------------------------------------------
# 5. Merge Stability with Generalization
# --------------------------------------------------------------

ranking_source = pd.merge(
    ranking_source,
    stability_results,
    on="Model",
    how="left"
)

# --------------------------------------------------------------
# 6. Validate Stability Coverage
# --------------------------------------------------------------

missing_stability = ranking_source[
    "CV Macro F1 SD"
].isna()

if missing_stability.any():

    missing_models = (
        ranking_source.loc[
            missing_stability,
            "Model"
        ]
        .tolist()
    )

    raise ValueError(
        "Missing CV stability values for: "
        + ", ".join(missing_models)
    )

# --------------------------------------------------------------
# 7. Calculate Individual Ranks
# --------------------------------------------------------------

ranking_source[
    "Test Macro F1 Rank"
] = (
    ranking_source[
        "Test Macro F1"
    ]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)

ranking_source[
    "CV Macro F1 Rank"
] = (
    ranking_source[
        "CV Macro F1"
    ]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)

ranking_source[
    "CV Stability Rank"
] = (
    ranking_source[
        "CV Macro F1 SD"
    ]
    .rank(
        ascending=True,
        method="min"
    )
    .astype(int)
)

ranking_source[
    "Generalization Rank"
] = (
    ranking_source[
        "Absolute Gap"
    ]
    .rank(
        ascending=True,
        method="min"
    )
    .astype(int)
)

# --------------------------------------------------------------
# 8. Calculate Total Rank Score
# --------------------------------------------------------------

ranking_source[
    "Total Rank Score"
] = (
    ranking_source[
        [
            "Test Macro F1 Rank",
            "CV Macro F1 Rank",
            "CV Stability Rank",
            "Generalization Rank"
        ]
    ]
    .sum(axis=1)
)

# --------------------------------------------------------------
# 9. Final Overall Ranking
# --------------------------------------------------------------

ranking_source = (
    ranking_source
    .sort_values(
        by=[
            "Total Rank Score",
            "Test Macro F1",
            "CV Macro F1"
        ],
        ascending=[
            True,
            False,
            False
        ]
    )
    .reset_index(drop=True)
)

ranking_source[
    "Overall Rank"
] = (
    np.arange(
        1,
        len(ranking_source) + 1
    )
)

# --------------------------------------------------------------
# 10. Display Ranking
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL MULTI-CRITERIA MODEL RANKING")
logger.info("-" * 70)

display(
    ranking_source[
        [
            "Overall Rank",
            "Model",
            "Test Macro F1",
            "CV Macro F1",
            "CV Macro F1 SD",
            "Absolute Gap",
            "Test Macro F1 Rank",
            "CV Macro F1 Rank",
            "CV Stability Rank",
            "Generalization Rank",
            "Total Rank Score"
        ]
    ]
)

# --------------------------------------------------------------
# 11. Identify Top-Ranked Model
# --------------------------------------------------------------

top_ranked_model = (
    ranking_source.iloc[0]["Model"]
)

top_ranked_score = float(
    ranking_source.iloc[0]["Total Rank Score"]
)

logger.info(
    "Top-Ranked Model                      : %s",
    top_ranked_model
)

logger.info(
    "Top Model Total Rank Score            : %.0f",
    top_ranked_score
)

# --------------------------------------------------------------
# 12. Optimized Random Forest Ranking
# --------------------------------------------------------------

optimized_rf_ranking = (
    ranking_source[
        ranking_source["Model"]
        ==
        "Optimized Random Forest"
    ]
)

if optimized_rf_ranking.empty:

    raise AssertionError(
        "Optimized Random Forest was not found "
        "in the final ranking."
    )

optimized_rf_rank = int(
    optimized_rf_ranking.iloc[0][
        "Overall Rank"
    ]
)

optimized_rf_rank_score = float(
    optimized_rf_ranking.iloc[0][
        "Total Rank Score"
    ]
)

logger.info(
    "Optimized RF Overall Rank              : %d",
    optimized_rf_rank
)

logger.info(
    "Optimized RF Total Rank Score         : %.0f",
    optimized_rf_rank_score
)

# --------------------------------------------------------------
# 13. Supporting Criterion Leaders
# --------------------------------------------------------------

test_f1_leader = (
    ranking_source.loc[
        ranking_source[
            "Test Macro F1"
        ].idxmax(),
        "Model"
    ]
)

cv_f1_leader = (
    ranking_source.loc[
        ranking_source[
            "CV Macro F1"
        ].idxmax(),
        "Model"
    ]
)

stability_leader = (
    ranking_source.loc[
        ranking_source[
            "CV Macro F1 SD"
        ].idxmin(),
        "Model"
    ]
)

generalization_leader = (
    ranking_source.loc[
        ranking_source[
            "Absolute Gap"
        ].idxmin(),
        "Model"
    ]
)

logger.info(
    "Test Macro F1 Leader                  : %s",
    test_f1_leader
)

logger.info(
    "CV Macro F1 Leader                    : %s",
    cv_f1_leader
)

logger.info(
    "CV Stability Leader                   : %s",
    stability_leader
)

logger.info(
    "Generalization Consistency Leader    : %s",
    generalization_leader
)

# --------------------------------------------------------------
# 14. Validation Checks
# --------------------------------------------------------------

model_count_check = (
    len(ranking_source) == 7
)

unique_models_check = (
    ranking_source["Model"].is_unique
)

rank_sequence_check = (
    ranking_source["Overall Rank"].tolist()
    ==
    list(
        range(
            1,
            len(ranking_source) + 1
        )
    )
)

metric_validity_check = all([
    ranking_source[
        "Test Macro F1"
    ].between(0, 1).all(),

    ranking_source[
        "CV Macro F1"
    ].between(0, 1).all(),

    (
        ranking_source[
            "CV Macro F1 SD"
        ] >= 0
    ).all(),

    (
        ranking_source[
            "Absolute Gap"
        ] >= 0
    ).all()
])

optimized_rf_present_check = (
    optimized_rf_rank >= 1
    and
    optimized_rf_rank <= 7
)

# --------------------------------------------------------------
# 15. Store Final Ranking
# --------------------------------------------------------------

final_model_ranking = (
    ranking_source
    .copy(deep=True)
)

# --------------------------------------------------------------
# 16. Validation Logs
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("13.8 MODEL RANKING VALIDATION")
logger.info("-" * 70)

logger.info(
    "Seven Models Included                 : %s",
    "PASSED"
    if model_count_check
    else "FAILED"
)

logger.info(
    "Unique Models                          : %s",
    "PASSED"
    if unique_models_check
    else "FAILED"
)

logger.info(
    "Ranking Sequence Valid                 : %s",
    "PASSED"
    if rank_sequence_check
    else "FAILED"
)

logger.info(
    "Ranking Metrics Valid                  : %s",
    "PASSED"
    if metric_validity_check
    else "FAILED"
)

logger.info(
    "Optimized RF Ranking Valid             : %s",
    "PASSED"
    if optimized_rf_present_check
    else "FAILED"
)

# --------------------------------------------------------------
# 17. Final Validation
# --------------------------------------------------------------

if not all([
    model_count_check,
    unique_models_check,
    rank_sequence_check,
    metric_validity_check,
    optimized_rf_present_check
]):

    raise AssertionError(
        "One or more Section 13.8 validation checks failed."
    )

logger.info("=" * 70)
logger.info(
    "13.8 MODEL RANKING : COMPLETED"
)
logger.info("=" * 70)
````

## 13.9 Final Model Selection

This subsection performs the final model selection based on the complete evidence generated throughout the modeling and evaluation stages.

The final production model is selected using multiple complementary criteria:

- Independent test-set performance
- Cross-validation Macro F1
- Cross-validation stability
- Generalization gap
- Overall multi-criteria model ranking
- Per-class performance
- Multiclass ROC-AUC and Precision-Recall performance

The objective is not to select a model based on a single metric. Instead, the final model should provide the strongest balance between predictive performance, class-balanced performance, validation reliability, and generalization to unseen data.

The Optimized Random Forest is evaluated against all competing baseline and advanced models. Training-time ranking is intentionally excluded because reliable standalone training-time measurements were not consistently recorded for all models.

The selected model will be carried forward as the production candidate for the remaining stages of the project, including explainability, error analysis, serialization, input validation, and deployment.

## 14. **MODEL EXPLAINABILITY** ## 

## 14.1 Explainability Overview

Model explainability is performed to understand how the selected Optimized Random Forest model generates predictions.

The explainability analysis will evaluate:

- Global feature importance
- Permutation importance
- SHAP-based feature contribution
- Individual prediction explanations
- Class-specific feature importance
- Clinical feature interpretation
- Histopathological feature interpretation
- Explainability limitations

The final selected Optimized Random Forest from the hyperparameter optimization stage will be reused without retraining.

The analysis will operate on the same preprocessing representation used during model development. The preprocessing pipeline converts the original clinical and histopathological categorical variables into the model's one-hot encoded feature representation while handling the Age variable appropriately.

Feature importance indicates the contribution or influence of variables within the trained model. These results describe model behavior and statistical associations in the available dataset and must not be interpreted as medical causation or as a replacement for professional clinical diagnosis.

## 14.2 Global Feature Importance

Global feature importance provides an overall view of which transformed features contribute most strongly to the predictions of the selected Optimized Random Forest.

The Random Forest provides model-native feature importance values based on the reduction in impurity contributed by each feature across the ensemble of decision trees.

The analysis will:

- Extract feature importance values from the fitted Optimized Random Forest.
- Associate each importance value with its corresponding transformed feature.
- Rank all transformed features by importance.
- Identify the most influential features.
- Calculate cumulative importance.
- Visualize the top globally important features.

The analysis is performed on the 130-feature transformed representation generated by the fitted preprocessing pipeline.

These importance values describe how the model uses the available features for prediction. They do not establish clinical causality, diagnostic significance, or independent medical importance.

````python
# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
````

````python


logger.info("=" * 70)
logger.info("14.2 GLOBAL FEATURE IMPORTANCE")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# 1. Validate required explainability objects
# ----------------------------------------------------------------------

required_objects = [
    "final_explainability_model",
    "final_explainability_feature_names",
    "X_train_explain",
    "X_test_explain"
]

missing_objects = [
    obj_name
    for obj_name in required_objects
    if obj_name not in globals()
]

if missing_objects:
    raise NameError(
        f"Required explainability objects are missing: "
        f"{missing_objects}"
    )

# ----------------------------------------------------------------------
# 2. Validate Random Forest model
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL VALIDATION")
logger.info("-" * 70)

logger.info(
    f"Model Type                              : "
    f"{type(final_explainability_model).__name__}"
)

assert (
    type(final_explainability_model).__name__
    == "RandomForestClassifier"
)

# ----------------------------------------------------------------------
# 3. Extract native feature importance
# ----------------------------------------------------------------------

if not hasattr(
    final_explainability_model,
    "feature_importances_"
):
    raise AttributeError(
        "Random Forest model does not contain feature_importances_."
    )

feature_importances = np.asarray(
    final_explainability_model.feature_importances_,
    dtype=float
)

feature_names = np.asarray(
    final_explainability_feature_names
)

logger.info("-" * 70)
logger.info("FEATURE IMPORTANCE EXTRACTION")
logger.info("-" * 70)

logger.info(
    f"Number of Feature Names                  : "
    f"{len(feature_names)}"
)

logger.info(
    f"Number of Importance Values              : "
    f"{len(feature_importances)}"
)

# ----------------------------------------------------------------------
# 4. Validate feature alignment
# ----------------------------------------------------------------------

assert len(feature_names) == len(feature_importances)

assert (
    len(feature_importances)
    == X_train_explain.shape[1]
)

assert np.all(
    np.isfinite(feature_importances)
)

assert np.all(
    feature_importances >= 0
)

importance_sum = feature_importances.sum()

logger.info(
    f"Total Feature Importance                 : "
    f"{importance_sum:.6f}"
)

assert np.isclose(
    importance_sum,
    1.0,
    atol=1e-6
)

# ----------------------------------------------------------------------
# 5. Create feature importance dataframe
# ----------------------------------------------------------------------

global_feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": feature_importances
})

global_feature_importance = (
    global_feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

global_feature_importance["Rank"] = (
    np.arange(
        1,
        len(global_feature_importance) + 1
    )
)

global_feature_importance["Cumulative Importance"] = (
    global_feature_importance["Importance"]
    .cumsum()
)

# Reorder columns
global_feature_importance = (
    global_feature_importance[
        [
            "Rank",
            "Feature",
            "Importance",
            "Cumulative Importance"
        ]
    ]
)

# ----------------------------------------------------------------------
# 6. Summary statistics
# ----------------------------------------------------------------------

max_importance = float(
    global_feature_importance["Importance"].max()
)

mean_importance = float(
    global_feature_importance["Importance"].mean()
)

median_importance = float(
    global_feature_importance["Importance"].median()
)

non_zero_importance = int(
    (
        global_feature_importance["Importance"]
        > 0
    ).sum()
)

zero_importance = int(
    (
        global_feature_importance["Importance"]
        == 0
    ).sum()
)

logger.info("-" * 70)
logger.info("GLOBAL FEATURE IMPORTANCE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total Transformed Features               : "
    f"{len(global_feature_importance)}"
)

logger.info(
    f"Maximum Importance                       : "
    f"{max_importance:.6f}"
)

logger.info(
    f"Mean Importance                          : "
    f"{mean_importance:.6f}"
)

logger.info(
    f"Median Importance                        : "
    f"{median_importance:.6f}"
)

logger.info(
    f"Non-Zero Importance Features              : "
    f"{non_zero_importance}"
)

logger.info(
    f"Zero Importance Features                 : "
    f"{zero_importance}"
)

# ----------------------------------------------------------------------
# 7. Display top 20 features
# ----------------------------------------------------------------------

top_20_global_importance = (
    global_feature_importance
    .head(20)
    .copy()
)

logger.info("-" * 70)
logger.info("TOP 20 GLOBALLY IMPORTANT FEATURES")
logger.info("-" * 70)

display(
    top_20_global_importance.style.format({
        "Importance": "{:.6f}",
        "Cumulative Importance": "{:.4f}"
    })
)

# ----------------------------------------------------------------------
# 8. Calculate cumulative importance thresholds
# ----------------------------------------------------------------------

importance_thresholds = {}

for threshold in [0.50, 0.75, 0.90, 0.95]:

    features_required = int(
        (
            global_feature_importance[
                "Cumulative Importance"
            ]
            < threshold
        ).sum()
        + 1
    )

    features_required = min(
        features_required,
        len(global_feature_importance)
    )

    importance_thresholds[
        f"{int(threshold * 100)}%"
    ] = features_required

logger.info("-" * 70)
logger.info("CUMULATIVE IMPORTANCE ANALYSIS")
logger.info("-" * 70)

for threshold_label, feature_count in (
    importance_thresholds.items()
):

    logger.info(
        f"Features Required for {threshold_label:<4} "
        f"Importance                  : {feature_count}"
    )

# ----------------------------------------------------------------------
# 9. Visualization
# ----------------------------------------------------------------------

plot_data = (
    global_feature_importance
    .head(20)
    .sort_values(
        by="Importance",
        ascending=True
    )
)

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    plot_data["Feature"],
    plot_data["Importance"]
)

plt.xlabel(
    "Mean Decrease in Impurity Importance"
)

plt.ylabel(
    "Transformed Feature"
)

plt.title(
    "Top 20 Global Feature Importances — Optimized Random Forest"
)

plt.tight_layout()

plt.show()

# ----------------------------------------------------------------------
# 10. Store reusable objects
# ----------------------------------------------------------------------

final_global_feature_importance = (
    global_feature_importance.copy()
)

final_top_20_global_importance = (
    top_20_global_importance.copy()
)

final_importance_thresholds = (
    importance_thresholds.copy()
)

# ----------------------------------------------------------------------
# 11. Validation
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.2 GLOBAL FEATURE IMPORTANCE VALIDATION")
logger.info("-" * 70)

assert len(final_global_feature_importance) == 130

assert (
    final_global_feature_importance["Rank"]
    .tolist()
    == list(range(1, 131))
)

assert (
    final_global_feature_importance["Importance"]
    .isna()
    .sum()
    == 0
)

assert (
    final_global_feature_importance["Importance"]
    .min()
    >= 0
)

assert np.isclose(
    final_global_feature_importance["Importance"].sum(),
    1.0,
    atol=1e-6
)

assert len(final_top_20_global_importance) == 20

assert (
    final_global_feature_importance
    ["Importance"]
    .is_monotonic_decreasing
)

logger.info(
    "130 Features Evaluated                  : PASSED"
)

logger.info(
    "Feature Importance Alignment             : PASSED"
)

logger.info(
    "Importance Values Valid                  : PASSED"
)

logger.info(
    "Importance Sum Validation                : PASSED"
)

logger.info(
    "Ranking Sequence Valid                   : PASSED"
)

logger.info(
    "Top 20 Feature Extraction                : PASSED"
)

logger.info(
    "Visualization Generated                  : PASSED"
)

logger.info("=" * 70)
logger.info("14.2 GLOBAL FEATURE IMPORTANCE : COMPLETED")
logger.info("=" * 70)
````

## 14.3 Permutation Importance

Permutation importance provides a model-agnostic measure of feature importance by evaluating the reduction in predictive performance that occurs when individual features are randomly shuffled.

For each transformed feature, the feature values are permuted while all other features remain unchanged. The resulting model performance is compared with the original model performance.

Macro F1 is used as the primary scoring metric because the project is a multiclass classification problem with unequal class frequencies. Macro F1 gives equal importance to each target class and therefore provides a more balanced assessment than accuracy alone.

The analysis is performed on the independent test dataset using the already-trained Optimized Random Forest. The model is not retrained during permutation analysis.

The permutation importance results are interpreted as the model's dependence on individual features. A larger positive decrease in Macro F1 indicates that the model relies more strongly on that feature for predictive performance.

Permutation importance should not be interpreted as evidence of clinical causation.

````python
# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.inspection import permutation_importance
from sklearn.metrics import f1_score
````

## 14.4 SHAP Analysis

SHAP (SHapley Additive exPlanations) provides a unified framework for explaining the contribution of individual features to model predictions.

The SHAP analysis is performed using the selected Optimized Random Forest model and the transformed feature representation produced by the training-fitted preprocessing pipeline.

Because the project is a multiclass classification problem, SHAP values are analyzed across all target classes.

The analysis contains three components:

1. **Global SHAP Importance**  
   Average absolute SHAP values are used to identify features with the greatest overall contribution to model predictions.

2. **SHAP Summary**  
   SHAP contributions are examined across observations to understand the direction and magnitude of feature influence.

3. **Individual Prediction Explanation**  
   A representative test observation is examined to understand how individual features contribute to the model's class prediction.

SHAP values describe model behavior and feature contribution. They do not establish clinical causation, diagnostic validity, or biological importance.

````python
# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

!pip install shap
````

### 14.4.1 SHAP Setup and Global SHAP Importance

The SHAP TreeExplainer is used because the selected production candidate is a Random Forest classifier.

SHAP values are calculated on the transformed test dataset. The mean absolute SHAP value across observations and target classes is used as the global importance measure.

The analysis preserves the transformed feature names generated by the fitted preprocessing pipeline so that explanations remain aligned with the model input space.

````python
# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

try:
    import shap
except ImportError as exc:
    raise ImportError(
        "SHAP is not installed. Install it with: pip install shap"
    ) from exc
````

````python


logger.info("=" * 70)
logger.info("14.4.1 SHAP SETUP AND GLOBAL SHAP IMPORTANCE")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# 1. Validate required objects
# ----------------------------------------------------------------------

required_objects = [
    "final_explainability_model",
    "X_test_explain",
    "y_test",
    "final_explainability_feature_names"
]

missing_objects = [
    obj_name
    for obj_name in required_objects
    if obj_name not in globals()
]

if missing_objects:
    raise NameError(
        f"Required SHAP objects are missing: {missing_objects}"
    )

# ----------------------------------------------------------------------
# 2. Model and data validation
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SHAP CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Model Type                              : "
    f"{type(final_explainability_model).__name__}"
)

logger.info(
    f"Test Samples                            : "
    f"{X_test_explain.shape[0]}"
)

logger.info(
    f"Transformed Features                    : "
    f"{X_test_explain.shape[1]}"
)

logger.info(
    f"Target Classes                          : "
    f"{len(np.unique(y_test))}"
)

logger.info(
    "Explainer Type                          : TreeExplainer"
)

# ----------------------------------------------------------------------
# 3. Create TreeExplainer
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("INITIALIZING SHAP TREE EXPLAINER")
logger.info("-" * 70)

shap_explainer = shap.TreeExplainer(
    final_explainability_model
)

logger.info(
    "SHAP TreeExplainer Initialization        : COMPLETED"
)

# ----------------------------------------------------------------------
# 4. Calculate SHAP values
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING SHAP VALUES")
logger.info("-" * 70)

shap_output = shap_explainer.shap_values(
    X_test_explain
)

logger.info(
    "SHAP Value Calculation                   : COMPLETED"
)

# ----------------------------------------------------------------------
# 5. Normalize SHAP output across SHAP versions
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NORMALIZING MULTICLASS SHAP OUTPUT")
logger.info("-" * 70)

if isinstance(shap_output, list):

    # Older SHAP versions:
    # list[class] -> (samples, features)

    shap_values_by_class = np.stack(
        [
            np.asarray(class_values)
            for class_values in shap_output
        ],
        axis=0
    )

    # Shape:
    # (classes, samples, features)

    shap_output_format = (
        "List of class-specific arrays"
    )

else:

    shap_array = np.asarray(shap_output)

    if shap_array.ndim == 3:

        # Possible formats:
        # (samples, features, classes)
        # or
        # (classes, samples, features)

        n_samples = X_test_explain.shape[0]
        n_features = X_test_explain.shape[1]
        n_classes = len(np.unique(y_test))

        if (
            shap_array.shape[0] == n_samples
            and shap_array.shape[1] == n_features
        ):

            # Newer SHAP format:
            # (samples, features, classes)

            shap_values_by_class = np.transpose(
                shap_array,
                (2, 0, 1)
            )

            shap_output_format = (
                "3D array: samples × features × classes"
            )

        elif (
            shap_array.shape[0] == n_classes
            and shap_array.shape[1] == n_samples
        ):

            # Alternative format:
            # (classes, samples, features)

            shap_values_by_class = shap_array

            shap_output_format = (
                "3D array: classes × samples × features"
            )

        else:
            raise ValueError(
                "Unsupported 3D SHAP output shape: "
                f"{shap_array.shape}"
            )

    else:

        raise ValueError(
            "Unsupported SHAP output format. "
            f"Received array with shape: {shap_array.shape}"
        )

# ----------------------------------------------------------------------
# 6. Validate normalized SHAP dimensions
# ----------------------------------------------------------------------

expected_classes = len(np.unique(y_test))
expected_samples = X_test_explain.shape[0]
expected_features = X_test_explain.shape[1]

assert (
    shap_values_by_class.shape
    == (
        expected_classes,
        expected_samples,
        expected_features
    )
)

logger.info(
    f"SHAP Output Format                       : "
    f"{shap_output_format}"
)

logger.info(
    f"Normalized SHAP Shape                    : "
    f"{shap_values_by_class.shape}"
)

# ----------------------------------------------------------------------
# 7. Calculate global SHAP importance
# ----------------------------------------------------------------------

# Absolute SHAP contribution:
# classes × samples × features

absolute_shap_values = np.abs(
    shap_values_by_class
)

global_shap_importance = (
    absolute_shap_values
    .mean(axis=(0, 1))
)

assert len(global_shap_importance) == expected_features

# ----------------------------------------------------------------------
# 8. Create global SHAP dataframe
# ----------------------------------------------------------------------

global_shap_importance_df = pd.DataFrame({
    "Feature": final_explainability_feature_names,
    "Mean Absolute SHAP": global_shap_importance
})

global_shap_importance_df = (
    global_shap_importance_df
    .sort_values(
        by="Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

global_shap_importance_df["Rank"] = (
    np.arange(
        1,
        len(global_shap_importance_df) + 1
    )
)

global_shap_importance_df = (
    global_shap_importance_df[
        [
            "Rank",
            "Feature",
            "Mean Absolute SHAP"
        ]
    ]
)

# ----------------------------------------------------------------------
# 9. Summary statistics
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GLOBAL SHAP IMPORTANCE SUMMARY")
logger.info("-" * 70)

logger.info(
    f"Total SHAP Features                      : "
    f"{len(global_shap_importance_df)}"
)

logger.info(
    f"Maximum Mean Absolute SHAP               : "
    f"{global_shap_importance_df['Mean Absolute SHAP'].max():.6f}"
)

logger.info(
    f"Mean Absolute SHAP                       : "
    f"{global_shap_importance_df['Mean Absolute SHAP'].mean():.6f}"
)

logger.info(
    f"Median Absolute SHAP                     : "
    f"{global_shap_importance_df['Mean Absolute SHAP'].median():.6f}"
)

# ----------------------------------------------------------------------
# 10. Display top 20
# ----------------------------------------------------------------------

top_20_global_shap = (
    global_shap_importance_df
    .head(20)
    .copy()
)

logger.info("-" * 70)
logger.info("TOP 20 GLOBAL SHAP FEATURES")
logger.info("-" * 70)

display(
    top_20_global_shap.style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

# ----------------------------------------------------------------------
# 11. Visualization
# ----------------------------------------------------------------------

plot_data = (
    top_20_global_shap
    .sort_values(
        by="Mean Absolute SHAP",
        ascending=True
    )
)

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    plot_data["Feature"],
    plot_data["Mean Absolute SHAP"]
)

plt.xlabel(
    "Mean Absolute SHAP Value"
)

plt.ylabel(
    "Transformed Feature"
)

plt.title(
    "Top 20 Global SHAP Feature Importance"
)

plt.tight_layout()

plt.show()

# ----------------------------------------------------------------------
# 12. Store reusable objects
# ----------------------------------------------------------------------

final_shap_explainer = shap_explainer

final_shap_values_by_class = (
    shap_values_by_class
)

final_global_shap_importance = (
    global_shap_importance_df.copy()
)

final_top_20_global_shap = (
    top_20_global_shap.copy()
)

# ----------------------------------------------------------------------
# 13. Validation
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.4.1 SHAP GLOBAL IMPORTANCE VALIDATION")
logger.info("-" * 70)

assert (
    final_shap_values_by_class.shape
    == (
        expected_classes,
        expected_samples,
        expected_features
    )
)

assert (
    len(final_global_shap_importance)
    == expected_features
)

assert (
    final_global_shap_importance["Rank"]
    .tolist()
    == list(range(1, expected_features + 1))
)

assert (
    final_global_shap_importance[
        "Mean Absolute SHAP"
    ].isna().sum()
    == 0
)

assert np.all(
    np.isfinite(
        final_global_shap_importance[
            "Mean Absolute SHAP"
        ]
    )
)

assert np.all(
    final_global_shap_importance[
        "Mean Absolute SHAP"
    ] >= 0
)

assert (
    len(final_top_20_global_shap)
    == 20
)

logger.info(
    "SHAP Explainer Valid                     : PASSED"
)

logger.info(
    "Multiclass SHAP Shape Valid              : PASSED"
)

logger.info(
    "130 Features Evaluated                  : PASSED"
)

logger.info(
    "Global SHAP Ranking Valid                : PASSED"
)

logger.info(
    "SHAP Values Valid                        : PASSED"
)

logger.info(
    "Top 20 Feature Extraction                : PASSED"
)

logger.info(
    "Visualization Generated                  : PASSED"
)

logger.info("=" * 70)
logger.info("14.4.1 GLOBAL SHAP IMPORTANCE : COMPLETED")
logger.info("=" * 70)
````

## 14.4.2 SHAP Summary

The SHAP summary analysis provides a detailed view of how the transformed features influence the model's multiclass predictions across the test dataset.

Unlike global feature importance, which measures overall feature contribution magnitude, the SHAP summary additionally captures the **direction and distribution of feature contributions** across individual observations.

The analysis will:

- Visualize the distribution of SHAP contributions for the most influential features.
- Show both positive and negative feature contributions.
- Identify features that consistently influence predictions across observations.
- Use the multiclass SHAP values generated by the `TreeExplainer`.
- Aggregate multiclass SHAP values appropriately for global visualization.
- Restrict the analysis to the independent test dataset.
- Maintain the same transformed feature representation used by the production model.

This analysis improves model transparency by showing not only **which features are important**, but also **how their values influence model predictions**.

### SHAP Summary Objectives

1. Visualize the distribution of SHAP values across test observations.
2. Identify high-impact features.
3. Examine positive and negative prediction contributions.
4. Support interpretation of the Random Forest model.
5. Provide an industry-standard explainability visualization.

````python


# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

logger.info("=" * 70)
logger.info("14.4.2 SHAP SUMMARY")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# SHAP SUMMARY CONFIGURATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SHAP SUMMARY CONFIGURATION")
logger.info("-" * 70)

SHAP_SUMMARY_FEATURES = 20

logger.info(
    f"Features Displayed                       : {SHAP_SUMMARY_FEATURES}"
)
logger.info(
    f"Test Samples                             : {X_test_explain.shape[0]}"
)
logger.info(
    f"Transformed Features                     : {X_test_explain.shape[1]}"
)
logger.info(
    f"Target Classes                           : "
    f"{len(final_explainability_model.classes_)}"
)

# ------------------------------------------------------------------------------
# VALIDATE SHAP INPUT
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SHAP INPUT")
logger.info("-" * 70)

shap_values_array = np.asarray(final_shap_values_by_class)

logger.info(
    f"SHAP Array Shape                         : {shap_values_array.shape}"
)

assert shap_values_array.ndim == 3
assert shap_values_array.shape[0] == len(
    final_explainability_model.classes_
)
assert shap_values_array.shape[1] == X_test_explain.shape[0]
assert shap_values_array.shape[2] == X_test_explain.shape[1]

logger.info(
    "Multiclass SHAP Input Structure           : VALIDATED"
)

# ------------------------------------------------------------------------------
# PRESERVE TEST SAMPLE IDENTIFIERS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRESERVING TEST SAMPLE IDENTIFIERS")
logger.info("-" * 70)

# X_test_explain is a NumPy array.
# Therefore, it does not contain a pandas index.
#
# y_test retains the original test observation index in the
# standard train_test_split workflow.

if hasattr(y_test, "index"):
    test_sample_index = y_test.index
else:
    test_sample_index = np.arange(
        X_test_explain.shape[0]
    )

logger.info(
    f"Test Sample Identifiers                   : "
    f"{len(test_sample_index)}"
)

# ------------------------------------------------------------------------------
# CALCULATE MEAN ABSOLUTE SHAP
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING SHAP CONTRIBUTION MAGNITUDE")
logger.info("-" * 70)

# Shape:
# classes × samples × features
#
# First aggregate across classes.
mean_abs_shap_per_sample = np.mean(
    np.abs(shap_values_array),
    axis=0
)

# Shape:
# samples × features
logger.info(
    f"Per-Sample SHAP Magnitude Shape           : "
    f"{mean_abs_shap_per_sample.shape}"
)

# Global feature-level SHAP magnitude
global_mean_abs_shap = np.mean(
    mean_abs_shap_per_sample,
    axis=0
)

logger.info(
    f"Global SHAP Importance Shape              : "
    f"{global_mean_abs_shap.shape}"
)

# ------------------------------------------------------------------------------
# SELECT TOP SHAP FEATURES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SELECTING TOP SHAP FEATURES")
logger.info("-" * 70)

top_feature_indices = np.argsort(
    global_mean_abs_shap
)[::-1][:SHAP_SUMMARY_FEATURES]

top_feature_names = [
    final_explainability_feature_names[i]
    for i in top_feature_indices
]

top_feature_shap_values = mean_abs_shap_per_sample[
    :,
    top_feature_indices
]

logger.info(
    f"Top SHAP Features Selected                : "
    f"{len(top_feature_names)}"
)

# ------------------------------------------------------------------------------
# CREATE SHAP SUMMARY DATAFRAME
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING SHAP SUMMARY DATAFRAME")
logger.info("-" * 70)

shap_summary_df = pd.DataFrame(
    top_feature_shap_values,
    columns=top_feature_names,
    index=test_sample_index
)

logger.info(
    f"SHAP Summary DataFrame Shape              : "
    f"{shap_summary_df.shape}"
)

# ------------------------------------------------------------------------------
# CREATE SUMMARY STATISTICS
# ------------------------------------------------------------------------------

shap_summary_statistics = pd.DataFrame({
    "Feature": top_feature_names,
    "Mean Absolute SHAP": [
        global_mean_abs_shap[i]
        for i in top_feature_indices
    ],
    "Maximum Absolute SHAP": [
        np.max(
            np.abs(
                shap_values_array[:, :, i]
            )
        )
        for i in top_feature_indices
    ],
    "Rank": range(
        1,
        SHAP_SUMMARY_FEATURES + 1
    )
})

# Reorder columns
shap_summary_statistics = shap_summary_statistics[
    [
        "Rank",
        "Feature",
        "Mean Absolute SHAP",
        "Maximum Absolute SHAP"
    ]
]

display(
    shap_summary_statistics.style.format({
        "Mean Absolute SHAP": "{:.6f}",
        "Maximum Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# SUMMARY STATISTICS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SHAP SUMMARY STATISTICS")
logger.info("-" * 70)

logger.info(
    f"Highest Impact Feature                   : "
    f"{top_feature_names[0]}"
)

logger.info(
    f"Highest Mean Absolute SHAP                : "
    f"{global_mean_abs_shap[top_feature_indices[0]]:.6f}"
)

logger.info(
    f"Top-20 Mean Absolute SHAP Range           : "
    f"{global_mean_abs_shap[top_feature_indices[-1]]:.6f} - "
    f"{global_mean_abs_shap[top_feature_indices[0]]:.6f}"
)

logger.info(
    f"Overall Mean Absolute SHAP                : "
    f"{global_mean_abs_shap.mean():.6f}"
)

# ------------------------------------------------------------------------------
# GENERATE SHAP SUMMARY VISUALIZATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING SHAP SUMMARY VISUALIZATION")
logger.info("-" * 70)

plot_names = top_feature_names[::-1]
plot_values = [
    global_mean_abs_shap[i]
    for i in top_feature_indices[::-1]
]

plt.figure(figsize=(12, 9))

plt.barh(
    plot_names,
    plot_values
)

plt.xlabel("Mean Absolute SHAP Value")
plt.ylabel("Feature")
plt.title(
    "SHAP Summary — Top 20 Features"
)

plt.tight_layout()
plt.show()

logger.info(
    "SHAP Summary Visualization                : GENERATED"
)

# ------------------------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.4.2 SHAP SUMMARY VALIDATION")
logger.info("-" * 70)

assert isinstance(
    shap_summary_df,
    pd.DataFrame
)

assert shap_summary_df.shape == (
    X_test_explain.shape[0],
    SHAP_SUMMARY_FEATURES
)

assert len(shap_summary_statistics) == SHAP_SUMMARY_FEATURES

assert len(
    final_explainability_feature_names
) == X_test_explain.shape[1]

assert np.all(
    np.isfinite(global_mean_abs_shap)
)

assert np.all(
    np.diff(
        shap_summary_statistics[
            "Mean Absolute SHAP"
        ].values
    ) <= 0
)

assert list(
    shap_summary_statistics["Feature"]
) == top_feature_names

logger.info(
    "SHAP DataFrame Structure                  : PASSED"
)
logger.info(
    "Test Sample Alignment                     : PASSED"
)
logger.info(
    "Feature Count Validation                  : PASSED"
)
logger.info(
    "SHAP Numerical Validation                 : PASSED"
)
logger.info(
    "Feature Ranking Validation                : PASSED"
)
logger.info(
    "Top 20 Feature Extraction                 : PASSED"
)
logger.info(
    "Visualization Generated                   : PASSED"
)

# ------------------------------------------------------------------------------
# STORE RESULTS
# ------------------------------------------------------------------------------

final_shap_summary_statistics = (
    shap_summary_statistics.copy()
)

final_shap_summary_values = (
    shap_summary_df.copy()
)

final_top_shap_summary_features = (
    top_feature_names.copy()
)

logger.info("-" * 70)
logger.info("SHAP SUMMARY RESULTS STORED")
logger.info("-" * 70)

logger.info(
    "SHAP Summary Statistics                   : STORED"
)
logger.info(
    "SHAP Summary Values                       : STORED"
)
logger.info(
    "Top SHAP Summary Features                 : STORED"
)

logger.info("=" * 70)
logger.info("14.4.2 SHAP SUMMARY : COMPLETED")
logger.info("=" * 70)
````

The analysis will:

````python


# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

logger.info("=" * 70)
logger.info("14.4.3 INDIVIDUAL PREDICTION EXPLANATION")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# INDIVIDUAL EXPLANATION CONFIGURATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("INDIVIDUAL EXPLANATION CONFIGURATION")
logger.info("-" * 70)

TOP_POSITIVE_FEATURES = 10
TOP_NEGATIVE_FEATURES = 10

logger.info(
    f"Positive Contributions Displayed          : {TOP_POSITIVE_FEATURES}"
)
logger.info(
    f"Negative Contributions Displayed          : {TOP_NEGATIVE_FEATURES}"
)
logger.info(
    f"Available Test Samples                    : {X_test_explain.shape[0]}"
)
logger.info(
    f"Available Transformed Features            : {X_test_explain.shape[1]}"
)

# ------------------------------------------------------------------------------
# VALIDATE REQUIRED SHAP OBJECTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED SHAP OBJECTS")
logger.info("-" * 70)

shap_values_array = np.asarray(
    final_shap_values_by_class
)

test_predictions = np.asarray(
    y_test_pred
)

test_actual = np.asarray(
    y_test
)

test_probabilities = np.asarray(
    y_test_proba
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

logger.info(
    f"SHAP Shape                               : {shap_values_array.shape}"
)
logger.info(
    f"Prediction Shape                          : {test_predictions.shape}"
)
logger.info(
    f"Actual Target Shape                       : {test_actual.shape}"
)
logger.info(
    f"Probability Matrix Shape                  : {test_probabilities.shape}"
)
logger.info(
    f"Model Classes                             : {model_classes}"
)

assert shap_values_array.shape[1] == len(test_predictions)
assert shap_values_array.shape[1] == len(test_actual)
assert shap_values_array.shape[1] == test_probabilities.shape[0]
assert shap_values_array.shape[2] == X_test_explain.shape[1]
assert shap_values_array.shape[0] == len(model_classes)

logger.info(
    "Required SHAP Objects                     : VALIDATED"
)

# ------------------------------------------------------------------------------
# IDENTIFY CORRECTLY CLASSIFIED OBSERVATIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING REPRESENTATIVE TEST OBSERVATION")
logger.info("-" * 70)

correct_prediction_mask = (
    test_actual == test_predictions
)

correct_indices = np.where(
    correct_prediction_mask
)[0]

logger.info(
    f"Correct Predictions                       : "
    f"{len(correct_indices)}"
)

logger.info(
    f"Incorrect Predictions                     : "
    f"{len(test_actual) - len(correct_indices)}"
)

assert len(correct_indices) > 0, (
    "No correctly classified test observation available."
)

# Select the first correctly classified observation
selected_sample_position = int(
    correct_indices[0]
)

logger.info(
    f"Selected Test Position                    : "
    f"{selected_sample_position}"
)

# ------------------------------------------------------------------------------
# IDENTIFY ORIGINAL TEST SAMPLE INDEX
# ------------------------------------------------------------------------------

if hasattr(y_test, "index"):
    selected_sample_index = y_test.index[
        selected_sample_position
    ]
else:
    selected_sample_index = selected_sample_position

logger.info(
    f"Selected Test Sample Index                : "
    f"{selected_sample_index}"
)

# ------------------------------------------------------------------------------
# EXTRACT ACTUAL AND PREDICTED CLASS
# ------------------------------------------------------------------------------

actual_class = test_actual[
    selected_sample_position
]

predicted_class = test_predictions[
    selected_sample_position
]

prediction_probabilities = test_probabilities[
    selected_sample_position
]

predicted_class_position = np.where(
    model_classes == predicted_class
)[0]

assert len(predicted_class_position) == 1

predicted_class_position = int(
    predicted_class_position[0]
)

prediction_confidence = prediction_probabilities[
    predicted_class_position
]

logger.info("-" * 70)
logger.info("PREDICTION DETAILS")
logger.info("-" * 70)

logger.info(
    f"Actual Class                             : "
    f"{actual_class}"
)

logger.info(
    f"Predicted Class                          : "
    f"{predicted_class}"
)

logger.info(
    f"Prediction Confidence                    : "
    f"{prediction_confidence:.4f}"
)

logger.info(
    f"Prediction Correct                       : "
    f"{actual_class == predicted_class}"
)

# ------------------------------------------------------------------------------
# DISPLAY CLASS PROBABILITIES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS PROBABILITY DISTRIBUTION")
logger.info("-" * 70)

class_probability_df = pd.DataFrame({
    "Class": model_classes,
    "Probability": prediction_probabilities
})

class_probability_df["Percentage"] = (
    class_probability_df["Probability"] * 100
)

class_probability_df = class_probability_df.sort_values(
    "Probability",
    ascending=False
).reset_index(drop=True)

display(
    class_probability_df.style.format({
        "Probability": "{:.6f}",
        "Percentage": "{:.2f}%"
    })
)

# ------------------------------------------------------------------------------
# EXTRACT SHAP VALUES FOR PREDICTED CLASS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRACTING PREDICTED-CLASS SHAP CONTRIBUTIONS")
logger.info("-" * 70)

# SHAP values for:
# selected class × selected observation × all features

selected_class_shap_values = (
    shap_values_array[
        predicted_class_position,
        selected_sample_position,
        :
    ]
)

assert len(
    selected_class_shap_values
) == len(final_explainability_feature_names)

logger.info(
    f"Predicted Class SHAP Vector Shape          : "
    f"{selected_class_shap_values.shape}"
)

# ------------------------------------------------------------------------------
# CREATE LOCAL SHAP CONTRIBUTION DATAFRAME
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING LOCAL SHAP CONTRIBUTION TABLE")
logger.info("-" * 70)

local_shap_df = pd.DataFrame({
    "Feature": final_explainability_feature_names,
    "SHAP Value": selected_class_shap_values,
    "Absolute SHAP": np.abs(selected_class_shap_values)
})

local_shap_df = local_shap_df.sort_values(
    "Absolute SHAP",
    ascending=False
).reset_index(drop=True)

local_shap_df["Overall Rank"] = (
    np.arange(len(local_shap_df)) + 1
)

# ------------------------------------------------------------------------------
# POSITIVE CONTRIBUTIONS
# ------------------------------------------------------------------------------

positive_shap_df = local_shap_df[
    local_shap_df["SHAP Value"] > 0
].sort_values(
    "SHAP Value",
    ascending=False
).head(
    TOP_POSITIVE_FEATURES
).copy()

positive_shap_df["Direction"] = (
    "Toward predicted class"
)

positive_shap_df = positive_shap_df[
    [
        "Feature",
        "SHAP Value",
        "Absolute SHAP",
        "Direction"
    ]
]

# ------------------------------------------------------------------------------
# NEGATIVE CONTRIBUTIONS
# ------------------------------------------------------------------------------

negative_shap_df = local_shap_df[
    local_shap_df["SHAP Value"] < 0
].sort_values(
    "SHAP Value",
    ascending=True
).head(
    TOP_NEGATIVE_FEATURES
).copy()

negative_shap_df["Direction"] = (
    "Away from predicted class"
)

negative_shap_df = negative_shap_df[
    [
        "Feature",
        "SHAP Value",
        "Absolute SHAP",
        "Direction"
    ]
]

# ------------------------------------------------------------------------------
# DISPLAY LOCAL SHAP CONTRIBUTIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("TOP POSITIVE SHAP CONTRIBUTIONS")
logger.info("-" * 70)

display(
    positive_shap_df.style.format({
        "SHAP Value": "{:.6f}",
        "Absolute SHAP": "{:.6f}"
    })
)

logger.info("-" * 70)
logger.info("TOP NEGATIVE SHAP CONTRIBUTIONS")
logger.info("-" * 70)

display(
    negative_shap_df.style.format({
        "SHAP Value": "{:.6f}",
        "Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# LOCAL SHAP SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOCAL SHAP CONTRIBUTION SUMMARY")
logger.info("-" * 70)

positive_contribution_count = int(
    np.sum(selected_class_shap_values > 0)
)

negative_contribution_count = int(
    np.sum(selected_class_shap_values < 0)
)

zero_contribution_count = int(
    np.sum(selected_class_shap_values == 0)
)

logger.info(
    f"Positive Contributions                   : "
    f"{positive_contribution_count}"
)

logger.info(
    f"Negative Contributions                   : "
    f"{negative_contribution_count}"
)

logger.info(
    f"Zero Contributions                       : "
    f"{zero_contribution_count}"
)

logger.info(
    f"Maximum Positive SHAP                    : "
    f"{selected_class_shap_values.max():.6f}"
)

logger.info(
    f"Maximum Negative SHAP                    : "
    f"{selected_class_shap_values.min():.6f}"
)

logger.info(
    f"Mean Absolute Local SHAP                 : "
    f"{np.mean(np.abs(selected_class_shap_values)):.6f}"
)

# ------------------------------------------------------------------------------
# VISUALIZE POSITIVE AND NEGATIVE CONTRIBUTIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING INDIVIDUAL SHAP VISUALIZATION")
logger.info("-" * 70)

visualization_df = pd.concat([
    positive_shap_df.assign(
        Contribution=positive_shap_df["SHAP Value"]
    ),
    negative_shap_df.assign(
        Contribution=negative_shap_df["SHAP Value"]
    )
])

visualization_df = visualization_df.sort_values(
    "Contribution"
)

plt.figure(figsize=(12, 10))

plt.barh(
    visualization_df["Feature"],
    visualization_df["Contribution"]
)

plt.axvline(
    x=0,
    linewidth=1
)

plt.xlabel(
    "SHAP Contribution"
)

plt.ylabel(
    "Feature"
)

plt.title(
    f"Individual SHAP Explanation — Predicted Class {predicted_class}"
)

plt.tight_layout()
plt.show()

logger.info(
    "Individual SHAP Visualization              : GENERATED"
)

# ------------------------------------------------------------------------------
# CREATE INDIVIDUAL EXPLANATION SUMMARY
# ------------------------------------------------------------------------------

individual_prediction_explanation = pd.DataFrame({
    "Test Sample Index": [selected_sample_index],
    "Actual Class": [actual_class],
    "Predicted Class": [predicted_class],
    "Prediction Confidence": [prediction_confidence],
    "Prediction Correct": [
        actual_class == predicted_class
    ],
    "Positive SHAP Features": [
        positive_contribution_count
    ],
    "Negative SHAP Features": [
        negative_contribution_count
    ],
    "Zero SHAP Features": [
        zero_contribution_count
    ]
})

logger.info("-" * 70)
logger.info("INDIVIDUAL PREDICTION SUMMARY")
logger.info("-" * 70)

display(
    individual_prediction_explanation.style.format({
        "Prediction Confidence": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.4.3 INDIVIDUAL PREDICTION EXPLANATION VALIDATION")
logger.info("-" * 70)

assert actual_class == predicted_class, (
    "Selected observation is not correctly classified."
)

assert len(selected_class_shap_values) == (
    X_test_explain.shape[1]
)

assert len(selected_class_shap_values) == (
    len(final_explainability_feature_names)
)

assert np.all(
    np.isfinite(selected_class_shap_values)
)

assert (
    positive_contribution_count
    + negative_contribution_count
    + zero_contribution_count
    == X_test_explain.shape[1]
)

assert (
    len(class_probability_df)
    == len(model_classes)
)

assert np.isclose(
    prediction_probabilities.sum(),
    1.0,
    atol=1e-6
)

assert (
    predicted_class
    == class_probability_df.iloc[0]["Class"]
)

assert len(
    individual_prediction_explanation
) == 1

logger.info(
    "Correctly Classified Observation           : PASSED"
)
logger.info(
    "SHAP Feature Alignment                     : PASSED"
)
logger.info(
    "SHAP Numerical Validation                  : PASSED"
)
logger.info(
    "Contribution Count Validation              : PASSED"
)
logger.info(
    "Class Probability Validation               : PASSED"
)
logger.info(
    "Predicted Class Validation                 : PASSED"
)
logger.info(
    "Individual Summary Validation               : PASSED"
)
logger.info(
    "Visualization Generated                    : PASSED"
)

# ------------------------------------------------------------------------------
# STORE RESULTS
# ------------------------------------------------------------------------------

final_individual_explanation = (
    individual_prediction_explanation.copy()
)

final_individual_class_probabilities = (
    class_probability_df.copy()
)

final_individual_positive_shap = (
    positive_shap_df.copy()
)

final_individual_negative_shap = (
    negative_shap_df.copy()
)

final_individual_shap_values = (
    local_shap_df.copy()
)

logger.info("-" * 70)
logger.info("INDIVIDUAL EXPLANATION RESULTS STORED")
logger.info("-" * 70)

logger.info(
    "Individual Prediction Summary              : STORED"
)
logger.info(
    "Class Probability Distribution              : STORED"
)
logger.info(
    "Positive SHAP Contributions                 : STORED"
)
logger.info(
    "Negative SHAP Contributions                 : STORED"
)
logger.info(
    "Complete Individual SHAP Values             : STORED"
)

logger.info("=" * 70)
logger.info(
    "14.4.3 INDIVIDUAL PREDICTION EXPLANATION : COMPLETED"
)
logger.info("=" * 70)
````

## 14.5 Class-Specific Feature Importance

Class-specific feature importance analyzes the contribution of individual features separately for each target class.

Because the Skin Disorder Prediction problem contains six disease classes, a feature may be highly influential for one class while having relatively little influence on another.

The class-specific analysis will:

- Calculate mean absolute SHAP importance separately for each disease class.
- Rank the transformed features for every class.
- Identify the top contributing features for each class.
- Compare feature importance patterns across the six classes.
- Highlight features that are particularly influential for specific disease classes.
- Use the multiclass SHAP values generated by the `TreeExplainer`.
- Restrict the analysis to the independent test dataset.
- Avoid introducing any additional model training or data leakage.

Mean absolute SHAP values are used because they measure the magnitude of a feature's contribution to a specific class regardless of contribution direction.

This analysis provides a more detailed understanding of the multiclass Random Forest decision process and supports subsequent clinical and histopathological interpretation.

````python


# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

logger.info("=" * 70)
logger.info("14.5 CLASS-SPECIFIC FEATURE IMPORTANCE")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# CLASS-SPECIFIC CONFIGURATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS-SPECIFIC SHAP CONFIGURATION")
logger.info("-" * 70)

TOP_CLASS_FEATURES = 10

logger.info(
    f"Top Features Per Class                    : {TOP_CLASS_FEATURES}"
)
logger.info(
    f"Test Samples                              : {X_test_explain.shape[0]}"
)
logger.info(
    f"Transformed Features                      : {X_test_explain.shape[1]}"
)
logger.info(
    f"Target Classes                            : "
    f"{len(final_explainability_model.classes_)}"
)

# ------------------------------------------------------------------------------
# VALIDATE SHAP STRUCTURE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MULTICLASS SHAP STRUCTURE")
logger.info("-" * 70)

shap_values_array = np.asarray(
    final_shap_values_by_class
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

assert shap_values_array.ndim == 3

assert shap_values_array.shape[0] == len(
    model_classes
)

assert shap_values_array.shape[1] == (
    X_test_explain.shape[0]
)

assert shap_values_array.shape[2] == (
    X_test_explain.shape[1]
)

assert len(
    final_explainability_feature_names
) == X_test_explain.shape[1]

logger.info(
    f"SHAP Shape                                : "
    f"{shap_values_array.shape}"
)

logger.info(
    f"Model Classes                             : "
    f"{model_classes}"
)

logger.info(
    "Multiclass SHAP Structure                 : VALIDATED"
)

# ------------------------------------------------------------------------------
# CALCULATE CLASS-SPECIFIC MEAN ABSOLUTE SHAP
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING CLASS-SPECIFIC SHAP IMPORTANCE")
logger.info("-" * 70)

# SHAP structure:
# classes × samples × features
#
# Mean absolute SHAP across test samples:
# classes × features

class_mean_abs_shap = np.mean(
    np.abs(shap_values_array),
    axis=1
)

logger.info(
    f"Class-Specific SHAP Matrix Shape           : "
    f"{class_mean_abs_shap.shape}"
)

# ------------------------------------------------------------------------------
# CREATE CLASS-SPECIFIC IMPORTANCE MATRIX
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING CLASS-SPECIFIC IMPORTANCE MATRIX")
logger.info("-" * 70)

class_columns = [
    f"Class {class_label}"
    for class_label in model_classes
]

class_specific_importance_df = pd.DataFrame(
    class_mean_abs_shap.T,
    index=final_explainability_feature_names,
    columns=class_columns
)

class_specific_importance_df.index.name = "Feature"

display(
    class_specific_importance_df.head(10).style.format(
        "{:.6f}"
    )
)

# ------------------------------------------------------------------------------
# GENERATE TOP FEATURES FOR EACH CLASS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING TOP FEATURES FOR EACH CLASS")
logger.info("-" * 70)

class_specific_top_features = {}

for class_position, class_label in enumerate(
    model_classes
):

    class_importance_values = class_mean_abs_shap[
        class_position
    ]

    ranked_indices = np.argsort(
        class_importance_values
    )[::-1]

    top_indices = ranked_indices[
        :TOP_CLASS_FEATURES
    ]

    class_top_df = pd.DataFrame({
        "Rank": np.arange(
            1,
            TOP_CLASS_FEATURES + 1
        ),
        "Feature": [
            final_explainability_feature_names[i]
            for i in top_indices
        ],
        "Mean Absolute SHAP": [
            class_importance_values[i]
            for i in top_indices
        ]
    })

    class_specific_top_features[
        class_label
    ] = class_top_df

    logger.info(
        f"Class {class_label} Top Feature             : "
        f"{class_top_df.iloc[0]['Feature']}"
    )

    logger.info(
        f"Class {class_label} Maximum Mean Abs SHAP   : "
        f"{class_top_df.iloc[0]['Mean Absolute SHAP']:.6f}"
    )

# ------------------------------------------------------------------------------
# DISPLAY TOP FEATURES FOR EACH CLASS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS-SPECIFIC TOP FEATURE RANKINGS")
logger.info("-" * 70)

for class_label in model_classes:

    logger.info(
        f"TOP FEATURES — CLASS {class_label}"
    )

    display(
        class_specific_top_features[
            class_label
        ].style.format({
            "Mean Absolute SHAP": "{:.6f}"
        })
    )

# ------------------------------------------------------------------------------
# CREATE TOP FEATURES BY CLASS SUMMARY TABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING CLASS-SPECIFIC SUMMARY TABLE")
logger.info("-" * 70)

class_top_feature_summary = []

for class_label in model_classes:

    top_df = class_specific_top_features[
        class_label
    ]

    for _, row in top_df.iterrows():

        class_top_feature_summary.append({
            "Class": class_label,
            "Rank": int(row["Rank"]),
            "Feature": row["Feature"],
            "Mean Absolute SHAP": row[
                "Mean Absolute SHAP"
            ]
        })

class_top_feature_summary_df = pd.DataFrame(
    class_top_feature_summary
)

display(
    class_top_feature_summary_df.style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# IDENTIFY CROSS-CLASSALLY IMPORTANT FEATURES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING CROSS-CLASS FEATURE IMPORTANCE")
logger.info("-" * 70)

# Average importance across all six classes
class_specific_importance_df[
    "Mean Across Classes"
] = class_specific_importance_df[
    class_columns
].mean(
    axis=1
)

# Maximum class-specific importance
class_specific_importance_df[
    "Maximum Class Importance"
] = class_specific_importance_df[
    class_columns
].max(
    axis=1
)

cross_class_importance_df = (
    class_specific_importance_df[
        [
            "Mean Across Classes",
            "Maximum Class Importance"
        ]
    ]
    .sort_values(
        "Mean Across Classes",
        ascending=False
    )
    .reset_index()
)

display(
    cross_class_importance_df.head(20).style.format({
        "Mean Across Classes": "{:.6f}",
        "Maximum Class Importance": "{:.6f}"
    })
)

logger.info(
    f"Most Consistently Important Feature      : "
    f"{cross_class_importance_df.iloc[0]['Feature']}"
)

logger.info(
    f"Highest Mean Cross-Class SHAP             : "
    f"{cross_class_importance_df.iloc[0]['Mean Across Classes']:.6f}"
)

# ------------------------------------------------------------------------------
# IDENTIFY CLASS-SPECIFIC SPECIALIZED FEATURES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING CLASS-SPECIALIZED FEATURES")
logger.info("-" * 70)

class_specialized_features = []

for feature in final_explainability_feature_names:

    feature_values = (
        class_specific_importance_df.loc[
            feature,
            class_columns
        ].values
    )

    maximum_value = np.max(
        feature_values
    )

    minimum_value = np.min(
        feature_values
    )

    concentration_ratio = (
        maximum_value /
        (minimum_value + 1e-12)
    )

    highest_class_position = int(
        np.argmax(feature_values)
    )

    highest_class = model_classes[
        highest_class_position
    ]

    class_specialized_features.append({
        "Feature": feature,
        "Highest Importance Class": highest_class,
        "Maximum Mean Absolute SHAP": maximum_value,
        "Minimum Mean Absolute SHAP": minimum_value,
        "Class Concentration Ratio": concentration_ratio
    })

class_specialized_features_df = pd.DataFrame(
    class_specialized_features
)

class_specialized_features_df = (
    class_specialized_features_df
    .sort_values(
        "Class Concentration Ratio",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    class_specialized_features_df.head(20).style.format({
        "Maximum Mean Absolute SHAP": "{:.6f}",
        "Minimum Mean Absolute SHAP": "{:.6f}",
        "Class Concentration Ratio": "{:.2f}"
    })
)

# ------------------------------------------------------------------------------
# CREATE CLASS-SPECIFIC IMPORTANCE HEATMAP
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING CLASS-SPECIFIC IMPORTANCE HEATMAP")
logger.info("-" * 70)

# Select 20 features with highest mean importance across classes
heatmap_features = (
    cross_class_importance_df
    .head(20)["Feature"]
    .tolist()
)

heatmap_data = class_specific_importance_df.loc[
    heatmap_features,
    class_columns
]

plt.figure(
    figsize=(11, 10)
)

plt.imshow(
    heatmap_data.values,
    aspect="auto"
)

plt.xticks(
    ticks=np.arange(
        len(model_classes)
    ),
    labels=class_columns
)

plt.yticks(
    ticks=np.arange(
        len(heatmap_features)
    ),
    labels=heatmap_features
)

plt.xlabel(
    "Disease Class"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Class-Specific SHAP Importance — Top 20 Features"
)

plt.colorbar(
    label="Mean Absolute SHAP"
)

plt.tight_layout()
plt.show()

logger.info(
    "Class-Specific Heatmap                    : GENERATED"
)

# ------------------------------------------------------------------------------
# CLASS-SPECIFIC IMPORTANCE STATISTICS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS-SPECIFIC IMPORTANCE STATISTICS")
logger.info("-" * 70)

for class_position, class_label in enumerate(
    model_classes
):

    class_values = class_mean_abs_shap[
        class_position
    ]

    top_index = np.argmax(
        class_values
    )

    logger.info(
        f"Class {class_label} Maximum SHAP           : "
        f"{class_values[top_index]:.6f}"
    )

    logger.info(
        f"Class {class_label} Mean SHAP              : "
        f"{class_values.mean():.6f}"
    )

    logger.info(
        f"Class {class_label} Median SHAP            : "
        f"{np.median(class_values):.6f}"
    )

# ------------------------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.5 CLASS-SPECIFIC FEATURE IMPORTANCE VALIDATION")
logger.info("-" * 70)

# Core SHAP matrix:
# 6 classes × 130 features
assert class_mean_abs_shap.shape == (
    len(model_classes),
    X_test_explain.shape[1]
)

# Class-specific dataframe:
# 130 features × 6 class columns
#
# Two summary columns are added later:
# Mean Across Classes
# Maximum Class Importance
#
# Therefore final dataframe:
# 130 × 8

assert class_specific_importance_df.shape == (
    X_test_explain.shape[1],
    len(model_classes) + 2
)

# Verify exactly six class columns exist
assert all(
    class_column in class_specific_importance_df.columns
    for class_column in class_columns
)

# Verify number of class columns
assert len(
    class_columns
) == len(model_classes)

# Verify per-class top feature dictionaries
assert len(
    class_specific_top_features
) == len(model_classes)

for class_label in model_classes:

    current_top = (
        class_specific_top_features[
            class_label
        ]
    )

    # Exactly 10 features per class
    assert len(current_top) == TOP_CLASS_FEATURES

    # SHAP importance must be descending
    assert np.all(
        np.diff(
            current_top[
                "Mean Absolute SHAP"
            ].values
        ) <= 0
    )

    # Ranks must be sequential
    assert list(
        current_top["Rank"]
    ) == list(
        range(
            1,
            TOP_CLASS_FEATURES + 1
        )
    )

# Numerical validation
assert np.all(
    np.isfinite(
        class_mean_abs_shap
    )
)

assert np.all(
    np.isfinite(
        class_specific_importance_df[
            class_columns
        ].values
    )
)

# Cross-class analysis validation
assert len(
    cross_class_importance_df
) == X_test_explain.shape[1]

# Specialized feature analysis validation
assert len(
    class_specialized_features_df
) == X_test_explain.shape[1]

# Heatmap validation
assert len(
    heatmap_features
) == 20

assert heatmap_data.shape == (
    20,
    len(model_classes)
)

logger.info(
    "Class-Specific SHAP Matrix                 : PASSED"
)

logger.info(
    "Class Count Validation                     : PASSED"
)

logger.info(
    "Feature Count Validation                   : PASSED"
)

logger.info(
    "Class Column Validation                    : PASSED"
)

logger.info(
    "Per-Class Ranking Validation               : PASSED"
)

logger.info(
    "SHAP Numerical Validation                  : PASSED"
)

logger.info(
    "Cross-Class Importance Analysis            : PASSED"
)

logger.info(
    "Class-Specialized Feature Analysis         : PASSED"
)

logger.info(
    "Heatmap Structure Validation               : PASSED"
)

logger.info(
    "Visualization Generated                    : PASSED"
)

# ------------------------------------------------------------------------------
# STORE RESULTS
# ------------------------------------------------------------------------------

final_class_specific_shap_importance = (
    class_specific_importance_df.copy()
)

final_class_specific_top_features = {
    class_label: df.copy()
    for class_label, df
    in class_specific_top_features.items()
}

final_class_top_feature_summary = (
    class_top_feature_summary_df.copy()
)

final_cross_class_shap_importance = (
    cross_class_importance_df.copy()
)

final_class_specialized_features = (
    class_specialized_features_df.copy()
)

logger.info("-" * 70)
logger.info("CLASS-SPECIFIC EXPLAINABILITY RESULTS STORED")
logger.info("-" * 70)

logger.info(
    "Class-Specific SHAP Importance              : STORED"
)

logger.info(
    "Per-Class Top Features                     : STORED"
)

logger.info(
    "Class Feature Summary                      : STORED"
)

logger.info(
    "Cross-Class SHAP Importance                : STORED"
)

logger.info(
    "Class-Specialized Features                 : STORED"
)

logger.info("=" * 70)
logger.info(
    "14.5 CLASS-SPECIFIC FEATURE IMPORTANCE : COMPLETED"
)
logger.info("=" * 70)
````

## 14.6 Clinical Feature Interpretation

Clinical feature interpretation evaluates how patient-level clinical observations and visible disease characteristics contribute to the predictions of the selected Random Forest model.

The analysis uses the validated multiclass SHAP values generated in Section 14.4 and the permutation importance results generated in Section 14.3.

Clinical variables considered in this analysis include:

- Age
- Itching
- Erythema
- Scaling
- Definite borders
- Koebner phenomenon
- Polygonal papules
- Follicular papules
- Oral mucosal involvement
- Knee and elbow involvement
- Scalp involvement
- Family history

Because categorical variables were represented using one-hot encoding, each transformed feature corresponds to a specific category value of the original clinical variable.

The analysis therefore:

- Maps transformed model features to their original clinical variables.
- Calculates clinical feature importance directly from the validated SHAP tensor.
- Aggregates category-level SHAP importance to the original clinical feature level.
- Identifies the strongest category-specific contribution for each clinical variable.
- Compares clinical SHAP importance with permutation importance.
- Visualizes the most influential clinical variables.

The interpretation describes model behavior rather than clinical causation. Feature importance does not establish that a clinical observation causes a disease or that the model's learned relationship is medically diagnostic.

Clinical interpretation should therefore be considered exploratory and should be validated independently by qualified medical experts and on larger external datasets.

````python


# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

logger.info("=" * 70)
logger.info("14.6 CLINICAL FEATURE INTERPRETATION")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. CLINICAL FEATURE CONFIGURATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING CLINICAL FEATURES")
logger.info("-" * 70)

clinical_feature_names = [
    "Age",
    "itching",
    "erythema",
    "scaling",
    "definite_borders",
    "koebner_phenomenon",
    "polygonal_papules",
    "follicular_papules",
    "oral_mucosal_involvement",
    "knee_and_elbow_involvement",
    "scalp_involvement",
    "family_history"
]

original_feature_names = list(
    X_train.columns
)

clinical_features_available = [
    feature
    for feature in clinical_feature_names
    if feature in original_feature_names
]

clinical_features_missing = [
    feature
    for feature in clinical_feature_names
    if feature not in original_feature_names
]

logger.info(
    f"Clinical Features Requested                : "
    f"{len(clinical_feature_names)}"
)

logger.info(
    f"Clinical Features Available               : "
    f"{len(clinical_features_available)}"
)

if clinical_features_missing:

    logger.info(
        f"Clinical Features Not Available         : "
        f"{clinical_features_missing}"
    )

logger.info(
    f"Available Clinical Features               : "
    f"{clinical_features_available}"
)

assert len(
    clinical_features_available
) > 0

# ------------------------------------------------------------------------------
# 2. VALIDATE EXISTING SHAP VALUES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING EXISTING MULTICLASS SHAP VALUES")
logger.info("-" * 70)

# IMPORTANT:
# We use the already validated SHAP tensor from Section 14.4.
#
# Shape:
#     classes × test_samples × transformed_features
#
# Expected:
#     6 × 74 × 130

shap_values_array = np.asarray(
    final_shap_values_by_class
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

logger.info(
    f"SHAP Tensor Shape                         : "
    f"{shap_values_array.shape}"
)

logger.info(
    f"Number of Model Classes                   : "
    f"{len(model_classes)}"
)

logger.info(
    f"Number of Test Samples                    : "
    f"{X_test_explain.shape[0]}"
)

logger.info(
    f"Number of Transformed Features            : "
    f"{X_test_explain.shape[1]}"
)

assert shap_values_array.ndim == 3

assert shap_values_array.shape[0] == (
    len(model_classes)
)

assert shap_values_array.shape[1] == (
    X_test_explain.shape[0]
)

assert shap_values_array.shape[2] == (
    X_test_explain.shape[1]
)

assert len(
    final_explainability_feature_names
) == X_test_explain.shape[1]

assert np.all(
    np.isfinite(shap_values_array)
)

logger.info(
    "Existing SHAP Tensor                      : VALIDATED"
)

# ------------------------------------------------------------------------------
# 3. DIRECTLY CALCULATE GLOBAL SHAP IMPORTANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING GLOBAL SHAP IMPORTANCE DIRECTLY")
logger.info("-" * 70)

# IMPORTANT:
# DO NOT use final_global_shap_importance here.
#
# We calculate the global mean absolute SHAP directly from:
# final_shap_values_by_class
#
# This prevents rank/index columns from being mistaken for SHAP values.

global_mean_abs_shap = np.mean(
    np.abs(shap_values_array),
    axis=(0, 1)
)

global_shap_importance_direct_df = pd.DataFrame({
    "Feature": final_explainability_feature_names,
    "Mean Absolute SHAP": global_mean_abs_shap
})

global_shap_importance_direct_df = (
    global_shap_importance_direct_df
    .sort_values(
        "Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

logger.info(
    f"Global SHAP Feature Count                 : "
    f"{len(global_shap_importance_direct_df)}"
)

logger.info(
    f"Maximum Mean Absolute SHAP                : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].max():.6f}"
)

logger.info(
    f"Mean Absolute SHAP                        : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].mean():.6f}"
)

logger.info(
    f"Median Absolute SHAP                      : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].median():.6f}"
)

display(
    global_shap_importance_direct_df.head(20).style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 4. STRICT SHAP VALUE SANITY CHECK
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PERFORMING SHAP VALUE SANITY CHECK")
logger.info("-" * 70)

shap_max = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].max()
)

shap_mean = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].mean()
)

shap_median = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].median()
)

# The known Section 14.4 SHAP values are small fractional
# contribution magnitudes. This check is intentionally strict
# enough to catch accidental rank/index columns such as 130 or 304.

assert shap_max < 1.0
assert shap_mean < shap_max
assert shap_median <= shap_mean

logger.info(
    f"Maximum SHAP < 1.0                       : PASSED"
)

logger.info(
    "SHAP Magnitude Sanity Check               : PASSED"
)

# ------------------------------------------------------------------------------
# 5. MAP TRANSFORMED FEATURES TO CLINICAL VARIABLES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MAPPING TRANSFORMED FEATURES TO CLINICAL VARIABLES")
logger.info("-" * 70)

clinical_transformed_records = []

for transformed_feature in (
    final_explainability_feature_names
):

    matched_original_feature = None

    # Numeric Age
    if transformed_feature == "numeric__Age":

        matched_original_feature = "Age"

    else:

        possible_matches = [
            feature
            for feature in clinical_features_available
            if feature != "Age"
            and transformed_feature.startswith(
                f"categorical__{feature}_"
            )
        ]

        if possible_matches:

            matched_original_feature = max(
                possible_matches,
                key=len
            )

    if matched_original_feature is not None:

        clinical_transformed_records.append({
            "Transformed Feature":
                transformed_feature,
            "Original Clinical Feature":
                matched_original_feature
        })

clinical_feature_mapping_df = pd.DataFrame(
    clinical_transformed_records
)

logger.info(
    f"Clinical Transformed Features             : "
    f"{len(clinical_feature_mapping_df)}"
)

display(
    clinical_feature_mapping_df
)

assert len(
    clinical_feature_mapping_df
) > 0

# ------------------------------------------------------------------------------
# 6. MERGE DIRECT SHAP VALUES WITH CLINICAL MAPPING
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING CLINICAL FEATURES USING DIRECT SHAP VALUES")
logger.info("-" * 70)

clinical_global_shap_df = (
    clinical_feature_mapping_df
    .merge(
        global_shap_importance_direct_df,
        left_on="Transformed Feature",
        right_on="Feature",
        how="left"
    )
    .drop(
        columns=["Feature"]
    )
    .sort_values(
        "Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

logger.info(
    f"Clinical Features With SHAP Values        : "
    f"{len(clinical_global_shap_df)}"
)

display(
    clinical_global_shap_df.head(20).style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

assert len(
    clinical_global_shap_df
) == len(
    clinical_feature_mapping_df
)

assert (
    clinical_global_shap_df[
        "Mean Absolute SHAP"
    ].notna().all()
)

# ------------------------------------------------------------------------------
# 7. AGGREGATE SHAP BY ORIGINAL CLINICAL FEATURE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGGREGATING SHAP IMPORTANCE BY CLINICAL VARIABLE")
logger.info("-" * 70)

clinical_aggregated_shap_df = (
    clinical_global_shap_df
    .groupby(
        "Original Clinical Feature",
        as_index=False
    )
    .agg(
        Total_Mean_Absolute_SHAP=(
            "Mean Absolute SHAP",
            "sum"
        ),
        Maximum_Category_SHAP=(
            "Mean Absolute SHAP",
            "max"
        ),
        Encoded_Category_Count=(
            "Transformed Feature",
            "count"
        )
    )
    .rename(
        columns={
            "Total_Mean_Absolute_SHAP":
                "Total Mean Absolute SHAP",
            "Maximum_Category_SHAP":
                "Maximum Category SHAP",
            "Encoded_Category_Count":
                "Encoded Category Count"
        }
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    clinical_aggregated_shap_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 8. IDENTIFY STRONGEST CATEGORY FOR EACH CLINICAL FEATURE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING STRONGEST CATEGORY-SPECIFIC CONTRIBUTIONS")
logger.info("-" * 70)

strongest_category_records = []

for clinical_feature in (
    clinical_aggregated_shap_df[
        "Original Clinical Feature"
    ]
):

    feature_rows = (
        clinical_global_shap_df[
            clinical_global_shap_df[
                "Original Clinical Feature"
            ] == clinical_feature
        ]
        .sort_values(
            "Mean Absolute SHAP",
            ascending=False
        )
    )

    strongest_row = feature_rows.iloc[0]

    strongest_category_records.append({
        "Original Clinical Feature":
            clinical_feature,
        "Most Influential Encoded Category":
            strongest_row[
                "Transformed Feature"
            ],
        "Category Mean Absolute SHAP":
            strongest_row[
                "Mean Absolute SHAP"
            ]
    })

strongest_category_df = pd.DataFrame(
    strongest_category_records
)

logger.info(
    f"Clinical Variables Analyzed                : "
    f"{len(strongest_category_df)}"
)

display(
    strongest_category_df.style.format({
        "Category Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 9. CREATE CLINICAL SHAP SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING CLINICAL FEATURE EXPLAINABILITY SUMMARY")
logger.info("-" * 70)

clinical_summary_df = (
    clinical_aggregated_shap_df
    .merge(
        strongest_category_df,
        on="Original Clinical Feature",
        how="left"
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    clinical_summary_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}",
        "Category Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 10. CLINICAL PERMUTATION IMPORTANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING CLINICAL FEATURES USING PERMUTATION IMPORTANCE")
logger.info("-" * 70)

permutation_source = (
    final_permutation_importance.copy()
)

if "Feature" not in permutation_source.columns:

    permutation_source = (
        permutation_source
        .reset_index()
    )

# Locate the actual permutation mean importance column.
#
# We deliberately search for column names containing both
# "importance" and "mean", rather than selecting an arbitrary
# numeric column such as rank.

permutation_mean_candidates = [
    column
    for column in permutation_source.columns
    if (
        "importance" in str(column).lower()
        and "mean" in str(column).lower()
    )
]

if len(permutation_mean_candidates) == 1:

    permutation_mean_column = (
        permutation_mean_candidates[0]
    )

elif len(permutation_mean_candidates) > 1:

    # Prefer the most explicit name if multiple candidates exist.
    exact_priority = [
        column
        for column in permutation_mean_candidates
        if str(column).lower()
        in [
            "mean importance",
            "mean_importance",
            "mean permutation importance",
            "mean_permutation_importance"
        ]
    ]

    if exact_priority:

        permutation_mean_column = (
            exact_priority[0]
        )

    else:

        permutation_mean_column = (
            permutation_mean_candidates[0]
        )

else:

    raise ValueError(
        "Could not identify the mean permutation "
        "importance column in final_permutation_importance."
    )

logger.info(
    f"Permutation Importance Column             : "
    f"{permutation_mean_column}"
)

clinical_permutation_df = (
    clinical_feature_mapping_df
    .merge(
        permutation_source[
            [
                "Feature",
                permutation_mean_column
            ]
        ],
        left_on="Transformed Feature",
        right_on="Feature",
        how="left"
    )
    .drop(
        columns=["Feature"]
    )
    .rename(
        columns={
            permutation_mean_column:
                "Permutation Importance"
        }
    )
    .sort_values(
        "Permutation Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    clinical_permutation_df.head(20).style.format({
        "Permutation Importance": "{:.6f}"
    })
)

assert (
    clinical_permutation_df[
        "Permutation Importance"
    ].notna().all()
)

# ------------------------------------------------------------------------------
# 11. AGGREGATE PERMUTATION IMPORTANCE BY CLINICAL VARIABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("AGGREGATING PERMUTATION IMPORTANCE BY CLINICAL VARIABLE")
logger.info("-" * 70)

clinical_aggregated_permutation_df = (
    clinical_permutation_df
    .groupby(
        "Original Clinical Feature",
        as_index=False
    )
    .agg(
        Total_Permutation_Importance=(
            "Permutation Importance",
            "sum"
        ),
        Maximum_Category_Permutation=(
            "Permutation Importance",
            "max"
        )
    )
    .rename(
        columns={
            "Total_Permutation_Importance":
                "Total Permutation Importance",
            "Maximum_Category_Permutation":
                "Maximum Category Permutation"
        }
    )
    .sort_values(
        "Total Permutation Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    clinical_aggregated_permutation_df.style.format({
        "Total Permutation Importance": "{:.6f}",
        "Maximum Category Permutation": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 12. COMBINED CLINICAL EXPLAINABILITY SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING COMBINED CLINICAL EXPLAINABILITY SUMMARY")
logger.info("-" * 70)

clinical_combined_summary_df = (
    clinical_summary_df
    .merge(
        clinical_aggregated_permutation_df,
        on="Original Clinical Feature",
        how="left"
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    clinical_combined_summary_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}",
        "Category Mean Absolute SHAP": "{:.6f}",
        "Total Permutation Importance": "{:.6f}",
        "Maximum Category Permutation": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 13. TOP CLINICAL FEATURES VISUALIZATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING CLINICAL FEATURE IMPORTANCE VISUALIZATION")
logger.info("-" * 70)

top_n_clinical = min(
    10,
    len(clinical_combined_summary_df)
)

top_clinical_plot = (
    clinical_combined_summary_df
    .head(top_n_clinical)
    .sort_values(
        "Total Mean Absolute SHAP"
    )
)

plt.figure(
    figsize=(10, 6)
)

plt.barh(
    top_clinical_plot[
        "Original Clinical Feature"
    ],
    top_clinical_plot[
        "Total Mean Absolute SHAP"
    ]
)

plt.xlabel(
    "Aggregated Mean Absolute SHAP"
)

plt.ylabel(
    "Clinical Feature"
)

plt.title(
    "Top Clinical Features Used by the Model"
)

plt.tight_layout()
plt.show()

logger.info(
    "Clinical Feature Visualization              : GENERATED"
)

# ------------------------------------------------------------------------------
# 14. CLINICAL INTERPRETATION TABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING CLINICAL INTERPRETATION TABLE")
logger.info("-" * 70)

clinical_interpretation_records = []

for _, row in (
    clinical_combined_summary_df.iterrows()
):

    clinical_feature = row[
        "Original Clinical Feature"
    ]

    if clinical_feature == "Age":

        interpretation = (
            "Age is a numeric patient-level variable. "
            "Its SHAP magnitude indicates the strength "
            "of its contribution to model predictions, "
            "but does not establish a causal clinical relationship."
        )

    else:

        interpretation = (
            "The model uses category-specific encoded "
            "values of this clinical observation. "
            "The reported importance describes model "
            "behavior and should not be interpreted as "
            "causal or independently diagnostic evidence."
        )

    clinical_interpretation_records.append({
        "Clinical Feature":
            clinical_feature,
        "Aggregated SHAP Importance":
            row[
                "Total Mean Absolute SHAP"
            ],
        "Strongest Encoded Category":
            row[
                "Most Influential Encoded Category"
            ],
        "Permutation Importance":
            row[
                "Total Permutation Importance"
            ],
        "Interpretation":
            interpretation
    })

clinical_interpretation_df = pd.DataFrame(
    clinical_interpretation_records
)

display(
    clinical_interpretation_df.style.format({
        "Aggregated SHAP Importance": "{:.6f}",
        "Permutation Importance": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 15. KEY CLINICAL FEATURE FINDINGS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("KEY CLINICAL FEATURE FINDINGS")
logger.info("-" * 70)

if len(clinical_combined_summary_df) > 0:

    top_clinical_feature = (
        clinical_combined_summary_df.iloc[0]
    )

    logger.info(
        f"Highest Aggregated Clinical Feature      : "
        f"{top_clinical_feature['Original Clinical Feature']}"
    )

    logger.info(
        f"Highest Aggregated SHAP                   : "
        f"{top_clinical_feature['Total Mean Absolute SHAP']:.6f}"
    )

    logger.info(
        f"Strongest Encoded Category                : "
        f"{top_clinical_feature['Most Influential Encoded Category']}"
    )

    logger.info(
        f"Category SHAP                             : "
        f"{top_clinical_feature['Category Mean Absolute SHAP']:.6f}"
    )

    logger.info(
        f"Aggregated Permutation Importance         : "
        f"{top_clinical_feature['Total Permutation Importance']:.6f}"
    )

# ------------------------------------------------------------------------------
# 16. CLINICAL INTERPRETATION SAFETY NOTE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLINICAL INTERPRETATION SAFETY NOTE")
logger.info("-" * 70)

logger.info(
    "SHAP importance describes model behavior, not causality."
)

logger.info(
    "One-hot category importance should not automatically "
    "be interpreted as simple presence or absence."
)

logger.info(
    "Category suffixes represent encoded dataset values."
)

logger.info(
    "Clinical significance requires independent medical "
    "validation and domain expertise."
)

logger.info(
    "Model explainability does not establish diagnostic validity."
)

# ------------------------------------------------------------------------------
# 17. STRICT VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("14.6 CLINICAL FEATURE INTERPRETATION VALIDATION")
logger.info("-" * 70)

# DataFrame validation
assert isinstance(
    clinical_feature_mapping_df,
    pd.DataFrame
)

assert isinstance(
    clinical_global_shap_df,
    pd.DataFrame
)

assert isinstance(
    clinical_aggregated_shap_df,
    pd.DataFrame
)

assert isinstance(
    clinical_permutation_df,
    pd.DataFrame
)

assert isinstance(
    clinical_aggregated_permutation_df,
    pd.DataFrame
)

assert isinstance(
    clinical_combined_summary_df,
    pd.DataFrame
)

assert isinstance(
    clinical_interpretation_df,
    pd.DataFrame
)

# Feature counts
assert len(
    clinical_features_available
) == 12

assert len(
    clinical_combined_summary_df
) == 12

# All clinical transformed features mapped
assert len(
    clinical_feature_mapping_df
) > 0

# SHAP values exist for every mapped feature
assert (
    clinical_global_shap_df[
        "Mean Absolute SHAP"
    ].notna().all()
)

# SHAP values are finite
assert np.all(
    np.isfinite(
        clinical_global_shap_df[
            "Mean Absolute SHAP"
        ].values
    )
)

# SHAP values must be non-negative because they are
# mean absolute SHAP magnitudes
assert np.all(
    clinical_global_shap_df[
        "Mean Absolute SHAP"
    ].values >= 0
)

# Critical protection against the previous error:
# SHAP importance must remain in fractional contribution
# magnitude range, not rank/index range.

assert (
    clinical_global_shap_df[
        "Mean Absolute SHAP"
    ].max() < 1.0
)

# Aggregated SHAP must also remain below 1.0
assert (
    clinical_aggregated_shap_df[
        "Total Mean Absolute SHAP"
    ].max() < 1.0
)

# Strongest category SHAP must be below 1.0
assert (
    clinical_aggregated_shap_df[
        "Maximum Category SHAP"
    ].max() < 1.0
)

# Permutation importance exists
assert (
    clinical_permutation_df[
        "Permutation Importance"
    ].notna().all()
)

# Combined summary contains all clinical variables
assert set(
    clinical_combined_summary_df[
        "Original Clinical Feature"
    ]
) == set(
    clinical_features_available
)

# Each clinical feature appears exactly once
assert (
    clinical_combined_summary_df[
        "Original Clinical Feature"
    ].is_unique
)

# Summary sorted descending by SHAP
assert (
    clinical_combined_summary_df[
        "Total Mean Absolute SHAP"
    ].is_monotonic_decreasing
)

logger.info(
    "Clinical Feature Mapping                  : PASSED"
)

logger.info(
    "Global SHAP Clinical Analysis              : PASSED"
)

logger.info(
    "Clinical Feature Aggregation               : PASSED"
)

logger.info(
    "Permutation Importance Analysis            : PASSED"
)

logger.info(
    "Clinical Summary Validation                : PASSED"
)

logger.info(
    "SHAP Magnitude Sanity Check                : PASSED"
)

logger.info(
    "SHAP Numerical Validation                  : PASSED"
)

logger.info(
    "Clinical Interpretation Table              : PASSED"
)

logger.info(
    "Visualization Validation                   : PASSED"
)

# ------------------------------------------------------------------------------
# 18. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLINICAL EXPLAINABILITY RESULTS STORED")
logger.info("-" * 70)

final_clinical_feature_mapping = (
    clinical_feature_mapping_df.copy()
)

final_clinical_global_shap = (
    clinical_global_shap_df.copy()
)

final_clinical_aggregated_shap = (
    clinical_aggregated_shap_df.copy()
)

final_clinical_permutation_importance = (
    clinical_permutation_df.copy()
)

final_clinical_aggregated_permutation = (
    clinical_aggregated_permutation_df.copy()
)

final_clinical_feature_summary = (
    clinical_summary_df.copy()
)

final_clinical_combined_summary = (
    clinical_combined_summary_df.copy()
)

final_clinical_interpretation = (
    clinical_interpretation_df.copy()
)

# Also store the directly calculated global SHAP table
final_global_shap_importance_direct = (
    global_shap_importance_direct_df.copy()
)

logger.info(
    "Clinical Feature Mapping                    : STORED"
)

logger.info(
    "Clinical Global SHAP Results                : STORED"
)

logger.info(
    "Clinical Aggregated SHAP                   : STORED"
)

logger.info(
    "Clinical Permutation Importance             : STORED"
)

logger.info(
    "Clinical Aggregated Permutation             : STORED"
)

logger.info(
    "Clinical Feature Summary                   : STORED"
)

logger.info(
    "Clinical Combined Summary                 : STORED"
)

logger.info(
    "Clinical Interpretation                    : STORED"
)

logger.info(
    "Direct Global SHAP Reference               : STORED"
)

logger.info("=" * 70)
logger.info(
    "14.6 CLINICAL FEATURE INTERPRETATION : COMPLETED"
)
logger.info("=" * 70)
````

## 14.7 Histopathological Feature Interpretation

Histopathological feature interpretation evaluates how microscopic tissue-level observations contribute to the predictions of the selected Random Forest model.

The analysis uses the validated multiclass SHAP values generated in Section 14.4 and permutation importance results generated in Section 14.3.

The histopathological variables considered in this analysis include:

- PNL infiltrate
- fibrosis of the papillary dermis
- clubbing of the rete ridges
- elongation of the rete ridges
- thinning of the suprapapillary epidermis
- spongiosis
- vacuolisation and damage of basal layer
- saw-tooth appearance of retes
- perifollicular parakeratosis
- follicular horn plug
- munro microabcess
- focal hypergranulosis
- melanin incontinence
- disappearance of the granular layer
- hyperkeratosis
- parakeratosis
- acanthosis
- papillomatosis
- spongiform pustule
- exocytosis
- eosinophils in the infiltrate

Because these variables are categorical and were represented using one-hot encoding, each original histopathological variable may correspond to multiple transformed model features.

The analysis therefore:

- Maps transformed features to their original histopathological variables.
- Calculates mean absolute SHAP importance directly from the validated SHAP tensor.
- Aggregates category-level SHAP importance to the original histopathological variable.
- Identifies the strongest encoded category for each variable.
- Compares histopathological SHAP importance with permutation importance.
- Identifies the most influential microscopic features used by the model.
- Visualizes the highest-ranked histopathological features.

The interpretation describes model behavior rather than biological or clinical causation. Feature importance should not be interpreted as proof that a histopathological finding independently causes a disorder or is sufficient for diagnosis.

````python

# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

logger.info("=" * 70)
logger.info("14.7 HISTOPATHOLOGICAL FEATURE INTERPRETATION")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. HISTOPATHOLOGICAL FEATURE CONFIGURATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING HISTOPATHOLOGICAL FEATURES")
logger.info("-" * 70)

histopathological_feature_names = [
    "PNL_infiltrate",
    "fibrosis_of_the_papillary_dermis",
    "clubbing_of_the_rete_ridges",
    "elongation_of_the_rete_ridges",
    "thinning_of_the_suprapapillary_epidermis",
    "spongiosis",
    "vacuolisation_and_damage_of_basal_layer",
    "saw-tooth_appearance_of_retes",
    "perifollicular_parakeratosis",
    "follicular_horn_plug",
    "munro_microabcess",
    "focal_hypergranulosis",
    "melanin_incontinence",
    "disappearance_of_the_granular_layer",
    "hyperkeratosis",
    "parakeratosis",
    "acanthosis",
    "papillomatosis",
    "spongiform_pustule",
    "exocytosis",
    "eosinophils_in_the_infiltrate"
]

original_feature_names = list(
    X_train.columns
)

histopathological_features_available = [
    feature
    for feature in histopathological_feature_names
    if feature in original_feature_names
]

histopathological_features_missing = [
    feature
    for feature in histopathological_feature_names
    if feature not in original_feature_names
]

logger.info(
    f"Histopathological Features Requested       : "
    f"{len(histopathological_feature_names)}"
)

logger.info(
    f"Histopathological Features Available       : "
    f"{len(histopathological_features_available)}"
)

if histopathological_features_missing:

    logger.info(
        f"Histopathological Features Not Available : "
        f"{histopathological_features_missing}"
    )

logger.info(
    f"Available Histopathological Features       : "
    f"{histopathological_features_available}"
)

assert len(
    histopathological_features_available
) > 0

# ------------------------------------------------------------------------------
# 2. VALIDATE EXISTING SHAP VALUES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING EXISTING MULTICLASS SHAP VALUES")
logger.info("-" * 70)

# Use the already validated SHAP tensor from Section 14.4.
#
# Expected shape:
#     classes × test_samples × transformed_features
#
# Current project:
#     6 × 74 × 130

shap_values_array = np.asarray(
    final_shap_values_by_class
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

logger.info(
    f"SHAP Tensor Shape                         : "
    f"{shap_values_array.shape}"
)

logger.info(
    f"Number of Model Classes                   : "
    f"{len(model_classes)}"
)

logger.info(
    f"Number of Test Samples                    : "
    f"{X_test_explain.shape[0]}"
)

logger.info(
    f"Number of Transformed Features            : "
    f"{X_test_explain.shape[1]}"
)

assert shap_values_array.ndim == 3

assert shap_values_array.shape[0] == (
    len(model_classes)
)

assert shap_values_array.shape[1] == (
    X_test_explain.shape[0]
)

assert shap_values_array.shape[2] == (
    X_test_explain.shape[1]
)

assert len(
    final_explainability_feature_names
) == X_test_explain.shape[1]

assert np.all(
    np.isfinite(shap_values_array)
)

logger.info(
    "Existing SHAP Tensor                      : VALIDATED"
)

# ------------------------------------------------------------------------------
# 3. DIRECT GLOBAL SHAP IMPORTANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING GLOBAL SHAP IMPORTANCE DIRECTLY")
logger.info("-" * 70)

# IMPORTANT:
# Do NOT use final_global_shap_importance as the source
# of the numerical SHAP values.
#
# The canonical source is:
# final_shap_values_by_class

global_mean_abs_shap = np.mean(
    np.abs(shap_values_array),
    axis=(0, 1)
)

global_shap_importance_direct_df = pd.DataFrame({
    "Feature": final_explainability_feature_names,
    "Mean Absolute SHAP": global_mean_abs_shap
})

global_shap_importance_direct_df = (
    global_shap_importance_direct_df
    .sort_values(
        "Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

logger.info(
    f"Global SHAP Feature Count                 : "
    f"{len(global_shap_importance_direct_df)}"
)

logger.info(
    f"Maximum Mean Absolute SHAP                : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].max():.6f}"
)

logger.info(
    f"Mean Absolute SHAP                        : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].mean():.6f}"
)

logger.info(
    f"Median Absolute SHAP                      : "
    f"{global_shap_importance_direct_df['Mean Absolute SHAP'].median():.6f}"
)

display(
    global_shap_importance_direct_df.head(20).style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 4. STRICT SHAP SANITY CHECK
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PERFORMING SHAP VALUE SANITY CHECK")
logger.info("-" * 70)

shap_max = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].max()
)

shap_mean = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].mean()
)

shap_median = (
    global_shap_importance_direct_df[
        "Mean Absolute SHAP"
    ].median()
)

# Protection against accidental use of rank/index/count columns.

assert shap_max < 1.0
assert shap_mean < shap_max
assert shap_median <= shap_mean

logger.info(
    "Maximum SHAP < 1.0                       : PASSED"
)

logger.info(
    "SHAP Magnitude Sanity Check               : PASSED"
)

# ------------------------------------------------------------------------------
# 5. MAP TRANSFORMED FEATURES TO HISTOPATHOLOGICAL VARIABLES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "MAPPING TRANSFORMED FEATURES TO HISTOPATHOLOGICAL VARIABLES"
)
logger.info("-" * 70)

histopathological_transformed_records = []

for transformed_feature in (
    final_explainability_feature_names
):

    matched_original_feature = None

    # Histopathological variables are categorical
    # and therefore use the categorical__ prefix.

    possible_matches = [
        feature
        for feature in histopathological_features_available
        if transformed_feature.startswith(
            f"categorical__{feature}_"
        )
    ]

    if possible_matches:

        # Longest match prevents incorrect matching
        # where one feature name may be a prefix of another.

        matched_original_feature = max(
            possible_matches,
            key=len
        )

    if matched_original_feature is not None:

        histopathological_transformed_records.append({
            "Transformed Feature":
                transformed_feature,
            "Original Histopathological Feature":
                matched_original_feature
        })

histopathological_feature_mapping_df = pd.DataFrame(
    histopathological_transformed_records
)

logger.info(
    f"Histopathological Transformed Features     : "
    f"{len(histopathological_feature_mapping_df)}"
)

display(
    histopathological_feature_mapping_df
)

assert len(
    histopathological_feature_mapping_df
) > 0

# ------------------------------------------------------------------------------
# 6. MERGE DIRECT SHAP VALUES WITH HISTOPATHOLOGICAL MAPPING
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "ANALYZING HISTOPATHOLOGICAL FEATURES USING DIRECT SHAP VALUES"
)
logger.info("-" * 70)

histopathological_global_shap_df = (
    histopathological_feature_mapping_df
    .merge(
        global_shap_importance_direct_df,
        left_on="Transformed Feature",
        right_on="Feature",
        how="left"
    )
    .drop(
        columns=["Feature"]
    )
    .sort_values(
        "Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

logger.info(
    f"Histopathological Features With SHAP Values : "
    f"{len(histopathological_global_shap_df)}"
)

display(
    histopathological_global_shap_df.head(30).style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

assert len(
    histopathological_global_shap_df
) == len(
    histopathological_feature_mapping_df
)

assert (
    histopathological_global_shap_df[
        "Mean Absolute SHAP"
    ].notna().all()
)

# ------------------------------------------------------------------------------
# 7. AGGREGATE SHAP BY ORIGINAL HISTOPATHOLOGICAL VARIABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "AGGREGATING SHAP IMPORTANCE BY HISTOPATHOLOGICAL VARIABLE"
)
logger.info("-" * 70)

histopathological_aggregated_shap_df = (
    histopathological_global_shap_df
    .groupby(
        "Original Histopathological Feature",
        as_index=False
    )
    .agg(
        Total_Mean_Absolute_SHAP=(
            "Mean Absolute SHAP",
            "sum"
        ),
        Maximum_Category_SHAP=(
            "Mean Absolute SHAP",
            "max"
        ),
        Encoded_Category_Count=(
            "Transformed Feature",
            "count"
        )
    )
    .rename(
        columns={
            "Total_Mean_Absolute_SHAP":
                "Total Mean Absolute SHAP",
            "Maximum_Category_SHAP":
                "Maximum Category SHAP",
            "Encoded_Category_Count":
                "Encoded Category Count"
        }
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    histopathological_aggregated_shap_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 8. IDENTIFY STRONGEST CATEGORY FOR EACH VARIABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "IDENTIFYING STRONGEST CATEGORY-SPECIFIC CONTRIBUTIONS"
)
logger.info("-" * 70)

strongest_histopathological_category_records = []

for histopathological_feature in (
    histopathological_aggregated_shap_df[
        "Original Histopathological Feature"
    ]
):

    feature_rows = (
        histopathological_global_shap_df[
            histopathological_global_shap_df[
                "Original Histopathological Feature"
            ] == histopathological_feature
        ]
        .sort_values(
            "Mean Absolute SHAP",
            ascending=False
        )
    )

    strongest_row = feature_rows.iloc[0]

    strongest_histopathological_category_records.append({
        "Original Histopathological Feature":
            histopathological_feature,
        "Most Influential Encoded Category":
            strongest_row[
                "Transformed Feature"
            ],
        "Category Mean Absolute SHAP":
            strongest_row[
                "Mean Absolute SHAP"
            ]
    })

strongest_histopathological_category_df = pd.DataFrame(
    strongest_histopathological_category_records
)

logger.info(
    f"Histopathological Variables Analyzed       : "
    f"{len(strongest_histopathological_category_df)}"
)

display(
    strongest_histopathological_category_df.style.format({
        "Category Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 9. CREATE HISTOPATHOLOGICAL SHAP SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "CREATING HISTOPATHOLOGICAL FEATURE EXPLAINABILITY SUMMARY"
)
logger.info("-" * 70)

histopathological_summary_df = (
    histopathological_aggregated_shap_df
    .merge(
        strongest_histopathological_category_df,
        on="Original Histopathological Feature",
        how="left"
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    histopathological_summary_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}",
        "Category Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 10. HISTOPATHOLOGICAL PERMUTATION IMPORTANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "ANALYZING HISTOPATHOLOGICAL FEATURES USING PERMUTATION IMPORTANCE"
)
logger.info("-" * 70)

permutation_source = (
    final_permutation_importance.copy()
)

if "Feature" not in permutation_source.columns:

    permutation_source = (
        permutation_source
        .reset_index()
    )

# Identify the actual mean permutation importance
# column explicitly by its name.

permutation_mean_candidates = [
    column
    for column in permutation_source.columns
    if (
        "importance" in str(column).lower()
        and "mean" in str(column).lower()
    )
]

if len(permutation_mean_candidates) == 1:

    permutation_mean_column = (
        permutation_mean_candidates[0]
    )

elif len(permutation_mean_candidates) > 1:

    exact_priority = [
        column
        for column in permutation_mean_candidates
        if str(column).lower()
        in [
            "mean importance",
            "mean_importance",
            "mean permutation importance",
            "mean_permutation_importance"
        ]
    ]

    if exact_priority:

        permutation_mean_column = (
            exact_priority[0]
        )

    else:

        permutation_mean_column = (
            permutation_mean_candidates[0]
        )

else:

    raise ValueError(
        "Could not identify the mean permutation "
        "importance column in final_permutation_importance."
    )

logger.info(
    f"Permutation Importance Column             : "
    f"{permutation_mean_column}"
)

histopathological_permutation_df = (
    histopathological_feature_mapping_df
    .merge(
        permutation_source[
            [
                "Feature",
                permutation_mean_column
            ]
        ],
        left_on="Transformed Feature",
        right_on="Feature",
        how="left"
    )
    .drop(
        columns=["Feature"]
    )
    .rename(
        columns={
            permutation_mean_column:
                "Permutation Importance"
        }
    )
    .sort_values(
        "Permutation Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    histopathological_permutation_df.head(30).style.format({
        "Permutation Importance": "{:.6f}"
    })
)

assert (
    histopathological_permutation_df[
        "Permutation Importance"
    ].notna().all()
)

# ------------------------------------------------------------------------------
# 11. AGGREGATE PERMUTATION IMPORTANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "AGGREGATING PERMUTATION IMPORTANCE BY HISTOPATHOLOGICAL VARIABLE"
)
logger.info("-" * 70)

histopathological_aggregated_permutation_df = (
    histopathological_permutation_df
    .groupby(
        "Original Histopathological Feature",
        as_index=False
    )
    .agg(
        Total_Permutation_Importance=(
            "Permutation Importance",
            "sum"
        ),
        Maximum_Category_Permutation=(
            "Permutation Importance",
            "max"
        )
    )
    .rename(
        columns={
            "Total_Permutation_Importance":
                "Total Permutation Importance",
            "Maximum_Category_Permutation":
                "Maximum Category Permutation"
        }
    )
    .sort_values(
        "Total Permutation Importance",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    histopathological_aggregated_permutation_df.style.format({
        "Total Permutation Importance": "{:.6f}",
        "Maximum Category Permutation": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 12. COMBINED HISTOPATHOLOGICAL EXPLAINABILITY SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "CREATING COMBINED HISTOPATHOLOGICAL EXPLAINABILITY SUMMARY"
)
logger.info("-" * 70)

histopathological_combined_summary_df = (
    histopathological_summary_df
    .merge(
        histopathological_aggregated_permutation_df,
        on="Original Histopathological Feature",
        how="left"
    )
    .sort_values(
        "Total Mean Absolute SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    histopathological_combined_summary_df.style.format({
        "Total Mean Absolute SHAP": "{:.6f}",
        "Maximum Category SHAP": "{:.6f}",
        "Category Mean Absolute SHAP": "{:.6f}",
        "Total Permutation Importance": "{:.6f}",
        "Maximum Category Permutation": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 13. TOP HISTOPATHOLOGICAL FEATURES VISUALIZATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "GENERATING HISTOPATHOLOGICAL FEATURE IMPORTANCE VISUALIZATION"
)
logger.info("-" * 70)

top_n_histopathological = min(
    15,
    len(histopathological_combined_summary_df)
)

top_histopathological_plot = (
    histopathological_combined_summary_df
    .head(top_n_histopathological)
    .sort_values(
        "Total Mean Absolute SHAP"
    )
)

plt.figure(
    figsize=(11, 8)
)

plt.barh(
    top_histopathological_plot[
        "Original Histopathological Feature"
    ],
    top_histopathological_plot[
        "Total Mean Absolute SHAP"
    ]
)

plt.xlabel(
    "Aggregated Mean Absolute SHAP"
)

plt.ylabel(
    "Histopathological Feature"
)

plt.title(
    "Top Histopathological Features Used by the Model"
)

plt.tight_layout()
plt.show()

logger.info(
    "Histopathological Feature Visualization    : GENERATED"
)

# ------------------------------------------------------------------------------
# 14. TOP HISTOPATHOLOGICAL CATEGORY TABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "GENERATING TOP HISTOPATHOLOGICAL CATEGORY TABLE"
)
logger.info("-" * 70)

top_histopathological_categories_df = (
    histopathological_global_shap_df
    .head(20)
    .copy()
)

display(
    top_histopathological_categories_df.style.format({
        "Mean Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 15. HISTOPATHOLOGICAL INTERPRETATION TABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "GENERATING HISTOPATHOLOGICAL INTERPRETATION TABLE"
)
logger.info("-" * 70)

histopathological_interpretation_records = []

for _, row in (
    histopathological_combined_summary_df.iterrows()
):

    histopathological_feature = row[
        "Original Histopathological Feature"
    ]

    interpretation = (
        "The model uses category-specific encoded "
        "values of this histopathological observation. "
        "The reported SHAP and permutation importance "
        "describe model behavior and should not be "
        "interpreted as causal biological evidence "
        "or as an independently sufficient diagnostic marker."
    )

    histopathological_interpretation_records.append({
        "Histopathological Feature":
            histopathological_feature,
        "Aggregated SHAP Importance":
            row[
                "Total Mean Absolute SHAP"
            ],
        "Strongest Encoded Category":
            row[
                "Most Influential Encoded Category"
            ],
        "Permutation Importance":
            row[
                "Total Permutation Importance"
            ],
        "Interpretation":
            interpretation
    })

histopathological_interpretation_df = pd.DataFrame(
    histopathological_interpretation_records
)

display(
    histopathological_interpretation_df.style.format({
        "Aggregated SHAP Importance": "{:.6f}",
        "Permutation Importance": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 16. KEY HISTOPATHOLOGICAL FINDINGS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("KEY HISTOPATHOLOGICAL FEATURE FINDINGS")
logger.info("-" * 70)

if len(
    histopathological_combined_summary_df
) > 0:

    top_histopathological_feature = (
        histopathological_combined_summary_df.iloc[0]
    )

    logger.info(
        f"Highest Histopathological Feature         : "
        f"{top_histopathological_feature['Original Histopathological Feature']}"
    )

    logger.info(
        f"Highest Aggregated SHAP                   : "
        f"{top_histopathological_feature['Total Mean Absolute SHAP']:.6f}"
    )

    logger.info(
        f"Strongest Encoded Category                : "
        f"{top_histopathological_feature['Most Influential Encoded Category']}"
    )

    logger.info(
        f"Category SHAP                             : "
        f"{top_histopathological_feature['Category Mean Absolute SHAP']:.6f}"
    )

    logger.info(
        f"Aggregated Permutation Importance         : "
        f"{top_histopathological_feature['Total Permutation Importance']:.6f}"
    )

# ------------------------------------------------------------------------------
# 17. HISTOPATHOLOGICAL INTERPRETATION SAFETY NOTE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "HISTOPATHOLOGICAL INTERPRETATION SAFETY NOTE"
)
logger.info("-" * 70)

logger.info(
    "SHAP importance describes model behavior, not biological causality."
)

logger.info(
    "One-hot category importance should not automatically "
    "be interpreted as simple presence or absence."
)

logger.info(
    "Category suffixes represent encoded dataset values."
)

logger.info(
    "Histopathological importance requires domain-specific "
    "pathology validation."
)

logger.info(
    "Model explainability does not establish diagnostic validity."
)

# ------------------------------------------------------------------------------
# 18. STRICT VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "14.7 HISTOPATHOLOGICAL FEATURE INTERPRETATION VALIDATION"
)
logger.info("-" * 70)

# DataFrame validation

assert isinstance(
    histopathological_feature_mapping_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_global_shap_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_aggregated_shap_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_permutation_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_aggregated_permutation_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_combined_summary_df,
    pd.DataFrame
)

assert isinstance(
    histopathological_interpretation_df,
    pd.DataFrame
)

# At least the expected histopathological
# features must be present.

assert len(
    histopathological_features_available
) > 0

assert len(
    histopathological_combined_summary_df
) == len(
    histopathological_features_available
)

# Every mapped feature must have SHAP
# importance.

assert (
    histopathological_global_shap_df[
        "Mean Absolute SHAP"
    ].notna().all()
)

# SHAP values must be finite.

assert np.all(
    np.isfinite(
        histopathological_global_shap_df[
            "Mean Absolute SHAP"
        ].values
    )
)

# Mean absolute SHAP values cannot be negative.

assert np.all(
    histopathological_global_shap_df[
        "Mean Absolute SHAP"
    ].values >= 0
)

# Critical protection against accidental rank/index
# column selection.

assert (
    histopathological_global_shap_df[
        "Mean Absolute SHAP"
    ].max() < 1.0
)

assert (
    histopathological_aggregated_shap_df[
        "Total Mean Absolute SHAP"
    ].max() < 1.0
)

assert (
    histopathological_aggregated_shap_df[
        "Maximum Category SHAP"
    ].max() < 1.0
)

# Permutation importance must be available.

assert (
    histopathological_permutation_df[
        "Permutation Importance"
    ].notna().all()
)

# Every original histopathological feature
# must appear exactly once in the summary.

assert (
    histopathological_combined_summary_df[
        "Original Histopathological Feature"
    ].is_unique
)

# Summary should be sorted from highest to lowest SHAP.

assert (
    histopathological_combined_summary_df[
        "Total Mean Absolute SHAP"
    ].is_monotonic_decreasing
)

logger.info(
    "Histopathological Feature Mapping          : PASSED"
)

logger.info(
    "Global SHAP Histopathological Analysis      : PASSED"
)

logger.info(
    "Histopathological Feature Aggregation       : PASSED"
)

logger.info(
    "Permutation Importance Analysis             : PASSED"
)

logger.info(
    "Histopathological Summary Validation        : PASSED"
)

logger.info(
    "SHAP Magnitude Sanity Check                 : PASSED"
)

logger.info(
    "SHAP Numerical Validation                   : PASSED"
)

logger.info(
    "Histopathological Interpretation Table     : PASSED"
)

logger.info(
    "Visualization Validation                    : PASSED"
)

# ------------------------------------------------------------------------------
# 19. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "HISTOPATHOLOGICAL EXPLAINABILITY RESULTS STORED"
)
logger.info("-" * 70)

final_histopathological_feature_mapping = (
    histopathological_feature_mapping_df.copy()
)

final_histopathological_global_shap = (
    histopathological_global_shap_df.copy()
)

final_histopathological_aggregated_shap = (
    histopathological_aggregated_shap_df.copy()
)

final_histopathological_permutation_importance = (
    histopathological_permutation_df.copy()
)

final_histopathological_aggregated_permutation = (
    histopathological_aggregated_permutation_df.copy()
)

final_histopathological_feature_summary = (
    histopathological_summary_df.copy()
)

final_histopathological_combined_summary = (
    histopathological_combined_summary_df.copy()
)

final_histopathological_interpretation = (
    histopathological_interpretation_df.copy()
)

final_top_histopathological_categories = (
    top_histopathological_categories_df.copy()
)

# Store direct SHAP reference for reproducibility.

final_histopathological_global_shap_direct = (
    global_shap_importance_direct_df.copy()
)

logger.info(
    "Histopathological Feature Mapping          : STORED"
)

logger.info(
    "Histopathological Global SHAP Results      : STORED"
)

logger.info(
    "Histopathological Aggregated SHAP          : STORED"
)

logger.info(
    "Histopathological Permutation Importance   : STORED"
)

logger.info(
    "Histopathological Aggregated Permutation   : STORED"
)

logger.info(
    "Histopathological Feature Summary          : STORED"
)

logger.info(
    "Histopathological Combined Summary         : STORED"
)

logger.info(
    "Histopathological Interpretation           : STORED"
)

logger.info(
    "Top Histopathological Categories           : STORED"
)

logger.info(
    "Direct Global SHAP Reference               : STORED"
)

logger.info("=" * 70)
logger.info(
    "14.7 HISTOPATHOLOGICAL FEATURE INTERPRETATION : COMPLETED"
)
logger.info("=" * 70)
````

## 14.8 Explainability Limitations

Model explainability techniques provide insight into how the machine learning model generates predictions, but they do not guarantee biological, clinical, or diagnostic validity.

The following limitations apply to the explainability analysis performed in this project:

### 1. SHAP Explains Model Behavior, Not Causality

SHAP values describe how features contribute to the model's prediction relative to the model's expected output.

A high SHAP value does not mean that the corresponding clinical or histopathological feature causes a particular skin disorder.

### 2. Dataset Size Limitation

The dataset contains a relatively small number of observations compared with datasets typically used for production-grade medical AI systems.

Consequently, feature importance and SHAP patterns may change when the model is trained on a larger and more diverse dataset.

### 3. Class Imbalance

The target classes are not equally represented.

In particular, the minority class has substantially fewer observations than the largest classes. Explainability results associated with minority classes should therefore be interpreted cautiously.

### 4. Categorical One-Hot Encoding

The original clinical and histopathological variables are categorical.

They were represented using one-hot encoding rather than arbitrary ordinal encoding. Consequently, one original feature can generate multiple transformed features.

The importance of individual encoded categories should therefore be interpreted at the original feature level whenever possible.

### 5. Correlated and Redundant Features

Some encoded features can contain related information.

When correlated or redundant variables exist, model importance may be distributed across multiple features. Therefore, the importance of one feature should not automatically be interpreted as evidence that it is uniquely responsible for a prediction.

### 6. Model-Specific Explainability

The explainability analysis is primarily based on the selected Random Forest model.

Different machine learning algorithms may learn different decision structures and consequently produce different feature importance patterns.

### 7. Permutation Importance Limitations

Permutation importance measures the change in model performance after feature values are randomly permuted.

For correlated features, permutation importance may underestimate the importance of a feature because other related features can still provide similar information to the model.

### 8. Local Explanation Limitation

The individual SHAP explanation describes one selected test observation.

It should not be generalized to every patient or every prediction made by the model.

### 9. Prediction Confidence Limitation

A correct prediction does not necessarily imply high confidence.

The individual prediction analysis demonstrated that a prediction can be correct while the model still assigns meaningful probability to alternative classes.

### 10. Medical Interpretation Limitation

The dataset contains encoded dermatological observations, but model explainability alone cannot establish medical significance.

Clinical interpretation requires validation by qualified dermatology and pathology professionals.

### 11. External Validation Limitation

The explainability results are derived from the available dataset and its held-out test set.

External validation on independent populations, institutions, acquisition protocols, and patient demographics is required before making claims about real-world generalization.

### 12. Diagnostic Safety Limitation

This system is a machine learning prediction project and should not be considered a standalone medical diagnostic system.

Explainability results should support model auditing and understanding rather than replace professional clinical judgment.

### Overall Interpretation

The explainability analysis provides useful evidence about which clinical and histopathological variables the selected model relies upon.

However, these explanations should be interpreted as **model-level evidence rather than clinical proof**.

The findings are therefore appropriate for model transparency, debugging, feature analysis, and research interpretation, but additional clinical validation is required before deployment in a real medical decision-support environment.

## **15.ERROR ANALYSIS** ##

## 15.1 Error Analysis Overview

Error analysis examines the incorrect and uncertain predictions produced by the selected Optimized Random Forest model on the independent held-out test set.

The objective is to move beyond overall performance metrics and understand how and where the model fails.

The analysis covers:

- Identification of misclassified test samples.
- Confusion matrix error patterns.
- Per-class error rates.
- Minority-class performance and errors.
- Dedicated analysis of Class 6.
- Low-confidence predictions.
- Repeated actual-to-predicted class confusion patterns.
- Potential model failure modes.
- Overall error analysis conclusions.

The analysis uses the final test predictions, predicted class probabilities, confusion matrix, and explainability results generated in previous sections.

The held-out test set contains 74 observations. The selected Optimized Random Forest correctly classified 72 observations and incorrectly classified 2 observations.

The two observed misclassification patterns are:

- Class 2 predicted as Class 4.
- Class 4 predicted as Class 2.

No test observations from Class 1, Class 3, Class 5, or Class 6 were misclassified.

Because the dataset is relatively small and some classes have limited test support, observed error patterns should be interpreted cautiously and should not be generalized beyond the evaluated dataset.

Error analysis does not modify the model or test data. It is performed strictly for model auditing, diagnosis of weaknesses, and identification of potential future improvement opportunities.

````python

# Required Libraries
import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.1 ERROR ANALYSIS OVERVIEW")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. VALIDATE FINAL TEST PREDICTIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING FINAL TEST PREDICTIONS")
logger.info("-" * 70)

assert len(y_test) == len(y_test_pred)
assert len(y_test) == len(y_test_proba)

assert len(y_test) == 74

logger.info(
    f"Test Samples                              : {len(y_test)}"
)

logger.info(
    f"Model Classes                             : "
    f"{list(final_explainability_model.classes_)}"
)

# ------------------------------------------------------------------------------
# 2. CREATE ERROR ANALYSIS BASE TABLE
# ------------------------------------------------------------------------------

error_overview_df = pd.DataFrame({
    "Sample Index": y_test.index,
    "Actual Class": np.asarray(y_test),
    "Predicted Class": np.asarray(y_test_pred)
})

error_overview_df["Correct"] = (
    error_overview_df["Actual Class"]
    == error_overview_df["Predicted Class"]
)

error_overview_df["Error"] = (
    ~error_overview_df["Correct"]
)

# ------------------------------------------------------------------------------
# 3. CALCULATE OVERALL ERROR STATISTICS
# ------------------------------------------------------------------------------

total_samples = len(
    error_overview_df
)

correct_count = int(
    error_overview_df["Correct"].sum()
)

error_count = int(
    error_overview_df["Error"].sum()
)

accuracy = (
    correct_count / total_samples
)

error_rate = (
    error_count / total_samples
)

logger.info(
    f"Total Test Samples                       : {total_samples}"
)

logger.info(
    f"Correct Predictions                      : {correct_count}"
)

logger.info(
    f"Incorrect Predictions                    : {error_count}"
)

logger.info(
    f"Test Accuracy                            : {accuracy:.4f}"
)

logger.info(
    f"Overall Error Rate                       : {error_rate:.4f}"
)

# ------------------------------------------------------------------------------
# 4. ERROR OVERVIEW SUMMARY
# ------------------------------------------------------------------------------

error_overview_summary_df = pd.DataFrame({
    "Metric": [
        "Total Test Samples",
        "Correct Predictions",
        "Incorrect Predictions",
        "Test Accuracy",
        "Overall Error Rate"
    ],
    "Value": [
        total_samples,
        correct_count,
        error_count,
        accuracy,
        error_rate
    ]
})

display(
    error_overview_summary_df
)

# ------------------------------------------------------------------------------
# 5. MISCLASSIFICATION PAIRS
# ------------------------------------------------------------------------------

misclassification_pairs = (
    error_overview_df[
        error_overview_df["Error"]
    ]
    .groupby(
        ["Actual Class", "Predicted Class"]
    )
    .size()
    .reset_index(
        name="Error Count"
    )
    .sort_values(
        "Error Count",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    misclassification_pairs
)

# ------------------------------------------------------------------------------
# 6. VALIDATE KNOWN PROJECT RESULTS
# ------------------------------------------------------------------------------

assert correct_count == 72
assert error_count == 2

assert np.isclose(
    accuracy,
    0.9730,
    atol=0.001
)

assert np.isclose(
    error_rate,
    2 / 74,
    atol=1e-10
)

assert len(
    misclassification_pairs
) == 2

assert set(
    zip(
        misclassification_pairs["Actual Class"],
        misclassification_pairs["Predicted Class"]
    )
) == {
    (2, 4),
    (4, 2)
}

logger.info("-" * 70)
logger.info("15.1 ERROR ANALYSIS VALIDATION")
logger.info("-" * 70)

logger.info(
    "Test Sample Count Validation              : PASSED"
)

logger.info(
    "Prediction Length Validation              : PASSED"
)

logger.info(
    "Correct Prediction Count Validation       : PASSED"
)

logger.info(
    "Error Count Validation                    : PASSED"
)

logger.info(
    "Known Misclassification Pair Validation   : PASSED"
)

logger.info(
    "Overall Error Rate Validation             : PASSED"
)

# ------------------------------------------------------------------------------
# 7. STORE OVERVIEW RESULTS
# ------------------------------------------------------------------------------

final_error_overview = (
    error_overview_df.copy()
)

final_error_overview_summary = (
    error_overview_summary_df.copy()
)

final_error_misclassification_pairs = (
    misclassification_pairs.copy()
)

final_error_count = error_count
final_error_rate = error_rate

logger.info("-" * 70)
logger.info("STORING 15.1 RESULTS")
logger.info("-" * 70)

logger.info(
    "Error Overview Dataset                    : STORED"
)

logger.info(
    "Error Overview Summary                   : STORED"
)

logger.info(
    "Misclassification Pair Summary           : STORED"
)

logger.info(
    "Error Count                               : STORED"
)

logger.info(
    "Error Rate                                : STORED"
)

logger.info("=" * 70)
logger.info(
    "15.1 ERROR ANALYSIS OVERVIEW : COMPLETED"
)
logger.info("=" * 70)
````

## 15.2 Misclassified Samples

This subsection performs a detailed examination of every incorrectly classified observation in the held-out test set.

The selected Optimized Random Forest produced two incorrect predictions:

- Actual Class 2 → Predicted Class 4
- Actual Class 4 → Predicted Class 2

For each misclassified observation, the analysis examines:

- Original test sample index.
- Actual class.
- Predicted class.
- Prediction confidence.
- Probability assigned to every class.
- Difference between the highest and second-highest predicted probabilities.
- Original clinical and histopathological feature values.
- Whether the error represents confusion between similar classes.
- Local SHAP contributions for the predicted class.

The analysis uses the already validated test predictions and probability outputs from Section 12.

No test observations are modified during this analysis.

Because the test set contains only two errors, the findings are treated as case-level observations rather than statistically generalizable failure patterns.

````python


# Required Libraries
import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.2 MISCLASSIFIED SAMPLES")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. VALIDATE REQUIRED DATA
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED ERROR ANALYSIS DATA")
logger.info("-" * 70)

assert len(y_test) == len(y_test_pred)
assert len(y_test) == len(y_test_proba)

assert len(
    final_explainability_feature_names
) == X_test_explain.shape[1]

assert len(
    y_test
) == X_test_explain.shape[0]

logger.info(
    f"Test Samples                              : {len(y_test)}"
)

logger.info(
    f"Transformed Features                      : "
    f"{X_test_explain.shape[1]}"
)

# ------------------------------------------------------------------------------
# 2. IDENTIFY MISCLASSIFIED SAMPLE POSITIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING MISCLASSIFIED SAMPLES")
logger.info("-" * 70)

y_test_array = np.asarray(
    y_test
)

y_test_pred_array = np.asarray(
    y_test_pred
)

misclassified_positions = np.where(
    y_test_array != y_test_pred_array
)[0]

logger.info(
    f"Misclassified Sample Count                : "
    f"{len(misclassified_positions)}"
)

assert len(
    misclassified_positions
) == 2

# ------------------------------------------------------------------------------
# 3. CREATE MISCLASSIFICATION SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING MISCLASSIFICATION SUMMARY")
logger.info("-" * 70)

misclassified_records = []

for position in misclassified_positions:

    actual_class = y_test_array[position]

    predicted_class = y_test_pred_array[position]

    probabilities = np.asarray(
        y_test_proba[position]
    )

    sorted_probability_indices = np.argsort(
        probabilities
    )[::-1]

    highest_probability = (
        probabilities[
            sorted_probability_indices[0]
        ]
    )

    second_highest_probability = (
        probabilities[
            sorted_probability_indices[1]
        ]
    )

    probability_margin = (
        highest_probability -
        second_highest_probability
    )

    misclassified_records.append({
        "Test Position": position,
        "Sample Index": y_test.index[position],
        "Actual Class": actual_class,
        "Predicted Class": predicted_class,
        "Prediction Confidence": highest_probability,
        "Second Highest Probability":
            second_highest_probability,
        "Probability Margin":
            probability_margin
    })

misclassified_summary_df = pd.DataFrame(
    misclassified_records
)

display(
    misclassified_summary_df.style.format({
        "Prediction Confidence": "{:.4f}",
        "Second Highest Probability": "{:.4f}",
        "Probability Margin": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 4. COMPLETE CLASS PROBABILITY DISTRIBUTIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("COMPLETE PROBABILITY DISTRIBUTIONS")
logger.info("-" * 70)

model_classes = np.asarray(
    final_explainability_model.classes_
)

probability_records = []

for position in misclassified_positions:

    probabilities = np.asarray(
        y_test_proba[position]
    )

    record = {
        "Test Position": position,
        "Sample Index": y_test.index[position],
        "Actual Class":
            y_test_array[position],
        "Predicted Class":
            y_test_pred_array[position]
    }

    for class_label, probability in zip(
        model_classes,
        probabilities
    ):

        record[
            f"Probability Class {class_label}"
        ] = probability

    probability_records.append(
        record
    )

misclassified_probability_df = pd.DataFrame(
    probability_records
)

display(
    misclassified_probability_df.style.format(
        {
            column: "{:.4f}"
            for column in
            misclassified_probability_df.columns
            if "Probability Class" in column
        }
    )
)

# ------------------------------------------------------------------------------
# 5. ORIGINAL FEATURE VALUES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ORIGINAL FEATURE VALUES OF MISCLASSIFIED SAMPLES")
logger.info("-" * 70)

# Retrieve original test rows.
#
# X_test contains the raw, original feature representation.
# It has not been modified by the explainability transformation.

misclassified_original_samples = (
    X_test.iloc[
        misclassified_positions
    ]
    .copy()
)

misclassified_original_samples.insert(
    0,
    "Sample Index",
    y_test.index[
        misclassified_positions
    ]
)

misclassified_original_samples.insert(
    1,
    "Actual Class",
    y_test_array[
        misclassified_positions
    ]
)

misclassified_original_samples.insert(
    2,
    "Predicted Class",
    y_test_pred_array[
        misclassified_positions
    ]
)

display(
    misclassified_original_samples
)

# ------------------------------------------------------------------------------
# 6. COMPARE THE TWO MISCLASSIFIED CASES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISCLASSIFICATION CASE COMPARISON")
logger.info("-" * 70)

case_comparison_records = []

for position in misclassified_positions:

    actual_class = y_test_array[position]

    predicted_class = y_test_pred_array[position]

    confidence = np.max(
        y_test_proba[position]
    )

    case_comparison_records.append({
        "Test Position": position,
        "Sample Index": y_test.index[position],
        "Error Pattern":
            f"{actual_class} → {predicted_class}",
        "Prediction Confidence":
            confidence
    })

case_comparison_df = pd.DataFrame(
    case_comparison_records
)

display(
    case_comparison_df.style.format({
        "Prediction Confidence": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 7. LOCAL SHAP ANALYSIS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOCAL SHAP ANALYSIS OF MISCLASSIFIED SAMPLES")
logger.info("-" * 70)

shap_values_array = np.asarray(
    final_shap_values_by_class
)

local_shap_records = []

for position in misclassified_positions:

    predicted_class = y_test_pred_array[position]

    class_position = np.where(
        model_classes == predicted_class
    )[0]

    assert len(class_position) == 1

    class_position = class_position[0]

    local_shap_values = (
        shap_values_array[
            class_position,
            position,
            :
        ]
    )

    absolute_local_shap = np.abs(
        local_shap_values
    )

    top_indices = np.argsort(
        absolute_local_shap
    )[::-1][:10]

    for rank, feature_index in enumerate(
        top_indices,
        start=1
    ):

        local_shap_records.append({
            "Test Position": position,
            "Sample Index": y_test.index[position],
            "Actual Class": y_test_array[position],
            "Predicted Class": predicted_class,
            "Rank": rank,
            "Feature":
                final_explainability_feature_names[
                    feature_index
                ],
            "SHAP Value":
                local_shap_values[
                    feature_index
                ],
            "Absolute SHAP":
                absolute_local_shap[
                    feature_index
                ]
        })

misclassified_local_shap_df = pd.DataFrame(
    local_shap_records
)

display(
    misclassified_local_shap_df.style.format({
        "SHAP Value": "{:.6f}",
        "Absolute SHAP": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 8. POSITIVE AND NEGATIVE SHAP CONTRIBUTIONS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "STRONGEST POSITIVE AND NEGATIVE SHAP CONTRIBUTIONS"
)
logger.info("-" * 70)

local_shap_direction_records = []

for position in misclassified_positions:

    predicted_class = y_test_pred_array[position]

    class_position = np.where(
        model_classes == predicted_class
    )[0][0]

    local_shap_values = (
        shap_values_array[
            class_position,
            position,
            :
        ]
    )

    positive_indices = np.where(
        local_shap_values > 0
    )[0]

    negative_indices = np.where(
        local_shap_values < 0
    )[0]

    positive_sorted = positive_indices[
        np.argsort(
            local_shap_values[
                positive_indices
            ]
        )[::-1]
    ][:5]

    negative_sorted = negative_indices[
        np.argsort(
            np.abs(
                local_shap_values[
                    negative_indices
                ]
            )
        )[::-1]
    ][:5]

    for feature_index in positive_sorted:

        local_shap_direction_records.append({
            "Test Position": position,
            "Sample Index": y_test.index[position],
            "Actual Class": y_test_array[position],
            "Predicted Class": predicted_class,
            "Direction": "Positive",
            "Feature":
                final_explainability_feature_names[
                    feature_index
                ],
            "SHAP Value":
                local_shap_values[
                    feature_index
                ]
        })

    for feature_index in negative_sorted:

        local_shap_direction_records.append({
            "Test Position": position,
            "Sample Index": y_test.index[position],
            "Actual Class": y_test_array[position],
            "Predicted Class": predicted_class,
            "Direction": "Negative",
            "Feature":
                final_explainability_feature_names[
                    feature_index
                ],
            "SHAP Value":
                local_shap_values[
                    feature_index
                ]
        })

misclassified_shap_direction_df = pd.DataFrame(
    local_shap_direction_records
)

display(
    misclassified_shap_direction_df.style.format({
        "SHAP Value": "{:.6f}"
    })
)

# ------------------------------------------------------------------------------
# 9. ERROR CONFIDENCE CLASSIFICATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "ASSESSING CONFIDENCE OF MISCLASSIFIED PREDICTIONS"
)
logger.info("-" * 70)

confidence_thresholds = {
    "Very Low": 0.40,
    "Low": 0.60,
    "Moderate": 0.80,
    "High": 1.00
}

def classify_prediction_confidence(confidence):

    if confidence < 0.40:
        return "Very Low"

    elif confidence < 0.60:
        return "Low"

    elif confidence < 0.80:
        return "Moderate"

    else:
        return "High"


misclassified_summary_df[
    "Confidence Category"
] = (
    misclassified_summary_df[
        "Prediction Confidence"
    ]
    .apply(
        classify_prediction_confidence
    )
)

display(
    misclassified_summary_df.style.format({
        "Prediction Confidence": "{:.4f}",
        "Second Highest Probability": "{:.4f}",
        "Probability Margin": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 10. MISCLASSIFICATION INTERPRETATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MISCLASSIFICATION INTERPRETATION")
logger.info("-" * 70)

for _, row in misclassified_summary_df.iterrows():

    logger.info(
        f"Sample {row['Sample Index']}: "
        f"Actual Class {row['Actual Class']} → "
        f"Predicted Class {row['Predicted Class']} | "
        f"Confidence = "
        f"{row['Prediction Confidence']:.4f} | "
        f"Margin = "
        f"{row['Probability Margin']:.4f} | "
        f"Category = "
        f"{row['Confidence Category']}"
    )

# ------------------------------------------------------------------------------
# 11. VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("15.2 MISCLASSIFIED SAMPLES VALIDATION")
logger.info("-" * 70)

assert len(
    misclassified_summary_df
) == 2

assert (
    misclassified_summary_df[
        "Actual Class"
    ].tolist()
    == [2, 4]
    or
    misclassified_summary_df[
        "Actual Class"
    ].tolist()
    == [4, 2]
)

assert set(
    zip(
        misclassified_summary_df["Actual Class"],
        misclassified_summary_df["Predicted Class"]
    )
) == {
    (2, 4),
    (4, 2)
}

assert (
    misclassified_original_samples.shape[0]
    == 2
)

assert (
    misclassified_probability_df.shape[0]
    == 2
)

assert (
    misclassified_local_shap_df.shape[0]
    == 20
)

assert (
    misclassified_shap_direction_df.shape[0]
    == 20
)

assert (
    misclassified_summary_df[
        "Prediction Confidence"
    ].between(0, 1).all()
)

assert (
    misclassified_summary_df[
        "Probability Margin"
    ].between(0, 1).all()
)

assert np.allclose(
    y_test_proba.sum(axis=1),
    1.0,
    atol=1e-6
)

logger.info(
    "Misclassified Sample Count Validation      : PASSED"
)

logger.info(
    "Actual → Predicted Pair Validation         : PASSED"
)

logger.info(
    "Original Feature Retrieval Validation     : PASSED"
)

logger.info(
    "Probability Distribution Validation        : PASSED"
)

logger.info(
    "Local SHAP Validation                      : PASSED"
)

logger.info(
    "SHAP Directional Analysis Validation       : PASSED"
)

logger.info(
    "Confidence Range Validation                : PASSED"
)

logger.info(
    "Probability Sum Validation                 : PASSED"
)

# ------------------------------------------------------------------------------
# 12. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STORING 15.2 RESULTS")
logger.info("-" * 70)

final_misclassified_samples = (
    misclassified_summary_df.copy()
)

final_misclassified_probabilities = (
    misclassified_probability_df.copy()
)

final_misclassified_original_features = (
    misclassified_original_samples.copy()
)

final_misclassified_case_comparison = (
    case_comparison_df.copy()
)

final_misclassified_local_shap = (
    misclassified_local_shap_df.copy()
)

final_misclassified_shap_direction = (
    misclassified_shap_direction_df.copy()
)

logger.info(
    "Misclassified Sample Summary               : STORED"
)

logger.info(
    "Misclassified Probability Distributions    : STORED"
)

logger.info(
    "Misclassified Original Features            : STORED"
)

logger.info(
    "Misclassified Case Comparison              : STORED"
)

logger.info(
    "Misclassified Local SHAP                   : STORED"
)

logger.info(
    "Misclassified SHAP Direction Analysis      : STORED"
)

logger.info("=" * 70)
logger.info(
    "15.2 MISCLASSIFIED SAMPLES : COMPLETED"
)
logger.info("=" * 70)
````

## 15.3 Confusion Matrix Error Analysis

This subsection analyzes the confusion matrix of the final Optimized Random Forest on the held-out test set.

The analysis focuses on identifying:

- Correct predictions for each class.
- Incorrect predictions between classes.
- Class-level error concentration.
- Symmetric versus asymmetric confusion.
- The relationship between the confusion matrix and the previously identified misclassified samples.

The final test set contains six classes.

The confusion matrix is analyzed using the original test labels and final model predictions only.

No training data, test data transformation, or model parameters are modified.

Because the test set contains only 74 observations, confusion patterns are interpreted descriptively rather than as statistically conclusive evidence of class similarity.

````python


# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.metrics import confusion_matrix

logger.info("=" * 70)
logger.info("15.3 CONFUSION MATRIX ERROR ANALYSIS")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. VALIDATE INPUT DATA
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CONFUSION MATRIX INPUTS")
logger.info("-" * 70)

y_test_array = np.asarray(
    y_test
)

y_test_pred_array = np.asarray(
    y_test_pred
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert len(model_classes) == 6

assert set(
    model_classes
) == {1, 2, 3, 4, 5, 6}

logger.info(
    f"Test Samples                              : "
    f"{len(y_test_array)}"
)

logger.info(
    f"Model Classes                             : "
    f"{model_classes.tolist()}"
)

# ------------------------------------------------------------------------------
# 2. COMPUTE CONFUSION MATRIX
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("COMPUTING CONFUSION MATRIX")
logger.info("-" * 70)

cm = confusion_matrix(
    y_test_array,
    y_test_pred_array,
    labels=model_classes
)

confusion_matrix_df = pd.DataFrame(
    cm,
    index=[
        f"Actual {class_label}"
        for class_label in model_classes
    ],
    columns=[
        f"Predicted {class_label}"
        for class_label in model_classes
    ]
)

logger.info(
    f"Confusion Matrix Shape                    : "
    f"{cm.shape}"
)

display(
    confusion_matrix_df
)

# ------------------------------------------------------------------------------
# 3. IDENTIFY CORRECT AND INCORRECT CELLS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING CORRECT AND INCORRECT CONFUSION MATRIX CELLS")
logger.info("-" * 70)

diagonal_correct = np.diag(
    cm
)

total_correct_from_cm = int(
    diagonal_correct.sum()
)

total_samples_from_cm = int(
    cm.sum()
)

total_errors_from_cm = (
    total_samples_from_cm -
    total_correct_from_cm
)

logger.info(
    f"Correct Predictions                      : "
    f"{total_correct_from_cm}"
)

logger.info(
    f"Incorrect Predictions                    : "
    f"{total_errors_from_cm}"
)

# ------------------------------------------------------------------------------
# 4. MISCLASSIFICATION CELL ANALYSIS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING MISCLASSIFICATION CELLS")
logger.info("-" * 70)

misclassification_cells = []

for actual_position, actual_class in enumerate(
    model_classes
):

    for predicted_position, predicted_class in enumerate(
        model_classes
    ):

        count = int(
            cm[
                actual_position,
                predicted_position
            ]
        )

        if (
            actual_class != predicted_class
            and count > 0
        ):

            misclassification_cells.append({
                "Actual Class": actual_class,
                "Predicted Class": predicted_class,
                "Error Count": count
            })

misclassification_cells_df = pd.DataFrame(
    misclassification_cells
)

display(
    misclassification_cells_df
)

# ------------------------------------------------------------------------------
# 5. CLASS-WISE CONFUSION SUMMARY
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS-WISE CONFUSION SUMMARY")
logger.info("-" * 70)

class_confusion_records = []

for position, class_label in enumerate(
    model_classes
):

    actual_count = int(
        cm[position].sum()
    )

    correct_count = int(
        cm[position, position]
    )

    error_count = (
        actual_count -
        correct_count
    )

    class_error_rate = (
        error_count / actual_count
        if actual_count > 0
        else 0.0
    )

    class_confusion_records.append({
        "Class": class_label,
        "Actual Samples": actual_count,
        "Correct Predictions": correct_count,
        "Incorrect Predictions": error_count,
        "Error Rate": class_error_rate
    })

class_confusion_summary_df = pd.DataFrame(
    class_confusion_records
)

display(
    class_confusion_summary_df.style.format({
        "Error Rate": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 6. PREDICTION CONFUSION DIRECTION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING CONFUSION DIRECTION")
logger.info("-" * 70)

direction_records = []

for _, row in misclassification_cells_df.iterrows():

    actual_class = row["Actual Class"]
    predicted_class = row["Predicted Class"]
    error_count = row["Error Count"]

    reverse_count = int(
        cm[
            np.where(
                model_classes == predicted_class
            )[0][0],
            np.where(
                model_classes == actual_class
            )[0][0]
        ]
    )

    direction_records.append({
        "Actual → Predicted":
            f"{actual_class} → {predicted_class}",
        "Error Count":
            int(error_count),
        "Reverse Direction Count":
            reverse_count,
        "Symmetric Confusion":
            error_count == reverse_count
    })

confusion_direction_df = pd.DataFrame(
    direction_records
)

display(
    confusion_direction_df
)

# ------------------------------------------------------------------------------
# 7. ERROR CONCENTRATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING ERROR CONCENTRATION")
logger.info("-" * 70)

if len(misclassification_cells_df) > 0:

    total_confusion_errors = int(
        misclassification_cells_df[
            "Error Count"
        ].sum()
    )

    for _, row in misclassification_cells_df.iterrows():

        proportion = (
            row["Error Count"] /
            total_confusion_errors
        )

        logger.info(
            f"Class {row['Actual Class']} → "
            f"Class {row['Predicted Class']} | "
            f"Errors = {int(row['Error Count'])} | "
            f"Error Share = {proportion:.4f}"
        )

else:

    total_confusion_errors = 0

    logger.info(
        "No off-diagonal confusion detected."
    )

# ------------------------------------------------------------------------------
# 8. VISUALIZE CONFUSION MATRIX
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING CONFUSION MATRIX VISUALIZATION")
logger.info("-" * 70)

fig, ax = plt.subplots(
    figsize=(8, 6)
)

image = ax.imshow(
    cm,
    interpolation="nearest"
)

ax.set_title(
    "Optimized Random Forest — Test Confusion Matrix"
)

ax.set_xlabel(
    "Predicted Class"
)

ax.set_ylabel(
    "Actual Class"
)

ax.set_xticks(
    np.arange(len(model_classes))
)

ax.set_yticks(
    np.arange(len(model_classes))
)

ax.set_xticklabels(
    model_classes
)

ax.set_yticklabels(
    model_classes
)

threshold = (
    cm.max() / 2.0
)

for row in range(
    cm.shape[0]
):

    for column in range(
        cm.shape[1]
    ):

        ax.text(
            column,
            row,
            str(cm[row, column]),
            ha="center",
            va="center",
            color="white"
            if cm[row, column] > threshold
            else "black"
        )

plt.tight_layout()

display(
    fig
)

plt.close(
    fig
)

logger.info(
    "Confusion Matrix Visualization            : GENERATED"
)

# ------------------------------------------------------------------------------
# 9. VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("15.3 CONFUSION MATRIX ERROR ANALYSIS VALIDATION")
logger.info("-" * 70)

assert cm.shape == (6, 6)

assert total_samples_from_cm == 74

assert total_correct_from_cm == 72

assert total_errors_from_cm == 2

assert set(
    zip(
        misclassification_cells_df[
            "Actual Class"
        ],
        misclassification_cells_df[
            "Predicted Class"
        ]
    )
) == {
    (2, 4),
    (4, 2)
}

assert (
    misclassification_cells_df[
        "Error Count"
    ].sum()
    == 2
)

assert (
    class_confusion_summary_df[
        "Incorrect Predictions"
    ].sum()
    == 2
)

assert (
    confusion_direction_df[
        "Symmetric Confusion"
    ].all()
)

logger.info(
    "Confusion Matrix Shape Validation          : PASSED"
)

logger.info(
    "Total Sample Count Validation              : PASSED"
)

logger.info(
    "Correct Prediction Count Validation        : PASSED"
)

logger.info(
    "Error Count Validation                     : PASSED"
)

logger.info(
    "Misclassification Pair Validation          : PASSED"
)

logger.info(
    "Class-Wise Error Count Validation           : PASSED"
)

logger.info(
    "Symmetric Confusion Validation              : PASSED"
)

# ------------------------------------------------------------------------------
# 10. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STORING 15.3 RESULTS")
logger.info("-" * 70)

final_confusion_matrix = (
    cm.copy()
)

final_confusion_matrix_df = (
    confusion_matrix_df.copy()
)

final_confusion_misclassification_cells = (
    misclassification_cells_df.copy()
)

final_confusion_class_summary = (
    class_confusion_summary_df.copy()
)

final_confusion_direction_analysis = (
    confusion_direction_df.copy()
)

logger.info(
    "Confusion Matrix                          : STORED"
)

logger.info(
    "Confusion Matrix DataFrame                : STORED"
)

logger.info(
    "Misclassification Cell Analysis           : STORED"
)

logger.info(
    "Class-Wise Confusion Summary              : STORED"
)

logger.info(
    "Confusion Direction Analysis              : STORED"
)

logger.info("=" * 70)
logger.info(
    "15.3 CONFUSION MATRIX ERROR ANALYSIS : COMPLETED"
)
logger.info("=" * 70)
````

## 15.4 Per-Class Error Analysis

This subsection evaluates the final Optimized Random Forest separately for each target class on the held-out test set.

The analysis examines:

- Number of test samples per class.
- Correct predictions per class.
- Incorrect predictions per class.
- Class-specific error rate.
- Class-specific accuracy.
- Precision, recall, and F1-score.
- Number of false positives and false negatives.
- Relative error contribution of each class.

The purpose is to identify whether prediction errors are concentrated in particular target classes rather than relying only on overall accuracy.

The analysis uses the final test predictions and the corresponding held-out test labels.

No training observations or test observations are modified.

````python


# Required Libraries
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.metrics import (
    precision_recall_fscore_support
)

logger.info("=" * 70)
logger.info("15.4 PER-CLASS ERROR ANALYSIS")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. VALIDATE REQUIRED DATA
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PER-CLASS ANALYSIS INPUTS")
logger.info("-" * 70)

y_test_array = np.asarray(
    y_test
)

y_test_pred_array = np.asarray(
    y_test_pred
)

model_classes = np.asarray(
    final_explainability_model.classes_
)

assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert len(model_classes) == 6
assert set(model_classes) == {1, 2, 3, 4, 5, 6}

logger.info(
    f"Test Samples                              : "
    f"{len(y_test_array)}"
)

logger.info(
    f"Number of Classes                         : "
    f"{len(model_classes)}"
)

# ------------------------------------------------------------------------------
# 2. CALCULATE PER-CLASS METRICS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING PER-CLASS PERFORMANCE")
logger.info("-" * 70)

precision, recall, f1_score, support = (
    precision_recall_fscore_support(
        y_test_array,
        y_test_pred_array,
        labels=model_classes,
        zero_division=0
    )
)

per_class_records = []

for position, class_label in enumerate(
    model_classes
):

    actual_samples = int(
        support[position]
    )

    correct_predictions = int(
        np.sum(
            (
                y_test_array == class_label
            )
            &
            (
                y_test_pred_array == class_label
            )
        )
    )

    incorrect_predictions = (
        actual_samples -
        correct_predictions
    )

    class_error_rate = (
        incorrect_predictions /
        actual_samples
        if actual_samples > 0
        else 0.0
    )

    class_accuracy = (
        correct_predictions /
        actual_samples
        if actual_samples > 0
        else 0.0
    )

    # False positives:
    # predicted as this class but actual class was different.
    false_positives = int(
        np.sum(
            (
                y_test_pred_array == class_label
            )
            &
            (
                y_test_array != class_label
            )
        )
    )

    # False negatives:
    # actual class is this class but predicted as another class.
    false_negatives = int(
        np.sum(
            (
                y_test_array == class_label
            )
            &
            (
                y_test_pred_array != class_label
            )
        )
    )

    per_class_records.append({
        "Class": class_label,
        "Test Samples": actual_samples,
        "Correct Predictions": correct_predictions,
        "Incorrect Predictions": incorrect_predictions,
        "Class Accuracy": class_accuracy,
        "Error Rate": class_error_rate,
        "Precision": precision[position],
        "Recall": recall[position],
        "F1 Score": f1_score[position],
        "False Positives": false_positives,
        "False Negatives": false_negatives
    })

per_class_error_df = pd.DataFrame(
    per_class_records
)

display(
    per_class_error_df.style.format({
        "Class Accuracy": "{:.4f}",
        "Error Rate": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 3. ERROR RATE RANKING
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANKING CLASSES BY ERROR RATE")
logger.info("-" * 70)

per_class_error_ranking_df = (
    per_class_error_df
    .sort_values(
        by=[
            "Error Rate",
            "Incorrect Predictions"
        ],
        ascending=[
            False,
            False
        ]
    )
    .reset_index(drop=True)
)

display(
    per_class_error_ranking_df[
        [
            "Class",
            "Test Samples",
            "Correct Predictions",
            "Incorrect Predictions",
            "Error Rate",
            "F1 Score"
        ]
    ].style.format({
        "Error Rate": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 4. ERROR CONTRIBUTION BY CLASS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING ERROR CONTRIBUTION BY CLASS")
logger.info("-" * 70)

total_errors = int(
    np.sum(
        y_test_array != y_test_pred_array
    )
)

per_class_error_df[
    "Error Contribution"
] = np.where(
    total_errors > 0,
    per_class_error_df[
        "Incorrect Predictions"
    ] / total_errors,
    0.0
)

display(
    per_class_error_df[
        [
            "Class",
            "Incorrect Predictions",
            "Error Contribution"
        ]
    ].style.format({
        "Error Contribution": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 5. IDENTIFY BEST AND WORST PERFORMING CLASSES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING BEST AND WORST PERFORMING CLASSES")
logger.info("-" * 70)

worst_error_rate = (
    per_class_error_df[
        "Error Rate"
    ].max()
)

best_error_rate = (
    per_class_error_df[
        "Error Rate"
    ].min()
)

worst_classes = (
    per_class_error_df.loc[
        per_class_error_df[
            "Error Rate"
        ] == worst_error_rate,
        "Class"
    ]
    .tolist()
)

best_classes = (
    per_class_error_df.loc[
        per_class_error_df[
            "Error Rate"
        ] == best_error_rate,
        "Class"
    ]
    .tolist()
)

logger.info(
    f"Highest Class Error Rate                  : "
    f"{worst_error_rate:.4f}"
)

logger.info(
    f"Highest Error Rate Classes                : "
    f"{worst_classes}"
)

logger.info(
    f"Lowest Class Error Rate                   : "
    f"{best_error_rate:.4f}"
)

logger.info(
    f"Lowest Error Rate Classes                 : "
    f"{best_classes}"
)

# ------------------------------------------------------------------------------
# 6. PRECISION / RECALL ERROR INTERPRETATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "ANALYZING FALSE POSITIVES AND FALSE NEGATIVES"
)
logger.info("-" * 70)

for _, row in per_class_error_df.iterrows():

    logger.info(
        f"Class {int(row['Class'])} | "
        f"FP = {int(row['False Positives'])} | "
        f"FN = {int(row['False Negatives'])} | "
        f"Precision = {row['Precision']:.4f} | "
        f"Recall = {row['Recall']:.4f} | "
        f"F1 = {row['F1 Score']:.4f}"
    )

# ------------------------------------------------------------------------------
# 7. PER-CLASS ERROR STATUS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ASSIGNING PER-CLASS ERROR STATUS")
logger.info("-" * 70)

def assign_error_status(error_rate):

    if error_rate == 0:
        return "No Test Errors"

    elif error_rate <= 0.05:
        return "Very Low Error"

    elif error_rate <= 0.10:
        return "Low Error"

    elif error_rate <= 0.20:
        return "Moderate Error"

    else:
        return "High Error"


per_class_error_df[
    "Error Status"
] = (
    per_class_error_df[
        "Error Rate"
    ]
    .apply(assign_error_status)
)

display(
    per_class_error_df[
        [
            "Class",
            "Test Samples",
            "Incorrect Predictions",
            "Error Rate",
            "Error Status"
        ]
    ].style.format({
        "Error Rate": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 8. VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("15.4 PER-CLASS ERROR ANALYSIS VALIDATION")
logger.info("-" * 70)

assert per_class_error_df.shape[0] == 6

assert (
    per_class_error_df[
        "Test Samples"
    ].sum()
    == 74
)

assert (
    per_class_error_df[
        "Correct Predictions"
    ].sum()
    == 72
)

assert (
    per_class_error_df[
        "Incorrect Predictions"
    ].sum()
    == 2
)

assert set(
    per_class_error_df[
        "Class"
    ]
) == {1, 2, 3, 4, 5, 6}

# Known class-specific error pattern.
class_2_row = per_class_error_df[
    per_class_error_df["Class"] == 2
].iloc[0]

class_4_row = per_class_error_df[
    per_class_error_df["Class"] == 4
].iloc[0]

assert int(
    class_2_row["Incorrect Predictions"]
) == 1

assert int(
    class_4_row["Incorrect Predictions"]
) == 1

for class_label in [1, 3, 5, 6]:

    row = per_class_error_df[
        per_class_error_df[
            "Class"
        ] == class_label
    ].iloc[0]

    assert int(
        row["Incorrect Predictions"]
    ) == 0

assert np.isclose(
    class_2_row["Error Rate"],
    1 / 12,
    atol=1e-6
)

assert np.isclose(
    class_4_row["Error Rate"],
    1 / 10,
    atol=1e-6
)

logger.info(
    "Class Count Validation                     : PASSED"
)

logger.info(
    "Test Sample Distribution Validation        : PASSED"
)

logger.info(
    "Correct Prediction Distribution Validation : PASSED"
)

logger.info(
    "Error Distribution Validation              : PASSED"
)

logger.info(
    "Class Label Validation                     : PASSED"
)

logger.info(
    "Class 2 Error Validation                   : PASSED"
)

logger.info(
    "Class 4 Error Validation                   : PASSED"
)

logger.info(
    "Zero-Error Class Validation                : PASSED"
)

logger.info(
    "Class Error Rate Validation                : PASSED"
)

# ------------------------------------------------------------------------------
# 9. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STORING 15.4 RESULTS")
logger.info("-" * 70)

final_per_class_error_analysis = (
    per_class_error_df.copy()
)

final_per_class_error_ranking = (
    per_class_error_ranking_df.copy()
)

logger.info(
    "Per-Class Error Analysis                   : STORED"
)

logger.info(
    "Per-Class Error Ranking                    : STORED"
)

logger.info("=" * 70)
logger.info(
    "15.4 PER-CLASS ERROR ANALYSIS : COMPLETED"
)
logger.info("=" * 70)
````

## 15.5 Minority-Class Error Analysis

This subsection evaluates the performance of the final Optimized Random Forest on the minority classes.

The analysis focuses on classes with comparatively smaller representation in the held-out test set and identifies whether class imbalance is associated with increased prediction errors.

The analysis examines:

- Test-set representation of each class.
- Minority-class sample counts.
- Correct and incorrect predictions.
- Class-specific error rates.
- Precision, recall, and F1-score.
- False positives and false negatives.
- Minority-class contribution to total test errors.
- Whether the smallest class, Class 6, experienced any observed test errors.

The original dataset is imbalanced, with Class 6 representing the smallest target class.

Because the held-out test set contains only 4 Class 6 observations, its observed zero-error result should not be interpreted as proof of strong generalization to unseen Class 6 cases.

This analysis is descriptive and is intended to identify potential minority-class risks for further validation.

````python


# Required Libraries
import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.5 MINORITY-CLASS ERROR ANALYSIS")
logger.info("=" * 70)

# ------------------------------------------------------------------------------
# 1. VALIDATE REQUIRED DATA
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MINORITY-CLASS ANALYSIS INPUTS")
logger.info("-" * 70)

y_test_array = np.asarray(y_test)
y_test_pred_array = np.asarray(y_test_pred)

model_classes = np.asarray(
    final_explainability_model.classes_
)

assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert set(model_classes) == {1, 2, 3, 4, 5, 6}

logger.info(
    f"Test Samples                              : "
    f"{len(y_test_array)}"
)

logger.info(
    f"Target Classes                            : "
    f"{model_classes.tolist()}"
)

# ------------------------------------------------------------------------------
# 2. IDENTIFY MINORITY CLASSES
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING MINORITY CLASSES")
logger.info("-" * 70)

class_counts = (
    pd.Series(y_test_array)
    .value_counts(sort=False)
    .reindex(model_classes, fill_value=0)
)

class_count_df = pd.DataFrame({
    "Class": model_classes,
    "Test Samples": class_counts.values
})

minority_reference_count = int(
    class_count_df["Test Samples"].median()
)

minority_classes = (
    class_count_df.loc[
        class_count_df["Test Samples"]
        < minority_reference_count,
        "Class"
    ]
    .tolist()
)

smallest_test_count = int(
    class_count_df["Test Samples"].min()
)

smallest_classes = (
    class_count_df.loc[
        class_count_df["Test Samples"]
        == smallest_test_count,
        "Class"
    ]
    .tolist()
)

logger.info(
    f"Median Test Class Count                   : "
    f"{minority_reference_count}"
)

logger.info(
    f"Minority Classes Below Median              : "
    f"{minority_classes}"
)

logger.info(
    f"Smallest Test Class Count                  : "
    f"{smallest_test_count}"
)

logger.info(
    f"Smallest Test Classes                      : "
    f"{smallest_classes}"
)

# ------------------------------------------------------------------------------
# 3. BUILD MINORITY-CLASS PERFORMANCE TABLE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING MINORITY-CLASS PERFORMANCE")
logger.info("-" * 70)

minority_records = []

for class_label in model_classes:

    actual_mask = (
        y_test_array == class_label
    )

    predicted_mask = (
        y_test_pred_array == class_label
    )

    actual_count = int(
        actual_mask.sum()
    )

    correct_count = int(
        (
            actual_mask
            &
            predicted_mask
        ).sum()
    )

    incorrect_count = int(
        (
            actual_mask
            &
            ~predicted_mask
        ).sum()
    )

    false_positive_count = int(
        (
            ~actual_mask
            &
            predicted_mask
        ).sum()
    )

    false_negative_count = incorrect_count

    error_rate = (
        incorrect_count / actual_count
        if actual_count > 0
        else 0.0
    )

    minority_records.append({
        "Class": class_label,
        "Test Samples": actual_count,
        "Correct Predictions": correct_count,
        "Incorrect Predictions": incorrect_count,
        "Error Rate": error_rate,
        "False Positives": false_positive_count,
        "False Negatives": false_negative_count,
        "Minority Class":
            class_label in minority_classes,
        "Smallest Class":
            class_label in smallest_classes
    })

minority_analysis_df = pd.DataFrame(
    minority_records
)

# ------------------------------------------------------------------------------
# 4. ADD PRECISION / RECALL / F1
# ------------------------------------------------------------------------------

minority_analysis_df = (
    minority_analysis_df
    .merge(
        final_per_class_error_analysis[
            [
                "Class",
                "Precision",
                "Recall",
                "F1 Score"
            ]
        ],
        on="Class",
        how="left"
    )
)

display(
    minority_analysis_df.style.format({
        "Error Rate": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 5. FOCUS ON SMALLEST CLASS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ANALYZING SMALLEST TARGET CLASS")
logger.info("-" * 70)

smallest_class_rows = (
    minority_analysis_df[
        minority_analysis_df["Smallest Class"]
    ]
    .copy()
)

display(
    smallest_class_rows.style.format({
        "Error Rate": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

for _, row in smallest_class_rows.iterrows():

    logger.info(
        f"Class {int(row['Class'])} | "
        f"Samples = {int(row['Test Samples'])} | "
        f"Errors = {int(row['Incorrect Predictions'])} | "
        f"Recall = {row['Recall']:.4f} | "
        f"F1 = {row['F1 Score']:.4f}"
    )

# ------------------------------------------------------------------------------
# 6. CALCULATE MINORITY-CLASS ERROR CONTRIBUTION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING MINORITY-CLASS ERROR CONTRIBUTION")
logger.info("-" * 70)

total_test_errors = int(
    np.sum(
        y_test_array != y_test_pred_array
    )
)

minority_error_count = int(
    minority_analysis_df.loc[
        minority_analysis_df["Minority Class"],
        "Incorrect Predictions"
    ].sum()
)

minority_error_share = (
    minority_error_count / total_test_errors
    if total_test_errors > 0
    else 0.0
)

logger.info(
    f"Total Test Errors                        : "
    f"{total_test_errors}"
)

logger.info(
    f"Minority-Class Errors                    : "
    f"{minority_error_count}"
)

logger.info(
    f"Minority-Class Error Share               : "
    f"{minority_error_share:.4f}"
)

# ------------------------------------------------------------------------------
# 7. COMPARE MINORITY AND NON-MINORITY PERFORMANCE
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info(
    "COMPARING MINORITY AND NON-MINORITY CLASSES"
)
logger.info("-" * 70)

minority_mask = (
    minority_analysis_df["Minority Class"]
)

minority_correct = int(
    minority_analysis_df.loc[
        minority_mask,
        "Correct Predictions"
    ].sum()
)

minority_samples = int(
    minority_analysis_df.loc[
        minority_mask,
        "Test Samples"
    ].sum()
)

minority_errors = int(
    minority_analysis_df.loc[
        minority_mask,
        "Incorrect Predictions"
    ].sum()
)

non_minority_correct = int(
    minority_analysis_df.loc[
        ~minority_mask,
        "Correct Predictions"
    ].sum()
)

non_minority_samples = int(
    minority_analysis_df.loc[
        ~minority_mask,
        "Test Samples"
    ].sum()
)

non_minority_errors = int(
    minority_analysis_df.loc[
        ~minority_mask,
        "Incorrect Predictions"
    ].sum()
)

minority_accuracy = (
    minority_correct / minority_samples
    if minority_samples > 0
    else 0.0
)

non_minority_accuracy = (
    non_minority_correct / non_minority_samples
    if non_minority_samples > 0
    else 0.0
)

minority_comparison_df = pd.DataFrame({
    "Group": [
        "Minority Classes",
        "Non-Minority Classes"
    ],
    "Test Samples": [
        minority_samples,
        non_minority_samples
    ],
    "Correct Predictions": [
        minority_correct,
        non_minority_correct
    ],
    "Incorrect Predictions": [
        minority_errors,
        non_minority_errors
    ],
    "Accuracy": [
        minority_accuracy,
        non_minority_accuracy
    ]
})

display(
    minority_comparison_df.style.format({
        "Accuracy": "{:.4f}"
    })
)

logger.info(
    f"Minority-Class Accuracy                  : "
    f"{minority_accuracy:.4f}"
)

logger.info(
    f"Non-Minority-Class Accuracy              : "
    f"{non_minority_accuracy:.4f}"
)

# ------------------------------------------------------------------------------
# 8. ASSIGN MINORITY-CLASS RISK STATUS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ASSESSING MINORITY-CLASS RISK")
logger.info("-" * 70)

def minority_risk_status(
    sample_count,
    error_rate
):

    if sample_count <= 5:

        if error_rate == 0:
            return "Limited Evidence — No Observed Test Errors"

        return "High Monitoring Priority"

    elif error_rate > 0.10:

        return "High Monitoring Priority"

    elif error_rate > 0:

        return "Moderate Monitoring Priority"

    else:

        return "Low Observed Error"


minority_analysis_df[
    "Risk Status"
] = minority_analysis_df.apply(
    lambda row:
        minority_risk_status(
            int(row["Test Samples"]),
            float(row["Error Rate"])
        ),
    axis=1
)

display(
    minority_analysis_df[
        [
            "Class",
            "Test Samples",
            "Incorrect Predictions",
            "Error Rate",
            "Recall",
            "F1 Score",
            "Minority Class",
            "Risk Status"
        ]
    ].style.format({
        "Error Rate": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}"
    })
)

# ------------------------------------------------------------------------------
# 9. VALIDATION
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("15.5 MINORITY-CLASS ERROR ANALYSIS VALIDATION")
logger.info("-" * 70)

assert minority_analysis_df.shape[0] == 6

assert (
    minority_analysis_df["Test Samples"].sum()
    == 74
)

assert (
    minority_analysis_df["Correct Predictions"].sum()
    == 72
)

assert (
    minority_analysis_df["Incorrect Predictions"].sum()
    == 2
)

assert (
    minority_reference_count == 11
)

assert minority_classes == [4, 5, 6]

assert smallest_test_count == 4

assert smallest_classes == [6]

# Class 6 validation
class_6_row = minority_analysis_df[
    minority_analysis_df["Class"] == 6
].iloc[0]

assert int(
    class_6_row["Test Samples"]
) == 4

assert int(
    class_6_row["Incorrect Predictions"]
) == 0

assert np.isclose(
    class_6_row["Recall"],
    1.0,
    atol=1e-6
)

assert np.isclose(
    class_6_row["F1 Score"],
    1.0,
    atol=1e-6
)

# Minority-class validation
# Classes 4, 5 and 6 are minority classes.
# Class 4 contributes the only minority-class error.
assert minority_error_count == 1

assert np.isclose(
    minority_error_share,
    0.5,
    atol=1e-6
)

assert total_test_errors == 2

# Minority class samples:
# Class 4 = 10
# Class 5 = 10
# Class 6 = 4
assert minority_samples == 24

# Minority correct:
# Class 4 = 9
# Class 5 = 10
# Class 6 = 4
assert minority_correct == 23

assert minority_errors == 1

assert np.isclose(
    minority_accuracy,
    23 / 24,
    atol=1e-6
)

# Non-minority classes:
# Class 1 = 23
# Class 2 = 12
# Class 3 = 15
assert non_minority_samples == 50

assert non_minority_correct == 49

assert non_minority_errors == 1

assert np.isclose(
    non_minority_accuracy,
    49 / 50,
    atol=1e-6
)

logger.info(
    "Minority Analysis Shape Validation         : PASSED"
)

logger.info(
    "Test Sample Distribution Validation        : PASSED"
)

logger.info(
    "Prediction Distribution Validation         : PASSED"
)

logger.info(
    "Minority Class Identification Validation   : PASSED"
)

logger.info(
    "Smallest Class Identification Validation   : PASSED"
)

logger.info(
    "Class 6 Sample Count Validation            : PASSED"
)

logger.info(
    "Class 6 Error Count Validation             : PASSED"
)

logger.info(
    "Class 6 Recall Validation                  : PASSED"
)

logger.info(
    "Class 6 F1 Validation                      : PASSED"
)

logger.info(
    "Minority Error Count Validation            : PASSED"
)

logger.info(
    "Minority Error Contribution Validation     : PASSED"
)

logger.info(
    "Minority Accuracy Validation               : PASSED"
)

logger.info(
    "Non-Minority Accuracy Validation           : PASSED"
)

# ------------------------------------------------------------------------------
# 10. STORE FINAL RESULTS
# ------------------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STORING 15.5 RESULTS")
logger.info("-" * 70)

final_minority_class_error_analysis = (
    minority_analysis_df.copy()
)

final_minority_class_comparison = (
    minority_comparison_df.copy()
)

final_minority_class_labels = (
    minority_classes.copy()
)

final_smallest_class_labels = (
    smallest_classes.copy()
)

logger.info(
    "Minority-Class Error Analysis              : STORED"
)

logger.info(
    "Minority vs Non-Minority Comparison        : STORED"
)

logger.info(
    "Minority Class Labels                      : STORED"
)

logger.info(
    "Smallest Class Labels                      : STORED"
)

logger.info("=" * 70)
logger.info(
    "15.5 MINORITY-CLASS ERROR ANALYSIS : COMPLETED"
)
logger.info("=" * 70)
````

## 15.6 Class 6 Error Analysis

Class 6 is the smallest target class in the held-out test set, with only 4 observations.

This subsection performs a focused error analysis of Class 6 using the final Optimized Random Forest model.

The analysis evaluates:

- Number of Class 6 test observations.
- Correct and incorrect Class 6 predictions.
- Class 6 precision, recall, and F1-score.
- False positives and false negatives.
- Class 6 prediction confidence.
- Prediction probability distribution for Class 6 observations.
- Whether any Class 6 observations were confused with another class.
- Confidence and margin of Class 6 predictions.
- Overall Class 6 error status.

Because only 4 Class 6 observations are present in the test set, the observed zero-error result provides limited statistical evidence of generalization. Therefore, the result should not be interpreted as proof that the model will perform perfectly on unseen Class 6 cases.

This analysis is descriptive and focuses on model behavior rather than clinical causality.

````python


import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.6 CLASS 6 ERROR ANALYSIS")
logger.info("=" * 70)


# ============================================================================
# REQUIRED LIBRARIES
# ============================================================================
# numpy
# pandas
# IPython.display.display


# ============================================================================
# 1. VALIDATE CLASS 6 ANALYSIS INPUTS
# ============================================================================

logger.info("-" * 70)
logger.info("VALIDATING CLASS 6 ANALYSIS INPUTS")
logger.info("-" * 70)

# Convert required arrays safely
y_test_array = np.asarray(y_test)
y_test_pred_array = np.asarray(y_test_pred)
y_test_proba_array = np.asarray(y_test_proba)

# Model class labels
model_classes = np.asarray(final_explainability_model.classes_)

logger.info(
    f"Test Samples                              : {len(y_test_array)}"
)

logger.info(
    f"Probability Matrix Shape                  : {y_test_proba_array.shape}"
)

logger.info(
    f"Model Classes                             : {model_classes.tolist()}"
)

# Input validations
assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert y_test_proba_array.shape == (74, 6)
assert np.array_equal(model_classes, np.array([1, 2, 3, 4, 5, 6]))


# ============================================================================
# 2. IDENTIFY CLASS 6 TEST OBSERVATIONS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING CLASS 6 TEST OBSERVATIONS")
logger.info("-" * 70)

class_6_actual_mask = y_test_array == 6

class_6_test_positions = np.where(class_6_actual_mask)[0]

class_6_count = int(
    class_6_actual_mask.sum()
)

logger.info(
    f"Class 6 Test Samples                     : {class_6_count}"
)

logger.info(
    f"Class 6 Test Indices                     : "
    f"{class_6_test_positions.tolist()}"
)

assert class_6_count == 4


# ============================================================================
# 3. CALCULATE CLASS 6 PREDICTION PERFORMANCE
# ============================================================================

logger.info("-" * 70)
logger.info("CALCULATING CLASS 6 PREDICTION PERFORMANCE")
logger.info("-" * 70)

# Correct Class 6 predictions
class_6_correct_mask = (
    class_6_actual_mask
    &
    (y_test_pred_array == 6)
)

class_6_incorrect_mask = (
    class_6_actual_mask
    &
    (y_test_pred_array != 6)
)

class_6_correct = int(
    class_6_correct_mask.sum()
)

class_6_incorrect = int(
    class_6_incorrect_mask.sum()
)

# Class 6 recall
class_6_recall = (
    class_6_correct / class_6_count
    if class_6_count > 0
    else np.nan
)

# --------------------------------------------------------------------------
# IMPORTANT:
# False positive means:
# Actual class is NOT 6
# AND predicted class IS 6
# --------------------------------------------------------------------------

class_6_false_positive_mask = (
    (y_test_array != 6)
    &
    (y_test_pred_array == 6)
)

class_6_false_positive = int(
    class_6_false_positive_mask.sum()
)

# False negative means:
# Actual class IS 6
# AND predicted class is NOT 6

class_6_false_negative = int(
    class_6_incorrect_mask.sum()
)

logger.info(
    f"Class 6 Correct Predictions              : {class_6_correct}"
)

logger.info(
    f"Class 6 Incorrect Predictions            : {class_6_incorrect}"
)

logger.info(
    f"Class 6 Recall                           : {class_6_recall:.4f}"
)

logger.info(
    f"Class 6 False Positives                  : {class_6_false_positive}"
)

logger.info(
    f"Class 6 False Negatives                  : {class_6_false_negative}"
)


# ============================================================================
# 4. BUILD CLASS 6 PREDICTION DETAILS
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING CLASS 6 PREDICTION DETAILS")
logger.info("-" * 70)

class_6_probability_index = int(
    np.where(model_classes == 6)[0][0]
)

class_6_probabilities = (
    y_test_proba_array[class_6_test_positions, class_6_probability_index]
)

class_6_prediction_confidence = (
    np.max(
        y_test_proba_array[class_6_test_positions],
        axis=1
    )
)

# Sort probabilities to calculate prediction margin
class_6_sorted_probabilities = np.sort(
    y_test_proba_array[class_6_test_positions],
    axis=1
)

class_6_prediction_margin = (
    class_6_sorted_probabilities[:, -1]
    -
    class_6_sorted_probabilities[:, -2]
)

class_6_detail_df = pd.DataFrame({
    "Test Position": class_6_test_positions,
    "Actual Class": y_test_array[class_6_test_positions],
    "Predicted Class": y_test_pred_array[class_6_test_positions],
    "Correct": class_6_correct_mask[class_6_test_positions],
    "Class 6 Probability": class_6_probabilities,
    "Prediction Confidence": class_6_prediction_confidence,
    "Prediction Margin": class_6_prediction_margin
})

display(class_6_detail_df)


# ============================================================================
# 5. ANALYZE CLASS 6 PROBABILITIES
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING CLASS 6 PROBABILITIES")
logger.info("-" * 70)

class_6_min_probability = float(
    np.min(class_6_probabilities)
)

class_6_max_probability = float(
    np.max(class_6_probabilities)
)

class_6_mean_probability = float(
    np.mean(class_6_probabilities)
)

class_6_min_confidence = float(
    np.min(class_6_prediction_confidence)
)

class_6_max_confidence = float(
    np.max(class_6_prediction_confidence)
)

class_6_mean_confidence = float(
    np.mean(class_6_prediction_confidence)
)

class_6_min_margin = float(
    np.min(class_6_prediction_margin)
)

class_6_mean_margin = float(
    np.mean(class_6_prediction_margin)
)

logger.info(
    f"Minimum Class 6 Probability             : "
    f"{class_6_min_probability:.4f}"
)

logger.info(
    f"Maximum Class 6 Probability             : "
    f"{class_6_max_probability:.4f}"
)

logger.info(
    f"Mean Class 6 Probability                : "
    f"{class_6_mean_probability:.4f}"
)

logger.info(
    f"Minimum Prediction Confidence           : "
    f"{class_6_min_confidence:.4f}"
)

logger.info(
    f"Maximum Prediction Confidence           : "
    f"{class_6_max_confidence:.4f}"
)

logger.info(
    f"Mean Prediction Confidence              : "
    f"{class_6_mean_confidence:.4f}"
)

logger.info(
    f"Minimum Prediction Margin               : "
    f"{class_6_min_margin:.4f}"
)

logger.info(
    f"Mean Prediction Margin                  : "
    f"{class_6_mean_margin:.4f}"
)


# ============================================================================
# 6. ANALYZE CLASS 6 MISCLASSIFICATION PATTERNS
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING CLASS 6 MISCLASSIFICATION PATTERNS")
logger.info("-" * 70)

class_6_error_df = class_6_detail_df[
    class_6_detail_df["Correct"] == False
].copy()

class_6_error_count = len(
    class_6_error_df
)

if class_6_error_count == 0:
    logger.info(
        "Class 6 Misclassifications               : NONE"
    )
else:
    logger.info(
        f"Class 6 Misclassifications               : "
        f"{class_6_error_count}"
    )

    display(class_6_error_df)


# ============================================================================
# 7. ANALYZE CLASS 6 COMPETING PREDICTIONS
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING CLASS 6 COMPETING PREDICTIONS")
logger.info("-" * 70)

class_6_competitor_records = []

for position in class_6_test_positions:

    probability_row = y_test_proba_array[position].copy()

    # Remove Class 6 probability
    probability_row[class_6_probability_index] = -np.inf

    strongest_competitor_index = int(
        np.argmax(probability_row)
    )

    strongest_competing_class = int(
        model_classes[strongest_competitor_index]
    )

    strongest_competing_probability = float(
        probability_row[strongest_competitor_index]
    )

    class_6_probability = float(
        y_test_proba_array[
            position,
            class_6_probability_index
        ]
    )

    class_6_competitor_records.append({
        "Test Position": int(position),
        "Actual Class": int(y_test_array[position]),
        "Predicted Class": int(y_test_pred_array[position]),
        "Class 6 Probability": class_6_probability,
        "Strongest Competing Class": strongest_competing_class,
        "Competing Probability": strongest_competing_probability
    })

class_6_competitor_df = pd.DataFrame(
    class_6_competitor_records
)

display(class_6_competitor_df)


# ============================================================================
# 8. ASSIGN CLASS 6 ERROR STATUS
# ============================================================================

logger.info("-" * 70)
logger.info("ASSIGNING CLASS 6 ERROR STATUS")
logger.info("-" * 70)

if class_6_error_count == 0:

    class_6_status = (
        "No Observed Test Errors — Limited Evidence"
    )

else:

    class_6_status = (
        "Observed Class 6 Errors — Requires Monitoring"
    )

logger.info(
    f"Class 6 Error Status                    : "
    f"{class_6_status}"
)


# ============================================================================
# 9. CLASS 6 ERROR ANALYSIS VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("15.6 CLASS 6 ERROR ANALYSIS VALIDATION")
logger.info("-" * 70)

# Sample count
assert class_6_count == 4

logger.info(
    "Class 6 Sample Count Validation            : PASSED"
)


# Correct predictions
assert class_6_correct == 4

logger.info(
    "Class 6 Correct Prediction Validation      : PASSED"
)


# Error count
assert class_6_incorrect == 0

logger.info(
    "Class 6 Error Count Validation             : PASSED"
)


# Recall
assert class_6_recall == 1.0

logger.info(
    "Class 6 Recall Validation                  : PASSED"
)


# False negatives
assert class_6_false_negative == 0

logger.info(
    "Class 6 False Negative Validation          : PASSED"
)


# False positives
assert class_6_false_positive == 0

logger.info(
    "Class 6 False Positive Validation          : PASSED"
)


# Detail table shape
assert class_6_detail_df.shape[0] == 4

logger.info(
    "Class 6 Detail Shape Validation            : PASSED"
)


# All Class 6 observations must actually be Class 6
assert (
    class_6_detail_df["Actual Class"] == 6
).all()

logger.info(
    "Class 6 Label Validation                   : PASSED"
)


# All Class 6 observations correctly predicted
assert (
    class_6_detail_df["Correct"] == True
).all()

logger.info(
    "Class 6 Correctness Validation             : PASSED"
)


# No Class 6 misclassification records
assert class_6_error_df.empty

logger.info(
    "Class 6 Misclassification Validation       : PASSED"
)


# Status
assert (
    class_6_status
    ==
    "No Observed Test Errors — Limited Evidence"
)

logger.info(
    "Class 6 Status Validation                  : PASSED"
)


# Probability range
assert (
    (class_6_probabilities >= 0)
    &
    (class_6_probabilities <= 1)
).all()

logger.info(
    "Class 6 Probability Range Validation       : PASSED"
)


# Confidence range
assert (
    (class_6_prediction_confidence >= 0)
    &
    (class_6_prediction_confidence <= 1)
).all()

logger.info(
    "Class 6 Confidence Range Validation        : PASSED"
)


# Prediction margin
assert (
    class_6_prediction_margin >= 0
).all()

logger.info(
    "Class 6 Margin Validation                  : PASSED"
)


# Competitor probability range
assert (
    (class_6_competitor_df["Competing Probability"] >= 0)
    &
    (class_6_competitor_df["Competing Probability"] <= 1)
).all()

logger.info(
    "Class 6 Competitor Probability Validation  : PASSED"
)


# ============================================================================
# 10. STORE 15.6 RESULTS
# ============================================================================

logger.info("-" * 70)
logger.info("STORING 15.6 RESULTS")
logger.info("-" * 70)

final_class_6_error_analysis = {
    "test_sample_count": class_6_count,
    "correct_predictions": class_6_correct,
    "incorrect_predictions": class_6_incorrect,
    "recall": class_6_recall,
    "false_positives": class_6_false_positive,
    "false_negatives": class_6_false_negative,
    "minimum_probability": class_6_min_probability,
    "maximum_probability": class_6_max_probability,
    "mean_probability": class_6_mean_probability,
    "minimum_confidence": class_6_min_confidence,
    "maximum_confidence": class_6_max_confidence,
    "mean_confidence": class_6_mean_confidence,
    "minimum_margin": class_6_min_margin,
    "mean_margin": class_6_mean_margin,
    "misclassification_count": class_6_error_count,
    "status": class_6_status
}

final_class_6_competitor_analysis = (
    class_6_competitor_df.copy()
)

final_class_6_errors = (
    class_6_error_df.copy()
)

final_class_6_status = class_6_status


logger.info(
    "Class 6 Error Analysis                    : STORED"
)

logger.info(
    "Class 6 Competitor Analysis               : STORED"
)

logger.info(
    "Class 6 Error Records                     : STORED"
)

logger.info(
    "Class 6 Error Status                      : STORED"
)


# ============================================================================
# 11. FINAL SECTION STATUS
# ============================================================================

logger.info("=" * 70)
logger.info("15.6 CLASS 6 ERROR ANALYSIS : COMPLETED")
logger.info("=" * 70)
````

## 15.7 Low-Confidence Predictions

### Objective

Prediction confidence provides an additional perspective on model reliability by measuring how strongly the final model favors its predicted class.

This subsection identifies low-confidence predictions from the held-out test set and examines their relationship with observed model errors.

### Analysis Scope

* Calculate prediction confidence for every test observation.
* Calculate the prediction margin between the highest and second-highest class probabilities.
* Identify low-confidence predictions using a predefined confidence threshold.
* Compare low-confidence predictions with correctly and incorrectly classified observations.
* Determine whether the observed misclassified samples are associated with low confidence.
* Analyze the most uncertain class predictions.
* Assess whether confidence-based monitoring could support a production warning mechanism.

### Interpretation

Low confidence does not necessarily mean that a prediction is incorrect, and high confidence does not guarantee correctness. Confidence values produced by a machine-learning classifier should therefore be treated as model-specific signals rather than calibrated probabilities unless formal probability calibration has been performed.

The analysis is performed only on the held-out test set and does not modify model predictions or introduce additional training.


````python


import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.7 LOW-CONFIDENCE PREDICTIONS")
logger.info("=" * 70)





# ============================================================================
# 1. VALIDATE LOW-CONFIDENCE ANALYSIS INPUTS
# ============================================================================

logger.info("-" * 70)
logger.info("VALIDATING LOW-CONFIDENCE ANALYSIS INPUTS")
logger.info("-" * 70)

# Convert arrays safely
y_test_array = np.asarray(y_test)
y_test_pred_array = np.asarray(y_test_pred)
y_test_proba_array = np.asarray(y_test_proba)

# Model classes
model_classes = np.asarray(
    final_explainability_model.classes_
)

logger.info(
    f"Test Samples                              : {len(y_test_array)}"
)

logger.info(
    f"Prediction Array Shape                    : "
    f"{y_test_pred_array.shape}"
)

logger.info(
    f"Probability Matrix Shape                  : "
    f"{y_test_proba_array.shape}"
)

logger.info(
    f"Model Classes                             : "
    f"{model_classes.tolist()}"
)

# --------------------------------------------------------------------------
# Input validations
# --------------------------------------------------------------------------

assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert y_test_proba_array.shape == (74, 6)

assert np.array_equal(
    model_classes,
    np.array([1, 2, 3, 4, 5, 6])
)

# Probability rows should sum approximately to 1
probability_row_sums = (
    y_test_proba_array.sum(axis=1)
)

assert np.allclose(
    probability_row_sums,
    1.0,
    atol=1e-6
)

logger.info(
    "Low-Confidence Input Validation             : PASSED"
)


# ============================================================================
# 2. CALCULATE PREDICTION CONFIDENCE
# ============================================================================

logger.info("-" * 70)
logger.info("CALCULATING PREDICTION CONFIDENCE")
logger.info("-" * 70)

# Prediction confidence = highest predicted probability
prediction_confidence = (
    np.max(
        y_test_proba_array,
        axis=1
    )
)

# Index of predicted class
prediction_probability_index = (
    np.argmax(
        y_test_proba_array,
        axis=1
    )
)

# Class corresponding to maximum probability
probability_based_prediction = (
    model_classes[prediction_probability_index]
)

# Verify probability-based prediction agrees with model prediction
assert np.array_equal(
    probability_based_prediction,
    y_test_pred_array
)

# Sort probabilities to obtain the two strongest classes
sorted_probabilities = np.sort(
    y_test_proba_array,
    axis=1
)

highest_probability = (
    sorted_probabilities[:, -1]
)

second_highest_probability = (
    sorted_probabilities[:, -2]
)

# Prediction margin
prediction_margin = (
    highest_probability
    -
    second_highest_probability
)

logger.info(
    f"Minimum Prediction Confidence           : "
    f"{prediction_confidence.min():.4f}"
)

logger.info(
    f"Maximum Prediction Confidence           : "
    f"{prediction_confidence.max():.4f}"
)

logger.info(
    f"Mean Prediction Confidence              : "
    f"{prediction_confidence.mean():.4f}"
)

logger.info(
    f"Minimum Prediction Margin               : "
    f"{prediction_margin.min():.4f}"
)

logger.info(
    f"Mean Prediction Margin                  : "
    f"{prediction_margin.mean():.4f}"
)


# ============================================================================
# 3. BUILD COMPLETE CONFIDENCE ANALYSIS TABLE
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING COMPLETE CONFIDENCE ANALYSIS TABLE")
logger.info("-" * 70)

prediction_correct = (
    y_test_array == y_test_pred_array
)

confidence_analysis_df = pd.DataFrame({
    "Test Position": np.arange(len(y_test_array)),
    "Actual Class": y_test_array,
    "Predicted Class": y_test_pred_array,
    "Correct": prediction_correct,
    "Prediction Confidence": prediction_confidence,
    "Second Highest Probability": second_highest_probability,
    "Prediction Margin": prediction_margin
})

display(confidence_analysis_df.head(10))


# ============================================================================
# 4. DEFINE LOW-CONFIDENCE THRESHOLD
# ============================================================================

logger.info("-" * 70)
logger.info("DEFINING LOW-CONFIDENCE THRESHOLD")
logger.info("-" * 70)

# --------------------------------------------------------------------------
# Threshold rationale
#
# For this six-class problem:
# Random-guess baseline = 1 / 6 = 0.1667
#
# A 0.60 threshold identifies predictions where the model's winning
# probability is below a moderate confidence level.
#
# This is an analytical monitoring threshold, NOT a clinical decision
# threshold and NOT a calibrated probability cutoff.
# --------------------------------------------------------------------------

LOW_CONFIDENCE_THRESHOLD = 0.60

logger.info(
    f"Low-Confidence Threshold                 : "
    f"{LOW_CONFIDENCE_THRESHOLD:.2f}"
)


# ============================================================================
# 5. IDENTIFY LOW-CONFIDENCE PREDICTIONS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING LOW-CONFIDENCE PREDICTIONS")
logger.info("-" * 70)

low_confidence_mask = (
    prediction_confidence
    <
    LOW_CONFIDENCE_THRESHOLD
)

low_confidence_positions = np.where(
    low_confidence_mask
)[0]

low_confidence_count = int(
    low_confidence_mask.sum()
)

logger.info(
    f"Low-Confidence Predictions              : "
    f"{low_confidence_count}"
)

logger.info(
    f"Low-Confidence Percentage                : "
    f"{(low_confidence_count / len(y_test_array)) * 100:.2f}%"
)


# ============================================================================
# 6. BUILD LOW-CONFIDENCE PREDICTION TABLE
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING LOW-CONFIDENCE PREDICTION TABLE")
logger.info("-" * 70)

low_confidence_predictions_df = (
    confidence_analysis_df[
        low_confidence_mask
    ]
    .copy()
    .sort_values(
        by="Prediction Confidence",
        ascending=True
    )
    .reset_index(drop=True)
)

display(
    low_confidence_predictions_df
)


# ============================================================================
# 7. ANALYZE LOW-CONFIDENCE CORRECT / INCORRECT PREDICTIONS
# ============================================================================

logger.info("-" * 70)
logger.info(
    "ANALYZING LOW-CONFIDENCE CORRECT / INCORRECT PREDICTIONS"
)
logger.info("-" * 70)

low_confidence_correct_count = int(
    (
        low_confidence_mask
        &
        prediction_correct
    ).sum()
)

low_confidence_incorrect_count = int(
    (
        low_confidence_mask
        &
        (~prediction_correct)
    ).sum()
)

if low_confidence_count > 0:

    low_confidence_accuracy = (
        low_confidence_correct_count
        /
        low_confidence_count
    )

else:

    low_confidence_accuracy = np.nan

logger.info(
    f"Low-Confidence Correct Predictions       : "
    f"{low_confidence_correct_count}"
)

logger.info(
    f"Low-Confidence Incorrect Predictions     : "
    f"{low_confidence_incorrect_count}"
)

logger.info(
    f"Low-Confidence Group Accuracy            : "
    f"{low_confidence_accuracy:.4f}"
    if not np.isnan(low_confidence_accuracy)
    else
    "Low-Confidence Group Accuracy            : N/A"
)


# ============================================================================
# 8. CHECK OVERLAP WITH MISCLASSIFIED SAMPLES
# ============================================================================

logger.info("-" * 70)
logger.info("CHECKING LOW-CONFIDENCE / ERROR OVERLAP")
logger.info("-" * 70)

misclassified_mask = (
    ~prediction_correct
)

misclassified_positions = np.where(
    misclassified_mask
)[0]

low_confidence_error_mask = (
    low_confidence_mask
    &
    misclassified_mask
)

low_confidence_error_positions = np.where(
    low_confidence_error_mask
)[0]

low_confidence_error_count = int(
    low_confidence_error_mask.sum()
)

total_error_count = int(
    misclassified_mask.sum()
)

if total_error_count > 0:

    error_capture_rate = (
        low_confidence_error_count
        /
        total_error_count
    )

else:

    error_capture_rate = np.nan

logger.info(
    f"Total Misclassified Predictions         : "
    f"{total_error_count}"
)

logger.info(
    f"Misclassified Predictions in "
    f"Low-Confidence Group                    : "
    f"{low_confidence_error_count}"
)

logger.info(
    f"Error Capture Rate by Low Confidence    : "
    f"{error_capture_rate:.4f}"
    if not np.isnan(error_capture_rate)
    else
    "Error Capture Rate by Low Confidence    : N/A"
)


# ============================================================================
# 9. BUILD LOW-CONFIDENCE ERROR TABLE
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING LOW-CONFIDENCE ERROR TABLE")
logger.info("-" * 70)

low_confidence_error_df = (
    confidence_analysis_df[
        low_confidence_error_mask
    ]
    .copy()
    .sort_values(
        by="Prediction Confidence",
        ascending=True
    )
    .reset_index(drop=True)
)

if low_confidence_error_df.empty:

    logger.info(
        "Low-Confidence Misclassified Samples     : NONE"
    )

else:

    display(
        low_confidence_error_df
    )


# ============================================================================
# 10. ANALYZE ALL MISCLASSIFIED SAMPLES BY CONFIDENCE
# ============================================================================

logger.info("-" * 70)
logger.info(
    "ANALYZING ALL MISCLASSIFIED SAMPLES BY CONFIDENCE"
)
logger.info("-" * 70)

misclassified_confidence_df = (
    confidence_analysis_df[
        misclassified_mask
    ]
    .copy()
    .sort_values(
        by="Prediction Confidence",
        ascending=True
    )
    .reset_index(drop=True)
)

display(
    misclassified_confidence_df
)


# ============================================================================
# 11. IDENTIFY MOST UNCERTAIN PREDICTIONS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING MOST UNCERTAIN PREDICTIONS")
logger.info("-" * 70)

# Rank all observations from lowest to highest confidence
most_uncertain_predictions_df = (
    confidence_analysis_df
    .sort_values(
        by=[
            "Prediction Confidence",
            "Prediction Margin"
        ],
        ascending=[
            True,
            True
        ]
    )
    .reset_index(drop=True)
)

# Display the 10 least confident predictions
top_uncertain_predictions_df = (
    most_uncertain_predictions_df
    .head(10)
    .copy()
)

display(
    top_uncertain_predictions_df
)


# ============================================================================
# 12. ANALYZE CONFIDENCE BY CORRECTNESS
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING CONFIDENCE BY PREDICTION CORRECTNESS")
logger.info("-" * 70)

correct_confidence = (
    prediction_confidence[
        prediction_correct
    ]
)

incorrect_confidence = (
    prediction_confidence[
        misclassified_mask
    ]
)

correct_mean_confidence = float(
    np.mean(correct_confidence)
)

incorrect_mean_confidence = float(
    np.mean(incorrect_confidence)
)

correct_min_confidence = float(
    np.min(correct_confidence)
)

incorrect_max_confidence = float(
    np.max(incorrect_confidence)
)

logger.info(
    f"Correct Prediction Count                : "
    f"{len(correct_confidence)}"
)

logger.info(
    f"Incorrect Prediction Count              : "
    f"{len(incorrect_confidence)}"
)

logger.info(
    f"Mean Confidence — Correct               : "
    f"{correct_mean_confidence:.4f}"
)

logger.info(
    f"Mean Confidence — Incorrect             : "
    f"{incorrect_mean_confidence:.4f}"
)

logger.info(
    f"Minimum Confidence — Correct            : "
    f"{correct_min_confidence:.4f}"
)

logger.info(
    f"Maximum Confidence — Incorrect          : "
    f"{incorrect_max_confidence:.4f}"
)


# ============================================================================
# 13. IDENTIFY CONFIDENCE RISK LEVELS
# ============================================================================

logger.info("-" * 70)
logger.info("ASSIGNING CONFIDENCE RISK LEVELS")
logger.info("-" * 70)

def assign_confidence_level(confidence):

    if confidence < 0.60:
        return "Low"

    elif confidence < 0.80:
        return "Moderate"

    else:
        return "High"


confidence_analysis_df["Confidence Level"] = (
    confidence_analysis_df[
        "Prediction Confidence"
    ]
    .apply(assign_confidence_level)
)

confidence_level_summary = (
    confidence_analysis_df[
        "Confidence Level"
    ]
    .value_counts()
    .reindex(
        ["Low", "Moderate", "High"],
        fill_value=0
    )
    .rename_axis("Confidence Level")
    .reset_index(
        name="Prediction Count"
    )
)

confidence_level_summary[
    "Percentage"
] = (
    confidence_level_summary["Prediction Count"]
    /
    len(confidence_analysis_df)
    *
    100
)

display(
    confidence_level_summary
)

for _, row in confidence_level_summary.iterrows():

    logger.info(
        f"{row['Confidence Level']} Confidence Predictions"
        f"                  : "
        f"{int(row['Prediction Count'])}"
    )


# ============================================================================
# 14. BUILD LOW-CONFIDENCE CLASS SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING LOW-CONFIDENCE CLASS SUMMARY")
logger.info("-" * 70)

if low_confidence_count > 0:

    low_confidence_class_summary = (
        low_confidence_predictions_df
        .groupby(
            "Predicted Class"
        )
        .agg(
            Low_Confidence_Count=(
                "Predicted Class",
                "size"
            ),
            Mean_Confidence=(
                "Prediction Confidence",
                "mean"
            ),
            Minimum_Confidence=(
                "Prediction Confidence",
                "min"
            ),
            Error_Count=(
                "Correct",
                lambda x: int((~x).sum())
            )
        )
        .reset_index()
        .sort_values(
            by="Low_Confidence_Count",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    low_confidence_class_summary = pd.DataFrame(
        columns=[
            "Predicted Class",
            "Low_Confidence_Count",
            "Mean_Confidence",
            "Minimum_Confidence",
            "Error_Count"
        ]
    )

display(
    low_confidence_class_summary
)


# ============================================================================
# 15. ASSIGN LOW-CONFIDENCE ANALYSIS STATUS
# ============================================================================

logger.info("-" * 70)
logger.info("ASSIGNING LOW-CONFIDENCE ANALYSIS STATUS")
logger.info("-" * 70)

if low_confidence_error_count == total_error_count:

    low_confidence_status = (
        "All Observed Errors Are Low Confidence"
    )

elif low_confidence_error_count > 0:

    low_confidence_status = (
        "Some Observed Errors Are Low Confidence"
    )

else:

    low_confidence_status = (
        "Observed Errors Are Not Low Confidence"
    )

logger.info(
    f"Low-Confidence Analysis Status          : "
    f"{low_confidence_status}"
)


# ============================================================================
# 16. VALIDATE LOW-CONFIDENCE ANALYSIS
# ============================================================================

logger.info("-" * 70)
logger.info("15.7 LOW-CONFIDENCE PREDICTIONS VALIDATION")
logger.info("-" * 70)

# Test sample count
assert len(confidence_analysis_df) == 74

logger.info(
    "Test Sample Count Validation              : PASSED"
)


# Prediction confidence range
assert (
    (prediction_confidence >= 0)
    &
    (prediction_confidence <= 1)
).all()

logger.info(
    "Prediction Confidence Range Validation   : PASSED"
)


# Prediction margin range
assert (
    (prediction_margin >= 0)
    &
    (prediction_margin <= 1)
).all()

logger.info(
    "Prediction Margin Range Validation       : PASSED"
)


# Probability-based predictions must match model predictions
assert np.array_equal(
    probability_based_prediction,
    y_test_pred_array
)

logger.info(
    "Probability Prediction Validation        : PASSED"
)


# Correct + incorrect must equal total
assert (
    int(prediction_correct.sum())
    +
    int(misclassified_mask.sum())
    ==
    74
)

logger.info(
    "Correctness Count Validation             : PASSED"
)


# Low-confidence count
assert (
    low_confidence_count
    ==
    int(low_confidence_mask.sum())
)

logger.info(
    "Low-Confidence Count Validation          : PASSED"
)


# Low-confidence table
assert (
    len(low_confidence_predictions_df)
    ==
    low_confidence_count
)

logger.info(
    "Low-Confidence Table Validation          : PASSED"
)


# Low-confidence errors cannot exceed total errors
assert (
    low_confidence_error_count
    <=
    total_error_count
)

logger.info(
    "Low-Confidence Error Count Validation    : PASSED"
)


# Error capture rate range
if not np.isnan(error_capture_rate):

    assert (
        0
        <=
        error_capture_rate
        <=
        1
    )

logger.info(
    "Error Capture Rate Validation            : PASSED"
)


# Confidence level summary count
assert (
    confidence_level_summary["Prediction Count"].sum()
    ==
    74
)

logger.info(
    "Confidence Level Summary Validation      : PASSED"
)


# Threshold validation
assert (
    LOW_CONFIDENCE_THRESHOLD
    ==
    0.60
)

logger.info(
    "Low-Confidence Threshold Validation      : PASSED"
)


# ============================================================================
# 17. STORE 15.7 RESULTS
# ============================================================================

logger.info("-" * 70)
logger.info("STORING 15.7 RESULTS")
logger.info("-" * 70)

final_low_confidence_analysis = {
    "test_sample_count": int(len(y_test_array)),
    "threshold": float(LOW_CONFIDENCE_THRESHOLD),
    "low_confidence_count": int(low_confidence_count),
    "low_confidence_percentage": float(
        low_confidence_count
        /
        len(y_test_array)
        *
        100
    ),
    "low_confidence_correct_count": int(
        low_confidence_correct_count
    ),
    "low_confidence_incorrect_count": int(
        low_confidence_incorrect_count
    ),
    "low_confidence_accuracy": (
        float(low_confidence_accuracy)
        if not np.isnan(low_confidence_accuracy)
        else np.nan
    ),
    "total_error_count": int(total_error_count),
    "low_confidence_error_count": int(
        low_confidence_error_count
    ),
    "error_capture_rate": (
        float(error_capture_rate)
        if not np.isnan(error_capture_rate)
        else np.nan
    ),
    "minimum_confidence": float(
        prediction_confidence.min()
    ),
    "maximum_confidence": float(
        prediction_confidence.max()
    ),
    "mean_confidence": float(
        prediction_confidence.mean()
    ),
    "minimum_margin": float(
        prediction_margin.min()
    ),
    "mean_margin": float(
        prediction_margin.mean()
    ),
    "correct_mean_confidence": float(
        correct_mean_confidence
    ),
    "incorrect_mean_confidence": float(
        incorrect_mean_confidence
    ),
    "correct_min_confidence": float(
        correct_min_confidence
    ),
    "incorrect_max_confidence": float(
        incorrect_max_confidence
    ),
    "status": low_confidence_status
}

final_low_confidence_predictions = (
    low_confidence_predictions_df.copy()
)

final_low_confidence_errors = (
    low_confidence_error_df.copy()
)

final_most_uncertain_predictions = (
    top_uncertain_predictions_df.copy()
)

final_confidence_level_summary = (
    confidence_level_summary.copy()
)

final_low_confidence_class_summary = (
    low_confidence_class_summary.copy()
)

final_confidence_analysis = (
    confidence_analysis_df.copy()
)

final_low_confidence_status = (
    low_confidence_status
)


logger.info(
    "Low-Confidence Analysis                  : STORED"
)

logger.info(
    "Low-Confidence Predictions               : STORED"
)

logger.info(
    "Low-Confidence Error Records             : STORED"
)

logger.info(
    "Most Uncertain Predictions                : STORED"
)

logger.info(
    "Confidence Level Summary                 : STORED"
)

logger.info(
    "Low-Confidence Class Summary             : STORED"
)

logger.info(
    "Complete Confidence Analysis             : STORED"
)

logger.info(
    "Low-Confidence Analysis Status           : STORED"
)


# ============================================================================
# 18. FINAL SECTION STATUS
# ============================================================================

logger.info("=" * 70)
logger.info(
    "15.7 LOW-CONFIDENCE PREDICTIONS : COMPLETED"
)
logger.info("=" * 70)
````

## 15.8 Common Misclassification Patterns

### Objective

This subsection analyzes recurring patterns among the model's observed misclassifications on the held-out test set.

The analysis focuses on the direction, frequency, confidence, probability competition, and class-level characteristics of the observed errors.

### Analysis Scope

* Identify all observed misclassification pairs.
* Quantify the frequency of each confusion direction.
* Determine whether confusion is symmetric between classes.
* Compare prediction confidence and probability margins across error directions.
* Identify the strongest competing classes involved in each error.
* Compare the observed error pattern with overall class performance.
* Determine whether sufficient evidence exists to describe a recurring model failure pattern.

### Interpretation

A misclassification pattern should only be considered a meaningful recurring pattern when sufficient observations support it. In this test set, only two prediction errors are available. Therefore, repeated confusion involving the same pair of classes can be reported as an observed pattern, but should not be generalized as a definitive model limitation.

The analysis describes model behavior on the held-out test set and does not establish clinical similarity or disease-level causality between classes.


````python


import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.8 COMMON MISCLASSIFICATION PATTERNS")
logger.info("=" * 70)


# ============================================================================
# REQUIRED LIBRARIES
# ============================================================================
# numpy
# pandas
# IPython.display.display


# ============================================================================
# 1. VALIDATE MISCLASSIFICATION ANALYSIS INPUTS
# ============================================================================

logger.info("-" * 70)
logger.info("VALIDATING MISCLASSIFICATION PATTERN INPUTS")
logger.info("-" * 70)

y_test_array = np.asarray(y_test)
y_test_pred_array = np.asarray(y_test_pred)
y_test_proba_array = np.asarray(y_test_proba)

model_classes = np.asarray(
    final_explainability_model.classes_
)

logger.info(
    f"Test Samples                              : "
    f"{len(y_test_array)}"
)

logger.info(
    f"Probability Matrix Shape                  : "
    f"{y_test_proba_array.shape}"
)

logger.info(
    f"Model Classes                             : "
    f"{model_classes.tolist()}"
)

assert len(y_test_array) == 74
assert len(y_test_pred_array) == 74
assert y_test_proba_array.shape == (74, 6)

assert np.array_equal(
    model_classes,
    np.array([1, 2, 3, 4, 5, 6])
)

logger.info(
    "Misclassification Input Validation        : PASSED"
)


# ============================================================================
# 2. IDENTIFY ALL MISCLASSIFIED OBSERVATIONS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING MISCLASSIFIED OBSERVATIONS")
logger.info("-" * 70)

misclassified_mask = (
    y_test_array != y_test_pred_array
)

misclassified_positions = np.where(
    misclassified_mask
)[0]

misclassified_count = int(
    misclassified_mask.sum()
)

logger.info(
    f"Total Misclassified Predictions         : "
    f"{misclassified_count}"
)

assert misclassified_count == 2


# ============================================================================
# 3. BUILD DETAILED MISCLASSIFICATION TABLE
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING MISCLASSIFICATION DETAIL TABLE")
logger.info("-" * 70)

misclassification_records = []

for position in misclassified_positions:

    probability_row = (
        y_test_proba_array[position]
    )

    predicted_index = int(
        np.argmax(probability_row)
    )

    predicted_class = int(
        model_classes[predicted_index]
    )

    # Second-highest probability
    sorted_indices = np.argsort(
        probability_row
    )

    second_index = int(
        sorted_indices[-2]
    )

    second_class = int(
        model_classes[second_index]
    )

    prediction_confidence = float(
        probability_row[predicted_index]
    )

    second_probability = float(
        probability_row[second_index]
    )

    prediction_margin = (
        prediction_confidence
        -
        second_probability
    )

    actual_class = int(
        y_test_array[position]
    )

    misclassification_records.append({
        "Test Position": int(position),
        "Actual Class": actual_class,
        "Predicted Class": predicted_class,
        "Prediction Confidence": prediction_confidence,
        "Second Highest Class": second_class,
        "Second Highest Probability": second_probability,
        "Prediction Margin": prediction_margin
    })

misclassification_detail_df = pd.DataFrame(
    misclassification_records
)

display(
    misclassification_detail_df
)


# ============================================================================
# 4. IDENTIFY MISCLASSIFICATION DIRECTIONS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING MISCLASSIFICATION DIRECTIONS")
logger.info("-" * 70)

misclassification_detail_df[
    "Misclassification Direction"
] = (
    misclassification_detail_df[
        "Actual Class"
    ].astype(str)
    +
    " → "
    +
    misclassification_detail_df[
        "Predicted Class"
    ].astype(str)
)

direction_counts = (
    misclassification_detail_df[
        "Misclassification Direction"
    ]
    .value_counts()
    .rename_axis(
        "Misclassification Direction"
    )
    .reset_index(
        name="Error Count"
    )
)

direction_counts[
    "Error Share"
] = (
    direction_counts["Error Count"]
    /
    misclassified_count
)

display(
    direction_counts
)

for _, row in direction_counts.iterrows():

    logger.info(
        f"{row['Misclassification Direction']:<40} : "
        f"{int(row['Error Count'])}"
    )


# ============================================================================
# 5. IDENTIFY UNIQUE MISCLASSIFICATION PAIRS
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING UNIQUE MISCLASSIFICATION PAIRS")
logger.info("-" * 70)

pair_counts = (
    misclassification_detail_df
    .groupby(
        [
            "Actual Class",
            "Predicted Class"
        ]
    )
    .size()
    .reset_index(
        name="Error Count"
    )
    .sort_values(
        by="Error Count",
        ascending=False
    )
    .reset_index(drop=True)
)

pair_counts[
    "Error Share"
] = (
    pair_counts["Error Count"]
    /
    misclassified_count
)

display(
    pair_counts
)


# ============================================================================
# 6. CHECK FOR SYMMETRIC CLASS CONFUSION
# ============================================================================

logger.info("-" * 70)
logger.info("CHECKING FOR SYMMETRIC CLASS CONFUSION")
logger.info("-" * 70)

symmetric_records = []

for _, row in pair_counts.iterrows():

    actual_class = int(
        row["Actual Class"]
    )

    predicted_class = int(
        row["Predicted Class"]
    )

    reverse_count_series = pair_counts[
        (
            pair_counts["Actual Class"]
            ==
            predicted_class
        )
        &
        (
            pair_counts["Predicted Class"]
            ==
            actual_class
        )
    ]["Error Count"]

    reverse_count = int(
        reverse_count_series.iloc[0]
    ) if not reverse_count_series.empty else 0

    symmetric_records.append({
        "Class Pair": (
            f"{actual_class} ↔ {predicted_class}"
        ),
        "Forward Direction": (
            f"{actual_class} → {predicted_class}"
        ),
        "Forward Error Count": int(
            row["Error Count"]
        ),
        "Reverse Direction": (
            f"{predicted_class} → {actual_class}"
        ),
        "Reverse Error Count": reverse_count,
        "Symmetric Confusion": (
            reverse_count > 0
        )
    })

symmetric_confusion_df = pd.DataFrame(
    symmetric_records
)

# Remove duplicate unordered pairs
symmetric_confusion_df[
    "Normalized Pair"
] = symmetric_confusion_df.apply(
    lambda row: tuple(
        sorted([
            int(
                row["Forward Direction"].split(" → ")[0]
            ),
            int(
                row["Forward Direction"].split(" → ")[1]
            )
        ])
    ),
    axis=1
)

symmetric_confusion_df = (
    symmetric_confusion_df
    .drop_duplicates(
        subset=["Normalized Pair"]
    )
    .drop(
        columns=["Normalized Pair"]
    )
    .reset_index(drop=True)
)

display(
    symmetric_confusion_df
)


# ============================================================================
# 7. ANALYZE PROBABILITY COMPETITION
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING PROBABILITY COMPETITION")
logger.info("-" * 70)

probability_competition_records = []

for position in misclassified_positions:

    probability_row = (
        y_test_proba_array[position]
    )

    actual_class = int(
        y_test_array[position]
    )

    predicted_class = int(
        y_test_pred_array[position]
    )

    predicted_index = int(
        np.where(
            model_classes == predicted_class
        )[0][0]
    )

    actual_index = int(
        np.where(
            model_classes == actual_class
        )[0][0]
    )

    predicted_probability = float(
        probability_row[predicted_index]
    )

    actual_probability = float(
        probability_row[actual_index]
    )

    probability_difference = (
        predicted_probability
        -
        actual_probability
    )

    probability_competition_records.append({
        "Test Position": int(position),
        "Actual Class": actual_class,
        "Predicted Class": predicted_class,
        "Predicted Probability": predicted_probability,
        "Actual Class Probability": actual_probability,
        "Probability Difference": probability_difference
    })

probability_competition_df = pd.DataFrame(
    probability_competition_records
)

display(
    probability_competition_df
)


# ============================================================================
# 8. ANALYZE ERROR CONFIDENCE AND MARGIN
# ============================================================================

logger.info("-" * 70)
logger.info("ANALYZING ERROR CONFIDENCE AND MARGIN")
logger.info("-" * 70)

error_mean_confidence = float(
    misclassification_detail_df[
        "Prediction Confidence"
    ].mean()
)

error_min_confidence = float(
    misclassification_detail_df[
        "Prediction Confidence"
    ].min()
)

error_max_confidence = float(
    misclassification_detail_df[
        "Prediction Confidence"
    ].max()
)

error_mean_margin = float(
    misclassification_detail_df[
        "Prediction Margin"
    ].mean()
)

error_min_margin = float(
    misclassification_detail_df[
        "Prediction Margin"
    ].min()
)

error_max_margin = float(
    misclassification_detail_df[
        "Prediction Margin"
    ].max()
)

logger.info(
    f"Mean Error Confidence                    : "
    f"{error_mean_confidence:.4f}"
)

logger.info(
    f"Minimum Error Confidence                 : "
    f"{error_min_confidence:.4f}"
)

logger.info(
    f"Maximum Error Confidence                 : "
    f"{error_max_confidence:.4f}"
)

logger.info(
    f"Mean Error Prediction Margin             : "
    f"{error_mean_margin:.4f}"
)

logger.info(
    f"Minimum Error Prediction Margin          : "
    f"{error_min_margin:.4f}"
)

logger.info(
    f"Maximum Error Prediction Margin          : "
    f"{error_max_margin:.4f}"
)


# ============================================================================
# 9. IDENTIFY DOMINANT MISCLASSIFICATION PAIR
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING DOMINANT MISCLASSIFICATION PATTERN")
logger.info("-" * 70)

max_error_count = int(
    pair_counts["Error Count"].max()
)

dominant_pairs_df = pair_counts[
    pair_counts["Error Count"] == max_error_count
].copy()

dominant_pairs = [
    (
        int(row["Actual Class"]),
        int(row["Predicted Class"])
    )
    for _, row in dominant_pairs_df.iterrows()
]

dominant_pair_share = (
    max_error_count
    /
    misclassified_count
)

logger.info(
    f"Maximum Errors in One Direction           : "
    f"{max_error_count}"
)

logger.info(
    f"Dominant Pair Error Share                 : "
    f"{dominant_pair_share:.4f}"
)

logger.info(
    f"Dominant Misclassification Pairs          : "
    f"{dominant_pairs}"
)


# ============================================================================
# 10. IDENTIFY RECURRING PATTERN STATUS
# ============================================================================

logger.info("-" * 70)
logger.info("ASSIGNING MISCLASSIFICATION PATTERN STATUS")
logger.info("-" * 70)

unique_error_directions = len(
    direction_counts
)

if misclassified_count == 0:

    misclassification_pattern_status = (
        "No Observed Misclassification Pattern"
    )

elif unique_error_directions == 1:

    misclassification_pattern_status = (
        "Single Observed Misclassification Direction — "
        "Limited Evidence"
    )

elif (
    len(
        symmetric_confusion_df[
            symmetric_confusion_df[
                "Symmetric Confusion"
            ]
        ]
    )
    > 0
):

    misclassification_pattern_status = (
        "Symmetric Class Confusion Observed — "
        "Limited Evidence"
    )

else:

    misclassification_pattern_status = (
        "Multiple Misclassification Directions — "
        "Limited Evidence"
    )

logger.info(
    f"Misclassification Pattern Status        : "
    f"{misclassification_pattern_status}"
)


# ============================================================================
# 11. BUILD PATTERN SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING COMMON MISCLASSIFICATION PATTERN SUMMARY")
logger.info("-" * 70)

pattern_summary_records = []

for _, row in pair_counts.iterrows():

    actual_class = int(
        row["Actual Class"]
    )

    predicted_class = int(
        row["Predicted Class"]
    )

    pair_mask = (
        (misclassification_detail_df["Actual Class"] == actual_class)
        &
        (misclassification_detail_df["Predicted Class"] == predicted_class)
    )

    pair_confidence = (
        misclassification_detail_df.loc[
            pair_mask,
            "Prediction Confidence"
        ]
    )

    pair_margin = (
        misclassification_detail_df.loc[
            pair_mask,
            "Prediction Margin"
        ]
    )

    reverse_mask = (
        (misclassification_detail_df["Actual Class"] == predicted_class)
        &
        (misclassification_detail_df["Predicted Class"] == actual_class)
    )

    reverse_count = int(
        reverse_mask.sum()
    )

    pattern_summary_records.append({
        "Actual Class": actual_class,
        "Predicted Class": predicted_class,
        "Error Count": int(row["Error Count"]),
        "Error Share": float(row["Error Share"]),
        "Mean Confidence": float(
            pair_confidence.mean()
        ),
        "Mean Margin": float(
            pair_margin.mean()
        ),
        "Reverse Direction Errors": reverse_count,
        "Symmetric": reverse_count > 0
    })

misclassification_pattern_summary_df = pd.DataFrame(
    pattern_summary_records
)

display(
    misclassification_pattern_summary_df
)


# ============================================================================
# 12. VALIDATE COMMON MISCLASSIFICATION PATTERNS
# ============================================================================

logger.info("-" * 70)
logger.info("15.8 COMMON MISCLASSIFICATION PATTERNS VALIDATION")
logger.info("-" * 70)

# Total errors
assert misclassified_count == 2

logger.info(
    "Total Error Count Validation              : PASSED"
)


# Pair counts must sum to total errors
assert (
    pair_counts["Error Count"].sum()
    ==
    misclassified_count
)

logger.info(
    "Pair Count Validation                     : PASSED"
)


# Direction counts must sum to total errors
assert (
    direction_counts["Error Count"].sum()
    ==
    misclassified_count
)

logger.info(
    "Direction Count Validation                : PASSED"
)


# Every error must have actual != predicted
assert (
    misclassification_detail_df[
        "Actual Class"
    ]
    !=
    misclassification_detail_df[
        "Predicted Class"
    ]
).all()

logger.info(
    "Misclassification Direction Validation    : PASSED"
)


# Confidence range
assert (
    (
        misclassification_detail_df[
            "Prediction Confidence"
        ] >= 0
    )
    &
    (
        misclassification_detail_df[
            "Prediction Confidence"
        ] <= 1
    )
).all()

logger.info(
    "Error Confidence Range Validation         : PASSED"
)


# Margin range
assert (
    (
        misclassification_detail_df[
            "Prediction Margin"
        ] >= 0
    )
    &
    (
        misclassification_detail_df[
            "Prediction Margin"
        ] <= 1
    )
).all()

logger.info(
    "Error Margin Range Validation             : PASSED"
)


# Probability difference
assert (
    probability_competition_df[
        "Probability Difference"
    ] >= 0
).all()

logger.info(
    "Probability Competition Validation        : PASSED"
)


# Error share
assert np.isclose(
    direction_counts["Error Share"].sum(),
    1.0
)

logger.info(
    "Error Share Validation                    : PASSED"
)


# Known observed confusion directions
observed_directions = set(
    direction_counts[
        "Misclassification Direction"
    ]
)

expected_directions = {
    "2 → 4",
    "4 → 2"
}

assert (
    observed_directions
    ==
    expected_directions
)

logger.info(
    "Observed Error Direction Validation       : PASSED"
)


# Symmetric confusion
assert (
    symmetric_confusion_df[
        "Symmetric Confusion"
    ]
    .any()
)

logger.info(
    "Symmetric Confusion Validation            : PASSED"
)


# Dominant pair count
assert max_error_count == 1

logger.info(
    "Dominant Pair Count Validation            : PASSED"
)


# Pattern status
assert (
    misclassification_pattern_status
    ==
    "Symmetric Class Confusion Observed — "
    "Limited Evidence"
)

logger.info(
    "Pattern Status Validation                 : PASSED"
)


# ============================================================================
# 13. STORE 15.8 RESULTS
# ============================================================================

logger.info("-" * 70)
logger.info("STORING 15.8 RESULTS")
logger.info("-" * 70)

final_misclassification_patterns = (
    misclassification_pattern_summary_df.copy()
)

final_misclassification_direction_counts = (
    direction_counts.copy()
)

final_misclassification_pair_counts = (
    pair_counts.copy()
)

final_symmetric_confusion_analysis = (
    symmetric_confusion_df.copy()
)

final_misclassification_probability_competition = (
    probability_competition_df.copy()
)

final_misclassification_detail = (
    misclassification_detail_df.copy()
)

final_misclassification_pattern_status = (
    misclassification_pattern_status
)

final_misclassification_pattern_metrics = {
    "total_errors": int(misclassified_count),
    "unique_error_directions": int(
        unique_error_directions
    ),
    "maximum_pair_error_count": int(
        max_error_count
    ),
    "dominant_pair_share": float(
        dominant_pair_share
    ),
    "mean_error_confidence": float(
        error_mean_confidence
    ),
    "minimum_error_confidence": float(
        error_min_confidence
    ),
    "maximum_error_confidence": float(
        error_max_confidence
    ),
    "mean_error_margin": float(
        error_mean_margin
    ),
    "minimum_error_margin": float(
        error_min_margin
    ),
    "maximum_error_margin": float(
        error_max_margin
    ),
    "status": misclassification_pattern_status
}


logger.info(
    "Misclassification Pattern Analysis       : STORED"
)

logger.info(
    "Misclassification Direction Counts       : STORED"
)

logger.info(
    "Misclassification Pair Counts            : STORED"
)

logger.info(
    "Symmetric Confusion Analysis             : STORED"
)

logger.info(
    "Probability Competition Analysis         : STORED"
)

logger.info(
    "Misclassification Detail Records         : STORED"
)

logger.info(
    "Misclassification Pattern Metrics        : STORED"
)

logger.info(
    "Misclassification Pattern Status         : STORED"
)


# ============================================================================
# 14. FINAL SECTION STATUS
# ============================================================================

logger.info("=" * 70)
logger.info(
    "15.8 COMMON MISCLASSIFICATION PATTERNS : COMPLETED"
)
logger.info("=" * 70)
````

## 15.9 Model Failure Analysis

### Objective

This subsection investigates the observed failure modes of the final optimized Random Forest model on the held-out test set.

The analysis combines:

* misclassification records,
* prediction confidence,
* prediction margins,
* class-level error rates,
* confusion patterns,
* low-confidence prediction analysis,
* probability competition,
* and previously generated explainability results.

### Failure Analysis Dimensions

The model is evaluated for:

1. **Class Confusion**

   * Which classes are confused by the model?
   * Is the confusion directional or symmetric?

2. **Prediction Uncertainty**

   * Are incorrect predictions associated with low confidence?
   * How large is the decision margin?

3. **Class-Specific Weakness**

   * Which classes have the highest observed error rates?
   * Are minority classes disproportionately affected?

4. **Probability Competition**

   * How closely does the predicted class probability compete with the actual class probability?

5. **Explainability Evidence**

   * Do the observed errors occur in cases where important model features provide competing signals?

6. **Generalization Risk**

   * Is the observed failure pattern consistent with a broader model limitation, or is the available evidence too limited?

### Interpretation Principles

Model failure analysis describes observed model behavior and should not be interpreted as clinical causality.

Because only two test observations are misclassified, conclusions must remain conservative. An observed failure pattern may indicate an area requiring monitoring without proving that the model systematically fails for that class or clinical condition.

The objective is therefore to identify **evidence-based risk areas**, rather than to manufacture explanations for individual errors.


````python


import numpy as np
import pandas as pd
from IPython.display import display

logger.info("=" * 70)
logger.info("15.9 MODEL FAILURE ANALYSIS")
logger.info("=" * 70)


# ============================================================================
# REQUIRED LIBRARIES
# ============================================================================
# numpy
# pandas
# IPython.display.display


# ============================================================================
# 1. VALIDATE CORE MODEL OUTPUTS
# ============================================================================

logger.info("-" * 70)
logger.info("VALIDATING MODEL FAILURE ANALYSIS INPUTS")
logger.info("-" * 70)

y_test_array = np.asarray(y_test)
y_test_pred_array = np.asarray(y_test_pred)
y_test_proba_array = np.asarray(y_test_proba)

model_classes = np.asarray(
    final_explainability_model.classes_
)

misclassified_mask = (
    y_test_array != y_test_pred_array
)

misclassified_positions = np.where(
    misclassified_mask
)[0]

total_test_samples = len(
    y_test_array
)

total_errors = int(
    misclassified_mask.sum()
)

total_correct = (
    total_test_samples
    -
    total_errors
)

logger.info(
    f"Test Samples                              : "
    f"{total_test_samples}"
)

logger.info(
    f"Correct Predictions                       : "
    f"{total_correct}"
)

logger.info(
    f"Incorrect Predictions                    : "
    f"{total_errors}"
)

assert total_test_samples == 74
assert total_correct == 72
assert total_errors == 2

assert y_test_proba_array.shape == (
    total_test_samples,
    len(model_classes)
)

logger.info(
    "Core Model Output Validation              : PASSED"
)


# ============================================================================
# 2. BUILD OBSERVED FAILURE RECORDS
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING OBSERVED FAILURE RECORDS")
logger.info("-" * 70)

failure_records = []

for position in misclassified_positions:

    actual_class = int(
        y_test_array[position]
    )

    predicted_class = int(
        y_test_pred_array[position]
    )

    probability_row = (
        y_test_proba_array[position]
    )

    predicted_index = int(
        np.argmax(probability_row)
    )

    sorted_indices = np.argsort(
        probability_row
    )

    second_index = int(
        sorted_indices[-2]
    )

    predicted_probability = float(
        probability_row[predicted_index]
    )

    actual_index = int(
        np.where(
            model_classes == actual_class
        )[0][0]
    )

    actual_probability = float(
        probability_row[actual_index]
    )

    second_probability = float(
        probability_row[second_index]
    )

    second_class = int(
        model_classes[second_index]
    )

    prediction_margin = (
        predicted_probability
        -
        second_probability
    )

    failure_records.append({
        "Test Position": int(position),
        "Actual Class": actual_class,
        "Predicted Class": predicted_class,
        "Predicted Probability": predicted_probability,
        "Actual Class Probability": actual_probability,
        "Second Highest Class": second_class,
        "Second Highest Probability": second_probability,
        "Prediction Margin": prediction_margin,
        "Misclassification Direction": (
            f"{actual_class} → {predicted_class}"
        )
    })

failure_df = pd.DataFrame(
    failure_records
)

display(
    failure_df
)

assert len(failure_df) == total_errors

logger.info(
    "Failure Record Construction               : PASSED"
)


# ============================================================================
# 3. FAILURE MODE: CLASS CONFUSION
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 1 — CLASS CONFUSION")
logger.info("-" * 70)

confusion_failure_summary = (
    failure_df
    .groupby(
        [
            "Actual Class",
            "Predicted Class"
        ]
    )
    .size()
    .reset_index(
        name="Failure Count"
    )
)

confusion_failure_summary[
    "Failure Share"
] = (
    confusion_failure_summary[
        "Failure Count"
    ]
    /
    total_errors
)

display(
    confusion_failure_summary
)

logger.info(
    "Observed confusion pattern                : "
    "Class 2 ↔ Class 4"
)

logger.info(
    "Confusion structure                       : "
    "Symmetric"
)


# ============================================================================
# 4. FAILURE MODE: PREDICTION UNCERTAINTY
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 2 — PREDICTION UNCERTAINTY")
logger.info("-" * 70)

failure_mean_confidence = float(
    failure_df[
        "Predicted Probability"
    ].mean()
)

failure_min_confidence = float(
    failure_df[
        "Predicted Probability"
    ].min()
)

failure_max_confidence = float(
    failure_df[
        "Predicted Probability"
    ].max()
)

failure_mean_margin = float(
    failure_df[
        "Prediction Margin"
    ].mean()
)

failure_min_margin = float(
    failure_df[
        "Prediction Margin"
    ].min()
)

failure_max_margin = float(
    failure_df[
        "Prediction Margin"
    ].max()
)

logger.info(
    f"Mean Failure Confidence                   : "
    f"{failure_mean_confidence:.4f}"
)

logger.info(
    f"Minimum Failure Confidence                : "
    f"{failure_min_confidence:.4f}"
)

logger.info(
    f"Maximum Failure Confidence                : "
    f"{failure_max_confidence:.4f}"
)

logger.info(
    f"Mean Failure Margin                      : "
    f"{failure_mean_margin:.4f}"
)

logger.info(
    f"Minimum Failure Margin                   : "
    f"{failure_min_margin:.4f}"
)

logger.info(
    f"Maximum Failure Margin                   : "
    f"{failure_max_margin:.4f}"
)


# ============================================================================
# 5. LOW-CONFIDENCE FAILURE CAPTURE
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 3 — LOW-CONFIDENCE ERROR CAPTURE")
logger.info("-" * 70)

LOW_CONFIDENCE_THRESHOLD = 0.60

low_confidence_failure_mask = (
    failure_df[
        "Predicted Probability"
    ]
    <
    LOW_CONFIDENCE_THRESHOLD
)

low_confidence_error_count = int(
    low_confidence_failure_mask.sum()
)

error_capture_rate = (
    low_confidence_error_count
    /
    total_errors
)

logger.info(
    f"Low-Confidence Threshold                 : "
    f"{LOW_CONFIDENCE_THRESHOLD:.2f}"
)

logger.info(
    f"Errors Below Threshold                   : "
    f"{low_confidence_error_count}"
)

logger.info(
    f"Error Capture Rate                       : "
    f"{error_capture_rate:.4f}"
)


# ============================================================================
# 6. CLASS-SPECIFIC FAILURE RISK
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 4 — CLASS-SPECIFIC FAILURE RISK")
logger.info("-" * 70)

class_failure_records = []

for class_label in model_classes:

    class_label = int(class_label)

    class_mask = (
        y_test_array
        ==
        class_label
    )

    class_support = int(
        class_mask.sum()
    )

    class_errors = int(
        (
            class_mask
            &
            misclassified_mask
        ).sum()
    )

    class_correct = (
        class_support
        -
        class_errors
    )

    class_error_rate = (
        class_errors
        /
        class_support
        if class_support > 0
        else np.nan
    )

    class_accuracy = (
        class_correct
        /
        class_support
        if class_support > 0
        else np.nan
    )

    class_failure_records.append({
        "Class": class_label,
        "Test Support": class_support,
        "Correct": class_correct,
        "Errors": class_errors,
        "Error Rate": class_error_rate,
        "Accuracy": class_accuracy
    })

class_failure_df = pd.DataFrame(
    class_failure_records
)

class_failure_df = (
    class_failure_df
    .sort_values(
        by=[
            "Error Rate",
            "Errors"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    class_failure_df
)

highest_class_error_rate = float(
    class_failure_df[
        "Error Rate"
    ].max()
)

highest_risk_classes = (
    class_failure_df[
        class_failure_df[
            "Error Rate"
        ]
        ==
        highest_class_error_rate
    ]["Class"]
    .astype(int)
    .tolist()
)

logger.info(
    f"Highest Observed Class Error Rate        : "
    f"{highest_class_error_rate:.4f}"
)

logger.info(
    f"Highest Observed Risk Classes            : "
    f"{highest_risk_classes}"
)


# ============================================================================
# 7. MINORITY-CLASS FAILURE CHECK
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 5 — MINORITY-CLASS RISK")
logger.info("-" * 70)

test_class_counts = (
    pd.Series(y_test_array)
    .value_counts()
    .sort_index()
)

median_test_class_count = float(
    test_class_counts.median()
)

minority_classes = (
    test_class_counts[
        test_class_counts
        <
        median_test_class_count
    ]
    .index
    .astype(int)
    .tolist()
)

minority_error_count = int(
    sum(
        (
            y_test_array[position]
            in
            minority_classes
        )
        for position in misclassified_positions
    )
)

minority_error_share = (
    minority_error_count
    /
    total_errors
)

logger.info(
    f"Median Test Class Count                  : "
    f"{median_test_class_count:.1f}"
)

logger.info(
    f"Minority Classes                         : "
    f"{minority_classes}"
)

logger.info(
    f"Minority-Class Errors                    : "
    f"{minority_error_count}"
)

logger.info(
    f"Minority-Class Error Share               : "
    f"{minority_error_share:.4f}"
)


# ============================================================================
# 8. PROBABILITY COMPETITION FAILURE ANALYSIS
# ============================================================================

logger.info("-" * 70)
logger.info("FAILURE MODE 6 — PROBABILITY COMPETITION")
logger.info("-" * 70)

probability_competition = failure_df[
    [
        "Test Position",
        "Actual Class",
        "Predicted Class",
        "Predicted Probability",
        "Actual Class Probability",
        "Prediction Margin"
    ]
].copy()

probability_competition[
    "Predicted vs Actual Probability Gap"
] = (
    probability_competition[
        "Predicted Probability"
    ]
    -
    probability_competition[
        "Actual Class Probability"
    ]
)

display(
    probability_competition
)

logger.info(
    "Both observed errors show competition "
    "between the actual and predicted classes."
)

logger.info(
    "The narrowest observed probability gap "
    "is treated as the strongest uncertainty signal."
)


# ============================================================================
# 9. IDENTIFY STRONGEST OBSERVED FAILURE CASE
# ============================================================================

logger.info("-" * 70)
logger.info("IDENTIFYING STRONGEST OBSERVED FAILURE CASE")
logger.info("-" * 70)

strongest_failure_position = int(
    probability_competition.loc[
        probability_competition[
            "Predicted vs Actual Probability Gap"
        ].idxmin(),
        "Test Position"
    ]
)

strongest_failure_gap = float(
    probability_competition[
        "Predicted vs Actual Probability Gap"
    ].min()
)

strongest_failure_row = (
    probability_competition[
        probability_competition[
            "Test Position"
        ]
        ==
        strongest_failure_position
    ]
    .iloc[0]
)

logger.info(
    f"Strongest Uncertainty Test Position      : "
    f"{strongest_failure_position}"
)

logger.info(
    f"Actual Class                             : "
    f"{int(strongest_failure_row['Actual Class'])}"
)

logger.info(
    f"Predicted Class                          : "
    f"{int(strongest_failure_row['Predicted Class'])}"
)

logger.info(
    f"Predicted Probability                    : "
    f"{strongest_failure_row['Predicted Probability']:.4f}"
)

logger.info(
    f"Actual Class Probability                 : "
    f"{strongest_failure_row['Actual Class Probability']:.4f}"
)

logger.info(
    f"Probability Gap                          : "
    f"{strongest_failure_gap:.4f}"
)


# ============================================================================
# 10. FAILURE SEVERITY CLASSIFICATION
# ============================================================================

logger.info("-" * 70)
logger.info("ASSIGNING FAILURE SEVERITY")
logger.info("-" * 70)

if total_errors == 0:

    failure_severity_status = (
        "No Observed Test Errors"
    )

elif total_errors <= 2:

    failure_severity_status = (
        "Low Observed Failure Count — "
        "Monitoring Required"
    )

elif total_errors <= 5:

    failure_severity_status = (
        "Moderate Observed Failure Count — "
        "Investigation Required"
    )

else:

    failure_severity_status = (
        "High Observed Failure Count — "
        "Model Review Required"
    )

logger.info(
    f"Failure Severity Status                  : "
    f"{failure_severity_status}"
)


# ============================================================================
# 11. DETERMINE OVERALL FAILURE MODE
# ============================================================================

logger.info("-" * 70)
logger.info("DETERMINING OVERALL MODEL FAILURE MODE")
logger.info("-" * 70)

if (
    total_errors == 2
    and
    set(
        failure_df[
            "Misclassification Direction"
        ]
    )
    ==
    {
        "2 → 4",
        "4 → 2"
    }
):

    overall_failure_mode = (
        "Low-Frequency Symmetric Class 2 ↔ Class 4 "
        "Decision-Boundary Confusion"
    )

else:

    overall_failure_mode = (
        "Multiple Observed Failure Modes"
    )

logger.info(
    f"Overall Failure Mode                    : "
    f"{overall_failure_mode}"
)


# ============================================================================
# 12. DETERMINE EVIDENCE STRENGTH
# ============================================================================

logger.info("-" * 70)
logger.info("ASSESSING FAILURE ANALYSIS EVIDENCE")
logger.info("-" * 70)

if total_errors <= 2:

    failure_evidence_strength = (
        "Limited Evidence"
    )

elif total_errors <= 5:

    failure_evidence_strength = (
        "Moderate Evidence"
    )

else:

    failure_evidence_strength = (
        "Stronger Observational Evidence"
    )

logger.info(
    f"Failure Evidence Strength                 : "
    f"{failure_evidence_strength}"
)


# ============================================================================
# 13. BUILD MODEL FAILURE SUMMARY
# ============================================================================

logger.info("-" * 70)
logger.info("BUILDING MODEL FAILURE SUMMARY")
logger.info("-" * 70)

model_failure_summary = pd.DataFrame([
    {
        "Failure Dimension": "Class Confusion",
        "Observed Finding": "Class 2 ↔ Class 4",
        "Evidence": "1 error in each direction",
        "Risk Level": "Moderate Monitoring"
    },
    {
        "Failure Dimension": "Prediction Uncertainty",
        "Observed Finding": "Both errors low confidence",
        "Evidence": f"Mean confidence = {failure_mean_confidence:.4f}",
        "Risk Level": "Monitoring"
    },
    {
        "Failure Dimension": "Decision Margin",
        "Observed Finding": "Narrow error margins",
        "Evidence": f"Minimum margin = {failure_min_margin:.4f}",
        "Risk Level": "Monitoring"
    },
    {
        "Failure Dimension": "Low-Confidence Capture",
        "Observed Finding": "All observed errors captured",
        "Evidence": f"Capture rate = {error_capture_rate:.4f}",
        "Risk Level": "Positive Signal"
    },
    {
        "Failure Dimension": "Class-Specific Risk",
        "Observed Finding": "Class 4 highest error rate",
        "Evidence": f"Error rate = {highest_class_error_rate:.4f}",
        "Risk Level": "Moderate Monitoring"
    },
    {
        "Failure Dimension": "Minority-Class Risk",
        "Observed Finding": "One minority-class error",
        "Evidence": f"Share = {minority_error_share:.4f}",
        "Risk Level": "Monitoring"
    },
    {
        "Failure Dimension": "Overall Failure Frequency",
        "Observed Finding": "2 / 74 test samples",
        "Evidence": f"Error rate = {total_errors / total_test_samples:.4f}",
        "Risk Level": "Low Observed Frequency"
    }
])

display(
    model_failure_summary
)


# ============================================================================
# 14. FINAL FAILURE ANALYSIS CONCLUSION
# ============================================================================

logger.info("-" * 70)
logger.info("FINAL MODEL FAILURE ANALYSIS CONCLUSION")
logger.info("-" * 70)

logger.info(
    "The final model produced only two observed "
    "test-set errors."
)

logger.info(
    "Both errors involved reciprocal confusion "
    "between Class 2 and Class 4."
)

logger.info(
    "Both incorrect predictions had confidence "
    "below the analytical low-confidence threshold."
)

logger.info(
    "The strongest uncertainty case had only a "
    "small probability advantage for the predicted class."
)

logger.info(
    "Class 4 had the highest observed class-specific "
    "error rate."
)

logger.info(
    "The available evidence is limited because only "
    "two test errors were observed."
)


# ============================================================================
# 15. VALIDATION
# ============================================================================

logger.info("-" * 70)
logger.info("15.9 MODEL FAILURE ANALYSIS VALIDATION")
logger.info("-" * 70)

# Core counts
assert total_test_samples == 74
assert total_correct == 72
assert total_errors == 2

logger.info(
    "Core Test Count Validation                 : PASSED"
)


# Failure records
assert len(failure_df) == 2

logger.info(
    "Failure Record Count Validation            : PASSED"
)


# All records are genuine errors
assert (
    failure_df["Actual Class"]
    !=
    failure_df["Predicted Class"]
).all()

logger.info(
    "Failure Direction Validation               : PASSED"
)


# Expected failure directions
assert set(
    failure_df[
        "Misclassification Direction"
    ]
) == {
    "2 → 4",
    "4 → 2"
}

logger.info(
    "Failure Pattern Validation                 : PASSED"
)


# Both errors low confidence
assert (
    failure_df[
        "Predicted Probability"
    ]
    <
    LOW_CONFIDENCE_THRESHOLD
).all()

logger.info(
    "Low-Confidence Failure Validation          : PASSED"
)


# Error capture rate
assert np.isclose(
    error_capture_rate,
    1.0
)

logger.info(
    "Error Capture Rate Validation              : PASSED"
)


# Class 4 highest observed error rate
assert 4 in highest_risk_classes

logger.info(
    "Highest Risk Class Validation              : PASSED"
)


# Probability gap must be non-negative
assert (
    probability_competition[
        "Predicted vs Actual Probability Gap"
    ]
    >=
    0
).all()

logger.info(
    "Probability Gap Validation                 : PASSED"
)


# Limited evidence
assert (
    failure_evidence_strength
    ==
    "Limited Evidence"
)

logger.info(
    "Evidence Strength Validation               : PASSED"
)


# Overall failure mode
assert (
    overall_failure_mode
    ==
    "Low-Frequency Symmetric Class 2 ↔ Class 4 "
    "Decision-Boundary Confusion"
)

logger.info(
    "Overall Failure Mode Validation            : PASSED"
)


# ============================================================================
# 16. STORE 15.9 RESULTS
# ============================================================================

logger.info("-" * 70)
logger.info("STORING 15.9 RESULTS")
logger.info("-" * 70)

final_model_failure_records = (
    failure_df.copy()
)

final_model_failure_confusion = (
    confusion_failure_summary.copy()
)

final_model_failure_class_analysis = (
    class_failure_df.copy()
)

final_model_failure_probability_analysis = (
    probability_competition.copy()
)

final_model_failure_summary = (
    model_failure_summary.copy()
)

final_model_failure_metrics = {
    "test_samples": int(total_test_samples),
    "correct_predictions": int(total_correct),
    "incorrect_predictions": int(total_errors),
    "error_rate": float(
        total_errors
        /
        total_test_samples
    ),
    "mean_failure_confidence": float(
        failure_mean_confidence
    ),
    "minimum_failure_confidence": float(
        failure_min_confidence
    ),
    "maximum_failure_confidence": float(
        failure_max_confidence
    ),
    "mean_failure_margin": float(
        failure_mean_margin
    ),
    "minimum_failure_margin": float(
        failure_min_margin
    ),
    "maximum_failure_margin": float(
        failure_max_margin
    ),
    "low_confidence_threshold": float(
        LOW_CONFIDENCE_THRESHOLD
    ),
    "error_capture_rate": float(
        error_capture_rate
    ),
    "minority_error_count": int(
        minority_error_count
    ),
    "minority_error_share": float(
        minority_error_share
    ),
    "highest_class_error_rate": float(
        highest_class_error_rate
    ),
    "highest_risk_classes": (
        highest_risk_classes
    ),
    "strongest_failure_position": int(
        strongest_failure_position
    ),
    "strongest_failure_probability_gap": float(
        strongest_failure_gap
    ),
    "failure_severity_status": (
        failure_severity_status
    ),
    "overall_failure_mode": (
        overall_failure_mode
    ),
    "evidence_strength": (
        failure_evidence_strength
    )
}

final_model_failure_status = (
    "Low-Frequency Symmetric Class 2 ↔ Class 4 "
    "Decision-Boundary Confusion — Limited Evidence"
)

logger.info(
    "Model Failure Records                    : STORED"
)

logger.info(
    "Model Failure Confusion Analysis         : STORED"
)

logger.info(
    "Model Failure Class Analysis             : STORED"
)

logger.info(
    "Model Failure Probability Analysis       : STORED"
)

logger.info(
    "Model Failure Summary                    : STORED"
)

logger.info(
    "Model Failure Metrics                    : STORED"
)

logger.info(
    "Model Failure Status                     : STORED"
)


# ============================================================================
# 17. FINAL SECTION STATUS
# ============================================================================

logger.info("=" * 70)
logger.info(
    "15.9 MODEL FAILURE ANALYSIS : COMPLETED"
)
logger.info("=" * 70)
````

## 16.**MODEL ROBUSTNESS & STABILITY** ## 

## 16.1 Cross-Validation Stability

Cross-validation stability evaluates whether the selected production model maintains consistent performance across multiple validation folds.

For the finalized optimized Random Forest, stratified cross-validation results are assessed using multiclass-appropriate metrics:

- Macro F1
- Balanced Accuracy
- Accuracy
- Macro Precision
- Macro Recall

The analysis focuses on both the average performance and variation across folds.

A small standard deviation indicates that model performance is relatively stable across different training and validation subsets, while a larger standard deviation indicates greater sensitivity to the composition of the validation data.

The test set remains completely untouched during this analysis.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_validate


# ------------------------------------------------------------
# Logging Configuration
# ------------------------------------------------------------

logger = logging.getLogger("skin_disorder_robustness")

if not logger.handlers:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s - %(levelname)s - %(message)s"
    )


logger.info("=" * 70)
logger.info("16.1 CROSS-VALIDATION STABILITY")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Robustness Configuration
# ------------------------------------------------------------

ROBUSTNESS_CV_FOLDS = 5
ROBUSTNESS_RANDOM_STATE = 42

logger.info("-" * 70)
logger.info("ROBUSTNESS CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)

logger.info(
    f"Cross-Validation Folds                     : "
    f"{ROBUSTNESS_CV_FOLDS}"
)

logger.info(
    f"Random State                               : "
    f"{ROBUSTNESS_RANDOM_STATE}"
)

logger.info(
    f"Training Samples                           : "
    f"{len(X_train)}"
)


# ------------------------------------------------------------
# 2. Create CV-Safe Training Copy
# ------------------------------------------------------------
#
# IMPORTANT:
# X_train contains "?" in Age.
#
# We do NOT modify X_train.
#
# We only convert Age:
#
#     "45" -> 45
#     "60" -> 60
#     "?"  -> NaN
#
# The pipeline's median imputer will then fit independently
# inside every CV training fold.
# ------------------------------------------------------------

X_train_cv = X_train.copy(deep=True)

if "Age" in X_train_cv.columns:

    X_train_cv["Age"] = pd.to_numeric(
        X_train_cv["Age"],
        errors="coerce"
    )


# ------------------------------------------------------------
# 3. Validate Temporary CV Representation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("CV-SAFE INPUT VALIDATION")
logger.info("-" * 70)

assert list(
    X_train_cv.columns
) == list(
    X_train.columns
), "CV feature schema changed."

assert len(X_train_cv) == len(X_train), (
    "CV training row count changed."
)

if "Age" in X_train_cv.columns:

    assert pd.api.types.is_numeric_dtype(
        X_train_cv["Age"]
    ), "Age is not numeric after conversion."

logger.info(
    "CV Feature Schema Validation                 : PASSED"
)

logger.info(
    "CV Row Count Validation                       : PASSED"
)

logger.info(
    "Age Numeric Representation                   : PASSED"
)


if "Age" in X_train_cv.columns:

    logger.info(
        f"Age Missing Values for CV                  : "
        f"{X_train_cv['Age'].isna().sum()}"
    )


# ------------------------------------------------------------
# 4. Retrieve Production Pipeline
# ------------------------------------------------------------

robustness_model = (
    best_optimized_search.best_estimator_
)

logger.info("-" * 70)
logger.info("PRODUCTION MODEL")
logger.info("-" * 70)

logger.info(
    f"Model Type                                 : "
    f"{type(robustness_model).__name__}"
)

logger.info(
    f"Pipeline Steps                             : "
    f"{list(robustness_model.named_steps.keys())}"
)


# ------------------------------------------------------------
# 5. Stratified Cross-Validation
# ------------------------------------------------------------

robustness_cv = StratifiedKFold(
    n_splits=ROBUSTNESS_CV_FOLDS,
    shuffle=True,
    random_state=ROBUSTNESS_RANDOM_STATE
)


# ------------------------------------------------------------
# 6. Multiclass Evaluation Metrics
# ------------------------------------------------------------

robustness_scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro",
}


logger.info("-" * 70)
logger.info("CROSS-VALIDATION EVALUATION")
logger.info("-" * 70)

logger.info(
    "Evaluation Strategy                         : "
    "Stratified 5-Fold Cross-Validation"
)

logger.info(
    "Primary Stability Metric                    : "
    "Macro F1"
)


# ------------------------------------------------------------
# 7. Execute Cross-Validation
# ------------------------------------------------------------
#
# The complete pipeline is cloned for every fold.
#
# Therefore:
#
# - preprocessing is fitted only on the fold's training data
# - median imputation is fitted only on the fold's training data
# - one-hot encoding is fitted only on the fold's training data
# - Random Forest is fitted only on the fold's training data
#
# This preserves leakage prevention.
# ------------------------------------------------------------

robustness_cv_results = cross_validate(
    estimator=robustness_model,
    X=X_train_cv,
    y=y_train,
    cv=robustness_cv,
    scoring=robustness_scoring,
    return_train_score=True,
    n_jobs=-1,
    error_score="raise"
)


# ------------------------------------------------------------
# 8. Extract Fold-Level Results
# ------------------------------------------------------------

robustness_fold_results = pd.DataFrame({
    "Fold": np.arange(
        1,
        ROBUSTNESS_CV_FOLDS + 1
    ),
    "Accuracy": robustness_cv_results[
        "test_accuracy"
    ],
    "Balanced Accuracy": robustness_cv_results[
        "test_balanced_accuracy"
    ],
    "Macro Precision": robustness_cv_results[
        "test_precision_macro"
    ],
    "Macro Recall": robustness_cv_results[
        "test_recall_macro"
    ],
    "Macro F1": robustness_cv_results[
        "test_f1_macro"
    ],
})


# ------------------------------------------------------------
# 9. Display Fold-Level Results
# ------------------------------------------------------------

display(
    robustness_fold_results.round(4)
)


# ------------------------------------------------------------
# 10. Calculate Stability Statistics
# ------------------------------------------------------------

robustness_stability_summary = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Balanced Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
    ],
    "Mean": [
        np.mean(
            robustness_cv_results["test_accuracy"]
        ),
        np.mean(
            robustness_cv_results["test_balanced_accuracy"]
        ),
        np.mean(
            robustness_cv_results["test_precision_macro"]
        ),
        np.mean(
            robustness_cv_results["test_recall_macro"]
        ),
        np.mean(
            robustness_cv_results["test_f1_macro"]
        ),
    ],
    "Standard Deviation": [
        np.std(
            robustness_cv_results["test_accuracy"],
            ddof=1
        ),
        np.std(
            robustness_cv_results[
                "test_balanced_accuracy"
            ],
            ddof=1
        ),
        np.std(
            robustness_cv_results[
                "test_precision_macro"
            ],
            ddof=1
        ),
        np.std(
            robustness_cv_results[
                "test_recall_macro"
            ],
            ddof=1
        ),
        np.std(
            robustness_cv_results["test_f1_macro"],
            ddof=1
        ),
    ],
    "Minimum": [
        np.min(
            robustness_cv_results["test_accuracy"]
        ),
        np.min(
            robustness_cv_results[
                "test_balanced_accuracy"
            ]
        ),
        np.min(
            robustness_cv_results[
                "test_precision_macro"
            ]
        ),
        np.min(
            robustness_cv_results[
                "test_recall_macro"
            ]
        ),
        np.min(
            robustness_cv_results["test_f1_macro"]
        ),
    ],
    "Maximum": [
        np.max(
            robustness_cv_results["test_accuracy"]
        ),
        np.max(
            robustness_cv_results[
                "test_balanced_accuracy"
            ]
        ),
        np.max(
            robustness_cv_results[
                "test_precision_macro"
            ]
        ),
        np.max(
            robustness_cv_results[
                "test_recall_macro"
            ]
        ),
        np.max(
            robustness_cv_results["test_f1_macro"]
        ),
    ],
})


# ------------------------------------------------------------
# 11. Display Stability Summary
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("CROSS-VALIDATION STABILITY SUMMARY")
logger.info("-" * 70)

display(
    robustness_stability_summary.round(4)
)


# ------------------------------------------------------------
# 12. Primary Macro F1 Stability
# ------------------------------------------------------------

macro_f1_mean = np.mean(
    robustness_cv_results["test_f1_macro"]
)

macro_f1_std = np.std(
    robustness_cv_results["test_f1_macro"],
    ddof=1
)

macro_f1_min = np.min(
    robustness_cv_results["test_f1_macro"]
)

macro_f1_max = np.max(
    robustness_cv_results["test_f1_macro"]
)


logger.info(
    f"Macro F1 Mean                             : "
    f"{macro_f1_mean:.4f}"
)

logger.info(
    f"Macro F1 Standard Deviation               : "
    f"{macro_f1_std:.4f}"
)

logger.info(
    f"Macro F1 Minimum                           : "
    f"{macro_f1_min:.4f}"
)

logger.info(
    f"Macro F1 Maximum                           : "
    f"{macro_f1_max:.4f}"
)


# ------------------------------------------------------------
# 13. Stability Classification
# ------------------------------------------------------------

if macro_f1_std < 0.01:

    cv_stability_status = "High Stability"

elif macro_f1_std < 0.03:

    cv_stability_status = "Good Stability"

else:

    cv_stability_status = "Higher Fold Variability"


logger.info(
    f"Cross-Validation Stability Status          : "
    f"{cv_stability_status}"
)


# ------------------------------------------------------------
# 14. Leakage-Safety Validation
# ------------------------------------------------------------

# Confirm that the original X_train was not modified.

assert X_train.equals(
    X_train.copy(deep=True)
), "Original X_train was unexpectedly modified."

logger.info(
    "Original Training Data Preservation          : PASSED"
)


# ------------------------------------------------------------
# 15. Statistical Validation
# ------------------------------------------------------------

assert len(
    robustness_fold_results
) == ROBUSTNESS_CV_FOLDS

assert len(
    robustness_cv_results["test_f1_macro"]
) == ROBUSTNESS_CV_FOLDS

assert np.isfinite(
    robustness_cv_results["test_f1_macro"]
).all()

assert 0 <= macro_f1_mean <= 1

assert macro_f1_std >= 0

assert (
    macro_f1_min
    <= macro_f1_mean
    <= macro_f1_max
)


logger.info("-" * 70)
logger.info("16.1 VALIDATION")
logger.info("-" * 70)

logger.info(
    "Fold Count Validation                       : PASSED"
)

logger.info(
    "Metric Validity Validation                   : PASSED"
)

logger.info(
    "Macro F1 Range Validation                    : PASSED"
)

logger.info(
    "Stability Statistics Validation              : PASSED"
)

logger.info(
    "Leakage-Safety Representation Validation     : PASSED"
)


# ------------------------------------------------------------
# 16. Store Results for Later Subsections
# ------------------------------------------------------------

final_cv_stability_results = {
    "cv_folds": ROBUSTNESS_CV_FOLDS,
    "random_state": ROBUSTNESS_RANDOM_STATE,
    "fold_results": robustness_fold_results.copy(),
    "summary": robustness_stability_summary.copy(),
    "macro_f1_mean": float(macro_f1_mean),
    "macro_f1_std": float(macro_f1_std),
    "macro_f1_min": float(macro_f1_min),
    "macro_f1_max": float(macro_f1_max),
    "stability_status": cv_stability_status,
}


logger.info("=" * 70)
logger.info("16.1 CROSS-VALIDATION STABILITY : COMPLETE")
logger.info("=" * 70)
````

## 16.2 Mean ± Standard Deviation

This subsection summarizes cross-validation performance using the Mean ± Standard Deviation representation.

The mean represents the average model performance across the five stratified folds, while the standard deviation represents the variation in performance between folds.

A lower standard deviation indicates greater consistency across validation folds.

The following multiclass metrics are analyzed:

- Accuracy
- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1

Macro F1 is treated as the primary robustness metric because it gives equal importance to all classes and is therefore appropriate for the imbalanced six-class classification problem.

The results are based exclusively on the training dataset through cross-validation. The independent test set remains untouched.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd


# ------------------------------------------------------------
# Logging Configuration
# ------------------------------------------------------------

logger = logging.getLogger("skin_disorder_robustness")

if not logger.handlers:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s - %(levelname)s - %(message)s"
    )


logger.info("=" * 70)
logger.info("16.2 MEAN ± STANDARD DEVIATION")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Validate Previous Section Results
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CROSS-VALIDATION RESULTS")
logger.info("-" * 70)

assert "final_cv_stability_results" in globals(), (
    "16.1 cross-validation results are unavailable."
)

assert (
    final_cv_stability_results["cv_folds"] == 5
), "Expected 5-fold cross-validation results."

logger.info(
    "Cross-Validation Results Availability         : PASSED"
)

logger.info(
    "Five-Fold Configuration Validation            : PASSED"
)


# ------------------------------------------------------------
# 2. Retrieve Stability Summary
# ------------------------------------------------------------

cv_summary = (
    final_cv_stability_results["summary"]
    .copy()
)

logger.info(
    f"Metrics Available                           : "
    f"{len(cv_summary)}"
)


# ------------------------------------------------------------
# 3. Calculate Mean ± SD
# ------------------------------------------------------------

cv_summary["Mean ± SD"] = (
    cv_summary["Mean"].map(
        lambda value: f"{value:.4f}"
    )
    + " ± "
    + cv_summary["Standard Deviation"].map(
        lambda value: f"{value:.4f}"
    )
)


# ------------------------------------------------------------
# 4. Calculate Relative Variability
# ------------------------------------------------------------
#
# Coefficient of Variation:
#
#     CV = Standard Deviation / Mean
#
# This provides a normalized measure of variability.
#
# It is used here only as a descriptive stability indicator.
# ------------------------------------------------------------

cv_summary["Coefficient of Variation"] = (
    cv_summary["Standard Deviation"]
    / cv_summary["Mean"]
)


# ------------------------------------------------------------
# 5. Stability Ranking
# ------------------------------------------------------------
#
# Lower SD = greater fold-to-fold consistency.
# ------------------------------------------------------------

cv_summary["Stability Rank"] = (
    cv_summary["Standard Deviation"]
    .rank(
        method="min",
        ascending=True
    )
    .astype(int)
)


# ------------------------------------------------------------
# 6. Display Mean ± SD Summary
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MEAN ± STANDARD DEVIATION SUMMARY")
logger.info("-" * 70)

mean_sd_display = cv_summary[
    [
        "Metric",
        "Mean ± SD",
        "Coefficient of Variation",
        "Stability Rank",
    ]
].copy()

mean_sd_display[
    "Coefficient of Variation"
] = mean_sd_display[
    "Coefficient of Variation"
].map(
    lambda value: f"{value:.4f}"
)

display(mean_sd_display)


# ------------------------------------------------------------
# 7. Identify Most Stable Metric
# ------------------------------------------------------------

most_stable_index = (
    cv_summary["Standard Deviation"]
    .idxmin()
)

most_stable_metric = (
    cv_summary.loc[
        most_stable_index,
        "Metric"
    ]
)

most_stable_sd = (
    cv_summary.loc[
        most_stable_index,
        "Standard Deviation"
    ]
)


# ------------------------------------------------------------
# 8. Identify Least Stable Metric
# ------------------------------------------------------------

least_stable_index = (
    cv_summary["Standard Deviation"]
    .idxmax()
)

least_stable_metric = (
    cv_summary.loc[
        least_stable_index,
        "Metric"
    ]
)

least_stable_sd = (
    cv_summary.loc[
        least_stable_index,
        "Standard Deviation"
    ]
)


logger.info("-" * 70)
logger.info("STABILITY EXTREMES")
logger.info("-" * 70)

logger.info(
    f"Most Stable Metric                         : "
    f"{most_stable_metric}"
)

logger.info(
    f"Most Stable Standard Deviation              : "
    f"{most_stable_sd:.4f}"
)

logger.info(
    f"Least Stable Metric                        : "
    f"{least_stable_metric}"
)

logger.info(
    f"Least Stable Standard Deviation             : "
    f"{least_stable_sd:.4f}"
)


# ------------------------------------------------------------
# 9. Primary Metric — Macro F1
# ------------------------------------------------------------

macro_f1_row = cv_summary[
    cv_summary["Metric"] == "Macro F1"
]

assert len(macro_f1_row) == 1, (
    "Macro F1 result is unavailable or duplicated."
)

macro_f1_mean = float(
    macro_f1_row["Mean"].iloc[0]
)

macro_f1_sd = float(
    macro_f1_row["Standard Deviation"].iloc[0]
)

macro_f1_cv = float(
    macro_f1_row["Coefficient of Variation"].iloc[0]
)


logger.info("-" * 70)
logger.info("PRIMARY ROBUSTNESS METRIC")
logger.info("-" * 70)

logger.info(
    f"Primary Metric                             : "
    f"Macro F1"
)

logger.info(
    f"Macro F1 Mean                              : "
    f"{macro_f1_mean:.4f}"
)

logger.info(
    f"Macro F1 Standard Deviation                : "
    f"{macro_f1_sd:.4f}"
)

logger.info(
    f"Macro F1 Mean ± SD                         : "
    f"{macro_f1_mean:.4f} ± {macro_f1_sd:.4f}"
)

logger.info(
    f"Macro F1 Coefficient of Variation          : "
    f"{macro_f1_cv:.4f}"
)


# ------------------------------------------------------------
# 10. Stability Interpretation
# ------------------------------------------------------------

if macro_f1_sd < 0.01:

    macro_f1_stability = "High"

elif macro_f1_sd < 0.03:

    macro_f1_stability = "Good"

else:

    macro_f1_stability = "Higher Variability"


logger.info(
    f"Macro F1 Stability Level                   : "
    f"{macro_f1_stability}"
)


# ------------------------------------------------------------
# 11. Numerical Validation
# ------------------------------------------------------------

assert np.isfinite(
    cv_summary["Mean"]
).all()

assert np.isfinite(
    cv_summary["Standard Deviation"]
).all()

assert (
    cv_summary["Standard Deviation"] >= 0
).all()

assert (
    cv_summary["Coefficient of Variation"] >= 0
).all()

assert (
    0 <= macro_f1_mean <= 1
)

assert (
    macro_f1_sd >= 0
)

logger.info("-" * 70)
logger.info("16.2 NUMERICAL VALIDATION")
logger.info("-" * 70)

logger.info(
    "Mean Value Validation                       : PASSED"
)

logger.info(
    "Standard Deviation Validation               : PASSED"
)

logger.info(
    "Coefficient of Variation Validation         : PASSED"
)

logger.info(
    "Macro F1 Validation                         : PASSED"
)


# ------------------------------------------------------------
# 12. Store Results
# ------------------------------------------------------------

final_mean_sd_analysis = {
    "summary": cv_summary.copy(),
    "most_stable_metric": most_stable_metric,
    "most_stable_sd": float(most_stable_sd),
    "least_stable_metric": least_stable_metric,
    "least_stable_sd": float(least_stable_sd),
    "macro_f1_mean": macro_f1_mean,
    "macro_f1_sd": macro_f1_sd,
    "macro_f1_mean_sd": (
        f"{macro_f1_mean:.4f} ± {macro_f1_sd:.4f}"
    ),
    "macro_f1_cv": macro_f1_cv,
    "macro_f1_stability": macro_f1_stability,
}


# ------------------------------------------------------------
# 13. Final Status
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.2 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Mean ± SD Analysis                         : COMPLETE"
)

logger.info(
    f"Primary Stability Result                   : "
    f"{macro_f1_mean:.4f} ± {macro_f1_sd:.4f}"
)

logger.info("=" * 70)
logger.info("16.2 MEAN ± STANDARD DEVIATION : COMPLETE")
logger.info("=" * 70)
````

## 16.3 Per-Class Stability

This subsection evaluates whether model performance remains stable across individual target classes during cross-validation.

Per-class stability is important for this six-class dermatological classification problem because overall Macro F1 can hide variation affecting specific classes.

For each class, cross-validation predictions are generated using stratified folds and the following metrics are calculated for every fold:

- Precision
- Recall
- F1-score

The mean, standard deviation, minimum, and maximum F1-score are then calculated for each class.

The analysis focuses particularly on:

- Classes with higher fold-to-fold variation
- Classes with consistently strong performance
- Classes with relatively limited representation
- Potential instability affecting minority classes

Only the training dataset is used for this analysis. The independent test set remains untouched.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd

from IPython.display import display

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import precision_recall_fscore_support
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler


# --------------------------------------------------------------
# Logging Configuration
# --------------------------------------------------------------

logger = logging.getLogger("skin_disorder_robustness")

if not logger.handlers:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s - %(levelname)s - %(message)s"
    )


# --------------------------------------------------------------
# SECTION HEADER
# --------------------------------------------------------------

logger.info("=" * 70)
logger.info("16.3 PER-CLASS STABILITY")
logger.info("=" * 70)


# --------------------------------------------------------------
# 1. VALIDATE CORE DATA
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CORE DATA")
logger.info("-" * 70)


assert "X_train" in globals(), (
    "X_train is unavailable. Please execute the dataset preparation "
    "and train/test split cells before running Section 16.3."
)

assert "y_train" in globals(), (
    "y_train is unavailable. Please execute the dataset preparation "
    "and train/test split cells before running Section 16.3."
)


assert isinstance(X_train, pd.DataFrame), (
    "X_train must be a pandas DataFrame."
)


assert len(X_train) == len(y_train), (
    "X_train and y_train row counts do not match."
)


assert X_train.shape[0] > 0, (
    "X_train contains no observations."
)


logger.info(
    "X_train Availability                           : PASSED"
)

logger.info(
    "y_train Availability                           : PASSED"
)

logger.info(
    f"Training Samples                              : "
    f"{X_train.shape[0]}"
)

logger.info(
    f"Training Features                             : "
    f"{X_train.shape[1]}"
)


# --------------------------------------------------------------
# 2. PRESERVE ORIGINAL TRAINING DATA
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRESERVING ORIGINAL TRAINING DATA")
logger.info("-" * 70)


per_class_X_original = X_train.copy(deep=True)
per_class_y_original = (
    y_train.copy(deep=True)
    if hasattr(y_train, "copy")
    else np.array(y_train, copy=True)
)


logger.info(
    "Original Training Data Snapshot                : PASSED"
)


# --------------------------------------------------------------
# 3. IDENTIFY TARGET CLASSES
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING TARGET CLASSES")
logger.info("-" * 70)


per_class_labels = np.sort(
    pd.unique(y_train)
)


logger.info(
    f"Target Classes                                : "
    f"{per_class_labels.tolist()}"
)


expected_classes = np.array(
    [1, 2, 3, 4, 5, 6]
)


assert np.array_equal(
    per_class_labels,
    expected_classes
), (
    f"Unexpected target classes: "
    f"{per_class_labels.tolist()}"
)


logger.info(
    "Target Class Validation                       : PASSED"
)


# --------------------------------------------------------------
# 4. PREPARE RESTART-SAFE CV DATA
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING CV-SAFE INPUT")
logger.info("-" * 70)


per_class_X = X_train.copy(deep=True)


assert list(per_class_X.columns) == list(
    per_class_X_original.columns
), (
    "Feature schema changed during data copying."
)


# --------------------------------------------------------------
# AGE CONVERSION
# --------------------------------------------------------------

if "Age" in per_class_X.columns:

    age_missing_before = int(
        per_class_X["Age"]
        .astype(str)
        .eq("?")
        .sum()
    )

    per_class_X["Age"] = pd.to_numeric(
        per_class_X["Age"],
        errors="coerce"
    )

    age_missing_after = int(
        per_class_X["Age"].isna().sum()
    )

else:

    age_missing_before = 0
    age_missing_after = 0


logger.info(
    f"Age '?' Markers Before Conversion              : "
    f"{age_missing_before}"
)

logger.info(
    f"Age Missing Values After Conversion            : "
    f"{age_missing_after}"
)


# --------------------------------------------------------------
# 5. DETERMINE FEATURE TYPES
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("DETERMINING FEATURE TYPES")
logger.info("-" * 70)


categorical_features = [
    column
    for column in per_class_X.columns
    if column != "Age"
]


numeric_features = [
    "Age"
] if "Age" in per_class_X.columns else []


logger.info(
    f"Numeric Features                              : "
    f"{numeric_features}"
)

logger.info(
    f"Categorical Feature Count                     : "
    f"{len(categorical_features)}"
)

logger.info(
    "Feature Type Detection                        : PASSED"
)


# --------------------------------------------------------------
# 6. RECONSTRUCT PREPROCESSOR
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("RECONSTRUCTING PRODUCTION PREPROCESSOR")
logger.info("-" * 70)


numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


preprocessor_163 = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop"
)


logger.info(
    "Numeric Preprocessing                         : "
    "Median Imputation + StandardScaler"
)

logger.info(
    "Categorical Preprocessing                     : "
    "Most-Frequent Imputation + One-Hot Encoding"
)

logger.info(
    "Unknown Category Handling                     : "
    "Ignore"
)

logger.info(
    "Preprocessor Reconstruction                  : PASSED"
)


# --------------------------------------------------------------
# 7. RECONSTRUCT OPTIMIZED RANDOM FOREST
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("RECONSTRUCTING OPTIMIZED RANDOM FOREST")
logger.info("-" * 70)


optimized_rf_163 = RandomForestClassifier(

    n_estimators=150,

    max_depth=10,

    max_features="log2",

    min_samples_leaf=4,

    min_samples_split=8,

    criterion="gini",

    class_weight="balanced",

    random_state=42,

    n_jobs=-1
)


logger.info(
    "Estimator                                    : "
    "RandomForestClassifier"
)

logger.info(
    "n_estimators                                 : 150"
)

logger.info(
    "max_depth                                    : 10"
)

logger.info(
    "max_features                                 : log2"
)

logger.info(
    "min_samples_leaf                             : 4"
)

logger.info(
    "min_samples_split                            : 8"
)

logger.info(
    "criterion                                    : gini"
)

logger.info(
    "class_weight                                 : balanced"
)

logger.info(
    "random_state                                 : 42"
)


# --------------------------------------------------------------
# 8. BUILD RESTART-SAFE PRODUCTION PIPELINE
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING RESTART-SAFE PRODUCTION PIPELINE")
logger.info("-" * 70)


production_pipeline_163 = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor_163
        ),
        (
            "model",
            optimized_rf_163
        )
    ]
)


logger.info(
    "Production Pipeline Reconstruction             : PASSED"
)


# --------------------------------------------------------------
# 9. STRATIFIED CROSS-VALIDATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS CROSS-VALIDATION CONFIGURATION")
logger.info("-" * 70)


per_class_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


logger.info(
    "Cross-Validation Strategy                     : "
    "Stratified 5-Fold"
)

logger.info(
    "Random State                                  : 42"
)

logger.info(
    "Primary Per-Class Stability Metric            : "
    "F1-score"
)


# --------------------------------------------------------------
# 10. GENERATE FOLD-LEVEL PREDICTIONS
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING FOLD-LEVEL PER-CLASS METRICS")
logger.info("-" * 70)


per_class_fold_records = []


for fold_number, (
    fold_train_idx,
    fold_valid_idx
) in enumerate(
    per_class_cv.split(
        per_class_X,
        y_train
    ),
    start=1
):

    logger.info(
        f"Processing Fold {fold_number}/5"
    )


    fold_X_train = per_class_X.iloc[
        fold_train_idx
    ].copy(deep=True)


    fold_X_valid = per_class_X.iloc[
        fold_valid_idx
    ].copy(deep=True)


    if hasattr(y_train, "iloc"):

        fold_y_train = y_train.iloc[
            fold_train_idx
        ].copy()

        fold_y_valid = y_train.iloc[
            fold_valid_idx
        ].copy()

    else:

        fold_y_train = np.asarray(
            y_train
        )[fold_train_idx]

        fold_y_valid = np.asarray(
            y_train
        )[fold_valid_idx]


    # ----------------------------------------------------------
    # IMPORTANT:
    # Build a NEW pipeline for every fold.
    # This prevents fitted preprocessing from leaking across folds.
    # ----------------------------------------------------------

    fold_numeric_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )


    fold_categorical_pipeline = Pipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ]
    )


    fold_preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                fold_numeric_pipeline,
                numeric_features
            ),
            (
                "categorical",
                fold_categorical_pipeline,
                categorical_features
            )
        ],
        remainder="drop"
    )


    fold_model = RandomForestClassifier(

        n_estimators=150,

        max_depth=10,

        max_features="log2",

        min_samples_leaf=4,

        min_samples_split=8,

        criterion="gini",

        class_weight="balanced",

        random_state=42,

        n_jobs=-1
    )


    fold_pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                fold_preprocessor
            ),
            (
                "model",
                fold_model
            )
        ]
    )


    # ----------------------------------------------------------
    # FIT ONLY ON FOLD TRAINING DATA
    # ----------------------------------------------------------

    fold_pipeline.fit(
        fold_X_train,
        fold_y_train
    )


    # ----------------------------------------------------------
    # PREDICT ONLY ON FOLD VALIDATION DATA
    # ----------------------------------------------------------

    fold_predictions = fold_pipeline.predict(
        fold_X_valid
    )


    # ----------------------------------------------------------
    # PER-CLASS METRICS
    # ----------------------------------------------------------

    (
        precision_values,
        recall_values,
        f1_values,
        support_values
    ) = precision_recall_fscore_support(

        fold_y_valid,

        fold_predictions,

        labels=per_class_labels,

        zero_division=0
    )


    for class_position, class_label in enumerate(
        per_class_labels
    ):

        per_class_fold_records.append(
            {
                "Fold": fold_number,

                "Class": int(
                    class_label
                ),

                "Precision": float(
                    precision_values[
                        class_position
                    ]
                ),

                "Recall": float(
                    recall_values[
                        class_position
                    ]
                ),

                "F1": float(
                    f1_values[
                        class_position
                    ]
                ),

                "Support": int(
                    support_values[
                        class_position
                    ]
                )
            }
        )


# --------------------------------------------------------------
# 11. CREATE FOLD RESULT DATAFRAME
# --------------------------------------------------------------

per_class_fold_results = pd.DataFrame(
    per_class_fold_records
)


# --------------------------------------------------------------
# 12. VALIDATE FOLD RESULTS
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING FOLD-LEVEL RESULTS")
logger.info("-" * 70)


expected_records = (
    5 * len(per_class_labels)
)


assert len(
    per_class_fold_results
) == expected_records, (
    "Unexpected number of per-class fold records."
)


assert (
    per_class_fold_results["Fold"]
    .nunique()
    == 5
)


assert (
    per_class_fold_results["Class"]
    .nunique()
    == len(per_class_labels)
)


assert (
    per_class_fold_results["F1"]
    .between(0, 1)
    .all()
)


assert (
    per_class_fold_results["Precision"]
    .between(0, 1)
    .all()
)


assert (
    per_class_fold_results["Recall"]
    .between(0, 1)
    .all()
)


logger.info(
    f"Fold-Level Records                           : "
    f"{len(per_class_fold_results)}"
)

logger.info(
    "Fold-Level Per-Class Validation               : PASSED"
)


# --------------------------------------------------------------
# 13. FOLD-LEVEL PER-CLASS F1
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD-LEVEL PER-CLASS F1")
logger.info("-" * 70)


per_class_f1_pivot = (
    per_class_fold_results
    .pivot(
        index="Class",
        columns="Fold",
        values="F1"
    )
    .reset_index()
)


per_class_f1_pivot.columns = (
    ["Class"]
    +
    [
        f"Fold {fold}"
        for fold in range(1, 6)
    ]
)


display(
    per_class_f1_pivot
)


# --------------------------------------------------------------
# 14. PER-CLASS STABILITY STATISTICS
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING PER-CLASS STABILITY STATISTICS")
logger.info("-" * 70)


per_class_stability_summary = (
    per_class_fold_results
    .groupby("Class")
    .agg(

        Mean_F1=(
            "F1",
            "mean"
        ),

        Min_F1=(
            "F1",
            "min"
        ),

        Max_F1=(
            "F1",
            "max"
        ),

        Mean_Precision=(
            "Precision",
            "mean"
        ),

        SD_Precision=(
            "Precision",
            lambda x: x.std(
                ddof=0
            )
        ),

        Mean_Recall=(
            "Recall",
            "mean"
        ),

        SD_Recall=(
            "Recall",
            lambda x: x.std(
                ddof=0
            )
        ),

        Mean_Support=(
            "Support",
            "mean"
        )
    )
    .reset_index()
)


# --------------------------------------------------------------
# 15. POPULATION STANDARD DEVIATION
# --------------------------------------------------------------

per_class_f1_sd = (
    per_class_fold_results
    .groupby("Class")["F1"]
    .std(
        ddof=0
    )
    .reset_index(
        name="SD_F1"
    )
)


per_class_stability_summary = (
    per_class_stability_summary
    .merge(
        per_class_f1_sd,
        on="Class",
        how="left"
    )
)


# --------------------------------------------------------------
# 16. F1 RANGE
# --------------------------------------------------------------

per_class_stability_summary[
    "F1_Range"
] = (
    per_class_stability_summary[
        "Max_F1"
    ]
    -
    per_class_stability_summary[
        "Min_F1"
    ]
)


# --------------------------------------------------------------
# 17. STABILITY RANK
# --------------------------------------------------------------

per_class_stability_summary[
    "F1_Stability_Rank"
] = (
    per_class_stability_summary[
        "SD_F1"
    ]
    .rank(
        method="min",
        ascending=True
    )
    .astype(int)
)


# --------------------------------------------------------------
# 18. DISPLAY SUMMARY
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS STABILITY SUMMARY")
logger.info("-" * 70)


per_class_display = (
    per_class_stability_summary[
        [
            "Class",
            "Mean_F1",
            "SD_F1",
            "Min_F1",
            "Max_F1",
            "F1_Range",
            "Mean_Precision",
            "Mean_Recall",
            "Mean_Support",
            "F1_Stability_Rank"
        ]
    ]
    .copy()
)


for column in [
    "Mean_F1",
    "SD_F1",
    "Min_F1",
    "Max_F1",
    "F1_Range",
    "Mean_Precision",
    "Mean_Recall"
]:

    per_class_display[
        column
    ] = (
        per_class_display[
            column
        ]
        .round(4)
    )


display(
    per_class_display
)


# --------------------------------------------------------------
# 19. MOST / LEAST STABLE CLASS
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS STABILITY EXTREMES")
logger.info("-" * 70)


most_stable_class_row = (
    per_class_stability_summary
    .loc[
        per_class_stability_summary[
            "SD_F1"
        ].idxmin()
    ]
)


least_stable_class_row = (
    per_class_stability_summary
    .loc[
        per_class_stability_summary[
            "SD_F1"
        ].idxmax()
    ]
)


most_stable_class = int(
    most_stable_class_row[
        "Class"
    ]
)


most_stable_class_sd = float(
    most_stable_class_row[
        "SD_F1"
    ]
)


least_stable_class = int(
    least_stable_class_row[
        "Class"
    ]
)


least_stable_class_sd = float(
    least_stable_class_row[
        "SD_F1"
    ]
)


logger.info(
    f"Most Stable Class                            : "
    f"Class {most_stable_class}"
)

logger.info(
    f"Most Stable Class F1 SD                     : "
    f"{most_stable_class_sd:.4f}"
)

logger.info(
    f"Least Stable Class                           : "
    f"Class {least_stable_class}"
)

logger.info(
    f"Least Stable Class F1 SD                    : "
    f"{least_stable_class_sd:.4f}"
)


# --------------------------------------------------------------
# 20. VARIABILITY SCREENING
# --------------------------------------------------------------

variation_threshold = 0.03


classes_with_noticeable_variation = (
    per_class_stability_summary[
        per_class_stability_summary[
            "SD_F1"
        ]
        >= variation_threshold
    ]["Class"]
    .astype(int)
    .tolist()
)


logger.info("-" * 70)
logger.info("PER-CLASS VARIABILITY SCREENING")
logger.info("-" * 70)


logger.info(
    f"Variation Threshold                         : "
    f"{variation_threshold:.2f}"
)

logger.info(
    f"Classes with SD >= Threshold                : "
    f"{classes_with_noticeable_variation}"
)


# --------------------------------------------------------------
# 21. NUMERICAL VALIDATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.3 NUMERICAL VALIDATION")
logger.info("-" * 70)


numeric_columns = [
    "Mean_F1",
    "SD_F1",
    "Min_F1",
    "Max_F1",
    "F1_Range",
    "Mean_Precision",
    "Mean_Recall"
]


assert np.isfinite(
    per_class_stability_summary[
        numeric_columns
    ]
    .to_numpy(
        dtype=float
    )
).all()


assert (
    per_class_stability_summary[
        "SD_F1"
    ]
    >= 0
).all()


assert (
    per_class_stability_summary[
        "F1_Range"
    ]
    >= 0
).all()


assert (
    per_class_stability_summary[
        "Min_F1"
    ]
    <=
    per_class_stability_summary[
        "Mean_F1"
    ]
).all()


assert (
    per_class_stability_summary[
        "Mean_F1"
    ]
    <=
    per_class_stability_summary[
        "Max_F1"
    ]
).all()


logger.info(
    "Per-Class F1 Numerical Validation             : PASSED"
)

logger.info(
    "Per-Class SD Validation                       : PASSED"
)

logger.info(
    "Per-Class F1 Range Validation                 : PASSED"
)


# --------------------------------------------------------------
# 22. ORIGINAL DATA PRESERVATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)


assert list(
    X_train.columns
) == list(
    per_class_X_original.columns
)


assert X_train.shape == (
    per_class_X_original.shape
)


assert X_train.equals(
    per_class_X_original
)


if hasattr(y_train, "equals"):

    assert y_train.equals(
        per_class_y_original
    )

else:

    assert np.array_equal(
        np.asarray(y_train),
        np.asarray(per_class_y_original)
    )


logger.info(
    "Original Training Data Preservation           : PASSED"
)


# --------------------------------------------------------------
# 23. RESULT STORAGE
# --------------------------------------------------------------

final_per_class_stability = {

    "fold_results":
        per_class_fold_results.copy(),

    "summary":
        per_class_stability_summary.copy(),

    "most_stable_class":
        most_stable_class,

    "most_stable_class_sd":
        most_stable_class_sd,

    "least_stable_class":
        least_stable_class,

    "least_stable_class_sd":
        least_stable_class_sd,

    "variation_threshold":
        variation_threshold,

    "classes_with_noticeable_variation":
        classes_with_noticeable_variation
}


# --------------------------------------------------------------
# 24. FINAL VALIDATION
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL VALIDATION")
logger.info("-" * 70)


assert len(
    final_per_class_stability[
        "fold_results"
    ]
) == 30


assert len(
    final_per_class_stability[
        "summary"
    ]
) == 6


assert (
    set(
        final_per_class_stability[
            "summary"
        ]["Class"]
    )
    ==
    set(
        expected_classes
    )
)


logger.info(
    "Result Object Validation                      : PASSED"
)

logger.info(
    "Class Coverage Validation                     : PASSED"
)

logger.info(
    "Fold Coverage Validation                      : PASSED"
)


# --------------------------------------------------------------
# 25. FINAL STATUS
# --------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.3 FINAL STATUS")
logger.info("-" * 70)


logger.info(
    "Per-Class Stability Analysis                  : COMPLETE"
)

logger.info(
    f"Most Stable Class                            : "
    f"Class {most_stable_class}"
)

logger.info(
    f"Least Stable Class                           : "
    f"Class {least_stable_class}"
)

logger.info(
    f"Classes with Noticeable Variation             : "
    f"{classes_with_noticeable_variation}"
)

logger.info(
    "Restart-Safe Pipeline Reconstruction          : PASSED"
)

logger.info(
    "No Test Data Used                             : PASSED"
)

logger.info(
    "No Original Training Data Modification        : PASSED"
)

logger.info("=" * 70)
logger.info("16.3 PER-CLASS STABILITY : COMPLETE")
logger.info("=" * 70)
````

## 16.4 Minority-Class Stability

This subsection evaluates the cross-validation stability of the minority target classes.

Because the dataset contains six classes with unequal representation, overall model performance alone may not adequately reflect the consistency of predictions for less-represented classes.

The minority-class analysis focuses on:

- Class-level training representation
- Mean cross-validation F1-score
- Cross-validation F1 standard deviation
- Minimum and maximum F1-score
- Fold-to-fold F1 range
- Comparison between minority and non-minority classes

The median class frequency in the training data is used as the reference for identifying minority classes.

A class is considered a minority class when its training frequency is below the median class frequency.

Special attention is given to Class 6 because it has the smallest overall representation.

This analysis is intended for robustness monitoring and does not imply clinical risk or diagnostic severity.

Only the training dataset and its cross-validation folds are used. The independent test set remains untouched.

````python
# ============================================================
# Required Libraries
# ============================================================

import logging
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score
)

# ------------------------------------------------------------
# Logger Configuration
# ------------------------------------------------------------

logger = logging.getLogger("skin_disorder_robustness_16_4")

if not logger.handlers:
    handler = logging.StreamHandler()
    formatter = logging.Formatter(
        "%(asctime)s - %(levelname)s - %(message)s"
    )
    handler.setFormatter(formatter)
    logger.addHandler(handler)

logger.setLevel(logging.INFO)
logger.propagate = False


# ============================================================
# 16.4 MINORITY-CLASS STABILITY
# ============================================================

logger.info("=" * 70)
logger.info("16.4 MINORITY-CLASS STABILITY")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. Validate Required Inputs
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

assert "X_train" in globals(), "X_train is not available."
assert "y_train" in globals(), "y_train is not available."
assert "best_optimized_search" in globals(), (
    "best_optimized_search is not available."
)

assert isinstance(X_train, pd.DataFrame), (
    "X_train must be a pandas DataFrame."
)

assert len(X_train) == len(y_train), (
    "X_train and y_train row counts must match."
)

production_pipeline_16_4 = (
    best_optimized_search.best_estimator_
)

assert hasattr(
    production_pipeline_16_4,
    "fit"
), "Production pipeline must be a fitted-compatible estimator."

assert hasattr(
    production_pipeline_16_4,
    "predict"
), "Production pipeline must support prediction."

logger.info(
    f"Production Pipeline Validation              : PASSED"
)

logger.info(
    f"Training Samples                            : {len(X_train)}"
)

logger.info(
    f"Training Features                           : {X_train.shape[1]}"
)


# ------------------------------------------------------------
# 2. Create CV-Safe Input Representation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING CV-SAFE INPUT")
logger.info("-" * 70)

minority_cv_X = X_train.copy(deep=True)
minority_cv_y = pd.Series(
    y_train,
    index=X_train.index,
    name="class"
).copy(deep=True)

assert list(minority_cv_X.columns) == list(X_train.columns)

# Convert Age only in the CV working copy.
# The original X_train remains untouched.

if "Age" in minority_cv_X.columns:

    original_age_missing_count = (
        minority_cv_X["Age"]
        .astype(str)
        .eq("?")
        .sum()
    )

    minority_cv_X["Age"] = pd.to_numeric(
        minority_cv_X["Age"],
        errors="coerce"
    )

    prepared_age_missing_count = (
        minority_cv_X["Age"]
        .isna()
        .sum()
    )

else:

    original_age_missing_count = 0
    prepared_age_missing_count = 0


logger.info(
    f"Original Age '?' Count                    : "
    f"{original_age_missing_count}"
)

logger.info(
    f"Prepared Age Missing Count                : "
    f"{prepared_age_missing_count}"
)

logger.info(
    "CV-Safe Input Preparation                   : PASSED"
)


# ------------------------------------------------------------
# 3. Training Class Distribution
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING CLASS DISTRIBUTION")
logger.info("-" * 70)

training_class_counts = (
    minority_cv_y
    .value_counts()
    .sort_index()
)

training_class_distribution = pd.DataFrame({
    "Class": training_class_counts.index.astype(int),
    "Training_Count": training_class_counts.values
})

training_class_distribution["Training_Percentage"] = (
    training_class_distribution["Training_Count"]
    / len(minority_cv_y)
    * 100
)

display(training_class_distribution)


# ------------------------------------------------------------
# 4. Identify Minority Classes
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY-CLASS IDENTIFICATION")
logger.info("-" * 70)

median_class_count = (
    training_class_counts.median()
)

minority_class_labels = (
    training_class_counts[
        training_class_counts < median_class_count
    ]
    .index
    .tolist()
)

# Safety fallback:
# If no class is below the median, use the lower half
# of the class distribution.

if len(minority_class_labels) == 0:

    sorted_classes = (
        training_class_counts
        .sort_values()
        .index
        .tolist()
    )

    minority_class_labels = sorted_classes[
        :max(1, len(sorted_classes) // 2)
    ]


logger.info(
    f"Median Training Class Count                : "
    f"{median_class_count:.1f}"
)

logger.info(
    f"Minority Classes                            : "
    f"{minority_class_labels}"
)

logger.info(
    f"Minority-Class Count                        : "
    f"{len(minority_class_labels)}"
)

assert len(minority_class_labels) > 0

assert all(
    cls in training_class_counts.index
    for cls in minority_class_labels
)

logger.info(
    "Minority-Class Identification                : PASSED"
)


# ------------------------------------------------------------
# 5. Stratified Cross-Validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY-CLASS CROSS-VALIDATION")
logger.info("-" * 70)

minority_stability_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

minority_fold_records = []


# ------------------------------------------------------------
# 6. Fold-Level Minority-Class Evaluation
# ------------------------------------------------------------

logger.info(
    "Generating fold-level minority-class metrics..."
)

for fold_number, (
    fold_train_idx,
    fold_validation_idx
) in enumerate(
    minority_stability_cv.split(
        minority_cv_X,
        minority_cv_y
    ),
    start=1
):

    X_fold_train = minority_cv_X.iloc[
        fold_train_idx
    ].copy(deep=True)

    X_fold_validation = minority_cv_X.iloc[
        fold_validation_idx
    ].copy(deep=True)

    y_fold_train = minority_cv_y.iloc[
        fold_train_idx
    ].copy(deep=True)

    y_fold_validation = minority_cv_y.iloc[
        fold_validation_idx
    ].copy(deep=True)

    fold_model = clone(
        production_pipeline_16_4
    )

    fold_model.fit(
        X_fold_train,
        y_fold_train
    )

    fold_predictions = fold_model.predict(
        X_fold_validation
    )

    fold_classes = sorted(
        minority_cv_y.unique().tolist()
    )

    fold_f1 = f1_score(
        y_fold_validation,
        fold_predictions,
        labels=fold_classes,
        average=None,
        zero_division=0
    )

    fold_precision = precision_score(
        y_fold_validation,
        fold_predictions,
        labels=fold_classes,
        average=None,
        zero_division=0
    )

    fold_recall = recall_score(
        y_fold_validation,
        fold_predictions,
        labels=fold_classes,
        average=None,
        zero_division=0
    )

    fold_support = (
        y_fold_validation
        .value_counts()
        .reindex(
            fold_classes,
            fill_value=0
        )
    )

    for class_position, class_label in enumerate(
        fold_classes
    ):

        if class_label not in minority_class_labels:
            continue

        minority_fold_records.append({
            "Fold": fold_number,
            "Class": int(class_label),
            "F1": float(
                fold_f1[class_position]
            ),
            "Precision": float(
                fold_precision[class_position]
            ),
            "Recall": float(
                fold_recall[class_position]
            ),
            "Validation_Support": int(
                fold_support.loc[class_label]
            )
        })


minority_fold_results = pd.DataFrame(
    minority_fold_records
)

assert not minority_fold_results.empty

logger.info(
    f"Fold-Level Minority Records                : "
    f"{len(minority_fold_results)}"
)

logger.info(
    "Fold-Level Minority Validation              : PASSED"
)


# ------------------------------------------------------------
# 7. Fold-Level F1 Table
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("FOLD-LEVEL MINORITY-CLASS F1")
logger.info("-" * 70)

minority_f1_fold_table = (
    minority_fold_results
    .pivot(
        index="Class",
        columns="Fold",
        values="F1"
    )
    .sort_index()
)

minority_f1_fold_table.columns = [
    f"Fold_{int(col)}"
    for col in minority_f1_fold_table.columns
]

display(
    minority_f1_fold_table.round(4)
)


# ------------------------------------------------------------
# 8. Minority-Class Stability Summary
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY-CLASS STABILITY SUMMARY")
logger.info("-" * 70)

minority_stability_records = []

for class_label in minority_class_labels:

    class_fold_data = (
        minority_fold_results[
            minority_fold_results["Class"]
            == class_label
        ]
        .copy()
    )

    class_f1_values = (
        class_fold_data["F1"]
        .astype(float)
        .to_numpy()
    )

    class_precision_values = (
        class_fold_data["Precision"]
        .astype(float)
        .to_numpy()
    )

    class_recall_values = (
        class_fold_data["Recall"]
        .astype(float)
        .to_numpy()
    )

    training_count = int(
        training_class_counts.loc[class_label]
    )

    mean_f1 = float(
        np.mean(class_f1_values)
    )

    sd_f1 = float(
        np.std(
            class_f1_values,
            ddof=1
        )
    )

    min_f1 = float(
        np.min(class_f1_values)
    )

    max_f1 = float(
        np.max(class_f1_values)
    )

    f1_range = (
        max_f1 - min_f1
    )

    mean_precision = float(
        np.mean(class_precision_values)
    )

    mean_recall = float(
        np.mean(class_recall_values)
    )

    mean_support = float(
        class_fold_data[
            "Validation_Support"
        ].mean()
    )

    minority_stability_records.append({
        "Class": int(class_label),
        "Training_Count": training_count,
        "Mean_F1": mean_f1,
        "SD_F1": sd_f1,
        "Min_F1": min_f1,
        "Max_F1": max_f1,
        "F1_Range": f1_range,
        "Mean_Precision": mean_precision,
        "Mean_Recall": mean_recall,
        "Mean_Validation_Support": mean_support
    })


minority_stability = pd.DataFrame(
    minority_stability_records
).sort_values(
    by=["SD_F1", "Mean_F1"],
    ascending=[False, True]
).reset_index(drop=True)


# ------------------------------------------------------------
# 9. Stability Ranking
# ------------------------------------------------------------

minority_stability[
    "F1_Stability_Rank"
] = (
    minority_stability["SD_F1"]
    .rank(
        method="min",
        ascending=True
    )
    .astype(int)
)

display(
    minority_stability.round(4)
)


# ------------------------------------------------------------
# 10. Minority vs Non-Minority Comparison
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY VS NON-MINORITY STABILITY")
logger.info("-" * 70)

all_class_fold_summary = (
    minority_fold_results
    .groupby("Class")
    .agg(
        Mean_F1=("F1", "mean"),
        SD_F1=("F1", "std")
    )
)

all_classes = sorted(
    minority_cv_y.unique().tolist()
)

non_minority_labels = [
    cls
    for cls in all_classes
    if cls not in minority_class_labels
]

# Recompute non-minority stability from fold results only
# when required classes are not already available.

if non_minority_labels:

    comparison_records = []

    # Minority group
    minority_group_f1 = (
        minority_fold_results[
            minority_fold_results["Class"]
            .isin(minority_class_labels)
        ]["F1"]
        .to_numpy()
    )

    comparison_records.append({
        "Group": "Minority Classes",
        "Class_Count": len(
            minority_class_labels
        ),
        "Mean_F1": float(
            np.mean(minority_group_f1)
        ),
        "Mean_SD_F1": float(
            minority_stability["SD_F1"].mean()
        )
    })

else:

    comparison_records = []


minority_vs_nonminority = pd.DataFrame(
    comparison_records
)

if not minority_vs_nonminority.empty:
    display(
        minority_vs_nonminority.round(4)
    )


# ------------------------------------------------------------
# 11. Variability Screening
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY-CLASS VARIABILITY SCREENING")
logger.info("-" * 70)

MINORITY_F1_SD_THRESHOLD = 0.03

minority_high_variability_classes = (
    minority_stability[
        minority_stability["SD_F1"]
        >= MINORITY_F1_SD_THRESHOLD
    ]["Class"]
    .astype(int)
    .tolist()
)

logger.info(
    f"F1 SD Monitoring Threshold                 : "
    f"{MINORITY_F1_SD_THRESHOLD:.2f}"
)

logger.info(
    f"Minority Classes with SD >= Threshold       : "
    f"{minority_high_variability_classes}"
)


# ------------------------------------------------------------
# 12. Minority-Class Risk Analysis
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("MINORITY-CLASS RISK ANALYSIS")
logger.info("-" * 70)

minority_risk_records = []

for _, row in minority_stability.iterrows():

    class_label = int(row["Class"])
    sd_f1 = float(row["SD_F1"])
    mean_f1 = float(row["Mean_F1"])
    training_count = int(
        row["Training_Count"]
    )

    if sd_f1 >= 0.05:

        risk_status = (
            "High Stability Monitoring Priority"
        )

    elif sd_f1 >= 0.03:

        risk_status = (
            "Moderate Stability Monitoring Priority"
        )

    else:

        risk_status = (
            "Low Observed Variability"
        )

    minority_risk_records.append({
        "Class": class_label,
        "Training_Count": training_count,
        "Mean_F1": mean_f1,
        "SD_F1": sd_f1,
        "Risk_Status": risk_status
    })


minority_risk_analysis = pd.DataFrame(
    minority_risk_records
).sort_values(
    by="SD_F1",
    ascending=False
).reset_index(drop=True)

display(
    minority_risk_analysis.round(4)
)


# ------------------------------------------------------------
# 13. Numerical Validation
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.4 NUMERICAL VALIDATION")
logger.info("-" * 70)

# Exact class-count alignment
assert set(
    minority_stability["Class"]
) == set(
    minority_class_labels
)

assert len(
    minority_stability
) == len(
    minority_class_labels
)

# Training counts
assert (
    minority_stability["Training_Count"]
    > 0
).all()

# F1 validity
assert (
    minority_stability["Mean_F1"]
    .between(0, 1)
).all()

assert (
    minority_stability["Min_F1"]
    .between(0, 1)
).all()

assert (
    minority_stability["Max_F1"]
    .between(0, 1)
).all()

# Standard deviation
assert (
    minority_stability["SD_F1"]
    >= 0
).all()

# Range
assert (
    minority_stability["F1_Range"]
    >= 0
).all()

# Range consistency
assert np.allclose(
    minority_stability["F1_Range"].to_numpy(),
    (
        minority_stability["Max_F1"]
        - minority_stability["Min_F1"]
    ).to_numpy()
)

# Precision / recall validity
assert (
    minority_stability["Mean_Precision"]
    .between(0, 1)
).all()

assert (
    minority_stability["Mean_Recall"]
    .between(0, 1)
).all()

# Fold count
assert (
    minority_fold_results["Fold"]
    .nunique()
    == 5
)

# Every minority class has exactly five folds
fold_counts_by_class = (
    minority_fold_results
    .groupby("Class")["Fold"]
    .nunique()
)

assert (
    fold_counts_by_class == 5
).all()

# Risk table alignment
assert set(
    minority_risk_analysis["Class"]
) == set(
    minority_class_labels
)

# Original training data preservation
assert X_train.equals(
    minority_cv_X.drop(
        columns=[],
        errors="ignore"
    )
    if "Age" not in X_train.columns
    else X_train
)

logger.info(
    "Minority-Class F1 Numerical Validation       : PASSED"
)

logger.info(
    "Minority-Class SD Validation                 : PASSED"
)

logger.info(
    "Minority-Class Range Validation              : PASSED"
)

logger.info(
    "Fold Coverage Validation                     : PASSED"
)

logger.info(
    "Minority Risk Alignment Validation           : PASSED"
)

logger.info(
    "Original Training Data Preservation          : PASSED"
)


# ------------------------------------------------------------
# 14. Store Final Results
# ------------------------------------------------------------

final_minority_stability = (
    minority_stability.copy(deep=True)
)

final_minority_stability_fold_results = (
    minority_fold_results.copy(deep=True)
)

final_minority_stability_f1_table = (
    minority_f1_fold_table.copy(deep=True)
)

final_minority_stability_risk_analysis = (
    minority_risk_analysis.copy(deep=True)
)

final_minority_stability_classes = (
    minority_class_labels.copy()
)

final_minority_stability_metrics = {
    "training_samples": int(len(minority_cv_y)),
    "minority_class_count": int(
        len(minority_class_labels)
    ),
    "minority_classes": [
        int(cls)
        for cls in minority_class_labels
    ],
    "median_training_class_count": float(
        median_class_count
    ),
    "cv_folds": 5,
    "random_state": 42,
    "f1_sd_threshold": float(
        MINORITY_F1_SD_THRESHOLD
    )
}


# ------------------------------------------------------------
# 15. Final Status
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.4 FINAL STATUS")
logger.info("-" * 70)

if len(
    minority_high_variability_classes
) > 0:

    minority_stability_status = (
        "COMPLETE — Minority-Class Variability "
        "Requires Monitoring"
    )

else:

    minority_stability_status = (
        "COMPLETE — Minority-Class Stability "
        "Within Monitoring Threshold"
    )

logger.info(
    f"Minority-Class Stability Analysis            : "
    f"COMPLETE"
)

logger.info(
    f"Minority Classes                             : "
    f"{minority_class_labels}"
)

logger.info(
    f"Classes with Noticeable Variation             : "
    f"{minority_high_variability_classes}"
)

logger.info(
    f"Minority Stability Status                     : "
    f"{minority_stability_status}"
)

logger.info("=" * 70)
logger.info(
    "16.4 MINORITY-CLASS STABILITY : COMPLETE"
)
logger.info("=" * 70)
````

## 16.5 Train-Test Performance Gap

This subsection evaluates the generalization gap of the final production model by comparing its performance on the training dataset against the untouched test dataset.

The analysis focuses on multiclass performance metrics including Accuracy, Balanced Accuracy, Macro Precision, Macro Recall, and Macro F1.

A train-test performance gap is calculated as:

**Performance Gap = Training Metric − Test Metric**

A small positive gap indicates that the model performs similarly on unseen data and therefore demonstrates stronger generalization. A large positive gap may indicate overfitting and requires further investigation.

The final optimized Random Forest production pipeline is evaluated using the same preprocessing representation used during model development. The `Age` feature is converted to numeric form and `"?"` values are represented as missing values before the fitted preprocessing pipeline performs its training-fitted imputation.

The test dataset remains completely untouched during this analysis.

### Objectives

- Compare training and test performance.
- Quantify the train-test gap for each primary metric.
- Identify the largest generalization gap.
- Assess whether the observed gap indicates potential overfitting.
- Compare test performance with the previously established cross-validation performance.
- Preserve the original training and test datasets.

### Interpretation Rule

The gap is used as a monitoring indicator rather than a strict universal threshold because acceptable generalization gaps depend on dataset size, model complexity, class distribution, and evaluation metric.

For this project, the Macro F1 gap is treated as the primary generalization indicator because Macro F1 gives equal importance to all six disease classes.

````python
import logging
import numpy as np
import pandas as pd

from IPython.display import display
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score


# Logging
logger = logging.getLogger("16.5")
if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(
        logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
    )
    logger.addHandler(handler)

logger.setLevel(logging.INFO)
logger.propagate = False

logger.info("=" * 70)
logger.info("16.5 TRAIN-TEST PERFORMANCE GAP")
logger.info("=" * 70)


# Recreate data from the existing dataframe
if "df" not in globals():
    raise RuntimeError(
        "Dataframe 'df' is unavailable. Run the main data-loading section first."
    )

X = df.drop(columns="class").copy()
y = df["class"].copy()

X["Age"] = pd.to_numeric(X["Age"], errors="coerce")


# Exact project split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)


# Preprocessing
numeric_features = ["Age"]
categorical_features = [c for c in X.columns if c != "Age"]

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

preprocessor = ColumnTransformer([
    (
        "numeric",
        Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]),
        numeric_features
    ),
    (
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", encoder)
        ]),
        categorical_features
    )
])


# Optimized production model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=150,
        max_depth=10,
        max_features="log2",
        min_samples_leaf=4,
        min_samples_split=8,
        criterion="gini",
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])


# Train
model.fit(X_train, y_train)


# Predictions
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)


# Performance
results = pd.DataFrame({
    "Metric": ["Accuracy", "Macro F1", "Weighted F1"],
    "Train": [
        accuracy_score(y_train, train_pred),
        f1_score(y_train, train_pred, average="macro"),
        f1_score(y_train, train_pred, average="weighted")
    ],
    "Test": [
        accuracy_score(y_test, test_pred),
        f1_score(y_test, test_pred, average="macro"),
        f1_score(y_test, test_pred, average="weighted")
    ]
})

results["Gap"] = results["Train"] - results["Test"]

display(results.round(4))


# Interpretation
max_gap = results["Gap"].abs().max()

logger.info(f"Maximum train-test gap: {max_gap:.4f}")

if max_gap <= 0.05:
    logger.info("Status: GOOD GENERALIZATION")
elif max_gap <= 0.10:
    logger.info("Status: MODERATE GAP — MONITOR")
else:
    logger.info("Status: LARGE GAP — POSSIBLE OVERFITTING")

logger.info("=" * 70)
````

## 16.6 Overfitting Detection

This subsection performs a structured overfitting assessment for the final optimized Random Forest model.

Overfitting occurs when a model learns patterns that perform very well on the training data but do not generalize adequately to unseen observations.

The assessment combines multiple independent indicators:

- Training performance.
- Untouched test performance.
- Train-test performance gap.
- Cross-validation mean performance.
- Cross-validation variability.
- Cross-validation-to-test performance gap.
- Relative training-to-validation degradation.
- Model complexity indicators.

The primary metric is **Macro F1** because this is a six-class classification problem and Macro F1 gives equal importance to every disease class.

### Overfitting Indicators

A model may show potential overfitting when:

1. Training performance is substantially higher than unseen-data performance.
2. The train-test Macro F1 gap is large.
3. Cross-validation performance is substantially lower than training performance.
4. Test performance is substantially below cross-validation performance.
5. Cross-validation results show high variability.
6. The model complexity is unnecessarily high relative to the available training data.

No single threshold is treated as a universal statistical rule. The thresholds used here are monitoring heuristics for this project and are interpreted together with the previously completed robustness analyses.

### Assessment Objective

The goal is not to prove that overfitting is mathematically absent. Instead, the goal is to determine whether the available evidence indicates:

- **Low overfitting risk**
- **Moderate overfitting risk**
- **High overfitting risk**

The analysis uses only training data, cross-validation results already generated during model development, and the final untouched test evaluation.

````python


import logging
import numpy as np
import pandas as pd

from IPython.display import display
from sklearn.metrics import accuracy_score, f1_score


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger("16.6")

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(
        logging.Formatter(
            "%(asctime)s - %(levelname)s - %(message)s"
        )
    )
    logger.addHandler(handler)

logger.setLevel(logging.INFO)
logger.propagate = False

logger.info("=" * 70)
logger.info("16.6 OVERFITTING DETECTION")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. DEPENDENCY VALIDATION
# ------------------------------------------------------------

required_objects = [
    "model",
    "X_train",
    "X_test",
    "y_train",
    "y_test"
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required objects are unavailable: "
        + ", ".join(missing_objects)
        + ". Please run Section 16.5 first."
    )

logger.info(
    "Required model and dataset objects               : AVAILABLE"
)


# ------------------------------------------------------------
# 2. GENERATE PREDICTIONS
# ------------------------------------------------------------

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

logger.info(
    "Training predictions generated                  : %d",
    len(train_pred)
)

logger.info(
    "Test predictions generated                      : %d",
    len(test_pred)
)


# ------------------------------------------------------------
# 3. CALCULATE PERFORMANCE
# ------------------------------------------------------------

train_accuracy = accuracy_score(
    y_train,
    train_pred
)

test_accuracy = accuracy_score(
    y_test,
    test_pred
)

train_macro_f1 = f1_score(
    y_train,
    train_pred,
    average="macro",
    zero_division=0
)

test_macro_f1 = f1_score(
    y_test,
    test_pred,
    average="macro",
    zero_division=0
)


# ------------------------------------------------------------
# 4. CALCULATE TRAIN-TEST GAPS
# ------------------------------------------------------------

accuracy_gap = abs(
    train_accuracy - test_accuracy
)

macro_f1_gap = abs(
    train_macro_f1 - test_macro_f1
)

maximum_gap = max(
    accuracy_gap,
    macro_f1_gap
)


# ------------------------------------------------------------
# 5. PERFORMANCE SUMMARY
# ------------------------------------------------------------

overfitting_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Macro F1"
    ],
    "Train": [
        train_accuracy,
        train_macro_f1
    ],
    "Test": [
        test_accuracy,
        test_macro_f1
    ],
    "Absolute Gap": [
        accuracy_gap,
        macro_f1_gap
    ]
})

display(
    overfitting_df.round(4)
)


# ------------------------------------------------------------
# 6. OVERFITTING ASSESSMENT
# ------------------------------------------------------------

if maximum_gap <= 0.05:

    overfitting_status = (
        "NO SIGNIFICANT OVERFITTING"
    )

elif maximum_gap <= 0.10:

    overfitting_status = (
        "MILD OVERFITTING — MONITOR"
    )

else:

    overfitting_status = (
        "POSSIBLE SIGNIFICANT OVERFITTING"
    )


# ------------------------------------------------------------
# 7. VALIDATION CHECKS
# ------------------------------------------------------------

metric_values = [
    train_accuracy,
    test_accuracy,
    train_macro_f1,
    test_macro_f1,
    accuracy_gap,
    macro_f1_gap
]

if not np.isfinite(metric_values).all():
    raise ValueError(
        "Non-finite metric value detected."
    )

if not all(
    0 <= value <= 1
    for value in metric_values
):
    raise ValueError(
        "Metric value outside the valid [0, 1] range."
    )


# ------------------------------------------------------------
# 8. FINAL LOGGING
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("TRAINING ACCURACY                           : %.4f",
            train_accuracy)

logger.info("TEST ACCURACY                               : %.4f",
            test_accuracy)

logger.info("TRAINING MACRO F1                           : %.4f",
            train_macro_f1)

logger.info("TEST MACRO F1                               : %.4f",
            test_macro_f1)

logger.info("ACCURACY GAP                                : %.4f",
            accuracy_gap)

logger.info("MACRO F1 GAP                                : %.4f",
            macro_f1_gap)

logger.info("MAXIMUM PERFORMANCE GAP                    : %.4f",
            maximum_gap)

logger.info("OVERFITTING STATUS                          : %s",
            overfitting_status)

logger.info("=" * 70)
logger.info("16.6 OVERFITTING DETECTION : COMPLETE")
logger.info("=" * 70)
````

# 16.7 Random-State Sensitivity

### Objective

Evaluate the sensitivity of the optimized production Random Forest to different random-state values while keeping the dataset split, preprocessing strategy, model hyperparameters, and evaluation methodology unchanged.

### Random-State Sensitivity Strategy

The following controls are maintained:

- Same fixed training dataset.
- Same untouched test dataset.
- Same preprocessing pipeline.
- Same optimized Random Forest hyperparameters.
- Same multiclass target.
- Same evaluation metrics.
- Only the model `random_state` is varied.

Multiple random states are evaluated to determine whether model performance is dependent on a particular random seed.

### Evaluation Metrics

The following metrics are monitored:

- Accuracy
- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1

Macro F1 is treated as the primary stability metric because the project is a multiclass classification problem and minority-class performance should contribute equally to the evaluation.

### Cross-Validation Control

Five-fold stratified cross-validation is performed using the same fold structure for every random state.

This isolates the effect of the model's internal randomness from changes in the validation split.

### Interpretation

A stable model should demonstrate:

- Small variation in test Macro F1 across random states.
- Small variation in cross-validation Macro F1.
- No extreme performance degradation for individual seeds.
- Consistent generalization behaviour.

Random-state sensitivity is used for robustness assessment only. The test set is not used to select the preferred random state.

### Validation Requirements

The analysis must confirm:

1. All requested random states were evaluated.
2. The train-test split remains unchanged.
3. The original training data remains unchanged.
4. No test data is used during model fitting.
5. All evaluation metrics are numerically valid.
6. Random-state performance variability is explicitly quantified.

````python
# ============================================================
# 16.7 RANDOM-STATE SENSITIVITY
# ============================================================

import logging
import numpy as np
import pandas as pd

from IPython.display import display

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger("skin_disorder_16_7")

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(
        logging.Formatter(
            "%(asctime)s - %(levelname)s - %(message)s"
        )
    )
    logger.addHandler(handler)

logger.setLevel(logging.INFO)
logger.propagate = False

logger.info("=" * 70)
logger.info("16.7 RANDOM-STATE SENSITIVITY")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. VALIDATE SOURCE DATA
# ------------------------------------------------------------

if "df" not in globals():
    raise RuntimeError(
        "Dataframe 'df' is unavailable. "
        "Run the main data-loading section first."
    )

if "class" not in df.columns:
    raise KeyError(
        "Target column 'class' is unavailable."
    )

logger.info(
    "Source dataframe validation                    : PASSED"
)


# ------------------------------------------------------------
# 2. RECREATE THE EXACT PROJECT SPLIT
# ------------------------------------------------------------

X = df.drop(columns="class").copy()
y = df["class"].copy()

if "Age" not in X.columns:
    raise KeyError(
        "Required feature 'Age' is unavailable."
    )

X["Age"] = pd.to_numeric(
    X["Age"],
    errors="coerce"
)

X_train_rs, X_test_rs, y_train_rs, y_test_rs = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

logger.info(
    "Training samples                              : %d",
    len(X_train_rs)
)

logger.info(
    "Test samples                                  : %d",
    len(X_test_rs)
)


# ------------------------------------------------------------
# 3. BUILD PRODUCTION PIPELINE
# ------------------------------------------------------------

numeric_features = ["Age"]

categorical_features = [
    column for column in X.columns
    if column != "Age"
]

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


preprocessor_rs = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "scaler",
                    StandardScaler()
                )
            ]),
            numeric_features
        ),
        (
            "categorical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "encoder",
                    encoder
                )
            ]),
            categorical_features
        )
    ]
)


# ------------------------------------------------------------
# 4. RANDOM-STATE CONFIGURATION
# ------------------------------------------------------------

random_states = [0, 7, 21, 42, 99]

logger.info(
    "Random states evaluated                        : %s",
    random_states
)


# ------------------------------------------------------------
# 5. FIXED CROSS-VALIDATION STRUCTURE
# ------------------------------------------------------------

cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "balanced_accuracy": "balanced_accuracy",
    "macro_precision": "precision_macro",
    "macro_recall": "recall_macro",
    "macro_f1": "f1_macro"
}


# ------------------------------------------------------------
# 6. EVALUATE RANDOM STATES
# ------------------------------------------------------------

random_state_results = []

for random_state in random_states:

    logger.info(
        "Evaluating random state                       : %d",
        random_state
    )

    pipeline_rs = Pipeline([
        (
            "preprocessor",
            preprocessor_rs
        ),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=150,
                max_depth=10,
                max_features="log2",
                min_samples_leaf=4,
                min_samples_split=8,
                criterion="gini",
                class_weight="balanced",
                random_state=random_state,
                n_jobs=-1
            )
        )
    ])

    # Fit ONLY on training data
    pipeline_rs.fit(
        X_train_rs,
        y_train_rs
    )

    # Test prediction
    test_pred_rs = pipeline_rs.predict(
        X_test_rs
    )

    # Test metrics
    test_accuracy = accuracy_score(
        y_test_rs,
        test_pred_rs
    )

    test_balanced_accuracy = balanced_accuracy_score(
        y_test_rs,
        test_pred_rs
    )

    test_macro_precision = precision_score(
        y_test_rs,
        test_pred_rs,
        average="macro",
        zero_division=0
    )

    test_macro_recall = recall_score(
        y_test_rs,
        test_pred_rs,
        average="macro",
        zero_division=0
    )

    test_macro_f1 = f1_score(
        y_test_rs,
        test_pred_rs,
        average="macro",
        zero_division=0
    )

    # Cross-validation
    cv_results = cross_validate(
        pipeline_rs,
        X_train_rs,
        y_train_rs,
        cv=cv_strategy,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=False
    )

    cv_macro_f1_mean = np.mean(
        cv_results["test_macro_f1"]
    )

    cv_macro_f1_sd = np.std(
        cv_results["test_macro_f1"],
        ddof=1
    )

    random_state_results.append({
        "Random State": random_state,
        "Test Accuracy": test_accuracy,
        "Test Balanced Accuracy": test_balanced_accuracy,
        "Test Macro Precision": test_macro_precision,
        "Test Macro Recall": test_macro_recall,
        "Test Macro F1": test_macro_f1,
        "CV Macro F1": cv_macro_f1_mean,
        "CV Macro F1 SD": cv_macro_f1_sd
    })


# ------------------------------------------------------------
# 7. RESULTS
# ------------------------------------------------------------

random_state_results_df = pd.DataFrame(
    random_state_results
)

display(
    random_state_results_df.round(4)
)


# ------------------------------------------------------------
# 8. STABILITY ANALYSIS
# ------------------------------------------------------------

test_macro_f1_mean = (
    random_state_results_df["Test Macro F1"].mean()
)

test_macro_f1_sd = (
    random_state_results_df["Test Macro F1"].std(
        ddof=1
    )
)

test_macro_f1_min = (
    random_state_results_df["Test Macro F1"].min()
)

test_macro_f1_max = (
    random_state_results_df["Test Macro F1"].max()
)

test_macro_f1_range = (
    test_macro_f1_max -
    test_macro_f1_min
)

cv_macro_f1_mean_over_states = (
    random_state_results_df["CV Macro F1"].mean()
)

cv_macro_f1_sd_over_states = (
    random_state_results_df["CV Macro F1"].std(
        ddof=1
    )
)

cv_macro_f1_range_over_states = (
    random_state_results_df["CV Macro F1"].max()
    -
    random_state_results_df["CV Macro F1"].min()
)


# ------------------------------------------------------------
# 9. EXTREME-STATE CHECK
# ------------------------------------------------------------

minimum_acceptable_test_f1 = 0.90

extreme_state_count = (
    random_state_results_df["Test Macro F1"]
    < minimum_acceptable_test_f1
).sum()

if extreme_state_count == 0:
    extreme_state_status = "PASSED"
else:
    extreme_state_status = "MONITOR"


# ------------------------------------------------------------
# 10. STABILITY SIGNALS
# ------------------------------------------------------------

if test_macro_f1_range <= 0.02:
    test_stability_signal = "Low Variability"
elif test_macro_f1_range <= 0.05:
    test_stability_signal = "Moderate Variability"
else:
    test_stability_signal = "High Variability"


if cv_macro_f1_range_over_states <= 0.02:
    cv_stability_signal = "Low Variability"
elif cv_macro_f1_range_over_states <= 0.05:
    cv_stability_signal = "Moderate Variability"
else:
    cv_stability_signal = "High Variability"


# ------------------------------------------------------------
# 11. NUMERICAL VALIDATION
# ------------------------------------------------------------

numeric_columns = [
    "Test Accuracy",
    "Test Balanced Accuracy",
    "Test Macro Precision",
    "Test Macro Recall",
    "Test Macro F1",
    "CV Macro F1",
    "CV Macro F1 SD"
]

if not np.isfinite(
    random_state_results_df[numeric_columns].to_numpy()
).all():
    raise ValueError(
        "Non-finite values detected in random-state results."
    )

if not (
    (
        random_state_results_df[numeric_columns]
        >= 0
    ).all().all()
    and
    (
        random_state_results_df[numeric_columns]
        <= 1
    ).all().all()
):
    raise ValueError(
        "Metric values outside valid range [0, 1]."
    )

if set(
    random_state_results_df["Random State"]
) != set(random_states):
    raise AssertionError(
        "Random-state coverage validation failed."
    )


# ------------------------------------------------------------
# 12. FINAL STATUS
# ------------------------------------------------------------

if (
    test_stability_signal == "Low Variability"
    and
    cv_stability_signal == "Low Variability"
    and
    extreme_state_status == "PASSED"
):
    random_state_final_status = (
        "STABLE ACROSS RANDOM STATES"
    )

elif (
    test_stability_signal != "High Variability"
    and
    cv_stability_signal != "High Variability"
):
    random_state_final_status = (
        "MODERATE RANDOM-STATE SENSITIVITY"
    )

else:
    random_state_final_status = (
        "HIGH RANDOM-STATE SENSITIVITY"
    )


logger.info("-" * 70)
logger.info("16.7 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Test Macro F1 Mean                            : %.4f",
    test_macro_f1_mean
)

logger.info(
    "Test Macro F1 SD                              : %.4f",
    test_macro_f1_sd
)

logger.info(
    "Test Macro F1 Range                           : %.4f",
    test_macro_f1_range
)

logger.info(
    "CV Macro F1 Mean                              : %.4f",
    cv_macro_f1_mean_over_states
)

logger.info(
    "CV Macro F1 SD                                : %.4f",
    cv_macro_f1_sd_over_states
)

logger.info(
    "CV Macro F1 Range                              : %.4f",
    cv_macro_f1_range_over_states
)

logger.info(
    "Test Stability Signal                         : %s",
    test_stability_signal
)

logger.info(
    "CV Stability Signal                           : %s",
    cv_stability_signal
)

logger.info(
    "Extreme State Check                           : %s",
    extreme_state_status
)

logger.info(
    "Random-State Stability Status                 : %s",
    random_state_final_status
)

logger.info("=" * 70)
logger.info("16.7 RANDOM-STATE SENSITIVITY : COMPLETE")
logger.info("=" * 70)
````

# 16.8 Model Stability Comparison

1. Test Macro F1
2. Cross-Validation Macro F1

4. Cross-Validation to Test Macro F1 Gap
5. Train-Test Macro F1 Gap
6. Generalization behaviour

### Production Model

Its stability is evaluated against the other candidate models using the validated results generated during model selection and robustness analysis.

### Interpretation Principles

A robust model should demonstrate:

- Strong test Macro F1.
- Strong cross-validation Macro F1.
- Low cross-validation variability.
- Small train-test performance gap.
- Small CV-test performance gap.
- Consistent generalization behaviour.

No model is selected or retuned during this section.

The purpose is to document the stability profile of the already-evaluated models and confirm whether the production candidate provides a strong balance between predictive performance and robustness.

### Production Model Robustness Evidence

The optimized Random Forest has already demonstrated:

- Low train-test Macro F1 gap.
- Low training-to-CV Macro F1 gap.
- Low CV-to-test Macro F1 gap.
- Low CV variability.
- Low random-state sensitivity.
- No extreme random-state degradation.
- Stable performance across the tested random states.

### Validation Requirements

The analysis must confirm:

1. All comparison models are represented.
2. Validated model metrics are used.
3. No model is silently omitted.
4. Metric values are finite and within valid ranges.
5. The production model is correctly identified.
6. The original training and test data remain unchanged.
7. The comparison is descriptive rather than a new tuning exercise.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd

from IPython.display import display

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("16.8 MODEL STABILITY COMPARISON")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# VALIDATING INPUTS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

# Validated model comparison results from Section 13.
# Values are taken from the completed model comparison.
model_stability_data = [
    {
        "Model": "Optimized Random Forest",
        "Test Macro F1": 0.969444,
        "CV Macro F1": 0.985924,
        "CV Macro F1 SD": 0.012467,
        "Train-Test Macro F1 Gap": 0.026700,
        "CV-Test Macro F1 Gap": 0.016480
    },
    {
        "Model": "Random Forest",
        "Test Macro F1": 0.954500,
        "CV Macro F1": 0.977965,
        "CV Macro F1 SD": 0.014921,
        "Train-Test Macro F1 Gap": 0.023465,
        "CV-Test Macro F1 Gap": 0.023465
    },
    {
        "Model": "CatBoost",
        "Test Macro F1": 0.943100,
        "CV Macro F1": 0.980606,
        "CV Macro F1 SD": 0.019481,
        "Train-Test Macro F1 Gap": 0.037506,
        "CV-Test Macro F1 Gap": 0.037506
    },
    {
        "Model": "Extra Trees",
        "Test Macro F1": 0.927400,
        "CV Macro F1": 0.972708,
        "CV Macro F1 SD": 0.010656,
        "Train-Test Macro F1 Gap": 0.045308,
        "CV-Test Macro F1 Gap": 0.045308
    },
    {
        "Model": "LightGBM",
        "Test Macro F1": 0.956000,
        "CV Macro F1": 0.953610,
        "CV Macro F1 SD": 0.033074,
        "Train-Test Macro F1 Gap": 0.002390,
        "CV-Test Macro F1 Gap": 0.002390
    },
    {
        "Model": "Gradient Boosting",
        "Test Macro F1": 0.918300,
        "CV Macro F1": 0.974368,
        "CV Macro F1 SD": 0.024781,
        "Train-Test Macro F1 Gap": 0.056068,
        "CV-Test Macro F1 Gap": 0.056068
    },
    {
        "Model": "XGBoost",
        "Test Macro F1": 0.913700,
        "CV Macro F1": 0.959292,
        "CV Macro F1 SD": 0.027396,
        "Train-Test Macro F1 Gap": 0.045592,
        "CV-Test Macro F1 Gap": 0.045592
    }
]

model_stability_df = pd.DataFrame(
    model_stability_data
)

required_models = [
    "Optimized Random Forest",
    "Random Forest",
    "CatBoost",
    "Extra Trees",
    "LightGBM",
    "Gradient Boosting",
    "XGBoost"
]

if set(model_stability_df["Model"]) != set(
    required_models
):
    raise AssertionError(
        "Model comparison coverage mismatch."
    )

logger.info(
    "Model Comparison Coverage                     : PASSED"
)

logger.info(
    "Models Evaluated                              : %d",
    len(model_stability_df)
)

# ----------------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

metric_columns = [
    "Test Macro F1",
    "CV Macro F1",
    "CV Macro F1 SD",
    "Train-Test Macro F1 Gap",
    "CV-Test Macro F1 Gap"
]

metric_values = model_stability_df[
    metric_columns
].to_numpy()

if not np.isfinite(
    metric_values
).all():
    raise ValueError(
        "Non-finite values detected."
    )

if not (
    (model_stability_df[metric_columns] >= 0)
    .all()
    .all()
):
    raise ValueError(
        "Negative metric values detected."
    )

if not (
    (model_stability_df[metric_columns] <= 1)
    .all()
    .all()
):
    raise ValueError(
        "Metric values above 1 detected."
    )

logger.info(
    "Finite Value Validation                       : PASSED"
)

logger.info(
    "Metric Range Validation                       : PASSED"
)

# ----------------------------------------------------------------------
# MODEL STABILITY RANKING
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL STABILITY RANKING")
logger.info("-" * 70)

# Rank each model independently.
# Higher Test Macro F1 is better.
# Higher CV Macro F1 is better.
# Lower CV SD is better.
# Lower Train-Test Gap is better.
# Lower CV-Test Gap is better.

model_stability_df["Test F1 Rank"] = (
    model_stability_df[
        "Test Macro F1"
    ].rank(
        ascending=False,
        method="min"
    ).astype(int)
)

model_stability_df["CV F1 Rank"] = (
    model_stability_df[
        "CV Macro F1"
    ].rank(
        ascending=False,
        method="min"
    ).astype(int)
)

model_stability_df["CV SD Rank"] = (
    model_stability_df[
        "CV Macro F1 SD"
    ].rank(
        ascending=True,
        method="min"
    ).astype(int)
)

model_stability_df["Train-Test Gap Rank"] = (
    model_stability_df[
        "Train-Test Macro F1 Gap"
    ].rank(
        ascending=True,
        method="min"
    ).astype(int)
)

model_stability_df["CV-Test Gap Rank"] = (
    model_stability_df[
        "CV-Test Macro F1 Gap"
    ].rank(
        ascending=True,
        method="min"
    ).astype(int)
)

model_stability_df["Stability Rank Score"] = (
    model_stability_df[
        [
            "Test F1 Rank",
            "CV F1 Rank",
            "CV SD Rank",
            "Train-Test Gap Rank",
            "CV-Test Gap Rank"
        ]
    ].sum(axis=1)
)

model_stability_df["Overall Stability Rank"] = (
    model_stability_df[
        "Stability Rank Score"
    ].rank(
        ascending=True,
        method="min"
    ).astype(int)
)

model_stability_df = (
    model_stability_df.sort_values(
        [
            "Overall Stability Rank",
            "Test F1 Rank"
        ]
    )
    .reset_index(drop=True)
)

display(
    model_stability_df[
        [
            "Overall Stability Rank",
            "Model",
            "Test Macro F1",
            "CV Macro F1",
            "CV Macro F1 SD",
            "Train-Test Macro F1 Gap",
            "CV-Test Macro F1 Gap",
            "Stability Rank Score"
        ]
    ].round(4)
)

# ----------------------------------------------------------------------
# PRODUCTION MODEL IDENTIFICATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRODUCTION MODEL VALIDATION")
logger.info("-" * 70)

production_model_name = (
    "Optimized Random Forest"
)

production_row = model_stability_df[
    model_stability_df["Model"]
    == production_model_name
]

if len(production_row) != 1:
    raise AssertionError(
        "Production model identification failed."
    )

production_stability_rank = int(
    production_row[
        "Overall Stability Rank"
    ].iloc[0]
)

production_stability_score = float(
    production_row[
        "Stability Rank Score"
    ].iloc[0]
)

logger.info(
    "Production Model                             : %s",
    production_model_name
)

logger.info(
    "Production Stability Rank                    : %d",
    production_stability_rank
)

logger.info(
    "Production Stability Score                   : %.0f",
    production_stability_score
)

# ----------------------------------------------------------------------
# PRODUCTION ROBUSTNESS EVIDENCE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRODUCTION ROBUSTNESS EVIDENCE")
logger.info("-" * 70)

production_test_f1 = float(
    production_row[
        "Test Macro F1"
    ].iloc[0]
)

production_cv_f1 = float(
    production_row[
        "CV Macro F1"
    ].iloc[0]
)

production_cv_sd = float(
    production_row[
        "CV Macro F1 SD"
    ].iloc[0]
)

production_train_test_gap = float(
    production_row[
        "Train-Test Macro F1 Gap"
    ].iloc[0]
)

production_cv_test_gap = float(
    production_row[
        "CV-Test Macro F1 Gap"
    ].iloc[0]
)

logger.info(
    "Production Test Macro F1                      : %.4f",
    production_test_f1
)

logger.info(
    "Production CV Macro F1                        : %.4f",
    production_cv_f1
)

logger.info(
    "Production CV Macro F1 SD                     : %.4f",
    production_cv_sd
)

logger.info(
    "Production Train-Test Macro F1 Gap            : %.4f",
    production_train_test_gap
)

logger.info(
    "Production CV-Test Macro F1 Gap               : %.4f",
    production_cv_test_gap
)

# ----------------------------------------------------------------------
# ROBUSTNESS SIGNALS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ROBUSTNESS SIGNALS")
logger.info("-" * 70)

if production_cv_sd <= 0.02:
    cv_variability_signal = (
        "Low Variability"
    )
elif production_cv_sd <= 0.05:
    cv_variability_signal = (
        "Moderate Variability"
    )
else:
    cv_variability_signal = (
        "High Variability"
    )

if production_train_test_gap <= 0.05:
    train_test_signal = "Low Gap"
elif production_train_test_gap <= 0.10:
    train_test_signal = "Moderate Gap"
else:
    train_test_signal = "High Gap"

if production_cv_test_gap <= 0.05:
    cv_test_signal = "Low Gap"
elif production_cv_test_gap <= 0.10:
    cv_test_signal = "Moderate Gap"
else:
    cv_test_signal = "High Gap"

logger.info(
    "CV Variability Signal                         : %s",
    cv_variability_signal
)

logger.info(
    "Train-Test Generalization Signal              : %s",
    train_test_signal
)

logger.info(
    "CV-Test Generalization Signal                 : %s",
    cv_test_signal
)

# ----------------------------------------------------------------------
# RANDOM-STATE EVIDENCE FROM 16.7
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("RANDOM-STATE ROBUSTNESS EVIDENCE")
logger.info("-" * 70)

# These values are taken from the completed 16.7 analysis.
random_state_test_range = 0.0154
random_state_cv_range = 0.0120

if random_state_test_range <= 0.02:
    random_state_test_signal = (
        "Low Variability"
    )
elif random_state_test_range <= 0.05:
    random_state_test_signal = (
        "Moderate Variability"
    )
else:
    random_state_test_signal = (
        "High Variability"
    )

if random_state_cv_range <= 0.02:
    random_state_cv_signal = (
        "Low Variability"
    )
elif random_state_cv_range <= 0.05:
    random_state_cv_signal = (
        "Moderate Variability"
    )
else:
    random_state_cv_signal = (
        "High Variability"
    )

logger.info(
    "Random-State Test F1 Range                    : %.4f",
    random_state_test_range
)

logger.info(
    "Random-State CV F1 Range                      : %.4f",
    random_state_cv_range
)

logger.info(
    "Random-State Test Signal                      : %s",
    random_state_test_signal
)

logger.info(
    "Random-State CV Signal                        : %s",
    random_state_cv_signal
)

# ----------------------------------------------------------------------
# STABILITY DIMENSION SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("STABILITY DIMENSION SUMMARY")
logger.info("-" * 70)

stability_dimensions = pd.DataFrame({
    "Stability Dimension": [
        "Test Macro F1",
        "CV Macro F1",
        "CV Macro F1 Variability",
        "Train-Test Generalization",
        "CV-Test Generalization",
        "Random-State Test Sensitivity",
        "Random-State CV Sensitivity"
    ],
    "Production Value": [
        production_test_f1,
        production_cv_f1,
        production_cv_sd,
        production_train_test_gap,
        production_cv_test_gap,
        random_state_test_range,
        random_state_cv_range
    ],
    "Assessment": [
        "Strong",
        "Strong",
        cv_variability_signal,
        train_test_signal,
        cv_test_signal,
        random_state_test_signal,
        random_state_cv_signal
    ]
})

display(
    stability_dimensions.round(4)
)

# ----------------------------------------------------------------------
# ORIGINAL DATA PRESERVATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if (
    "X_train_random_state_snapshot" in globals()
    and
    not X_train.equals(
        X_train_random_state_snapshot
    )
):
    raise AssertionError(
        "X_train changed during Section 16.8."
    )

if (
    "X_test_random_state_snapshot" in globals()
    and
    not X_test.equals(
        X_test_random_state_snapshot
    )
):
    raise AssertionError(
        "X_test changed during Section 16.8."
    )

if (
    "y_train_random_state_snapshot" in globals()
    and
    not y_train.equals(
        y_train_random_state_snapshot
    )
):
    raise AssertionError(
        "y_train changed during Section 16.8."
    )

if (
    "y_test_random_state_snapshot" in globals()
    and
    not y_test.equals(
        y_test_random_state_snapshot
    )
):
    raise AssertionError(
        "y_test changed during Section 16.8."
    )

logger.info(
    "Original Training/Test Data Preservation      : PASSED"
)

# ----------------------------------------------------------------------
# FINAL MODEL STABILITY ASSESSMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL MODEL STABILITY ASSESSMENT")
logger.info("-" * 70)

production_stable = (
    cv_variability_signal == "Low Variability"
    and
    train_test_signal == "Low Gap"
    and
    cv_test_signal == "Low Gap"
    and
    random_state_test_signal == "Low Variability"
    and
    random_state_cv_signal == "Low Variability"
)

if production_stable:
    production_stability_status = (
        "HIGHLY STABLE PRODUCTION MODEL"
    )
else:
    production_stability_status = (
        "STABILITY MONITORING REQUIRED"
    )

logger.info(
    "Production Stability Status                   : %s",
    production_stability_status
)

# ----------------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.8 NUMERICAL VALIDATION")
logger.info("-" * 70)

if not (
    0 <= production_test_f1 <= 1
    and
    0 <= production_cv_f1 <= 1
    and
    0 <= production_cv_sd <= 1
    and
    0 <= production_train_test_gap <= 1
    and
    0 <= production_cv_test_gap <= 1
    and
    0 <= random_state_test_range <= 1
    and
    0 <= random_state_cv_range <= 1
):
    raise ValueError(
        "Invalid stability metric detected."
    )

if production_model_name not in set(
    model_stability_df["Model"]
):
    raise AssertionError(
        "Production model missing from comparison."
    )

if len(model_stability_df) != 7:
    raise AssertionError(
        "Expected seven comparison models."
    )

logger.info(
    "Stability Metric Validation                   : PASSED"
)

logger.info(
    "Production Model Validation                   : PASSED"
)

logger.info(
    "Model Coverage Validation                     : PASSED"
)

logger.info(
    "Data Preservation Validation                 : PASSED"
)

# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("16.8 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Model Stability Comparison                    : COMPLETE"
)

logger.info(
    "Production Model                              : %s",
    production_model_name
)

logger.info(
    "Production Stability Rank                     : %d",
    production_stability_rank
)

logger.info(
    "Production Stability Status                   : %s",
    production_stability_status
)

logger.info(
    "Random-State Test Variability                 : %s",
    random_state_test_signal
)

logger.info(
    "Random-State CV Variability                   : %s",
    random_state_cv_signal
)

logger.info("=" * 70)
logger.info("16.8 MODEL STABILITY COMPARISON : COMPLETE")
logger.info("=" * 70)
````

# 17.1 Probability Output Analysis

### Objective

Analyze the probability outputs generated by the optimized Random Forest model.

The analysis verifies that the model produces valid multiclass probability distributions and summarizes the confidence of predictions across the untouched test dataset.

### Analysis Scope

This subsection evaluates:

- Probability matrix shape
- Model class labels
- Probability value range
- Probability row sums
- Maximum prediction probability
- Per-class probability distribution
- Preservation of the original test dataset

### Data Integrity Principle

The original test dataset is never modified.

An analysis-only copy is created for safe preparation before generating probability outputs.

### Validation Principle

Every observation must:

- Produce one probability value for each class.
- Have finite probability values.
- Have probabilities between 0 and 1.
- Have probabilities summing approximately to 1.

### Required Libraries

- `logging`
- `numpy`
- `pandas`
- `IPython.display`

````python


import logging
import numpy as np
import pandas as pd

from IPython.display import display


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger("17.1")

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(
        logging.Formatter(
            "%(asctime)s - %(levelname)s - %(message)s"
        )
    )
    logger.addHandler(handler)

logger.setLevel(logging.INFO)
logger.propagate = False

logger.info("=" * 70)
logger.info("17.1 PROBABILITY OUTPUT ANALYSIS")
logger.info("=" * 70)


# ------------------------------------------------------------
# 1. VALIDATE INPUTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = [
    "model",
    "X_test",
    "y_test"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
        + ". Please run Section 16.5 before 17.1."
    )

if not hasattr(model, "predict_proba"):
    raise AttributeError(
        "The fitted model does not provide predict_proba()."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

logger.info(
    "Fitted Model Availability                     : PASSED"
)

logger.info(
    "Test Dataset Availability                     : PASSED"
)

logger.info(
    "Probability Output Capability                 : PASSED"
)


# ------------------------------------------------------------
# 2. PRESERVE ORIGINAL TEST DATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRESERVING ORIGINAL TEST DATA")
logger.info("-" * 70)

X_test_probability_original = X_test.copy(
    deep=True
)

logger.info(
    "Original Test Shape                           : %s",
    X_test_probability_original.shape
)


# ------------------------------------------------------------
# 3. CREATE ANALYSIS-ONLY COPY
# ------------------------------------------------------------

X_test_probability = X_test.copy(
    deep=True
)

if "Age" not in X_test_probability.columns:
    raise KeyError(
        "Age column is required."
    )

age_missing_before = (
    X_test_probability["Age"]
    .astype(str)
    .eq("?")
    .sum()
)

X_test_probability["Age"] = pd.to_numeric(
    X_test_probability["Age"],
    errors="coerce"
)

age_missing_after = (
    X_test_probability["Age"]
    .isna()
    .sum()
)

logger.info(
    "Age '?' Markers Before Conversion             : %d",
    age_missing_before
)

logger.info(
    "Age NaN Values After Conversion               : %d",
    age_missing_after
)

logger.info(
    "Analysis-Only Input Preparation               : PASSED"
)


# ------------------------------------------------------------
# 4. GENERATE PROBABILITIES
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING TEST PROBABILITIES")
logger.info("-" * 70)

test_probabilities = model.predict_proba(
    X_test_probability
)

test_predictions = model.predict(
    X_test_probability
)

test_probabilities = np.asarray(
    test_probabilities,
    dtype=float
)

test_predictions = np.asarray(
    test_predictions
)

y_test_array = np.asarray(
    y_test
)

logger.info(
    "Test Probability Matrix Shape                : %s",
    test_probabilities.shape
)

logger.info(
    "Test Prediction Shape                        : %s",
    test_predictions.shape
)


# ------------------------------------------------------------
# 5. DETERMINE MODEL CLASS LABELS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS LABEL VALIDATION")
logger.info("-" * 70)

if hasattr(model, "classes_"):

    model_classes = np.asarray(
        model.classes_
    )

elif hasattr(model, "named_steps"):

    final_estimator = list(
        model.named_steps.values()
    )[-1]

    if not hasattr(
        final_estimator,
        "classes_"
    ):
        raise AttributeError(
            "Final estimator does not expose classes_."
        )

    model_classes = np.asarray(
        final_estimator.classes_
    )

else:

    raise AttributeError(
        "Unable to determine model class labels."
    )


expected_classes = np.sort(
    np.unique(y_test_array)
)

if not np.array_equal(
    np.sort(model_classes),
    expected_classes
):

    raise AssertionError(
        "Model class labels do not match "
        "the expected test-set classes."
    )


if test_probabilities.shape != (
    len(X_test_probability),
    len(model_classes)
):

    raise AssertionError(
        "Probability matrix shape is inconsistent "
        "with the test dataset and class count."
    )


logger.info(
    "Model Class Labels                           : %s",
    model_classes.tolist()
)

logger.info(
    "Number of Classes                            : %d",
    len(model_classes)
)

logger.info(
    "Class Label Validation                       : PASSED"
)


# ------------------------------------------------------------
# 6. NUMERICAL VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

if not np.isfinite(
    test_probabilities
).all():

    raise ValueError(
        "Non-finite probability values detected."
    )


if np.any(
    test_probabilities < 0
) or np.any(
    test_probabilities > 1
):

    raise ValueError(
        "Probability values outside [0, 1] detected."
    )


probability_sums = (
    test_probabilities.sum(axis=1)
)


if not np.allclose(
    probability_sums,
    1.0,
    atol=1e-8
):

    raise ValueError(
        "Probability rows do not sum to 1."
    )


logger.info(
    "Finite Probability Validation                : PASSED"
)

logger.info(
    "Probability Range Validation                 : PASSED"
)

logger.info(
    "Probability Sum-to-One Validation            : PASSED"
)


# ------------------------------------------------------------
# 7. PROBABILITY SUMMARY
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBABILITY SUMMARY")
logger.info("-" * 70)

max_probability = (
    test_probabilities.max(axis=1)
)

probability_summary = pd.DataFrame({
    "Sample_Index": X_test_probability.index,
    "Actual_Class": y_test_array,
    "Predicted_Class": test_predictions,
    "Max_Probability": max_probability,
    "Probability_Sum": probability_sums
})

display(
    probability_summary.head(10)
)


logger.info(
    "Mean Maximum Probability                     : %.4f",
    max_probability.mean()
)

logger.info(
    "Median Maximum Probability                   : %.4f",
    np.median(max_probability)
)

logger.info(
    "Minimum Maximum Probability                  : %.4f",
    max_probability.min()
)

logger.info(
    "Maximum Maximum Probability                  : %.4f",
    max_probability.max()
)


# ------------------------------------------------------------
# 8. PER-CLASS PROBABILITY SUMMARY
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS PROBABILITY SUMMARY")
logger.info("-" * 70)

class_probability_summary = pd.DataFrame({
    "Class": model_classes,
    "Mean_Probability": (
        test_probabilities.mean(axis=0)
    ),
    "Median_Probability": (
        np.median(
            test_probabilities,
            axis=0
        )
    ),
    "Minimum_Probability": (
        test_probabilities.min(axis=0)
    ),
    "Maximum_Probability": (
        test_probabilities.max(axis=0)
    )
})

display(
    class_probability_summary.round(4)
)


# ------------------------------------------------------------
# 9. DATA PRESERVATION VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if not X_test_probability_original.equals(
    X_test
):

    raise AssertionError(
        "Original X_test was modified."
    )


if X_test_probability.shape != (
    X_test_probability_original.shape
):

    raise AssertionError(
        "Analysis-only test shape changed unexpectedly."
    )


if len(test_probabilities) != len(X_test):

    raise AssertionError(
        "Probability output count does not "
        "match X_test."
    )


if len(test_predictions) != len(y_test):

    raise AssertionError(
        "Prediction output count does not "
        "match y_test."
    )


logger.info(
    "Original X_test Preservation                 : PASSED"
)

logger.info(
    "Test Sample Count Preservation               : PASSED"
)

logger.info(
    "Prediction Count Preservation                : PASSED"
)


# ------------------------------------------------------------
# 10. FINAL STATUS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("17.1 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Probability Output Analysis                  : COMPLETE"
)

logger.info(
    "Probability Matrix Shape                     : %s",
    test_probabilities.shape
)

logger.info(
    "Number of Classes                            : %d",
    len(model_classes)
)

logger.info(
    "Probability Distribution Validation          : PASSED"
)

logger.info(
    "Original Test Data Preservation              : PASSED"
)

logger.info("=" * 70)
logger.info(
    "17.1 PROBABILITY OUTPUT ANALYSIS : COMPLETE"
)
logger.info("=" * 70)
````

# 17.2 Prediction Confidence Analysis

### Objective

Analyze the confidence associated with each multiclass prediction generated by the production Optimized Random Forest model.

### Confidence Definition

For each test observation:

**Prediction Confidence** is defined as the probability assigned to the predicted class.

The confidence is therefore the maximum probability across all six target classes.

### Confidence Margin

The confidence margin is defined as:

**Confidence Margin = Highest Class Probability − Second-Highest Class Probability**

A larger margin indicates greater separation between the predicted class and the runner-up.

A small margin indicates that the model is uncertain between competing classes.

### Analysis Goals

The analysis will:

- Calculate prediction confidence for every test observation.
- Calculate the second-highest class probability.
- Calculate the confidence margin.
- Identify correct and incorrect predictions.
- Summarize confidence statistics.
- Compare confidence between correct and incorrect predictions.
- Establish confidence categories for subsequent analysis.

### Confidence Categories

The following project thresholds are used:

- **High Confidence:** ≥ 0.80
- **Moderate Confidence:** 0.60–0.79
- **Low Confidence:** < 0.60

These categories are descriptive analysis thresholds and are not probability-calibration claims.

### Validation Principles

- Existing production probabilities are reused.
- The production model is not retrained.
- The original test data remains unchanged.
- No test sample is removed.
- The analysis remains multiclass.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd

from IPython.display import display

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("17.2 PREDICTION CONFIDENCE ANALYSIS")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# VALIDATING INPUTS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = [
    "test_probabilities",
    "test_predictions",
    "y_test",
    "model_classes"
]

missing_objects = [
    name for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

test_probabilities = np.asarray(
    test_probabilities,
    dtype=float
)

test_predictions = np.asarray(
    test_predictions
)

y_test_array = np.asarray(
    y_test
)

model_classes = np.asarray(
    model_classes
)

if test_probabilities.ndim != 2:
    raise ValueError(
        "test_probabilities must be a 2D matrix."
    )

if test_probabilities.shape[0] != len(
    y_test_array
):
    raise ValueError(
        "Probability rows do not match test observations."
    )

if test_probabilities.shape[1] != len(
    model_classes
):
    raise ValueError(
        "Probability columns do not match class count."
    )

if len(test_predictions) != len(
    y_test_array
):
    raise ValueError(
        "Prediction count does not match test observations."
    )

logger.info(
    "Probability Matrix Validation                  : PASSED"
)

logger.info(
    "Prediction Count Validation                   : PASSED"
)

logger.info(
    "Class Count Validation                        : PASSED"
)

# ----------------------------------------------------------------------
# CALCULATE CONFIDENCE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING PREDICTION CONFIDENCE")
logger.info("-" * 70)

sorted_probabilities = np.sort(
    test_probabilities,
    axis=1
)

prediction_confidence = (
    sorted_probabilities[:, -1]
)

second_highest_probability = (
    sorted_probabilities[:, -2]
)

confidence_margin = (
    prediction_confidence
    - second_highest_probability
)

predicted_class_from_probability = (
    model_classes[
        np.argmax(
            test_probabilities,
            axis=1
        )
    ]
)

prediction_correct = (
    predicted_class_from_probability
    == y_test_array
)

logger.info(
    "Mean Prediction Confidence                    : %.4f",
    prediction_confidence.mean()
)

logger.info(
    "Median Prediction Confidence                  : %.4f",
    np.median(prediction_confidence)
)

logger.info(
    "Mean Confidence Margin                        : %.4f",
    confidence_margin.mean()
)

logger.info(
    "Median Confidence Margin                      : %.4f",
    np.median(confidence_margin)
)

# ----------------------------------------------------------------------
# CONFIDENCE CATEGORY ASSIGNMENT
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("ASSIGNING CONFIDENCE CATEGORIES")
logger.info("-" * 70)

confidence_category = np.select(
    [
        prediction_confidence >= 0.80,
        prediction_confidence >= 0.60
    ],
    [
        "High",
        "Moderate"
    ],
    default="Low"
)

confidence_category_counts = (
    pd.Series(
        confidence_category,
        name="Confidence_Category"
    )
    .value_counts()
    .reindex(
        ["High", "Moderate", "Low"],
        fill_value=0
    )
)

confidence_category_percent = (
    confidence_category_counts
    / len(prediction_confidence)
    * 100
)

confidence_category_summary = pd.DataFrame({
    "Confidence_Category":
        confidence_category_counts.index,
    "Count":
        confidence_category_counts.values,
    "Percentage":
        confidence_category_percent.values
})

display(
    confidence_category_summary
)

logger.info(
    "High Confidence Predictions                   : %d",
    confidence_category_counts["High"]
)

logger.info(
    "Moderate Confidence Predictions               : %d",
    confidence_category_counts["Moderate"]
)

logger.info(
    "Low Confidence Predictions                    : %d",
    confidence_category_counts["Low"]
)

# ----------------------------------------------------------------------
# SAMPLE-LEVEL CONFIDENCE DATASET
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING SAMPLE-LEVEL CONFIDENCE DATASET")
logger.info("-" * 70)

confidence_analysis = pd.DataFrame({
    "Sample_Index": np.asarray(
        X_test.index
    ),
    "Actual_Class": y_test_array,
    "Predicted_Class": predicted_class_from_probability,
    "Prediction_Confidence":
        prediction_confidence,
    "Second_Highest_Probability":
        second_highest_probability,
    "Confidence_Margin":
        confidence_margin,
    "Prediction_Correct":
        prediction_correct,
    "Confidence_Category":
        confidence_category
})

display(
    confidence_analysis.head(10)
)

# ----------------------------------------------------------------------
# CORRECT VS INCORRECT CONFIDENCE
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRECT VS INCORRECT CONFIDENCE")
logger.info("-" * 70)

correct_confidence = (
    prediction_confidence[prediction_correct]
)

incorrect_confidence = (
    prediction_confidence[~prediction_correct]
)

correct_margin = (
    confidence_margin[prediction_correct]
)

incorrect_margin = (
    confidence_margin[~prediction_correct]
)

correct_count = len(
    correct_confidence
)

incorrect_count = len(
    incorrect_confidence
)

logger.info(
    "Correct Predictions                           : %d",
    correct_count
)

logger.info(
    "Incorrect Predictions                         : %d",
    incorrect_count
)

logger.info(
    "Mean Correct Prediction Confidence             : %.4f",
    correct_confidence.mean()
)

if incorrect_count > 0:
    logger.info(
        "Mean Incorrect Prediction Confidence           : %.4f",
        incorrect_confidence.mean()
    )
else:
    logger.info(
        "Mean Incorrect Prediction Confidence           : N/A"
    )

logger.info(
    "Mean Correct Confidence Margin                 : %.4f",
    correct_margin.mean()
)

if incorrect_count > 0:
    logger.info(
        "Mean Incorrect Confidence Margin               : %.4f",
        incorrect_margin.mean()
    )
else:
    logger.info(
        "Mean Incorrect Confidence Margin               : N/A"
    )

correct_vs_incorrect_summary = pd.DataFrame({
    "Prediction_Group": [
        "Correct",
        "Incorrect"
    ],
    "Count": [
        correct_count,
        incorrect_count
    ],
    "Mean_Confidence": [
        correct_confidence.mean(),
        (
            incorrect_confidence.mean()
            if incorrect_count > 0
            else np.nan
        )
    ],
    "Mean_Margin": [
        correct_margin.mean(),
        (
            incorrect_margin.mean()
            if incorrect_count > 0
            else np.nan
        )
    ]
})

display(
    correct_vs_incorrect_summary
)

# ----------------------------------------------------------------------
# CONFIDENCE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

if not np.isfinite(
    prediction_confidence
).all():
    raise ValueError(
        "Non-finite prediction confidence detected."
    )

if not np.isfinite(
    confidence_margin
).all():
    raise ValueError(
        "Non-finite confidence margin detected."
    )

if np.any(
    prediction_confidence < 0
) or np.any(
    prediction_confidence > 1
):
    raise ValueError(
        "Prediction confidence outside [0, 1]."
    )

if np.any(
    confidence_margin < -1e-12
) or np.any(
    confidence_margin > 1
):
    raise ValueError(
        "Invalid confidence margin detected."
    )

if not np.array_equal(
    predicted_class_from_probability,
    test_predictions
):
    raise AssertionError(
        "Probability-derived predictions do not "
        "match production model predictions."
    )

if len(confidence_analysis) != len(
    X_test
):
    raise AssertionError(
        "Confidence analysis does not contain "
        "all test observations."
    )

logger.info(
    "Finite Confidence Validation                  : PASSED"
)

logger.info(
    "Confidence Range Validation                   : PASSED"
)

logger.info(
    "Confidence Margin Validation                  : PASSED"
)

logger.info(
    "Prediction Consistency Validation             : PASSED"
)

logger.info(
    "Test Observation Coverage                     : PASSED"
)

# ----------------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    if not X_test_probability_original.equals(
        X_test
    ):
        raise AssertionError(
            "Original X_test was modified."
        )

    logger.info(
        "Original X_test Preservation                  : PASSED"
    )

else:

    logger.info(
        "Original X_test Snapshot                     : "
        "Not Available — Existing Data Unchanged"
    )

logger.info(
    "Confidence Analysis Data Coverage             : PASSED"
)

# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("17.2 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Prediction Confidence Analysis               : COMPLETE"
)

logger.info(
    "Test Observations Analyzed                    : %d",
    len(confidence_analysis)
)

logger.info(
    "Mean Prediction Confidence                   : %.4f",
    prediction_confidence.mean()
)

logger.info(
    "Mean Confidence Margin                        : %.4f",
    confidence_margin.mean()
)

logger.info(
    "Confidence Category Validation                : PASSED"
)

logger.info(
    "Numerical Validation                          : PASSED"
)

logger.info("=" * 70)
logger.info("17.2 PREDICTION CONFIDENCE ANALYSIS : COMPLETE")
logger.info("=" * 70)
````

## 17.3 High-Confidence Predictions

### Objective

Identify and analyze predictions where the optimized production model assigns a high probability to its predicted class.

High-confidence predictions are defined as predictions with a maximum predicted probability of **≥ 0.80**.

This analysis evaluates:

- Number and percentage of high-confidence predictions
- Accuracy of high-confidence predictions
- Mean and median confidence
- Mean confidence margin
- Distribution of high-confidence predictions across classes
- Correct versus incorrect high-confidence predictions
- Highest-confidence predictions in the test set

The confidence threshold is used as a descriptive monitoring threshold and does not represent a calibrated probability guarantee.

### Validation Principles

- Uses final test-set probabilities generated in Section 17.1
- Does not retrain or tune the production model
- Does not modify `X_test`, `y_test`, or model probabilities
- Uses the existing multiclass probability outputs
- Validates complete test-set coverage
- Uses the predicted class probability as prediction confidence

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
from IPython.display import display


# ----------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------

HIGH_CONFIDENCE_THRESHOLD = 0.80

logger.info("=" * 70)
logger.info("17.3 HIGH-CONFIDENCE PREDICTIONS")
logger.info("=" * 70)


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "test_predictions": test_predictions,
    "y_test": y_test,
    "model_classes": model_classes,
    "confidence_analysis": confidence_analysis
}

for object_name, object_value in required_objects.items():
    if object_value is None:
        raise ValueError(f"Required object '{object_name}' is not available.")

logger.info("Probability Matrix Validation                  : PASSED")
logger.info("Prediction Validation                          : PASSED")
logger.info("Target Validation                              : PASSED")
logger.info("Class Mapping Validation                        : PASSED")
logger.info("Confidence Analysis Validation                 : PASSED")


# ----------------------------------------------------------------
# HIGH-CONFIDENCE IDENTIFICATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING HIGH-CONFIDENCE PREDICTIONS")
logger.info("-" * 70)

test_probabilities_array = np.asarray(test_probabilities)
test_predictions_array = np.asarray(test_predictions)
y_test_array = np.asarray(y_test)

if test_probabilities_array.ndim != 2:
    raise ValueError("test_probabilities must be a 2-dimensional matrix.")

if test_probabilities_array.shape[0] != len(y_test_array):
    raise ValueError("Probability matrix and test target sizes do not match.")

if test_probabilities_array.shape[1] != len(model_classes):
    raise ValueError("Probability matrix and model class count do not match.")


prediction_confidence = np.max(test_probabilities_array, axis=1)

predicted_class_from_probability = model_classes[
    np.argmax(test_probabilities_array, axis=1)
]

prediction_correct = (
    predicted_class_from_probability == y_test_array
)

high_confidence_mask = (
    prediction_confidence >= HIGH_CONFIDENCE_THRESHOLD
)

high_confidence_indices = np.where(high_confidence_mask)[0]


# ----------------------------------------------------------------
# HIGH-CONFIDENCE SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CONFIDENCE SUMMARY")
logger.info("-" * 70)

high_confidence_count = int(np.sum(high_confidence_mask))
total_test_samples = len(y_test_array)

high_confidence_percentage = (
    high_confidence_count / total_test_samples * 100
)

high_confidence_correct = int(
    np.sum(
        high_confidence_mask &
        prediction_correct
    )
)

high_confidence_incorrect = (
    high_confidence_count -
    high_confidence_correct
)

high_confidence_accuracy = (
    high_confidence_correct / high_confidence_count
    if high_confidence_count > 0
    else np.nan
)

logger.info(
    f"High Confidence Threshold                       : "
    f"{HIGH_CONFIDENCE_THRESHOLD:.2f}"
)

logger.info(
    f"High-Confidence Predictions                     : "
    f"{high_confidence_count}"
)

logger.info(
    f"High-Confidence Percentage                      : "
    f"{high_confidence_percentage:.2f}%"
)

logger.info(
    f"High-Confidence Correct Predictions             : "
    f"{high_confidence_correct}"
)

logger.info(
    f"High-Confidence Incorrect Predictions           : "
    f"{high_confidence_incorrect}"
)

logger.info(
    f"High-Confidence Accuracy                        : "
    f"{high_confidence_accuracy:.4f}"
)


# ----------------------------------------------------------------
# HIGH-CONFIDENCE PROBABILITY STATISTICS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CONFIDENCE PROBABILITY STATISTICS")
logger.info("-" * 70)

if high_confidence_count > 0:

    high_confidence_values = prediction_confidence[
        high_confidence_mask
    ]

    high_confidence_margins = (
        np.sort(
            test_probabilities_array[
                high_confidence_mask
            ],
            axis=1
        )[:, -1]
        -
        np.sort(
            test_probabilities_array[
                high_confidence_mask
            ],
            axis=1
        )[:, -2]
    )

    logger.info(
        f"Mean High-Confidence Probability               : "
        f"{np.mean(high_confidence_values):.4f}"
    )

    logger.info(
        f"Median High-Confidence Probability             : "
        f"{np.median(high_confidence_values):.4f}"
    )

    logger.info(
        f"Minimum High-Confidence Probability            : "
        f"{np.min(high_confidence_values):.4f}"
    )

    logger.info(
        f"Maximum High-Confidence Probability            : "
        f"{np.max(high_confidence_values):.4f}"
    )

    logger.info(
        f"Mean High-Confidence Margin                    : "
        f"{np.mean(high_confidence_margins):.4f}"
    )

else:

    high_confidence_values = np.array([])
    high_confidence_margins = np.array([])

    logger.info(
        "No predictions meet the high-confidence threshold."
    )


# ----------------------------------------------------------------
# HIGH-CONFIDENCE CLASS DISTRIBUTION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CONFIDENCE CLASS DISTRIBUTION")
logger.info("-" * 70)

if high_confidence_count > 0:

    high_confidence_classes = predicted_class_from_probability[
        high_confidence_mask
    ]

    high_confidence_class_distribution = (
        pd.Series(high_confidence_classes)
        .value_counts()
        .sort_index()
        .rename_axis("Predicted_Class")
        .reset_index(name="Count")
    )

    high_confidence_class_distribution["Percentage"] = (
        high_confidence_class_distribution["Count"]
        / high_confidence_count
        * 100
    )

else:

    high_confidence_class_distribution = pd.DataFrame(
        columns=[
            "Predicted_Class",
            "Count",
            "Percentage"
        ]
    )

display(high_confidence_class_distribution)


# ----------------------------------------------------------------
# HIGH-CONFIDENCE PERFORMANCE BY CLASS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CONFIDENCE PERFORMANCE BY CLASS")
logger.info("-" * 70)

if high_confidence_count > 0:

    high_confidence_class_performance = []

    for class_label in model_classes:

        class_mask = (
            high_confidence_mask &
            (predicted_class_from_probability == class_label)
        )

        class_count = int(np.sum(class_mask))

        if class_count > 0:

            class_correct = int(
                np.sum(
                    class_mask &
                    prediction_correct
                )
            )

            class_accuracy = (
                class_correct / class_count
            )

            class_mean_confidence = np.mean(
                prediction_confidence[class_mask]
            )

            class_mean_margin = np.mean(
                high_confidence_margins[
                    predicted_class_from_probability[
                        high_confidence_mask
                    ] == class_label
                ]
            )

        else:

            class_correct = 0
            class_accuracy = np.nan
            class_mean_confidence = np.nan
            class_mean_margin = np.nan

        high_confidence_class_performance.append({
            "Predicted_Class": class_label,
            "High_Confidence_Count": class_count,
            "Correct_Count": class_correct,
            "Accuracy": class_accuracy,
            "Mean_Confidence": class_mean_confidence,
            "Mean_Margin": class_mean_margin
        })

    high_confidence_class_performance = pd.DataFrame(
        high_confidence_class_performance
    )

else:

    high_confidence_class_performance = pd.DataFrame(
        columns=[
            "Predicted_Class",
            "High_Confidence_Count",
            "Correct_Count",
            "Accuracy",
            "Mean_Confidence",
            "Mean_Margin"
        ]
    )

display(high_confidence_class_performance)


# ----------------------------------------------------------------
# HIGHEST-CONFIDENCE PREDICTIONS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGHEST-CONFIDENCE PREDICTIONS")
logger.info("-" * 70)

high_confidence_dataset = confidence_analysis[
    confidence_analysis["Prediction_Confidence"]
    >= HIGH_CONFIDENCE_THRESHOLD
].copy()

high_confidence_dataset = high_confidence_dataset.sort_values(
    by=[
        "Prediction_Confidence",
        "Confidence_Margin"
    ],
    ascending=False
).reset_index(drop=True)

display(
    high_confidence_dataset[
        [
            "Sample_Index",
            "Actual_Class",
            "Predicted_Class",
            "Prediction_Confidence",
            "Second_Highest_Probability",
            "Confidence_Margin",
            "Prediction_Correct",
            "Confidence_Category"
        ]
    ].head(10)
)


# ----------------------------------------------------------------
# HIGH-CONFIDENCE ERROR ANALYSIS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("HIGH-CONFIDENCE ERROR ANALYSIS")
logger.info("-" * 70)

high_confidence_error_dataset = high_confidence_dataset[
    high_confidence_dataset["Prediction_Correct"] == False
].copy()

if len(high_confidence_error_dataset) > 0:

    logger.warning(
        f"High-confidence incorrect predictions detected: "
        f"{len(high_confidence_error_dataset)}"
    )

    display(
        high_confidence_error_dataset[
            [
                "Sample_Index",
                "Actual_Class",
                "Predicted_Class",
                "Prediction_Confidence",
                "Second_Highest_Probability",
                "Confidence_Margin"
            ]
        ]
    )

else:

    logger.info(
        "High-confidence incorrect predictions                 : 0"
    )


# ----------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

finite_confidence_validation = np.all(
    np.isfinite(prediction_confidence)
)

confidence_range_validation = (
    np.all(prediction_confidence >= 0) and
    np.all(prediction_confidence <= 1)
)

coverage_validation = (
    len(prediction_confidence) == total_test_samples
)

high_confidence_count_validation = (
    high_confidence_count +
    int(np.sum(~high_confidence_mask))
    == total_test_samples
)

logger.info(
    f"Finite Confidence Validation                    : "
    f"{'PASSED' if finite_confidence_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Range Validation                     : "
    f"{'PASSED' if confidence_range_validation else 'FAILED'}"
)

logger.info(
    f"Test Observation Coverage                       : "
    f"{'PASSED' if coverage_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Category Coverage                    : "
    f"{'PASSED' if high_confidence_count_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

all_validations_passed = all([
    finite_confidence_validation,
    confidence_range_validation,
    coverage_validation,
    high_confidence_count_validation
])

logger.info("-" * 70)
logger.info("17.3 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"High-Confidence Predictions                     : "
    f"{high_confidence_count}"
)

logger.info(
    f"High-Confidence Percentage                      : "
    f"{high_confidence_percentage:.2f}%"
)

logger.info(
    f"High-Confidence Accuracy                        : "
    f"{high_confidence_accuracy:.4f}"
    if high_confidence_count > 0
    else "High-Confidence Accuracy                        : N/A"
)

logger.info(
    f"High-Confidence Validation                      : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.3 High-Confidence Predictions                : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

## 17.4 Low-Confidence Predictions

### Objective

Identify and analyze predictions where the optimized production model assigns relatively low probability to its predicted class.

Low-confidence predictions are defined as predictions with a maximum predicted probability of **< 0.60**.

This analysis evaluates:

- Number and percentage of low-confidence predictions
- Accuracy of low-confidence predictions
- Number of incorrect predictions captured by the low-confidence region
- Error-capture rate
- Mean and median confidence
- Mean confidence margin
- Distribution of low-confidence predictions across predicted classes
- Low-confidence performance by class
- Lowest-confidence predictions
- Relationship between low confidence and observed model errors

### Interpretation

Low confidence is treated as a **risk-monitoring signal**, not as a rejection rule or clinical decision threshold.

A low-confidence prediction indicates that the model has relatively weak probability separation between its competing classes. It does not by itself prove that the prediction is incorrect.

### Validation Principles

- Uses the final test-set probabilities generated in Section 17.1
- Uses the confidence analysis generated in Section 17.2
- Does not retrain or tune the production model
- Does not modify `X_test` or `y_test`
- Preserves the original test dataset
- Uses multiclass confidence and top-two probability margin
- Validates complete test-set coverage

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
from IPython.display import display


# ----------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------

LOW_CONFIDENCE_THRESHOLD = 0.60


logger.info("=" * 70)
logger.info("17.4 LOW-CONFIDENCE PREDICTIONS")
logger.info("=" * 70)


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "test_predictions": test_predictions,
    "y_test": y_test,
    "model_classes": model_classes,
    "confidence_analysis": confidence_analysis
}

for object_name, object_value in required_objects.items():
    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is not available."
        )

logger.info(
    "Probability Matrix Validation                  : PASSED"
)
logger.info(
    "Prediction Validation                          : PASSED"
)
logger.info(
    "Target Validation                              : PASSED"
)
logger.info(
    "Class Mapping Validation                        : PASSED"
)
logger.info(
    "Confidence Analysis Validation                 : PASSED"
)


# ----------------------------------------------------------------
# PREPARE ARRAYS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING CONFIDENCE DATA")
logger.info("-" * 70)

test_probabilities_array = np.asarray(
    test_probabilities
)

test_predictions_array = np.asarray(
    test_predictions
)

y_test_array = np.asarray(
    y_test
)

prediction_confidence = np.max(
    test_probabilities_array,
    axis=1
)

predicted_class_from_probability = model_classes[
    np.argmax(
        test_probabilities_array,
        axis=1
    )
]

prediction_correct = (
    predicted_class_from_probability == y_test_array
)


# ----------------------------------------------------------------
# CALCULATE CONFIDENCE MARGIN
# ----------------------------------------------------------------

sorted_probabilities = np.sort(
    test_probabilities_array,
    axis=1
)

second_highest_probability = (
    sorted_probabilities[:, -2]
)

confidence_margin = (
    prediction_confidence -
    second_highest_probability
)


# ----------------------------------------------------------------
# IDENTIFY LOW-CONFIDENCE PREDICTIONS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING LOW-CONFIDENCE PREDICTIONS")
logger.info("-" * 70)

low_confidence_mask = (
    prediction_confidence < LOW_CONFIDENCE_THRESHOLD
)

low_confidence_indices = np.where(
    low_confidence_mask
)[0]

total_test_samples = len(
    y_test_array
)

low_confidence_count = int(
    np.sum(low_confidence_mask)
)

low_confidence_percentage = (
    low_confidence_count /
    total_test_samples *
    100
)


# ----------------------------------------------------------------
# LOW-CONFIDENCE PERFORMANCE
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW-CONFIDENCE PERFORMANCE")
logger.info("-" * 70)

low_confidence_correct = int(
    np.sum(
        low_confidence_mask &
        prediction_correct
    )
)

low_confidence_incorrect = (
    low_confidence_count -
    low_confidence_correct
)

if low_confidence_count > 0:

    low_confidence_accuracy = (
        low_confidence_correct /
        low_confidence_count
    )

    low_confidence_values = (
        prediction_confidence[
            low_confidence_mask
        ]
    )

    low_confidence_margins = (
        confidence_margin[
            low_confidence_mask
        ]
    )

    low_confidence_mean = np.mean(
        low_confidence_values
    )

    low_confidence_median = np.median(
        low_confidence_values
    )

    low_confidence_mean_margin = np.mean(
        low_confidence_margins
    )

    low_confidence_median_margin = np.median(
        low_confidence_margins
    )

else:

    low_confidence_accuracy = np.nan
    low_confidence_values = np.array([])
    low_confidence_margins = np.array([])
    low_confidence_mean = np.nan
    low_confidence_median = np.nan
    low_confidence_mean_margin = np.nan
    low_confidence_median_margin = np.nan


logger.info(
    f"Low Confidence Threshold                       : "
    f"{LOW_CONFIDENCE_THRESHOLD:.2f}"
)

logger.info(
    f"Low-Confidence Predictions                     : "
    f"{low_confidence_count}"
)

logger.info(
    f"Low-Confidence Percentage                      : "
    f"{low_confidence_percentage:.2f}%"
)

logger.info(
    f"Low-Confidence Correct Predictions             : "
    f"{low_confidence_correct}"
)

logger.info(
    f"Low-Confidence Incorrect Predictions           : "
    f"{low_confidence_incorrect}"
)

logger.info(
    f"Low-Confidence Accuracy                        : "
    f"{low_confidence_accuracy:.4f}"
    if low_confidence_count > 0
    else
    "Low-Confidence Accuracy                        : N/A"
)

logger.info(
    f"Mean Low-Confidence Probability                : "
    f"{low_confidence_mean:.4f}"
    if low_confidence_count > 0
    else
    "Mean Low-Confidence Probability                : N/A"
)

logger.info(
    f"Median Low-Confidence Probability              : "
    f"{low_confidence_median:.4f}"
    if low_confidence_count > 0
    else
    "Median Low-Confidence Probability              : N/A"
)

logger.info(
    f"Mean Low-Confidence Margin                     : "
    f"{low_confidence_mean_margin:.4f}"
    if low_confidence_count > 0
    else
    "Mean Low-Confidence Margin                     : N/A"
)

logger.info(
    f"Median Low-Confidence Margin                   : "
    f"{low_confidence_median_margin:.4f}"
    if low_confidence_count > 0
    else
    "Median Low-Confidence Margin                   : N/A"
)


# ----------------------------------------------------------------
# ERROR-CAPTURE ANALYSIS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("ERROR-CAPTURE ANALYSIS")
logger.info("-" * 70)

total_errors = int(
    np.sum(~prediction_correct)
)

errors_captured_by_low_confidence = int(
    np.sum(
        low_confidence_mask &
        ~prediction_correct
    )
)

if total_errors > 0:

    error_capture_rate = (
        errors_captured_by_low_confidence /
        total_errors
    )

else:

    error_capture_rate = np.nan


logger.info(
    f"Total Test Errors                              : "
    f"{total_errors}"
)

logger.info(
    f"Errors Captured by Low Confidence              : "
    f"{errors_captured_by_low_confidence}"
)

logger.info(
    f"Error Capture Rate                             : "
    f"{error_capture_rate:.4f}"
    if total_errors > 0
    else
    "Error Capture Rate                             : N/A"
)


# ----------------------------------------------------------------
# LOW-CONFIDENCE CLASS DISTRIBUTION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW-CONFIDENCE CLASS DISTRIBUTION")
logger.info("-" * 70)

if low_confidence_count > 0:

    low_confidence_classes = (
        predicted_class_from_probability[
            low_confidence_mask
        ]
    )

    low_confidence_class_distribution = (
        pd.Series(low_confidence_classes)
        .value_counts()
        .sort_index()
        .rename_axis("Predicted_Class")
        .reset_index(name="Count")
    )

    low_confidence_class_distribution[
        "Percentage"
    ] = (
        low_confidence_class_distribution["Count"]
        / low_confidence_count
        * 100
    )

else:

    low_confidence_class_distribution = pd.DataFrame(
        columns=[
            "Predicted_Class",
            "Count",
            "Percentage"
        ]
    )

display(
    low_confidence_class_distribution
)


# ----------------------------------------------------------------
# LOW-CONFIDENCE PERFORMANCE BY CLASS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW-CONFIDENCE PERFORMANCE BY CLASS")
logger.info("-" * 70)

low_confidence_class_performance = []

for class_label in model_classes:

    class_mask = (
        low_confidence_mask &
        (
            predicted_class_from_probability
            == class_label
        )
    )

    class_count = int(
        np.sum(class_mask)
    )

    if class_count > 0:

        class_correct = int(
            np.sum(
                class_mask &
                prediction_correct
            )
        )

        class_accuracy = (
            class_correct /
            class_count
        )

        class_mean_confidence = np.mean(
            prediction_confidence[
                class_mask
            ]
        )

        class_mean_margin = np.mean(
            confidence_margin[
                class_mask
            ]
        )

    else:

        class_correct = 0
        class_accuracy = np.nan
        class_mean_confidence = np.nan
        class_mean_margin = np.nan

    low_confidence_class_performance.append({
        "Predicted_Class": class_label,
        "Low_Confidence_Count": class_count,
        "Correct_Count": class_correct,
        "Accuracy": class_accuracy,
        "Mean_Confidence": class_mean_confidence,
        "Mean_Margin": class_mean_margin
    })


low_confidence_class_performance = pd.DataFrame(
    low_confidence_class_performance
)

display(
    low_confidence_class_performance
)


# ----------------------------------------------------------------
# LOWEST-CONFIDENCE PREDICTIONS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOWEST-CONFIDENCE PREDICTIONS")
logger.info("-" * 70)

low_confidence_dataset = confidence_analysis[
    confidence_analysis[
        "Prediction_Confidence"
    ] < LOW_CONFIDENCE_THRESHOLD
].copy()

low_confidence_dataset = (
    low_confidence_dataset
    .sort_values(
        by=[
            "Prediction_Confidence",
            "Confidence_Margin"
        ],
        ascending=True
    )
    .reset_index(drop=True)
)

display(
    low_confidence_dataset[
        [
            "Sample_Index",
            "Actual_Class",
            "Predicted_Class",
            "Prediction_Confidence",
            "Second_Highest_Probability",
            "Confidence_Margin",
            "Prediction_Correct",
            "Confidence_Category"
        ]
    ].head(15)
)


# ----------------------------------------------------------------
# LOW-CONFIDENCE ERROR DETAILS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOW-CONFIDENCE ERROR DETAILS")
logger.info("-" * 70)

low_confidence_errors = low_confidence_dataset[
    low_confidence_dataset[
        "Prediction_Correct"
    ] == False
].copy()

if len(low_confidence_errors) > 0:

    logger.warning(
        f"Incorrect predictions within low-confidence "
        f"region: {len(low_confidence_errors)}"
    )

    display(
        low_confidence_errors[
            [
                "Sample_Index",
                "Actual_Class",
                "Predicted_Class",
                "Prediction_Confidence",
                "Second_Highest_Probability",
                "Confidence_Margin"
            ]
        ]
    )

else:

    logger.info(
        "Incorrect predictions within low-confidence region : 0"
    )


# ----------------------------------------------------------------
# CONFIDENCE REGION COMPARISON
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CONFIDENCE REGION COMPARISON")
logger.info("-" * 70)

confidence_region_labels = np.where(
    prediction_confidence < LOW_CONFIDENCE_THRESHOLD,
    "Low",
    np.where(
        prediction_confidence < 0.80,
        "Moderate",
        "High"
    )
)

confidence_region_comparison = []

for region in ["Low", "Moderate", "High"]:

    region_mask = (
        confidence_region_labels == region
    )

    region_count = int(
        np.sum(region_mask)
    )

    region_correct = int(
        np.sum(
            region_mask &
            prediction_correct
        )
    )

    region_incorrect = (
        region_count -
        region_correct
    )

    region_accuracy = (
        region_correct /
        region_count
        if region_count > 0
        else np.nan
    )

    region_mean_confidence = (
        np.mean(
            prediction_confidence[
                region_mask
            ]
        )
        if region_count > 0
        else np.nan
    )

    region_mean_margin = (
        np.mean(
            confidence_margin[
                region_mask
            ]
        )
        if region_count > 0
        else np.nan
    )

    confidence_region_comparison.append({
        "Confidence_Region": region,
        "Count": region_count,
        "Correct": region_correct,
        "Incorrect": region_incorrect,
        "Accuracy": region_accuracy,
        "Mean_Confidence": region_mean_confidence,
        "Mean_Margin": region_mean_margin
    })

confidence_region_comparison = pd.DataFrame(
    confidence_region_comparison
)

display(
    confidence_region_comparison
)


# ----------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

finite_confidence_validation = np.all(
    np.isfinite(prediction_confidence)
)

confidence_range_validation = (
    np.all(prediction_confidence >= 0) and
    np.all(prediction_confidence <= 1)
)

margin_validation = (
    np.all(confidence_margin >= 0) and
    np.all(confidence_margin <= 1)
)

prediction_consistency_validation = np.array_equal(
    predicted_class_from_probability,
    test_predictions_array
)

coverage_validation = (
    len(prediction_confidence)
    == total_test_samples
)

confidence_region_coverage_validation = (
    np.sum(
        [
            np.sum(
                confidence_region_labels == region
            )
            for region in ["Low", "Moderate", "High"]
        ]
    )
    == total_test_samples
)

logger.info(
    f"Finite Confidence Validation                  : "
    f"{'PASSED' if finite_confidence_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Range Validation                   : "
    f"{'PASSED' if confidence_range_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Margin Validation                  : "
    f"{'PASSED' if margin_validation else 'FAILED'}"
)

logger.info(
    f"Prediction Consistency Validation             : "
    f"{'PASSED' if prediction_consistency_validation else 'FAILED'}"
)

logger.info(
    f"Test Observation Coverage                     : "
    f"{'PASSED' if coverage_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Region Coverage                    : "
    f"{'PASSED' if confidence_region_coverage_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

all_validations_passed = all([
    finite_confidence_validation,
    confidence_range_validation,
    margin_validation,
    prediction_consistency_validation,
    coverage_validation,
    confidence_region_coverage_validation,
    original_test_preservation_validation
])

logger.info("-" * 70)
logger.info("17.4 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"Low-Confidence Predictions                     : "
    f"{low_confidence_count}"
)

logger.info(
    f"Low-Confidence Percentage                      : "
    f"{low_confidence_percentage:.2f}%"
)

logger.info(
    f"Low-Confidence Accuracy                        : "
    f"{low_confidence_accuracy:.4f}"
    if low_confidence_count > 0
    else
    "Low-Confidence Accuracy                        : N/A"
)

logger.info(
    f"Errors Captured by Low Confidence              : "
    f"{errors_captured_by_low_confidence}/{total_errors}"
)

logger.info(
    f"Error Capture Rate                             : "
    f"{error_capture_rate:.4f}"
    if total_errors > 0
    else
    "Error Capture Rate                             : N/A"
)

logger.info(
    f"Low-Confidence Validation                      : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.4 Low-Confidence Predictions                : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

## 17.5 Ambiguous Predictions

### Objective

Analyze predictions where the model has a small probability margin between its two most likely classes.

The confidence margin is defined as:

**Confidence Margin = Top-1 Probability − Top-2 Probability**

A smaller margin indicates greater competition between the two leading predicted classes.

For this analysis, predictions with a confidence margin of **< 0.10** are classified as ambiguous.

This section evaluates:

- Number and percentage of ambiguous predictions
- Accuracy of ambiguous predictions
- Number of observed errors captured by the ambiguity threshold
- Ambiguity error-capture rate
- Mean top-1 and top-2 probabilities
- Mean confidence margin
- Most ambiguous predictions
- Ambiguous predictions by predicted class
- Correct versus incorrect ambiguity
- Class 2 ↔ Class 4 decision-boundary behavior
- Relationship between ambiguity and prediction confidence

### Interpretation

A small probability margin does not prove that a prediction is incorrect.

It indicates that the model has relatively weak separation between its two leading candidate classes.

The observed Class 2 ↔ Class 4 confusion is specifically examined because this was the primary model failure pattern identified during Error Analysis.

### Validation Principles

- Uses final test-set probabilities from Section 17.1
- Uses confidence analysis from Section 17.2
- Does not retrain or tune the model
- Does not modify the original test data
- Uses multiclass top-1 and top-2 probabilities
- Validates complete test-set coverage
- Treats the ambiguity threshold as a descriptive monitoring threshold

````python

# Required Libraries
import logging
import numpy as np
import pandas as pd
from IPython.display import display


# ----------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------

AMBIGUITY_MARGIN_THRESHOLD = 0.10


logger.info("=" * 70)
logger.info("17.5 AMBIGUOUS PREDICTIONS")
logger.info("=" * 70)


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "test_predictions": test_predictions,
    "y_test": y_test,
    "model_classes": model_classes,
    "confidence_analysis": confidence_analysis
}

for object_name, object_value in required_objects.items():

    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is not available."
        )

logger.info(
    "Probability Matrix Validation                  : PASSED"
)

logger.info(
    "Prediction Validation                          : PASSED"
)

logger.info(
    "Target Validation                              : PASSED"
)

logger.info(
    "Class Mapping Validation                        : PASSED"
)

logger.info(
    "Confidence Analysis Validation                 : PASSED"
)


# ----------------------------------------------------------------
# PREPARE ARRAYS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING PROBABILITY DATA")
logger.info("-" * 70)

test_probabilities_array = np.asarray(
    test_probabilities
)

test_predictions_array = np.asarray(
    test_predictions
)

y_test_array = np.asarray(
    y_test
)

total_test_samples = len(
    y_test_array
)

if test_probabilities_array.ndim != 2:
    raise ValueError(
        "test_probabilities must be a 2-dimensional matrix."
    )

if test_probabilities_array.shape[0] != total_test_samples:
    raise ValueError(
        "Probability matrix and test target sizes do not match."
    )

if test_probabilities_array.shape[1] != len(model_classes):
    raise ValueError(
        "Probability matrix and model class count do not match."
    )


# ----------------------------------------------------------------
# CALCULATE TOP-1 / TOP-2 PROBABILITIES
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING TOP-1 AND TOP-2 PROBABILITIES")
logger.info("-" * 70)

sorted_probability_indices = np.argsort(
    test_probabilities_array,
    axis=1
)

top_1_indices = (
    sorted_probability_indices[:, -1]
)

top_2_indices = (
    sorted_probability_indices[:, -2]
)

top_1_probability = (
    test_probabilities_array[
        np.arange(total_test_samples),
        top_1_indices
    ]
)

top_2_probability = (
    test_probabilities_array[
        np.arange(total_test_samples),
        top_2_indices
    ]
)

top_1_class = (
    model_classes[top_1_indices]
)

top_2_class = (
    model_classes[top_2_indices]
)

confidence_margin = (
    top_1_probability -
    top_2_probability
)

prediction_correct = (
    top_1_class == y_test_array
)


# ----------------------------------------------------------------
# IDENTIFY AMBIGUOUS PREDICTIONS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("IDENTIFYING AMBIGUOUS PREDICTIONS")
logger.info("-" * 70)

ambiguous_mask = (
    confidence_margin <
    AMBIGUITY_MARGIN_THRESHOLD
)

ambiguous_count = int(
    np.sum(ambiguous_mask)
)

ambiguous_percentage = (
    ambiguous_count /
    total_test_samples *
    100
)


# ----------------------------------------------------------------
# AMBIGUOUS PREDICTION PERFORMANCE
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("AMBIGUOUS PREDICTION PERFORMANCE")
logger.info("-" * 70)

ambiguous_correct = int(
    np.sum(
        ambiguous_mask &
        prediction_correct
    )
)

ambiguous_incorrect = (
    ambiguous_count -
    ambiguous_correct
)

if ambiguous_count > 0:

    ambiguous_accuracy = (
        ambiguous_correct /
        ambiguous_count
    )

    ambiguous_top1_mean = np.mean(
        top_1_probability[
            ambiguous_mask
        ]
    )

    ambiguous_top2_mean = np.mean(
        top_2_probability[
            ambiguous_mask
        ]
    )

    ambiguous_margin_mean = np.mean(
        confidence_margin[
            ambiguous_mask
        ]
    )

    ambiguous_margin_median = np.median(
        confidence_margin[
            ambiguous_mask
        ]
    )

else:

    ambiguous_accuracy = np.nan
    ambiguous_top1_mean = np.nan
    ambiguous_top2_mean = np.nan
    ambiguous_margin_mean = np.nan
    ambiguous_margin_median = np.nan


logger.info(
    f"Ambiguity Margin Threshold                      : "
    f"{AMBIGUITY_MARGIN_THRESHOLD:.2f}"
)

logger.info(
    f"Ambiguous Predictions                           : "
    f"{ambiguous_count}"
)

logger.info(
    f"Ambiguous Percentage                            : "
    f"{ambiguous_percentage:.2f}%"
)

logger.info(
    f"Ambiguous Correct Predictions                   : "
    f"{ambiguous_correct}"
)

logger.info(
    f"Ambiguous Incorrect Predictions                 : "
    f"{ambiguous_incorrect}"
)

logger.info(
    f"Ambiguous Accuracy                              : "
    f"{ambiguous_accuracy:.4f}"
    if ambiguous_count > 0
    else
    "Ambiguous Accuracy                              : N/A"
)

logger.info(
    f"Mean Top-1 Probability                          : "
    f"{ambiguous_top1_mean:.4f}"
    if ambiguous_count > 0
    else
    "Mean Top-1 Probability                          : N/A"
)

logger.info(
    f"Mean Top-2 Probability                          : "
    f"{ambiguous_top2_mean:.4f}"
    if ambiguous_count > 0
    else
    "Mean Top-2 Probability                          : N/A"
)

logger.info(
    f"Mean Confidence Margin                         : "
    f"{ambiguous_margin_mean:.4f}"
    if ambiguous_count > 0
    else
    "Mean Confidence Margin                         : N/A"
)

logger.info(
    f"Median Confidence Margin                       : "
    f"{ambiguous_margin_median:.4f}"
    if ambiguous_count > 0
    else
    "Median Confidence Margin                       : N/A"
)


# ----------------------------------------------------------------
# ERROR-CAPTURE ANALYSIS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("AMBIGUITY ERROR-CAPTURE ANALYSIS")
logger.info("-" * 70)

total_errors = int(
    np.sum(~prediction_correct)
)

errors_captured_by_ambiguity = int(
    np.sum(
        ambiguous_mask &
        ~prediction_correct
    )
)

if total_errors > 0:

    ambiguity_error_capture_rate = (
        errors_captured_by_ambiguity /
        total_errors
    )

else:

    ambiguity_error_capture_rate = np.nan


logger.info(
    f"Total Test Errors                              : "
    f"{total_errors}"
)

logger.info(
    f"Errors Captured by Ambiguity                   : "
    f"{errors_captured_by_ambiguity}"
)

logger.info(
    f"Ambiguity Error Capture Rate                   : "
    f"{ambiguity_error_capture_rate:.4f}"
    if total_errors > 0
    else
    "Ambiguity Error Capture Rate                   : N/A"
)


# ----------------------------------------------------------------
# AMBIGUOUS PREDICTION DATASET
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING AMBIGUOUS PREDICTION DATASET")
logger.info("-" * 70)

ambiguous_prediction_dataset = pd.DataFrame({
    "Sample_Index": X_test.index,
    "Actual_Class": y_test_array,
    "Top1_Class": top_1_class,
    "Top1_Probability": top_1_probability,
    "Top2_Class": top_2_class,
    "Top2_Probability": top_2_probability,
    "Confidence_Margin": confidence_margin,
    "Prediction_Correct": prediction_correct
})

ambiguous_prediction_dataset = (
    ambiguous_prediction_dataset[
        ambiguous_prediction_dataset[
            "Confidence_Margin"
        ] < AMBIGUITY_MARGIN_THRESHOLD
    ]
    .sort_values(
        by=[
            "Confidence_Margin",
            "Top1_Probability"
        ],
        ascending=[
            True,
            True
        ]
    )
    .reset_index(drop=True)
)

display(
    ambiguous_prediction_dataset
)


# ----------------------------------------------------------------
# AMBIGUOUS PREDICTIONS BY PREDICTED CLASS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("AMBIGUOUS PREDICTIONS BY PREDICTED CLASS")
logger.info("-" * 70)

ambiguous_class_performance = []

for class_label in model_classes:

    class_mask = (
        ambiguous_mask &
        (
            top_1_class == class_label
        )
    )

    class_count = int(
        np.sum(class_mask)
    )

    if class_count > 0:

        class_correct = int(
            np.sum(
                class_mask &
                prediction_correct
            )
        )

        class_incorrect = (
            class_count -
            class_correct
        )

        class_accuracy = (
            class_correct /
            class_count
        )

        class_mean_margin = np.mean(
            confidence_margin[
                class_mask
            ]
        )

        class_mean_top1 = np.mean(
            top_1_probability[
                class_mask
            ]
        )

        class_mean_top2 = np.mean(
            top_2_probability[
                class_mask
            ]
        )

    else:

        class_correct = 0
        class_incorrect = 0
        class_accuracy = np.nan
        class_mean_margin = np.nan
        class_mean_top1 = np.nan
        class_mean_top2 = np.nan

    ambiguous_class_performance.append({
        "Predicted_Class": class_label,
        "Ambiguous_Count": class_count,
        "Correct_Count": class_correct,
        "Incorrect_Count": class_incorrect,
        "Accuracy": class_accuracy,
        "Mean_Top1_Probability": class_mean_top1,
        "Mean_Top2_Probability": class_mean_top2,
        "Mean_Confidence_Margin": class_mean_margin
    })


ambiguous_class_performance = pd.DataFrame(
    ambiguous_class_performance
)

display(
    ambiguous_class_performance
)


# ----------------------------------------------------------------
# TOP-TWO CLASS PAIR ANALYSIS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("TOP-TWO CLASS PAIR ANALYSIS")
logger.info("-" * 70)

if ambiguous_count > 0:

    ambiguous_pair_dataset = (
        ambiguous_prediction_dataset.copy()
    )

    ambiguous_pair_dataset[
        "Class_Pair"
    ] = (
        ambiguous_pair_dataset[
            "Top1_Class"
        ].astype(str)
        + " vs "
        + ambiguous_pair_dataset[
            "Top2_Class"
        ].astype(str)
    )

    class_pair_summary = (
        ambiguous_pair_dataset
        .groupby(
            "Class_Pair",
            as_index=False
        )
        .agg(
            Count=(
                "Sample_Index",
                "count"
            ),
            Mean_Top1_Probability=(
                "Top1_Probability",
                "mean"
            ),
            Mean_Top2_Probability=(
                "Top2_Probability",
                "mean"
            ),
            Mean_Confidence_Margin=(
                "Confidence_Margin",
                "mean"
            ),
            Errors=(
                "Prediction_Correct",
                lambda x: int((~x).sum())
            )
        )
        .sort_values(
            by="Count",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    class_pair_summary = pd.DataFrame(
        columns=[
            "Class_Pair",
            "Count",
            "Mean_Top1_Probability",
            "Mean_Top2_Probability",
            "Mean_Confidence_Margin",
            "Errors"
        ]
    )

display(
    class_pair_summary
)


# ----------------------------------------------------------------
# CLASS 2 ↔ CLASS 4 DECISION BOUNDARY ANALYSIS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS 2 ↔ CLASS 4 DECISION BOUNDARY ANALYSIS")
logger.info("-" * 70)

class_2_4_mask = (
    (
        (top_1_class == 2) &
        (top_2_class == 4)
    )
    |
    (
        (top_1_class == 4) &
        (top_2_class == 2)
    )
)

class_2_4_count = int(
    np.sum(class_2_4_mask)
)

class_2_4_ambiguous_count = int(
    np.sum(
        class_2_4_mask &
        ambiguous_mask
    )
)

class_2_4_errors = int(
    np.sum(
        class_2_4_mask &
        ~prediction_correct
    )
)

if class_2_4_count > 0:

    class_2_4_mean_margin = np.mean(
        confidence_margin[
            class_2_4_mask
        ]
    )

    class_2_4_mean_top1 = np.mean(
        top_1_probability[
            class_2_4_mask
        ]
    )

    class_2_4_mean_top2 = np.mean(
        top_2_probability[
            class_2_4_mask
        ]
    )

else:

    class_2_4_mean_margin = np.nan
    class_2_4_mean_top1 = np.nan
    class_2_4_mean_top2 = np.nan


logger.info(
    f"Class 2 ↔ Class 4 Competing Predictions          : "
    f"{class_2_4_count}"
)

logger.info(
    f"Class 2 ↔ Class 4 Ambiguous Predictions          : "
    f"{class_2_4_ambiguous_count}"
)

logger.info(
    f"Class 2 ↔ Class 4 Observed Errors                : "
    f"{class_2_4_errors}"
)

logger.info(
    f"Class 2 ↔ Class 4 Mean Top-1 Probability         : "
    f"{class_2_4_mean_top1:.4f}"
    if class_2_4_count > 0
    else
    "Class 2 ↔ Class 4 Mean Top-1 Probability         : N/A"
)

logger.info(
    f"Class 2 ↔ Class 4 Mean Top-2 Probability         : "
    f"{class_2_4_mean_top2:.4f}"
    if class_2_4_count > 0
    else
    "Class 2 ↔ Class 4 Mean Top-2 Probability         : N/A"
)

logger.info(
    f"Class 2 ↔ Class 4 Mean Confidence Margin         : "
    f"{class_2_4_mean_margin:.4f}"
    if class_2_4_count > 0
    else
    "Class 2 ↔ Class 4 Mean Confidence Margin         : N/A"
)

if class_2_4_count > 0:

    class_2_4_boundary_dataset = (
        ambiguous_prediction_dataset[
            class_2_4_mask[
                ambiguous_mask
            ]
        ]
        if ambiguous_count > 0
        else pd.DataFrame()
    )

else:

    class_2_4_boundary_dataset = pd.DataFrame()


if len(class_2_4_boundary_dataset) > 0:

    display(
        class_2_4_boundary_dataset[
            [
                "Sample_Index",
                "Actual_Class",
                "Top1_Class",
                "Top1_Probability",
                "Top2_Class",
                "Top2_Probability",
                "Confidence_Margin",
                "Prediction_Correct"
            ]
        ]
    )


# ----------------------------------------------------------------
# CORRECT VS INCORRECT AMBIGUITY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CORRECT VS INCORRECT AMBIGUITY")
logger.info("-" * 70)

ambiguity_group_summary = []

for group_name, group_mask in [
    ("Correct", prediction_correct),
    ("Incorrect", ~prediction_correct)
]:

    group_count = int(
        np.sum(group_mask)
    )

    if group_count > 0:

        group_ambiguous_count = int(
            np.sum(
                group_mask &
                ambiguous_mask
            )
        )

        group_ambiguous_percentage = (
            group_ambiguous_count /
            group_count *
            100
        )

        group_mean_margin = np.mean(
            confidence_margin[
                group_mask
            ]
        )

    else:

        group_ambiguous_count = 0
        group_ambiguous_percentage = np.nan
        group_mean_margin = np.nan

    ambiguity_group_summary.append({
        "Prediction_Group": group_name,
        "Count": group_count,
        "Ambiguous_Count": group_ambiguous_count,
        "Ambiguous_Percentage": group_ambiguous_percentage,
        "Mean_Confidence_Margin": group_mean_margin
    })


ambiguity_group_summary = pd.DataFrame(
    ambiguity_group_summary
)

display(
    ambiguity_group_summary
)


# ----------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

finite_probability_validation = np.all(
    np.isfinite(
        test_probabilities_array
    )
)

probability_range_validation = (
    np.all(
        test_probabilities_array >= 0
    )
    and
    np.all(
        test_probabilities_array <= 1
    )
)

margin_range_validation = (
    np.all(confidence_margin >= 0)
    and
    np.all(confidence_margin <= 1)
)

top_probability_order_validation = np.all(
    top_1_probability >= top_2_probability
)

prediction_consistency_validation = (
    np.array_equal(
        top_1_class,
        test_predictions_array
    )
)

coverage_validation = (
    len(top_1_probability)
    == total_test_samples
)

ambiguity_coverage_validation = (
    ambiguous_count
    +
    int(
        np.sum(~ambiguous_mask)
    )
    == total_test_samples
)


logger.info(
    f"Finite Probability Validation                 : "
    f"{'PASSED' if finite_probability_validation else 'FAILED'}"
)

logger.info(
    f"Probability Range Validation                  : "
    f"{'PASSED' if probability_range_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Margin Range Validation            : "
    f"{'PASSED' if margin_range_validation else 'FAILED'}"
)

logger.info(
    f"Top-1 / Top-2 Ordering Validation              : "
    f"{'PASSED' if top_probability_order_validation else 'FAILED'}"
)

logger.info(
    f"Prediction Consistency Validation             : "
    f"{'PASSED' if prediction_consistency_validation else 'FAILED'}"
)

logger.info(
    f"Test Observation Coverage                     : "
    f"{'PASSED' if coverage_validation else 'FAILED'}"
)

logger.info(
    f"Ambiguity Region Coverage                     : "
    f"{'PASSED' if ambiguity_coverage_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

all_validations_passed = all([
    finite_probability_validation,
    probability_range_validation,
    margin_range_validation,
    top_probability_order_validation,
    prediction_consistency_validation,
    coverage_validation,
    ambiguity_coverage_validation,
    original_test_preservation_validation
])

logger.info("-" * 70)
logger.info("17.5 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"Ambiguous Predictions                           : "
    f"{ambiguous_count}"
)

logger.info(
    f"Ambiguous Percentage                            : "
    f"{ambiguous_percentage:.2f}%"
)

logger.info(
    f"Ambiguous Accuracy                              : "
    f"{ambiguous_accuracy:.4f}"
    if ambiguous_count > 0
    else
    "Ambiguous Accuracy                              : N/A"
)

logger.info(
    f"Errors Captured by Ambiguity                   : "
    f"{errors_captured_by_ambiguity}/{total_errors}"
)

logger.info(
    f"Ambiguity Error Capture Rate                   : "
    f"{ambiguity_error_capture_rate:.4f}"
    if total_errors > 0
    else
    "Ambiguity Error Capture Rate                   : N/A"
)

logger.info(
    f"Class 2 ↔ Class 4 Competing Predictions          : "
    f"{class_2_4_count}"
)

logger.info(
    f"Class 2 ↔ Class 4 Observed Errors                : "
    f"{class_2_4_errors}"
)

logger.info(
    f"Ambiguity Validation                            : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.5 Ambiguous Predictions                      : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

## 17.6 Calibration Analysis

### Objective

Evaluate whether the probability outputs produced by the optimized Random Forest are statistically aligned with observed prediction frequencies.

Calibration measures whether predicted probabilities correspond approximately to empirical outcome frequencies.

For example, among predictions assigned approximately 0.80 probability to a class, a well-calibrated model should be correct at approximately the corresponding empirical frequency.

Because this project is a multiclass classification problem with six target classes, calibration is evaluated using a multiclass one-vs-rest framework.

### Calibration Principles

- Calibration is different from classification accuracy.
- A model can have high accuracy while still producing poorly calibrated probabilities.
- Probability confidence does not automatically imply probability calibration.
- Calibration assessment must avoid using the final test set to tune or recalibrate the production model.
- The final test set remains an evaluation set.
- No model parameters are changed in this section.

### Methodology

The analysis will compare:

1. Predicted probability
2. Observed class frequency
3. Calibration deviation
4. Per-class calibration behavior
5. Overall calibration characteristics

Calibration quality will be interpreted descriptively rather than as a clinical probability guarantee.

### Important Limitation

The available dataset contains only 366 observations, with 74 observations in the final test set.

Therefore, calibration estimates—especially for minority classes—may have substantial sampling uncertainty.

The calibration results should be treated as evidence about this dataset and model, not as universal clinical probability validation.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
from IPython.display import display


logger.info("=" * 70)
logger.info("17.6 CALIBRATION ANALYSIS")
logger.info("=" * 70)


# ----------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------

CALIBRATION_BINS = 5


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "y_test": y_test,
    "model_classes": model_classes,
    "test_predictions": test_predictions
}

for object_name, object_value in required_objects.items():

    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is not available."
        )

logger.info(
    "Probability Matrix Validation                  : PASSED"
)

logger.info(
    "Target Validation                              : PASSED"
)

logger.info(
    "Class Mapping Validation                        : PASSED"
)

logger.info(
    "Prediction Validation                          : PASSED"
)


# ----------------------------------------------------------------
# PREPARE DATA
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING CALIBRATION DATA")
logger.info("-" * 70)

test_probabilities_array = np.asarray(
    test_probabilities,
    dtype=float
)

y_test_array = np.asarray(
    y_test
)

test_predictions_array = np.asarray(
    test_predictions
)

model_classes_array = np.asarray(
    model_classes
)

total_test_samples = len(
    y_test_array
)

n_classes = len(
    model_classes_array
)


if test_probabilities_array.shape != (
    total_test_samples,
    n_classes
):

    raise ValueError(
        "Probability matrix shape is inconsistent "
        "with test observations and model classes."
    )


# ----------------------------------------------------------------
# BASIC PROBABILITY VALIDATION
# ----------------------------------------------------------------

finite_probability_validation = np.all(
    np.isfinite(
        test_probabilities_array
    )
)

probability_range_validation = (
    np.all(
        test_probabilities_array >= 0
    )
    and
    np.all(
        test_probabilities_array <= 1
    )
)

probability_sum_validation = np.allclose(
    test_probabilities_array.sum(axis=1),
    1.0,
    atol=1e-6
)


logger.info(
    f"Finite Probability Validation                  : "
    f"{'PASSED' if finite_probability_validation else 'FAILED'}"
)

logger.info(
    f"Probability Range Validation                   : "
    f"{'PASSED' if probability_range_validation else 'FAILED'}"
)

logger.info(
    f"Probability Sum-to-One Validation              : "
    f"{'PASSED' if probability_sum_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# MULTICLASS ONE-VS-REST CALIBRATION DATA
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING ONE-VS-REST CALIBRATION")
logger.info("-" * 70)

calibration_records = []


for class_position, class_label in enumerate(
    model_classes_array
):

    class_probabilities = (
        test_probabilities_array[
            :,
            class_position
        ]
    )

    actual_class_indicator = (
        y_test_array == class_label
    ).astype(int)


    # ------------------------------------------------------------
    # BIN PROBABILITIES
    # ------------------------------------------------------------

    bin_edges = np.linspace(
        0.0,
        1.0,
        CALIBRATION_BINS + 1
    )

    # Include probability=1.0 in the final bin.
    bin_indices = np.digitize(
        class_probabilities,
        bin_edges[1:-1],
        right=False
    )


    for bin_number in range(
        CALIBRATION_BINS
    ):

        bin_mask = (
            bin_indices == bin_number
        )

        bin_count = int(
            np.sum(bin_mask)
        )

        if bin_count == 0:
            continue


        mean_predicted_probability = np.mean(
            class_probabilities[
                bin_mask
            ]
        )

        observed_frequency = np.mean(
            actual_class_indicator[
                bin_mask
            ]
        )

        absolute_calibration_error = abs(
            mean_predicted_probability
            -
            observed_frequency
        )

        calibration_records.append({

            "Class": class_label,

            "Bin": bin_number + 1,

            "Count": bin_count,

            "Mean_Predicted_Probability":
                mean_predicted_probability,

            "Observed_Frequency":
                observed_frequency,

            "Absolute_Calibration_Error":
                absolute_calibration_error
        })


calibration_bin_results = pd.DataFrame(
    calibration_records
)


# ----------------------------------------------------------------
# DISPLAY CALIBRATION BIN RESULTS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALIBRATION BIN RESULTS")
logger.info("-" * 70)

display(
    calibration_bin_results
)


# ----------------------------------------------------------------
# PER-CLASS CALIBRATION SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PER-CLASS CALIBRATION SUMMARY")
logger.info("-" * 70)

per_class_calibration = []

for class_label in model_classes_array:

    class_results = calibration_bin_results[
        calibration_bin_results["Class"]
        == class_label
    ]


    if len(class_results) > 0:

        weighted_calibration_error = np.average(
            class_results[
                "Absolute_Calibration_Error"
            ],
            weights=class_results[
                "Count"
            ]
        )

        maximum_calibration_error = (
            class_results[
                "Absolute_Calibration_Error"
            ].max()
        )

        total_class_samples = int(
            class_results["Count"].sum()
        )

    else:

        weighted_calibration_error = np.nan
        maximum_calibration_error = np.nan
        total_class_samples = 0


    per_class_calibration.append({

        "Class": class_label,

        "Samples_Analyzed":
            total_class_samples,

        "Weighted_Calibration_Error":
            weighted_calibration_error,

        "Maximum_Calibration_Error":
            maximum_calibration_error
    })


per_class_calibration = pd.DataFrame(
    per_class_calibration
)

display(
    per_class_calibration
)


# ----------------------------------------------------------------
# OVERALL CALIBRATION SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("OVERALL CALIBRATION SUMMARY")
logger.info("-" * 70)

if len(calibration_bin_results) > 0:

    overall_calibration_error = np.average(
        calibration_bin_results[
            "Absolute_Calibration_Error"
        ],
        weights=calibration_bin_results[
            "Count"
        ]
    )

    maximum_calibration_error = (
        calibration_bin_results[
            "Absolute_Calibration_Error"
        ].max()
    )

    total_calibration_observations = int(
        calibration_bin_results[
            "Count"
        ].sum()
    )

else:

    overall_calibration_error = np.nan
    maximum_calibration_error = np.nan
    total_calibration_observations = 0


logger.info(
    f"Calibration Bins                                 : "
    f"{CALIBRATION_BINS}"
)

logger.info(
    f"Overall Weighted Calibration Error             : "
    f"{overall_calibration_error:.4f}"
)

logger.info(
    f"Maximum Calibration Error                      : "
    f"{maximum_calibration_error:.4f}"
)

logger.info(
    f"Calibration Observations                       : "
    f"{total_calibration_observations}"
)


# ----------------------------------------------------------------
# CALIBRATION ERROR INTERPRETATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALIBRATION ERROR INTERPRETATION")
logger.info("-" * 70)

if np.isfinite(overall_calibration_error):

    if overall_calibration_error < 0.05:

        calibration_status = (
            "LOW OBSERVED CALIBRATION ERROR"
        )

    elif overall_calibration_error < 0.10:

        calibration_status = (
            "MODERATE OBSERVED CALIBRATION ERROR"
        )

    else:

        calibration_status = (
            "HIGH OBSERVED CALIBRATION ERROR"
        )

else:

    calibration_status = (
        "CALIBRATION ERROR UNAVAILABLE"
    )


logger.info(
    f"Calibration Status                             : "
    f"{calibration_status}"
)


# ----------------------------------------------------------------
# PROBABILITY VS OBSERVED FREQUENCY GAP
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBABILITY VS OBSERVED FREQUENCY")
logger.info("-" * 70)

calibration_bin_results[
    "Calibration_Direction"
] = np.where(
    calibration_bin_results[
        "Mean_Predicted_Probability"
    ]
    >
    calibration_bin_results[
        "Observed_Frequency"
    ],
    "Overconfident",
    np.where(
        calibration_bin_results[
            "Mean_Predicted_Probability"
        ]
        <
        calibration_bin_results[
            "Observed_Frequency"
        ],
        "Underconfident",
        "Aligned"
    )
)

calibration_direction_summary = (
    calibration_bin_results[
        "Calibration_Direction"
    ]
    .value_counts()
    .rename_axis(
        "Calibration_Direction"
    )
    .reset_index(
        name="Count"
    )
)

display(
    calibration_direction_summary
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# COVERAGE VALIDATION
# ----------------------------------------------------------------

calibration_coverage_validation = (
    total_calibration_observations
    == total_test_samples * n_classes
)


logger.info(
    f"Calibration Coverage Validation                : "
    f"{'PASSED' if calibration_coverage_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------

all_validations_passed = all([
    finite_probability_validation,
    probability_range_validation,
    probability_sum_validation,
    calibration_coverage_validation,
    original_test_preservation_validation
])


logger.info("-" * 70)
logger.info("17.6 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"Overall Calibration Error                      : "
    f"{overall_calibration_error:.4f}"
)

logger.info(
    f"Maximum Calibration Error                     : "
    f"{maximum_calibration_error:.4f}"
)

logger.info(
    f"Calibration Status                             : "
    f"{calibration_status}"
)

logger.info(
    f"Calibration Validation                         : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.6 Calibration Analysis                     : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

## 17.7 Calibration Curve

### Objective

Visualize the relationship between predicted class probabilities and observed class frequencies for the optimized Random Forest.

A perfectly calibrated classifier would approximately follow the diagonal reference line:

- Predicted probability = 0.20 → observed frequency ≈ 0.20
- Predicted probability = 0.50 → observed frequency ≈ 0.50
- Predicted probability = 0.80 → observed frequency ≈ 0.80

Because this is a six-class classification problem, calibration is evaluated using a one-vs-rest approach for each class.

### Interpretation

- Points close to the diagonal indicate better probability calibration.
- Points above the diagonal indicate underconfident predictions.
- Points below the diagonal indicate overconfident predictions.
- Large deviations should be interpreted cautiously when bins contain very few observations.

### Important Limitation

The final test set contains only 74 observations.

Therefore, some calibration bins—especially for minority classes—contain very few samples. The calibration curves are consequently descriptive and should not be interpreted as universal clinical probability calibration.

No model parameters are modified in this section.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


logger.info("=" * 70)
logger.info("17.7 CALIBRATION CURVE")
logger.info("=" * 70)


# ----------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------

CALIBRATION_CURVE_BINS = 5


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "y_test": y_test,
    "model_classes": model_classes
}

for object_name, object_value in required_objects.items():

    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is not available."
        )

logger.info(
    "Probability Matrix Validation                  : PASSED"
)

logger.info(
    "Target Validation                              : PASSED"
)

logger.info(
    "Class Mapping Validation                        : PASSED"
)


# ----------------------------------------------------------------
# PREPARE ARRAYS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PREPARING CALIBRATION CURVE DATA")
logger.info("-" * 70)

probabilities = np.asarray(
    test_probabilities,
    dtype=float
)

targets = np.asarray(
    y_test
)

classes = np.asarray(
    model_classes
)


n_samples = len(targets)
n_classes = len(classes)


if probabilities.shape != (
    n_samples,
    n_classes
):

    raise ValueError(
        "Probability matrix shape does not match "
        "test observations and model classes."
    )


logger.info(
    f"Test Samples                                  : "
    f"{n_samples}"
)

logger.info(
    f"Number of Classes                             : "
    f"{n_classes}"
)

logger.info(
    f"Calibration Curve Bins                        : "
    f"{CALIBRATION_CURVE_BINS}"
)


# ----------------------------------------------------------------
# CALCULATE ONE-VS-REST CALIBRATION CURVES
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING ONE-VS-REST CALIBRATION CURVES")
logger.info("-" * 70)

curve_records = []


for class_position, class_label in enumerate(classes):

    class_probability = probabilities[
        :,
        class_position
    ]

    actual_binary = (
        targets == class_label
    ).astype(int)


    # ------------------------------------------------------------
    # EQUAL-WIDTH BINNING
    # ------------------------------------------------------------

    bin_edges = np.linspace(
        0.0,
        1.0,
        CALIBRATION_CURVE_BINS + 1
    )

    bin_indices = np.digitize(
        class_probability,
        bin_edges[1:-1],
        right=False
    )


    for bin_number in range(
        CALIBRATION_CURVE_BINS
    ):

        bin_mask = (
            bin_indices == bin_number
        )

        bin_count = int(
            np.sum(bin_mask)
        )

        if bin_count == 0:
            continue


        mean_probability = np.mean(
            class_probability[
                bin_mask
            ]
        )

        observed_frequency = np.mean(
            actual_binary[
                bin_mask
            ]
        )


        curve_records.append({

            "Class":
                class_label,

            "Bin":
                bin_number + 1,

            "Count":
                bin_count,

            "Mean_Predicted_Probability":
                mean_probability,

            "Observed_Frequency":
                observed_frequency,

            "Calibration_Gap":
                observed_frequency
                -
                mean_probability
        })


calibration_curve_data = pd.DataFrame(
    curve_records
)


# ----------------------------------------------------------------
# DISPLAY CURVE DATA
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALIBRATION CURVE DATA")
logger.info("-" * 70)

display(
    calibration_curve_data
)


# ----------------------------------------------------------------
# PLOT CALIBRATION CURVES
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("GENERATING CALIBRATION CURVE")
logger.info("-" * 70)

plt.figure(
    figsize=(10, 8)
)

# Perfect calibration reference
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect Calibration"
)


for class_label in classes:

    class_curve = calibration_curve_data[
        calibration_curve_data["Class"]
        == class_label
    ].sort_values(
        "Mean_Predicted_Probability"
    )


    if len(class_curve) == 0:
        continue


    plt.plot(
        class_curve[
            "Mean_Predicted_Probability"
        ],
        class_curve[
            "Observed_Frequency"
        ],
        marker="o",
        label=f"Class {class_label}"
    )


plt.xlabel(
    "Mean Predicted Probability"
)

plt.ylabel(
    "Observed Frequency"
)

plt.title(
    "Multiclass One-vs-Rest Calibration Curve"
)

plt.xlim(
    0,
    1
)

plt.ylim(
    0,
    1
)

plt.grid(
    True,
    alpha=0.3
)

plt.legend(
    loc="best"
)

plt.tight_layout()

plt.show()


# ----------------------------------------------------------------
# CALIBRATION GAP SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALIBRATION GAP SUMMARY")
logger.info("-" * 70)

calibration_gap_summary = (
    calibration_curve_data
    .groupby("Class")
    .agg(
        Mean_Absolute_Gap=(
            "Calibration_Gap",
            lambda x: np.mean(np.abs(x))
        ),
        Maximum_Absolute_Gap=(
            "Calibration_Gap",
            lambda x: np.max(np.abs(x))
        ),
        Mean_Gap=(
            "Calibration_Gap",
            "mean"
        ),
        Bins_Observed=(
            "Bin",
            "count"
        ),
        Samples_Analyzed=(
            "Count",
            "sum"
        )
    )
    .reset_index()
)


display(
    calibration_gap_summary
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# CURVE COVERAGE VALIDATION
# ----------------------------------------------------------------

curve_coverage_validation = (
    calibration_curve_data["Count"].sum()
    ==
    n_samples * n_classes
)


logger.info(
    f"Calibration Curve Coverage Validation          : "
    f"{'PASSED' if curve_coverage_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

finite_curve_validation = (
    np.isfinite(
        calibration_curve_data[
            "Mean_Predicted_Probability"
        ]
    ).all()
    and
    np.isfinite(
        calibration_curve_data[
            "Observed_Frequency"
        ]
    ).all()
)

probability_range_validation = (
    calibration_curve_data[
        "Mean_Predicted_Probability"
    ].between(0, 1).all()
)

frequency_range_validation = (
    calibration_curve_data[
        "Observed_Frequency"
    ].between(0, 1).all()
)


logger.info(
    f"Finite Calibration Values Validation           : "
    f"{'PASSED' if finite_curve_validation else 'FAILED'}"
)

logger.info(
    f"Probability Range Validation                    : "
    f"{'PASSED' if probability_range_validation else 'FAILED'}"
)

logger.info(
    f"Observed Frequency Range Validation             : "
    f"{'PASSED' if frequency_range_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

all_validations_passed = all([
    finite_curve_validation,
    probability_range_validation,
    frequency_range_validation,
    curve_coverage_validation,
    original_test_preservation_validation
])


logger.info("-" * 70)
logger.info("17.7 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"Calibration Curves Generated                    : "
    f"{calibration_curve_data['Class'].nunique()}"
)

logger.info(
    f"Calibration Status                              : "
    f"{'VISUALIZATION GENERATED' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info(
    f"Calibration Validation                          : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.7 Calibration Curve                         : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

## 17.8 Brier Score / Probability Reliability

### Objective

Evaluate the reliability of the model's multiclass probability predictions using the multiclass Brier score.

The Brier score measures the mean squared difference between the predicted probability distribution and the actual one-hot encoded outcome.

For each observation:

- The true class receives a target probability of 1.
- All other classes receive a target probability of 0.
- The model's complete six-class probability vector is compared against this target vector.

Lower Brier scores indicate better probabilistic accuracy.

### Additional Probability Reliability Measures

The analysis also evaluates:

1. Multiclass Brier score
2. Per-class one-vs-rest Brier scores
3. Multiclass logarithmic loss
4. Mean maximum prediction confidence
5. Mean confidence for correct predictions
6. Mean confidence for incorrect predictions
7. Probability normalization

### Interpretation

The Brier score evaluates the quality of the complete probability distribution rather than only whether the highest-probability class was correct.

This complements:

- Classification accuracy
- Confidence analysis
- Ambiguity analysis
- Calibration error
- Calibration curves

### Important Limitation

The final test set contains only 74 observations.

Therefore, Brier score and probability reliability estimates are dataset-specific and should not be interpreted as universal clinical probability reliability.

No probability calibration or model modification is performed in this section.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
from sklearn.metrics import log_loss
from IPython.display import display


logger.info("=" * 70)
logger.info("17.8 BRIER SCORE / PROBABILITY RELIABILITY")
logger.info("=" * 70)


# ----------------------------------------------------------------
# INPUT VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING INPUTS")
logger.info("-" * 70)

required_objects = {
    "test_probabilities": test_probabilities,
    "y_test": y_test,
    "model_classes": model_classes,
    "test_predictions": test_predictions
}

for object_name, object_value in required_objects.items():

    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is not available."
        )


probabilities = np.asarray(
    test_probabilities,
    dtype=float
)

targets = np.asarray(
    y_test
)

classes = np.asarray(
    model_classes
)

predictions = np.asarray(
    test_predictions
)


n_samples = len(targets)
n_classes = len(classes)


probability_shape_validation = (
    probabilities.shape
    ==
    (n_samples, n_classes)
)

prediction_shape_validation = (
    predictions.shape[0]
    ==
    n_samples
)

finite_probability_validation = np.all(
    np.isfinite(probabilities)
)

probability_range_validation = (
    np.all(probabilities >= 0)
    and
    np.all(probabilities <= 1)
)

probability_sum_validation = np.allclose(
    probabilities.sum(axis=1),
    1.0,
    atol=1e-6
)


logger.info(
    f"Probability Matrix Validation                  : "
    f"{'PASSED' if probability_shape_validation else 'FAILED'}"
)

logger.info(
    f"Prediction Shape Validation                    : "
    f"{'PASSED' if prediction_shape_validation else 'FAILED'}"
)

logger.info(
    f"Finite Probability Validation                  : "
    f"{'PASSED' if finite_probability_validation else 'FAILED'}"
)

logger.info(
    f"Probability Range Validation                   : "
    f"{'PASSED' if probability_range_validation else 'FAILED'}"
)

logger.info(
    f"Probability Sum-to-One Validation              : "
    f"{'PASSED' if probability_sum_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# ONE-HOT TARGET MATRIX
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CREATING ONE-HOT TARGET MATRIX")
logger.info("-" * 70)

target_matrix = np.zeros(
    (n_samples, n_classes),
    dtype=float
)


for class_position, class_label in enumerate(classes):

    target_matrix[:, class_position] = (
        targets == class_label
    ).astype(float)


target_encoding_validation = np.allclose(
    target_matrix.sum(axis=1),
    1.0
)

logger.info(
    f"One-Hot Target Encoding Validation              : "
    f"{'PASSED' if target_encoding_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# MULTICLASS BRIER SCORE
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING MULTICLASS BRIER SCORE")
logger.info("-" * 70)

squared_probability_errors = (
    probabilities
    -
    target_matrix
) ** 2


multiclass_brier_score = np.mean(
    np.sum(
        squared_probability_errors,
        axis=1
    )
)


logger.info(
    f"Multiclass Brier Score                         : "
    f"{multiclass_brier_score:.6f}"
)


# ----------------------------------------------------------------
# PER-CLASS ONE-VS-REST BRIER SCORES
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING PER-CLASS BRIER SCORES")
logger.info("-" * 70)

per_class_brier_records = []


for class_position, class_label in enumerate(classes):

    class_probability = probabilities[
        :,
        class_position
    ]

    class_target = target_matrix[
        :,
        class_position
    ]


    class_brier_score = np.mean(
        (
            class_probability
            -
            class_target
        ) ** 2
    )


    class_observed_frequency = np.mean(
        class_target
    )


    per_class_brier_records.append({

        "Class":
            class_label,

        "Brier_Score":
            class_brier_score,

        "Observed_Class_Frequency":
            class_observed_frequency,

        "Mean_Predicted_Probability":
            np.mean(class_probability)
    })


per_class_brier = pd.DataFrame(
    per_class_brier_records
)


display(
    per_class_brier
)


# ----------------------------------------------------------------
# LOG LOSS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING MULTICLASS LOG LOSS")
logger.info("-" * 70)

multiclass_log_loss = log_loss(
    targets,
    probabilities,
    labels=classes
)


logger.info(
    f"Multiclass Log Loss                            : "
    f"{multiclass_log_loss:.6f}"
)


# ----------------------------------------------------------------
# CONFIDENCE RELIABILITY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALCULATING CONFIDENCE RELIABILITY")
logger.info("-" * 70)

maximum_probabilities = probabilities.max(
    axis=1
)

predicted_class_positions = probabilities.argmax(
    axis=1
)

predicted_classes_from_probability = classes[
    predicted_class_positions
]


prediction_consistency_validation = np.array_equal(
    predicted_classes_from_probability,
    predictions
)


correct_mask = (
    predictions == targets
)

incorrect_mask = ~correct_mask


mean_max_probability = np.mean(
    maximum_probabilities
)

median_max_probability = np.median(
    maximum_probabilities
)

mean_correct_confidence = (
    np.mean(
        maximum_probabilities[
            correct_mask
        ]
    )
    if np.any(correct_mask)
    else np.nan
)

mean_incorrect_confidence = (
    np.mean(
        maximum_probabilities[
            incorrect_mask
        ]
    )
    if np.any(incorrect_mask)
    else np.nan
)


logger.info(
    f"Mean Maximum Probability                       : "
    f"{mean_max_probability:.4f}"
)

logger.info(
    f"Median Maximum Probability                     : "
    f"{median_max_probability:.4f}"
)

logger.info(
    f"Mean Correct-Prediction Confidence              : "
    f"{mean_correct_confidence:.4f}"
)

logger.info(
    f"Mean Incorrect-Prediction Confidence            : "
    f"{mean_incorrect_confidence:.4f}"
)


# ----------------------------------------------------------------
# PROBABILITY RELIABILITY SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBABILITY RELIABILITY SUMMARY")
logger.info("-" * 70)

reliability_summary = pd.DataFrame({

    "Metric": [

        "Multiclass Brier Score",

        "Multiclass Log Loss",

        "Mean Maximum Probability",

        "Median Maximum Probability",

        "Mean Correct Confidence",

        "Mean Incorrect Confidence",

        "Test Accuracy"
    ],

    "Value": [

        multiclass_brier_score,

        multiclass_log_loss,

        mean_max_probability,

        median_max_probability,

        mean_correct_confidence,

        mean_incorrect_confidence,

        np.mean(correct_mask)
    ]
})


display(
    reliability_summary
)


# ----------------------------------------------------------------
# PROBABILITY QUALITY INTERPRETATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBABILITY QUALITY INTERPRETATION")
logger.info("-" * 70)

if multiclass_brier_score < 0.10:

    brier_status = (
        "LOW BRIER SCORE / GOOD PROBABILITY QUALITY"
    )

elif multiclass_brier_score < 0.20:

    brier_status = (
        "MODERATE BRIER SCORE / ACCEPTABLE PROBABILITY QUALITY"
    )

else:

    brier_status = (
        "HIGH BRIER SCORE / PROBABILITY QUALITY REQUIRES REVIEW"
    )


logger.info(
    f"Brier Score Interpretation                     : "
    f"{brier_status}"
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)

if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# NUMERICAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("NUMERICAL VALIDATION")
logger.info("-" * 70)

brier_finite_validation = np.isfinite(
    multiclass_brier_score
)

log_loss_finite_validation = np.isfinite(
    multiclass_log_loss
)

brier_range_validation = (
    multiclass_brier_score >= 0
)

log_loss_range_validation = (
    multiclass_log_loss >= 0
)

confidence_range_validation = (
    np.all(
        maximum_probabilities >= 0
    )
    and
    np.all(
        maximum_probabilities <= 1
    )
)

per_class_brier_validation = (
    len(per_class_brier)
    ==
    n_classes
)

prediction_consistency_validation_log = (
    prediction_consistency_validation
)


logger.info(
    f"Brier Score Finite Validation                  : "
    f"{'PASSED' if brier_finite_validation else 'FAILED'}"
)

logger.info(
    f"Log Loss Finite Validation                     : "
    f"{'PASSED' if log_loss_finite_validation else 'FAILED'}"
)

logger.info(
    f"Brier Score Range Validation                   : "
    f"{'PASSED' if brier_range_validation else 'FAILED'}"
)

logger.info(
    f"Log Loss Range Validation                      : "
    f"{'PASSED' if log_loss_range_validation else 'FAILED'}"
)

logger.info(
    f"Confidence Range Validation                    : "
    f"{'PASSED' if confidence_range_validation else 'FAILED'}"
)

logger.info(
    f"Per-Class Brier Coverage Validation            : "
    f"{'PASSED' if per_class_brier_validation else 'FAILED'}"
)

logger.info(
    f"Prediction Consistency Validation              : "
    f"{'PASSED' if prediction_consistency_validation_log else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

all_validations_passed = all([

    probability_shape_validation,

    prediction_shape_validation,

    finite_probability_validation,

    probability_range_validation,

    probability_sum_validation,

    target_encoding_validation,

    brier_finite_validation,

    log_loss_finite_validation,

    brier_range_validation,

    log_loss_range_validation,

    confidence_range_validation,

    per_class_brier_validation,

    prediction_consistency_validation,

    original_test_preservation_validation
])


logger.info("-" * 70)
logger.info("17.8 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    f"Multiclass Brier Score                         : "
    f"{multiclass_brier_score:.6f}"
)

logger.info(
    f"Multiclass Log Loss                            : "
    f"{multiclass_log_loss:.6f}"
)

logger.info(
    f"Probability Quality                            : "
    f"{brier_status}"
)

logger.info(
    f"Probability Reliability Validation              : "
    f"{'PASSED' if all_validations_passed else 'FAILED'}"
)

logger.info(
    f"17.8 Brier Score / Probability Reliability      : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

The optimized Random Forest demonstrates:

- Moderate probability reliability according to the multiclass Brier score.

### Probability Reliability Conclusion

The probability outputs are useful for:

- Ranking prediction confidence.
- Identifying potentially uncertain predictions.
- Supporting model monitoring.
- Detecting ambiguous decision-boundary cases.

However, the raw probabilities should not be interpreted as directly calibrated clinical probabilities.

### Production Recommendation

The optimized Random Forest remains a strong production candidate for classification on this dataset.

For any real-world clinical deployment:

- External validation is required.
- Calibration should be reassessed on an independent representative dataset.
- Probability calibration methods may be considered if reliable probability estimates are required.
- Minority classes require continued monitoring.
- Class 2 and Class 4 should receive particular monitoring because of their observed reciprocal confusion.
- The model should not be treated as an autonomous clinical diagnostic system without appropriate clinical validation and governance.

### Overall Status

The confidence and calibration analysis provides useful evidence about probability behavior while maintaining appropriate limitations regarding sample size and clinical generalization.

````python


# Required Libraries
import logging
import numpy as np
import pandas as pd
from IPython.display import display


logger.info("=" * 70)
logger.info("17.9 CONFIDENCE & CALIBRATION SUMMARY")
logger.info("=" * 70)


# ----------------------------------------------------------------
# VALIDATE REQUIRED SECTION 17 OBJECTS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SECTION 17 RESULTS")
logger.info("-" * 70)


required_section_objects = {

    "test_probabilities":
        test_probabilities,

    "test_predictions":
        test_predictions,

    "y_test":
        y_test,

    "model_classes":
        model_classes,

    "confidence_analysis":
        confidence_analysis,

    "ambiguous_prediction_dataset":
        ambiguous_prediction_dataset,

    "per_class_calibration":
        per_class_calibration,

    "calibration_curve_data":
        calibration_curve_data,

    "per_class_brier":
        per_class_brier,

    "multiclass_brier_score":
        multiclass_brier_score,

    "multiclass_log_loss":
        multiclass_log_loss
}


missing_objects = [
    object_name
    for object_name, object_value
    in required_section_objects.items()
    if object_value is None
]


if len(missing_objects) > 0:

    raise ValueError(
        "Missing required Section 17 objects: "
        + ", ".join(missing_objects)
    )


logger.info(
    "Section 17 Object Availability                  : PASSED"
)


# ----------------------------------------------------------------
# CORE PERFORMANCE VALUES
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("EXTRACTING CORE CONFIDENCE METRICS")
logger.info("-" * 70)


probabilities = np.asarray(
    test_probabilities,
    dtype=float
)

targets = np.asarray(
    y_test
)

predictions = np.asarray(
    test_predictions
)

classes = np.asarray(
    model_classes
)


maximum_probabilities = probabilities.max(
    axis=1
)


prediction_correct = (
    predictions == targets
)


total_test_samples = len(
    targets
)

total_correct_predictions = int(
    prediction_correct.sum()
)

total_incorrect_predictions = (
    total_test_samples
    -
    total_correct_predictions
)


test_accuracy = (
    total_correct_predictions
    /
    total_test_samples
)


mean_confidence = np.mean(
    maximum_probabilities
)

median_confidence = np.median(
    maximum_probabilities
)


mean_correct_confidence = np.mean(
    maximum_probabilities[
        prediction_correct
    ]
)


mean_incorrect_confidence = np.mean(
    maximum_probabilities[
        ~prediction_correct
    ]
)


logger.info(
    f"Test Samples                                  : "
    f"{total_test_samples}"
)

logger.info(
    f"Test Accuracy                                 : "
    f"{test_accuracy:.4f}"
)

logger.info(
    f"Mean Prediction Confidence                    : "
    f"{mean_confidence:.4f}"
)

logger.info(
    f"Median Prediction Confidence                  : "
    f"{median_confidence:.4f}"
)

logger.info(
    f"Mean Correct-Prediction Confidence             : "
    f"{mean_correct_confidence:.4f}"
)

logger.info(
    f"Mean Incorrect-Prediction Confidence           : "
    f"{mean_incorrect_confidence:.4f}"
)


# ----------------------------------------------------------------
# HIGH / LOW CONFIDENCE METRICS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CONFIDENCE REGION SUMMARY")
logger.info("-" * 70)


high_confidence_threshold = 0.80
low_confidence_threshold = 0.60


high_confidence_mask = (
    maximum_probabilities
    >= high_confidence_threshold
)

low_confidence_mask = (
    maximum_probabilities
    < low_confidence_threshold
)


high_confidence_count = int(
    high_confidence_mask.sum()
)

low_confidence_count = int(
    low_confidence_mask.sum()
)


high_confidence_accuracy = np.mean(
    prediction_correct[
        high_confidence_mask
    ]
)


low_confidence_accuracy = np.mean(
    prediction_correct[
        low_confidence_mask
    ]
)


high_confidence_errors = int(
    np.sum(
        ~prediction_correct[
            high_confidence_mask
        ]
    )
)

low_confidence_errors = int(
    np.sum(
        ~prediction_correct[
            low_confidence_mask
        ]
    )
)


total_errors = total_incorrect_predictions


if total_errors > 0:

    low_confidence_error_capture_rate = (
        low_confidence_errors
        /
        total_errors
    )

else:

    low_confidence_error_capture_rate = np.nan


logger.info(
    f"High-Confidence Predictions                   : "
    f"{high_confidence_count}"
)

logger.info(
    f"High-Confidence Accuracy                      : "
    f"{high_confidence_accuracy:.4f}"
)

logger.info(
    f"High-Confidence Errors                        : "
    f"{high_confidence_errors}"
)

logger.info(
    f"Low-Confidence Predictions                    : "
    f"{low_confidence_count}"
)

logger.info(
    f"Low-Confidence Accuracy                       : "
    f"{low_confidence_accuracy:.4f}"
)

logger.info(
    f"Low-Confidence Errors                         : "
    f"{low_confidence_errors}"
)

logger.info(
    f"Low-Confidence Error Capture Rate             : "
    f"{low_confidence_error_capture_rate:.4f}"
)


# ----------------------------------------------------------------
# AMBIGUITY METRICS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("AMBIGUITY SUMMARY")
logger.info("-" * 70)


ambiguity_threshold = 0.10


if "Confidence_Margin" in ambiguous_prediction_dataset.columns:

    ambiguous_mask = (
        ambiguous_prediction_dataset[
            "Confidence_Margin"
        ]
        <
        ambiguity_threshold
    )

    ambiguous_count = int(
        ambiguous_mask.sum()
    )

    ambiguous_correct_count = int(
        ambiguous_prediction_dataset.loc[
            ambiguous_mask,
            "Prediction_Correct"
        ].sum()
    )

    ambiguous_incorrect_count = (
        ambiguous_count
        -
        ambiguous_correct_count
    )

else:

    ambiguous_count = 0
    ambiguous_correct_count = 0
    ambiguous_incorrect_count = 0


ambiguous_percentage = (
    ambiguous_count
    /
    total_test_samples
)


if ambiguous_count > 0:

    ambiguous_accuracy = (
        ambiguous_correct_count
        /
        ambiguous_count
    )

else:

    ambiguous_accuracy = np.nan


if total_errors > 0:

    ambiguity_error_capture_rate = (
        ambiguous_incorrect_count
        /
        total_errors
    )

else:

    ambiguity_error_capture_rate = np.nan


logger.info(
    f"Ambiguity Margin Threshold                    : "
    f"{ambiguity_threshold:.2f}"
)

logger.info(
    f"Ambiguous Predictions                         : "
    f"{ambiguous_count}"
)

logger.info(
    f"Ambiguous Percentage                          : "
    f"{ambiguous_percentage:.4%}"
)

logger.info(
    f"Ambiguous Accuracy                            : "
    f"{ambiguous_accuracy:.4f}"
)

logger.info(
    f"Ambiguity Error Capture Rate                  : "
    f"{ambiguity_error_capture_rate:.4f}"
)


# ----------------------------------------------------------------
# CLASS 2 / CLASS 4 BOUNDARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS 2 ↔ CLASS 4 BOUNDARY SUMMARY")
logger.info("-" * 70)


class_2_4_competing_mask = (
    (
        ambiguous_prediction_dataset[
            "Top1_Class"
        ].isin([2, 4])
    )
    &
    (
        ambiguous_prediction_dataset[
            "Top2_Class"
        ].isin([2, 4])
    )
)


# The full test-set probability dataset is used to identify
# all predictions where Classes 2 and 4 are the top two competing
# classes.

top_two_positions = np.argsort(
    probabilities,
    axis=1
)[:, -2:][:, ::-1]


top1_boundary_classes = classes[
    top_two_positions[:, 0]
]

top2_boundary_classes = classes[
    top_two_positions[:, 1]
]


class_2_4_full_mask = (
    (
        (
            top1_boundary_classes == 2
        )
        &
        (
            top2_boundary_classes == 4
        )
    )
    |
    (
        (
            top1_boundary_classes == 4
        )
        &
        (
            top2_boundary_classes == 2
        )
    )
)


class_2_4_competing_count = int(
    class_2_4_full_mask.sum()
)


class_2_4_error_count = int(
    np.sum(
        ~prediction_correct[
            class_2_4_full_mask
        ]
    )
)


logger.info(
    f"Class 2 ↔ Class 4 Competing Predictions          : "
    f"{class_2_4_competing_count}"
)

logger.info(
    f"Class 2 ↔ Class 4 Observed Errors                : "
    f"{class_2_4_error_count}"
)


# ----------------------------------------------------------------
# CALIBRATION METRICS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("CALIBRATION SUMMARY")
logger.info("-" * 70)


overall_calibration_error = np.average(
    calibration_bin_results[
        "Absolute_Calibration_Error"
    ],
    weights=calibration_bin_results[
        "Count"
    ]
)


maximum_calibration_error = (
    calibration_bin_results[
        "Absolute_Calibration_Error"
    ].max()
)


calibration_status = (
    "MODERATE OBSERVED CALIBRATION ERROR"
    if overall_calibration_error < 0.10
    else
    "HIGH OBSERVED CALIBRATION ERROR"
)


logger.info(
    f"Overall Calibration Error                     : "
    f"{overall_calibration_error:.4f}"
)

logger.info(
    f"Maximum Calibration Error                     : "
    f"{maximum_calibration_error:.4f}"
)

logger.info(
    f"Calibration Status                             : "
    f"{calibration_status}"
)


# ----------------------------------------------------------------
# BRIER / LOG LOSS SUMMARY
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PROBABILITY RELIABILITY SUMMARY")
logger.info("-" * 70)


logger.info(
    f"Multiclass Brier Score                        : "
    f"{multiclass_brier_score:.6f}"
)

logger.info(
    f"Multiclass Log Loss                           : "
    f"{multiclass_log_loss:.6f}"
)


# ----------------------------------------------------------------
# SUMMARY TABLE
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL CONFIDENCE & CALIBRATION SUMMARY TABLE")
logger.info("-" * 70)


confidence_calibration_summary = pd.DataFrame({

    "Category": [

        "Test Accuracy",

        "Mean Prediction Confidence",

        "Median Prediction Confidence",

        "Mean Correct Confidence",

        "Mean Incorrect Confidence",

        "High-Confidence Accuracy",

        "High-Confidence Errors",

        "Low-Confidence Accuracy",

        "Low-Confidence Error Capture Rate",

        "Ambiguous Predictions",

        "Ambiguous Accuracy",

        "Ambiguity Error Capture Rate",

        "Class 2 ↔ Class 4 Competing Predictions",

        "Class 2 ↔ Class 4 Observed Errors",

        "Overall Calibration Error",

        "Maximum Calibration Error",

        "Multiclass Brier Score",

        "Multiclass Log Loss"
    ],

    "Value": [

        test_accuracy,

        mean_confidence,

        median_confidence,

        mean_correct_confidence,

        mean_incorrect_confidence,

        high_confidence_accuracy,

        high_confidence_errors,

        low_confidence_accuracy,

        low_confidence_error_capture_rate,

        ambiguous_count,

        ambiguous_accuracy,

        ambiguity_error_capture_rate,

        class_2_4_competing_count,

        class_2_4_error_count,

        overall_calibration_error,

        maximum_calibration_error,

        multiclass_brier_score,

        multiclass_log_loss
    ]
})


display(
    confidence_calibration_summary
)


# ----------------------------------------------------------------
# PRODUCTION ASSESSMENT
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("PRODUCTION PROBABILITY ASSESSMENT")
logger.info("-" * 70)


probability_output_valid = (
    np.all(
        np.isfinite(probabilities)
    )
    and
    np.all(
        probabilities >= 0
    )
    and
    np.all(
        probabilities <= 1
    )
    and
    np.allclose(
        probabilities.sum(axis=1),
        1.0,
        atol=1e-6
    )
)


high_confidence_reliable = (
    high_confidence_errors == 0
)

low_confidence_monitoring_useful = (
    low_confidence_error_capture_rate >= 0.50
)

calibration_requires_monitoring = (
    overall_calibration_error >= 0.05
)


logger.info(
    f"Probability Output Validity                    : "
    f"{'PASSED' if probability_output_valid else 'FAILED'}"
)

logger.info(
    f"High-Confidence Prediction Reliability         : "
    f"{'STRONG' if high_confidence_reliable else 'REQUIRES REVIEW'}"
)

logger.info(
    f"Low-Confidence Monitoring Utility              : "
    f"{'USEFUL' if low_confidence_monitoring_useful else 'LIMITED'}"
)

logger.info(
    f"Calibration Monitoring Requirement             : "
    f"{'REQUIRED' if calibration_requires_monitoring else 'LOW PRIORITY'}"
)


# ----------------------------------------------------------------
# DATA PRESERVATION VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("DATA PRESERVATION VALIDATION")
logger.info("-" * 70)


if "X_test_probability_original" in globals():

    original_test_preservation_validation = (
        X_test.equals(
            X_test_probability_original
        )
    )

else:

    original_test_preservation_validation = True


logger.info(
    f"Original X_test Preservation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL VALIDATION
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL VALIDATION")
logger.info("-" * 70)


numerical_validation = all([

    probability_output_valid,

    np.isfinite(
        overall_calibration_error
    ),

    np.isfinite(
        multiclass_brier_score
    ),

    np.isfinite(
        multiclass_log_loss
    )
])


coverage_validation = all([

    total_test_samples == 74,

    len(predictions) == total_test_samples,

    len(maximum_probabilities) == total_test_samples
])


all_validations_passed = all([

    numerical_validation,

    coverage_validation,

    original_test_preservation_validation
])


logger.info(
    f"Numerical Validation                          : "
    f"{'PASSED' if numerical_validation else 'FAILED'}"
)

logger.info(
    f"Test Coverage Validation                      : "
    f"{'PASSED' if coverage_validation else 'FAILED'}"
)

logger.info(
    f"Data Preservation Validation                  : "
    f"{'PASSED' if original_test_preservation_validation else 'FAILED'}"
)


# ----------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------

logger.info("-" * 70)
logger.info("17.9 FINAL STATUS")
logger.info("-" * 70)


logger.info(
    f"Confidence Analysis                            : COMPLETE"
)

logger.info(
    f"Calibration Analysis                           : COMPLETE"
)

logger.info(
    f"Probability Reliability Analysis               : COMPLETE"
)

logger.info(
    f"Overall Probability Output Assessment          : "
    f"{'STRONG WITH CALIBRATION MONITORING' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info(
    f"17.9 Confidence & Calibration Summary          : "
    f"{'COMPLETE' if all_validations_passed else 'REQUIRES REVIEW'}"
)

logger.info("=" * 70)
````

# 18.1 Final Model Selection

This subsection formally establishes the final production model that will be
serialized and carried forward into deployment.

The final model is selected based on the evidence established during the
modeling and evaluation stages, including:

- Independent test-set performance
- Cross-validation performance
- Cross-validation stability
- Generalization performance
- Per-class performance
- Multiclass ROC-AUC
- Precision-Recall performance
- Overall model evaluation

The selected model is the Optimized Random Forest represented by the
`model` object.

No retraining is performed in this subsection. The purpose is to validate
that the existing fitted model is available and suitable for serialization.

Required Libraries:
- logging
- numpy
- pandas
- scikit-learn
- IPython.display

````python
# Required Libraries
import logging
import numpy as np
import pandas as pd

from IPython.display import display
from sklearn.pipeline import Pipeline

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("18.1 FINAL MODEL SELECTION")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING FINAL MODEL")
logger.info("-" * 70)

required_objects = [
    "model"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

if model is None:
    raise ValueError(
        "The final model object is None."
    )

if not hasattr(model, "predict"):
    raise AttributeError(
        "The final model does not provide predict()."
    )

if not hasattr(model, "predict_proba"):
    raise AttributeError(
        "The final model does not provide predict_proba()."
    )

if not hasattr(model, "fit"):
    raise AttributeError(
        "The final model does not provide fit()."
    )

logger.info(
    "Final Model Object Availability                 : PASSED"
)

logger.info(
    "Prediction Capability                           : PASSED"
)

logger.info(
    "Probability Prediction Capability               : PASSED"
)

# ----------------------------------------------------------------------
# MODEL TYPE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("MODEL TYPE VALIDATION")
logger.info("-" * 70)

if not isinstance(model, Pipeline):
    raise TypeError(
        "The final production model must be a "
        "scikit-learn Pipeline."
    )

logger.info(
    "Model Type                                     : %s",
    type(model).__name__
)

logger.info(
    "Pipeline Validation                            : PASSED"
)

# ----------------------------------------------------------------------
# PIPELINE STRUCTURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("PIPELINE STRUCTURE VALIDATION")
logger.info("-" * 70)

if not hasattr(model, "named_steps"):
    raise AttributeError(
        "Final model pipeline does not contain named steps."
    )

pipeline_steps = list(
    model.named_steps.keys()
)

if len(pipeline_steps) < 2:
    raise AssertionError(
        "Final production pipeline must contain "
        "preprocessing and model stages."
    )

final_estimator = list(
    model.named_steps.values()
)[-1]

if not hasattr(final_estimator, "predict"):
    raise AttributeError(
        "Final pipeline estimator does not provide predict()."
    )

logger.info(
    "Pipeline Steps                                  : %s",
    pipeline_steps
)

logger.info(
    "Final Estimator                                 : %s",
    type(final_estimator).__name__
)

logger.info(
    "Pipeline Structure Validation                  : PASSED"
)

# ----------------------------------------------------------------------
# CLASS VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("CLASS VALIDATION")
logger.info("-" * 70)

if not hasattr(final_estimator, "classes_"):
    raise AttributeError(
        "Final estimator does not expose classes_."
    )

model_classes = np.asarray(
    final_estimator.classes_
)

if model_classes.size == 0:
    raise AssertionError(
        "Final model contains no class labels."
    )

if "y_test" in globals():
    expected_classes = np.sort(
        np.unique(
            np.asarray(y_test)
        )
    )

    if not np.array_equal(
        np.sort(model_classes),
        expected_classes
    ):
        raise AssertionError(
            "Final model class labels do not match "
            "the test-set class labels."
        )

logger.info(
    "Final Model Classes                            : %s",
    model_classes.tolist()
)

logger.info(
    "Class Label Validation                         : PASSED"
)

# ----------------------------------------------------------------------
# FINAL MODEL SUMMARY
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("FINAL MODEL SUMMARY")
logger.info("-" * 70)

final_model_summary = pd.DataFrame({
    "Property": [
        "Production Candidate",
        "Pipeline Type",
        "Final Estimator",
        "Number of Classes",
        "Number of Pipeline Steps",
        "Probability Support",
        "Serialization Status"
    ],
    "Value": [
        "Optimized Random Forest",
        type(model).__name__,
        type(final_estimator).__name__,
        len(model_classes),
        len(pipeline_steps),
        "Available",
        "Ready"
    ]
})

display(
    final_model_summary
)

# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.1 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Selected Production Model                     : Optimized Random Forest"
)

logger.info(
    "Final Estimator                               : %s",
    type(final_estimator).__name__
)

logger.info(
    "Number of Classes                             : %d",
    len(model_classes)
)

logger.info(
    "Model Serialization Readiness                 : PASSED"
)

logger.info("=" * 70)
logger.info("18.1 FINAL MODEL SELECTION : COMPLETE")
logger.info("=" * 70)
````

# 18.2 Save Production Model

This subsection serializes the final fitted production model for
reproducible inference and deployment.

The complete fitted scikit-learn Pipeline is saved rather than only the
Random Forest estimator. This preserves the complete inference workflow,
including:

- Feature preprocessing
- Missing-value handling
- Feature scaling
- Categorical encoding
- Final Random Forest classifier

The serialized artifact will be stored in the standardized model artifact
directory established earlier in the project.

The saved model can subsequently be loaded without retraining and used for
prediction on compatible raw input data.

Required Libraries:
- logging
- pathlib
- joblib
- scikit-learn

````python
# Required Libraries
import logging
from pathlib import Path

import joblib

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("18.2 SAVE PRODUCTION MODEL")
logger.info("=" * 70)

# ----------------------------------------------------------------------
# VALIDATING REQUIRED OBJECTS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

required_objects = [
    "model",
    "PRODUCTION_MODEL_PATH"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise NameError(
        "Missing required objects: "
        + ", ".join(missing_objects)
    )

if model is None:
    raise ValueError(
        "Production model is None."
    )

if not hasattr(model, "predict"):
    raise AttributeError(
        "Production model does not provide predict()."
    )

if not hasattr(model, "predict_proba"):
    raise AttributeError(
        "Production model does not provide predict_proba()."
    )

logger.info(
    "Production Model Availability                   : PASSED"
)

logger.info(
    "Prediction Capability                           : PASSED"
)

logger.info(
    "Probability Prediction Capability               : PASSED"
)

# ----------------------------------------------------------------------
# VALIDATE ARTIFACT PATH
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT PATH")
logger.info("-" * 70)

production_model_path = Path(
    PRODUCTION_MODEL_PATH
)

production_model_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

logger.info(
    "Production Model Path                           : %s",
    production_model_path
)

logger.info(
    "Model Artifact Directory                        : %s",
    production_model_path.parent
)

logger.info(
    "Artifact Directory Validation                   : PASSED"
)

# ----------------------------------------------------------------------
# SERIALIZE PRODUCTION MODEL
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("SERIALIZING PRODUCTION MODEL")
logger.info("-" * 70)

joblib.dump(
    model,
    production_model_path
)

logger.info(
    "Production Model Serialization                  : PASSED"
)

# ----------------------------------------------------------------------
# FILE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SERIALIZED ARTIFACT")
logger.info("-" * 70)

if not production_model_path.exists():
    raise FileNotFoundError(
        "Serialized production model was not created."
    )

if not production_model_path.is_file():
    raise AssertionError(
        "Production model path does not reference a file."
    )

artifact_size_bytes = (
    production_model_path.stat().st_size
)

if artifact_size_bytes <= 0:
    raise ValueError(
        "Serialized production model file is empty."
    )

logger.info(
    "Serialized Artifact Exists                      : PASSED"
)

logger.info(
    "Serialized Artifact Is File                    : PASSED"
)

logger.info(
    "Serialized Artifact Size                       : %d bytes",
    artifact_size_bytes
)

logger.info(
    "Serialized Artifact Size Validation             : PASSED"
)

# ----------------------------------------------------------------------
# LOAD-BACK VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL LOAD-BACK")
logger.info("-" * 70)

loaded_production_model = joblib.load(
    production_model_path
)

if loaded_production_model is None:
    raise ValueError(
        "Loaded production model is None."
    )

if not hasattr(
    loaded_production_model,
    "predict"
):
    raise AttributeError(
        "Loaded production model does not provide predict()."
    )

if not hasattr(
    loaded_production_model,
    "predict_proba"
):
    raise AttributeError(
        "Loaded production model does not provide predict_proba()."
    )

logger.info(
    "Serialized Model Load-Back                      : PASSED"
)

logger.info(
    "Loaded Prediction Capability                   : PASSED"
)

logger.info(
    "Loaded Probability Capability                  : PASSED"
)

# ----------------------------------------------------------------------
# MODEL STRUCTURE VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED MODEL STRUCTURE")
logger.info("-" * 70)

if not hasattr(
    loaded_production_model,
    "named_steps"
):
    raise AssertionError(
        "Loaded production model is not a Pipeline."
    )

loaded_pipeline_steps = list(
    loaded_production_model.named_steps.keys()
)

if loaded_pipeline_steps != list(
    model.named_steps.keys()
):
    raise AssertionError(
        "Loaded pipeline structure does not match "
        "the original production model."
    )

logger.info(
    "Loaded Pipeline Steps                          : %s",
    loaded_pipeline_steps
)

logger.info(
    "Pipeline Structure Preservation                : PASSED"
)

# ----------------------------------------------------------------------
# CLASS VALIDATION
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED MODEL CLASSES")
logger.info("-" * 70)

original_estimator = list(
    model.named_steps.values()
)[-1]

loaded_estimator = list(
    loaded_production_model.named_steps.values()
)[-1]

if not hasattr(
    original_estimator,
    "classes_"
):
    raise AttributeError(
        "Original production estimator does not expose classes_."
    )

if not hasattr(
    loaded_estimator,
    "classes_"
):
    raise AttributeError(
        "Loaded production estimator does not expose classes_."
    )

if not np.array_equal(
    np.asarray(original_estimator.classes_),
    np.asarray(loaded_estimator.classes_)
):
    raise AssertionError(
        "Loaded model class labels do not match "
        "the original model."
    )

logger.info(
    "Loaded Model Classes                           : %s",
    np.asarray(
        loaded_estimator.classes_
    ).tolist()
)

logger.info(
    "Class Preservation                             : PASSED"
)

# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.2 FINAL STATUS")
logger.info("-" * 70)

logger.info(
    "Production Model                               : SAVED"
)

logger.info(
    "Artifact Path                                  : %s",
    production_model_path
)

logger.info(
    "Artifact Size                                  : %d bytes",
    artifact_size_bytes
)

logger.info(
    "Load-Back Validation                           : PASSED"
)

logger.info(
    "Pipeline Structure Preservation                : PASSED"
)

logger.info(
    "Class Preservation                             : PASSED"
)

logger.info(
    "Production Model Serialization                 : COMPLETE"
)

logger.info("=" * 70)
logger.info("18.2 SAVE PRODUCTION MODEL : COMPLETE")
logger.info("=" * 70)
````

`artifacts/models/preprocessing_pipeline.pkl`

````python


import logging
from pathlib import Path

import joblib

logger = logging.getLogger(__name__)

logger.info("=" * 70)
logger.info("18.3 SAVE PREPROCESSING PIPELINE")
logger.info("=" * 70)

# ------------------------------------------------------------
# VALIDATING REQUIRED OBJECTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

if "model" not in globals():
    raise NameError("Required object 'model' is not available.")

if "PREPROCESSING_PIPELINE_PATH" not in globals():
    raise NameError("Required path 'PREPROCESSING_PIPELINE_PATH' is not available.")

if not hasattr(model, "named_steps"):
    raise TypeError("Production model does not expose named pipeline steps.")

if "preprocessor" not in model.named_steps:
    raise KeyError("Fitted 'preprocessor' step not found in production model.")

preprocessing_pipeline = model.named_steps["preprocessor"]

logger.info("Production Model Availability                   : PASSED")
logger.info("Preprocessing Pipeline Availability             : PASSED")

# ------------------------------------------------------------
# VALIDATING PREPROCESSOR
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PREPROCESSING PIPELINE")
logger.info("-" * 70)

if not hasattr(preprocessing_pipeline, "transform"):
    raise TypeError("Preprocessing pipeline does not support transform().")

logger.info(
    "Preprocessing Pipeline Type                      : %s",
    type(preprocessing_pipeline).__name__
)

if hasattr(preprocessing_pipeline, "transformers_"):
    transformer_names = [
        name for name, _, _ in preprocessing_pipeline.transformers_
    ]
    logger.info(
        "Preprocessing Components                        : %s",
        transformer_names
    )

logger.info("Preprocessing Transform Capability              : PASSED")

# ------------------------------------------------------------
# VALIDATING ARTIFACT PATH
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT PATH")
logger.info("-" * 70)

preprocessing_pipeline_path = Path(PREPROCESSING_PIPELINE_PATH)

preprocessing_pipeline_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

logger.info(
    "Preprocessing Pipeline Path                      : %s",
    preprocessing_pipeline_path
)

logger.info(
    "Preprocessing Artifact Directory                 : %s",
    preprocessing_pipeline_path.parent
)

logger.info("Artifact Directory Validation                    : PASSED")

# ------------------------------------------------------------
# SERIALIZING PREPROCESSING PIPELINE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("SERIALIZING PREPROCESSING PIPELINE")
logger.info("-" * 70)

joblib.dump(
    preprocessing_pipeline,
    preprocessing_pipeline_path
)

logger.info("Preprocessing Pipeline Serialization            : PASSED")

# ------------------------------------------------------------
# VALIDATING SERIALIZED ARTIFACT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SERIALIZED ARTIFACT")
logger.info("-" * 70)

if not preprocessing_pipeline_path.exists():
    raise FileNotFoundError(
        "Serialized preprocessing pipeline was not created."
    )

if not preprocessing_pipeline_path.is_file():
    raise IOError(
        "Serialized preprocessing pipeline path is not a file."
    )

artifact_size = preprocessing_pipeline_path.stat().st_size

if artifact_size <= 0:
    raise IOError(
        "Serialized preprocessing pipeline artifact is empty."
    )

logger.info("Serialized Artifact Exists                      : PASSED")
logger.info("Serialized Artifact Is File                    : PASSED")
logger.info(
    "Serialized Artifact Size                       : %d bytes",
    artifact_size
)
logger.info("Serialized Artifact Size Validation             : PASSED")

# ------------------------------------------------------------
# VALIDATING PIPELINE LOAD-BACK
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PREPROCESSING PIPELINE LOAD-BACK")
logger.info("-" * 70)

loaded_preprocessing_pipeline = joblib.load(
    preprocessing_pipeline_path
)

if loaded_preprocessing_pipeline is None:
    raise ValueError(
        "Loaded preprocessing pipeline is None."
    )

if not hasattr(loaded_preprocessing_pipeline, "transform"):
    raise TypeError(
        "Loaded preprocessing pipeline does not support transform()."
    )

logger.info("Serialized Pipeline Load-Back                    : PASSED")
logger.info("Loaded Transform Capability                     : PASSED")

# ------------------------------------------------------------
# VALIDATING LOADED PIPELINE STRUCTURE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED PIPELINE STRUCTURE")
logger.info("-" * 70)

original_type = type(preprocessing_pipeline)
loaded_type = type(loaded_preprocessing_pipeline)

if original_type != loaded_type:
    raise TypeError(
        "Loaded preprocessing pipeline type does not match original."
    )

logger.info(
    "Loaded Pipeline Type                            : %s",
    loaded_type.__name__
)

if hasattr(preprocessing_pipeline, "transformers"):
    original_transformers = [
        name for name, _, _ in preprocessing_pipeline.transformers
    ]
    loaded_transformers = [
        name for name, _, _ in loaded_preprocessing_pipeline.transformers
    ]

    if original_transformers != loaded_transformers:
        raise ValueError(
            "Preprocessing transformer structure was not preserved."
        )

    logger.info(
        "Loaded Transformer Structure                    : %s",
        loaded_transformers
    )

logger.info("Pipeline Structure Preservation                 : PASSED")

# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.3 FINAL STATUS")
logger.info("-" * 70)

logger.info("Preprocessing Pipeline                         : SAVED")
logger.info(
    "Artifact Path                                  : %s",
    preprocessing_pipeline_path
)
logger.info(
    "Artifact Size                                  : %d bytes",
    artifact_size
)
logger.info("Load-Back Validation                           : PASSED")
logger.info("Pipeline Structure Preservation                : PASSED")
logger.info("Preprocessing Pipeline Serialization           : COMPLETE")

logger.info("=" * 70)
logger.info("18.3 SAVE PREPROCESSING PIPELINE : COMPLETE")
logger.info("=" * 70)
````

## 18.5 Save Class Mapping

### Objective
Create a production-ready class mapping artifact that preserves the exact target
class labels learned by the final model.

### Class Mapping Requirements
- Preserve all six model classes.
- Store classes in the exact model order.
- Store both index-to-label and label-to-index mappings.
- Serialize the mapping as JSON.
- Validate artifact existence, size, load-back, and mapping preservation.
- Ensure the saved mapping matches the final production model classes.

### Required Libraries
- `json`

````python


import json
import logging
from pathlib import Path

import numpy as np


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger(__name__)


# ------------------------------------------------------------
# SECTION HEADER
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("18.5 SAVE CLASS MAPPING")
logger.info("=" * 70)


# ------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

required_objects = {
    "model": model,
    "CLASS_MAPPING_PATH": CLASS_MAPPING_PATH,
}

for object_name, object_value in required_objects.items():
    if object_value is None:
        raise ValueError(f"Required object '{object_name}' is unavailable.")

logger.info("Production Model Availability                   : PASSED")
logger.info("Class Mapping Path Availability                : PASSED")


# ------------------------------------------------------------
# VALIDATE MODEL CLASS INFORMATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL CLASS INFORMATION")
logger.info("-" * 70)

if not hasattr(model, "named_steps"):
    raise TypeError("Production model must be a sklearn Pipeline.")

if "classifier" not in model.named_steps:
    raise ValueError("Production pipeline does not contain 'classifier' step.")

classifier = model.named_steps["classifier"]

if not hasattr(classifier, "classes_"):
    raise AttributeError(
        "Final classifier does not expose classes_."
    )

model_classes = np.asarray(classifier.classes_)

if model_classes.size == 0:
    raise ValueError("No model classes detected.")

logger.info("Final Estimator Type                           : %s",
            type(classifier).__name__)
logger.info("Model Class Count                             : %d",
            len(model_classes))
logger.info("Model Classes                                 : %s",
            [str(value) for value in model_classes])


# ------------------------------------------------------------
# NORMALIZE CLASS LABELS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("NORMALIZING CLASS LABELS")
logger.info("-" * 70)

class_labels = [str(value).strip() for value in model_classes]

if len(class_labels) != len(set(class_labels)):
    raise ValueError("Duplicate class labels detected.")

if not all(class_labels):
    raise ValueError("Empty class label detected.")

logger.info("Class Label Normalization                     : PASSED")
logger.info("Normalized Class Labels                       : %s",
            class_labels)


# ------------------------------------------------------------
# BUILD CLASS MAPPING
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING CLASS MAPPING")
logger.info("-" * 70)

index_to_label = {
    str(index): label
    for index, label in enumerate(class_labels)
}

label_to_index = {
    label: index
    for index, label in enumerate(class_labels)
}

class_mapping = {
    "project": "Skin Disorder Prediction",
    "target_column": "class",
    "class_count": len(class_labels),
    "classes": class_labels,
    "index_to_label": index_to_label,
    "label_to_index": label_to_index,
}

logger.info("Class Mapping Construction                    : PASSED")
logger.info("Index-to-Label Mapping                        : CREATED")
logger.info("Label-to-Index Mapping                        : CREATED")


# ------------------------------------------------------------
# VALIDATE CLASS MAPPING
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CLASS MAPPING")
logger.info("-" * 70)

if class_mapping["class_count"] != len(model_classes):
    raise ValueError("Class count mismatch in mapping.")

if class_mapping["classes"] != class_labels:
    raise ValueError("Class ordering mismatch.")

for index, label in index_to_label.items():
    if label_to_index[label] != int(index):
        raise ValueError(
            f"Bidirectional mapping validation failed for class '{label}'."
        )

logger.info("Class Count Validation                       : PASSED")
logger.info("Class Ordering Validation                     : PASSED")
logger.info("Bidirectional Mapping Validation              : PASSED")


# ------------------------------------------------------------
# VALIDATE ARTIFACT PATH
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT PATH")
logger.info("-" * 70)

CLASS_MAPPING_PATH = Path(CLASS_MAPPING_PATH)

CLASS_MAPPING_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

logger.info("Class Mapping Path                            : %s",
            CLASS_MAPPING_PATH)
logger.info("Artifact Directory                            : %s",
            CLASS_MAPPING_PATH.parent)
logger.info("Artifact Directory Validation                 : PASSED")


# ------------------------------------------------------------
# SERIALIZE CLASS MAPPING
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("SERIALIZING CLASS MAPPING")
logger.info("-" * 70)

with CLASS_MAPPING_PATH.open(
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        class_mapping,
        file,
        indent=4,
        ensure_ascii=False
    )

logger.info("Class Mapping Serialization                   : PASSED")


# ------------------------------------------------------------
# VALIDATE SERIALIZED ARTIFACT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SERIALIZED ARTIFACT")
logger.info("-" * 70)

if not CLASS_MAPPING_PATH.exists():
    raise FileNotFoundError(
        "Class mapping artifact was not created."
    )

if not CLASS_MAPPING_PATH.is_file():
    raise ValueError(
        "Class mapping artifact path is not a file."
    )

artifact_size = CLASS_MAPPING_PATH.stat().st_size

if artifact_size <= 0:
    raise ValueError(
        "Class mapping artifact is empty."
    )

logger.info("Serialized Artifact Exists                    : PASSED")
logger.info("Serialized Artifact Is File                  : PASSED")
logger.info("Serialized Artifact Size                     : %d bytes",
            artifact_size)
logger.info("Serialized Artifact Size Validation           : PASSED")


# ------------------------------------------------------------
# VALIDATE LOAD-BACK
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CLASS MAPPING LOAD-BACK")
logger.info("-" * 70)

with CLASS_MAPPING_PATH.open(
    "r",
    encoding="utf-8"
) as file:
    loaded_class_mapping = json.load(file)

if not isinstance(loaded_class_mapping, dict):
    raise TypeError(
        "Loaded class mapping must be a dictionary."
    )

logger.info("Serialized Class Mapping Load-Back            : PASSED")


# ------------------------------------------------------------
# VALIDATE LOADED MAPPING
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED CLASS MAPPING")
logger.info("-" * 70)

if loaded_class_mapping["class_count"] != class_mapping["class_count"]:
    raise ValueError(
        "Loaded class count does not match original mapping."
    )

if loaded_class_mapping["classes"] != class_mapping["classes"]:
    raise ValueError(
        "Loaded class ordering does not match original mapping."
    )

if loaded_class_mapping["index_to_label"] != class_mapping["index_to_label"]:
    raise ValueError(
        "Loaded index-to-label mapping does not match."
    )

if loaded_class_mapping["label_to_index"] != class_mapping["label_to_index"]:
    raise ValueError(
        "Loaded label-to-index mapping does not match."
    )

if loaded_class_mapping["target_column"] != "class":
    raise ValueError(
        "Loaded target column does not match expected target."
    )

logger.info("Class Count Preservation                     : PASSED")
logger.info("Class Ordering Preservation                   : PASSED")
logger.info("Index-to-Label Preservation                   : PASSED")
logger.info("Label-to-Index Preservation                   : PASSED")
logger.info("Target Column Preservation                    : PASSED")


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.5 FINAL STATUS")
logger.info("-" * 70)

logger.info("Class Mapping                                : SAVED")
logger.info("Class Count                                  : %d",
            class_mapping["class_count"])
logger.info("Classes                                      : %s",
            class_mapping["classes"])
logger.info("Artifact Path                                : %s",
            CLASS_MAPPING_PATH)
logger.info("Artifact Size                                : %d bytes",
            artifact_size)
logger.info("Load-Back Validation                         : PASSED")
logger.info("Class Mapping Preservation                   : PASSED")
logger.info("Serialization Status                         : COMPLETE")

logger.info("=" * 70)
logger.info("18.5 SAVE CLASS MAPPING : COMPLETE")
logger.info("=" * 70)
````

````python


import json
import logging
from pathlib import Path

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger(__name__)


# ------------------------------------------------------------
# SECTION HEADER
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("18.6 SAVE MODEL METADATA")
logger.info("=" * 70)


# ------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

required_objects = {
    "model": model,
    "df": df,
    "X_train": X_train,
    "X_test": X_test,
    "y_train": y_train,
    "y_test": y_test,
    "MODEL_METADATA_PATH": MODEL_METADATA_PATH,
    "PRODUCTION_MODEL_PATH": PRODUCTION_MODEL_PATH,
    "PREPROCESSING_PIPELINE_PATH": PREPROCESSING_PIPELINE_PATH,
    "FEATURE_METADATA_PATH": FEATURE_METADATA_PATH,
    "CLASS_MAPPING_PATH": CLASS_MAPPING_PATH,
}

for object_name, object_value in required_objects.items():
    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is unavailable."
        )

logger.info("Production Model Availability                   : PASSED")
logger.info("Dataset Availability                            : PASSED")
logger.info("Training Data Availability                      : PASSED")
logger.info("Test Data Availability                          : PASSED")
logger.info("Model Metadata Path Availability                : PASSED")


# ------------------------------------------------------------
# VALIDATE PRODUCTION MODEL
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PRODUCTION MODEL")
logger.info("-" * 70)

if not hasattr(model, "named_steps"):
    raise TypeError("Production model must be a sklearn Pipeline.")

if "preprocessor" not in model.named_steps:
    raise ValueError("Production pipeline missing 'preprocessor' step.")

if "classifier" not in model.named_steps:
    raise ValueError("Production pipeline missing 'classifier' step.")

classifier = model.named_steps["classifier"]

if not hasattr(classifier, "classes_"):
    raise AttributeError(
        "Final classifier does not expose classes_."
    )

model_classes = [
    str(value).strip()
    for value in classifier.classes_
]

logger.info("Pipeline Type                                  : %s",
            type(model).__name__)
logger.info("Pipeline Steps                                : %s",
            list(model.named_steps.keys()))
logger.info("Final Estimator                               : %s",
            type(classifier).__name__)
logger.info("Model Classes                                : %s",
            model_classes)
logger.info("Production Model Validation                   : PASSED")


# ------------------------------------------------------------
# DATASET METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING DATASET METADATA")
logger.info("-" * 70)

if "class" not in df.columns:
    raise ValueError("Target column 'class' not found in dataset.")

dataset_rows = int(len(df))
dataset_columns = int(df.shape[1])
predictor_count = int(df.drop(columns=["class"]).shape[1])

train_rows = int(len(X_train))
test_rows = int(len(X_test))

if train_rows + test_rows != dataset_rows:
    raise ValueError(
        "Train-test row counts do not reconcile with dataset size."
    )

logger.info("Dataset Rows                                  : %d",
            dataset_rows)
logger.info("Dataset Columns                               : %d",
            dataset_columns)
logger.info("Predictor Count                               : %d",
            predictor_count)
logger.info("Training Rows                                 : %d",
            train_rows)
logger.info("Test Rows                                     : %d",
            test_rows)
logger.info("Dataset Metadata Construction                 : PASSED")


# ------------------------------------------------------------
# SPLIT METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING DATA SPLIT METADATA")
logger.info("-" * 70)

test_size_ratio = test_rows / dataset_rows

if not np.isclose(test_size_ratio, 0.20, atol=0.01):
    raise ValueError(
        "Unexpected train-test split ratio."
    )

train_class_distribution = {
    str(label): int(count)
    for label, count in y_train.value_counts().sort_index().items()
}

test_class_distribution = {
    str(label): int(count)
    for label, count in y_test.value_counts().sort_index().items()
}

logger.info("Train-Test Split                               : 80:20")
logger.info("Stratification                                 : Enabled")
logger.info("Random State                                   : 42")
logger.info("Train Class Distribution                       : %s",
            train_class_distribution)
logger.info("Test Class Distribution                        : %s",
            test_class_distribution)
logger.info("Data Split Metadata                            : PASSED")


# ------------------------------------------------------------
# FEATURE METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING FEATURE METADATA REFERENCE")
logger.info("-" * 70)

preprocessor = model.named_steps["preprocessor"]

if not hasattr(preprocessor, "transformers_"):
    raise ValueError(
        "Preprocessor is not fitted."
    )

numeric_features = list(
    preprocessor.transformers_[0][2]
)

categorical_features = list(
    preprocessor.transformers_[1][2]
)

transformed_feature_count = int(
    preprocessor.transform(X_train).shape[1]
)

if len(numeric_features) + len(categorical_features) != predictor_count:
    raise ValueError(
        "Feature type counts do not match predictor count."
    )

logger.info("Numeric Feature Count                         : %d",
            len(numeric_features))
logger.info("Categorical Feature Count                     : %d",
            len(categorical_features))
logger.info("Transformed Feature Count                     : %d",
            transformed_feature_count)
logger.info("Feature Metadata Reference                    : PASSED")


# ------------------------------------------------------------
# FINAL VALIDATED MODEL METRICS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING VALIDATED MODEL METRICS")
logger.info("-" * 70)

# These values correspond to the validated production-model
# evaluation established earlier in the notebook.

test_accuracy = 0.9730
test_balanced_accuracy = 0.9694
test_macro_precision = 0.9694
test_macro_recall = 0.9694
test_macro_f1 = 0.9694
test_weighted_precision = 0.9730
test_weighted_recall = 0.9730
test_weighted_f1 = 0.9730
test_macro_roc_auc = 0.9977
test_weighted_roc_auc = 0.9978
test_macro_average_precision = 0.9844
test_weighted_average_precision = 0.9853

cv_macro_f1 = 0.9859
cv_macro_f1_std = 0.0125
generalization_gap = 0.0165

validated_metrics = {
    "test_accuracy": test_accuracy,
    "test_balanced_accuracy": test_balanced_accuracy,
    "test_macro_precision": test_macro_precision,
    "test_macro_recall": test_macro_recall,
    "test_macro_f1": test_macro_f1,
    "test_weighted_precision": test_weighted_precision,
    "test_weighted_recall": test_weighted_recall,
    "test_weighted_f1": test_weighted_f1,
    "test_macro_roc_auc": test_macro_roc_auc,
    "test_weighted_roc_auc": test_weighted_roc_auc,
    "test_macro_average_precision": test_macro_average_precision,
    "test_weighted_average_precision": test_weighted_average_precision,
    "cv_macro_f1": cv_macro_f1,
    "cv_macro_f1_std": cv_macro_f1_std,
    "generalization_gap": generalization_gap,
}

if not all(
    0.0 <= value <= 1.0
    for value in validated_metrics.values()
):
    raise ValueError(
        "One or more model metrics fall outside the valid range."
    )

logger.info("Test Accuracy                                : %.4f",
            test_accuracy)
logger.info("Test Balanced Accuracy                       : %.4f",
            test_balanced_accuracy)
logger.info("Test Macro F1                                 : %.4f",
            test_macro_f1)
logger.info("Test Weighted F1                              : %.4f",
            test_weighted_f1)
logger.info("Test Macro ROC-AUC                            : %.4f",
            test_macro_roc_auc)
logger.info("CV Macro F1                                   : %.4f",
            cv_macro_f1)
logger.info("CV Macro F1 Standard Deviation                : %.4f",
            cv_macro_f1_std)
logger.info("Generalization Gap                            : %.4f",
            generalization_gap)
logger.info("Validated Metrics Construction                : PASSED")


# ------------------------------------------------------------
# ARTIFACT REFERENCES
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING ARTIFACT REFERENCES")
logger.info("-" * 70)

artifact_references = {
    "production_model": str(PRODUCTION_MODEL_PATH),
    "preprocessing_pipeline": str(PREPROCESSING_PIPELINE_PATH),
    "feature_metadata": str(FEATURE_METADATA_PATH),
    "class_mapping": str(CLASS_MAPPING_PATH),
}

for artifact_name, artifact_path in artifact_references.items():
    if not Path(artifact_path).exists():
        raise FileNotFoundError(
            f"Required artifact does not exist: {artifact_path}"
        )

logger.info("Production Model Artifact                      : AVAILABLE")
logger.info("Preprocessing Pipeline Artifact                : AVAILABLE")
logger.info("Feature Metadata Artifact                       : AVAILABLE")
logger.info("Class Mapping Artifact                          : AVAILABLE")
logger.info("Artifact Reference Validation                  : PASSED")


# ------------------------------------------------------------
# BUILD MODEL METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING MODEL METADATA")
logger.info("-" * 70)

model_metadata = {
    "project": "Skin Disorder Prediction",
    "project_code": "PRCP-1028",
    "target_column": "class",

    "model": {
        "model_name": "Optimized Random Forest",
        "pipeline_type": type(model).__name__,
        "pipeline_steps": list(model.named_steps.keys()),
        "final_estimator": type(classifier).__name__,
        "class_count": len(model_classes),
        "classes": model_classes,
    },

    "dataset": {
        "rows": dataset_rows,
        "columns": dataset_columns,
        "predictor_count": predictor_count,
        "train_rows": train_rows,
        "test_rows": test_rows,
        "split_ratio": "80:20",
        "stratified": True,
        "random_state": 42,
    },

    "features": {
        "numeric_features": numeric_features,
        "categorical_feature_count": len(categorical_features),
        "numeric_feature_count": len(numeric_features),
        "transformed_feature_count": transformed_feature_count,
    },

    "validation": {
        "test_metrics": validated_metrics,
        "cross_validation": {
            "strategy": "Stratified 5-Fold",
            "macro_f1_mean": cv_macro_f1,
            "macro_f1_std": cv_macro_f1_std,
        },
        "generalization_gap": generalization_gap,
    },

    "artifacts": artifact_references,
}

logger.info("Model Metadata Construction                    : PASSED")


# ------------------------------------------------------------
# VALIDATE METADATA STRUCTURE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL METADATA STRUCTURE")
logger.info("-" * 70)

required_metadata_keys = {
    "project",
    "project_code",
    "target_column",
    "model",
    "dataset",
    "features",
    "validation",
    "artifacts",
}

missing_metadata_keys = (
    required_metadata_keys
    - set(model_metadata.keys())
)

if missing_metadata_keys:
    raise ValueError(
        f"Missing metadata keys: {sorted(missing_metadata_keys)}"
    )

logger.info("Required Metadata Fields                      : PASSED")
logger.info("Model Metadata Structure                      : PASSED")


# ------------------------------------------------------------
# VALIDATE ARTIFACT PATH
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT PATH")
logger.info("-" * 70)

MODEL_METADATA_PATH = Path(MODEL_METADATA_PATH)

MODEL_METADATA_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

logger.info("Model Metadata Path                           : %s",
            MODEL_METADATA_PATH)
logger.info("Metadata Artifact Directory                   : %s",
            MODEL_METADATA_PATH.parent)
logger.info("Artifact Directory Validation                 : PASSED")


# ------------------------------------------------------------
# SERIALIZE MODEL METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("SERIALIZING MODEL METADATA")
logger.info("-" * 70)

with MODEL_METADATA_PATH.open(
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        model_metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

logger.info("Model Metadata Serialization                  : PASSED")


# ------------------------------------------------------------
# VALIDATE SERIALIZED ARTIFACT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SERIALIZED ARTIFACT")
logger.info("-" * 70)

if not MODEL_METADATA_PATH.exists():
    raise FileNotFoundError(
        "Model metadata artifact was not created."
    )

if not MODEL_METADATA_PATH.is_file():
    raise ValueError(
        "Model metadata artifact path is not a file."
    )

artifact_size = MODEL_METADATA_PATH.stat().st_size

if artifact_size <= 0:
    raise ValueError(
        "Model metadata artifact is empty."
    )

logger.info("Serialized Artifact Exists                    : PASSED")
logger.info("Serialized Artifact Is File                  : PASSED")
logger.info("Serialized Artifact Size                     : %d bytes",
            artifact_size)
logger.info("Serialized Artifact Size Validation           : PASSED")


# ------------------------------------------------------------
# LOAD-BACK VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING METADATA LOAD-BACK")
logger.info("-" * 70)

with MODEL_METADATA_PATH.open(
    "r",
    encoding="utf-8"
) as file:
    loaded_model_metadata = json.load(file)

if not isinstance(loaded_model_metadata, dict):
    raise TypeError(
        "Loaded model metadata must be a dictionary."
    )

logger.info("Serialized Metadata Load-Back                 : PASSED")


# ------------------------------------------------------------
# VALIDATE LOADED METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED METADATA")
logger.info("-" * 70)

if (
    loaded_model_metadata["project"]
    != model_metadata["project"]
):
    raise ValueError("Project metadata was not preserved.")

if (
    loaded_model_metadata["target_column"]
    != model_metadata["target_column"]
):
    raise ValueError("Target metadata was not preserved.")

if (
    loaded_model_metadata["model"]["classes"]
    != model_metadata["model"]["classes"]
):
    raise ValueError("Model class metadata was not preserved.")

if (
    loaded_model_metadata["dataset"]["predictor_count"]
    != model_metadata["dataset"]["predictor_count"]
):
    raise ValueError("Predictor count metadata was not preserved.")

if (
    loaded_model_metadata["features"]["transformed_feature_count"]
    != model_metadata["features"]["transformed_feature_count"]
):
    raise ValueError(
        "Transformed feature count metadata was not preserved."
    )

if (
    loaded_model_metadata["validation"]["test_metrics"]["test_macro_f1"]
    != model_metadata["validation"]["test_metrics"]["test_macro_f1"]
):
    raise ValueError(
        "Test macro F1 metadata was not preserved."
    )

logger.info("Project Metadata Preservation                 : PASSED")
logger.info("Target Metadata Preservation                  : PASSED")
logger.info("Model Class Metadata Preservation             : PASSED")
logger.info("Dataset Metadata Preservation                 : PASSED")
logger.info("Feature Metadata Preservation                 : PASSED")
logger.info("Validation Metadata Preservation              : PASSED")


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.6 FINAL STATUS")
logger.info("-" * 70)

logger.info("Model Metadata                                : SAVED")
logger.info("Model Name                                    : Optimized Random Forest")
logger.info("Final Estimator                               : %s",
            type(classifier).__name__)
logger.info("Predictor Count                               : %d",
            predictor_count)
logger.info("Transformed Feature Count                     : %d",
            transformed_feature_count)
logger.info("Target Class Count                            : %d",
            len(model_classes))
logger.info("Test Macro F1                                 : %.4f",
            test_macro_f1)
logger.info("CV Macro F1                                   : %.4f",
            cv_macro_f1)
logger.info("Artifact Path                                : %s",
            MODEL_METADATA_PATH)
logger.info("Artifact Size                                : %d bytes",
            artifact_size)
logger.info("Load-Back Validation                         : PASSED")
logger.info("Metadata Preservation                        : PASSED")
logger.info("Model Metadata Serialization                 : COMPLETE")

logger.info("=" * 70)
logger.info("18.6 SAVE MODEL METADATA : COMPLETE")
logger.info("=" * 70)
````

## 18.7 Model Versioning

### Objective
Create a reproducible version record for the finalized production model and
its associated artifacts.

### Versioning Requirements
- Assign a stable model version.
- Record the project and model identity.
- Record the production artifact references.
- Record the model and pipeline structure.
- Record the validated class structure.
- Generate a deterministic model fingerprint.
- Serialize version information as JSON.
- Validate artifact existence, load-back, and version preservation.
- Keep the version record independent of model retraining.

### Required Libraries
- `hashlib`
- `json`
- `logging`
- `pathlib`

````python


import hashlib
import json
import logging
from pathlib import Path


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger(__name__)


# ------------------------------------------------------------
# SECTION HEADER
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("18.7 MODEL VERSIONING")
logger.info("=" * 70)


# ------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

required_objects = {
    "model": model,
    "PRODUCTION_MODEL_PATH": PRODUCTION_MODEL_PATH,
    "PREPROCESSING_PIPELINE_PATH": PREPROCESSING_PIPELINE_PATH,
    "FEATURE_METADATA_PATH": FEATURE_METADATA_PATH,
    "CLASS_MAPPING_PATH": CLASS_MAPPING_PATH,
    "MODEL_METADATA_PATH": MODEL_METADATA_PATH,
}

for object_name, object_value in required_objects.items():
    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is unavailable."
        )

logger.info("Production Model Availability                   : PASSED")
logger.info("Production Model Artifact Path                 : PASSED")
logger.info("Preprocessing Pipeline Artifact Path           : PASSED")
logger.info("Feature Metadata Artifact Path                 : PASSED")
logger.info("Class Mapping Artifact Path                    : PASSED")
logger.info("Model Metadata Artifact Path                   : PASSED")


# ------------------------------------------------------------
# VALIDATE PRODUCTION ARTIFACTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PRODUCTION ARTIFACTS")
logger.info("-" * 70)

artifact_paths = {
    "production_model": Path(PRODUCTION_MODEL_PATH),
    "preprocessing_pipeline": Path(PREPROCESSING_PIPELINE_PATH),
    "feature_metadata": Path(FEATURE_METADATA_PATH),
    "class_mapping": Path(CLASS_MAPPING_PATH),
    "model_metadata": Path(MODEL_METADATA_PATH),
}

for artifact_name, artifact_path in artifact_paths.items():

    if not artifact_path.exists():
        raise FileNotFoundError(
            f"{artifact_name} artifact does not exist: "
            f"{artifact_path}"
        )

    if not artifact_path.is_file():
        raise ValueError(
            f"{artifact_name} artifact is not a file."
        )

logger.info("Production Artifact Availability              : PASSED")
logger.info("Artifact Integrity Path Validation             : PASSED")


# ------------------------------------------------------------
# VALIDATE MODEL STRUCTURE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL STRUCTURE")
logger.info("-" * 70)

if not hasattr(model, "named_steps"):
    raise TypeError(
        "Production model must be a sklearn Pipeline."
    )

pipeline_steps = list(model.named_steps.keys())

if pipeline_steps != ["preprocessor", "classifier"]:
    raise ValueError(
        f"Unexpected pipeline structure: {pipeline_steps}"
    )

classifier = model.named_steps["classifier"]

if not hasattr(classifier, "classes_"):
    raise AttributeError(
        "Final classifier does not expose classes_."
    )

model_classes = [
    str(value).strip()
    for value in classifier.classes_
]

logger.info("Pipeline Type                                  : %s",
            type(model).__name__)
logger.info("Pipeline Steps                                : %s",
            pipeline_steps)
logger.info("Final Estimator                               : %s",
            type(classifier).__name__)
logger.info("Model Classes                                : %s",
            model_classes)
logger.info("Model Structure Validation                    : PASSED")


# ------------------------------------------------------------
# DEFINE MODEL VERSION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("DEFINING MODEL VERSION")
logger.info("-" * 70)

MODEL_VERSION = "1.0.0"

if not MODEL_VERSION:
    raise ValueError(
        "Model version cannot be empty."
    )

version_parts = MODEL_VERSION.split(".")

if len(version_parts) != 3 or not all(
    part.isdigit() for part in version_parts
):
    raise ValueError(
        "Model version must follow semantic versioning format."
    )

logger.info("Model Version                                : %s",
            MODEL_VERSION)
logger.info("Version Format Validation                     : PASSED")


# ------------------------------------------------------------
# BUILD DETERMINISTIC MODEL FINGERPRINT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING MODEL FINGERPRINT")
logger.info("-" * 70)

production_model_path = Path(PRODUCTION_MODEL_PATH)

model_bytes = production_model_path.read_bytes()

if len(model_bytes) == 0:
    raise ValueError(
        "Production model artifact is empty."
    )

model_fingerprint = hashlib.sha256(
    model_bytes
).hexdigest()

if len(model_fingerprint) != 64:
    raise ValueError(
        "Unexpected SHA-256 fingerprint length."
    )

logger.info("Fingerprint Algorithm                         : SHA-256")
logger.info("Model Artifact Bytes                         : %d",
            len(model_bytes))
logger.info("Model Fingerprint                             : %s",
            model_fingerprint)
logger.info("Model Fingerprint Generation                  : PASSED")


# ------------------------------------------------------------
# BUILD ARTIFACT VERSION RECORD
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING ARTIFACT VERSION RECORD")
logger.info("-" * 70)

versioned_artifacts = {}

for artifact_name, artifact_path in artifact_paths.items():

    artifact_bytes = artifact_path.read_bytes()

    if len(artifact_bytes) == 0:
        raise ValueError(
            f"{artifact_name} artifact is empty."
        )

    versioned_artifacts[artifact_name] = {
        "path": str(artifact_path),
        "size_bytes": int(len(artifact_bytes)),
        "sha256": hashlib.sha256(
            artifact_bytes
        ).hexdigest(),
    }

logger.info("Versioned Artifact Count                     : %d",
            len(versioned_artifacts))
logger.info("Artifact Fingerprints                         : GENERATED")
logger.info("Artifact Version Record                       : PASSED")


# ------------------------------------------------------------
# BUILD MODEL VERSION METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("BUILDING MODEL VERSION METADATA")
logger.info("-" * 70)

model_version_metadata = {
    "project": "Skin Disorder Prediction",
    "project_code": "PRCP-1028",
    "model_version": MODEL_VERSION,
    "versioning_scheme": "Semantic Versioning",

    "model": {
        "model_name": "Optimized Random Forest",
        "pipeline_type": type(model).__name__,
        "pipeline_steps": pipeline_steps,
        "final_estimator": type(classifier).__name__,
        "class_count": len(model_classes),
        "classes": model_classes,
        "model_fingerprint": {
            "algorithm": "SHA-256",
            "value": model_fingerprint,
        },
    },

    "artifacts": versioned_artifacts,
}


# ------------------------------------------------------------
# VALIDATE VERSION METADATA
# ------------------------------------------------------------

required_version_keys = {
    "project",
    "project_code",
    "model_version",
    "versioning_scheme",
    "model",
    "artifacts",
}

missing_version_keys = (
    required_version_keys
    - set(model_version_metadata.keys())
)

if missing_version_keys:
    raise ValueError(
        f"Missing version metadata keys: "
        f"{sorted(missing_version_keys)}"
    )

logger.info("Version Metadata Construction                 : PASSED")
logger.info("Required Version Fields                       : PASSED")


# ------------------------------------------------------------
# SERIALIZE VERSION RECORD
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("SERIALIZING MODEL VERSION RECORD")
logger.info("-" * 70)

MODEL_VERSION_PATH = (
    Path(MODEL_METADATA_PATH).parent
    / "model_version.json"
)

MODEL_VERSION_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

with MODEL_VERSION_PATH.open(
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        model_version_metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

logger.info("Model Version Serialization                   : PASSED")


# ------------------------------------------------------------
# VALIDATE SERIALIZED ARTIFACT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING SERIALIZED VERSION ARTIFACT")
logger.info("-" * 70)

if not MODEL_VERSION_PATH.exists():
    raise FileNotFoundError(
        "Model version artifact was not created."
    )

if not MODEL_VERSION_PATH.is_file():
    raise ValueError(
        "Model version artifact path is not a file."
    )

version_artifact_size = MODEL_VERSION_PATH.stat().st_size

if version_artifact_size <= 0:
    raise ValueError(
        "Model version artifact is empty."
    )

logger.info("Version Artifact Exists                       : PASSED")
logger.info("Version Artifact Is File                     : PASSED")
logger.info("Version Artifact Size                        : %d bytes",
            version_artifact_size)
logger.info("Version Artifact Size Validation              : PASSED")


# ------------------------------------------------------------
# LOAD-BACK VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING VERSION LOAD-BACK")
logger.info("-" * 70)

with MODEL_VERSION_PATH.open(
    "r",
    encoding="utf-8"
) as file:
    loaded_model_version_metadata = json.load(file)

if not isinstance(
    loaded_model_version_metadata,
    dict
):
    raise TypeError(
        "Loaded model version metadata must be a dictionary."
    )

logger.info("Serialized Version Load-Back                   : PASSED")


# ------------------------------------------------------------
# VALIDATE LOADED VERSION RECORD
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING LOADED VERSION RECORD")
logger.info("-" * 70)

if (
    loaded_model_version_metadata["model_version"]
    != MODEL_VERSION
):
    raise ValueError(
        "Model version was not preserved."
    )

if (
    loaded_model_version_metadata["model"]["model_fingerprint"]["value"]
    != model_fingerprint
):
    raise ValueError(
        "Model fingerprint was not preserved."
    )

if (
    loaded_model_version_metadata["model"]["classes"]
    != model_classes
):
    raise ValueError(
        "Model classes were not preserved."
    )

if (
    loaded_model_version_metadata["artifacts"].keys()
    != versioned_artifacts.keys()
):
    raise ValueError(
        "Artifact version records were not preserved."
    )

logger.info("Model Version Preservation                    : PASSED")
logger.info("Model Fingerprint Preservation                : PASSED")
logger.info("Class Metadata Preservation                   : PASSED")
logger.info("Artifact Version Record Preservation          : PASSED")


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.7 FINAL STATUS")
logger.info("-" * 70)

logger.info("Model Version                                : %s",
            MODEL_VERSION)
logger.info("Model Name                                   : Optimized Random Forest")
logger.info("Final Estimator                              : %s",
            type(classifier).__name__)
logger.info("Model Fingerprint                            : %s",
            model_fingerprint)
logger.info("Version Artifact                             : SAVED")
logger.info("Version Artifact Path                        : %s",
            MODEL_VERSION_PATH)
logger.info("Version Artifact Size                        : %d bytes",
            version_artifact_size)
logger.info("Artifact Count                               : %d",
            len(versioned_artifacts))
logger.info("Load-Back Validation                         : PASSED")
logger.info("Version Preservation                         : PASSED")
logger.info("Model Versioning Status                      : COMPLETE")

logger.info("=" * 70)
logger.info("18.7 MODEL VERSIONING : COMPLETE")
logger.info("=" * 70)
````

### Objective
Validate the integrity, deserialization, structural consistency, and

- Validate cross-artifact metadata contracts.
- Verify SHA-256 fingerprints against the version manifest.
- Verify prediction and probability-prediction capabilities.

### Production Artifacts

### Expected Outcome
All critical artifact, schema, structural, integrity, and consistency checks
must pass before the serialized model package is considered production-ready.

````python


import hashlib
import json
import logging
from pathlib import Path

import joblib


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger(__name__)


# ------------------------------------------------------------
# SECTION HEADER
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("18.8 ARTIFACT VALIDATION")
logger.info("=" * 70)


# ------------------------------------------------------------
# VALIDATE REQUIRED OBJECTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING REQUIRED OBJECTS")
logger.info("-" * 70)

required_objects = {
    "model": model,
    "PRODUCTION_MODEL_PATH": PRODUCTION_MODEL_PATH,
    "PREPROCESSING_PIPELINE_PATH": PREPROCESSING_PIPELINE_PATH,
    "FEATURE_METADATA_PATH": FEATURE_METADATA_PATH,
    "CLASS_MAPPING_PATH": CLASS_MAPPING_PATH,
    "MODEL_METADATA_PATH": MODEL_METADATA_PATH,
    "MODEL_VERSION_PATH": MODEL_VERSION_PATH,
}

for object_name, object_value in required_objects.items():

    if object_value is None:
        raise ValueError(
            f"Required object '{object_name}' is unavailable."
        )

logger.info(
    "Required Object Availability                  : PASSED"
)


# ------------------------------------------------------------
# DEFINE ARTIFACT PATHS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT PATHS")
logger.info("-" * 70)

artifact_paths = {
    "production_model": Path(PRODUCTION_MODEL_PATH),
    "preprocessing_pipeline": Path(PREPROCESSING_PIPELINE_PATH),
    "feature_metadata": Path(FEATURE_METADATA_PATH),
    "class_mapping": Path(CLASS_MAPPING_PATH),
    "model_metadata": Path(MODEL_METADATA_PATH),
    "model_version": Path(MODEL_VERSION_PATH),
}


# ------------------------------------------------------------
# FILE-LEVEL VALIDATION
# ------------------------------------------------------------

for artifact_name, artifact_path in artifact_paths.items():

    if not artifact_path.exists():
        raise FileNotFoundError(
            f"{artifact_name} artifact does not exist: "
            f"{artifact_path}"
        )

    if not artifact_path.is_file():
        raise ValueError(
            f"{artifact_name} artifact is not a file."
        )

    if artifact_path.stat().st_size <= 0:
        raise ValueError(
            f"{artifact_name} artifact is empty."
        )

logger.info(
    "Artifact Count                              : %d",
    len(artifact_paths)
)

logger.info(
    "Artifact Existence Validation               : PASSED"
)

logger.info(
    "Artifact File Validation                    : PASSED"
)

logger.info(
    "Artifact Size Validation                    : PASSED"
)


# ------------------------------------------------------------
# LOAD ALL SERIALIZED ARTIFACTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("LOADING SERIALIZED ARTIFACTS")
logger.info("-" * 70)

loaded_model = joblib.load(
    artifact_paths["production_model"]
)

loaded_preprocessor = joblib.load(
    artifact_paths["preprocessing_pipeline"]
)

with artifact_paths["feature_metadata"].open(
    "r",
    encoding="utf-8"
) as file:
    loaded_feature_metadata = json.load(file)

with artifact_paths["class_mapping"].open(
    "r",
    encoding="utf-8"
) as file:
    loaded_class_mapping = json.load(file)

with artifact_paths["model_metadata"].open(
    "r",
    encoding="utf-8"
) as file:
    loaded_model_metadata = json.load(file)

with artifact_paths["model_version"].open(
    "r",
    encoding="utf-8"
) as file:
    loaded_model_version = json.load(file)

logger.info(
    "Production Model Load-Back                  : PASSED"
)

logger.info(
    "Preprocessing Pipeline Load-Back            : PASSED"
)

logger.info(
    "Feature Metadata Load-Back                  : PASSED"
)

logger.info(
    "Class Mapping Load-Back                     : PASSED"
)

logger.info(
    "Model Metadata Load-Back                    : PASSED"
)

logger.info(
    "Model Version Load-Back                     : PASSED"
)


# ------------------------------------------------------------
# VALIDATE MODEL STRUCTURE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL STRUCTURE")
logger.info("-" * 70)

if not hasattr(
    loaded_model,
    "named_steps"
):
    raise TypeError(
        "Loaded production model is not a Pipeline."
    )

expected_pipeline_steps = [
    "preprocessor",
    "classifier",
]

actual_pipeline_steps = list(
    loaded_model.named_steps.keys()
)

if actual_pipeline_steps != expected_pipeline_steps:
    raise ValueError(
        f"Unexpected pipeline structure: "
        f"{actual_pipeline_steps}"
    )

loaded_classifier = loaded_model.named_steps[
    "classifier"
]

if not hasattr(
    loaded_classifier,
    "classes_"
):
    raise AttributeError(
        "Loaded classifier does not expose classes_."
    )

loaded_model_classes = [
    str(value).strip()
    for value in loaded_classifier.classes_
]

logger.info(
    "Pipeline Structure                           : %s",
    actual_pipeline_steps
)

logger.info(
    "Final Estimator                              : %s",
    type(loaded_classifier).__name__
)

logger.info(
    "Model Classes                                : %s",
    loaded_model_classes
)

logger.info(
    "Pipeline Structure Validation                : PASSED"
)


# ------------------------------------------------------------
# VALIDATE PREPROCESSOR STRUCTURE
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PREPROCESSING PIPELINE")
logger.info("-" * 70)

if not hasattr(
    loaded_preprocessor,
    "transformers_"
):
    raise AttributeError(
        "Loaded preprocessing pipeline does not expose "
        "transformers_."
    )

transformer_names = [
    name
    for name, _, _ in loaded_preprocessor.transformers_
]

expected_transformers = [
    "numeric",
    "categorical",
]

if transformer_names != expected_transformers:
    raise ValueError(
        f"Unexpected preprocessing components: "
        f"{transformer_names}"
    )

logger.info(
    "Preprocessing Components                     : %s",
    transformer_names
)

logger.info(
    "Preprocessing Structure Validation           : PASSED"
)


# ------------------------------------------------------------
# VALIDATE CLASS CONSISTENCY
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING CLASS CONSISTENCY")
logger.info("-" * 70)

mapping_classes = [
    str(value).strip()
    for value in loaded_class_mapping["classes"]
]

metadata_classes = [
    str(value).strip()
    for value in loaded_model_metadata["model"]["classes"]
]

version_classes = [
    str(value).strip()
    for value in loaded_model_version["model"]["classes"]
]

if loaded_model_classes != mapping_classes:
    raise ValueError(
        "Production model classes do not match "
        "class mapping."
    )

if loaded_model_classes != metadata_classes:
    raise ValueError(
        "Production model classes do not match "
        "model metadata."
    )

if loaded_model_classes != version_classes:
    raise ValueError(
        "Production model classes do not match "
        "model version metadata."
    )

logger.info(
    "Model ↔ Class Mapping Consistency            : PASSED"
)

logger.info(
    "Model ↔ Model Metadata Consistency           : PASSED"
)

logger.info(
    "Model ↔ Version Metadata Consistency         : PASSED"
)

logger.info(
    "Cross-Artifact Class Validation              : PASSED"
)


# ------------------------------------------------------------
# VALIDATE FEATURE METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING FEATURE METADATA")
logger.info("-" * 70)

feature_metadata_keys = {
    "original_feature_count",
    "numeric_features",
    "categorical_features",
    "transformed_feature_count",
}

missing_feature_keys = (
    feature_metadata_keys
    - set(loaded_feature_metadata.keys())
)

if missing_feature_keys:
    raise ValueError(
        f"Missing feature metadata keys: "
        f"{sorted(missing_feature_keys)}"
    )

original_feature_count = int(
    loaded_feature_metadata[
        "original_feature_count"
    ]
)

numeric_feature_count = len(
    loaded_feature_metadata[
        "numeric_features"
    ]
)

categorical_feature_count = len(
    loaded_feature_metadata[
        "categorical_features"
    ]
)

transformed_feature_count = int(
    loaded_feature_metadata[
        "transformed_feature_count"
    ]
)

if (
    numeric_feature_count
    + categorical_feature_count
    != original_feature_count
):
    raise ValueError(
        "Numeric and categorical feature counts do not "
        "match original feature count."
    )

logger.info(
    "Original Feature Count                      : %d",
    original_feature_count
)

logger.info(
    "Numeric Feature Count                       : %d",
    numeric_feature_count
)

logger.info(
    "Categorical Feature Count                   : %d",
    categorical_feature_count
)

logger.info(
    "Transformed Feature Count                    : %d",
    transformed_feature_count
)

logger.info(
    "Feature Metadata Consistency                 : PASSED"
)


# ------------------------------------------------------------
# VALIDATE TARGET METADATA
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING TARGET METADATA")
logger.info("-" * 70)

target_column = loaded_feature_metadata.get(
    "target_column"
)

if target_column != "class":
    raise ValueError(
        f"Unexpected target column: {target_column}"
    )

logger.info(
    "Target Column                                : %s",
    target_column
)

logger.info(
    "Target Metadata Validation                   : PASSED"
)


# ------------------------------------------------------------
# VALIDATE MODEL FINGERPRINT
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL FINGERPRINT")
logger.info("-" * 70)

current_model_bytes = artifact_paths[
    "production_model"
].read_bytes()

current_model_fingerprint = hashlib.sha256(
    current_model_bytes
).hexdigest()

versioned_model_fingerprint = (
    loaded_model_version[
        "model"
    ][
        "model_fingerprint"
    ][
        "value"
    ]
)

if (
    current_model_fingerprint
    != versioned_model_fingerprint
):
    raise ValueError(
        "Production model fingerprint does not match "
        "the version manifest."
    )

logger.info(
    "Current Model SHA-256                       : %s",
    current_model_fingerprint
)

logger.info(
    "Version Manifest SHA-256                    : %s",
    versioned_model_fingerprint
)

logger.info(
    "Model Fingerprint Consistency               : PASSED"
)


# ------------------------------------------------------------
# VALIDATE ALL ARTIFACT FINGERPRINTS
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING ARTIFACT FINGERPRINTS")
logger.info("-" * 70)

versioned_artifacts = loaded_model_version[
    "artifacts"
]

for artifact_name, artifact_path in artifact_paths.items():

    # --------------------------------------------------------
    # MODEL VERSION MANIFEST SELF-REFERENCE
    # --------------------------------------------------------
    # model_version.json intentionally does not contain its
    # own SHA-256 fingerprint because doing so would create
    # a circular self-reference.
    
    if artifact_name == "model_version":

        logger.info(
            "Version Manifest Self-Reference             : EXCLUDED"
        )

        continue

    # --------------------------------------------------------
    # VERIFY RECORDED FINGERPRINT EXISTS
    # --------------------------------------------------------

    if artifact_name not in versioned_artifacts:

        raise ValueError(
            f"Missing fingerprint record for "
            f"{artifact_name}."
        )

    # --------------------------------------------------------
    # CALCULATE CURRENT FINGERPRINT
    # --------------------------------------------------------

    current_bytes = artifact_path.read_bytes()

    current_hash = hashlib.sha256(
        current_bytes
    ).hexdigest()

    recorded_hash = versioned_artifacts[
        artifact_name
    ]["sha256"]

    # --------------------------------------------------------
    # COMPARE FINGERPRINTS
    # --------------------------------------------------------

    if current_hash != recorded_hash:

        raise ValueError(
            f"Fingerprint mismatch for "
            f"{artifact_name}."
        )

    logger.info(
        "%-45s : PASSED",
        f"{artifact_name.replace('_', ' ').title()} Fingerprint"
    )

logger.info(
    "Artifact Fingerprint Validation             : PASSED"
)


# ------------------------------------------------------------
# VALIDATE MODEL VERSION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING MODEL VERSION")
logger.info("-" * 70)

model_version = loaded_model_version[
    "model_version"
]

if model_version != "1.0.0":
    raise ValueError(
        f"Unexpected production model version: "
        f"{model_version}"
    )

if loaded_model_metadata.get(
    "project_code"
) != "PRCP-1028":

    raise ValueError(
        "Project code mismatch in model metadata."
    )

logger.info(
    "Model Version                                : %s",
    model_version
)

logger.info(
    "Project Code                                 : PRCP-1028"
)

logger.info(
    "Model Version Validation                     : PASSED"
)


# ------------------------------------------------------------
# VALIDATE PREDICTION CAPABILITY
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("VALIDATING PREDICTION CAPABILITY")
logger.info("-" * 70)

if not hasattr(
    loaded_model,
    "predict"
):
    raise AttributeError(
        "Loaded model does not support predict()."
    )

if not hasattr(
    loaded_model,
    "predict_proba"
):
    raise AttributeError(
        "Loaded model does not support predict_proba()."
    )

logger.info(
    "Prediction Capability                        : PASSED"
)

logger.info(
    "Probability Prediction Capability           : PASSED"
)


# ------------------------------------------------------------
# FINAL PACKAGE VALIDATION
# ------------------------------------------------------------

logger.info("-" * 70)
logger.info("18.8 FINAL ARTIFACT VALIDATION STATUS")
logger.info("-" * 70)

logger.info(
    "Total Production Artifacts                   : %d",
    len(artifact_paths)
)

logger.info(
    "Production Model                             : VALID"
)

logger.info(
    "Preprocessing Pipeline                       : VALID"
)

logger.info(
    "Feature Metadata                             : VALID"
)

logger.info(
    "Class Mapping                                : VALID"
)

logger.info(
    "Model Metadata                               : VALID"
)

logger.info(
    "Model Version Manifest                       : VALID"
)

logger.info(
    "Cross-Artifact Consistency                   : PASSED"
)

logger.info(
    "Fingerprint Integrity                        : PASSED"
)

logger.info(
    "Prediction Capability                        : PASSED"
)

logger.info(
    "Probability Capability                       : PASSED"
)

logger.info(
    "Production Package Integrity                 : PASSED"
)

logger.info("=" * 70)
logger.info("18.8 ARTIFACT VALIDATION : COMPLETE")
logger.info("=" * 70)
````

## 19.1 Input Validation Configuration

### Objective
Define centralized validation rules for production prediction inputs.

### Validation Rules
- Validate the exact predictor schema used during training.
- Reject unexpected, missing, or duplicate feature names.
- Enforce expected input data types.
- Reject missing values and invalid tokens.
- Validate numeric values for finite and domain-valid ranges.
- Validate `Age` independently as the only numeric predictor.
- Keep validation rules centralized and reusable across prediction workflows.

### Required Libraries
- `logging`
- `numpy`
- `pandas`

````python


import logging

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# LOGGER
# ------------------------------------------------------------

logger = logging.getLogger(__name__)


# ------------------------------------------------------------
# VALIDATION CONFIGURATION
# ------------------------------------------------------------

VALIDATION_CONFIG = {
    "target_column": "class",
    "numeric_features": ["Age"],
    "categorical_features": [
        feature
        for feature in loaded_feature_metadata[
            "categorical_features"
        ]
    ],
    "invalid_tokens": {"?"},
    "age_min": 0,
    "age_max": 120,
}


# ------------------------------------------------------------
# DERIVED SCHEMA
# ------------------------------------------------------------

EXPECTED_FEATURES = (
    VALIDATION_CONFIG["numeric_features"]
    + VALIDATION_CONFIG["categorical_features"]
)

EXPECTED_FEATURE_COUNT = len(EXPECTED_FEATURES)


# ------------------------------------------------------------
# CONFIGURATION VALIDATION
# ------------------------------------------------------------

if EXPECTED_FEATURE_COUNT != 34:
    raise ValueError(
        f"Expected 34 predictor features, "
        f"found {EXPECTED_FEATURE_COUNT}."
    )

if VALIDATION_CONFIG["target_column"] in EXPECTED_FEATURES:
    raise ValueError(
        "Target column must not be included in prediction features."
    )

if len(EXPECTED_FEATURES) != len(set(EXPECTED_FEATURES)):
    raise ValueError(
        "Duplicate features detected in expected schema."
    )

logger.info("=" * 70)
logger.info("19.1 INPUT VALIDATION CONFIGURATION")
logger.info("=" * 70)

logger.info(
    "Expected Predictor Features                   : %d",
    EXPECTED_FEATURE_COUNT
)

logger.info(
    "Numeric Features                             : %s",
    VALIDATION_CONFIG["numeric_features"]
)

logger.info(
    "Categorical Feature Count                    : %d",
    len(VALIDATION_CONFIG["categorical_features"])
)

logger.info(
    "Invalid Tokens                               : %s",
    sorted(VALIDATION_CONFIG["invalid_tokens"])
)

logger.info(
    "Age Valid Range                              : %d-%d",
    VALIDATION_CONFIG["age_min"],
    VALIDATION_CONFIG["age_max"]
)

logger.info(
    "Target Column Excluded                       : %s",
    VALIDATION_CONFIG["target_column"]
)

logger.info(
    "Input Validation Configuration               : PASSED"
)
````

## 19.3 Feature Name Validation

### Objective
Validate that incoming prediction data contains the exact predictor feature
schema expected by the trained production pipeline.

### Why It Matters
Machine learning models depend on a fixed feature schema established during
training. A missing, unexpected, duplicated, or incorrectly ordered feature
can result in invalid predictions, preprocessing errors, or silent data
mismatch. Feature-name validation provides a contract between incoming data
and the serialized production model.

### Validation Scope
The validation checks:

- All expected predictor features are present.
- No unexpected features are supplied.
- No feature name appears more than once.
- The target column `class` is not included in prediction input.
- Feature ordering exactly matches the training schema.

### Production Behavior
Valid inputs continue to the next validation stage. Any schema mismatch
raises a validation error before the input reaches the model.

### Schema Source
The expected feature schema is derived from the production feature metadata
generated during model serialization, ensuring that validation remains aligned
with the trained model package.

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
The prediction input must match the production feature names and ordering
exactly before it is accepted for further validation and prediction.

````python


def validate_feature_names(
    input_data: pd.DataFrame
) -> None:
    """
    Validate prediction input feature names and ordering.
    """

    if not isinstance(
        input_data,
        pd.DataFrame
    ):
        raise TypeError(
            "Input must be a pandas DataFrame."
        )

    actual_features = list(
        input_data.columns
    )

    expected_features = list(
        EXPECTED_FEATURES
    )

    # --------------------------------------------------------
    # DUPLICATE FEATURE VALIDATION
    # --------------------------------------------------------

    duplicate_features = (
        input_data.columns[
            input_data.columns.duplicated()
        ]
        .tolist()
    )

    if duplicate_features:
        raise ValueError(
            f"Duplicate feature names detected: "
            f"{duplicate_features}"
        )

    # --------------------------------------------------------
    # TARGET COLUMN VALIDATION
    # --------------------------------------------------------

    target_column = VALIDATION_CONFIG[
        "target_column"
    ]

    if target_column in actual_features:
        raise ValueError(
            f"Target column '{target_column}' "
            f"must not be included in prediction input."
        )

    # --------------------------------------------------------
    # MISSING FEATURE VALIDATION
    # --------------------------------------------------------

    missing_features = sorted(
        set(expected_features)
        - set(actual_features)
    )

    if missing_features:
        raise ValueError(
            f"Missing feature names: "
            f"{missing_features}"
        )

    # --------------------------------------------------------
    # UNEXPECTED FEATURE VALIDATION
    # --------------------------------------------------------

    unexpected_features = sorted(
        set(actual_features)
        - set(expected_features)
    )

    if unexpected_features:
        raise ValueError(
            f"Unexpected feature names: "
            f"{unexpected_features}"
        )

    # --------------------------------------------------------
    # FEATURE ORDER VALIDATION
    # --------------------------------------------------------

    if actual_features != expected_features:
        raise ValueError(
            "Feature names are valid but their order "
            "does not match the production schema."
        )


# ------------------------------------------------------------
# VALIDATION TEST INPUT
# ------------------------------------------------------------

validation_input = pd.DataFrame(
    columns=EXPECTED_FEATURES
)


# ------------------------------------------------------------
# RUN VALIDATION
# ------------------------------------------------------------

validate_feature_names(
    validation_input
)


# ------------------------------------------------------------
# LOG RESULTS
# ------------------------------------------------------------

logger.info("=" * 70)
logger.info("19.3 FEATURE NAME VALIDATION")
logger.info("=" * 70)

logger.info(
    "Expected Feature Count                      : %d",
    len(EXPECTED_FEATURES)
)

logger.info(
    "Received Feature Count                      : %d",
    len(validation_input.columns)
)

logger.info(
    "Missing Features                            : %d",
    len(
        set(EXPECTED_FEATURES)
        - set(validation_input.columns)
    )
)

logger.info(
    "Unexpected Features                         : %d",
    len(
        set(validation_input.columns)
        - set(EXPECTED_FEATURES)
    )
)

logger.info(
    "Duplicate Features                          : %d",
    validation_input.columns.duplicated().sum()
)

logger.info(
    "Feature Order                               : VALID"
)

logger.info(
    "Feature Name Validation                     : PASSED"
)
````

## 19.4 Data Type Validation

### Objective
Validate that incoming prediction data uses data types compatible with the
production preprocessing pipeline and the expected feature schema.

### Why It Matters
Data type mismatches can cause preprocessing failures, incorrect conversions,
or unexpected model behavior during production inference. Since the trained
pipeline expects `Age` as a numeric feature and all remaining predictors as
categorical features, the validation layer must verify these expectations
before prediction.

### Validation Scope
The validation checks:

- `Age` is supplied as a numeric-compatible feature.
- All categorical features are represented using categorical-compatible
  values.
- No unsupported complex data types are present.
- The target column `class` is not required for prediction input.
- The validation is performed against the feature schema already established
  by the production model package.

### Production Behavior
Valid input data continues to the next validation stage. If a feature contains
an incompatible data type, the validation layer raises an explicit error before
the input reaches the preprocessing pipeline or model.

### Schema Source
Numeric and categorical feature definitions are derived from
`VALIDATION_CONFIG`, which was created from the serialized feature metadata.
This keeps data-type validation aligned with the trained production schema.

### Required Libraries
- `logging`
- `numpy`
- `pandas`

### Expected Outcome
The validation layer must confirm that every incoming predictor has a data type
compatible with its role in the production preprocessing pipeline.

````python
import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


def validate_data_types(input_data: pd.DataFrame) -> None:
    """
    Validate predictor data types against the production feature schema.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    if VALIDATION_CONFIG["target_column"] in input_data.columns:
        raise ValueError(
            f"Target column '{VALIDATION_CONFIG['target_column']}' "
            "must not be included in prediction input."
        )

    for feature in VALIDATION_CONFIG["numeric_features"]:
        if feature not in input_data.columns:
            raise ValueError(
                f"Required numeric feature '{feature}' is missing."
            )

        if not pd.api.types.is_numeric_dtype(input_data[feature]):
            raise TypeError(
                f"Feature '{feature}' must be numeric. "
                f"Received dtype: {input_data[feature].dtype}"
            )

    for feature in VALIDATION_CONFIG["categorical_features"]:
        if feature not in input_data.columns:
            raise ValueError(
                f"Required categorical feature '{feature}' is missing."
            )

        dtype = input_data[feature].dtype

        if pd.api.types.is_complex_dtype(input_data[feature]):
            raise TypeError(
                f"Categorical feature '{feature}' contains unsupported "
                f"complex data type: {dtype}"
            )

    logger.info("All predictor data types are compatible with the production schema.")


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

# Age must be numeric.
validation_input["Age"] = 45

# Categorical predictors use string-compatible values.
for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

validate_data_types(validation_input)

logger.info("=" * 70)
logger.info("19.4 DATA TYPE VALIDATION")
logger.info("=" * 70)
logger.info(
    "Numeric Features Checked                       : %d",
    len(VALIDATION_CONFIG["numeric_features"])
)
logger.info(
    "Categorical Features Checked                   : %d",
    len(VALIDATION_CONFIG["categorical_features"])
)
logger.info(
    "Age Data Type                                  : %s",
    validation_input["Age"].dtype
)
logger.info(
    "Categorical Data Types Valid                   : %s",
    all(
        not pd.api.types.is_complex_dtype(validation_input[feature])
        for feature in VALIDATION_CONFIG["categorical_features"]
    )
)
logger.info("Data Type Validation                         : PASSED")
````

## 19.5 Missing Value Validation

### Objective
Validate incoming prediction data for missing values before it is passed to the
production preprocessing pipeline.

### Why It Matters
Missing values are a common source of inference-time failures and unexpected
model behavior. Although the trained preprocessing pipeline contains
imputation logic, the input-validation layer should explicitly detect and
report missing values so that data-quality issues are visible before prediction.

### Validation Scope
The validation checks:

- Missing values across all expected predictor features.
- Total number of missing values in the input.
- Missing values by feature.
- Missing values in the numeric feature `Age`.
- Missing values in categorical predictors.
- The target column `class` is not accepted as prediction input.

### Production Behavior
The validation layer reports the location and quantity of missing values. Inputs
containing missing values are rejected at this validation stage rather than
silently allowing incomplete production data to proceed to inference.

### Schema Source
The expected predictor features are derived from `EXPECTED_FEATURES`, while
numeric and categorical feature groups are derived from `VALIDATION_CONFIG`.
This ensures missing-value validation remains synchronized with the production
model schema.

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
A valid prediction input must contain zero missing values across all predictor
features before it is accepted for production inference.

````python
import logging
import pandas as pd

logger = logging.getLogger(__name__)


def validate_missing_values(input_data: pd.DataFrame) -> None:
    """
    Validate that prediction input contains no missing values.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    target_column = VALIDATION_CONFIG["target_column"]

    if target_column in input_data.columns:
        raise ValueError(
            f"Target column '{target_column}' must not be included "
            "in prediction input."
        )

    missing_by_feature = input_data[EXPECTED_FEATURES].isna().sum()
    total_missing = int(missing_by_feature.sum())

    if total_missing > 0:
        missing_features = {
            feature: int(count)
            for feature, count in missing_by_feature.items()
            if count > 0
        }

        raise ValueError(
            f"Missing values detected in prediction input: "
            f"{missing_features}"
        )


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

# Valid numeric value for Age.
validation_input["Age"] = 45

# Valid categorical values.
for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

missing_by_feature = validation_input[EXPECTED_FEATURES].isna().sum()
total_missing = int(missing_by_feature.sum())
age_missing = int(validation_input["Age"].isna().sum())
categorical_missing = int(
    validation_input[VALIDATION_CONFIG["categorical_features"]]
    .isna()
    .sum()
    .sum()
)

validate_missing_values(validation_input)

logger.info("=" * 70)
logger.info("19.5 MISSING VALUE VALIDATION")
logger.info("=" * 70)
logger.info(
    "Expected Feature Count                      : %d",
    EXPECTED_FEATURE_COUNT
)
logger.info(
    "Total Missing Values                        : %d",
    total_missing
)
logger.info(
    "Features With Missing Values                : %d",
    int((missing_by_feature > 0).sum())
)
logger.info(
    "Missing Values in Age                       : %d",
    age_missing
)
logger.info(
    "Missing Values in Categorical Features      : %d",
    categorical_missing
)
logger.info("Missing Value Validation                    : PASSED")
````

## 19.6 Invalid Token Validation

### Objective
Validate incoming prediction data for known invalid placeholder tokens before it
is passed to the production preprocessing pipeline.

### Why It Matters
The training dataset contains the invalid placeholder token `?`, which is used
to represent unavailable or improperly recorded values. If such tokens reach
the production preprocessing layer without validation, they can cause data-type
conversion failures or introduce unexpected categorical values that were not
intended to represent valid observations.

### Validation Scope
The validation checks:

- All expected predictor features for configured invalid tokens.
- The invalid token set defined in `VALIDATION_CONFIG`.
- The numeric feature `Age` for invalid token occurrences.
- All categorical predictors for invalid token occurrences.
- The target column `class` is not accepted as prediction input.
- The exact number of invalid-token occurrences detected in the input.

### Production Behavior
Prediction input containing a configured invalid token is rejected before it
reaches the preprocessing pipeline. The validation error identifies the
affected features and occurrence counts, making the data-quality issue
traceable during production inference.

### Schema Source
The invalid-token rules are derived from `VALIDATION_CONFIG`, while the
features being inspected are derived from `EXPECTED_FEATURES`. This keeps the
validation contract synchronized with the production input schema.

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
A valid prediction input must contain zero occurrences of the configured
invalid token `?` across all predictor features.

````python
import logging
import pandas as pd

logger = logging.getLogger(__name__)


def validate_invalid_tokens(input_data: pd.DataFrame) -> None:
    """
    Validate that prediction input does not contain configured invalid tokens.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    target_column = VALIDATION_CONFIG["target_column"]

    if target_column in input_data.columns:
        raise ValueError(
            f"Target column '{target_column}' must not be included "
            "in prediction input."
        )

    invalid_tokens = VALIDATION_CONFIG["invalid_tokens"]
    invalid_occurrences = {}

    for feature in EXPECTED_FEATURES:
        feature_values = input_data[feature].astype("string")

        for token in invalid_tokens:
            count = int((feature_values == token).sum())

            if count > 0:
                invalid_occurrences[feature] = (
                    invalid_occurrences.get(feature, 0) + count
                )

    if invalid_occurrences:
        raise ValueError(
            f"Invalid token(s) detected in prediction input: "
            f"{invalid_occurrences}"
        )


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

# Valid numeric value for Age.
validation_input["Age"] = 45

# Valid categorical values.
for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# CHECK INVALID TOKEN OCCURRENCES
# ----------------------------------------------------------------------

invalid_tokens = VALIDATION_CONFIG["invalid_tokens"]
invalid_token_counts = {}

for feature in EXPECTED_FEATURES:
    feature_values = validation_input[feature].astype("string")

    count = sum(
        int((feature_values == token).sum())
        for token in invalid_tokens
    )

    if count > 0:
        invalid_token_counts[feature] = count

total_invalid_tokens = sum(invalid_token_counts.values())


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

validate_invalid_tokens(validation_input)

logger.info("=" * 70)
logger.info("19.6 INVALID TOKEN VALIDATION")
logger.info("=" * 70)
logger.info(
    "Expected Feature Count                      : %d",
    EXPECTED_FEATURE_COUNT
)
logger.info(
    "Configured Invalid Tokens                   : %s",
    sorted(invalid_tokens)
)
logger.info(
    "Features With Invalid Tokens                : %d",
    len(invalid_token_counts)
)
logger.info(
    "Total Invalid Token Occurrences             : %d",
    total_invalid_tokens
)
logger.info(
    "Age Invalid Token Occurrences              : %d",
    sum(
        int(
            (validation_input["Age"].astype("string") == token).sum()
        )
        for token in invalid_tokens
    )
)
logger.info("Invalid Token Validation                    : PASSED")
````

## 19.7 Numeric Range Validation

### Objective
Validate that numeric predictor values fall within the acceptable production
range defined for the model input schema.

### Why It Matters
A numeric value can be syntactically valid and still be operationally invalid.
Out-of-range values may represent data-entry errors, corrupted records, or
unrealistic observations. Detecting these values before inference prevents
invalid inputs from reaching the preprocessing pipeline and production model.

### Validation Scope
The validation checks:

- All configured numeric features.
- The minimum and maximum allowed value for each numeric feature.
- The configured production range for `Age`.
- Values below the configured minimum.
- Values above the configured maximum.
- Missing numeric values are handled separately by the missing-value validation
  stage.
- The target column `class` is not accepted as prediction input.

### Production Behavior
If a numeric feature contains a value outside its configured production range,
the validation layer raises an explicit error identifying the affected feature
and observed range violation. Valid values continue to the next validation
stage.

### Schema Source
Numeric features and their permitted ranges are derived from
`VALIDATION_CONFIG`. For this project, `Age` is the numeric predictor and its
configured acceptable range is 0 to 120.

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
All numeric predictor values must remain within their configured production
ranges before the input is accepted for inference.

````python
import logging
import pandas as pd

logger = logging.getLogger(__name__)


def validate_numeric_ranges(input_data: pd.DataFrame) -> None:
    """
    Validate numeric predictor values against configured production ranges.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    target_column = VALIDATION_CONFIG["target_column"]

    if target_column in input_data.columns:
        raise ValueError(
            f"Target column '{target_column}' must not be included "
            "in prediction input."
        )

    range_violations = {}

    for feature in VALIDATION_CONFIG["numeric_features"]:
        if feature not in input_data.columns:
            raise ValueError(
                f"Required numeric feature '{feature}' is missing."
            )

        feature_values = input_data[feature]

        if not pd.api.types.is_numeric_dtype(feature_values):
            raise TypeError(
                f"Numeric feature '{feature}' must contain numeric values."
            )

        if feature == "Age":
            minimum = VALIDATION_CONFIG["age_min"]
            maximum = VALIDATION_CONFIG["age_max"]

            below_minimum = feature_values < minimum
            above_maximum = feature_values > maximum

            violation_count = int(
                (below_minimum | above_maximum).sum()
            )

            if violation_count > 0:
                range_violations[feature] = {
                    "below_minimum": int(below_minimum.sum()),
                    "above_maximum": int(above_maximum.sum()),
                }

    if range_violations:
        raise ValueError(
            f"Numeric range violations detected: {range_violations}"
        )


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

# Valid Age value within the configured 0-120 range.
validation_input["Age"] = 45

# Valid categorical values.
for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# RANGE CHECK SUMMARY
# ----------------------------------------------------------------------

age_values = validation_input["Age"]

age_minimum = VALIDATION_CONFIG["age_min"]
age_maximum = VALIDATION_CONFIG["age_max"]

observed_age_minimum = float(age_values.min())
observed_age_maximum = float(age_values.max())

below_minimum_count = int(
    (age_values < age_minimum).sum()
)

above_maximum_count = int(
    (age_values > age_maximum).sum()
)

total_range_violations = (
    below_minimum_count + above_maximum_count
)


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

validate_numeric_ranges(validation_input)

logger.info("=" * 70)
logger.info("19.7 NUMERIC RANGE VALIDATION")
logger.info("=" * 70)
logger.info(
    "Numeric Features Checked                       : %d",
    len(VALIDATION_CONFIG["numeric_features"])
)
logger.info(
    "Age Allowed Range                              : %d-%d",
    age_minimum,
    age_maximum
)
logger.info(
    "Observed Age Minimum                          : %.1f",
    observed_age_minimum
)
logger.info(
    "Observed Age Maximum                          : %.1f",
    observed_age_maximum
)
logger.info(
    "Values Below Minimum                          : %d",
    below_minimum_count
)
logger.info(
    "Values Above Maximum                          : %d",
    above_maximum_count
)
logger.info(
    "Total Range Violations                        : %d",
    total_range_violations
)
logger.info("Numeric Range Validation                      : PASSED")
````

## 19.8 Age Validation

### Objective
Perform dedicated validation of the `Age` feature before prediction to ensure
that age values are numeric, present, finite, and within the configured
production range.

### Why It Matters
`Age` requires special handling in this dataset because the original training
data contained `?` placeholders and the feature is processed as a numeric
variable. A dedicated validation layer prevents malformed age values from
reaching the production preprocessing pipeline.

### Validation Scope
The validation checks:

- `Age` is present in the prediction input.
- `Age` contains numeric-compatible values.
- `Age` does not contain the configured invalid token.
- `Age` does not contain missing values.
- `Age` does not contain infinite values.
- `Age` remains within the configured range of 0 to 120.

### Production Behavior
Any invalid, missing, non-finite, or out-of-range age value causes validation
to fail before the input reaches the production model. Valid age values are
accepted for subsequent validation and inference.

### Schema Source
The `Age` feature and its acceptable range are derived from
`VALIDATION_CONFIG`, keeping this dedicated validation aligned with the
production input contract.

### Required Libraries
- `logging`
- `numpy`
- `pandas`

### Expected Outcome
The production input must contain a valid finite numeric `Age` value within
the configured 0–120 range.

````python
import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


def validate_age(input_data: pd.DataFrame) -> None:
    """
    Perform dedicated production validation for the Age feature.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    age_feature = "Age"

    if age_feature not in input_data.columns:
        raise ValueError("Required feature 'Age' is missing.")

    age_values = input_data[age_feature]

    if not pd.api.types.is_numeric_dtype(age_values):
        raise TypeError(
            f"Age must be numeric. Received dtype: {age_values.dtype}"
        )

    if age_values.isna().any():
        raise ValueError("Age contains missing values.")

    if not np.isfinite(age_values.to_numpy(dtype=float)).all():
        raise ValueError("Age contains NaN or infinite values.")

    minimum = VALIDATION_CONFIG["age_min"]
    maximum = VALIDATION_CONFIG["age_max"]

    below_minimum = age_values < minimum
    above_maximum = age_values > maximum

    if below_minimum.any() or above_maximum.any():
        raise ValueError(
            f"Age values must remain within the range "
            f"{minimum}-{maximum}."
        )


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

validation_input["Age"] = 45

for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# AGE VALIDATION SUMMARY
# ----------------------------------------------------------------------

age_values = validation_input["Age"]

age_minimum = VALIDATION_CONFIG["age_min"]
age_maximum = VALIDATION_CONFIG["age_max"]

age_is_numeric = pd.api.types.is_numeric_dtype(age_values)
age_missing_count = int(age_values.isna().sum())
age_non_finite_count = int(
    (~np.isfinite(age_values.to_numpy(dtype=float))).sum()
)
age_below_minimum_count = int(
    (age_values < age_minimum).sum()
)
age_above_maximum_count = int(
    (age_values > age_maximum).sum()
)


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

validate_age(validation_input)

logger.info("=" * 70)
logger.info("19.8 AGE VALIDATION")
logger.info("=" * 70)
logger.info(
    "Age Data Type                                 : %s",
    age_values.dtype
)
logger.info(
    "Age Numeric Validation                        : %s",
    age_is_numeric
)
logger.info(
    "Age Missing Values                            : %d",
    age_missing_count
)
logger.info(
    "Age Non-Finite Values                         : %d",
    age_non_finite_count
)
logger.info(
    "Age Allowed Range                             : %d-%d",
    age_minimum,
    age_maximum
)
logger.info(
    "Age Range Violations                          : %d",
    age_below_minimum_count + age_above_maximum_count
)
logger.info("Age Validation                               : PASSED")
````

## 19.9 NaN / Infinity Validation

### Objective
Validate all incoming predictor features for non-finite numeric values before
the input reaches the production preprocessing pipeline.

### Why It Matters
`NaN`, positive infinity, and negative infinity are invalid numerical states
for production inference. They can originate from upstream calculations,
data-conversion errors, corrupted records, or external data sources. Explicit
detection prevents these values from propagating into preprocessing and model
inference.

### Validation Scope
The validation checks:

- All expected predictor features for `NaN` values.
- Numeric features for positive and negative infinity.
- Non-finite numeric values before model inference.
- The `Age` feature as the project's numeric predictor.
- Categorical features are checked for missing values without attempting
  numeric conversion.
- The target column `class` is not accepted as prediction input.

### Production Behavior
Any non-finite numeric value causes validation to fail before the input reaches
the production model. The validation layer reports the affected feature and
number of invalid values so that upstream data-quality problems can be
identified.

### Schema Source
The numeric and categorical feature definitions are derived from
`VALIDATION_CONFIG`, while the complete predictor schema is derived from
`EXPECTED_FEATURES`.

### Required Libraries
- `logging`
- `numpy`
- `pandas`

### Expected Outcome
All numeric predictor values must be finite, with zero `NaN` or infinite values,
before the input is accepted for production inference.

````python
import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


def validate_nan_infinity(input_data: pd.DataFrame) -> None:
    """
    Validate numeric predictors for NaN and infinite values.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    target_column = VALIDATION_CONFIG["target_column"]

    if target_column in input_data.columns:
        raise ValueError(
            f"Target column '{target_column}' must not be included "
            "in prediction input."
        )

    non_finite_occurrences = {}

    for feature in VALIDATION_CONFIG["numeric_features"]:
        if feature not in input_data.columns:
            raise ValueError(
                f"Required numeric feature '{feature}' is missing."
            )

        if not pd.api.types.is_numeric_dtype(input_data[feature]):
            raise TypeError(
                f"Numeric feature '{feature}' must contain numeric values."
            )

        values = input_data[feature].to_numpy(dtype=float)

        nan_count = int(np.isnan(values).sum())
        infinity_count = int(np.isinf(values).sum())

        if nan_count > 0 or infinity_count > 0:
            non_finite_occurrences[feature] = {
                "nan": nan_count,
                "infinity": infinity_count,
            }

    if non_finite_occurrences:
        raise ValueError(
            f"NaN or infinity values detected: {non_finite_occurrences}"
        )


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

validation_input["Age"] = 45

for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# NON-FINITE VALUE SUMMARY
# ----------------------------------------------------------------------

nan_count = 0
infinity_count = 0

for feature in VALIDATION_CONFIG["numeric_features"]:
    values = validation_input[feature].to_numpy(dtype=float)

    nan_count += int(np.isnan(values).sum())
    infinity_count += int(np.isinf(values).sum())

total_non_finite = nan_count + infinity_count


# ----------------------------------------------------------------------
# EXECUTE VALIDATION
# ----------------------------------------------------------------------

validate_nan_infinity(validation_input)

logger.info("=" * 70)
logger.info("19.9 NAN / INFINITY VALIDATION")
logger.info("=" * 70)
logger.info(
    "Numeric Features Checked                       : %d",
    len(VALIDATION_CONFIG["numeric_features"])
)
logger.info(
    "NaN Values Detected                           : %d",
    nan_count
)
logger.info(
    "Infinity Values Detected                      : %d",
    infinity_count
)
logger.info(
    "Total Non-Finite Values                       : %d",
    total_non_finite
)
logger.info("NaN / Infinity Validation                     : PASSED")
````

## 19.10 Schema Validation

### Objective
Perform a consolidated validation of the production prediction-input schema
before data is passed to the model.

### Why It Matters
Individual validation checks protect specific aspects of the input, but a
production inference interface also requires a single schema-level contract.
This validation confirms that the incoming data contains exactly the expected
predictor structure and is compatible with the established validation rules.

### Validation Scope
The schema validation checks:

- Input is provided as a pandas DataFrame.
- Exactly 34 predictor features are supplied.
- Expected feature names are present.
- No unexpected or duplicate features are present.
- Feature ordering matches the production schema.
- The target column `class` is excluded.
- Data types are compatible with the production schema.
- Missing values are absent.
- Configured invalid tokens are absent.
- Numeric values remain within their configured ranges.
- Numeric values contain no `NaN` or infinite values.
- `Age` satisfies its dedicated validation requirements.

### Production Behavior
Schema validation acts as the consolidated gate before model inference. Any
schema or data-quality violation causes validation to fail before the input is
passed to the production preprocessing pipeline and model.

### Schema Source
The production schema is derived from the serialized feature metadata through
`EXPECTED_FEATURES` and `VALIDATION_CONFIG`. No independent hardcoded feature
schema is introduced.

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
A prediction input is accepted only when its complete structure and data
quality satisfy all production validation requirements.

````python
import logging
import pandas as pd

logger = logging.getLogger(__name__)


def validate_schema(input_data: pd.DataFrame) -> None:
    """
    Execute the complete production input-schema validation gate.
    """

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    validate_feature_count(input_data)
    validate_feature_names(input_data)
    validate_data_types(input_data)
    validate_missing_values(input_data)
    validate_invalid_tokens(input_data)
    validate_numeric_ranges(input_data)
    validate_age(input_data)
    validate_nan_infinity(input_data)


# ----------------------------------------------------------------------
# VALIDATION TEST INPUT
# ----------------------------------------------------------------------

validation_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

validation_input["Age"] = 45

for feature in VALIDATION_CONFIG["categorical_features"]:
    validation_input[feature] = "1"


# ----------------------------------------------------------------------
# EXECUTE CONSOLIDATED SCHEMA VALIDATION
# ----------------------------------------------------------------------

validate_schema(validation_input)

logger.info("=" * 70)
logger.info("19.10 SCHEMA VALIDATION")
logger.info("=" * 70)
logger.info(
    "Expected Feature Count                      : %d",
    EXPECTED_FEATURE_COUNT
)
logger.info(
    "Received Feature Count                      : %d",
    len(validation_input.columns)
)
logger.info(
    "Target Column Included                      : %s",
    VALIDATION_CONFIG["target_column"] in validation_input.columns
)
logger.info(
    "Schema Feature Order                        : VALID"
)
logger.info(
    "Data Type Validation                        : PASSED"
)
logger.info(
    "Missing Value Validation                    : PASSED"
)
logger.info(
    "Invalid Token Validation                    : PASSED"
)
logger.info(
    "Numeric Range Validation                    : PASSED"
)
logger.info(
    "Age Validation                              : PASSED"
)
logger.info(
    "NaN / Infinity Validation                   : PASSED"
)
logger.info("Schema Validation                           : PASSED")
````

## 19.11 Input Validation Testing

### Objective
Test the complete production input-validation layer using both valid and
intentionally invalid prediction inputs.

### Why It Matters
A validation system is only production-ready when it demonstrates two
behaviors: valid inputs must be accepted, while invalid inputs must be
reliably rejected before reaching the model. Testing these failure conditions
provides evidence that the validation rules are actively enforcing the
production input contract.

### Validation Scope
The testing covers representative validation failures for:

- Valid production input.
- Incorrect feature count.
- Missing feature.
- Unexpected feature.
- Duplicate feature.
- Target column inclusion.
- Missing values.
- Invalid token `?`.
- Out-of-range `Age`.
- Non-finite `Age`.
- Invalid `Age` data type.

Each test is executed independently so that one rejected input does not
interrupt the remaining validation tests.

### Production Behavior
A valid input must pass the complete schema validation gate. Invalid inputs
must be rejected with an explicit validation error before model inference.

### Test Design
The test cases intentionally modify a known-valid prediction input one
condition at a time. This isolates individual validation rules and makes the
result easier to audit and troubleshoot.

### Required Libraries
- `logging`
- `numpy`
- `pandas`

### Expected Outcome
The valid input must be accepted, while every intentionally invalid test case
must be rejected by the appropriate validation rule. The final test summary
must confirm that the complete validation layer behaves as expected.

````python
import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# CREATE BASE VALID INPUT
# ----------------------------------------------------------------------

base_input = pd.DataFrame({
    feature: [0]
    for feature in EXPECTED_FEATURES
})

base_input["Age"] = 45

for feature in VALIDATION_CONFIG["categorical_features"]:
    base_input[feature] = "1"


# ----------------------------------------------------------------------
# DEFINE TEST CASES
# ----------------------------------------------------------------------

test_cases = {}


# 1. Valid input
test_cases["Valid Input"] = base_input.copy()


# 2. Incorrect feature count
invalid_feature_count = base_input.drop(
    columns=[EXPECTED_FEATURES[-1]]
)
test_cases["Incorrect Feature Count"] = invalid_feature_count


# 3. Missing feature
missing_feature_input = base_input.drop(
    columns=[EXPECTED_FEATURES[1]]
)
test_cases["Missing Feature"] = missing_feature_input


# 4. Unexpected feature
unexpected_feature_input = base_input.copy()
unexpected_feature_input["unexpected_feature"] = "1"
test_cases["Unexpected Feature"] = unexpected_feature_input


# 5. Duplicate feature
duplicate_feature_input = base_input.copy()
duplicate_feature_input.columns = list(
    duplicate_feature_input.columns[:-1]
) + [EXPECTED_FEATURES[0]]
test_cases["Duplicate Feature"] = duplicate_feature_input


# 6. Target column included
target_column_input = base_input.copy()
target_column_input[VALIDATION_CONFIG["target_column"]] = 1
test_cases["Target Column Included"] = target_column_input


# 7. Missing value
missing_value_input = base_input.copy()
missing_value_input.loc[0, "Age"] = np.nan
test_cases["Missing Value"] = missing_value_input


# 8. Invalid token
invalid_token_input = base_input.copy()
invalid_token_input.loc[
    0,
    VALIDATION_CONFIG["categorical_features"][0]
] = "?"
test_cases["Invalid Token"] = invalid_token_input


# 9. Out-of-range Age
out_of_range_age_input = base_input.copy()
out_of_range_age_input.loc[0, "Age"] = 121
test_cases["Out-of-Range Age"] = out_of_range_age_input


# 10. Infinite Age
# Convert Age to float before assigning infinity to avoid
# pandas incompatible-dtype FutureWarning.
infinite_age_input = base_input.copy()
infinite_age_input["Age"] = infinite_age_input["Age"].astype(float)
infinite_age_input.loc[0, "Age"] = np.inf
test_cases["Infinite Age"] = infinite_age_input


# 11. Invalid Age data type
invalid_age_type_input = base_input.copy()
invalid_age_type_input["Age"] = "invalid"
test_cases["Invalid Age Data Type"] = invalid_age_type_input


# ----------------------------------------------------------------------
# EXECUTE VALIDATION TESTS
# ----------------------------------------------------------------------

test_results = []

for test_name, test_input in test_cases.items():

    expected_result = (
        "ACCEPTED"
        if test_name == "Valid Input"
        else "REJECTED"
    )

    try:
        validate_schema(test_input)
        actual_result = "ACCEPTED"

    except (TypeError, ValueError):
        actual_result = "REJECTED"

    test_passed = actual_result == expected_result

    test_results.append({
        "Test Case": test_name,
        "Expected": expected_result,
        "Actual": actual_result,
        "Result": "PASS" if test_passed else "FAIL"
    })


# ----------------------------------------------------------------------
# TEST SUMMARY
# ----------------------------------------------------------------------

test_results_df = pd.DataFrame(test_results)

total_tests = len(test_results_df)

passed_tests = int(
    (test_results_df["Result"] == "PASS").sum()
)

failed_tests = int(
    (test_results_df["Result"] == "FAIL").sum()
)

valid_input_passed = bool(
    test_results_df.loc[
        test_results_df["Test Case"] == "Valid Input",
        "Result"
    ].eq("PASS").all()
)

invalid_inputs_rejected = bool(
    test_results_df.loc[
        test_results_df["Test Case"] != "Valid Input",
        "Actual"
    ].eq("REJECTED").all()
)


# ----------------------------------------------------------------------
# DISPLAY TEST RESULTS
# ----------------------------------------------------------------------

display(test_results_df)


# ----------------------------------------------------------------------
# LOG VALIDATION TEST SUMMARY
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("19.11 INPUT VALIDATION TESTING")
logger.info("=" * 70)

logger.info(
    "Total Test Cases                             : %d",
    total_tests
)

logger.info(
    "Passed Test Cases                           : %d",
    passed_tests
)

logger.info(
    "Failed Test Cases                           : %d",
    failed_tests
)

logger.info(
    "Valid Input Accepted                        : %s",
    valid_input_passed
)

logger.info(
    "Invalid Inputs Rejected                     : %s",
    invalid_inputs_rejected
)


if failed_tests != 0:
    raise AssertionError(
        "One or more input-validation tests failed."
    )

logger.info(
    "Input Validation Testing                    : PASSED"
)

logger.info(
    "19. INPUT VALIDATION                        : COMPLETE"
)
````

## 20.4 Load Feature Metadata

### Objective
Load the serialized feature metadata associated with the production model package.

### Why It Matters
Feature metadata defines the production input contract and records the feature structure used during model development. Loading and validating this metadata ensures that production prediction remains aligned with the trained model's feature schema.

### Validation Scope
The feature metadata will be validated for:

- File existence.
- Successful JSON loading.
- Correct metadata structure.
- Consistency between numeric and categorical features.
- Consistency with the expected production feature schema.
- Consistency of the target column.

### Production Behavior

### Artifact Source
The feature metadata is loaded from `feature_metadata.json`, generated and validated during Section 18.

### Required Libraries
- `json`
- `logging`

### Expected Outcome
The serialized feature metadata is successfully loaded and verified against the production feature schema.

````python


import json
import logging

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# VALIDATE FEATURE METADATA ARTIFACT
# ----------------------------------------------------------------------

feature_metadata_path = PREDICTION_CONFIG[
    "feature_metadata_path"
]

if not feature_metadata_path.exists():
    raise FileNotFoundError(
        f"Feature metadata not found: {feature_metadata_path}"
    )

if feature_metadata_path.stat().st_size == 0:
    raise ValueError(
        "Feature metadata artifact is empty."
    )


# ----------------------------------------------------------------------
# LOAD FEATURE METADATA
# ----------------------------------------------------------------------

with open(
    feature_metadata_path,
    "r",
    encoding="utf-8"
) as file:
    production_feature_metadata = json.load(file)


# ----------------------------------------------------------------------
# VALIDATE METADATA STRUCTURE
# ----------------------------------------------------------------------

if not isinstance(
    production_feature_metadata,
    dict
):
    raise TypeError(
        "Feature metadata must be a dictionary."
    )


required_metadata_keys = {
    "original_feature_count",
    "numeric_features",
    "categorical_features",
    "transformed_feature_count",
    "target_column",
}

missing_metadata_keys = (
    required_metadata_keys
    - set(production_feature_metadata.keys())
)

if missing_metadata_keys:
    raise ValueError(
        f"Missing feature metadata keys: "
        f"{sorted(missing_metadata_keys)}"
    )


# ----------------------------------------------------------------------
# VALIDATE FEATURE SCHEMA CONSISTENCY
# ----------------------------------------------------------------------

metadata_features = (
    production_feature_metadata["numeric_features"]
    + production_feature_metadata["categorical_features"]
)

if metadata_features != EXPECTED_FEATURES:
    raise ValueError(
        "Feature metadata does not match the production "
        "feature schema."
    )


if (
    production_feature_metadata["target_column"]
    != PREDICTION_CONFIG["target_column"]
):
    raise ValueError(
        "Feature metadata target column does not match "
        "prediction configuration."
    )


if (
    production_feature_metadata["original_feature_count"]
    != EXPECTED_FEATURE_COUNT
):
    raise ValueError(
        "Original feature count does not match "
        "the production feature count."
    )


# ----------------------------------------------------------------------
# LOG FEATURE METADATA STATUS
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.4 LOAD FEATURE METADATA")
logger.info("=" * 70)

logger.info(
    "Metadata Artifact Exists                  : %s",
    feature_metadata_path.exists()
)

logger.info(
    "Metadata Artifact Size                   : %d bytes",
    feature_metadata_path.stat().st_size
)

logger.info(
    "Original Feature Count                   : %d",
    production_feature_metadata[
        "original_feature_count"
    ]
)

logger.info(
    "Numeric Features                         : %d",
    len(
        production_feature_metadata[
            "numeric_features"
        ]
    )
)

logger.info(
    "Categorical Features                     : %d",
    len(
        production_feature_metadata[
            "categorical_features"
        ]
    )
)

logger.info(
    "Transformed Feature Count                : %d",
    production_feature_metadata[
        "transformed_feature_count"
    ]
)

logger.info(
    "Target Column                            : %s",
    production_feature_metadata[
        "target_column"
    ]
)

logger.info(
    "Feature Metadata Loading                 : PASSED"
)
````

## 20.5 Load Class Mapping

### Objective
Load the serialized class mapping associated with the production model package.

### Why It Matters
The production model predicts one of the six trained disorder classes. The class mapping provides a consistent interpretation of model outputs and ensures that predicted classes remain aligned with the class definitions established during model training and serialization.

### Validation Scope
The class mapping will be validated for:

- File existence.
- Non-empty artifact size.
- Successful JSON loading.
- Correct mapping structure.
- Consistency with the production model classes.
- Complete representation of the six supported classes.

### Production Behavior
If the class mapping is missing, malformed, incomplete, or inconsistent with the production model classes, the prediction pipeline must stop before returning a prediction.

### Artifact Source
The class mapping is loaded from `class_mapping.json`, generated and validated during Section 18.5.

### Required Libraries
- `json`
- `logging`

### Expected Outcome
The serialized class mapping is successfully loaded and verified against the production model's class definitions.

````python


import json
import logging

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# VALIDATE CLASS MAPPING ARTIFACT
# ----------------------------------------------------------------------

class_mapping_path = PREDICTION_CONFIG[
    "class_mapping_path"
]

if not class_mapping_path.exists():
    raise FileNotFoundError(
        f"Class mapping not found: {class_mapping_path}"
    )

if class_mapping_path.stat().st_size == 0:
    raise ValueError(
        "Class mapping artifact is empty."
    )


# ----------------------------------------------------------------------
# LOAD CLASS MAPPING
# ----------------------------------------------------------------------

with open(
    class_mapping_path,
    "r",
    encoding="utf-8"
) as file:
    production_class_mapping = json.load(file)


# ----------------------------------------------------------------------
# VALIDATE MAPPING OBJECT
# ----------------------------------------------------------------------

if not isinstance(
    production_class_mapping,
    dict
):
    raise TypeError(
        "Class mapping artifact must contain a JSON object."
    )


# ----------------------------------------------------------------------
# EXTRACT PRODUCTION MODEL CLASSES
# ----------------------------------------------------------------------

model_classes = [
    str(value)
    for value in production_model.classes_
]


# ----------------------------------------------------------------------
# IDENTIFY CLASS INFORMATION IN THE ARTIFACT
# ----------------------------------------------------------------------

class_mapping_values = []

for key, value in production_class_mapping.items():

    if isinstance(value, dict):

        for nested_key, nested_value in value.items():

            if isinstance(nested_value, dict):

                class_mapping_values.extend(
                    str(item)
                    for item in nested_value.keys()
                )

                class_mapping_values.extend(
                    str(item)
                    for item in nested_value.values()
                )

            elif isinstance(nested_value, (list, tuple)):

                class_mapping_values.extend(
                    str(item)
                    for item in nested_value
                )

            else:

                class_mapping_values.append(
                    str(nested_value)
                )

    elif isinstance(value, (list, tuple)):

        class_mapping_values.extend(
            str(item)
            for item in value
        )

    else:

        class_mapping_values.append(
            str(value)
        )


# ----------------------------------------------------------------------
# VALIDATE CLASS REPRESENTATION
# ----------------------------------------------------------------------

artifact_text = json.dumps(
    production_class_mapping
).lower()

missing_model_classes = [
    class_label
    for class_label in model_classes
    if class_label.lower() not in artifact_text
]

if missing_model_classes:

    raise ValueError(
        "Class mapping artifact does not contain "
        f"all production classes: {missing_model_classes}"
    )


# ----------------------------------------------------------------------
# VALIDATE EXPECTED CLASS COUNT
# ----------------------------------------------------------------------

if len(model_classes) != 6:

    raise ValueError(
        "Expected exactly six production classes."
    )


# ----------------------------------------------------------------------
# VALIDATE CLASS ORDER
# ----------------------------------------------------------------------

expected_class_order = [
    "1",
    "2",
    "3",
    "4",
    "5",
    "6"
]

if model_classes != expected_class_order:

    raise ValueError(
        "Production model class order does not match "
        f"expected classes: {expected_class_order}"
    )


# ----------------------------------------------------------------------
# FINAL CLASS MAPPING VALIDATION
# ----------------------------------------------------------------------

class_mapping_validation_passed = (
    len(model_classes) == 6
    and not missing_model_classes
    and model_classes == expected_class_order
)


if not class_mapping_validation_passed:

    raise ValueError(
        "Class mapping validation failed."
    )


# ----------------------------------------------------------------------
# LOG CLASS MAPPING STATUS
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.5 LOAD CLASS MAPPING")
logger.info("=" * 70)

logger.info(
    "Mapping Artifact Exists                    : %s",
    class_mapping_path.exists()
)

logger.info(
    "Mapping Artifact Size                     : %d bytes",
    class_mapping_path.stat().st_size
)

logger.info(
    "Mapping Artifact Type                     : %s",
    type(production_class_mapping).__name__
)

logger.info(
    "Production Classes                        : %s",
    model_classes
)

logger.info(
    "Class Count                               : %d",
    len(model_classes)
)

logger.info(
    "Missing Production Classes                : %s",
    missing_model_classes
)

logger.info(
    "Class Mapping Consistency                 : PASSED"
)

logger.info(
    "Class Mapping Loading                     : PASSED"
)
````

## 20.6 Prediction Function

### Objective
Create a reusable production prediction function that accepts a validated input DataFrame and generates a skin-disorder class prediction using the serialized production model.

### Why It Matters
A dedicated prediction function provides a controlled inference interface between validated user input and the production model. It ensures that prediction is performed only after the input has passed the established Section 19 validation checks.

### Validation Scope / Methodology
The prediction function will:

- Confirm that the input is a pandas DataFrame.
- Validate the input using the production input-validation function.
- Ensure exactly one input record is supplied.
- Generate the prediction using the serialized production model.
- Validate that the predicted class belongs to the supported production classes.
- Return the predicted class in a consistent string format.

### Production Behavior
The serialized `production_model` is an end-to-end sklearn Pipeline containing preprocessing followed by the Random Forest classifier.

Therefore, the function will pass the validated raw input directly to `production_model.predict()`.

Manual preprocessing using `production_preprocessing.transform()` must not be performed before prediction because this would result in duplicate preprocessing.

### Schema Source / Artifact Source
- Input feature schema: `EXPECTED_FEATURES`
- Validation configuration: `VALIDATION_CONFIG`
- Input validation: Section 19
- Production model: `production_model`
- Production model artifact: `skin_disorder_model.pkl`

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
A reusable `predict_skin_disorder()` function is created that validates a single input record and safely returns the predicted skin-disorder class.

````python
 

import logging
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# PREDICTION FUNCTION
# ----------------------------------------------------------------------

def predict_skin_disorder(input_data: pd.DataFrame) -> str:
    """
    Validate one input record and generate a production prediction.

    Parameters
    ----------
    input_data : pandas.DataFrame
        One raw input record following the production feature schema.

    Returns
    -------
    str
        Predicted skin-disorder class.
    """

    # --------------------------------------------------------------
    # 1. INPUT TYPE VALIDATION
    # --------------------------------------------------------------

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError(
            "Input data must be provided as a pandas DataFrame."
        )

    # --------------------------------------------------------------
    # 2. SINGLE RECORD VALIDATION
    # --------------------------------------------------------------

    if len(input_data) != 1:
        raise ValueError(
            "Prediction requires exactly one input record."
        )

    # --------------------------------------------------------------
    # 3. FEATURE COUNT VALIDATION
    # --------------------------------------------------------------

    if input_data.shape[1] != EXPECTED_FEATURE_COUNT:
        raise ValueError(
            f"Expected {EXPECTED_FEATURE_COUNT} predictor features, "
            f"received {input_data.shape[1]}."
        )

    # --------------------------------------------------------------
    # 4. FEATURE NAME VALIDATION
    # --------------------------------------------------------------

    received_features = list(input_data.columns)

    if received_features != EXPECTED_FEATURES:
        missing_features = [
            feature
            for feature in EXPECTED_FEATURES
            if feature not in received_features
        ]

        unexpected_features = [
            feature
            for feature in received_features
            if feature not in EXPECTED_FEATURES
        ]

        raise ValueError(
            "Input feature schema does not match production schema. "
            f"Missing={missing_features}, "
            f"Unexpected={unexpected_features}"
        )

    # --------------------------------------------------------------
    # 5. TARGET COLUMN PROTECTION
    # --------------------------------------------------------------

    target_column = PREDICTION_CONFIG["target_column"]

    if target_column in input_data.columns:
        raise ValueError(
            f"Target column '{target_column}' must not be supplied "
            "during prediction."
        )

    # --------------------------------------------------------------
    # 6. MISSING VALUE VALIDATION
    # --------------------------------------------------------------

    if input_data.isnull().any().any():
        raise ValueError(
            "Input contains missing values."
        )

    # --------------------------------------------------------------
    # 7. INVALID TOKEN VALIDATION
    # --------------------------------------------------------------

    invalid_tokens = VALIDATION_CONFIG["invalid_tokens"]

    invalid_token_locations = []

    for column in input_data.columns:

        matches = input_data[column].astype(str).isin(
            invalid_tokens
        )

        if matches.any():
            invalid_token_locations.append(column)

    if invalid_token_locations:
        raise ValueError(
            "Input contains invalid tokens in columns: "
            f"{invalid_token_locations}"
        )

    # --------------------------------------------------------------
    # 8. AGE VALIDATION
    # --------------------------------------------------------------

    age_column = "Age"

    if age_column not in input_data.columns:
        raise ValueError(
            "Age feature is missing from the production schema."
        )

    age_values = pd.to_numeric(
        input_data[age_column],
        errors="coerce"
    )

    if age_values.isna().any():
        raise ValueError(
            "Age contains invalid numeric values."
        )

    age_min = VALIDATION_CONFIG["age_min"]
    age_max = VALIDATION_CONFIG["age_max"]

    if (
        (age_values < age_min).any()
        or
        (age_values > age_max).any()
    ):
        raise ValueError(
            f"Age must be between {age_min} and {age_max}."
        )

    # --------------------------------------------------------------
    # 9. NaN / INFINITY VALIDATION
    # --------------------------------------------------------------

    numeric_columns = input_data.select_dtypes(
        include=[np.number]
    ).columns

    if len(numeric_columns) > 0:

        numeric_values = input_data[numeric_columns].to_numpy(
            dtype=float
        )

        if not np.isfinite(numeric_values).all():
            raise ValueError(
                "Input contains NaN or infinite numeric values."
            )

    # --------------------------------------------------------------
    # 10. GENERATE PRODUCTION PREDICTION
    # --------------------------------------------------------------

    prediction = production_model.predict(
        input_data
    )

    if len(prediction) != 1:
        raise ValueError(
            "Production model returned an unexpected number "
            "of predictions."
        )

    predicted_class = str(
        prediction[0]
    )

    # --------------------------------------------------------------
    # 11. PREDICTED CLASS VALIDATION
    # --------------------------------------------------------------

    if predicted_class not in model_classes:
        raise ValueError(
            f"Unsupported predicted class: {predicted_class}"
        )

    # --------------------------------------------------------------
    # 12. LOG RESULT
    # --------------------------------------------------------------

    logger.info("=" * 70)
    logger.info("20.6 PREDICTION FUNCTION")
    logger.info("=" * 70)
    logger.info(
        "Input Data Type                           : %s",
        type(input_data).__name__
    )
    logger.info(
        "Input Record Count                        : %d",
        len(input_data)
    )
    logger.info(
        "Feature Count                              : %d",
        input_data.shape[1]
    )
    logger.info(
        "Input Validation                           : PASSED"
    )
    logger.info(
        "Production Prediction                      : PASSED"
    )
    logger.info(
        "Predicted Class                            : %s",
        predicted_class
    )
    logger.info(
        "Prediction Function                        : PASSED"
    )

    return predicted_class


logger.info("=" * 70)
logger.info("20.6 PREDICTION FUNCTION : READY")
logger.info("=" * 70)
logger.info(
    "Function Name                              : predict_skin_disorder"
)
logger.info(
    "Single Record Prediction                    : ENABLED"
)
logger.info(
    "Production Model                            : %s",
    type(production_model).__name__
)
logger.info(
    "Supported Classes                           : %s",
    model_classes
)
logger.info(
    "Prediction Function                         : PASSED"
)
````

## 20.7 Probability Prediction

### Objective
Generate class probabilities for a single validated skin-disorder input using the production model.

### Why It Matters
Class probabilities provide the confidence distribution across all supported skin-disorder classes instead of returning only the predicted class. This information is required for confidence scoring and safe prediction behavior.

### Validation Scope / Methodology
The probability prediction will:

- Accept the same validated single-record input used by `predict_skin_disorder()`.
- Generate probabilities using the production model's `predict_proba()` method.
- Verify that probabilities are returned for all production classes.
- Verify that all probability values are finite.
- Verify that probabilities are within the range 0 to 1.
- Verify that probabilities sum to approximately 1.
- Preserve the production model class order.

### Production Behavior
The serialized `production_model` is an end-to-end sklearn Pipeline.

Therefore, the raw validated input is passed directly to:

`production_model.predict_proba()`

Manual preprocessing must not be performed before probability prediction.

### Schema Source / Artifact Source
- Feature schema: `EXPECTED_FEATURES`
- Production model: `production_model`
- Production classes: `model_classes`
- Prediction function: `predict_skin_disorder()`

### Required Libraries
- `logging`
- `numpy`
- `pandas`

### Expected Outcome
A reusable `predict_skin_disorder_probabilities()` function is created that returns a probability distribution across production classes and validates the probability output before returning it.

````python


import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# PROBABILITY PREDICTION FUNCTION
# ----------------------------------------------------------------------

def predict_skin_disorder_probabilities(
    input_data: pd.DataFrame
) -> dict:
    """
    Validate input data and generate production class probabilities.

    Parameters
    ----------
    input_data : pandas.DataFrame
        One raw input record following the production feature schema.

    Returns
    -------
    dict
        Dictionary containing class probabilities.
    """

    # --------------------------------------------------------------
    # 1. INPUT TYPE VALIDATION
    # --------------------------------------------------------------

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError(
            "Input data must be provided as a pandas DataFrame."
        )

    # --------------------------------------------------------------
    # 2. SINGLE RECORD VALIDATION
    # --------------------------------------------------------------

    if len(input_data) != 1:
        raise ValueError(
            "Probability prediction requires exactly one input record."
        )

    # --------------------------------------------------------------
    # 3. VALIDATE INPUT USING PRODUCTION PREDICTION FUNCTION
    # --------------------------------------------------------------

    predict_skin_disorder(input_data)

    # --------------------------------------------------------------
    # 4. GENERATE PROBABILITIES
    # --------------------------------------------------------------

    probabilities = production_model.predict_proba(
        input_data
    )

    # --------------------------------------------------------------
    # 5. VALIDATE PROBABILITY OUTPUT SHAPE
    # --------------------------------------------------------------

    if probabilities.ndim != 2:
        raise ValueError(
            "Probability output must be a two-dimensional array."
        )

    if probabilities.shape[0] != 1:
        raise ValueError(
            "Probability prediction must contain exactly one record."
        )

    if probabilities.shape[1] != len(model_classes):
        raise ValueError(
            "Probability count does not match the production class count."
        )

    # --------------------------------------------------------------
    # 6. EXTRACT SINGLE PROBABILITY VECTOR
    # --------------------------------------------------------------

    probability_vector = probabilities[0].astype(float)

    # --------------------------------------------------------------
    # 7. VALIDATE FINITE VALUES
    # --------------------------------------------------------------

    if not np.isfinite(probability_vector).all():
        raise ValueError(
            "Probability output contains NaN or infinite values."
        )

    # --------------------------------------------------------------
    # 8. VALIDATE PROBABILITY RANGE
    # --------------------------------------------------------------

    if (
        (probability_vector < 0).any()
        or
        (probability_vector > 1).any()
    ):
        raise ValueError(
            "Probability values must be between 0 and 1."
        )

    # --------------------------------------------------------------
    # 9. VALIDATE PROBABILITY SUM
    # --------------------------------------------------------------

    probability_sum = float(
        probability_vector.sum()
    )

    if not np.isclose(
        probability_sum,
        1.0,
        atol=1e-6
    ):
        raise ValueError(
            f"Probability values must sum to 1. "
            f"Observed sum: {probability_sum:.10f}"
        )

    # --------------------------------------------------------------
    # 10. CREATE CLASS-PROBABILITY MAPPING
    # --------------------------------------------------------------

    class_probabilities = {
        class_label: float(probability)
        for class_label, probability
        in zip(model_classes, probability_vector)
    }

    # --------------------------------------------------------------
    # 11. LOG PROBABILITY STATUS
    # --------------------------------------------------------------

    logger.info("=" * 70)
    logger.info("20.7 PROBABILITY PREDICTION")
    logger.info("=" * 70)

    logger.info(
        "Input Record Count                        : %d",
        len(input_data)
    )

    logger.info(
        "Production Classes                        : %d",
        len(model_classes)
    )

    logger.info(
        "Probability Vector Shape                  : %s",
        probabilities.shape
    )

    logger.info(
        "Probability Sum                           : %.6f",
        probability_sum
    )

    logger.info(
        "Probability Range                         : %.6f - %.6f",
        probability_vector.min(),
        probability_vector.max()
    )

    logger.info(
        "Probability Validation                    : PASSED"
    )

    logger.info(
        "Probability Prediction                   : PASSED"
    )

    return class_probabilities


logger.info("=" * 70)
logger.info("20.7 PROBABILITY PREDICTION : READY")
logger.info("=" * 70)
logger.info(
    "Function Name                              : "
    "predict_skin_disorder_probabilities"
)
logger.info(
    "Production Probability Capability          : %s",
    hasattr(production_model, "predict_proba")
)
logger.info(
    "Supported Classes                           : %s",
    model_classes
)
logger.info(
    "Probability Prediction                     : READY"
)
````

## 20.8 Predicted Class

### Objective
Create a reusable function that identifies the final predicted skin-disorder class from the production probability distribution.

### Why It Matters
The probability distribution contains the model's estimated probability for every supported class. The predicted class is the class with the highest probability.

### Validation Scope / Methodology
The function will:

- Accept a validated single-record input.
- Generate the production class probabilities.
- Identify the class with the highest probability.
- Verify that the predicted class belongs to the supported production classes.
- Verify that the selected class probability matches the maximum probability.
- Return the predicted class in a consistent string format.

### Production Behavior
The prediction is derived from the production model's probability output.

The class with the maximum probability is selected as the predicted skin-disorder class.

### Schema Source / Artifact Source
- Production model: `production_model`
- Production classes: `model_classes`
- Probability function: `predict_skin_disorder_probabilities()`
- Prediction configuration: `PREDICTION_CONFIG`

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
A reusable `get_predicted_class()` function is created that returns the production model's highest-probability skin-disorder class.

````python


import logging
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# PREDICTED CLASS FUNCTION
# ----------------------------------------------------------------------

def get_predicted_class(input_data: pd.DataFrame) -> str:
    """
    Generate the final predicted skin-disorder class.

    Parameters
    ----------
    input_data : pandas.DataFrame
        One validated production input record.

    Returns
    -------
    str
        Predicted skin-disorder class.
    """

    # --------------------------------------------------------------
    # 1. INPUT TYPE VALIDATION
    # --------------------------------------------------------------

    if not isinstance(input_data, pd.DataFrame):
        raise TypeError(
            "Input data must be provided as a pandas DataFrame."
        )

    # --------------------------------------------------------------
    # 2. SINGLE RECORD VALIDATION
    # --------------------------------------------------------------

    if len(input_data) != 1:
        raise ValueError(
            "Predicted class requires exactly one input record."
        )

    # --------------------------------------------------------------
    # 3. GENERATE PROBABILITIES
    # --------------------------------------------------------------

    class_probabilities = (
        predict_skin_disorder_probabilities(
            input_data
        )
    )

    # --------------------------------------------------------------
    # 4. VALIDATE PROBABILITY MAPPING
    # --------------------------------------------------------------

    if not isinstance(
        class_probabilities,
        dict
    ):
        raise TypeError(
            "Class probabilities must be returned as a dictionary."
        )

    if set(class_probabilities.keys()) != set(model_classes):
        raise ValueError(
            "Probability classes do not match production classes."
        )

    # --------------------------------------------------------------
    # 5. IDENTIFY HIGHEST-PROBABILITY CLASS
    # --------------------------------------------------------------

    predicted_class = max(
        class_probabilities,
        key=class_probabilities.get
    )

    predicted_probability = float(
        class_probabilities[predicted_class]
    )

    # --------------------------------------------------------------
    # 6. VALIDATE PREDICTED CLASS
    # --------------------------------------------------------------

    if predicted_class not in model_classes:
        raise ValueError(
            f"Unsupported predicted class: {predicted_class}"
        )

    # --------------------------------------------------------------
    # 7. VALIDATE MAXIMUM PROBABILITY
    # --------------------------------------------------------------

    maximum_probability = max(
        class_probabilities.values()
    )

    if predicted_probability != maximum_probability:
        raise ValueError(
            "Predicted class does not have the maximum probability."
        )

    # --------------------------------------------------------------
    # 8. LOG RESULT
    # --------------------------------------------------------------

    logger.info("=" * 70)
    logger.info("20.8 PREDICTED CLASS")
    logger.info("=" * 70)

    logger.info(
        "Input Record Count                        : %d",
        len(input_data)
    )

    logger.info(
        "Supported Classes                         : %s",
        model_classes
    )

    logger.info(
        "Predicted Class                           : %s",
        predicted_class
    )

    logger.info(
        "Maximum Class Probability                 : %.6f",
        predicted_probability
    )

    logger.info(
        "Predicted Class Validation                : PASSED"
    )

    logger.info(
        "Predicted Class Function                  : PASSED"
    )

    return predicted_class


logger.info("=" * 70)
logger.info("20.8 PREDICTED CLASS : READY")
logger.info("=" * 70)
logger.info(
    "Function Name                              : get_predicted_class"
)
logger.info(
    "Prediction Source                           : Production Probability Distribution"
)
logger.info(
    "Supported Classes                           : %s",
    model_classes
)
logger.info(
    "Predicted Class Function                    : READY"
)
````

### Schema Source / Artifact Source

### Expected Outcome
A real test record is converted into a valid production input, the class probabilities are generated successfully, and the notebook reports the predicted class, confidence score, threshold, and confidence status.

````python


import logging
import numpy as np
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. LOCATE A VALID PRODUCTION-SCHEMA RECORD
# ----------------------------------------------------------------------

if "X_test" not in globals():
    raise NameError(
        "X_test is not available. The test feature dataset is required."
    )

if not isinstance(X_test, pd.DataFrame):
    raise TypeError(
        "X_test must be a pandas DataFrame."
    )

if not set(EXPECTED_FEATURES).issubset(X_test.columns):
    missing_features = [
        feature
        for feature in EXPECTED_FEATURES
        if feature not in X_test.columns
    ]

    raise ValueError(
        f"X_test is missing production features: {missing_features}"
    )


# Keep only the production predictor schema.
confidence_input_source = X_test[
    EXPECTED_FEATURES
].copy()


# ----------------------------------------------------------------------
# 2. NORMALIZE AGE FOR PRODUCTION PREDICTION
# ----------------------------------------------------------------------

confidence_input_source["Age"] = pd.to_numeric(
    confidence_input_source["Age"],
    errors="coerce"
)


# ----------------------------------------------------------------------
# 3. FIND FIRST COMPLETELY VALID RECORD
# ----------------------------------------------------------------------

valid_record_mask = (
    confidence_input_source["Age"].notna()
    &
    confidence_input_source["Age"].between(
        VALIDATION_CONFIG["age_min"],
        VALIDATION_CONFIG["age_max"]
    )
    &
    ~confidence_input_source.isnull().any(axis=1)
)

for column in confidence_input_source.columns:

    if confidence_input_source[column].dtype == "object":

        valid_record_mask &= (
            ~confidence_input_source[column]
            .astype(str)
            .isin(VALIDATION_CONFIG["invalid_tokens"])
        )


valid_indices = confidence_input_source.index[
    valid_record_mask
]


if len(valid_indices) == 0:
    raise ValueError(
        "No valid production-schema record is available "
        "for confidence-score validation."
    )


prediction_input = confidence_input_source.loc[
    [valid_indices[0]]
].copy()


# ----------------------------------------------------------------------
# 4. FINAL INPUT VALIDATION
# ----------------------------------------------------------------------

if prediction_input.shape != (
    1,
    EXPECTED_FEATURE_COUNT
):
    raise ValueError(
        "Prediction input shape does not match "
        "the production feature schema."
    )


if list(prediction_input.columns) != EXPECTED_FEATURES:
    raise ValueError(
        "Prediction input feature order does not match "
        "the production feature schema."
    )


if prediction_input.isnull().any().any():
    raise ValueError(
        "Prediction input contains missing values."
    )


if (
    prediction_input.astype(str)
    .isin(VALIDATION_CONFIG["invalid_tokens"])
    .any()
    .any()
):
    raise ValueError(
        "Prediction input contains invalid tokens."
    )


# ----------------------------------------------------------------------
# 5. GENERATE PRODUCTION PROBABILITIES
# ----------------------------------------------------------------------

probabilities = production_model.predict_proba(
    prediction_input
)


# ----------------------------------------------------------------------
# 6. PROBABILITY VALIDATION
# ----------------------------------------------------------------------

if probabilities.shape != (
    1,
    len(model_classes)
):
    raise ValueError(
        "Production probability output shape is invalid."
    )


probability_vector = probabilities[0].astype(float)


if not np.isfinite(
    probability_vector
).all():
    raise ValueError(
        "Probability output contains NaN or infinite values."
    )


if (
    (probability_vector < 0).any()
    or
    (probability_vector > 1).any()
):
    raise ValueError(
        "Probability values must be between 0 and 1."
    )


probability_sum = float(
    probability_vector.sum()
)


if not np.isclose(
    probability_sum,
    1.0,
    atol=1e-6
):
    raise ValueError(
        f"Probability sum is invalid: {probability_sum:.10f}"
    )


# ----------------------------------------------------------------------
# 7. PREDICTED CLASS
# ----------------------------------------------------------------------

predicted_index = int(
    np.argmax(probability_vector)
)

predicted_class = str(
    model_classes[predicted_index]
)


# ----------------------------------------------------------------------
# 8. CONFIDENCE SCORE
# ----------------------------------------------------------------------

confidence_score = float(
    probability_vector[predicted_index]
)


confidence_threshold = float(
    PREDICTION_CONFIG[
        "confidence_threshold"
    ]
)


confidence_status = (
    "HIGH_CONFIDENCE"
    if confidence_score >= confidence_threshold
    else "LOW_CONFIDENCE"
)


# ----------------------------------------------------------------------
# 9. PROBABILITY TABLE
# ----------------------------------------------------------------------

probability_table = pd.DataFrame(
    {
        "Class": model_classes,
        "Probability": probability_vector,
        "Probability (%)": probability_vector * 100,
        "Predicted": [
            class_label == predicted_class
            for class_label in model_classes
        ],
    }
)


# ----------------------------------------------------------------------
# 10. CONFIDENCE VALIDATION
# ----------------------------------------------------------------------

if not 0.0 <= confidence_score <= 1.0:
    raise ValueError(
        "Confidence score must be between 0 and 1."
    )


if confidence_score != probability_table[
    "Probability"
].max():
    raise ValueError(
        "Confidence score does not match "
        "the maximum class probability."
    )


# ----------------------------------------------------------------------
# 11. STORE PRODUCTION PREDICTION RESULTS
# ----------------------------------------------------------------------

production_prediction_result = {
    "predicted_class": predicted_class,
    "confidence_score": confidence_score,
    "confidence_threshold": confidence_threshold,
    "confidence_status": confidence_status,
    "probability_sum": probability_sum,
    "input_index": valid_indices[0],
}


# ----------------------------------------------------------------------
# 12. DISPLAY ACTUAL PRODUCTION RESULT
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.9 CONFIDENCE SCORE")
logger.info("=" * 70)

logger.info(
    "Selected Test Record Index                 : %s",
    valid_indices[0]
)

logger.info(
    "Production Input Shape                     : %s",
    prediction_input.shape
)

logger.info(
    "Probability Vector Shape                   : %s",
    probabilities.shape
)

logger.info(
    "Probability Sum                            : %.6f",
    probability_sum
)

logger.info(
    "Predicted Class                            : %s",
    predicted_class
)

logger.info(
    "Confidence Score                           : %.6f",
    confidence_score
)

logger.info(
    "Confidence Threshold                       : %.2f",
    confidence_threshold
)

logger.info(
    "Confidence Status                          : %s",
    confidence_status
)

display(
    probability_table.style.format(
        {
            "Probability": "{:.6f}",
            "Probability (%)": "{:.2f}%"
        }
    )
)

logger.info(
    "Probability Validation                     : PASSED"
)

logger.info(
    "Confidence Score Validation                : PASSED"
)

logger.info(
    "20.9 CONFIDENCE SCORE                      : PASSED"
)
````

## 20.10 Prediction Metadata

### Objective
Create a structured prediction metadata object containing the key information required to interpret, audit, and monitor a production prediction.

### Why It Matters
A production prediction should contain more than a predicted class. Metadata provides traceability for the input record, model output, confidence level, threshold decision, probability distribution, and model version.

### Validation Scope / Methodology
The prediction metadata will include:

- Selected input record identifier.
- Predicted class.
- Complete class-probability distribution.
- Confidence score.
- Configured confidence threshold.
- Confidence status.
- Probability-sum validation result.
- Production model type.
- Model version.
- Project identifier.
- Prediction timestamp.

### Production Behavior
Metadata is generated from the actual production prediction results created in Section 20.9.

No prediction values are hardcoded.

The metadata object is designed to be serializable and suitable for downstream API responses, logging, monitoring, and audit trails.

### Schema Source / Artifact Source
- Prediction result: Section 20.9
- Production model: `production_model`
- Model version: `MODEL_VERSION_PATH`
- Project identifier: `PRCP-1028`
- Class mapping: `model_classes`
- Confidence threshold: `PREDICTION_CONFIG`

### Required Libraries
- `datetime`
- `json`
- `logging`

### Expected Outcome
A complete structured prediction metadata object is created and validated successfully, with the actual prediction class, confidence, probability distribution, model version, and prediction timestamp preserved.

````python


import json
import logging
from datetime import datetime, timezone

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. VALIDATE REQUIRED PREDICTION RESULTS
# ----------------------------------------------------------------------

required_prediction_variables = {
    "predicted_class": predicted_class,
    "confidence_score": confidence_score,
    "confidence_threshold": confidence_threshold,
    "confidence_status": confidence_status,
    "probability_sum": probability_sum,
    "probability_table": probability_table,
}

for variable_name, variable_value in required_prediction_variables.items():

    if variable_value is None:
        raise ValueError(
            f"Required prediction result is missing: {variable_name}"
        )


# ----------------------------------------------------------------------
# 2. LOAD MODEL VERSION ARTIFACT
# ----------------------------------------------------------------------

if not MODEL_VERSION_PATH.exists():
    raise FileNotFoundError(
        f"Model version artifact not found: {MODEL_VERSION_PATH}"
    )

if MODEL_VERSION_PATH.stat().st_size == 0:
    raise ValueError(
        "Model version artifact is empty."
    )

with open(
    MODEL_VERSION_PATH,
    "r",
    encoding="utf-8"
) as file:
    model_version_data = json.load(file)


if not isinstance(
    model_version_data,
    dict
):
    raise TypeError(
        "Model version artifact must contain a JSON object."
    )


# ----------------------------------------------------------------------
# 3. EXTRACT MODEL VERSION
# ----------------------------------------------------------------------

model_version = (
    model_version_data.get("version")
    or model_version_data.get("model_version")
)

if model_version is None:

    raise ValueError(
        "Model version is missing from model_version.json."
    )

model_version = str(model_version)


# ----------------------------------------------------------------------
# 4. BUILD CLASS PROBABILITY METADATA
# ----------------------------------------------------------------------

class_probability_metadata = {
    str(row["Class"]): float(row["Probability"])
    for _, row in probability_table.iterrows()
}


# ----------------------------------------------------------------------
# 5. CREATE PREDICTION METADATA
# ----------------------------------------------------------------------

prediction_metadata = {

    "project": "PRCP-1028",

    "prediction_timestamp_utc": datetime.now(
        timezone.utc
    ).isoformat(),

    "input": {

        "record_index": str(
            production_prediction_result["input_index"]
        ),

        "feature_count": int(
            prediction_input.shape[1]
        ),
    },

    "prediction": {

        "predicted_class": predicted_class,

        "confidence_score": float(
            confidence_score
        ),

        "confidence_threshold": float(
            confidence_threshold
        ),

        "confidence_status": confidence_status,

        "class_probabilities": class_probability_metadata,
    },

    "model": {

        "model_type": type(
            production_model.named_steps["classifier"]
        ).__name__,

        "pipeline_type": type(
            production_model
        ).__name__,

        "model_version": model_version,

        "class_count": len(
            model_classes
        ),
    },

    "validation": {

        "probability_sum": float(
            probability_sum
        ),

        "probability_sum_valid": bool(
            np.isclose(
                probability_sum,
                1.0,
                atol=1e-6
            )
        ),
    },
}


# ----------------------------------------------------------------------
# 6. METADATA VALIDATION
# ----------------------------------------------------------------------

if prediction_metadata["project"] != "PRCP-1028":

    raise ValueError(
        "Prediction metadata project identifier is invalid."
    )


if prediction_metadata[
    "prediction"
]["predicted_class"] not in model_classes:

    raise ValueError(
        "Prediction metadata contains an unsupported class."
    )


if not 0.0 <= prediction_metadata[
    "prediction"
]["confidence_score"] <= 1.0:

    raise ValueError(
        "Prediction metadata contains an invalid confidence score."
    )


if len(
    prediction_metadata[
        "prediction"
    ]["class_probabilities"]
) != len(model_classes):

    raise ValueError(
        "Prediction metadata class probability count is invalid."
    )


if not prediction_metadata[
    "validation"
]["probability_sum_valid"]:

    raise ValueError(
        "Prediction metadata probability validation failed."
    )


if prediction_metadata[
    "model"
]["model_version"] == "unknown":

    raise ValueError(
        "Production model version must not be unknown."
    )


# ----------------------------------------------------------------------
# 7. DISPLAY METADATA SUMMARY
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.10 PREDICTION METADATA")
logger.info("=" * 70)

logger.info(
    "Project                                    : %s",
    prediction_metadata["project"]
)

logger.info(
    "Input Record Index                         : %s",
    prediction_metadata["input"]["record_index"]
)

logger.info(
    "Feature Count                              : %d",
    prediction_metadata["input"]["feature_count"]
)

logger.info(
    "Predicted Class                            : %s",
    prediction_metadata["prediction"]["predicted_class"]
)

logger.info(
    "Confidence Score                           : %.6f",
    prediction_metadata["prediction"]["confidence_score"]
)

logger.info(
    "Confidence Threshold                       : %.2f",
    prediction_metadata["prediction"]["confidence_threshold"]
)

logger.info(
    "Confidence Status                          : %s",
    prediction_metadata["prediction"]["confidence_status"]
)

logger.info(
    "Model Type                                 : %s",
    prediction_metadata["model"]["model_type"]
)

logger.info(
    "Pipeline Type                              : %s",
    prediction_metadata["model"]["pipeline_type"]
)

logger.info(
    "Model Version                              : %s",
    prediction_metadata["model"]["model_version"]
)

logger.info(
    "Class Probability Count                   : %d",
    len(
        prediction_metadata[
            "prediction"
        ]["class_probabilities"]
    )
)

logger.info(
    "Probability Sum                            : %.6f",
    prediction_metadata[
        "validation"
    ]["probability_sum"]
)

logger.info(
    "Prediction Metadata Validation             : PASSED"
)

logger.info(
    "20.10 PREDICTION METADATA                  : PASSED"
)
````

## 20.11 Error Handling

### Objective
Implement controlled error handling for the production prediction workflow so invalid inputs, unavailable artifacts, prediction failures, and unexpected model outputs are handled safely.

### Why It Matters
Production prediction systems must fail safely and provide clear diagnostic information instead of exposing uncontrolled exceptions or returning unreliable predictions.

### Validation Scope / Methodology
The error-handling layer will:

- Validate that prediction input is a DataFrame.
- Validate that exactly one record is supplied.
- Validate the production feature schema.
- Reject missing values and invalid tokens.
- Reject invalid Age values.
- Catch production model prediction failures.
- Validate the predicted class.
- Return a controlled prediction response.
- Log errors using the configured application logger.

### Production Behavior
Valid inputs return a structured successful prediction response.

Invalid inputs return a controlled error response without producing an unsafe prediction.

Unexpected model or probability failures are captured and reported through the prediction response and application logs.

### Schema Source / Artifact Source
- Input schema: `EXPECTED_FEATURES`
- Validation configuration: `VALIDATION_CONFIG`
- Production model: `production_model`
- Prediction function: `predict_skin_disorder()`
- Probability function: `predict_skin_disorder_probabilities()`
- Class mapping: `model_classes`

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
The production prediction workflow handles valid and invalid inputs safely, produces controlled error responses, and passes the defined error-handling validation tests.

````python


import logging
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# SAFE PRODUCTION PREDICTION WRAPPER
# ----------------------------------------------------------------------

def safe_predict_skin_disorder(
    input_data: pd.DataFrame
) -> dict:
    """
    Execute a production prediction with controlled error handling.

    Returns a structured response containing either a successful
    prediction or a controlled error message.
    """

    try:

        # --------------------------------------------------------------
        # INPUT TYPE VALIDATION
        # --------------------------------------------------------------

        if not isinstance(
            input_data,
            pd.DataFrame
        ):
            raise TypeError(
                "Input data must be a pandas DataFrame."
            )

        # --------------------------------------------------------------
        # SINGLE RECORD VALIDATION
        # --------------------------------------------------------------

        if len(input_data) != 1:
            raise ValueError(
                "Prediction requires exactly one input record."
            )

        # --------------------------------------------------------------
        # FEATURE COUNT VALIDATION
        # --------------------------------------------------------------

        if input_data.shape[1] != EXPECTED_FEATURE_COUNT:
            raise ValueError(
                f"Expected {EXPECTED_FEATURE_COUNT} predictor features, "
                f"received {input_data.shape[1]}."
            )

        # --------------------------------------------------------------
        # FEATURE SCHEMA VALIDATION
        # --------------------------------------------------------------

        if list(input_data.columns) != EXPECTED_FEATURES:
            raise ValueError(
                "Input feature schema does not match "
                "the production schema."
            )

        # --------------------------------------------------------------
        # TARGET COLUMN PROTECTION
        # --------------------------------------------------------------

        if PREDICTION_CONFIG[
            "target_column"
        ] in input_data.columns:

            raise ValueError(
                "Target column must not be supplied during prediction."
            )

        # --------------------------------------------------------------
        # MISSING VALUE VALIDATION
        # --------------------------------------------------------------

        if input_data.isnull().any().any():
            raise ValueError(
                "Input contains missing values."
            )

        # --------------------------------------------------------------
        # INVALID TOKEN VALIDATION
        # --------------------------------------------------------------

        invalid_tokens = (
            VALIDATION_CONFIG[
                "invalid_tokens"
            ]
        )

        for column in input_data.columns:

            if (
                input_data[column]
                .astype(str)
                .isin(invalid_tokens)
                .any()
            ):

                raise ValueError(
                    f"Invalid token detected in feature: {column}"
                )

        # --------------------------------------------------------------
        # AGE VALIDATION
        # --------------------------------------------------------------

        age_values = pd.to_numeric(
            input_data["Age"],
            errors="coerce"
        )

        if age_values.isna().any():
            raise ValueError(
                "Age contains invalid numeric values."
            )

        if (
            (age_values < VALIDATION_CONFIG["age_min"]).any()
            or
            (age_values > VALIDATION_CONFIG["age_max"]).any()
        ):

            raise ValueError(
                "Age is outside the allowed production range."
            )

        # --------------------------------------------------------------
        # PRODUCTION PREDICTION
        # --------------------------------------------------------------

        predicted_class = predict_skin_disorder(
            input_data
        )

        # --------------------------------------------------------------
        # PROBABILITY PREDICTION
        # --------------------------------------------------------------

        class_probabilities = (
            predict_skin_disorder_probabilities(
                input_data
            )
        )

        confidence_score = float(
            max(class_probabilities.values())
        )

        confidence_threshold = float(
            PREDICTION_CONFIG[
                "confidence_threshold"
            ]
        )

        confidence_status = (
            "HIGH_CONFIDENCE"
            if confidence_score >= confidence_threshold
            else "LOW_CONFIDENCE"
        )

        # --------------------------------------------------------------
        # SUCCESS RESPONSE
        # --------------------------------------------------------------

        response = {
            "success": True,
            "predicted_class": predicted_class,
            "confidence_score": confidence_score,
            "confidence_threshold": confidence_threshold,
            "confidence_status": confidence_status,
            "class_probabilities": class_probabilities,
            "error": None,
        }

        return response

    except Exception as error:

        # --------------------------------------------------------------
        # CONTROLLED ERROR RESPONSE
        # --------------------------------------------------------------

        logger.error(
            "Prediction failed: %s",
            error
        )

        return {
            "success": False,
            "predicted_class": None,
            "confidence_score": None,
            "confidence_threshold": PREDICTION_CONFIG[
                "confidence_threshold"
            ],
            "confidence_status": "PREDICTION_ERROR",
            "class_probabilities": None,
            "error": str(error),
        }


# ----------------------------------------------------------------------
# ERROR HANDLING VALIDATION
# ----------------------------------------------------------------------

if not callable(
    safe_predict_skin_disorder
):
    raise TypeError(
        "Safe prediction function was not created successfully."
    )


# ----------------------------------------------------------------------
# VALID INPUT TEST
# ----------------------------------------------------------------------

valid_response = safe_predict_skin_disorder(
    prediction_input
)

if not valid_response["success"]:
    raise ValueError(
        "Valid production input was not accepted."
    )

if valid_response["predicted_class"] not in model_classes:
    raise ValueError(
        "Valid prediction returned an unsupported class."
    )


# ----------------------------------------------------------------------
# INVALID INPUT TEST
# ----------------------------------------------------------------------

invalid_input = prediction_input.copy()

invalid_input["Age"] = -1

invalid_response = safe_predict_skin_disorder(
    invalid_input
)

if invalid_response["success"]:
    raise ValueError(
        "Invalid production input was incorrectly accepted."
    )

if invalid_response["predicted_class"] is not None:
    raise ValueError(
        "Invalid input returned a prediction."
    )


# ----------------------------------------------------------------------
# DISPLAY VALIDATION RESULTS
# ----------------------------------------------------------------------

error_handling_results = pd.DataFrame(
    [
        {
            "Test Case": "Valid Production Input",
            "Accepted": valid_response["success"],
            "Prediction": valid_response["predicted_class"],
            "Status": "PASSED",
        },
        {
            "Test Case": "Invalid Age Input",
            "Accepted": invalid_response["success"],
            "Prediction": invalid_response["predicted_class"],
            "Status": "PASSED",
        },
    ]
)

display(
    error_handling_results
)


# ----------------------------------------------------------------------
# FINAL LOGGING
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.11 ERROR HANDLING")
logger.info("=" * 70)

logger.info(
    "Valid Input Accepted                       : %s",
    valid_response["success"]
)

logger.info(
    "Valid Prediction Class                    : %s",
    valid_response["predicted_class"]
)

logger.info(
    "Invalid Input Rejected                    : %s",
    not invalid_response["success"]
)

logger.info(
    "Invalid Input Prediction                  : %s",
    invalid_response["predicted_class"]
)

logger.info(
    "Controlled Error Response                 : %s",
    invalid_response["error"]
)

logger.info(
    "Error Handling Validation                 : PASSED"
)

logger.info(
    "20.11 ERROR HANDLING                      : PASSED"
)
````

## 20.12 Prediction Pipeline Testing

### Objective
Validate the complete production prediction pipeline from input validation through prediction, probability generation, confidence calculation, and controlled error handling.

### Why It Matters
A production prediction pipeline must be validated as an integrated workflow, not only through individual functions. End-to-end testing confirms that valid inputs produce reliable predictions and invalid inputs are rejected safely.

### Validation Scope / Methodology
The prediction pipeline will be tested using representative production scenarios:

- Valid production input.
- Invalid Age input.
- Missing-value input.
- Invalid token input.
- Incorrect feature schema.

Each test will verify whether the pipeline accepts or rejects the input as expected.

### Production Behavior
Valid production input must produce:

- A predicted class.
- Class probabilities.
- A confidence score.
- A confidence status.

Invalid production input must:

- Be rejected safely.
- Return no prediction.
- Return a controlled error message.

### Schema Source / Artifact Source
- Safe prediction function: `safe_predict_skin_disorder()`
- Production input schema: `EXPECTED_FEATURES`
- Validation configuration: `VALIDATION_CONFIG`
- Production model: `production_model`
- Prediction configuration: `PREDICTION_CONFIG`

### Required Libraries
- `logging`
- `pandas`

### Expected Outcome
All defined prediction pipeline test cases pass, valid input produces a prediction, invalid inputs are rejected, and the complete Section 20 safe prediction pipeline is validated successfully.

````python


import logging
import pandas as pd

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. PREPARE TEST INPUTS
# ----------------------------------------------------------------------

valid_input = prediction_input.copy()


invalid_age_input = prediction_input.copy()
invalid_age_input["Age"] = -1


missing_value_input = prediction_input.copy()
missing_value_input["Age"] = None


invalid_token_input = prediction_input.copy()

categorical_test_feature = (
    VALIDATION_CONFIG["categorical_features"][0]
)

invalid_token_input[
    categorical_test_feature
] = "?"


incorrect_schema_input = prediction_input.copy()

incorrect_schema_input = incorrect_schema_input.rename(
    columns={
        EXPECTED_FEATURES[0]: "Invalid_Feature"
    }
)


# ----------------------------------------------------------------------
# 2. DEFINE TEST CASES
# ----------------------------------------------------------------------

prediction_test_cases = [
    {
        "Test Case": "Valid Production Input",
        "Input": valid_input,
        "Expected Success": True,
    },
    {
        "Test Case": "Invalid Age",
        "Input": invalid_age_input,
        "Expected Success": False,
    },
    {
        "Test Case": "Missing Value",
        "Input": missing_value_input,
        "Expected Success": False,
    },
    {
        "Test Case": "Invalid Token",
        "Input": invalid_token_input,
        "Expected Success": False,
    },
    {
        "Test Case": "Incorrect Feature Schema",
        "Input": incorrect_schema_input,
        "Expected Success": False,
    },
]


# ----------------------------------------------------------------------
# 3. EXECUTE END-TO-END TESTS
# ----------------------------------------------------------------------

prediction_test_results = []

for test_case in prediction_test_cases:

    response = safe_predict_skin_disorder(
        test_case["Input"]
    )

    actual_success = response["success"]

    test_passed = (
        actual_success
        ==
        test_case["Expected Success"]
    )

    prediction_test_results.append(
        {
            "Test Case": test_case["Test Case"],
            "Expected Success": test_case["Expected Success"],
            "Actual Success": actual_success,
            "Predicted Class": response["predicted_class"],
            "Confidence": response["confidence_score"],
            "Status": (
                "PASSED"
                if test_passed
                else "FAILED"
            ),
        }
    )


# ----------------------------------------------------------------------
# 4. CREATE TEST RESULT TABLE
# ----------------------------------------------------------------------

prediction_test_results_df = pd.DataFrame(
    prediction_test_results
)


# ----------------------------------------------------------------------
# 5. VALIDATE COMPLETE TEST SUITE
# ----------------------------------------------------------------------

total_test_cases = len(
    prediction_test_results_df
)

passed_test_cases = int(
    (
        prediction_test_results_df["Status"]
        ==
        "PASSED"
    ).sum()
)

failed_test_cases = (
    total_test_cases
    -
    passed_test_cases
)


if failed_test_cases != 0:
    raise ValueError(
        f"Prediction pipeline testing failed: "
        f"{failed_test_cases} test case(s) failed."
    )


# ----------------------------------------------------------------------
# 6. DISPLAY RESULTS
# ----------------------------------------------------------------------

display(
    prediction_test_results_df.style.format(
        {
            "Confidence": (
                lambda value:
                "-"
                if pd.isna(value)
                else f"{value:.6f}"
            )
        }
    )
)


# ----------------------------------------------------------------------
# 7. FINAL VALIDATION LOGGING
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("20.12 PREDICTION PIPELINE TESTING")
logger.info("=" * 70)

logger.info(
    "Total Test Cases                           : %d",
    total_test_cases
)

logger.info(
    "Passed Test Cases                         : %d",
    passed_test_cases
)

logger.info(
    "Failed Test Cases                         : %d",
    failed_test_cases
)

logger.info(
    "Valid Input Prediction                    : %s",
    prediction_test_results_df.iloc[0]["Predicted Class"]
)

logger.info(
    "Valid Input Confidence                   : %.6f",
    prediction_test_results_df.iloc[0]["Confidence"]
)

logger.info(
    "Invalid Inputs Rejected                   : %s",
    all(
        prediction_test_results_df.iloc[
            1:
        ]["Actual Success"] == False
    )
)

logger.info(
    "Prediction Pipeline Testing               : PASSED"
)

logger.info(
    "20.12 PREDICTION PIPELINE TESTING         : PASSED"
)

logger.info(
    "=" * 70
)

logger.info(
    "20. SAFE PREDICTION PIPELINE              : COMPLETE"
)
````

## 21.1 API Architecture & Configuration

This subsection defines the production API configuration for the Skin Disorder Prediction service.

The configuration centralizes:

## 21.7 API Request Schema

This subsection defines the request schema for the FastAPI prediction endpoint.

The API request schema represents one patient record containing the 34 production predictor features.

The schema is designed to:

- Accept the canonical production feature set
- Preserve the production feature names
- Validate incoming data types through Pydantic
- Prevent the target column from being supplied as an input
- Provide a structured request contract for the `/predict` endpoint

The request model will be converted into a pandas DataFrame before being passed to the validated prediction pipeline.

````python
import logging
from typing import Any

import pandas as pd
from pydantic import BaseModel, ConfigDict

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. DEFINE API REQUEST MODEL
# ----------------------------------------------------------------------

class SkinDisorderPredictionRequest(BaseModel):
    model_config = ConfigDict(
        extra="forbid"
    )

    Age: Any

    clinical_features: dict[str, Any]


# ----------------------------------------------------------------------
# 2. VALIDATE REQUEST MODEL STRUCTURE
# ----------------------------------------------------------------------

if not issubclass(
    SkinDisorderPredictionRequest,
    BaseModel
):
    raise TypeError(
        "API request schema must inherit from BaseModel."
    )

request_model_fields = set(
    SkinDisorderPredictionRequest.model_fields.keys()
)

required_request_fields = {
    "Age",
    "clinical_features",
}

missing_request_fields = (
    required_request_fields
    -
    request_model_fields
)

if missing_request_fields:
    raise ValueError(
        "Required request schema fields are missing: "
        f"{sorted(missing_request_fields)}"
    )


# ----------------------------------------------------------------------
# 3. VALIDATE REQUEST FEATURE COUNT
# ----------------------------------------------------------------------

categorical_features = (
    API_CONFIG["expected_features"][1:]
)

if len(categorical_features) != 33:
    raise ValueError(
        "Expected exactly 33 categorical production features."
    )


# ----------------------------------------------------------------------
# 4. VALIDATE TARGET EXCLUSION
# ----------------------------------------------------------------------

if API_CONFIG["target_column"] in request_model_fields:
    raise ValueError(
        "Target column must not be part of the API request schema."
    )


# ----------------------------------------------------------------------
# 5. FINAL VALIDATION LOGGING
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("21.7 API REQUEST SCHEMA")
logger.info("=" * 70)

logger.info(
    "Request Model Type                         : %s",
    SkinDisorderPredictionRequest.__name__
)

logger.info(
    "Request Base Class                         : %s",
    BaseModel.__name__
)

logger.info(
    "Production Feature Count                   : %d",
    API_CONFIG["expected_feature_count"]
)

logger.info(
    "Numeric Feature Count                      : %d",
    len(API_CONFIG["expected_features"])
    - len(categorical_features)
)

logger.info(
    "Categorical Feature Count                  : %d",
    len(categorical_features)
)

logger.info(
    "Target Column Included                     : %s",
    API_CONFIG["target_column"] in request_model_fields
)

logger.info(
    "Extra Fields Policy                        : forbid"
)

logger.info(
    "API Request Schema Validation              : PASSED"
)

logger.info(
    "21.7 API REQUEST SCHEMA                    : PASSED"
)
````

## 21.8 Request Validation

This subsection validates incoming FastAPI prediction requests against the production input schema.

The validation layer enforces:

- Exact production feature names
- Exact production feature count
- Exact feature ordering
- Target column exclusion
- Missing-value rejection
- Invalid-token rejection
- Age numeric validation
- Age range validation
- NaN and infinity rejection
- Conversion into the pandas DataFrame format required by the production prediction pipeline

The validated DataFrame is passed to the existing `safe_predict_skin_disorder()` function so that API validation remains consistent with the production prediction logic established in Section 20.

````python
import logging
import numpy as np
import pandas as pd
from fastapi import HTTPException

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. DEFINE REQUEST VALIDATION FUNCTION
# ----------------------------------------------------------------------

def validate_api_request(
    request: SkinDisorderPredictionRequest
) -> pd.DataFrame:

    if not isinstance(
        request,
        SkinDisorderPredictionRequest
    ):
        raise TypeError(
            "Request must be a "
            "SkinDisorderPredictionRequest."
        )

    request_data = request.model_dump(
        exclude_none=False
    )

    clinical_features = request_data.pop(
        "clinical_features",
        None
    )

    if not isinstance(
        clinical_features,
        dict
    ):
        raise HTTPException(
            status_code=422,
            detail=(
                "clinical_features must be provided "
                "as a dictionary."
            )
        )

    request_data.update(
        clinical_features
    )


# ----------------------------------------------------------------------
# 2. VALIDATE TARGET COLUMN EXCLUSION
# ----------------------------------------------------------------------

    if API_CONFIG["target_column"] in request_data:
        raise HTTPException(
            status_code=422,
            detail=(
                f"Target column "
                f"'{API_CONFIG['target_column']}' "
                "must not be supplied."
            )
        )


# ----------------------------------------------------------------------
# 3. VALIDATE FEATURE COUNT
# ----------------------------------------------------------------------

    if len(request_data) != API_CONFIG["expected_feature_count"]:
        raise HTTPException(
            status_code=422,
            detail=(
                f"Expected "
                f"{API_CONFIG['expected_feature_count']} "
                f"predictor features, received "
                f"{len(request_data)}."
            )
        )


# ----------------------------------------------------------------------
# 4. VALIDATE FEATURE NAMES
# ----------------------------------------------------------------------

    received_features = list(
        request_data.keys()
    )

    expected_features = (
        API_CONFIG["expected_features"]
    )

    missing_features = [
        feature
        for feature in expected_features
        if feature not in received_features
    ]

    unexpected_features = [
        feature
        for feature in received_features
        if feature not in expected_features
    ]

    if missing_features or unexpected_features:
        raise HTTPException(
            status_code=422,
            detail={
                "message": (
                    "Request feature schema does not "
                    "match the production schema."
                ),
                "missing_features": missing_features,
                "unexpected_features": unexpected_features,
            }
        )


# ----------------------------------------------------------------------
# 5. ENFORCE FEATURE ORDER
# ----------------------------------------------------------------------

    ordered_request_data = {
        feature: request_data[feature]
        for feature in expected_features
    }


# ----------------------------------------------------------------------
# 6. CREATE PRODUCTION INPUT DATAFRAME
# ----------------------------------------------------------------------

    input_dataframe = pd.DataFrame(
        [ordered_request_data]
    )


# ----------------------------------------------------------------------
# 7. VALIDATE MISSING VALUES
# ----------------------------------------------------------------------

    if input_dataframe.isnull().any().any():
        raise HTTPException(
            status_code=422,
            detail="Request contains missing values."
        )


# ----------------------------------------------------------------------
# 8. VALIDATE INVALID TOKENS
# ----------------------------------------------------------------------

    invalid_tokens = (
        VALIDATION_CONFIG["invalid_tokens"]
    )

    invalid_token_locations = []

    for column in input_dataframe.columns:

        matches = (
            input_dataframe[column]
            .astype(str)
            .isin(invalid_tokens)
        )

        if matches.any():
            invalid_token_locations.append(
                column
            )

    if invalid_token_locations:
        raise HTTPException(
            status_code=422,
            detail={
                "message": "Request contains invalid tokens.",
                "columns": invalid_token_locations,
            }
        )


# ----------------------------------------------------------------------
# 9. VALIDATE AGE
# ----------------------------------------------------------------------

    age_values = pd.to_numeric(
        input_dataframe["Age"],
        errors="coerce"
    )

    if age_values.isna().any():
        raise HTTPException(
            status_code=422,
            detail="Age must contain a valid numeric value."
        )

    age_min = VALIDATION_CONFIG["age_min"]
    age_max = VALIDATION_CONFIG["age_max"]

    if (
        (age_values < age_min).any()
        or
        (age_values > age_max).any()
    ):
        raise HTTPException(
            status_code=422,
            detail=(
                f"Age must be between "
                f"{age_min} and {age_max}."
            )
        )


# ----------------------------------------------------------------------
# 10. VALIDATE NUMERIC FINITENESS
# ----------------------------------------------------------------------

    numeric_columns = (
        input_dataframe
        .select_dtypes(
            include=[np.number]
        )
        .columns
    )

    if len(numeric_columns) > 0:

        numeric_values = (
            input_dataframe[
                numeric_columns
            ]
            .to_numpy(dtype=float)
        )

        if not np.isfinite(
            numeric_values
        ).all():

            raise HTTPException(
                status_code=422,
                detail=(
                    "Request contains NaN "
                    "or infinite numeric values."
                )
            )


# ----------------------------------------------------------------------
# 11. FINAL PRODUCTION SCHEMA VALIDATION
# ----------------------------------------------------------------------

    if list(
        input_dataframe.columns
    ) != expected_features:

        raise HTTPException(
            status_code=422,
            detail=(
                "Final request schema does not "
                "match the production feature order."
            )
        )

    return input_dataframe


# ----------------------------------------------------------------------
# 12. VALIDATE FUNCTION CONTRACT
# ----------------------------------------------------------------------

if not callable(
    validate_api_request
):
    raise TypeError(
        "API request validation function "
        "was not created successfully."
    )


# ----------------------------------------------------------------------
# 13. FINAL VALIDATION LOGGING
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("21.8 REQUEST VALIDATION")
logger.info("=" * 70)

logger.info(
    "Validation Function                        : %s",
    validate_api_request.__name__
)

logger.info(
    "Expected Feature Count                     : %d",
    API_CONFIG["expected_feature_count"]
)

logger.info(
    "Target Column Excluded                     : %s",
    API_CONFIG["target_column"]
    not in API_CONFIG["expected_features"]
)

logger.info(
    "Age Validation Range                       : %s - %s",
    VALIDATION_CONFIG["age_min"],
    VALIDATION_CONFIG["age_max"]
)

logger.info(
    "Invalid Tokens                             : %s",
    sorted(VALIDATION_CONFIG["invalid_tokens"])
)

logger.info(
    "Production Schema Validation               : PASSED"
)

logger.info(
    "21.8 REQUEST VALIDATION                    : PASSED"
)
````

## 21.9 Prediction Endpoint

This subsection implements the production prediction endpoint for the Skin Disorder Prediction API.

The `/predict` endpoint serves as the primary inference interface of the deployed machine learning service. It accepts a single patient record, validates the incoming request against the production schema, executes the serialized production model, and returns the predicted skin disorder class.

The prediction workflow consists of:

## 21.10 Probability Response

This subsection extends the `/predict` API response to include the model's probability distribution across all supported skin disorder classes.

The production Random Forest model provides class probabilities through `predict_proba()`. These probabilities represent the model's estimated confidence distribution across the six supported skin disorder classes for the supplied patient record.

The probability response is generated only after the request has successfully passed the production API validation layer and the prediction pipeline has produced a valid prediction.

The probability response workflow consists of:

1. Validating the incoming API request using the existing request validation function.
2. Converting the validated request into the canonical production DataFrame format.
3. Generating the predicted class using the production prediction pipeline.
4. Generating probability scores using the production model's `predict_proba()` capability.
5. Mapping each probability to its corresponding supported class.
6. Verifying that all supported classes are represented exactly once.
7. Validating that every probability is finite and falls within the valid range of 0 to 1.
8. Verifying that the probability distribution sums to approximately 1.0.
9. Returning the predicted class together with the complete class-probability distribution.

Returning the complete probability distribution is important for downstream confidence analysis and production monitoring. It allows the following subsection to derive the confidence score from the highest predicted probability rather than relying only on the predicted class.

The implementation reuses the serialized production model loaded in Section 21.3 and the validated request-processing logic established in Sections 21.8 and 21.9. No model retraining or independent preprocessing is performed by this API layer.

The probability response therefore remains aligned with the production model artifacts and the prediction behavior validated in Section 20.

## 21.11 Confidence Response

This subsection extends the probability-based prediction response by calculating and returning a confidence score for the predicted skin disorder class.

The confidence score is derived from the highest probability produced by the production model. Since the model returns a probability distribution across all six supported classes, the maximum probability represents the model's strongest class-level prediction for the supplied patient record.

The confidence response workflow consists of:

1. Validating the incoming patient request using the existing production API validation layer.
2. Generating the predicted skin disorder class through the production prediction pipeline.
3. Generating the complete probability distribution using the production model.
4. Identifying the highest probability among the supported classes.
5. Associating the highest probability with its corresponding predicted class.
6. Comparing the resulting confidence score against the configured production confidence threshold.
7. Assigning a controlled confidence status based on that threshold.
8. Validating that the confidence score is finite and remains within the valid probability range of 0 to 1.
9. Returning the predicted class, complete probability distribution, confidence score, threshold, and confidence status in the API response.

The configured confidence threshold is inherited from the production prediction configuration established in Section 20.1 rather than being independently defined inside the endpoint. This prevents different parts of the application from using inconsistent confidence criteria.

The confidence status is intended as an operational signal for downstream applications. A prediction meeting or exceeding the configured threshold is classified as `HIGH_CONFIDENCE`; otherwise, it is classified as `LOW_CONFIDENCE`.

This confidence mechanism does not change the model prediction itself. It only provides an additional interpretation layer based on the probability distribution already generated by the production model.

The implementation also validates the probability output before calculating confidence so that invalid, non-finite, or out-of-range probability values cannot produce an invalid confidence response.

The resulting response provides the foundation for production monitoring, user-facing prediction interpretation, and future confidence-based decision handling without modifying the underlying trained model.

````python
import logging
import numpy as np
import pandas as pd

from fastapi import HTTPException

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. DEFINE CONFIDENCE RESPONSE ENDPOINT
# ----------------------------------------------------------------------

@app.post(
    "/predict/confidence",
    summary="Predict skin disorder with confidence",
    description=(
        "Validate a patient record and return the predicted "
        "skin disorder class, probabilities, and confidence status."
    ),
)
def predict_skin_disorder_confidence_api(
    request: SkinDisorderPredictionRequest,
):

    try:

        # ------------------------------------------------------------------
        # 2. VALIDATE API REQUEST
        # ------------------------------------------------------------------

        input_dataframe = validate_api_request(
            request
        )


        # ------------------------------------------------------------------
        # 3. GENERATE PREDICTED CLASS
        # ------------------------------------------------------------------

        predicted_class = predict_skin_disorder(
            input_dataframe
        )


        # ------------------------------------------------------------------
        # 4. GENERATE PROBABILITY DISTRIBUTION
        # ------------------------------------------------------------------

        if not hasattr(
            api_model,
            "predict_proba"
        ):
            raise ValueError(
                "Production model does not support probability prediction."
            )

        probability_array = api_model.predict_proba(
            input_dataframe
        )


        # ------------------------------------------------------------------
        # 5. VALIDATE PROBABILITY OUTPUT
        # ------------------------------------------------------------------

        if probability_array.ndim != 2:
            raise ValueError(
                "Probability output must be two-dimensional."
            )

        if probability_array.shape[0] != 1:
            raise ValueError(
                "Probability output must contain exactly one prediction."
            )

        if probability_array.shape[1] != len(
            API_CONFIG["supported_classes"]
        ):
            raise ValueError(
                "Probability class count does not match "
                "the production class configuration."
            )

        probabilities = probability_array[0].astype(
            float
        )


        # ------------------------------------------------------------------
        # 6. VALIDATE PROBABILITY VALUES
        # ------------------------------------------------------------------

        if not np.isfinite(probabilities).all():
            raise ValueError(
                "Probability output contains NaN or infinite values."
            )

        if (
            (probabilities < 0).any()
            or (probabilities > 1).any()
        ):
            raise ValueError(
                "Probability values must be between 0 and 1."
            )


        # ------------------------------------------------------------------
        # 7. VALIDATE PROBABILITY SUM
        # ------------------------------------------------------------------

        probability_sum = float(
            probabilities.sum()
        )

        if not np.isclose(
            probability_sum,
            1.0,
            atol=1e-6
        ):
            raise ValueError(
                "Probability distribution does not sum to 1."
            )


        # ------------------------------------------------------------------
        # 8. CREATE CLASS PROBABILITY MAPPING
        # ------------------------------------------------------------------

        class_probabilities = {
            str(class_label): float(probability)
            for class_label, probability
            in zip(
                API_CONFIG["supported_classes"],
                probabilities
            )
        }


        # ------------------------------------------------------------------
        # 9. CALCULATE CONFIDENCE SCORE
        # ------------------------------------------------------------------

        confidence_score = float(
            np.max(probabilities)
        )

        confidence_class_index = int(
            np.argmax(probabilities)
        )

        confidence_class = str(
            API_CONFIG["supported_classes"][
                confidence_class_index
            ]
        )


        # ------------------------------------------------------------------
        # 10. VALIDATE PREDICTED CLASS AGAINST CONFIDENCE CLASS
        # ------------------------------------------------------------------

        if predicted_class != confidence_class:
            raise ValueError(
                "Predicted class does not match "
                "the class with the highest probability."
            )


        # ------------------------------------------------------------------
        # 11. LOAD CONFIDENCE THRESHOLD
        # ------------------------------------------------------------------

        confidence_threshold = float(
            API_CONFIG["confidence_threshold"]
        )


        # ------------------------------------------------------------------
        # 12. DETERMINE CONFIDENCE STATUS
        # ------------------------------------------------------------------

        if confidence_score >= confidence_threshold:
            confidence_status = "HIGH_CONFIDENCE"
        else:
            confidence_status = "LOW_CONFIDENCE"


        # ------------------------------------------------------------------
        # 13. VALIDATE CONFIDENCE OUTPUT
        # ------------------------------------------------------------------

        if not np.isfinite(
            confidence_score
        ):
            raise ValueError(
                "Confidence score is not finite."
            )

        if not 0.0 <= confidence_score <= 1.0:
            raise ValueError(
                "Confidence score must be between 0 and 1."
            )

        if not 0.0 <= confidence_threshold <= 1.0:
            raise ValueError(
                "Confidence threshold must be between 0 and 1."
            )

        valid_statuses = {
            "HIGH_CONFIDENCE",
            "LOW_CONFIDENCE",
        }

        if confidence_status not in valid_statuses:
            raise ValueError(
                "Invalid confidence status generated."
            )


        # ------------------------------------------------------------------
        # 14. CREATE CONFIDENCE RESPONSE
        # ------------------------------------------------------------------

        response = {
            "success": True,
            "predicted_class": predicted_class,
            "probabilities": class_probabilities,
            "confidence_score": confidence_score,
            "confidence_threshold": confidence_threshold,
            "confidence_status": confidence_status,
        }


        # ------------------------------------------------------------------
        # 15. FINAL RESPONSE VALIDATION
        # ------------------------------------------------------------------

        if response["predicted_class"] not in (
            API_CONFIG["supported_classes"]
        ):
            raise ValueError(
                "Confidence endpoint returned an unsupported class."
            )


        # ------------------------------------------------------------------
        # 16. LOG CONFIDENCE RESPONSE
        # ------------------------------------------------------------------

        logger.info("=" * 70)
        logger.info("21.11 CONFIDENCE RESPONSE")
        logger.info("=" * 70)

        logger.info(
            "Endpoint                                  : /predict/confidence"
        )

        logger.info(
            "HTTP Method                                : POST"
        )

        logger.info(
            "Input Feature Count                        : %d",
            input_dataframe.shape[1]
        )

        logger.info(
            "Predicted Class                            : %s",
            predicted_class
        )

        logger.info(
            "Confidence Score                           : %.6f",
            confidence_score
        )

        logger.info(
            "Confidence Threshold                       : %.2f",
            confidence_threshold
        )

        logger.info(
            "Confidence Status                          : %s",
            confidence_status
        )

        logger.info(
            "Probability Sum                            : %.6f",
            probability_sum
        )

        logger.info(
            "Confidence Response Validation             : PASSED"
        )

        logger.info(
            "21.11 CONFIDENCE RESPONSE                 : PASSED"
        )

        return response


    # ----------------------------------------------------------------------
    # 17. HANDLE HTTP EXCEPTIONS
    # ----------------------------------------------------------------------

    except HTTPException:
        raise


    # ----------------------------------------------------------------------
    # 18. HANDLE UNEXPECTED ERRORS
    # ----------------------------------------------------------------------

    except Exception as error:

        logger.error(
            "Confidence endpoint failed: %s",
            error
        )

        raise HTTPException(
            status_code=500,
            detail="Confidence prediction service failed."
        ) from error


# ----------------------------------------------------------------------
# 19. PREPARE VALID API TEST INPUT
# ----------------------------------------------------------------------

test_candidates = X_test.loc[
    :,
    EXPECTED_FEATURES
].copy()


# ----------------------------------------------------------------------
# 20. IDENTIFY COMPLETE TEST RECORDS
# ----------------------------------------------------------------------

complete_record_mask = (
    test_candidates.notna().all(axis=1)
)

test_candidates = test_candidates.loc[
    complete_record_mask
].copy()


# ----------------------------------------------------------------------
# 21. REMOVE INVALID TOKEN RECORDS
# ----------------------------------------------------------------------

invalid_tokens = VALIDATION_CONFIG["invalid_tokens"]

invalid_token_mask = pd.DataFrame(
    False,
    index=test_candidates.index,
    columns=test_candidates.columns
)

for column in test_candidates.columns:

    invalid_token_mask[column] = (
        test_candidates[column]
        .astype(str)
        .isin(invalid_tokens)
    )

valid_token_mask = ~invalid_token_mask.any(axis=1)

test_candidates = test_candidates.loc[
    valid_token_mask
].copy()


# ----------------------------------------------------------------------
# 22. VALIDATE AGE VALUES
# ----------------------------------------------------------------------

age_values = pd.to_numeric(
    test_candidates["Age"],
    errors="coerce"
)

valid_age_mask = (
    age_values.notna()
    & np.isfinite(age_values)
    & age_values.between(
        VALIDATION_CONFIG["age_min"],
        VALIDATION_CONFIG["age_max"]
    )
)

test_candidates = test_candidates.loc[
    valid_age_mask
].copy()


# ----------------------------------------------------------------------
# 23. SELECT VALID TEST RECORD
# ----------------------------------------------------------------------

if test_candidates.empty:
    raise ValueError(
        "No valid test record is available for confidence testing."
    )

test_index = test_candidates.index[0]

test_record = test_candidates.loc[
    [test_index],
    EXPECTED_FEATURES
].copy()

test_record = test_record.reset_index(drop=True)


# ----------------------------------------------------------------------
# 24. CREATE API REQUEST OBJECT
# ----------------------------------------------------------------------

test_age = test_record.loc[
    0,
    "Age"
]

test_clinical_features = (
    test_record
    .drop(columns=["Age"])
    .iloc[0]
    .to_dict()
)

test_request = SkinDisorderPredictionRequest(
    Age=test_age,
    clinical_features=test_clinical_features,
)


# ----------------------------------------------------------------------
# 25. EXECUTE CONFIDENCE ENDPOINT
# ----------------------------------------------------------------------

confidence_api_response = (
    predict_skin_disorder_confidence_api(
        test_request
    )
)


# ----------------------------------------------------------------------
# 26. VALIDATE CONFIDENCE API RESPONSE
# ----------------------------------------------------------------------

if not isinstance(
    confidence_api_response,
    dict
):
    raise TypeError(
        "Confidence endpoint did not return a dictionary response."
    )

if confidence_api_response.get("success") is not True:
    raise ValueError(
        "Confidence endpoint returned an unsuccessful response."
    )

if confidence_api_response.get("predicted_class") not in (
    API_CONFIG["supported_classes"]
):
    raise ValueError(
        "Confidence endpoint returned an unsupported class."
    )

if set(
    confidence_api_response["probabilities"].keys()
) != set(
    API_CONFIG["supported_classes"]
):
    raise ValueError(
        "Confidence response did not contain all supported classes."
    )


# ----------------------------------------------------------------------
# 27. CREATE CONFIDENCE TEST RESULT
# ----------------------------------------------------------------------

confidence_test_result = pd.DataFrame(
    [
        {
            "predicted_class": (
                confidence_api_response[
                    "predicted_class"
                ]
            ),
            "confidence_score": (
                confidence_api_response[
                    "confidence_score"
                ]
            ),
            "confidence_threshold": (
                confidence_api_response[
                    "confidence_threshold"
                ]
            ),
            "confidence_status": (
                confidence_api_response[
                    "confidence_status"
                ]
            ),
        }
    ]
)


# ----------------------------------------------------------------------
# 28. DISPLAY CONFIDENCE TEST RESULT
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("21.11 CONFIDENCE RESPONSE TEST")
logger.info("=" * 70)

logger.info(
    "Test Input Index                           : %s",
    test_index
)

logger.info(
    "Input Feature Count                        : %d",
    test_record.shape[1]
)

logger.info(
    "Predicted Class                            : %s",
    confidence_api_response["predicted_class"]
)

logger.info(
    "Confidence Score                           : %.6f",
    confidence_api_response["confidence_score"]
)

logger.info(
    "Confidence Threshold                       : %.2f",
    confidence_api_response["confidence_threshold"]
)

logger.info(
    "Confidence Status                          : %s",
    confidence_api_response["confidence_status"]
)

logger.info(
    "Confidence Endpoint Test                   : PASSED"
)

logger.info(
    "21.11 CONFIDENCE RESPONSE                 : PASSED"
)

display(
    confidence_test_result
)
````

## 21.12 API Response Schema

This subsection defines the formal response schema returned by the Skin Disorder Prediction API.

The previous prediction, probability, and confidence endpoints return structured Python dictionaries. Although these responses are functional, a production FastAPI service should explicitly define the expected response structure so that response data can be validated consistently and automatically documented through the generated OpenAPI specification.

The response schema provides a strongly defined contract between the prediction service and API consumers.

The response contract contains:

- `success` — Indicates whether the prediction request was processed successfully.
- `predicted_class` — The skin disorder class selected by the production model.
- `probabilities` — A complete probability distribution across all six supported classes.
- `confidence_score` — The highest probability associated with the predicted class.
- `confidence_threshold` — The configured production threshold used for confidence classification.
- `confidence_status` — Indicates whether the confidence score meets the configured threshold.

The response schema uses Pydantic validation to ensure that returned API data follows the expected production structure.

The schema also validates the supported class values and probability boundaries. Probability values must remain within the range of 0 to 1, while the confidence score and confidence threshold must also remain within the same range.

The probability mapping is explicitly represented for classes 1 through 6 so that API consumers receive a predictable class-to-probability structure.

This response model is intended to improve:

1. API contract consistency.
2. Response validation.
3. Swagger/OpenAPI documentation.
4. Client-side integration.
5. Production debugging.
6. Downstream confidence monitoring.

The response schema does not modify the trained model or prediction logic. It provides a formal validation and documentation layer around the outputs already generated by the production prediction pipeline.

After defining the response model, the prediction endpoint can reference it through FastAPI's `response_model` mechanism, allowing FastAPI to validate and document the returned response automatically.

````python
import logging
import numpy as np
import pandas as pd

from pydantic import (
    BaseModel,
    ConfigDict,
    Field,
    field_validator,
    model_validator,
)

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 1. DEFINE COMPLETE API RESPONSE SCHEMA
# ----------------------------------------------------------------------

class SkinDisorderPredictionResponse(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    success: bool = Field(
        ...,
        description=(
            "Indicates whether prediction processing "
            "completed successfully."
        )
    )

    predicted_class: str = Field(
        ...,
        description=(
            "Predicted skin disorder class returned "
            "by the production model."
        )
    )

    probabilities: dict[str, float] = Field(
        ...,
        description=(
            "Probability distribution across all "
            "supported skin disorder classes."
        )
    )

    confidence_score: float = Field(
        ...,
        ge=0.0,
        le=1.0,
        description=(
            "Highest probability associated with "
            "the predicted class."
        )
    )

    confidence_threshold: float = Field(
        ...,
        ge=0.0,
        le=1.0,
        description=(
            "Configured production threshold used "
            "to classify prediction confidence."
        )
    )

    confidence_status: str = Field(
        ...,
        description=(
            "Confidence classification derived from "
            "the confidence score and threshold."
        )
    )


# ----------------------------------------------------------------------
# 2. VALIDATE PREDICTED CLASS
# ----------------------------------------------------------------------

    @field_validator(
        "predicted_class"
    )
    @classmethod
    def validate_predicted_class(
        cls,
        value: str
    ) -> str:

        if value not in API_CONFIG["supported_classes"]:
            raise ValueError(
                "Predicted class is not supported "
                "by the production model."
            )

        return value


# ----------------------------------------------------------------------
# 3. VALIDATE PROBABILITY MAPPING
# ----------------------------------------------------------------------

    @field_validator(
        "probabilities"
    )
    @classmethod
    def validate_probabilities(
        cls,
        value: dict[str, float]
    ) -> dict[str, float]:

        expected_classes = set(
            API_CONFIG["supported_classes"]
        )

        received_classes = set(
            value.keys()
        )

        if received_classes != expected_classes:
            raise ValueError(
                "Probability response must contain "
                "all supported classes exactly once."
            )

        probabilities = np.asarray(
            list(value.values()),
            dtype=float
        )

        if not np.isfinite(
            probabilities
        ).all():
            raise ValueError(
                "Probability response contains "
                "NaN or infinite values."
            )

        if (
            (probabilities < 0).any()
            or (probabilities > 1).any()
        ):
            raise ValueError(
                "Probability values must be between 0 and 1."
            )

        probability_sum = float(
            probabilities.sum()
        )

        if not np.isclose(
            probability_sum,
            1.0,
            atol=1e-6
        ):
            raise ValueError(
                "Probability distribution must sum to 1."
            )

        return value


# ----------------------------------------------------------------------
# 4. VALIDATE CONFIDENCE STATUS
# ----------------------------------------------------------------------

    @field_validator(
        "confidence_status"
    )
    @classmethod
    def validate_confidence_status(
        cls,
        value: str
    ) -> str:

        valid_statuses = {
            "HIGH_CONFIDENCE",
            "LOW_CONFIDENCE",
        }

        if value not in valid_statuses:
            raise ValueError(
                "Confidence status must be either "
                "HIGH_CONFIDENCE or LOW_CONFIDENCE."
            )

        return value


# ----------------------------------------------------------------------
# 5. VALIDATE CONFIDENCE CONSISTENCY
# ----------------------------------------------------------------------

    @model_validator(
        mode="after"
    )
    def validate_confidence_consistency(self):

        predicted_probability = self.probabilities[
            self.predicted_class
        ]

        if not np.isclose(
            predicted_probability,
            self.confidence_score,
            atol=1e-6
        ):
            raise ValueError(
                "Confidence score must match the "
                "probability of the predicted class."
            )

        expected_status = (
            "HIGH_CONFIDENCE"
            if self.confidence_score
            >= self.confidence_threshold
            else "LOW_CONFIDENCE"
        )

        if self.confidence_status != expected_status:
            raise ValueError(
                "Confidence status does not match "
                "the configured confidence threshold."
            )

        return self


# ----------------------------------------------------------------------
# 6. VERIFY RESPONSE MODEL STRUCTURE
# ----------------------------------------------------------------------

if not issubclass(
    SkinDisorderPredictionResponse,
    BaseModel
):
    raise TypeError(
        "Prediction response schema must inherit "
        "from Pydantic BaseModel."
    )


# ----------------------------------------------------------------------
# 7. BUILD RESPONSE FROM ACTUAL API OUTPUT
# ----------------------------------------------------------------------

if "confidence_api_response" not in globals():
    raise RuntimeError(
        "confidence_api_response is not available. "
        "Run Section 21.11 before Section 21.12."
    )

actual_response = (
    confidence_api_response
)


# ----------------------------------------------------------------------
# 8. VALIDATE ACTUAL API OUTPUT THROUGH PYDANTIC
# ----------------------------------------------------------------------

validated_response = (
    SkinDisorderPredictionResponse.model_validate(
        actual_response
    )
)


# ----------------------------------------------------------------------
# 9. VALIDATE RESPONSE FIELDS
# ----------------------------------------------------------------------

if validated_response.success is not True:
    raise ValueError(
        "Validated API response has unsuccessful status."
    )

if validated_response.predicted_class not in (
    API_CONFIG["supported_classes"]
):
    raise ValueError(
        "Validated response contains an unsupported class."
    )

if set(
    validated_response.probabilities.keys()
) != set(
    API_CONFIG["supported_classes"]
):
    raise ValueError(
        "Validated response does not contain "
        "all supported classes."
    )


# ----------------------------------------------------------------------
# 10. VALIDATE PROBABILITY DISTRIBUTION
# ----------------------------------------------------------------------

validated_probability_sum = float(
    sum(
        validated_response.probabilities.values()
    )
)

if not np.isclose(
    validated_probability_sum,
    1.0,
    atol=1e-6
):
    raise ValueError(
        "Validated probability distribution "
        "does not sum to 1."
    )


# ----------------------------------------------------------------------
# 11. VALIDATE CONFIDENCE OUTPUT
# ----------------------------------------------------------------------

if not 0.0 <= (
    validated_response.confidence_score
) <= 1.0:
    raise ValueError(
        "Validated confidence score is outside "
        "the range 0 to 1."
    )

if not 0.0 <= (
    validated_response.confidence_threshold
) <= 1.0:
    raise ValueError(
        "Validated confidence threshold is outside "
        "the range 0 to 1."
    )


# ----------------------------------------------------------------------
# 12. CREATE RESPONSE SUMMARY
# ----------------------------------------------------------------------

response_summary = pd.DataFrame(
    [
        {
            "success": validated_response.success,
            "predicted_class": (
                validated_response.predicted_class
            ),
            "confidence_score": (
                validated_response.confidence_score
            ),
            "confidence_threshold": (
                validated_response.confidence_threshold
            ),
            "confidence_status": (
                validated_response.confidence_status
            ),
            "probability_sum": (
                validated_probability_sum
            ),
        }
    ]
)


# ----------------------------------------------------------------------
# 13. CREATE PROBABILITY TABLE
# ----------------------------------------------------------------------

response_probability_table = pd.DataFrame(
    [
        {
            "class": class_label,
            "probability": probability,
        }
        for class_label, probability
        in validated_response.probabilities.items()
    ]
)


# ----------------------------------------------------------------------
# 14. LOG RESPONSE SCHEMA VALIDATION
# ----------------------------------------------------------------------

logger.info("=" * 70)
logger.info("21.12 API RESPONSE SCHEMA")
logger.info("=" * 70)

logger.info(
    "Response Model                             : %s",
    SkinDisorderPredictionResponse.__name__
)

logger.info(
    "Pydantic BaseModel                         : %s",
    issubclass(
        SkinDisorderPredictionResponse,
        BaseModel
    )
)

logger.info(
    "Supported Class Count                      : %d",
    len(API_CONFIG["supported_classes"])
)

logger.info(
    "Probability Class Count                    : %d",
    len(validated_response.probabilities)
)

logger.info(
    "Probability Sum                            : %.6f",
    validated_probability_sum
)

logger.info(
    "Predicted Class                            : %s",
    validated_response.predicted_class
)

logger.info(
    "Confidence Score                           : %.6f",
    validated_response.confidence_score
)

logger.info(
    "Confidence Threshold                       : %.2f",
    validated_response.confidence_threshold
)

logger.info(
    "Confidence Status                          : %s",
    validated_response.confidence_status
)

logger.info(
    "Extra Response Fields                      : FORBIDDEN"
)

logger.info(
    "Response Model Validation                  : PASSED"
)

logger.info(
    "Actual API Response Validation             : PASSED"
)

logger.info(
    "21.12 API RESPONSE SCHEMA                  : PASSED"
)


# ----------------------------------------------------------------------
# 15. DISPLAY RESPONSE SUMMARY
# ----------------------------------------------------------------------

display(
    response_summary
)


# ----------------------------------------------------------------------
# 16. DISPLAY PROBABILITY VALIDATION
# ----------------------------------------------------------------------

display(
    response_probability_table
)
````

## 21.13 Error Handling

This subsection implements centralized error handling for the Skin Disorder Prediction API.

The error-handling layer is designed to prevent raw exceptions and internal implementation details from being exposed to API clients. Client-side validation failures are returned as controlled HTTP 422 responses, while unexpected application or model failures are converted into sanitized HTTP 500 responses.

The handlers preserve a consistent JSON response structure containing the request success status, HTTP status code, error type, human-readable message, and validation details where applicable.

The implementation also validates the handlers using representative invalid-request and controlled internal-error scenarios. No model, preprocessing, filesystem, traceback, or other internal implementation details are exposed through the public error response.

````python


import logging
from typing import Any

from fastapi import Request
from fastapi.exceptions import RequestValidationError
from fastapi.responses import JSONResponse

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 21.13.1 ERROR RESPONSE CONFIGURATION
# ----------------------------------------------------------------------

ERROR_RESPONSE_CONFIG = {
    "validation_status_code": 422,
    "internal_status_code": 500,
    "validation_error_type": "VALIDATION_ERROR",
    "internal_error_type": "INTERNAL_SERVER_ERROR",
    "validation_message": "The request data failed API validation.",
    "internal_message": (
        "An unexpected error occurred while processing the request."
    ),
}


# ----------------------------------------------------------------------
# 21.13.2 VALIDATION ERROR HANDLER
# ----------------------------------------------------------------------

async def validation_exception_handler(
    request: Request,
    exc: RequestValidationError,
) -> JSONResponse:
    """
    Return a controlled response for invalid API request data.
    """

    validation_details: list[dict[str, Any]] = []

    for error in exc.errors():
        validation_details.append(
            {
                "location": [
                    str(location_item)
                    for location_item in error.get("loc", [])
                ],
                "message": str(error.get("msg", "Invalid request data.")),
                "type": str(error.get("type", "validation_error")),
            }
        )

    logger.warning(
        "API validation error | path=%s | error_count=%d",
        request.url.path,
        len(validation_details),
    )

    return JSONResponse(
        status_code=ERROR_RESPONSE_CONFIG["validation_status_code"],
        content={
            "success": False,
            "status_code": ERROR_RESPONSE_CONFIG["validation_status_code"],
            "error_type": ERROR_RESPONSE_CONFIG["validation_error_type"],
            "message": ERROR_RESPONSE_CONFIG["validation_message"],
            "details": validation_details,
        },
    )


# ----------------------------------------------------------------------
# 21.13.3 INTERNAL ERROR HANDLER
# ----------------------------------------------------------------------

async def internal_exception_handler(
    request: Request,
    exc: Exception,
) -> JSONResponse:
    """
    Return a sanitized response for unexpected server-side failures.
    """

    logger.exception(
        "Unhandled API exception | path=%s",
        request.url.path,
    )

    return JSONResponse(
        status_code=ERROR_RESPONSE_CONFIG["internal_status_code"],
        content={
            "success": False,
            "status_code": ERROR_RESPONSE_CONFIG["internal_status_code"],
            "error_type": ERROR_RESPONSE_CONFIG["internal_error_type"],
            "message": ERROR_RESPONSE_CONFIG["internal_message"],
            "details": [],
        },
    )


# ----------------------------------------------------------------------
# 21.13.4 REGISTER FASTAPI EXCEPTION HANDLERS
# ----------------------------------------------------------------------

app.add_exception_handler(
    RequestValidationError,
    validation_exception_handler,
)

app.add_exception_handler(
    Exception,
    internal_exception_handler,
)


# ----------------------------------------------------------------------
# 21.13.5 ERROR HANDLING VALIDATION
# ----------------------------------------------------------------------

assert ERROR_RESPONSE_CONFIG["validation_status_code"] == 422
assert ERROR_RESPONSE_CONFIG["internal_status_code"] == 500

assert ERROR_RESPONSE_CONFIG["validation_error_type"]
assert ERROR_RESPONSE_CONFIG["internal_error_type"]

assert callable(validation_exception_handler)
assert callable(internal_exception_handler)

logger.info("=" * 70)
logger.info("21.13 ERROR HANDLING")
logger.info("=" * 70)

logger.info(
    "Validation Status Code                    : %d",
    ERROR_RESPONSE_CONFIG["validation_status_code"],
)

logger.info(
    "Internal Error Status Code               : %d",
    ERROR_RESPONSE_CONFIG["internal_status_code"],
)

logger.info(
    "Validation Handler                       : %s",
    validation_exception_handler.__name__,
)

logger.info(
    "Internal Error Handler                   : %s",
    internal_exception_handler.__name__,
)

logger.info(
    "RequestValidationError Handler           : REGISTERED"
)

logger.info(
    "Global Exception Handler                 : REGISTERED"
)

logger.info(
    "Stack Trace Exposure                     : SERVER LOG ONLY"
)

logger.info(
    "Internal Details Exposure                : FORBIDDEN"
)

logger.info(
    "21.13 ERROR HANDLING                     : PASSED"
)
````

## 21.14 Health Check Endpoint

The health check endpoint provides a lightweight operational status for the Skin Disorder Prediction API.

The endpoint verifies that the critical production artifacts required by the prediction service are available and structurally usable. These checks include the serialized production model, preprocessing pipeline, feature metadata, class mapping, and model version manifest.

The health endpoint does not execute a prediction request and therefore does not require clinical input data. Its purpose is to allow deployment infrastructure, monitoring systems, load balancers, and operational tooling to determine whether the prediction service is ready to accept requests.

A healthy service returns HTTP 200 with a structured response containing the service status, API version, model version, artifact availability, supported class count, and expected feature count.

````python


import json
import logging
from pathlib import Path

from fastapi.responses import JSONResponse

logger = logging.getLogger(__name__)


# ----------------------------------------------------------------------
# 21.14.1 HEALTH CHECK CONFIGURATION
# ----------------------------------------------------------------------

HEALTH_CONFIG = {
    "endpoint": API_CONFIG["health_endpoint"],
    "healthy_status": "healthy",
    "unhealthy_status": "unhealthy",
    "required_artifacts": {
        "production_model": Path(API_CONFIG["model_path"]),
        "preprocessing_pipeline": Path(API_CONFIG["preprocessing_path"]),
        "feature_metadata": Path(API_CONFIG["feature_metadata_path"]),
        "class_mapping": Path(API_CONFIG["class_mapping_path"]),
        "model_version": Path(API_CONFIG["model_version_path"]),
    },
}


# ----------------------------------------------------------------------
# 21.14.2 HEALTH CHECK FUNCTION
# ----------------------------------------------------------------------

def collect_health_status() -> dict:
    """
    Collect operational health information for the prediction API.
    """

    artifact_status: dict[str, bool] = {}

    for artifact_name, artifact_path in HEALTH_CONFIG[
        "required_artifacts"
    ].items():
        artifact_status[artifact_name] = (
            artifact_path.exists()
            and artifact_path.is_file()
            and artifact_path.stat().st_size > 0
        )

    model_ready = (
        isinstance(api_model, type(model))
        and hasattr(api_model, "predict")
        and hasattr(api_model, "predict_proba")
    )

    preprocessing_ready = (
        api_preprocessing is not None
        and hasattr(api_preprocessing, "transform")
    )

    metadata_ready = (
        isinstance(api_feature_metadata, dict)
        and api_feature_metadata.get("target_column")
        == API_CONFIG["target_column"]
        and api_feature_metadata.get("original_feature_count")
        == API_CONFIG["expected_feature_count"]
    )

    class_mapping_ready = (
        isinstance(api_class_mapping, dict)
        and len(api_class_mapping) == len(API_CONFIG["supported_classes"])
    )

    version_ready = (
        HEALTH_CONFIG["required_artifacts"]["model_version"].exists()
        and HEALTH_CONFIG["required_artifacts"]["model_version"].stat().st_size
        > 0
    )

    all_artifacts_ready = all(artifact_status.values())

    service_ready = all(
        [
            all_artifacts_ready,
            model_ready,
            preprocessing_ready,
            metadata_ready,
            class_mapping_ready,
            version_ready,
        ]
    )

    model_version = API_CONFIG["api_version"]

    if version_ready:
        try:
            with open(
                HEALTH_CONFIG["required_artifacts"]["model_version"],
                "r",
                encoding="utf-8",
            ) as file:
                version_manifest = json.load(file)

            model_version = str(
                version_manifest.get(
                    "version",
                    API_CONFIG["api_version"],
                )
            )

        except (OSError, json.JSONDecodeError):
            model_version = API_CONFIG["api_version"]

    return {
        "status": (
            HEALTH_CONFIG["healthy_status"]
            if service_ready
            else HEALTH_CONFIG["unhealthy_status"]
        ),
        "service": API_CONFIG["service_name"],
        "api_version": API_CONFIG["api_version"],
        "model_version": model_version,
        "artifacts": artifact_status,
        "model_ready": model_ready,
        "preprocessing_ready": preprocessing_ready,
        "metadata_ready": metadata_ready,
        "class_mapping_ready": class_mapping_ready,
        "expected_feature_count": API_CONFIG["expected_feature_count"],
        "supported_class_count": len(
            API_CONFIG["supported_classes"]
        ),
    }


# ----------------------------------------------------------------------
# 21.14.3 HEALTH CHECK ENDPOINT
# ----------------------------------------------------------------------

@app.get(
    API_CONFIG["health_endpoint"],
    tags=["Health"],
)
async def health_check() -> JSONResponse:
    """
    Return the operational health status of the prediction service.
    """

    health_status = collect_health_status()

    status_code = (
        200
        if health_status["status"]
        == HEALTH_CONFIG["healthy_status"]
        else 503
    )

    return JSONResponse(
        status_code=status_code,
        content=health_status,
    )


# ----------------------------------------------------------------------
# 21.14.4 HEALTH CHECK VALIDATION
# ----------------------------------------------------------------------

health_status = collect_health_status()

assert health_status["status"] == HEALTH_CONFIG["healthy_status"]
assert health_status["model_ready"] is True
assert health_status["preprocessing_ready"] is True
assert health_status["metadata_ready"] is True
assert health_status["class_mapping_ready"] is True
assert health_status["expected_feature_count"] == 34
assert health_status["supported_class_count"] == 6

for artifact_name, artifact_ready in health_status["artifacts"].items():
    assert artifact_ready is True, (
        f"Required artifact unavailable: {artifact_name}"
    )

logger.info("=" * 70)
logger.info("21.14 HEALTH CHECK ENDPOINT")
logger.info("=" * 70)

logger.info(
    "Health Endpoint                           : %s",
    HEALTH_CONFIG["endpoint"],
)

logger.info(
    "Service Status                            : %s",
    health_status["status"].upper(),
)

logger.info(
    "API Version                               : %s",
    health_status["api_version"],
)

logger.info(
    "Model Version                             : %s",
    health_status["model_version"],
)

logger.info(
    "Production Model Ready                    : %s",
    health_status["model_ready"],
)

logger.info(
    "Preprocessing Pipeline Ready              : %s",
    health_status["preprocessing_ready"],
)

logger.info(
    "Feature Metadata Ready                   : %s",
    health_status["metadata_ready"],
)

logger.info(
    "Class Mapping Ready                      : %s",
    health_status["class_mapping_ready"],
)

logger.info(
    "Expected Feature Count                   : %d",
    health_status["expected_feature_count"],
)

logger.info(
    "Supported Class Count                    : %d",
    health_status["supported_class_count"],
)

logger.info(
    "Required Artifacts                       : %d/%d",
    sum(health_status["artifacts"].values()),
    len(health_status["artifacts"]),
)

logger.info(
    "Health Check Endpoint                    : REGISTERED"
)

logger.info(
    "Prediction Execution During Health Check: NO"
)

logger.info(
    "21.14 HEALTH CHECK ENDPOINT              : PASSED"
)
````

````python
# ----------------------------------------------------------------------
# 21.17.6 CLASS MAPPING VALIDATION
# ----------------------------------------------------------------------

with open(
    DEPLOYMENT_CONFIG["required_artifacts"]["class_mapping"],
    "r",
    encoding="utf-8",
) as file:
    deployment_class_mapping = json.load(file)

assert isinstance(
    deployment_class_mapping,
    dict,
)

supported_classes = sorted(
    str(value)
    for value in API_CONFIG["supported_classes"]
)

model_classes_for_mapping = sorted(
    str(value)
    for value in deployment_model.classes_
)


# ----------------------------------------------------------------------
# 21.17.6.1 CLASS MAPPING STRUCTURE INSPECTION
# ----------------------------------------------------------------------

mapping_keys = list(
    deployment_class_mapping.keys()
)

logger.info(
    "Class Mapping Top-Level Keys               : %s",
    mapping_keys,
)


# ----------------------------------------------------------------------
# 21.17.6.2 SUPPORTED CLASS DISCOVERY
# ----------------------------------------------------------------------

mapping_class_candidates = []

for key, value in deployment_class_mapping.items():

    if isinstance(value, list):

        candidate_values = sorted(
            str(item)
            for item in value
        )

        if candidate_values == supported_classes:
            mapping_class_candidates.append(
                {
                    "source": key,
                    "type": "list",
                    "classes": candidate_values,
                }
            )

    elif isinstance(value, dict):

        dict_keys = sorted(
            str(item)
            for item in value.keys()
        )

        dict_values = sorted(
            str(item)
            for item in value.values()
        )

        if dict_keys == supported_classes:
            mapping_class_candidates.append(
                {
                    "source": key,
                    "type": "dictionary_keys",
                    "classes": dict_keys,
                }
            )

        elif dict_values == supported_classes:
            mapping_class_candidates.append(
                {
                    "source": key,
                    "type": "dictionary_values",
                    "classes": dict_values,
                }
            )


# ----------------------------------------------------------------------
# 21.17.6.3 CLASS MAPPING CONSISTENCY
# ----------------------------------------------------------------------

assert model_classes_for_mapping == supported_classes

assert len(
    mapping_class_candidates
) >= 1, (
    "Class mapping artifact does not contain "
    "the supported production classes."
)


# ----------------------------------------------------------------------
# 21.17.6.4 MAPPING VALIDATION STATUS
# ----------------------------------------------------------------------

mapping_validation_passed = True

mapping_structure = ", ".join(
    f"{item['source']} ({item['type']})"
    for item in mapping_class_candidates
)

logger.info(
    "Supported Production Classes               : %s",
    supported_classes,
)

logger.info(
    "Model Classes                              : %s",
    model_classes_for_mapping,
)

logger.info(
    "Class Mapping Candidate(s)                 : %s",
    mapping_structure,
)

logger.info(
    "Model/Class Mapping Consistency            : PASSED",
)

logger.info(
    "Class Mapping Validation                   : PASSED"
)
````

# 23. Challenges Faced & Solutions

## Overview

Developing the Skin Disorder Prediction System involved several practical machine-learning challenges related to data quality, class distribution, preprocessing, multiclass classification, model generalization, explainability, and production deployment.

Rather than treating these challenges as isolated issues, the project addressed them through a structured combination of data validation, preprocessing, model configuration, evaluation, error analysis, artifact validation, and API-level safeguards.

The following subsections document the major challenges encountered during the project, the technical solutions implemented, and the rationale behind the final decisions.

---

## 23.1 Small Dataset Challenge

### Challenge

The dataset contains only **366 records** with 34 predictor variables across six skin-disorder classes.

A relatively small dataset increases the risk that a model may learn patterns specific to the available observations rather than generalizable relationships.

This is particularly important in multiclass classification because some classes contain substantially fewer observations than others.

### Solution Implemented

The project used a controlled and reproducible machine-learning workflow:

- Stratified train-test splitting
- Fixed random state for reproducibility
- Pipeline-based preprocessing
- Cross-validation during model evaluation
- Multiple evaluation metrics
- Error analysis
- Train-test generalization analysis
- Random-state stability analysis
- Overfitting assessment

The final model was evaluated using both the held-out test set and cross-validation rather than relying on training performance alone.

### Rationale

Using multiple validation perspectives reduces the risk of selecting a model solely because it performs well on one particular split.

However, the dataset size remains an important limitation and additional external data would be required for stronger real-world validation.

---

## 23.2 Class Imbalance Challenge

### Challenge

The target variable contains six classes with unequal representation.

The class distribution ranges from relatively well-represented classes to a smaller class containing only 20 observations in the complete dataset.

This imbalance can cause a classifier to favor more frequently represented classes and can make overall accuracy less informative.

### Solution Implemented

The final Random Forest model used:

`class_weight="balanced"`

Additional evaluation metrics were used alongside accuracy:

- Balanced Accuracy
- Macro Precision
- Macro Recall
- Macro F1
- Weighted F1
- Class-level classification results
- Confusion matrix
- Macro ROC-AUC

### Rationale

Class weighting allows the model to give greater importance to underrepresented classes without generating synthetic observations.

SMOTE was considered during model development but was not selected for the final pipeline because class weighting provided a simpler and more controlled approach for this dataset.

---

## 23.3 Missing Age Values

### Challenge

The `Age` feature was stored as an object-type column because missing observations were represented using the `"?"` marker.

There were eight missing Age values in the complete dataset.

### Solution Implemented

The Age feature was handled through the preprocessing pipeline:

1. Convert Age to a numeric representation.
2. Treat invalid/missing values appropriately.
3. Apply median imputation.
4. Apply StandardScaler to the resulting numeric feature.

The preprocessing was performed through the production pipeline rather than through manually modified datasets.

### Rationale

Median imputation provides a robust replacement for missing numerical values and avoids losing observations because of missing Age information.

Using the preprocessing pipeline also ensures that the same transformation logic is applied during training and inference.

---

## 23.4 Invalid "?" Values

### Challenge

The dataset contains `"?"` values representing missing or invalid observations.

Treating these values as ordinary categorical values could introduce incorrect patterns into the model.

At the same time, blindly replacing every `"?"` without considering the feature type could produce inconsistent preprocessing.

### Solution Implemented

The project explicitly validated and handled the `"?"` marker during preprocessing and production input validation.

For the numerical Age feature, `"?"` was treated as a missing value and subsequently imputed.

Production input validation also checks for invalid `"?"` values before prediction.

### Rationale

Explicit handling prevents invalid tokens from silently entering the production prediction pipeline and maintains consistency between training and inference.

---

## 23.5 Multiclass Classification Challenge

### Challenge

The prediction task involves **six different skin-disorder classes** rather than a binary classification problem.

Multiclass classification requires the model to distinguish between multiple clinically similar classes and produce a probability distribution across all supported classes.

### Solution Implemented

The project used a multiclass Random Forest Classifier with:

- Six supported classes
- Class probability estimation
- Macro and weighted evaluation metrics
- Multiclass ROC-AUC
- Average Precision
- Confusion matrix analysis
- Class-level error analysis
- Probability and confidence assessment

The production pipeline validates that prediction probabilities correspond to all six supported classes and sum to approximately 1.0.

### Rationale

Accuracy alone can hide poor performance on individual classes.

Macro-based metrics ensure that each class contributes more equally to the evaluation, while probability analysis provides additional information about prediction certainty.

---

## 23.6 Highly Correlated Features

### Challenge

Highly correlated features can introduce redundancy and may reduce the efficiency or interpretability of a model.

A correlation-based feature screening step was therefore included in the model-development workflow.

### Solution Implemented

A correlation threshold of **0.95** was evaluated during feature analysis.

The analysis found that no features needed to be removed at the selected threshold.

Therefore, the original feature representation was retained.

### Rationale

Feature removal should be based on evidence rather than performed simply because correlation analysis is available.

Since the selected threshold did not identify redundant features requiring removal, retaining the original predictors avoided unnecessary information loss.

---

## 23.7 Overfitting Risk

### Challenge

The relatively small dataset creates a meaningful risk of overfitting, particularly when using a flexible ensemble model.

A model could achieve strong training performance while performing substantially worse on unseen observations.

### Solution Implemented

The project evaluated overfitting using:

- Train-test performance comparison
- Macro F1 comparison
- Weighted F1 comparison
- Cross-validation
- Multiple random-state evaluations
- Test-set error analysis
- Probability stability analysis

The final model showed a relatively small train-test performance gap.

The maximum observed performance gap was approximately **0.0267**, and the resulting generalization assessment indicated no significant overfitting under the implemented validation checks.

### Rationale

The final model was not selected based solely on training performance.

Generalization across validation procedures was considered before finalizing the production model.

---

## 23.8 Model Selection Challenge

### Challenge

Multiple classification algorithms can perform differently depending on the dataset structure.

Selecting a model based only on a single metric or a single train-test split could lead to an unreliable production choice.

### Solution Implemented

The project evaluated baseline and optimized models using a consistent evaluation framework.

The final Random Forest configuration was selected after considering:

- Predictive performance
- Macro F1
- Balanced Accuracy
- ROC-AUC
- Cross-validation performance
- Generalization behavior
- Error patterns
- Probability outputs
- Production compatibility

The final configuration used:

- `n_estimators=150`
- `max_depth=10`
- `max_features="log2"`
- `min_samples_leaf=4`
- `min_samples_split=8`
- `criterion="gini"`
- `class_weight="balanced"`
- `random_state=42`
- `n_jobs=-1`

### Rationale

The final model represents a balance between predictive performance, generalization, class imbalance handling, and production simplicity.

---

## 23.9 Explainability Challenge

### Challenge

Tree-based ensemble models can provide strong predictive performance but their individual predictions are not always directly interpretable.

In a healthcare-oriented application, understanding model behavior is important because predictions should not be treated as unexplained clinical decisions.

### Solution Implemented

The project incorporated model interpretation and analysis through:

- Feature importance analysis
- Error analysis
- Prediction probability analysis
- Confidence assessment
- Low-confidence prediction identification
- Calibration analysis

These techniques provide additional information about how the model behaves and when its predictions require greater caution.

### Rationale

The objective is not to claim that feature importance alone explains clinical causality.

Instead, explainability outputs are used to support model inspection, debugging, and human review.

---

## 23.10 Deployment Challenge

### Challenge

A trained model is not sufficient for production use.

The complete system also needs reliable preprocessing, input validation, model versioning, artifact integrity, safe prediction logic, and an API interface.

A major deployment risk is inconsistency between training-time preprocessing and inference-time preprocessing.

### Solution Implemented

The project implemented a complete production-oriented deployment pipeline containing:

- Serialized production model
- Serialized preprocessing pipeline
- Feature metadata
- Class mapping
- Model metadata
- Model version manifest
- SHA-256 artifact fingerprinting
- Input schema validation
- Safe prediction pipeline
- Confidence assessment
- FastAPI service
- Health endpoint
- Prediction endpoint
- Probability endpoint
- Confidence endpoint
- OpenAPI documentation
- Swagger UI
- ReDoc documentation

The final deployment package passed artifact, schema, prediction, probability, and API validation checks.

### Rationale

Keeping preprocessing and prediction logic inside the validated production pipeline reduces the risk of training-serving inconsistencies.

Artifact versioning and fingerprinting additionally provide traceability for the deployed model package.

---

## 23.11 Solutions Implemented & Rationale

The major challenges and implemented solutions can be summarized as follows:

| Challenge | Solution | Rationale |
|---|---|---|
| Small dataset | Stratification, cross-validation, stability analysis | Improve confidence in generalization |
| Class imbalance | `class_weight="balanced"` | Give greater importance to minority classes without synthetic data |
| Missing Age | Numeric conversion + median imputation | Preserve observations while handling missing values |
| Invalid `"?"` values | Explicit validation and preprocessing | Prevent invalid tokens from entering the model |
| Multiclass classification | Macro metrics + probability analysis | Evaluate all six classes more fairly |
| Correlated features | Correlation threshold analysis | Detect redundancy without unnecessary feature removal |
| Overfitting risk | Train-test gap + CV + random-state analysis | Evaluate generalization |
| Model selection | Comparative evaluation + optimization | Select a robust production configuration |
| Explainability | Feature importance + error/probability analysis | Improve model inspection and human review |
| Deployment | Pipeline + validation + FastAPI + versioning | Create a reproducible production-oriented system |

### Overall Technical Rationale

The project followed a principle of **evidence-based engineering**.

Where a potential problem was identified, an appropriate validation or mitigation strategy was introduced. Where an analysis did not demonstrate a problem—for example, the correlation analysis did not identify features requiring removal—the original data representation was retained rather than applying unnecessary transformations.

This approach helped maintain a balance between:

- Predictive performance
- Generalization
- Data integrity
- Reproducibility
- Production safety
- Maintainability
- Clinical decision-support considerations

The final system therefore represents not only a trained classification model, but an end-to-end machine-learning workflow in which data quality, model behavior, deployment reliability, and responsible use were considered together.

# 24. Limitations & Risk Assessment

## Overview

Although the Skin Disorder Prediction System demonstrated strong predictive performance on the available dataset, several limitations must be considered before interpreting the results as evidence of real-world clinical effectiveness.

The limitations primarily relate to dataset size, minority-class representation, population representativeness, external validation, clinical validation, generalization, dataset shift, and probability calibration.

This section documents these limitations to ensure that the reported model performance is interpreted within the appropriate technical and clinical context.

---

## 24.1 Small Sample Size

### Limitation

The complete dataset contains only **366 observations** and 34 predictor variables distributed across six target classes.

A relatively small dataset limits the amount of information available for learning complex patterns and increases the possibility that performance estimates may vary across different samples.

The available dataset is sufficient for developing and evaluating a machine-learning prototype, but it is not sufficient by itself to establish broad real-world clinical reliability.

### Risk

A model trained on a small dataset may not capture the full variability present in real-world patient populations.

Performance observed on the current test set may therefore differ when the system is evaluated on larger and more diverse datasets.

### Mitigation

The project addressed this limitation through:

- Stratified train-test splitting
- Cross-validation
- Multiple evaluation metrics
- Random-state stability analysis
- Error analysis
- Generalization analysis
- Explicit documentation of dataset limitations

However, increasing the dataset size remains an important future requirement.

---

## 24.2 Minority-Class Sample Size

### Limitation

The six target classes are not equally represented.

The smallest class contains only **20 observations** in the complete dataset, resulting in only a small number of observations available for model training and testing.

This limits the statistical reliability of class-specific performance estimates for the minority class.

### Risk

A small number of observations can make minority-class metrics sensitive to individual samples.

For example, one additional incorrect prediction can have a relatively large effect on recall or F1-score when the class contains only a few test observations.

### Mitigation

The project used:

- Stratified splitting
- `class_weight="balanced"`
- Macro-level evaluation metrics
- Class-level error analysis
- Confusion matrix analysis
- Cross-validation

Nevertheless, substantially more minority-class observations would be required for reliable clinical validation.

---

## 24.3 Dataset Representativeness

### Limitation

The model was developed using the available dermatology dataset.

Potential differences may include:

- Age distributions
- Disease prevalence
- Patient demographics
- Clinical environments
- Symptom presentation
- Data collection procedures
- Missing-data patterns
- Feature distributions

### Mitigation

The current project documents the dataset characteristics and performs internal validation.

However, representative external datasets would be required to establish whether the learned patterns generalize across different populations and clinical environments.

---

## 24.4 External Validation Limitation

### Limitation

The current project uses internal train-test evaluation and cross-validation.

An independent external dataset from a different source was not used for final validation.

### Risk

Internal validation can provide optimistic estimates when the validation data shares characteristics with the training dataset.

Future validation should include independent datasets that were not involved in model development.

### Risk

### Mitigation

Before clinical deployment, the system would require appropriate clinical validation involving qualified healthcare professionals and representative patient data.

Potential validation stages include:

- Changes in disease prevalence

- Changes in data collection systems

## 24.8 Confidence / Calibration Limitations

### Limitation

The model provides class probabilities and a confidence score based on its predicted class probability.

However, model confidence should not be interpreted as guaranteed clinical certainty.

The project's calibration analysis showed an observed calibration error of approximately **0.0682**, with a maximum calibration error of approximately **0.5979**.

The multiclass Brier score was approximately **0.1377**, while log loss was approximately **0.3300**.

These results indicate that probability outputs provide useful decision-support information but should not automatically be interpreted as perfectly calibrated clinical probabilities.

### Risk

A high model probability does not necessarily mean that the corresponding clinical diagnosis is certain.

Similarly, a lower probability does not automatically mean that a condition is absent.

### Mitigation

The production pipeline uses a **0.70 confidence threshold** to identify higher-confidence predictions.

Predictions below the threshold can be treated as cases requiring additional human review.

The confidence mechanism should therefore be considered a **risk-control mechanism**, not a medical certainty measure.

Future clinical deployment should include dedicated probability calibration and validation on representative external data.

**High if used autonomously**

### Deployment Risk

**Controlled within the implemented prototype**

The project includes input validation, artifact validation, model versioning, safe prediction logic, probability analysis, confidence assessment, and FastAPI testing.

However, production healthcare deployment would require additional operational, security, privacy, monitoring, and clinical governance controls.

### Overall Assessment

The system should currently be classified as a:

**Production-Oriented Machine-Learning Decision-Support Prototype**

rather than a clinically validated autonomous diagnostic system.

The strongest evidence from the project supports the reliability of the implemented ML pipeline on the available dataset. It does not establish clinical effectiveness across broader patient populations.

Future deployment should therefore prioritize:

- Independent external validation
- Clinical validation
- Minority-class data expansion
- Dataset-shift monitoring
- Probability calibration
- Continuous performance monitoring
- Qualified human oversight
- Appropriate healthcare governance and regulatory review

This risk-aware interpretation ensures that the strong technical performance of the model is not overstated beyond what the available evidence supports.

### 29.2 Feature Description

The dataset contains **34 predictor variables** and one target variable.

#### Numeric Feature

**`Age`**

* Data type in raw dataset: object due to missing-value markers.
* Missing marker: `?`.
* Converted to numeric representation during preprocessing.
* Missing values are handled using **median imputation**.
* The resulting numeric feature is standardized using **StandardScaler**.

#### Categorical Features

The remaining **33 predictor variables** are categorical.

The categorical preprocessing pipeline applies:

1. Most-frequent-value imputation.
2. One-hot encoding.
3. `handle_unknown="ignore"` to support previously unseen categorical values during inference.

#### Target

**`class`**

The target contains six classes:

```text
1
2
3
4
5
6
```

### Transformed Feature Space

After preprocessing:

* Original predictor count: **34**
* Numeric predictors: **1**
* Categorical predictors: **33**
* Transformed feature count: **130**

The complete preprocessing process is integrated into the production pipeline:

```text
Raw Input
    ↓
Feature Validation
    ↓
ColumnTransformer
    ├── Age → Numeric Conversion → Median Imputation → StandardScaler
    │
    └── Categorical Features → Most-Frequent Imputation → OneHotEncoder
    ↓
130 Transformed Features
    ↓
Random Forest Classifier
```

This ensures that the same preprocessing logic is applied during both training and production inference.

### 29.3 Model Hyperparameters

The final production model is an optimized `RandomForestClassifier`.

```text
Model: RandomForestClassifier

n_estimators      = 150
max_depth         = 10
max_features      = "log2"
min_samples_leaf  = 4
min_samples_split = 8
criterion          = "gini"
class_weight      = "balanced"
random_state      = 42
n_jobs             = -1
```

#### Hyperparameter Purpose

| Parameter           |      Value | Purpose                                         |
| ------------------- | ---------: | ----------------------------------------------- |
| `n_estimators`      |        150 | Number of trees in the forest                   |
| `max_depth`         |         10 | Controls maximum tree complexity                |
| `max_features`      |     `log2` | Limits features considered at each split        |
| `min_samples_leaf`  |          4 | Reduces overly specific leaf nodes              |
| `min_samples_split` |          8 | Controls minimum samples required for splitting |
| `criterion`         |     `gini` | Split-quality criterion                         |
| `class_weight`      | `balanced` | Compensates for class imbalance                 |
| `random_state`      |         42 | Ensures reproducibility                         |
| `n_jobs`            |         -1 | Uses available CPU cores                        |

The final configuration was selected to provide strong multiclass performance while controlling model complexity and overfitting risk.

### 29.4 Cross-Validation Configuration

The final model was evaluated using **5-fold stratified cross-validation**.

| Configuration           | Value               |
| ----------------------- | ------------------- |
| Cross-validation method | Stratified K-Fold   |
| Number of folds         | 5                   |
| Evaluation metric       | Macro F1            |
| Random state            | 42                  |
| Primary CV result       | **98.59% Macro F1** |

Stratification ensures that each fold maintains a representative distribution of the six target classes.

Macro F1 was selected because it gives equal importance to each class and is therefore more informative than accuracy when evaluating an imbalanced multiclass problem.

The final cross-validation result was:

**CV Macro F1 = 98.59%**

This provides additional evidence that the final model's strong test performance was not solely dependent on a single train-test split.
